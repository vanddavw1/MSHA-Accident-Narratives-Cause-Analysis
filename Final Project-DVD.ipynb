## Data Import for MSHA Safety Data

````python
import csv
import pandas as pd

DATA_PATH = "accidents.txt"   # adjust to your AML folder path if needed

# --- 1. Detect the delimiter from the header row -------------------------
with open(DATA_PATH, "r", encoding="latin-1") as f:
    first_line = f.readline()

if "|" in first_line:
    sep = "|"
elif "\t" in first_line:
    sep = "\t"
else:
    sep = ","
print("Detected delimiter:", repr(sep))

# --- 2. Column typing ----------------------------------------------------
# ID/code fields stay as strings so leading zeros and '?' placeholders survive.
str_cols = [
    "MINE_ID", "CONTROLLER_ID", "OPERATOR_ID", "CONTRACTOR_ID", "DOCUMENT_NO",
    "SUBUNIT_CD", "FIPS_STATE_CD", "UG_LOCATION_CD", "UG_MINING_METHOD_CD",
    "MINING_EQUIP_CD", "EQUIP_MFR_CD", "EQUIP_MODEL_NO", "CLASSIFICATION_CD",
    "ACCIDENT_TYPE_CD", "DEGREE_INJURY_CD", "OCCUPATION_CD", "ACTIVITY_CD",
    "INJURY_SOURCE_CD", "NATURE_INJURY_CD", "INJ_BODY_PART_CD",
    "IMMED_NOTIFY_CD", "TRANS_TERM", "CLOSED_DOC_NO", "COAL_METAL_IND",
]

date_cols = ["ACCIDENT_DT", "RETURN_TO_WORK_DT", "INVEST_BEGIN_DT"]

# --- 3. Read the file ----------------------------------------------------
accidents = pd.read_csv(
    DATA_PATH,
    sep=sep,
    quotechar='"',
    encoding="latin-1",          # narratives contain non-UTF8 characters
    dtype={c: "string" for c in str_cols},
    na_values=["", "?", "NO VALUE FOUND", "No Value Found"],
    keep_default_na=True,
    low_memory=False,
    on_bad_lines="warn",
)

# --- 4. Parse dates (mm/dd/yyyy per the data dictionary) -----------------
for c in date_cols:
    if c in accidents.columns:
        accidents[c] = pd.to_datetime(accidents[c], format="%m/%d/%Y", errors="coerce")

# --- 5. Quick sanity check ----------------------------------------------
rows, cols = accidents.shape
print("Rows: {:,}".format(rows))
print("Columns: {}".format(cols))
print(accidents.columns.tolist())
accidents.head(3)
````

# EDA and Project Concept Development

## EDA Report Code

````python
# Full MSHA structured-data EDA and NLP PDF report
# Assumes a pandas DataFrame named `accidents` already exists in the notebook.

import os
import re
import textwrap
import warnings
from collections import Counter
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages

warnings.filterwarnings("ignore")

OUTPUT_PDF = "MSHA_EDA_Report.pdf"
TEXT_COL = "NARRATIVE"
DATE_COL = "ACCIDENT_DT"
TOP_N = 20
N_TOPICS = 8
MAX_FEATURES = 20000
RANDOM_STATE = 42

plt.rcParams.update({
    "figure.dpi": 110,
    "font.size": 9,
    "axes.grid": True,
    "grid.alpha": 0.25,
    "axes.spines.top": False,
    "axes.spines.right": False,
})

try:
    from sklearn.feature_extraction.text import TfidfVectorizer
    from sklearn.decomposition import NMF
    HAS_SKLEARN = True
except ImportError:
    HAS_SKLEARN = False

try:
    from wordcloud import WordCloud
    HAS_WORDCLOUD = True
except ImportError:
    HAS_WORDCLOUD = False

if "accidents" not in globals():
    raise NameError("Run the accidents.txt import cell first so a DataFrame named 'accidents' exists.")
if not isinstance(accidents, pd.DataFrame):
    raise TypeError("'accidents' must be a pandas DataFrame.")
if accidents.empty:
    raise ValueError("The 'accidents' DataFrame is empty.")

df = accidents.copy()
df.columns = df.columns.astype(str).str.strip().str.upper()

# Convert known MSHA placeholders to missing values.
PLACEHOLDERS = ["?", "NO VALUE FOUND", "No Value Found", "NOT MARKED", "Not Marked"]
df = df.replace(PLACEHOLDERS, np.nan)

# Convert known date columns.
for col in ["ACCIDENT_DT", "RETURN_TO_WORK_DT", "INVEST_BEGIN_DT"]:
    if col in df.columns:
        df[col] = pd.to_datetime(df[col], errors="coerce")

ID_CANDIDATES = [
    "MINE_ID", "CONTROLLER_ID", "OPERATOR_ID", "CONTRACTOR_ID",
    "DOCUMENT_NO", "CLOSED_DOC_NO", "EQUIP_MODEL_NO"
]
LABEL_CANDIDATES = [
    "SUBUNIT", "DEGREE_INJURY", "UG_LOCATION", "UG_MINING_METHOD",
    "MINING_EQUIP", "EQUIP_MFR_NAME", "CLASSIFICATION", "ACCIDENT_TYPE",
    "OCCUPATION", "ACTIVITY", "INJURY_SOURCE", "NATURE_INJURY",
    "INJ_BODY_PART", "IMMED_NOTIFY", "COAL_METAL_IND", "TRANS_TERM",
    "CONTROLLER_NAME", "OPERATOR_NAME"
]
ID_COLS = [c for c in ID_CANDIDATES if c in df.columns]
CODE_COLS = [c for c in df.columns if c.endswith("_CD")]
DATE_COLS = [c for c in df.columns if pd.api.types.is_datetime64_any_dtype(df[c])]
NUM_COLS = [c for c in df.columns if pd.api.types.is_numeric_dtype(df[c]) and c not in ID_COLS]
LABEL_COLS = [c for c in LABEL_CANDIDATES if c in df.columns]

pdf = PdfPages(OUTPUT_PDF)


def new_page(title=None, figsize=(8.5, 11), text_only=False):
    fig = plt.figure(figsize=figsize)
    if text_only:
        ax = fig.add_axes([0, 0, 1, 1])
        ax.axis("off")
        ax.grid(False)
    if title:
        fig.suptitle(title, fontsize=14, fontweight="bold", y=0.98)
    return fig


def save_page(fig):
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)


def add_text_pages(title, lines, font_size=8.5, lines_per_page=58):
    lines = lines or ["No information available."]
    chunks = [lines[i:i + lines_per_page] for i in range(0, len(lines), lines_per_page)]
    for page_no, chunk in enumerate(chunks):
        page_title = title if page_no == 0 else f"{title} (continued)"
        fig = new_page(page_title, text_only=True)
        fig.text(0.055, 0.935, "\n".join(chunk), va="top", ha="left",
                 fontsize=font_size, family="monospace")
        save_page(fig)


def dataframe_lines(frame, max_rows=100):
    with pd.option_context(
        "display.max_rows", max_rows,
        "display.max_columns", None,
        "display.width", 220,
        "display.max_colwidth", 40,
    ):
        return frame.head(max_rows).to_string().split("\n")


def barh(ax, labels, values, title, xlabel="Count", color="#2E5E8C"):
    y = np.arange(len(labels))
    ax.barh(y, values, color=color)
    ax.set_yticks(y)
    ax.set_yticklabels([str(x)[:48] for x in labels], fontsize=7.5)
    ax.invert_yaxis()
    ax.set_title(title, fontsize=10, fontweight="bold")
    ax.set_xlabel(xlabel, fontsize=8)


def safe_counts(series, n=TOP_N):
    return series.astype("string").fillna("<MISSING>").value_counts(dropna=False).head(n)


# 1. Cover page
fig = new_page(text_only=True)
fig.text(0.5, 0.75, "MSHA Accident and Injury Database", ha="center",
         fontsize=22, fontweight="bold")
fig.text(0.5, 0.69, "Structured-Data EDA and Narrative NLP Report", ha="center",
         fontsize=13, color="#444444")
fig.text(0.5, 0.64, datetime.now().strftime("Generated %B %d, %Y at %H:%M"),
         ha="center", fontsize=10, color="#666666")

date_coverage = "Not available"
if DATE_COL in df.columns and df[DATE_COL].notna().any():
    date_coverage = f"{df[DATE_COL].min():%Y-%m-%d} to {df[DATE_COL].max():%Y-%m-%d}"

cover_lines = [
    f"Records ...................... {len(df):,}",
    f"Columns ...................... {df.shape[1]:,}",
    f"Date coverage ................ {date_coverage}",
    f"Memory footprint ............. {df.memory_usage(deep=True).sum()/1_000_000:,.1f} MB",
    f"Exact duplicate rows ......... {df.duplicated().sum():,}",
    "",
    f"Identifier fields ............ {len(ID_COLS):,}",
    f"Code fields .................. {len(CODE_COLS):,}",
    f"Descriptive label fields ..... {len(LABEL_COLS):,}",
    f"Numeric fields ............... {len(NUM_COLS):,}",
    f"Date fields .................. {len(DATE_COLS):,}",
    "",
    "Report sections",
    "  1. Schema and completeness",
    "  2. Missingness",
    "  3. Numeric distributions and correlation",
    "  4. Categorical distributions",
    "  5. Code-to-label integrity",
    "  6. Temporal patterns",
    "  7. Narrative descriptive EDA",
    "  8. N-grams and word cloud",
    "  9. TF-IDF and NMF topics",
    " 10. Narrative/outcome comparisons",
    " 11. Findings and modeling recommendations",
]
fig.text(0.13, 0.54, "\n".join(cover_lines), va="top", fontsize=9.5, family="monospace")
save_page(fig)

# 2. Schema profile
profile_rows = []
for c in df.columns:
    example = str(df[c].dropna().iloc[0])[:35] if df[c].notna().any() else ""
    profile_rows.append({
        "column": c,
        "dtype": str(df[c].dtype),
        "non_null": int(df[c].notna().sum()),
        "missing": int(df[c].isna().sum()),
        "missing_%": round(100 * df[c].isna().mean(), 2),
        "unique": int(df[c].nunique(dropna=True)),
        "example": example,
    })
profile = pd.DataFrame(profile_rows).set_index("column")
add_text_pages("1. Schema and Data-Quality Profile",
               dataframe_lines(profile, len(profile)), font_size=7.0, lines_per_page=62)

# 3. Missingness
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_%": 100 * df.isna().mean(),
}).sort_values("missing_%", ascending=False)
missing = missing[missing["missing_count"] > 0]
if not missing.empty:
    fig = new_page("2a. Missing Values by Column")
    ax = fig.add_subplot(111)
    top = missing.head(40)
    barh(ax, top.index, top["missing_%"].values, "Columns with missing values",
         xlabel="Percent missing", color="#B4553C")
    ax.axvline(50, ls="--", lw=1, color="gray", label="50% missing")
    ax.legend(fontsize=8)
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    save_page(fig)
    add_text_pages("2b. Missing-Value Summary", dataframe_lines(missing.round(2), len(missing)),
                   font_size=7.5)

# 4. Numeric analysis
usable_num = [c for c in NUM_COLS if df[c].notna().any()]
if usable_num:
    numeric_summary = df[usable_num].describe().T
    numeric_summary["missing"] = df[usable_num].isna().sum()
    numeric_summary["missing_%"] = 100 * df[usable_num].isna().mean()
    numeric_summary["zero_%"] = 100 * (df[usable_num] == 0).mean()
    numeric_summary["skew"] = df[usable_num].skew(numeric_only=True)
    add_text_pages("3a. Numeric Variable Summary",
                   dataframe_lines(numeric_summary.round(2), len(numeric_summary)), font_size=7.0)

    plot_cols = usable_num[:15]
    nrows = int(np.ceil(len(plot_cols) / 3))
    fig = new_page("3b. Numeric Variable Distributions", figsize=(8.5, max(8, 2.6 * nrows)))
    for i, c in enumerate(plot_cols, 1):
        ax = fig.add_subplot(nrows, 3, i)
        s = pd.to_numeric(df[c], errors="coerce").dropna()
        if s.nunique() > 1:
            lo, hi = s.quantile([0.01, 0.99])
            ax.hist(s[s.between(lo, hi)], bins=35, color="#2E5E8C", edgecolor="white")
        ax.set_title(c, fontsize=8, fontweight="bold")
        ax.tick_params(labelsize=6.5)
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    save_page(fig)

    corr_cols = [c for c in usable_num if df[c].nunique(dropna=True) > 2][:15]
    if len(corr_cols) >= 3:
        corr = df[corr_cols].corr()
        fig = new_page("3c. Numeric Correlation Matrix", figsize=(8.5, 8.5))
        ax = fig.add_subplot(111)
        image = ax.imshow(corr, cmap="RdBu_r", vmin=-1, vmax=1)
        ax.set_xticks(range(len(corr_cols)))
        ax.set_xticklabels(corr_cols, rotation=90, fontsize=7)
        ax.set_yticks(range(len(corr_cols)))
        ax.set_yticklabels(corr_cols, fontsize=7)
        ax.grid(False)
        for r in range(len(corr_cols)):
            for c in range(len(corr_cols)):
                value = corr.iloc[r, c]
                ax.text(c, r, f"{value:.2f}", ha="center", va="center", fontsize=6,
                        color="white" if abs(value) > 0.55 else "black")
        fig.colorbar(image, ax=ax, shrink=0.75)
        fig.tight_layout(rect=[0, 0, 1, 0.96])
        save_page(fig)

# 5. Categorical analysis
cat_cols = [c for c in LABEL_COLS if 1 < df[c].nunique(dropna=True) <= 500]
for start in range(0, len(cat_cols), 4):
    chosen = cat_cols[start:start + 4]
    fig = new_page("4. Categorical Variable Distributions", figsize=(8.5, 11))
    for i, c in enumerate(chosen, 1):
        ax = fig.add_subplot(2, 2, i)
        counts = safe_counts(df[c])
        barh(ax, counts.index, counts.values, c, xlabel="Records")
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    save_page(fig)

# 6. Code-to-label integrity
pair_rows = []
for code_col in CODE_COLS:
    label_col = code_col[:-3]
    if label_col in df.columns:
        pair = df[[code_col, label_col]].dropna()
        multi_label = pair.groupby(code_col)[label_col].nunique() if not pair.empty else pd.Series(dtype=float)
        pair_rows.append({
            "code_column": code_col,
            "label_column": label_col,
            "unique_codes": int(pair[code_col].nunique()),
            "unique_labels": int(pair[label_col].nunique()),
            "codes_with_multiple_labels": int((multi_label > 1).sum()),
        })
if pair_rows:
    pair_summary = pd.DataFrame(pair_rows)
    add_text_pages("5. Code and Description Consistency",
                   dataframe_lines(pair_summary, len(pair_summary)), font_size=7.5)

# 7. Temporal analysis
if DATE_COL in df.columns and df[DATE_COL].notna().any():
    dated = df[df[DATE_COL].notna()].copy()
    fig = new_page("6. Temporal Accident Patterns", figsize=(8.5, 11))

    ax = fig.add_subplot(3, 1, 1)
    yearly = dated[DATE_COL].dt.year.value_counts().sort_index()
    ax.plot(yearly.index, yearly.values, marker="o", markersize=3, color="#2E5E8C")
    ax.set_title("Records by year", fontsize=10, fontweight="bold")
    ax.set_xlabel("Year")
    ax.set_ylabel("Records")

    ax = fig.add_subplot(3, 2, 3)
    monthly = dated[DATE_COL].dt.month.value_counts().sort_index()
    ax.bar(monthly.index, monthly.values, color="#2E5E8C")
    ax.set_xticks(range(1, 13))
    ax.set_title("Records by month", fontsize=10, fontweight="bold")

    ax = fig.add_subplot(3, 2, 4)
    weekday = dated[DATE_COL].dt.dayofweek.value_counts().sort_index()
    ax.bar(weekday.index, weekday.values, color="#2E5E8C")
    ax.set_xticks(range(7))
    ax.set_xticklabels(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"], fontsize=7)
    ax.set_title("Records by weekday", fontsize=10, fontweight="bold")

    if "ACCIDENT_TIME" in df.columns:
        ax = fig.add_subplot(3, 1, 3)
        t = pd.to_numeric(df["ACCIDENT_TIME"], errors="coerce")
        hour = (t // 100)
        hour = hour[hour.between(0, 23)]
        hourly = hour.value_counts().sort_index()
        ax.bar(hourly.index, hourly.values, color="#2E5E8C")
        ax.set_xticks(range(24))
        ax.set_title("Records by hour of day", fontsize=10, fontweight="bold")
        ax.set_xlabel("Hour on 24-hour clock")
        ax.set_ylabel("Records")

    fig.tight_layout(rect=[0, 0, 1, 0.96])
    save_page(fig)

# 8. NLP helpers
STOPWORDS = set("""
a about above after again against all am an and any are as at be because been before
being below between both but by can cannot could did do does doing down during each few
for from further had has have having he her here hers herself him himself his how i if
in into is it its itself just me more most my myself no nor not of off on once only or
other our ours ourselves out over own same she should so some such than that the their
theirs them themselves then there these they this those through to too under until up
very was we were what when where which while who whom why will with would you your yours
yourself yourselves
""".split())
DOMAIN_STOPWORDS = {
    "accident", "employee", "employees", "injured", "injury", "miner", "miners",
    "occurred", "reported", "resulted", "resulting", "stated", "sustained",
    "worker", "workers"
}
TOKEN_PATTERN = re.compile(r"[a-z][a-z'\-]+")


def tokenize_document(document, remove_domain_stopwords=False):
    tokens = TOKEN_PATTERN.findall(str(document).lower())
    blocked = STOPWORDS | DOMAIN_STOPWORDS if remove_domain_stopwords else STOPWORDS
    return [t for t in tokens if t not in blocked and len(t) > 2]


# 9. Narrative EDA and NLP
narratives = pd.Series(dtype="string")
valid_narrative_mask = pd.Series(False, index=df.index)
if TEXT_COL in df.columns:
    narrative_series = df[TEXT_COL].astype("string").fillna("").str.strip()
    valid_narrative_mask = narrative_series.str.len() > 0
    narratives = narrative_series.loc[valid_narrative_mask].copy()

if not narratives.empty:
    character_length = narratives.str.len()
    word_length = narratives.str.split().str.len()
    sentence_count = narratives.str.count(r"[.!?]+").clip(lower=1)
    words_per_sentence = word_length / sentence_count

    all_tokens = [token for doc in narratives for token in tokenize_document(doc)]
    vocabulary_counter = Counter(all_tokens)
    vocabulary_size = len(vocabulary_counter)
    total_tokens = len(all_tokens)
    single_use_terms = sum(1 for count in vocabulary_counter.values() if count == 1)

    # Regex calculations are intentionally outside f-string expressions.
    digit_pattern = r"\d"
    non_ascii_pattern = r"[^\x00-\x7F]"
    repeated_space_pattern = r"\s{2,}"
    ending_punctuation_pattern = r"[.!?]$"

    uppercase_count = int(narratives.str.isupper().sum())
    digits_count = int(narratives.str.contains(digit_pattern, regex=True, na=False).sum())
    non_ascii_count = int(narratives.str.contains(non_ascii_pattern, regex=True, na=False).sum())
    repeated_space_count = int(narratives.str.contains(repeated_space_pattern, regex=True, na=False).sum())
    ending_punctuation = narratives.str.contains(ending_punctuation_pattern, regex=True, na=False)
    no_ending_punctuation_count = int((~ending_punctuation).sum())

    n = len(narratives)
    duplicate_count = int(narratives.duplicated().sum())
    near_limit_count = int((character_length >= 383).sum())
    short_count = int((word_length < 5).sum())

    narrative_lines = [
        "NARRATIVE COMPLETENESS", "-" * 78,
        f"Narratives present ................. {n:,} of {len(df):,}",
        f"Completion rate .................... {100*n/len(df):.2f}%",
        f"Blank or missing .................. {len(df)-n:,}",
        f"Unique narratives ................. {narratives.nunique():,}",
        f"Exact duplicate narratives ........ {duplicate_count:,}",
        f"Duplicate percentage .............. {100*duplicate_count/n:.2f}%",
        "", "LENGTH IN CHARACTERS", "-" * 78,
        f"Mean / median ...................... {character_length.mean():,.1f} / {character_length.median():,.1f}",
        f"Minimum / maximum ................. {int(character_length.min()):,} / {int(character_length.max()):,}",
        f"P90 / P95 / P99 ................... {character_length.quantile(.90):,.1f} / {character_length.quantile(.95):,.1f} / {character_length.quantile(.99):,.1f}",
        f"At or near 384-character limit .... {near_limit_count:,} ({100*near_limit_count/n:.2f}%)",
        "", "LENGTH IN WORDS", "-" * 78,
        f"Mean / median ...................... {word_length.mean():,.1f} / {word_length.median():,.1f}",
        f"Minimum / maximum ................. {int(word_length.min()):,} / {int(word_length.max()):,}",
        f"Under five words .................. {short_count:,} ({100*short_count/n:.2f}%)",
        "", "SENTENCE STRUCTURE", "-" * 78,
        f"Mean sentences per narrative ...... {sentence_count.mean():,.2f}",
        f"Mean words per sentence ........... {words_per_sentence.mean():,.2f}",
        "", "VOCABULARY", "-" * 78,
        f"Total usable tokens ............... {total_tokens:,}",
        f"Unique terms ...................... {vocabulary_size:,}",
        f"Type-token ratio .................. {vocabulary_size/max(total_tokens,1):.5f}",
        f"Terms occurring once .............. {single_use_terms:,}",
        "", "FORMATTING", "-" * 78,
        f"All-uppercase narratives .......... {uppercase_count:,}",
        f"Narratives containing digits ...... {digits_count:,}",
        f"Narratives containing non-ASCII ... {non_ascii_count:,}",
        f"Narratives with repeated spaces ... {repeated_space_count:,}",
        f"Without ending punctuation ........ {no_ending_punctuation_count:,}",
    ]
    add_text_pages("7a. Narrative Descriptive Statistics", narrative_lines, font_size=8.5)

    # Narrative charts
    fig = new_page("7b. Narrative Length and Vocabulary", figsize=(8.5, 8.5))
    ax = fig.add_subplot(2, 2, 1)
    ax.hist(character_length, bins=60, color="#2E5E8C", edgecolor="white")
    ax.axvline(384, ls="--", lw=1.25, color="red", label="384-character limit")
    ax.set_title("Characters per narrative", fontsize=9, fontweight="bold")
    ax.legend(fontsize=7)

    ax = fig.add_subplot(2, 2, 2)
    ax.hist(word_length, bins=60, color="#2E5E8C", edgecolor="white")
    ax.set_title("Words per narrative", fontsize=9, fontweight="bold")

    frequencies = np.array(sorted(vocabulary_counter.values(), reverse=True))
    ax = fig.add_subplot(2, 2, 3)
    if len(frequencies):
        ranks = np.arange(1, len(frequencies) + 1)
        ax.loglog(ranks, frequencies, color="#2E5E8C")
    ax.set_title("Zipf curve", fontsize=9, fontweight="bold")
    ax.set_xlabel("Term rank")
    ax.set_ylabel("Frequency")

    ax = fig.add_subplot(2, 2, 4)
    if len(frequencies):
        coverage = np.cumsum(frequencies) / frequencies.sum()
        ax.plot(np.arange(1, len(frequencies) + 1), coverage, color="#2E5E8C")
        ax.set_xscale("log")
    ax.set_title("Cumulative vocabulary coverage", fontsize=9, fontweight="bold")
    ax.set_xlabel("Top N terms")
    ax.set_ylabel("Share of tokens")
    fig.tight_layout(rect=[0, 0, 1, 0.95])
    save_page(fig)

    # N-grams
    tokenized = [tokenize_document(doc, remove_domain_stopwords=True) for doc in narratives]
    unigram_counter = Counter(t for tokens in tokenized for t in tokens)
    bigram_counter = Counter(" ".join(pair) for tokens in tokenized for pair in zip(tokens, tokens[1:]))
    trigram_counter = Counter(" ".join(group) for tokens in tokenized for group in zip(tokens, tokens[1:], tokens[2:]))

    fig = new_page("8a. Most Frequent Terms and Phrases", figsize=(8.5, 11))
    for i, (counter, name) in enumerate([
        (unigram_counter, "Unigrams"),
        (bigram_counter, "Bigrams"),
        (trigram_counter, "Trigrams"),
    ], 1):
        ax = fig.add_subplot(3, 1, i)
        common = counter.most_common(TOP_N)
        barh(ax, [x[0] for x in common], [x[1] for x in common],
             f"Top {TOP_N} {name}", xlabel="Frequency")
    fig.tight_layout(rect=[0, 0, 1, 0.96])
    save_page(fig)

    if HAS_WORDCLOUD and unigram_counter:
        cloud = WordCloud(width=1600, height=900, background_color="white",
                          colormap="viridis", max_words=200).generate_from_frequencies(unigram_counter)
        fig = new_page("8b. Narrative Word Cloud", figsize=(8.5, 6))
        ax = fig.add_subplot(111)
        ax.imshow(cloud, interpolation="bilinear")
        ax.axis("off")
        ax.grid(False)
        fig.tight_layout(rect=[0, 0, 1, 0.95])
        save_page(fig)

    # TF-IDF and NMF
    if HAS_SKLEARN:
        min_df = 5 if n >= 100 else 2
        vectorizer = TfidfVectorizer(
            max_features=MAX_FEATURES,
            stop_words="english",
            ngram_range=(1, 2),
            min_df=min_df,
            sublinear_tf=True,
            strip_accents="unicode",
        )
        X = vectorizer.fit_transform(narratives)
        terms = np.array(vectorizer.get_feature_names_out())
        mean_scores = np.asarray(X.mean(axis=0)).ravel()
        order = mean_scores.argsort()[::-1][:TOP_N]

        fig = new_page("8c. Highest Mean TF-IDF Terms", figsize=(8.5, 6))
        ax = fig.add_subplot(111)
        barh(ax, terms[order], mean_scores[order], "Highest average TF-IDF terms",
             xlabel="Mean TF-IDF", color="#4B7F52")
        fig.tight_layout(rect=[0, 0, 1, 0.95])
        save_page(fig)

        n_docs = X.shape[0]
        n_features = X.shape[1]
        nonzero = X.nnz
        possible = n_docs * n_features
        sparsity = 100 * (1 - nonzero / possible) if possible else np.nan
        mean_nonzero = nonzero / n_docs if n_docs else np.nan
        tfidf_lines = [
            f"Document-term matrix shape ........ {n_docs:,} x {n_features:,}",
            f"Stored non-zero entries ........... {nonzero:,}",
            f"Matrix sparsity ................... {sparsity:.4f}%",
            f"Mean non-zero features/document ... {mean_nonzero:,.1f}",
            "",
            "Vectorizer settings",
            "  Features: unigrams and bigrams",
            "  English stopwords removed",
            f"  Minimum document frequency: {min_df:,}",
            "  Sublinear term-frequency scaling",
            f"  Maximum vocabulary size: {MAX_FEATURES:,}",
        ]
        add_text_pages("8d. TF-IDF Matrix Characteristics", tfidf_lines, font_size=9)

        valid_topics = min(N_TOPICS, X.shape[0], X.shape[1])
        if valid_topics >= 2:
            nmf = NMF(n_components=valid_topics, random_state=RANDOM_STATE,
                      init="nndsvda", max_iter=400)
            W = nmf.fit_transform(X)
            H = nmf.components_
            assigned = W.argmax(axis=1)
            topic_counts = pd.Series(assigned).value_counts().sort_index()
            topic_lines = []
            topic_labels = []
            topic_values = []
            for k in range(valid_topics):
                top_terms = terms[H[k].argsort()[::-1][:12]]
                count = int(topic_counts.get(k, 0))
                topic_lines.extend([
                    f"TOPIC {k+1}: {count:,} narratives ({100*count/n:.2f}%)",
                    "   " + ", ".join(top_terms),
                    "",
                ])
                topic_labels.append(f"Topic {k+1}: " + ", ".join(top_terms[:3]))
                topic_values.append(count)
            add_text_pages(f"9a. NMF Topic Model: {valid_topics} Topics", topic_lines, font_size=9)

            fig = new_page("9b. Topic Prevalence", figsize=(8.5, 6))
            ax = fig.add_subplot(111)
            barh(ax, topic_labels, topic_values, "Narratives assigned to dominant NMF topic",
                 xlabel="Narratives", color="#7A4E8C")
            fig.tight_layout(rect=[0, 0, 1, 0.95])
            save_page(fig)

    # Narrative length versus outcomes
    comparison_cols = [c for c in ["DEGREE_INJURY", "CLASSIFICATION", "COAL_METAL_IND"] if c in df.columns][:2]
    if comparison_cols:
        fig = new_page("10a. Narrative Length by Incident Outcome", figsize=(8.5, 9))
        for i, c in enumerate(comparison_cols, 1):
            ax = fig.add_subplot(len(comparison_cols), 1, i)
            comparison = pd.DataFrame({
                "category": df.loc[valid_narrative_mask, c].astype("string").fillna("<MISSING>"),
                "word_count": word_length,
            })
            categories = comparison["category"].value_counts().head(10).index
            values = [comparison.loc[comparison["category"] == category, "word_count"].values
                      for category in categories]
            ax.boxplot(values, vert=False, showfliers=False)
            ax.set_yticklabels([str(x)[:45] for x in categories], fontsize=7)
            ax.set_xlabel("Words per narrative")
            ax.set_title(f"Narrative length by {c}", fontsize=10, fontweight="bold")
        fig.tight_layout(rect=[0, 0, 1, 0.95])
        save_page(fig)

    # Sample narratives
    sample_lines = []
    for i, narrative in enumerate(narratives.sample(min(8, n), random_state=RANDOM_STATE), 1):
        sample_lines.append(f"[Sample {i}]")
        sample_lines.extend("    " + line for line in textwrap.wrap(str(narrative), width=95))
        sample_lines.append("")
    add_text_pages("10b. Random Sample of Narratives", sample_lines, font_size=8)
else:
    add_text_pages("7. Narrative Analysis", [
        f"No usable text was found in column {TEXT_COL}." if TEXT_COL in df.columns
        else f"The expected column {TEXT_COL} was not found.",
        "",
        "Available columns:",
        ", ".join(df.columns.astype(str).tolist()),
    ], font_size=8.5)

# 10. Findings and recommendations
high_missing = missing.index[missing["missing_%"] > 50].tolist() if not missing.empty else []
findings = [
    "DATA-QUALITY FINDINGS", "-" * 78,
    f"Columns over 50% missing ........... {len(high_missing):,}",
]
if high_missing:
    findings.append("  " + ", ".join(high_missing[:15]))
if not narratives.empty:
    findings.extend([
        f"Narratives near 384-char limit .... {int((narratives.str.len() >= 383).sum()):,}",
        f"Exact duplicate narratives ........ {int(narratives.duplicated().sum()):,}",
    ])
findings.extend([
    "", "DATA-HANDLING CONSIDERATIONS", "-" * 78,
    "* Keep identifiers as strings so leading zeros are preserved.",
    "* Review code and description pairs before using both in one model.",
    "* Validate whether ACCIDENT_TIME = 0 means midnight or not recorded.",
    "* Keep duplicate narratives within the same train/test partition.",
    "* Treat narratives near the source-field limit as potentially truncated.",
    "", "MODELING RECOMMENDATIONS", "-" * 78,
    "* Start with TF-IDF plus logistic regression or a linear SVM.",
    "* Candidate targets include DEGREE_INJURY, CLASSIFICATION, ACCIDENT_TYPE,",
    "  INJURY_SOURCE, and INJ_BODY_PART.",
    "* Use stratified evaluation and report macro-F1, precision, and recall.",
    "* Compare structured-only, narrative-only, and combined models.",
    "* Consider grouping train/test splits by MINE_ID to test generalization.",
    "* Treat NMF topics as exploratory themes requiring subject-matter review.",
])
add_text_pages("11. Findings and Modeling Recommendations", findings, font_size=9)

# Close file and report status.
metadata = pdf.infodict()
metadata["Title"] = "MSHA Accident and Injury EDA and NLP Report"
metadata["Author"] = "Python EDA and NLP analysis"
metadata["Subject"] = "Structured and narrative analysis of MSHA accident records"
metadata["CreationDate"] = datetime.now()
pdf.close()

output_path = os.path.abspath(OUTPUT_PDF)
print("=" * 80)
print("EDA report successfully created.")
print(f"PDF file: {output_path}")
print(f"Records analyzed: {len(df):,}")
print(f"Columns analyzed: {df.shape[1]:,}")
print(f"Nonblank narratives analyzed: {int(valid_narrative_mask.sum()):,}")
if not HAS_SKLEARN:
    print("TF-IDF and NMF were skipped because scikit-learn is not installed.")
if not HAS_WORDCLOUD:
    print("The word cloud was skipped because wordcloud is not installed.")
print("=" * 80)
````

````python
"""
MSHA NARRATIVE TERM AND ENTITY EXTRACTION COMPARISON
====================================================
Prerequisite: pandas DataFrame `accidents` is already loaded.
Output: MSHA_Narrative_Extraction_Final_Report.pdf

This final version:
- locks the analysis to the two supplied MSHA document numbers;
- includes the full stored narrative text and structured metadata;
- uses a reviewed manual ground-truth estimate that can be edited;
- includes the supplied FiveFilters, Dandelion, and Wordcount.com outputs;
- compares Count n-grams, TF-IDF, RAKE-like phrases, NLTK frequency,
  two online term methods, and two online entity/phrase methods;
- preserves ranks instead of inventing unavailable online scores;
- separates exact equivalence from broader domain relatedness;
- produces an organized PDF with final submission wording.
"""

import os
import re
import textwrap
import warnings
from collections import Counter
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

warnings.filterwarnings("ignore")

# =============================================================================
# 1. CONFIGURATION AND FINAL INPUTS
# =============================================================================

OUTPUT_PDF = "MSHA_Narrative_Extraction_Final_Report.pdf"
TEXT_COL = "NARRATIVE"
TOP_K = 20

# Lock the final report to the two reviewed records.
DOCUMENT_1_NO = "220021510007"
DOCUMENT_2_NO = "220122850018"

# Manual ground truth is a reasoned human-style estimate based on the two narratives.
# Edit any row if your own reading differs. Terms are not required to occur in both
# documents; the code calculates separate counts for each document.
MANUAL_GROUND_TRUTH = [
    # Shared domain entities/concepts
    {"term": "roof bolter", "kind": "ENTITY", "entity_type": "EQUIPMENT", "canonical": "roof-bolting machine"},
    {"term": "roof bolts", "kind": "ENTITY", "entity_type": "MATERIAL", "canonical": "roof bolt"},
    {"term": "loading supplies", "kind": "DOMAIN_CONCEPT", "entity_type": "ACTIVITY", "canonical": "material handling"},

    # Document 1
    {"term": "intake roof bolter", "kind": "ENTITY", "entity_type": "EQUIPMENT", "canonical": "roof-bolting machine"},
    {"term": "four foot roof bolts", "kind": "ENTITY", "entity_type": "MATERIAL", "canonical": "roof bolt"},
    {"term": "bundle straps", "kind": "ENTITY", "entity_type": "MATERIAL_RESTRAINT", "canonical": "retaining band"},
    {"term": "bands", "kind": "ENTITY", "entity_type": "MATERIAL_RESTRAINT", "canonical": "retaining band"},
    {"term": "removing these bolts", "kind": "TERM", "entity_type": "ACTIVITY", "canonical": "remove roof bolts"},
    {"term": "new bundle", "kind": "ENTITY", "entity_type": "MATERIAL", "canonical": "roof-bolt bundle"},
    {"term": "pulling these bolts off by hand", "kind": "DOMAIN_CONCEPT", "entity_type": "ACTIVITY", "canonical": "manual material handling"},
    {"term": "straps laying loose", "kind": "DOMAIN_CONCEPT", "entity_type": "HAZARD", "canonical": "loose retaining band"},
    {"term": "caught one of these straps", "kind": "DOMAIN_CONCEPT", "entity_type": "TRIGGER", "canonical": "strap caught during handling"},
    {"term": "strap", "kind": "ENTITY", "entity_type": "MATERIAL_RESTRAINT", "canonical": "retaining band"},
    {"term": "EE", "kind": "ENTITY", "entity_type": "PERSON_REFERENCE", "canonical": "employee"},

    # Document 2
    {"term": "load roof bolts", "kind": "TERM", "entity_type": "ACTIVITY", "canonical": "load roof bolts"},
    {"term": "supply tray", "kind": "ENTITY", "entity_type": "EQUIPMENT_COMPONENT", "canonical": "roof-bolter supply tray"},
    {"term": "scoop bucket", "kind": "ENTITY", "entity_type": "EQUIPMENT_COMPONENT", "canonical": "scoop bucket"},
    {"term": "cut the bands loose", "kind": "DOMAIN_CONCEPT", "entity_type": "TRIGGER", "canonical": "release retaining bands"},
    {"term": "roof bolt came up", "kind": "DOMAIN_CONCEPT", "entity_type": "HAZARD_EVENT", "canonical": "unexpected roof-bolt movement"},
    {"term": "struck him in the mouth", "kind": "DOMAIN_CONCEPT", "entity_type": "MECHANISM", "canonical": "struck-by event"},
    {"term": "mouth", "kind": "ENTITY", "entity_type": "BODY_PART", "canonical": "mouth"},
    {"term": "bridge", "kind": "ENTITY", "entity_type": "DENTAL_APPLIANCE", "canonical": "dental bridge"},
    {"term": "permanent tooth", "kind": "ENTITY", "entity_type": "BODY_PART", "canonical": "tooth"},
    {"term": "IE", "kind": "ENTITY", "entity_type": "PERSON_REFERENCE", "canonical": "injured employee"},
]

# Online term extractor results. Since the sites supplied ordered lists but no
# numerical scores or counts, the report stores display rank only.
ONLINE_TERM_RESULTS = {
    "FiveFilters": {
        "doc1": [
            "REMOVING THESE BOLTS", "LOADING SUPPLIES", "BOLTER ALONG",
            "BUNDLE STRAPS", "COULD PLACE", "NEW BUNDLE", "EE", "BOLTER",
            "BANDS", "BOLTS", "AS", "CAUGHT", "TOP", "TEND", "STRAP",
        ],
        "doc2": [
            "load roof bolts", "roof bolts", "supply tray", "roof bolter",
            "scoop bucket", "IE", "mouth", "bands", "bridge", "edge",
            "tooth", "incident",
        ],
    },
    "Wordcount.com": {
        "doc1": [
            "intake roof bolter", "loading supplies", "four foot roof bolts",
            "bundle straps", "bolts on the bolter", "removing bolts",
            "placing a new bundle", "pulling bolts off by hand",
            "strap laying loose", "strap tension", "tend of the strap struck",
            "caught one strap", "EE was removing bolts",
        ],
        "doc2": [
            "roof bolts", "supply tray", "roof bolter", "IE (incident employee)",
            "cut the bands loose", "roof bolt struck him in the mouth",
            "bridge in his mouth loosened", "edge of the scoop bucket",
            "set the edge of the scoop bucket down", "permanent tooth pulled",
            "permanent tooth", "incident injury", "loosened tooth during incident",
            "10/4/12",
        ],
    },
}

# Online entity extraction results. Dandelion is the named-entity engine.
# Wordcount.com is retained as the second online phrase/entity-oriented comparison
# because its output includes objects, equipment, body parts, and event phrases.
ONLINE_ENTITY_RESULTS = {
    "Dandelion": {
        "doc1": ["EE", "Bolts", "Bolter", "Strap"],
        "doc2": ["Roof", "Bolts", "bolter", "Scoop", "Bucket", "Permanent tooth", "Incident"],
    },
    "Wordcount.com phrase/entity": {
        "doc1": [
            "intake roof bolter", "four foot roof bolts", "bundle straps",
            "bolts on the bolter", "strap laying loose", "strap tension",
            "EE was removing bolts",
        ],
        "doc2": [
            "roof bolts", "supply tray", "roof bolter", "IE (incident employee)",
            "bridge in his mouth loosened", "edge of the scoop bucket",
            "permanent tooth", "10/4/12",
        ],
    },
}

# Exact-equivalence groups are stricter than general conceptual relatedness.
EXACT_EQUIVALENCE_GROUPS = [
    {"canonical": "retaining band", "forms": ["band", "bands", "strap", "straps", "bundle straps"]},
    {"canonical": "roof bolt", "forms": ["roof bolt", "roof bolts", "bolts"]},
    {"canonical": "roof-bolting machine", "forms": ["bolter", "roof bolter", "intake roof bolter"]},
    {"canonical": "employee", "forms": ["EE", "employee"]},
    {"canonical": "injured employee", "forms": ["IE", "injured employee"]},
]

# Broader concepts intentionally group related but not always synonymous entities.
DOMAIN_CONCEPTS = {
    "roof-bolting work": ["roof bolt", "roof bolts", "bolts", "roof bolter", "bolter", "intake roof bolter"],
    "material restraint": ["band", "bands", "strap", "straps", "bundle straps"],
    "material handling": ["loading supplies", "removing bolts", "load roof bolts", "pulling bolts off by hand"],
    "roof-bolter components": ["supply tray"],
    "mobile equipment components": ["scoop bucket", "bucket"],
    "unexpected material movement": ["roof bolt came up", "struck him in the mouth", "strap struck"],
    "oral injury": ["mouth", "bridge", "permanent tooth", "tooth"],
    "person reference": ["EE", "IE", "employee", "injured employee"],
}

STOPWORDS = set("""
a about above after again against all am an and any are as at be because been before
being below between both but by can could did do does doing down during each few for
from further had has have having he her here hers herself him himself his how i if in
into is it its itself just me more most my myself no nor not of off on once only or
other our ours ourselves out over own same she should so some such than that the their
theirs them themselves then there these they this those through to too under until up
very was we were what when where which while who whom why will with would you your yours
yourself yourselves employee employees miner miners worker workers accident incident
injured injury injuries occurred reported stated sustained resulting caused causing
""".split())

# =============================================================================
# 2. LOAD THE TWO FINAL DOCUMENTS
# =============================================================================

if "accidents" not in globals():
    raise NameError("A pandas DataFrame named 'accidents' must already be loaded.")
if not isinstance(accidents, pd.DataFrame) or accidents.empty:
    raise ValueError("'accidents' must be a non-empty pandas DataFrame.")

df = accidents.copy()
df.columns = df.columns.astype(str).str.strip().str.upper()
if TEXT_COL not in df.columns or "DOCUMENT_NO" not in df.columns:
    raise KeyError("The DataFrame must contain DOCUMENT_NO and NARRATIVE columns.")
df[TEXT_COL] = df[TEXT_COL].astype("string").fillna("").str.strip()
df["_DOCUMENT_NO_TEXT"] = df["DOCUMENT_NO"].astype(str).str.strip().str.replace(r"\.0$", "", regex=True)


def get_record(document_number):
    matches = df[df["_DOCUMENT_NO_TEXT"] == str(document_number)]
    if matches.empty:
        raise KeyError(f"DOCUMENT_NO={document_number} was not found in accidents.")
    return matches.iloc[0]


row1 = get_record(DOCUMENT_1_NO)
row2 = get_record(DOCUMENT_2_NO)
doc1 = re.sub(r"\s+", " ", str(row1[TEXT_COL])).strip()
doc2 = re.sub(r"\s+", " ", str(row2[TEXT_COL])).strip()
documents = [doc1, doc2]
topic_column = "CLASSIFICATION"
topic_value = str(row1.get("CLASSIFICATION", "HANDLING OF MATERIALS"))


def record_id(row):
    ids = []
    for col in ["DOCUMENT_NO", "MINE_ID"]:
        if col in row.index and pd.notna(row[col]):
            ids.append(f"{col}={row[col]}")
    return "; ".join(ids)

# =============================================================================
# 3. TEXT, TERM, PHRASE, AND ENTITY METHODS
# =============================================================================

TOKEN_RE = re.compile(r"[A-Za-z][A-Za-z'\-]+")


def phrase_count(text, phrase):
    pattern = r"(?<!\w)" + re.escape(str(phrase).lower()) + r"(?!\w)"
    return len(re.findall(pattern, text.lower()))


def normalize_term(value):
    return re.sub(r"\s+", " ", str(value)).strip().lower()


def ranked_list(values):
    return [(normalize_term(term), f"Rank {rank}", phrase_count(doc1 if False else "", term))
            for rank, term in enumerate(values, 1)]


def online_ranked(values, document_text):
    return [
        (normalize_term(term), f"Rank {rank}", phrase_count(document_text, term))
        for rank, term in enumerate(values, 1)
    ]


def nltk_analysis(text):
    try:
        import nltk
        from nltk import FreqDist
        from nltk.stem import PorterStemmer
        try:
            raw = nltk.word_tokenize(text)
            tokenizer = "nltk.word_tokenize"
        except LookupError:
            raw = nltk.tokenize.wordpunct_tokenize(text)
            tokenizer = "nltk.wordpunct_tokenize (offline fallback)"
        clean = [x.lower() for x in raw if re.fullmatch(r"[A-Za-z][A-Za-z'\-]+", x)]
        filtered = [x for x in clean if x not in STOPWORDS and len(x) > 2]
        stems = [PorterStemmer().stem(x) for x in filtered]
        return {"tokenizer": tokenizer, "raw": clean, "filtered": filtered,
                "stems": stems, "freq": FreqDist(filtered).most_common(TOP_K)}
    except ImportError:
        clean = [x.lower() for x in TOKEN_RE.findall(text)]
        filtered = [x for x in clean if x not in STOPWORDS and len(x) > 2]
        return {"tokenizer": "regex fallback; NLTK unavailable", "raw": clean,
                "filtered": filtered, "stems": filtered,
                "freq": Counter(filtered).most_common(TOP_K)}


def count_ngrams(text):
    vec = CountVectorizer(stop_words=list(STOPWORDS), ngram_range=(1, 3),
                          token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z'\-]+\b")
    matrix = vec.fit_transform([text])
    terms = np.array(vec.get_feature_names_out())
    counts = np.asarray(matrix.sum(axis=0)).ravel()
    lengths = np.array([len(x.split()) for x in terms])
    scores = counts * (1 + 0.20 * (lengths - 1))
    order = np.lexsort((-lengths, -scores))[:TOP_K]
    return [(terms[i], round(float(scores[i]), 4), int(counts[i])) for i in order]


def tfidf_terms(docs, target):
    vec = TfidfVectorizer(stop_words=list(STOPWORDS), ngram_range=(1, 3), min_df=1,
                          sublinear_tf=True,
                          token_pattern=r"(?u)\b[a-zA-Z][a-zA-Z'\-]+\b")
    matrix = vec.fit_transform(docs)
    terms = np.array(vec.get_feature_names_out())
    values = matrix[target].toarray().ravel()
    order = values.argsort()[::-1]
    return [(terms[i], round(float(values[i]), 4), phrase_count(docs[target], terms[i]))
            for i in order[:TOP_K] if values[i] > 0]


def rake_like(text):
    raw = [x.lower() for x in TOKEN_RE.findall(text)]
    phrases, current = [], []
    for word in raw:
        if word in STOPWORDS or len(word) <= 2:
            if current:
                phrases.append(current)
                current = []
        else:
            current.append(word)
    if current:
        phrases.append(current)
    freq, degree = Counter(), Counter()
    for phrase in phrases:
        for word in phrase:
            freq[word] += 1
            degree[word] += max(len(phrase) - 1, 0)
    word_score = {w: (degree[w] + freq[w]) / freq[w] for w in freq}
    scores, counts = Counter(), Counter()
    for phrase in phrases:
        for size in (1, 2, 3):
            for i in range(len(phrase) - size + 1):
                item = " ".join(phrase[i:i + size])
                scores[item] += sum(word_score[w] for w in phrase[i:i + size])
                counts[item] += 1
    ranked = sorted(scores, key=lambda x: (scores[x], len(x.split())), reverse=True)
    return [(x, round(float(scores[x]), 4), int(counts[x])) for x in ranked[:TOP_K]]


def domain_entities(text):
    results = []
    for concept, forms in DOMAIN_CONCEPTS.items():
        for form in forms:
            count = phrase_count(text, form)
            if count:
                results.append((normalize_term(form), concept, count))
    return sorted(results, key=lambda x: (x[2], len(x[0])), reverse=True)


count_doc1, count_doc2 = count_ngrams(doc1), count_ngrams(doc2)
tfidf_doc1, tfidf_doc2 = tfidf_terms(documents, 0), tfidf_terms(documents, 1)
rake_doc1, rake_doc2 = rake_like(doc1), rake_like(doc2)
nltk_doc1, nltk_doc2 = nltk_analysis(doc1), nltk_analysis(doc2)
domain_doc1, domain_doc2 = domain_entities(doc1), domain_entities(doc2)

fivefilters_doc1 = online_ranked(ONLINE_TERM_RESULTS["FiveFilters"]["doc1"], doc1)
fivefilters_doc2 = online_ranked(ONLINE_TERM_RESULTS["FiveFilters"]["doc2"], doc2)
wordcount_term_doc1 = online_ranked(ONLINE_TERM_RESULTS["Wordcount.com"]["doc1"], doc1)
wordcount_term_doc2 = online_ranked(ONLINE_TERM_RESULTS["Wordcount.com"]["doc2"], doc2)
dandelion_doc1 = online_ranked(ONLINE_ENTITY_RESULTS["Dandelion"]["doc1"], doc1)
dandelion_doc2 = online_ranked(ONLINE_ENTITY_RESULTS["Dandelion"]["doc2"], doc2)
wordcount_entity_doc1 = online_ranked(ONLINE_ENTITY_RESULTS["Wordcount.com phrase/entity"]["doc1"], doc1)
wordcount_entity_doc2 = online_ranked(ONLINE_ENTITY_RESULTS["Wordcount.com phrase/entity"]["doc2"], doc2)

# =============================================================================
# 4. GROUND TRUTH, NORMALIZATION, AND COMPARISON METRICS
# =============================================================================

ground_truth = pd.DataFrame(MANUAL_GROUND_TRUTH)
ground_truth["doc1_count"] = ground_truth["term"].map(lambda x: phrase_count(doc1, x))
ground_truth["doc2_count"] = ground_truth["term"].map(lambda x: phrase_count(doc2, x))

def result_dict(result):
    return {normalize_term(x[0]): x[1] if len(x) > 1 else "Returned" for x in result}


def exact_metrics(result, truth):
    extracted = set(result_dict(result))
    target = {normalize_term(x) for x in truth}
    hits = len(extracted & target)
    if not extracted or not target:
        return np.nan, np.nan, np.nan, hits
    precision = hits / len(extracted)
    recall = hits / len(target)
    f1 = 2 * precision * recall / (precision + recall) if precision + recall else 0
    return precision, recall, f1, hits


term_truth = ground_truth.loc[ground_truth["kind"].isin(["TERM", "DOMAIN_CONCEPT"]), "term"].tolist()
entity_truth = ground_truth.loc[ground_truth["kind"].eq("ENTITY"), "term"].tolist()

term_methods = {
    "Count n-grams D1": count_doc1,
    "Count n-grams D2": count_doc2,
    "TF-IDF D1": tfidf_doc1,
    "TF-IDF D2": tfidf_doc2,
    "RAKE-like D1": rake_doc1,
    "RAKE-like D2": rake_doc2,
    "NLTK FreqDist D1": [(x, c, c) for x, c in nltk_doc1["freq"]],
    "NLTK FreqDist D2": [(x, c, c) for x, c in nltk_doc2["freq"]],
    "FiveFilters D1": fivefilters_doc1,
    "FiveFilters D2": fivefilters_doc2,
    "Wordcount.com D1": wordcount_term_doc1,
    "Wordcount.com D2": wordcount_term_doc2,
}

entity_methods = {
    "Dandelion D1": dandelion_doc1,
    "Dandelion D2": dandelion_doc2,
    "Wordcount phrase/entity D1": wordcount_entity_doc1,
    "Wordcount phrase/entity D2": wordcount_entity_doc2,
    "Domain dictionary D1": domain_doc1,
    "Domain dictionary D2": domain_doc2,
}


def metrics_frame(methods, truth):
    rows = []
    for method, result in methods.items():
        p, r, f1, hits = exact_metrics(result, truth)
        rows.append({"method": method, "returned": len(result_dict(result)),
                     "ground_truth_hits": hits, "precision": p,
                     "recall": r, "f1": f1})
    return pd.DataFrame(rows)


term_metrics = metrics_frame(term_methods, term_truth)
entity_metrics = metrics_frame(entity_methods, entity_truth)

# Full comparison matrix, limited to ground-truth terms for readability.
master_rows = []
for _, gt in ground_truth.iterrows():
    item = normalize_term(gt["term"])
    row = {
        "manual_term_or_entity": gt["term"],
        "kind": gt["kind"],
        "canonical": gt["canonical"],
        "doc1_count": gt["doc1_count"],
        "doc2_count": gt["doc2_count"],
    }
    for method, result in {**term_methods, **entity_methods}.items():
        row[method] = result_dict(result).get(item, "")
    master_rows.append(row)
master_table = pd.DataFrame(master_rows)

# Exact-equivalence table.
equivalence_rows = []
for group in EXACT_EQUIVALENCE_GROUPS:
    forms = group["forms"]
    equivalence_rows.append({
        "canonical": group["canonical"],
        "equivalent_forms": ", ".join(forms),
        "document_1_forms": ", ".join(f"{x} ({phrase_count(doc1, x)})" for x in forms if phrase_count(doc1, x)) or "-",
        "document_2_forms": ", ".join(f"{x} ({phrase_count(doc2, x)})" for x in forms if phrase_count(doc2, x)) or "-",
    })
equivalence_table = pd.DataFrame(equivalence_rows)

# Broader-concept table.
concept_rows = []
for concept, forms in DOMAIN_CONCEPTS.items():
    one = [(x, phrase_count(doc1, x)) for x in forms if phrase_count(doc1, x)]
    two = [(x, phrase_count(doc2, x)) for x in forms if phrase_count(doc2, x)]
    if one or two:
        concept_rows.append({
            "broader_concept": concept,
            "document_1_related_terms": ", ".join(f"{x} ({c})" for x, c in one) or "-",
            "document_2_related_terms": ", ".join(f"{x} ({c})" for x, c in two) or "-",
        })
concept_table = pd.DataFrame(concept_rows)

preprocessing_table = pd.DataFrame([
    {"document": "Document 1", "method": "NLTK raw/lowercase", "tokens": len(nltk_doc1["raw"]),
     "unique": len(set(nltk_doc1["raw"])), "top_terms": ", ".join(x for x, _ in Counter(nltk_doc1["raw"]).most_common(12))},
    {"document": "Document 1", "method": "Stopword removal", "tokens": len(nltk_doc1["filtered"]),
     "unique": len(set(nltk_doc1["filtered"])), "top_terms": ", ".join(x for x, _ in nltk_doc1["freq"][:12])},
    {"document": "Document 1", "method": "Porter stemming", "tokens": len(nltk_doc1["stems"]),
     "unique": len(set(nltk_doc1["stems"])), "top_terms": ", ".join(x for x, _ in Counter(nltk_doc1["stems"]).most_common(12))},
    {"document": "Document 2", "method": "NLTK raw/lowercase", "tokens": len(nltk_doc2["raw"]),
     "unique": len(set(nltk_doc2["raw"])), "top_terms": ", ".join(x for x, _ in Counter(nltk_doc2["raw"]).most_common(12))},
    {"document": "Document 2", "method": "Stopword removal", "tokens": len(nltk_doc2["filtered"]),
     "unique": len(set(nltk_doc2["filtered"])), "top_terms": ", ".join(x for x, _ in nltk_doc2["freq"][:12])},
    {"document": "Document 2", "method": "Porter stemming", "tokens": len(nltk_doc2["stems"]),
     "unique": len(set(nltk_doc2["stems"])), "top_terms": ", ".join(x for x, _ in Counter(nltk_doc2["stems"]).most_common(12))},
])

# =============================================================================
# 5. PDF HELPERS
# =============================================================================

pdf = PdfPages(OUTPUT_PDF)
plt.rcParams.update({"figure.dpi": 110, "font.size": 9, "axes.grid": False})


def page(title=None, figsize=(8.5, 11)):
    fig = plt.figure(figsize=figsize)
    ax = fig.add_axes([0, 0, 1, 1])
    ax.axis("off")
    if title:
        fig.text(0.5, 0.965, title, ha="center", va="top", fontsize=14, fontweight="bold")
    return fig


def save(fig):
    pdf.savefig(fig, bbox_inches="tight")
    plt.close(fig)


def text_pages(title, lines, font_size=8.4, lines_per_page=56, width=106):
    wrapped = []
    for line in lines:
        wrapped.extend([""] if not line else textwrap.wrap(str(line), width=width,
                                                            break_long_words=False) or [""])
    chunks = [wrapped[i:i + lines_per_page] for i in range(0, len(wrapped), lines_per_page)] or [[""]]
    for i, chunk in enumerate(chunks):
        fig = page(title if i == 0 else title + " (continued)")
        fig.text(0.055, 0.925, "\n".join(chunk), va="top", ha="left",
                 fontsize=font_size, family="monospace", linespacing=1.2)
        save(fig)


def table_pages(title, frame, rows_per_page=20, font_size=6.8, col_width=24):
    if frame is None or frame.empty:
        text_pages(title, ["No results were returned."], 9)
        return
    printable = frame.copy()
    for col in printable.columns:
        printable[col] = printable[col].map(lambda x: "" if pd.isna(x) else str(x))
        printable[col] = printable[col].map(lambda x: "\n".join(textwrap.wrap(x, col_width)) if x else "")
    for start in range(0, len(printable), rows_per_page):
        chunk = printable.iloc[start:start + rows_per_page]
        fig = page(title if start == 0 else title + " (continued)")
        ax = fig.add_axes([0.025, 0.055, 0.95, 0.865])
        ax.axis("off")
        table = ax.table(cellText=chunk.values, colLabels=chunk.columns,
                         cellLoc="left", colLoc="left", loc="upper left")
        table.auto_set_font_size(False)
        table.set_fontsize(font_size)
        table.scale(1, 1.48)
        for (r, c), cell in table.get_celld().items():
            cell.set_edgecolor("#AAAAAA")
            if r == 0:
                cell.set_facecolor("#D9EAF7")
                cell.set_text_props(weight="bold")
        save(fig)


def full_narrative_page(number, row, narrative):
    fields = ["ACCIDENT_DT", "CLASSIFICATION", "ACCIDENT_TYPE", "ACTIVITY",
              "INJURY_SOURCE", "NATURE_INJURY", "INJ_BODY_PART", "DEGREE_INJURY",
              "OCCUPATION", "COAL_METAL_IND"]
    lines = [f"DOCUMENT {number}", "-" * 80, f"Record identifier: {record_id(row)}",
             f"Shared topic: {topic_column} = {topic_value}"]
    for field in fields:
        if field in row.index and pd.notna(row[field]):
            lines.append(f"{field}: {row[field]}")
    lines.extend(["", "FULL NARRATIVE TEXT", "-" * 80])
    lines.extend(textwrap.wrap(narrative, width=100, break_long_words=False))
    lines.extend(["", "-" * 80, f"Character count: {len(narrative):,}",
                  f"Word count: {len(narrative.split()):,}",
                  "Note: this is the complete stored database value; a 384-character record may still be source-truncated."])
    text_pages(f"Selected Narrative {number}: Complete Stored Text", lines, 8.4, 50, 105)


def extraction_frame(values, method):
    return pd.DataFrame([
        {"rank": i, "method": method, "term_or_entity": x[0],
         "engine_output": x[1] if len(x) > 1 else "Returned",
         "literal_document_count": x[2] if len(x) > 2 else ""}
        for i, x in enumerate(values, 1)
    ])


def formatted_metrics(frame):
    result = frame.copy()
    for col in ["precision", "recall", "f1"]:
        result[col] = result[col].map(lambda x: "N/A" if pd.isna(x) else f"{x:.3f}")
    return result

# =============================================================================
# 6. BUILD FINAL ORGANIZED REPORT
# =============================================================================

# Cover
fig = page()
fig.text(0.5, 0.76, "MSHA Narrative Extraction Comparison", ha="center",
         fontsize=22, fontweight="bold")
fig.text(0.5, 0.70, "Final Manual, Online, Programmatic, and Entity Analysis",
         ha="center", fontsize=13, color="#444444")
fig.text(0.5, 0.65, datetime.now().strftime("Generated %B %d, %Y at %H:%M"),
         ha="center", fontsize=10, color="#666666")
fig.text(0.10, 0.55, "\n".join([
    f"Shared topic: {topic_column} = {topic_value}",
    f"Document 1: {record_id(row1)}",
    f"Document 2: {record_id(row2)}",
    "",
    "Purpose:",
    "Compare manually identified terms and entities with online and Python extraction methods.",
    "Evaluate useful-term coverage, noise, phrase preservation, preprocessing effects,",
    "entity extraction, and normalization of equivalent surface forms.",
    "",
    "Important scoring note:",
    "The online tools returned ordered lists without numerical scores or counts.",
    "Their output is recorded as rank; literal occurrence counts are calculated separately.",
]), va="top", fontsize=9.5, family="monospace")
save(fig)

# Executive summary
text_pages("Executive Summary", [
    "Two coal-mine narratives classified as HANDLING OF MATERIALS were compared. Both involve",
    "roof-bolting materials, retaining bands or straps, unexpected movement of metal, and a",
    "struck-by injury mechanism. The narratives use different wording and describe different",
    "immediate objects: a strap in Document 1 and a roof bolt in Document 2.",
    "",
    "The manual ground truth emphasizes complete safety concepts rather than every frequent word.",
    "Online lists are evaluated using rank and exact ground-truth coverage because the services",
    "did not provide comparable scores, counts, or reliable entity types.",
    "",
    "FiveFilters returned a mix of useful phrases and generic fragments. Wordcount.com preserved",
    "longer event and equipment phrases and therefore provides more context. Dandelion returned",
    "a small set of short object labels. This demonstrates that a general entity extractor may",
    "have limited coverage in short technical incident narratives, where the most useful entities",
    "are equipment, materials, body parts, hazards, work activities, and event mechanisms.",
    "",
    "The report includes exact-match precision, recall, and F1, but those metrics are conservative:",
    "semantically similar phrases receive no exact-match credit. The equivalence and broader-concept",
    "tables are therefore needed to interpret related forms such as bolts/roof bolts and",
    "band/strap/bundle straps."
], 8.8, 52)

# Narratives
full_narrative_page(1, row1, doc1)
full_narrative_page(2, row2, doc2)

# Step 1
table_pages("Step 1 - Estimated Manual Ground Truth", ground_truth,
            rows_per_page=20, font_size=6.6, col_width=24)
text_pages("Step 1 - Ground-Truth Rationale", [
    "The manual ground truth is a reasoned estimate based on close reading of both narratives.",
    "It includes equipment, materials, material restraints, work activities, hazards, event",
    "triggers, injury mechanisms, person references, and body parts. Generic connecting words",
    "were excluded even when an automated extractor returned them.",
    "",
    "The document counts are literal whole-phrase counts. A zero does not necessarily mean the",
    "concept is absent; the narrative may express it with different wording. Exact phrase counts",
    "are intentionally kept separate from online rank and Python relevance scores."
], 8.8)

# Consolidated comparison, ground truth only for readable final submission.
base = ["manual_term_or_entity", "kind", "canonical", "doc1_count", "doc2_count"]
method_columns = [c for c in master_table.columns if c not in base]
for start in range(0, len(method_columns), 4):
    cols = base + method_columns[start:start + 4]
    table_pages("Consolidated Ground Truth and Engine Comparison",
                master_table[cols], rows_per_page=18, font_size=5.8, col_width=19)

# Step 2 online and Python term extraction
for title, values, method in [
    ("Step 2 - FiveFilters - Document 1", fivefilters_doc1, "FiveFilters"),
    ("Step 2 - FiveFilters - Document 2", fivefilters_doc2, "FiveFilters"),
    ("Step 2 - Wordcount.com - Document 1", wordcount_term_doc1, "Wordcount.com"),
    ("Step 2 - Wordcount.com - Document 2", wordcount_term_doc2, "Wordcount.com"),
    ("Python Count n-grams - Document 1", count_doc1, "Count n-grams"),
    ("Python Count n-grams - Document 2", count_doc2, "Count n-grams"),
    ("Python TF-IDF - Document 1", tfidf_doc1, "TF-IDF"),
    ("Python TF-IDF - Document 2", tfidf_doc2, "TF-IDF"),
]:
    table_pages(title, extraction_frame(values, method), 22, 7.0, 26)
table_pages("Term Extraction Metrics against Manual Ground Truth",
            formatted_metrics(term_metrics), 20, 7.0, 25)

# Step 3/4 NLTK and RAKE
text_pages("Step 3 - NLTK and Programmatic Methods", [
    f"Document 1 tokenizer: {nltk_doc1['tokenizer']}",
    f"Document 2 tokenizer: {nltk_doc2['tokenizer']}",
    "NLTK FreqDist provides directly interpretable token counts. RAKE-like extraction ranks",
    "candidate phrases using word frequency and co-occurrence degree. TF-IDF emphasizes terms",
    "that distinguish one document from the other. Their numerical outputs are not the same scale."
], 8.8)
for title, values in [
    ("Step 3 - RAKE-like Phrases - Document 1", rake_doc1),
    ("Step 3 - RAKE-like Phrases - Document 2", rake_doc2),
]:
    table_pages(title, extraction_frame(values, "RAKE-like"), 22, 7.0, 26)
table_pages("Step 4 - Preprocessing Comparison", preprocessing_table,
            12, 6.8, 35)

# Steps 5 and 7
table_pages("Step 5 - Related Terms under Broader Domain Concepts", concept_table,
            18, 7.0, 35)
table_pages("Step 7 - Exact or Contextual Equivalence", equivalence_table,
            18, 7.0, 35)
text_pages("Steps 5 and 7 - Interpretation", [
    "Exact or contextual equivalence is reported separately from broader domain relatedness.",
    "For example, band, bands, strap, and bundle straps refer to the same restraint concept in",
    "these narratives. Roof bolt and roof bolter are related to roof-bolting work but are not the",
    "same physical entity. Preserving that distinction is important for safety interpretation.",
    "",
    "EE and IE are treated as person-reference abbreviations based on their contextual use.",
    "If the course expects stricter synonymy, these abbreviation groups can be removed or noted",
    "as context-dependent normalization rather than universal equivalence."
], 8.8)

# Step 6 entities
for title, values, method in [
    ("Step 6 - Dandelion - Document 1", dandelion_doc1, "Dandelion"),
    ("Step 6 - Dandelion - Document 2", dandelion_doc2, "Dandelion"),
    ("Step 6 - Wordcount Phrase/Entity - Document 1", wordcount_entity_doc1, "Wordcount.com"),
    ("Step 6 - Wordcount Phrase/Entity - Document 2", wordcount_entity_doc2, "Wordcount.com"),
    ("Step 6 - Domain Dictionary - Document 1", domain_doc1, "Domain dictionary"),
    ("Step 6 - Domain Dictionary - Document 2", domain_doc2, "Domain dictionary"),
]:
    table_pages(title, extraction_frame(values, method), 22, 7.0, 26)
table_pages("Entity Extraction Metrics against Manual Entity Ground Truth",
            formatted_metrics(entity_metrics), 18, 7.0, 25)

# Final findings generated from metrics.
def best_row(frame):
    usable = frame.dropna(subset=["f1"]).sort_values(["f1", "recall", "precision"], ascending=False)
    return usable.iloc[0] if not usable.empty else None

best_term = best_row(term_metrics)
best_entity = best_row(entity_metrics)
findings = [
    "1. The two documents share a material-handling topic but use different surface wording.",
    "   Phrase-level comparison is therefore more meaningful than single-word overlap alone.",
    "",
    "2. FiveFilters returned useful material-handling terms but also generic fragments such as",
    "   AS, TOP, and TEND. Those outputs increase coverage but reduce interpretability.",
    "",
    "3. Wordcount.com preserved longer equipment, activity, hazard, and consequence phrases.",
    "   Its results are generally easier to connect to the event sequence, although several",
    "   outputs are full clauses rather than conventional named entities.",
    "",
    "4. Dandelion returned only short labels. That limited output is itself informative because",
    "   conventional named-entity tools are not optimized for short mining incident narratives.",
    "",
    "5. Stopword removal reduces noise. Porter stemming merges some morphological variants but",
    "   can reduce readability, so canonical concept mapping is more interpretable for this domain.",
    "",
    "6. Online ranks, literal counts, TF-IDF weights, and RAKE-like scores are different measures.",
    "   They should be compared by coverage and usefulness, not as directly equivalent numbers.",
]
if best_term is not None:
    findings.append(
        f"7. Highest exact-match term F1 in this run: {best_term['method']} "
        f"(precision={best_term['precision']:.3f}, recall={best_term['recall']:.3f}, F1={best_term['f1']:.3f})."
    )
if best_entity is not None:
    findings.append(
        f"8. Highest exact-match entity F1 in this run: {best_entity['method']} "
        f"(precision={best_entity['precision']:.3f}, recall={best_entity['recall']:.3f}, F1={best_entity['f1']:.3f})."
    )
findings.extend([
    "",
    "Exact-match metrics are conservative. For example, 'bolts' and 'roof bolts' may refer to",
    "the same material in context but will not match unless normalization is applied. The concept",
    "tables should therefore be read alongside precision, recall, and F1.",
])
text_pages("Final Findings and Discussion", findings, 8.8, 54)

metadata = pdf.infodict()
metadata["Title"] = "MSHA Narrative Extraction Final Report"
metadata["Author"] = "Python NLP comparison"
metadata["Subject"] = "Manual, online, Python term, phrase, preprocessing, and entity extraction"
metadata["CreationDate"] = datetime.now()
pdf.close()

print("=" * 80)
print("Final MSHA narrative extraction report completed.")
print(f"Document 1: {record_id(row1)}")
print(f"Document 2: {record_id(row2)}")
print(f"PDF output: {os.path.abspath(OUTPUT_PDF)}")
print("Online lists are represented by rank; no scores or types were invented.")
print("The complete stored narrative text for both records is included.")
print("=" * 80)
````

````python
# =============================================================================
# MSHA LOCAL HUGGING FACE LLM PILOT
# CAUSE, CONTRIBUTING FACTOR, AND RECOMMENDED ACTION CLASSIFICATION
#
# Run this entire block in one Azure Machine Learning Studio notebook cell.
#
# Prerequisite:
#   A pandas DataFrame named `accidents` is already loaded.
#
# This block:
#   1. Installs required Python packages.
#   2. Verifies that an NVIDIA GPU is available.
#   3. Downloads Qwen2.5-7B-Instruct from Hugging Face.
#   4. Loads the model in 4-bit mode.
#   5. Runs two built-in MSHA-style self-tests.
#   6. Selects a small stratified sample from `accidents`.
#   7. Generates one cause, factor, and action category per narrative.
#   8. Validates model output with Pydantic.
#   9. Checks that evidence excerpts occur in the narrative.
#  10. Saves CSV, Excel, JSONL, checkpoint, and review files.
#
# Important:
#   The labels represent reported event mechanisms and contributing factors.
#   They are not verified root-cause findings or final corrective actions.
# =============================================================================


# =============================================================================
# 1. INSTALL REQUIRED PACKAGES
# =============================================================================

import sys
import subprocess
import importlib
import importlib.util


REQUIRED_PACKAGES = [
    "transformers>=4.47.0",
    "accelerate>=1.0.0",
    "bitsandbytes>=0.43.0",
    "sentencepiece>=0.2.0",
    "pydantic>=2.0.0",
    "openpyxl>=3.1.0",
    "huggingface_hub>=0.27.0",
    "safetensors>=0.4.0",
]


print("=" * 90)
print("INSTALLING OR UPDATING REQUIRED PYTHON PACKAGES")
print("=" * 90)


subprocess.check_call(
    [
        sys.executable,
        "-m",
        "pip",
        "install",
        "--quiet",
        "--upgrade",
        *REQUIRED_PACKAGES,
    ]
)


importlib.invalidate_caches()


# =============================================================================
# 2. IMPORT LIBRARIES
# =============================================================================

import os
import re
import gc
import json
import time
import random
import warnings
from pathlib import Path
from datetime import datetime
from typing import Literal

import numpy as np
import pandas as pd
import torch

from pydantic import BaseModel, Field, ValidationError

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

warnings.filterwarnings("ignore")


# Display version information.
import transformers
import accelerate
import bitsandbytes


print("\nPACKAGE VERSIONS")
print("-" * 90)
print("Python:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)


# =============================================================================
# 3. USER SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

TEXT_COLUMN = "NARRATIVE"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

# Begin with a small test.
SAMPLE_SIZE = 20

# The sample will attempt to include diverse records across available fields.
STRATIFICATION_COLUMNS = [
    "CLASSIFICATION",
    "DEGREE_INJURY",
    "COAL_METAL_IND",
]

RANDOM_STATE = 42

PROMPT_VERSION = "MSHA_CAUSE_FACTOR_ACTION_HF_V1.0"

MAX_NEW_TOKENS = 500

MAX_RETRIES = 3

SOURCE_FIELD_LIMIT = 384

TRUNCATION_THRESHOLD = 383

LOW_CONFIDENCE_THRESHOLD = 0.70

# Set True only when intentionally rerunning records already in the checkpoint.
REPROCESS_COMPLETED = False

# Set True to run the two known roof-bolt narratives as self-tests.
RUN_BUILT_IN_SELF_TEST = True

# Output files.
OUTPUT_CSV = "MSHA_HF_Pilot_Classifications.csv"

OUTPUT_XLSX = "MSHA_HF_Pilot_Classifications.xlsx"

OUTPUT_JSONL = "MSHA_HF_Pilot_Classifications.jsonl"

CHECKPOINT_JSONL = "MSHA_HF_Pilot_Checkpoint.jsonl"

REVIEW_XLSX = "MSHA_HF_Pilot_Human_Review.xlsx"

SUMMARY_XLSX = "MSHA_HF_Pilot_Category_Summary.xlsx"


# =============================================================================
# 4. CONTROLLED TAXONOMIES
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_FALL",
    "STRUCK_BY_MOVING_OBJECT",
    "CAUGHT_IN_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "GROUND_OR_ROOF_FAILURE",
    "ELECTRICAL_CONTACT",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "CHEMICAL_OR_DUST_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


CONTRIBUTING_FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NOT_STATED",
]


ACTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 5. STRICT PYDANTIC RESPONSE SCHEMA
# =============================================================================

CauseLiteral = Literal[
    "SLIP_TRIP_FALL",
    "STRUCK_BY_MOVING_OBJECT",
    "CAUGHT_IN_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "GROUND_OR_ROOF_FAILURE",
    "ELECTRICAL_CONTACT",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "CHEMICAL_OR_DUST_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


FactorLiteral = Literal[
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NOT_STATED",
]


ActionLiteral = Literal[
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


class IncidentClassification(BaseModel):
    """
    Required structured classification for one MSHA narrative.
    """

    primary_cause_category: CauseLiteral

    primary_contributing_factor: FactorLiteral

    recommended_action_category: ActionLiteral

    cause_evidence_text: str = Field(
        description=(
            "A short verbatim excerpt from the narrative that supports the "
            "cause category. Use an empty string if no excerpt exists."
        )
    )

    contributing_factor_evidence_text: str = Field(
        description=(
            "A short verbatim excerpt from the narrative that supports the "
            "factor category. Use an empty string if no excerpt exists."
        )
    )

    action_rationale: str = Field(
        description=(
            "One short sentence explaining why the action category follows "
            "from the narrative. The recommendation requires human review."
        )
    )

    cause_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    action_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    insufficient_information_reason: str = ""

    reviewer_note: str = ""


# =============================================================================
# 6. VERIFY AZURE GPU
# =============================================================================

print("\n" + "=" * 90)
print("GPU CHECK")
print("=" * 90)


if not torch.cuda.is_available():

    raise RuntimeError(
        "PyTorch cannot access an NVIDIA GPU. "
        "Attach this notebook to an Azure ML GPU compute instance and rerun."
    )


gpu_name = torch.cuda.get_device_name(0)

gpu_properties = torch.cuda.get_device_properties(0)

gpu_total_memory_gb = (
    gpu_properties.total_memory
    / 1024**3
)

gpu_free_bytes, gpu_total_bytes = torch.cuda.mem_get_info(0)

gpu_free_memory_gb = (
    gpu_free_bytes
    / 1024**3
)


print("GPU:", gpu_name)

print(
    f"Total GPU memory: "
    f"{gpu_total_memory_gb:.2f} GB"
)

print(
    f"Currently free GPU memory: "
    f"{gpu_free_memory_gb:.2f} GB"
)


# =============================================================================
# 7. VALIDATE AND PREPARE THE ACCIDENT DATAFRAME
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The 'accidents' DataFrame is empty."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"Column {TEXT_COLUMN!r} was not found. "
        f"Available columns: {list(df.columns)}"
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = (
        df.index.astype(str)
    )


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df = df[
    df[TEXT_COLUMN].str.len() > 0
].copy()


df["_NARRATIVE_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN].str.len()
)


df["_NARRATIVE_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_NARRATIVE_MAY_BE_TRUNCATED"] = (
    df["_NARRATIVE_CHARACTER_COUNT"]
    >= TRUNCATION_THRESHOLD
)


print("\nINPUT DATA")

print("-" * 90)

print(
    f"Usable narratives: "
    f"{len(df):,}"
)

print(
    f"Identifier column: "
    f"{id_column}"
)

print(
    "Narratives flagged near field limit: "
    f"{int(df['_NARRATIVE_MAY_BE_TRUNCATED'].sum()):,}"
)


# =============================================================================
# 8. SELECT A SMALL DIVERSE PILOT SAMPLE
# =============================================================================

def select_pilot_sample(
    data,
    sample_size,
    stratification_columns,
    random_state,
):
    """
    Select one record from as many available strata as possible, then fill
    the remainder with a reproducible random sample.
    """

    sample_size = min(
        int(sample_size),
        len(data),
    )

    available_columns = [
        column
        for column in stratification_columns
        if column in data.columns
    ]

    if sample_size >= len(data):

        return (
            data.copy()
            .reset_index(drop=False)
            .rename(
                columns={
                    "index": "_ORIGINAL_INDEX"
                }
            )
        )

    if not available_columns:

        return (
            data.sample(
                n=sample_size,
                random_state=random_state,
            )
            .reset_index(drop=False)
            .rename(
                columns={
                    "index": "_ORIGINAL_INDEX"
                }
            )
        )

    working = data.copy()

    for column in available_columns:

        working[column] = (
            working[column]
            .astype("string")
            .fillna("<MISSING>")
        )

    grouped = working.groupby(
        available_columns,
        dropna=False,
        sort=False,
    )

    group_keys = list(
        grouped.groups.keys()
    )

    random_generator = random.Random(
        random_state
    )

    random_generator.shuffle(
        group_keys
    )

    selected_indexes = []

    for group_number, key in enumerate(
        group_keys
    ):

        if len(selected_indexes) >= sample_size:
            break

        group = grouped.get_group(
            key
        )

        selected_index = (
            group.sample(
                n=1,
                random_state=(
                    random_state
                    + group_number
                ),
            )
            .index[0]
        )

        selected_indexes.append(
            selected_index
        )

    remaining_count = (
        sample_size
        - len(selected_indexes)
    )

    if remaining_count > 0:

        remaining = data.drop(
            index=selected_indexes,
            errors="ignore",
        )

        extra_indexes = (
            remaining.sample(
                n=min(
                    remaining_count,
                    len(remaining),
                ),
                random_state=random_state,
            )
            .index
            .tolist()
        )

        selected_indexes.extend(
            extra_indexes
        )

    return (
        data.loc[selected_indexes]
        .copy()
        .reset_index(drop=False)
        .rename(
            columns={
                "index": "_ORIGINAL_INDEX"
            }
        )
    )


pilot_df = select_pilot_sample(
    data=df,
    sample_size=SAMPLE_SIZE,
    stratification_columns=STRATIFICATION_COLUMNS,
    random_state=RANDOM_STATE,
)


print(
    f"\nPilot narratives selected: "
    f"{len(pilot_df):,}"
)


# =============================================================================
# 9. DOWNLOAD AND LOAD QWEN2.5-7B-INSTRUCT
# =============================================================================

print("\n" + "=" * 90)
print("DOWNLOADING AND LOADING THE HUGGING FACE MODEL")
print("=" * 90)

print(
    f"Model: {MODEL_ID}"
)

print(
    "The first run downloads model files to the Azure compute instance."
)


quantization_configuration = (
    BitsAndBytesConfig(
        load_in_4bit=True,

        bnb_4bit_quant_type="nf4",

        bnb_4bit_use_double_quant=True,

        bnb_4bit_compute_dtype=torch.float16,
    )
)


tokenizer = (
    AutoTokenizer.from_pretrained(
        MODEL_ID,

        use_fast=True,

        trust_remote_code=False,
    )
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = (
        tokenizer.eos_token
    )


model = (
    AutoModelForCausalLM
    .from_pretrained(
        MODEL_ID,

        quantization_config=(
            quantization_configuration
        ),

        device_map="auto",

        torch_dtype=torch.float16,

        low_cpu_mem_usage=True,

        trust_remote_code=False,
    )
)


model.eval()


print(
    "Model loaded successfully."
)


print(
    "Model input device:",
    model.device,
)


# =============================================================================
# 10. DEFINE THE CLASSIFICATION PROMPT
# =============================================================================

SYSTEM_PROMPT = f"""
You are performing structured information extraction from short MSHA mine
accident narratives.

This is reported-event classification, not definitive root-cause analysis.

Use only information stated in the narrative.

Do not infer:
- missing training;
- missing PPE;
- a failed procedure;
- an absent guard;
- an absent lockout;
- a failed inspection;
- worker negligence;
- management failure;
- an organizational root cause.

Select exactly one category from each list.

PRIMARY CAUSE CATEGORY:
{json.dumps(CAUSE_CATEGORIES, indent=2)}

PRIMARY CONTRIBUTING FACTOR:
{json.dumps(CONTRIBUTING_FACTOR_CATEGORIES, indent=2)}

RECOMMENDED ACTION CATEGORY:
{json.dumps(ACTION_CATEGORIES, indent=2)}

Definitions:

primary_cause_category:
The main reported physical event, movement, contact, release, or
energy-transfer mechanism that produced the incident.

primary_contributing_factor:
The most important condition, body position, work method, equipment state,
material state, or environmental condition stated in the narrative.

recommended_action_category:
The strongest preliminary control category directly supported by the
narrative. This is decision support and requires review by qualified safety
and technical personnel.

Rules:

1. Select exactly one value from each category list.

2. Copy cause_evidence_text and contributing_factor_evidence_text verbatim
   from the narrative.

3. Do not rewrite or paraphrase evidence.

4. Keep each evidence excerpt short.

5. If there is no stated contributing factor, select NOT_STATED.

6. If the event mechanism cannot be determined, select
   INSUFFICIENT_INFORMATION.

7. If no corrective-action category is supportable, select
   INVESTIGATION_REQUIRED or INSUFFICIENT_INFORMATION.

8. Do not treat the final injury diagnosis as the causal mechanism.

9. Prefer physical, engineering, material, or energy controls when directly
   supported by the narrative.

10. Do not select training merely because an incident occurred.

11. Confidence values are model self-assessments from 0 to 1. They are not
    calibrated probabilities.

12. reviewer_note should identify ambiguity, competing interpretations, or
    likely narrative truncation.

Return only one valid JSON object.

The JSON object must contain exactly these keys:

{{
  "primary_cause_category": "CATEGORY",
  "primary_contributing_factor": "CATEGORY",
  "recommended_action_category": "CATEGORY",
  "cause_evidence_text": "verbatim narrative excerpt",
  "contributing_factor_evidence_text": "verbatim narrative excerpt or empty string",
  "action_rationale": "one short sentence",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "action_confidence": 0.0,
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


def build_user_prompt(
    narrative,
    narrative_may_be_truncated,
):
    """
    Build the user prompt for one accident narrative.
    """

    truncation_note = (
        "The stored narrative may be truncated near the source-field limit."
        if narrative_may_be_truncated
        else
        "The narrative is not flagged as being near the source-field limit."
    )

    return f"""
Classify this MSHA accident narrative.

DATA QUALITY NOTE:
{truncation_note}

NARRATIVE:
{narrative}

Return only the required JSON object.
""".strip()


# =============================================================================
# 11. JSON EXTRACTION AND RESPONSE VALIDATION
# =============================================================================

def extract_first_json_object(
    generated_text,
):
    """
    Extract the first balanced JSON object from model output.

    This avoids greedy regular expressions and tolerates occasional surrounding
    model text.
    """

    text = str(
        generated_text
    ).strip()

    start_index = text.find("{")

    if start_index < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0

    in_string = False

    escape_next = False

    for position in range(
        start_index,
        len(text),
    ):

        character = text[position]

        if escape_next:

            escape_next = False

            continue

        if character == "\\":

            if in_string:
                escape_next = True

            continue

        if character == '"':

            in_string = not in_string

            continue

        if in_string:
            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start_index:
                    position + 1
                ]

    raise ValueError(
        "A complete balanced JSON object was not found."
    )


def normalized_text(
    value,
):
    """
    Normalize case and whitespace for evidence checking.
    """

    return re.sub(
        r"\s+",
        " ",
        str(value),
    ).strip().lower()


def evidence_is_present(
    evidence,
    narrative,
):
    """
    Check whether an evidence excerpt is literally contained in the narrative.
    """

    evidence_normalized = normalized_text(
        evidence
    )

    narrative_normalized = normalized_text(
        narrative
    )

    if not evidence_normalized:
        return True

    return (
        evidence_normalized
        in narrative_normalized
    )


def determine_review_required(
    result,
    narrative_may_be_truncated,
    cause_evidence_valid,
    factor_evidence_valid,
):
    """
    Apply deterministic review rules independently of the model.
    """

    minimum_confidence = min(
        result.cause_confidence,
        result.factor_confidence,
        result.action_confidence,
    )

    low_confidence = (
        minimum_confidence
        < LOW_CONFIDENCE_THRESHOLD
    )

    incomplete_classification = (
        result.primary_cause_category
        == "INSUFFICIENT_INFORMATION"
        or result.primary_contributing_factor
        == "NOT_STATED"
        or result.recommended_action_category
        == "INSUFFICIENT_INFORMATION"
    )

    investigation_flag = (
        result.recommended_action_category
        in {
            "ENGINEERING_REVIEW_REQUIRED",
            "INVESTIGATION_REQUIRED",
        }
    )

    reviewer_note_present = bool(
        result.reviewer_note.strip()
    )

    return bool(
        narrative_may_be_truncated
        or low_confidence
        or incomplete_classification
        or investigation_flag
        or not cause_evidence_valid
        or not factor_evidence_valid
        or reviewer_note_present
    )


# =============================================================================
# 12. LOW-LEVEL MODEL GENERATION
# =============================================================================

def generate_model_response(
    messages,
    max_new_tokens=MAX_NEW_TOKENS,
):
    """
    Run one deterministic generation with the downloaded local model.
    """

    model_inputs = (
        tokenizer.apply_chat_template(
            messages,

            add_generation_prompt=True,

            tokenize=True,

            return_dict=True,

            return_tensors="pt",
        )
    )

    model_inputs = {
        key: value.to(
            model.device
        )
        for key, value
        in model_inputs.items()
    }

    input_token_count = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        output_ids = model.generate(
            **model_inputs,

            max_new_tokens=(
                max_new_tokens
            ),

            do_sample=False,

            repetition_penalty=1.03,

            pad_token_id=(
                tokenizer.pad_token_id
            ),

            eos_token_id=(
                tokenizer.eos_token_id
            ),

            use_cache=True,
        )

    generated_token_ids = output_ids[
        0,
        input_token_count:
    ]

    generated_text = tokenizer.decode(
        generated_token_ids,

        skip_special_tokens=True,
    )

    del model_inputs

    del output_ids

    del generated_token_ids

    return generated_text


# =============================================================================
# 13. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    narrative_may_be_truncated,
):
    """
    Classify one narrative with retries and Pydantic validation.
    """

    user_prompt = build_user_prompt(
        narrative=narrative,
        narrative_may_be_truncated=(
            narrative_may_be_truncated
        ),
    )

    base_messages = [
        {
            "role": "system",
            "content": SYSTEM_PROMPT,
        },
        {
            "role": "user",
            "content": user_prompt,
        },
    ]

    last_error = ""

    last_raw_response = ""

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        try:

            if attempt_number == 1:

                current_messages = (
                    base_messages
                )

            else:

                current_messages = (
                    base_messages
                    + [
                        {
                            "role": "user",
                            "content": (
                                "The previous output could not be validated. "
                                "Return only one valid JSON object using exactly "
                                "the required keys and permitted categories. "
                                "Do not add Markdown or commentary."
                            ),
                        }
                    ]
                )

            raw_response = (
                generate_model_response(
                    messages=(
                        current_messages
                    )
                )
            )

            last_raw_response = (
                raw_response
            )

            json_text = (
                extract_first_json_object(
                    raw_response
                )
            )

            parsed_result = (
                IncidentClassification
                .model_validate_json(
                    json_text
                )
            )

            cause_evidence_valid = (
                evidence_is_present(
                    parsed_result
                    .cause_evidence_text,

                    narrative,
                )
            )

            factor_evidence_valid = (
                evidence_is_present(
                    parsed_result
                    .contributing_factor_evidence_text,

                    narrative,
                )
            )

            review_required = (
                determine_review_required(
                    result=parsed_result,

                    narrative_may_be_truncated=(
                        narrative_may_be_truncated
                    ),

                    cause_evidence_valid=(
                        cause_evidence_valid
                    ),

                    factor_evidence_valid=(
                        factor_evidence_valid
                    ),
                )
            )

            return {
                **parsed_result.model_dump(),

                "CAUSE_EVIDENCE_VALID":
                    cause_evidence_valid,

                "FACTOR_EVIDENCE_VALID":
                    factor_evidence_valid,

                "REVIEW_REQUIRED":
                    review_required,

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            last_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} "
                f"failed: "
                f"{last_error[:300]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        "primary_cause_category":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NOT_STATED",

        "recommended_action_category":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "contributing_factor_evidence_text":
            "",

        "action_rationale":
            (
                "The local model output could not be "
                "validated and requires human review."
            ),

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "action_confidence":
            0.0,

        "insufficient_information_reason":
            (
                "Model generation or response "
                "validation failed."
            ),

        "reviewer_note":
            "",

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_VALID":
            False,

        "REVIEW_REQUIRED":
            True,

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            last_error,

        "RAW_MODEL_RESPONSE":
            last_raw_response,
    }


# =============================================================================
# 14. BUILT-IN SELF-TESTS
# =============================================================================

SELF_TEST_NARRATIVES = [
    {
        "SOURCE_RECORD_ID":
            "SELF_TEST_ROOF_BOLT_STRAP",

        "NARRATIVE":
            (
                "The employee was removing roof bolts from the bolter by hand. "
                "A loose bundle strap was caught while the bolts were being "
                "pulled, and the end of the strap struck the employee."
            ),

        "MAY_BE_TRUNCATED":
            False,
    },
    {
        "SOURCE_RECORD_ID":
            "SELF_TEST_RELEASED_ROOF_BOLT",

        "NARRATIVE":
            (
                "The employee was preparing to load roof bolts. The edge of a "
                "scoop bucket was resting on one roof bolt. When the retaining "
                "bands were cut, the roof bolt moved upward and struck the "
                "employee."
            ),

        "MAY_BE_TRUNCATED":
            False,
    },
]


if RUN_BUILT_IN_SELF_TEST:

    print("\n" + "=" * 90)
    print("RUNNING TWO BUILT-IN MODEL SELF-TESTS")
    print("=" * 90)

    self_test_records = []

    for self_test in SELF_TEST_NARRATIVES:

        print(
            "\nClassifying:",
            self_test[
                "SOURCE_RECORD_ID"
            ],
        )

        self_test_result = (
            classify_narrative(
                narrative=(
                    self_test[
                        "NARRATIVE"
                    ]
                ),

                narrative_may_be_truncated=(
                    self_test[
                        "MAY_BE_TRUNCATED"
                    ]
                ),
            )
        )

        self_test_records.append(
            {
                "SOURCE_RECORD_ID":
                    self_test[
                        "SOURCE_RECORD_ID"
                    ],

                "NARRATIVE":
                    self_test[
                        "NARRATIVE"
                    ],

                **self_test_result,
            }
        )

    self_test_df = pd.DataFrame(
        self_test_records
    )

    print("\nSELF-TEST RESULTS")

    display(
        self_test_df[
            [
                "SOURCE_RECORD_ID",
                "primary_cause_category",
                "primary_contributing_factor",
                "recommended_action_category",
                "cause_evidence_text",
                "contributing_factor_evidence_text",
                "CAUSE_EVIDENCE_VALID",
                "FACTOR_EVIDENCE_VALID",
                "REVIEW_REQUIRED",
                "PROCESSING_STATUS",
            ]
        ]
    )

    if not (
        self_test_df[
            "PROCESSING_STATUS"
        ]
        .eq("SUCCESS")
        .all()
    ):

        raise RuntimeError(
            "The model failed one or more built-in self-tests. "
            "Review the displayed errors before classifying the pilot sample."
        )


# =============================================================================
# 15. CHECKPOINT UTILITIES
# =============================================================================

def load_checkpoint(
    checkpoint_path,
):
    """
    Load completed JSONL records keyed by source record identifier.
    """

    checkpoint_file = Path(
        checkpoint_path
    )

    if not checkpoint_file.exists():

        return {}

    records = {}

    with checkpoint_file.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:
                continue

            try:

                record = json.loads(
                    line
                )

                source_record_id = str(
                    record[
                        "SOURCE_RECORD_ID"
                    ]
                )

                records[
                    source_record_id
                ] = record

            except Exception:

                continue

    return records


def append_checkpoint(
    record,
    checkpoint_path,
):
    """
    Append one completed classification to the checkpoint.
    """

    with open(
        checkpoint_path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


existing_checkpoint = (
    load_checkpoint(
        CHECKPOINT_JSONL
    )
)


print(
    "\nExisting checkpoint records:",
    len(existing_checkpoint),
)


# =============================================================================
# 16. CLASSIFY THE PILOT SAMPLE
# =============================================================================

context_columns = [
    column
    for column in [
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
    ]
    if column in pilot_df.columns
]


classification_records = []


for row_number, row in pilot_df.iterrows():

    source_record_id = str(
        row[id_column]
    )

    if (
        not REPROCESS_COMPLETED
        and source_record_id
        in existing_checkpoint
    ):

        print(
            f"[{row_number + 1}/"
            f"{len(pilot_df)}] "
            f"Checkpoint: "
            f"{source_record_id}"
        )

        classification_records.append(
            existing_checkpoint[
                source_record_id
            ]
        )

        continue

    narrative = str(
        row[TEXT_COLUMN]
    )

    narrative_may_be_truncated = bool(
        row[
            "_NARRATIVE_MAY_BE_TRUNCATED"
        ]
    )

    print(
        f"[{row_number + 1}/"
        f"{len(pilot_df)}] "
        f"Classifying "
        f"{source_record_id}"
    )

    started_at = time.time()

    classification = classify_narrative(
        narrative=narrative,

        narrative_may_be_truncated=(
            narrative_may_be_truncated
        ),
    )

    elapsed_seconds = (
        time.time()
        - started_at
    )

    output_record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "SOURCE_INDEX":
            row.get(
                "_ORIGINAL_INDEX",
                row_number,
            ),

        "NARRATIVE":
            narrative,

        "NARRATIVE_CHARACTER_COUNT":
            int(
                row[
                    "_NARRATIVE_CHARACTER_COUNT"
                ]
            ),

        "NARRATIVE_WORD_COUNT":
            int(
                row[
                    "_NARRATIVE_WORD_COUNT"
                ]
            ),

        "NARRATIVE_MAY_BE_TRUNCATED":
            narrative_may_be_truncated,

        **{
            column: row[column]
            for column in context_columns
        },

        **classification,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                elapsed_seconds,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    classification_records.append(
        output_record
    )

    append_checkpoint(
        record=output_record,

        checkpoint_path=(
            CHECKPOINT_JSONL
        ),
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 17. BUILD THE FINAL PILOT DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    classification_records
)


results_df = results_df.rename(
    columns={
        "primary_cause_category":
            "PRIMARY_CAUSE_CATEGORY",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "recommended_action_category":
            "RECOMMENDED_ACTION_CATEGORY",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "contributing_factor_evidence_text":
            "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",

        "action_rationale":
            "ACTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "action_confidence":
            "ACTION_CONFIDENCE",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


preferred_columns = [
    "SOURCE_RECORD_ID",
    "SOURCE_INDEX",
    "NARRATIVE",

    "PRIMARY_CAUSE_CATEGORY",
    "PRIMARY_CONTRIBUTING_FACTOR",
    "RECOMMENDED_ACTION_CATEGORY",

    "CAUSE_EVIDENCE_TEXT",
    "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",
    "ACTION_RATIONALE",

    "CAUSE_CONFIDENCE",
    "FACTOR_CONFIDENCE",
    "ACTION_CONFIDENCE",

    "CAUSE_EVIDENCE_VALID",
    "FACTOR_EVIDENCE_VALID",
    "REVIEW_REQUIRED",

    "INSUFFICIENT_INFORMATION_REASON",
    "REVIEWER_NOTE",

    "NARRATIVE_CHARACTER_COUNT",
    "NARRATIVE_WORD_COUNT",
    "NARRATIVE_MAY_BE_TRUNCATED",

    "PROCESSING_STATUS",
    "PROCESSING_ERROR",
    "PROCESSING_SECONDS",

    "MODEL_ID",
    "PROMPT_VERSION",
    "PROCESSED_AT",
]


preferred_columns += [
    column
    for column in context_columns
    if column not in preferred_columns
]


preferred_columns += [
    column
    for column in results_df.columns
    if column not in preferred_columns
]


results_df = results_df[
    preferred_columns
]


# =============================================================================
# 18. SAVE ALL PILOT OUTPUTS
# =============================================================================

results_df.to_csv(
    OUTPUT_CSV,

    index=False,

    encoding="utf-8-sig",
)


results_df.to_excel(
    OUTPUT_XLSX,

    index=False,

    engine="openpyxl",
)


results_df.to_json(
    OUTPUT_JSONL,

    orient="records",

    lines=True,

    force_ascii=False,
)


# =============================================================================
# 19. CREATE HUMAN-REVIEW WORKSHEET
# =============================================================================

review_columns = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",

    "PRIMARY_CAUSE_CATEGORY",
    "CAUSE_EVIDENCE_TEXT",
    "CAUSE_CONFIDENCE",

    "PRIMARY_CONTRIBUTING_FACTOR",
    "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",
    "FACTOR_CONFIDENCE",

    "RECOMMENDED_ACTION_CATEGORY",
    "ACTION_RATIONALE",
    "ACTION_CONFIDENCE",

    "NARRATIVE_MAY_BE_TRUNCATED",
    "REVIEW_REQUIRED",
    "REVIEWER_NOTE",

    "PROCESSING_STATUS",
    "PROCESSING_ERROR",
]


review_df = results_df[
    review_columns
].copy()


review_df[
    "HUMAN_CAUSE_CORRECT"
] = ""


review_df[
    "HUMAN_FACTOR_CORRECT"
] = ""


review_df[
    "HUMAN_ACTION_APPROPRIATE"
] = ""


review_df[
    "HUMAN_CORRECTED_CAUSE"
] = ""


review_df[
    "HUMAN_CORRECTED_FACTOR"
] = ""


review_df[
    "HUMAN_CORRECTED_ACTION"
] = ""


review_df[
    "HUMAN_REVIEW_COMMENTS"
] = ""


review_df.to_excel(
    REVIEW_XLSX,

    index=False,

    engine="openpyxl",
)


# =============================================================================
# 20. CREATE CATEGORY SUMMARIES
# =============================================================================

cause_summary = (
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "PRIMARY_CAUSE_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


factor_summary = (
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "PRIMARY_CONTRIBUTING_FACTOR"
    )
    .reset_index(
        name="COUNT"
    )
)


action_summary = (
    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "RECOMMENDED_ACTION_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


cause_factor_cross_tab = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
)


cause_action_cross_tab = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ],
)


with pd.ExcelWriter(
    SUMMARY_XLSX,

    engine="openpyxl",
) as writer:

    cause_summary.to_excel(
        writer,

        sheet_name="Cause Summary",

        index=False,
    )

    factor_summary.to_excel(
        writer,

        sheet_name="Factor Summary",

        index=False,
    )

    action_summary.to_excel(
        writer,

        sheet_name="Action Summary",

        index=False,
    )

    cause_factor_cross_tab.to_excel(
        writer,

        sheet_name="Cause by Factor",
    )

    cause_action_cross_tab.to_excel(
        writer,

        sheet_name="Cause by Action",
    )


# =============================================================================
# 21. FINAL QUALITY SUMMARY
# =============================================================================

successful_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq("SUCCESS")
    .sum()
)


failed_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq("FAILED")
    .sum()
)


review_required_count = int(
    results_df[
        "REVIEW_REQUIRED"
    ]
    .fillna(True)
    .sum()
)


valid_cause_evidence_count = int(
    results_df[
        "CAUSE_EVIDENCE_VALID"
    ]
    .fillna(False)
    .sum()
)


valid_factor_evidence_count = int(
    results_df[
        "FACTOR_EVIDENCE_VALID"
    ]
    .fillna(False)
    .sum()
)


average_processing_seconds = float(
    results_df[
        "PROCESSING_SECONDS"
    ]
    .mean()
)


print("\n" + "=" * 90)
print("HUGGING FACE MSHA PILOT COMPLETED")
print("=" * 90)

print(
    f"Model: "
    f"{MODEL_ID}"
)

print(
    f"GPU: "
    f"{gpu_name}"
)

print(
    f"Pilot records: "
    f"{len(results_df):,}"
)

print(
    f"Successful classifications: "
    f"{successful_count:,}"
)

print(
    f"Failed classifications: "
    f"{failed_count:,}"
)

print(
    f"Records requiring review: "
    f"{review_required_count:,}"
)

print(
    "Valid cause evidence excerpts: "
    f"{valid_cause_evidence_count:,}"
)

print(
    "Valid factor evidence excerpts: "
    f"{valid_factor_evidence_count:,}"
)

print(
    "Average processing seconds per narrative: "
    f"{average_processing_seconds:.3f}"
)

print(
    f"CSV: "
    f"{os.path.abspath(OUTPUT_CSV)}"
)

print(
    f"Excel: "
    f"{os.path.abspath(OUTPUT_XLSX)}"
)

print(
    f"JSONL: "
    f"{os.path.abspath(OUTPUT_JSONL)}"
)

print(
    f"Checkpoint: "
    f"{os.path.abspath(CHECKPOINT_JSONL)}"
)

print(
    f"Human review workbook: "
    f"{os.path.abspath(REVIEW_XLSX)}"
)

print(
    f"Category summary workbook: "
    f"{os.path.abspath(SUMMARY_XLSX)}"
)

print("=" * 90)


# =============================================================================
# 22. DISPLAY PILOT OUTPUTS
# =============================================================================

display_columns = [
    "SOURCE_RECORD_ID",

    "PRIMARY_CAUSE_CATEGORY",

    "PRIMARY_CONTRIBUTING_FACTOR",

    "RECOMMENDED_ACTION_CATEGORY",

    "CAUSE_CONFIDENCE",

    "FACTOR_CONFIDENCE",

    "ACTION_CONFIDENCE",

    "REVIEW_REQUIRED",

    "PROCESSING_STATUS",
]


display(
    results_df[
        display_columns
    ]
)


print("\nPRIMARY CAUSE DISTRIBUTION")

display(
    cause_summary
)


print("\nPRIMARY CONTRIBUTING-FACTOR DISTRIBUTION")

display(
    factor_summary
)


print("\nRECOMMENDED-ACTION DISTRIBUTION")

display(
    action_summary
)


print("\nRECORDS REQUIRING REVIEW")

display(
    results_df.loc[
        results_df[
            "REVIEW_REQUIRED"
        ].fillna(True),

        [
            "SOURCE_RECORD_ID",
            "NARRATIVE",
            "PRIMARY_CAUSE_CATEGORY",
            "PRIMARY_CONTRIBUTING_FACTOR",
            "RECOMMENDED_ACTION_CATEGORY",
            "REVIEWER_NOTE",
            "PROCESSING_STATUS",
            "PROCESSING_ERROR",
        ],
    ]
)
````

### Block 2 Initial test on Hugging Face model- Testing 10 Narratives

````python
# =============================================================================
# BLOCK 2 - MSHA HUGGING FACE PILOT, CORRECTED VERSION
#
# Run only after:
#   1. Package Block 1 completed.
#   2. The notebook kernel was restarted.
#   3. The accidents DataFrame was reloaded.
#
# This corrected version:
#   - uses the already downloaded Qwen2.5-7B-Instruct model;
#   - provides clearer field-specific category instructions;
#   - includes worked examples;
#   - returns validation feedback to the model during retries;
#   - repairs limited category-placement errors;
#   - does not stop the full pilot solely because one self-test requires review;
#   - uses a new checkpoint so earlier failed output is not reused.
#
# Outputs:
#   MSHA_HF_Pilot_Classifications_V1_2.csv
#   MSHA_HF_Pilot_Classifications_V1_2.xlsx
#   MSHA_HF_Pilot_Classifications_V1_2.jsonl
#   MSHA_HF_Pilot_Human_Review_V1_2.xlsx
#   MSHA_HF_Pilot_Checkpoint_V1_2.jsonl
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import sys
import os
import re
import gc
import json
import time
import random
import subprocess
from pathlib import Path
from datetime import datetime
from typing import Literal

import numpy as np
import pandas as pd
import torch

import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import (
    BaseModel,
    Field,
    ValidationError,
    model_validator,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)


# =============================================================================
# 2. VERIFY THE RESTARTED ENVIRONMENT
# =============================================================================

print("=" * 90)
print("RUNTIME ENVIRONMENT VERIFICATION")
print("=" * 90)

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)
print("Pydantic:", pydantic.__version__)
print("Pydantic Core:", pydantic_core.__version__)


pydantic_major_minor = tuple(
    int(part)
    for part in pydantic.__version__.split(".")[:2]
)

if pydantic_major_minor != (2, 12):

    raise RuntimeError(
        "The active Azure ML environment should use Pydantic 2.12.x, "
        f"but version {pydantic.__version__} is loaded. "
        "Run the package-installation block and restart the kernel."
    )


dependency_check = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "check",
    ],
    capture_output=True,
    text=True,
)

print("\nDependency check:")

print(
    dependency_check.stdout
    or dependency_check.stderr
    or "No dependency conflicts reported."
)

print(
    "\nNote: A pyarrow warning from azureml-dataset-runtime does not affect "
    "this pilot unless another notebook section specifically uses that runtime."
)


# =============================================================================
# 3. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

TEXT_COLUMN = "NARRATIVE"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

# Keep the first test small.
SAMPLE_SIZE = 10

STRATIFICATION_COLUMNS = [
    "CLASSIFICATION",
    "DEGREE_INJURY",
    "COAL_METAL_IND",
]

RANDOM_STATE = 42

PROMPT_VERSION = "MSHA_HF_CAUSE_FACTOR_ACTION_V1.2"

MAX_NEW_TOKENS = 500

MAX_RETRIES = 3

LOW_CONFIDENCE_THRESHOLD = 0.70

TRUNCATION_THRESHOLD = 383

RUN_FUNCTIONAL_TEST = True

# Use a new checkpoint file so the previous failed output is not reused.
CHECKPOINT_JSONL = "MSHA_HF_Pilot_Checkpoint_V1_2.jsonl"

OUTPUT_CSV = "MSHA_HF_Pilot_Classifications_V1_2.csv"

OUTPUT_XLSX = "MSHA_HF_Pilot_Classifications_V1_2.xlsx"

OUTPUT_JSONL = "MSHA_HF_Pilot_Classifications_V1_2.jsonl"

REVIEW_XLSX = "MSHA_HF_Pilot_Human_Review_V1_2.xlsx"

SUMMARY_XLSX = "MSHA_HF_Pilot_Category_Summary_V1_2.xlsx"

REPROCESS_COMPLETED = False


# =============================================================================
# 4. CONTROLLED TAXONOMIES
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_FALL",
    "STRUCK_BY_MOVING_OBJECT",
    "CAUGHT_IN_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "GROUND_OR_ROOF_FAILURE",
    "ELECTRICAL_CONTACT",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "CHEMICAL_OR_DUST_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NOT_STATED",
]


ACTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 5. PYDANTIC OUTPUT SCHEMA
# =============================================================================

class IncidentClassification(BaseModel):

    primary_cause_category: str

    primary_contributing_factor: str

    recommended_action_category: str

    cause_evidence_text: str = ""

    contributing_factor_evidence_text: str = ""

    action_rationale: str = ""

    cause_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    action_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_category_placement(self):

        if self.primary_cause_category not in CAUSE_CATEGORIES:

            raise ValueError(
                "primary_cause_category must be selected only from "
                f"CAUSE_CATEGORIES. Received: "
                f"{self.primary_cause_category}"
            )

        if self.primary_contributing_factor not in FACTOR_CATEGORIES:

            raise ValueError(
                "primary_contributing_factor must be selected only from "
                f"FACTOR_CATEGORIES. Received: "
                f"{self.primary_contributing_factor}"
            )

        if self.recommended_action_category not in ACTION_CATEGORIES:

            raise ValueError(
                "recommended_action_category must be selected only from "
                f"ACTION_CATEGORIES. Received: "
                f"{self.recommended_action_category}"
            )

        return self


# =============================================================================
# 6. VERIFY GPU
# =============================================================================

print("\n" + "=" * 90)
print("GPU VERIFICATION")
print("=" * 90)

if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. Attach the notebook to an "
        "Azure ML GPU compute instance."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0)
    .total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3

print("GPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


# =============================================================================
# 7. VALIDATE THE ACCIDENT DATAFRAME
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )

if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )

if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


df = accidents.copy()

df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"{TEXT_COLUMN!r} was not found. "
        f"Available columns: {list(df.columns)}"
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df = df[
    df[TEXT_COLUMN].str.len() > 0
].copy()


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN].str.len()
)


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_MAY_BE_TRUNCATED"] = (
    df["_CHARACTER_COUNT"]
    >= TRUNCATION_THRESHOLD
)


print(f"Usable narratives: {len(df):,}")
print(f"Record identifier: {id_column}")


# =============================================================================
# 8. SELECT A DIVERSE PILOT SAMPLE
# =============================================================================

def select_sample(
    data,
    sample_size,
    stratification_columns,
    random_state,
):

    sample_size = min(
        int(sample_size),
        len(data),
    )

    available_strata = [
        column
        for column in stratification_columns
        if column in data.columns
    ]

    if not available_strata:

        return (
            data.sample(
                n=sample_size,
                random_state=random_state,
            )
            .reset_index(drop=False)
            .rename(
                columns={
                    "index": "_ORIGINAL_INDEX"
                }
            )
        )

    working = data.copy()

    for column in available_strata:

        working[column] = (
            working[column]
            .astype("string")
            .fillna("<MISSING>")
        )

    grouped = working.groupby(
        available_strata,
        dropna=False,
        sort=False,
    )

    group_keys = list(
        grouped.groups.keys()
    )

    random.Random(
        random_state
    ).shuffle(
        group_keys
    )

    selected_indexes = []

    for group_number, group_key in enumerate(group_keys):

        if len(selected_indexes) >= sample_size:
            break

        group = grouped.get_group(
            group_key
        )

        selected_indexes.append(
            group.sample(
                n=1,
                random_state=(
                    random_state
                    + group_number
                ),
            ).index[0]
        )

    remaining_count = (
        sample_size
        - len(selected_indexes)
    )

    if remaining_count > 0:

        remaining = data.drop(
            index=selected_indexes,
            errors="ignore",
        )

        selected_indexes.extend(
            remaining.sample(
                n=min(
                    remaining_count,
                    len(remaining),
                ),
                random_state=random_state,
            )
            .index
            .tolist()
        )

    return (
        data.loc[selected_indexes]
        .copy()
        .reset_index(drop=False)
        .rename(
            columns={
                "index": "_ORIGINAL_INDEX"
            }
        )
    )


pilot_df = select_sample(
    data=df,
    sample_size=SAMPLE_SIZE,
    stratification_columns=STRATIFICATION_COLUMNS,
    random_state=RANDOM_STATE,
)


print(f"Pilot records selected: {len(pilot_df):,}")


# =============================================================================
# 9. LOAD THE MODEL
# =============================================================================

print("\n" + "=" * 90)
print("LOADING QWEN2.5-7B-INSTRUCT")
print("=" * 90)

print("Model:", MODEL_ID)


quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,

    bnb_4bit_quant_type="nf4",

    bnb_4bit_use_double_quant=True,

    bnb_4bit_compute_dtype=torch.float16,
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,

    use_fast=True,

    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = tokenizer.eos_token


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,

    quantization_config=quantization_config,

    device_map="auto",

    dtype=torch.float16,

    low_cpu_mem_usage=True,

    trust_remote_code=False,
)


model.eval()

print("Model loaded successfully.")
print("Model device:", model.device)


# =============================================================================
# 10. PROMPT WITH FIELD DEFINITIONS AND WORKED EXAMPLES
# =============================================================================

SYSTEM_PROMPT = f"""
You classify short MSHA accident narratives into three DIFFERENT types of
categories.

Do not confuse the category lists.

FIELD 1 - PRIMARY CAUSE CATEGORY:
This must describe the physical event or energy-transfer mechanism.
The only allowed values are:

{json.dumps(CAUSE_CATEGORIES, indent=2)}

FIELD 2 - PRIMARY CONTRIBUTING FACTOR:
This must describe the reported condition, body position, material state,
work method, equipment state, or environmental condition that contributed
to the event.
The only allowed values are:

{json.dumps(FACTOR_CATEGORIES, indent=2)}

FIELD 3 - RECOMMENDED ACTION CATEGORY:
This must describe the preliminary control category that directly addresses
the cause or contributing factor.
The only allowed values are:

{json.dumps(ACTION_CATEGORIES, indent=2)}

CRITICAL DISTINCTIONS:

- UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL is a CONTRIBUTING FACTOR.
  It is never a cause category.

- MATERIAL_RELEASE_OR_SHIFT is a CAUSE category.

- STRUCK_BY_MOVING_OBJECT is a CAUSE category.

- SECURE_OR_REPOSITION_MATERIAL is an ACTION category.

- REPAIR_REPLACE_OR_INSPECT_EQUIPMENT should be used only when the narrative
  supports an equipment defect, malfunction, damage, or inspection need.

- The presence of equipment does not by itself indicate equipment failure.

- If a bucket rests on a roof bolt and the roof bolt moves when retaining
  bands are cut:
    cause = MATERIAL_RELEASE_OR_SHIFT
    factor = UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL
    action = SECURE_OR_REPOSITION_MATERIAL

- If a loose strap is caught and the strap end moves and strikes a worker:
    cause = STRUCK_BY_MOVING_OBJECT
    factor = INADEQUATE_SECUREMENT_OR_RESTRAINT
    action = SECURE_OR_REPOSITION_MATERIAL

Use only information stated in the narrative.

Do not infer:
- worker negligence;
- missing training;
- missing PPE;
- management failure;
- equipment failure without evidence;
- a failed procedure unless it is described;
- an organizational root cause.

Copy evidence verbatim from the narrative.

Return only one JSON object with exactly these keys:

{{
  "primary_cause_category": "one CAUSE category",
  "primary_contributing_factor": "one FACTOR category",
  "recommended_action_category": "one ACTION category",
  "cause_evidence_text": "short verbatim excerpt",
  "contributing_factor_evidence_text": "short verbatim excerpt or empty string",
  "action_rationale": "one short sentence",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "action_confidence": 0.0,
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 11. JSON AND EVIDENCE UTILITIES
# =============================================================================

def extract_json_object(text):

    text = str(text).strip()

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0
    in_string = False
    escaped = False

    for position in range(start, len(text)):

        character = text[position]

        if escaped:

            escaped = False
            continue

        if character == "\\" and in_string:

            escaped = True
            continue

        if character == '"':

            in_string = not in_string
            continue

        if in_string:
            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    position + 1
                ]

    raise ValueError(
        "A complete JSON object was not found."
    )


def normalize_category(value):

    return (
        str(value)
        .strip()
        .upper()
        .replace("-", "_")
        .replace(" ", "_")
    )


def normalize_response_dictionary(response_dictionary):

    cleaned = dict(response_dictionary)

    for field_name in [
        "primary_cause_category",
        "primary_contributing_factor",
        "recommended_action_category",
    ]:

        if field_name in cleaned:

            cleaned[field_name] = (
                normalize_category(
                    cleaned[field_name]
                )
            )

    return cleaned


def normalize_text(value):

    return re.sub(
        r"\s+",
        " ",
        str(value),
    ).strip().lower()


def evidence_is_present(
    evidence,
    narrative,
):

    evidence_normalized = (
        normalize_text(evidence)
    )

    if not evidence_normalized:
        return True

    return (
        evidence_normalized
        in normalize_text(narrative)
    )


# =============================================================================
# 12. LIMITED CATEGORY-PLACEMENT REPAIR
# =============================================================================

def infer_cause_from_narrative(narrative):

    text = normalize_text(narrative)

    if any(
        phrase in text
        for phrase in [
            "came up",
            "shifted",
            "rolled",
            "fell from",
            "material moved",
            "load shifted",
            "released",
        ]
    ):

        return "MATERIAL_RELEASE_OR_SHIFT"

    if any(
        phrase in text
        for phrase in [
            "struck",
            "hit by",
            "struck by",
        ]
    ):

        return "STRUCK_BY_MOVING_OBJECT"

    if any(
        phrase in text
        for phrase in [
            "caught between",
            "pinched",
            "crushed between",
        ]
    ):

        return "CAUGHT_IN_BETWEEN"

    if any(
        phrase in text
        for phrase in [
            "slipped",
            "tripped",
            "fell",
        ]
    ):

        return "SLIP_TRIP_FALL"

    return "INSUFFICIENT_INFORMATION"


def infer_action_from_factor(factor_category):

    mapping = {
        "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
            "SECURE_OR_REPOSITION_MATERIAL",

        "INADEQUATE_SECUREMENT_OR_RESTRAINT":
            "SECURE_OR_REPOSITION_MATERIAL",

        "LINE_OF_FIRE_BODY_POSITION":
            "ESTABLISH_LINE_OF_FIRE_EXCLUSION",

        "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
            "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

        "ENERGY_ISOLATION_DEFICIENCY":
            "ISOLATE_OR_CONTROL_STORED_ENERGY",

        "EQUIPMENT_CONDITION_OR_FAILURE":
            "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

        "PROCEDURE_OR_WORK_METHOD":
            "REVISE_PROCEDURE_OR_JOB_PLAN",

        "COMMUNICATION_OR_COORDINATION":
            "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",

        "WORKSPACE_OR_ACCESS_LIMITATION":
            "REDESIGN_WORKSPACE_OR_ACCESS",
    }

    return mapping.get(
        factor_category,
        "INVESTIGATION_REQUIRED",
    )


def limited_category_repair(
    response_dictionary,
    narrative,
):
    """
    Repair only clear category-placement errors.

    The function does not freely invent categories. It handles cases in which
    the model placed a valid factor category in the cause field.
    """

    repaired = dict(
        response_dictionary
    )

    repairs = []

    cause_value = repaired.get(
        "primary_cause_category",
        "",
    )

    factor_value = repaired.get(
        "primary_contributing_factor",
        "",
    )

    action_value = repaired.get(
        "recommended_action_category",
        "",
    )

    # Specific error seen in the functional test:
    # the model placed a factor label in the cause field.
    if (
        cause_value in FACTOR_CATEGORIES
        and cause_value not in CAUSE_CATEGORIES
    ):

        misplaced_factor = cause_value

        repaired[
            "primary_cause_category"
        ] = infer_cause_from_narrative(
            narrative
        )

        if (
            factor_value not in FACTOR_CATEGORIES
            or factor_value == "NOT_STATED"
        ):

            repaired[
                "primary_contributing_factor"
            ] = misplaced_factor

        repairs.append(
            "Moved factor category from cause field and inferred "
            "the physical event category from narrative wording."
        )

    # If the cause value was placed in the factor field, move it back.
    factor_value = repaired.get(
        "primary_contributing_factor",
        "",
    )

    if (
        factor_value in CAUSE_CATEGORIES
        and factor_value not in FACTOR_CATEGORIES
    ):

        if (
            repaired.get(
                "primary_cause_category"
            )
            not in CAUSE_CATEGORIES
        ):

            repaired[
                "primary_cause_category"
            ] = factor_value

        repaired[
            "primary_contributing_factor"
        ] = "NOT_STATED"

        repairs.append(
            "Moved cause category from factor field."
        )

    # Repair an invalid action only when a valid factor gives a direct mapping.
    action_value = repaired.get(
        "recommended_action_category",
        "",
    )

    factor_value = repaired.get(
        "primary_contributing_factor",
        "",
    )

    if action_value not in ACTION_CATEGORIES:

        repaired[
            "recommended_action_category"
        ] = infer_action_from_factor(
            factor_value
        )

        repairs.append(
            "Replaced invalid action category using the validated "
            "contributing-factor category."
        )

    # The observed test response suggested repairing equipment even though the
    # narrative did not describe equipment failure. Replace this action only
    # when material instability was the factor.
    if (
        repaired.get(
            "recommended_action_category"
        )
        == "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT"
        and repaired.get(
            "primary_contributing_factor"
        )
        in {
            "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
            "INADEQUATE_SECUREMENT_OR_RESTRAINT",
        }
        and not any(
            phrase in normalize_text(narrative)
            for phrase in [
                "failed",
                "failure",
                "broken",
                "malfunction",
                "defective",
                "damaged",
            ]
        )
    ):

        repaired[
            "recommended_action_category"
        ] = "SECURE_OR_REPOSITION_MATERIAL"

        repairs.append(
            "Changed equipment-repair action to material-control action "
            "because no equipment defect was stated."
        )

    return repaired, repairs


# =============================================================================
# 13. MODEL GENERATION
# =============================================================================

def generate_response(messages):

    model_inputs = tokenizer.apply_chat_template(
        messages,

        add_generation_prompt=True,

        tokenize=True,

        return_dict=True,

        return_tensors="pt",
    )

    model_inputs = {
        key: value.to(model.device)
        for key, value in model_inputs.items()
    }

    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        generated_output = model.generate(
            **model_inputs,

            max_new_tokens=MAX_NEW_TOKENS,

            do_sample=False,

            repetition_penalty=1.03,

            pad_token_id=tokenizer.pad_token_id,

            eos_token_id=tokenizer.eos_token_id,

            use_cache=True,
        )

    generated_ids = generated_output[
        0,
        input_length:
    ]

    response_text = tokenizer.decode(
        generated_ids,

        skip_special_tokens=True,
    )

    del model_inputs
    del generated_output
    del generated_ids

    return response_text


# =============================================================================
# 14. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    may_be_truncated,
):

    truncation_note = (
        "The stored narrative may be truncated at the data-field limit."
        if may_be_truncated
        else
        "The narrative is not flagged as truncated."
    )

    user_prompt = f"""
{truncation_note}

NARRATIVE:
{narrative}

Classify the physical event, the separate contributing factor, and the
preliminary corrective-action category.

Return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_raw_response = ""

    all_repairs = []

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role": "system",
                "content": SYSTEM_PROMPT,
            },
            {
                "role": "user",
                "content": user_prompt,
            },
        ]

        if attempt_number > 1:

            repair_instruction = f"""
The previous JSON did not pass validation.

VALIDATION ERROR:
{prior_error}

PREVIOUS JSON:
{prior_response}

Correct the category placement.

Remember:
- cause must come only from CAUSE_CATEGORIES;
- factor must come only from FACTOR_CATEGORIES;
- action must come only from ACTION_CATEGORIES;
- UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL is a factor, not a cause;
- MATERIAL_RELEASE_OR_SHIFT is a cause;
- do not recommend equipment repair unless a defect or failure is stated.

Return only the corrected JSON object.
""".strip()

            messages.append(
                {
                    "role": "user",
                    "content": repair_instruction,
                }
            )

        try:

            raw_response = generate_response(
                messages
            )

            last_raw_response = raw_response

            json_text = extract_json_object(
                raw_response
            )

            response_dictionary = json.loads(
                json_text
            )

            response_dictionary = (
                normalize_response_dictionary(
                    response_dictionary
                )
            )

            response_dictionary, repairs = (
                limited_category_repair(
                    response_dictionary,
                    narrative,
                )
            )

            all_repairs.extend(
                repairs
            )

            result = (
                IncidentClassification
                .model_validate(
                    response_dictionary
                )
            )

            cause_evidence_valid = (
                evidence_is_present(
                    result.cause_evidence_text,
                    narrative,
                )
            )

            factor_evidence_valid = (
                evidence_is_present(
                    result
                    .contributing_factor_evidence_text,
                    narrative,
                )
            )

            minimum_confidence = min(
                result.cause_confidence,
                result.factor_confidence,
                result.action_confidence,
            )

            review_required = bool(
                may_be_truncated
                or minimum_confidence
                < LOW_CONFIDENCE_THRESHOLD
                or not cause_evidence_valid
                or not factor_evidence_valid
                or result.primary_cause_category
                == "INSUFFICIENT_INFORMATION"
                or result.primary_contributing_factor
                == "NOT_STATED"
                or result.recommended_action_category
                in {
                    "INVESTIGATION_REQUIRED",
                    "ENGINEERING_REVIEW_REQUIRED",
                    "INSUFFICIENT_INFORMATION",
                }
                or bool(
                    result.reviewer_note.strip()
                )
                or bool(all_repairs)
            )

            processing_status = (
                "SUCCESS_WITH_REPAIR"
                if all_repairs
                else "SUCCESS"
            )

            return {
                **result.model_dump(),

                "CAUSE_EVIDENCE_VALID":
                    cause_evidence_valid,

                "FACTOR_EVIDENCE_VALID":
                    factor_evidence_valid,

                "REVIEW_REQUIRED":
                    review_required,

                "AUTOMATIC_REPAIR_APPLIED":
                    bool(all_repairs),

                "AUTOMATIC_REPAIR_NOTES":
                    " | ".join(
                        dict.fromkeys(
                            all_repairs
                        )
                    ),

                "PROCESSING_STATUS":
                    processing_status,

                "PROCESSING_ERROR":
                    "",

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = (
                last_raw_response
            )

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:350]}"
            )

            gc.collect()
            torch.cuda.empty_cache()

    return {
        "primary_cause_category":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NOT_STATED",

        "recommended_action_category":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "contributing_factor_evidence_text":
            "",

        "action_rationale":
            (
                "The model output could not be validated and requires "
                "manual investigation."
            ),

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "action_confidence":
            0.0,

        "insufficient_information_reason":
            "Generation or validation failed.",

        "reviewer_note":
            "",

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_VALID":
            False,

        "REVIEW_REQUIRED":
            True,

        "AUTOMATIC_REPAIR_APPLIED":
            False,

        "AUTOMATIC_REPAIR_NOTES":
            "",

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "RAW_MODEL_RESPONSE":
            last_raw_response,
    }


# =============================================================================
# 15. CORRECTED FUNCTIONAL TEST
# =============================================================================

if RUN_FUNCTIONAL_TEST:

    print("\n" + "=" * 90)
    print("RUNNING CORRECTED MODEL FUNCTIONAL TEST")
    print("=" * 90)

    test_narrative = (
        "The employee was preparing to load roof bolts. "
        "The edge of a scoop bucket was resting on one roof bolt. "
        "When the retaining bands were cut, the roof bolt moved upward "
        "and struck the employee."
    )

    test_result = classify_narrative(
        narrative=test_narrative,
        may_be_truncated=False,
    )

    print(
        json.dumps(
            test_result,
            indent=2,
            default=str,
        )
    )

    acceptable_statuses = {
        "SUCCESS",
        "SUCCESS_WITH_REPAIR",
    }

    if (
        test_result[
            "PROCESSING_STATUS"
        ]
        not in acceptable_statuses
    ):

        print(
            "\nWARNING: The functional test did not produce a valid "
            "classification. The pilot will not be processed."
        )

        raise RuntimeError(
            "Corrected functional test failed. Review RAW_MODEL_RESPONSE "
            "and PROCESSING_ERROR."
        )

    print(
        "\nFunctional test passed with status:",
        test_result[
            "PROCESSING_STATUS"
        ],
    )


# =============================================================================
# 16. CHECKPOINT FUNCTIONS
# =============================================================================

def load_checkpoint(path):

    checkpoint_file = Path(path)

    if not checkpoint_file.exists():

        return {}

    checkpoint_records = {}

    with checkpoint_file.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:
                continue

            try:

                record = json.loads(
                    line
                )

                checkpoint_records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record

            except Exception:

                continue

    return checkpoint_records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


existing_checkpoint = load_checkpoint(
    CHECKPOINT_JSONL
)

print(
    "\nExisting V1.2 checkpoint records:",
    len(existing_checkpoint),
)


# =============================================================================
# 17. CLASSIFY THE PILOT SAMPLE
# =============================================================================

context_columns = [
    column
    for column in [
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
    ]
    if column in pilot_df.columns
]


records = []


for row_number, row in pilot_df.iterrows():

    source_record_id = str(
        row[id_column]
    )

    if (
        not REPROCESS_COMPLETED
        and source_record_id
        in existing_checkpoint
    ):

        print(
            f"[{row_number + 1}/{len(pilot_df)}] "
            f"Checkpoint: {source_record_id}"
        )

        records.append(
            existing_checkpoint[
                source_record_id
            ]
        )

        continue

    narrative = str(
        row[TEXT_COLUMN]
    )

    may_be_truncated = bool(
        row["_MAY_BE_TRUNCATED"]
    )

    print(
        f"[{row_number + 1}/{len(pilot_df)}] "
        f"Classifying {source_record_id}"
    )

    start_time = time.time()

    classification = classify_narrative(
        narrative=narrative,
        may_be_truncated=may_be_truncated,
    )

    record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "SOURCE_INDEX":
            row.get(
                "_ORIGINAL_INDEX",
                row_number,
            ),

        "NARRATIVE":
            narrative,

        "NARRATIVE_CHARACTER_COUNT":
            int(
                row["_CHARACTER_COUNT"]
            ),

        "NARRATIVE_WORD_COUNT":
            int(
                row["_WORD_COUNT"]
            ),

        "NARRATIVE_MAY_BE_TRUNCATED":
            may_be_truncated,

        **{
            column: row[column]
            for column in context_columns
        },

        **classification,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time() - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    records.append(
        record
    )

    append_checkpoint(
        record,
        CHECKPOINT_JSONL,
    )

    gc.collect()
    torch.cuda.empty_cache()


# =============================================================================
# 18. FORMAT THE FINAL DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    records
)


results_df = results_df.rename(
    columns={
        "primary_cause_category":
            "PRIMARY_CAUSE_CATEGORY",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "recommended_action_category":
            "RECOMMENDED_ACTION_CATEGORY",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "contributing_factor_evidence_text":
            "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",

        "action_rationale":
            "ACTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "action_confidence":
            "ACTION_CONFIDENCE",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


# =============================================================================
# 19. SAVE PILOT OUTPUTS
# =============================================================================

results_df.to_csv(
    OUTPUT_CSV,
    index=False,
    encoding="utf-8-sig",
)


results_df.to_excel(
    OUTPUT_XLSX,
    index=False,
    engine="openpyxl",
)


results_df.to_json(
    OUTPUT_JSONL,
    orient="records",
    lines=True,
    force_ascii=False,
)


# =============================================================================
# 20. HUMAN-REVIEW WORKBOOK
# =============================================================================

review_columns = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",

    "PRIMARY_CAUSE_CATEGORY",
    "CAUSE_EVIDENCE_TEXT",
    "CAUSE_CONFIDENCE",

    "PRIMARY_CONTRIBUTING_FACTOR",
    "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",
    "FACTOR_CONFIDENCE",

    "RECOMMENDED_ACTION_CATEGORY",
    "ACTION_RATIONALE",
    "ACTION_CONFIDENCE",

    "AUTOMATIC_REPAIR_APPLIED",
    "AUTOMATIC_REPAIR_NOTES",

    "NARRATIVE_MAY_BE_TRUNCATED",
    "REVIEW_REQUIRED",
    "REVIEWER_NOTE",

    "PROCESSING_STATUS",
    "PROCESSING_ERROR",
]


review_df = results_df[
    review_columns
].copy()


review_df["HUMAN_CAUSE_CORRECT"] = ""

review_df["HUMAN_FACTOR_CORRECT"] = ""

review_df["HUMAN_ACTION_APPROPRIATE"] = ""

review_df["HUMAN_CORRECTED_CAUSE"] = ""

review_df["HUMAN_CORRECTED_FACTOR"] = ""

review_df["HUMAN_CORRECTED_ACTION"] = ""

review_df["HUMAN_REVIEW_COMMENTS"] = ""


review_df.to_excel(
    REVIEW_XLSX,
    index=False,
    engine="openpyxl",
)


# =============================================================================
# 21. SUMMARY WORKBOOK
# =============================================================================

cause_summary = (
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "PRIMARY_CAUSE_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


factor_summary = (
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "PRIMARY_CONTRIBUTING_FACTOR"
    )
    .reset_index(
        name="COUNT"
    )
)


action_summary = (
    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "RECOMMENDED_ACTION_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


cause_factor_table = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
)


cause_action_table = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ],
)


with pd.ExcelWriter(
    SUMMARY_XLSX,
    engine="openpyxl",
) as writer:

    cause_summary.to_excel(
        writer,
        sheet_name="Cause Summary",
        index=False,
    )

    factor_summary.to_excel(
        writer,
        sheet_name="Factor Summary",
        index=False,
    )

    action_summary.to_excel(
        writer,
        sheet_name="Action Summary",
        index=False,
    )

    cause_factor_table.to_excel(
        writer,
        sheet_name="Cause by Factor",
    )

    cause_action_table.to_excel(
        writer,
        sheet_name="Cause by Action",
    )


# =============================================================================
# 22. FINAL QUALITY SUMMARY
# =============================================================================

successful_statuses = {
    "SUCCESS",
    "SUCCESS_WITH_REPAIR",
}


successful_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .isin(successful_statuses)
    .sum()
)


failed_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq("FAILED")
    .sum()
)


repair_count = int(
    results_df[
        "AUTOMATIC_REPAIR_APPLIED"
    ]
    .fillna(False)
    .sum()
)


review_required_count = int(
    results_df[
        "REVIEW_REQUIRED"
    ]
    .fillna(True)
    .sum()
)


print("\n" + "=" * 90)
print("CORRECTED HUGGING FACE PILOT COMPLETE")
print("=" * 90)

print("Model:", MODEL_ID)

print("GPU:", GPU_NAME)

print(
    "Pilot records:",
    len(results_df),
)

print(
    "Successful classifications:",
    successful_count,
)

print(
    "Failed classifications:",
    failed_count,
)

print(
    "Automatic repairs applied:",
    repair_count,
)

print(
    "Records requiring human review:",
    review_required_count,
)

print(
    "CSV:",
    os.path.abspath(OUTPUT_CSV),
)

print(
    "Excel:",
    os.path.abspath(OUTPUT_XLSX),
)

print(
    "JSONL:",
    os.path.abspath(OUTPUT_JSONL),
)

print(
    "Review workbook:",
    os.path.abspath(REVIEW_XLSX),
)

print(
    "Summary workbook:",
    os.path.abspath(SUMMARY_XLSX),
)

print(
    "Checkpoint:",
    os.path.abspath(CHECKPOINT_JSONL),
)

print("=" * 90)


# =============================================================================
# 23. DISPLAY RESULTS
# =============================================================================

display(
    results_df[
        [
            "SOURCE_RECORD_ID",

            "PRIMARY_CAUSE_CATEGORY",

            "PRIMARY_CONTRIBUTING_FACTOR",

            "RECOMMENDED_ACTION_CATEGORY",

            "CAUSE_CONFIDENCE",

            "FACTOR_CONFIDENCE",

            "ACTION_CONFIDENCE",

            "AUTOMATIC_REPAIR_APPLIED",

            "REVIEW_REQUIRED",

            "PROCESSING_STATUS",
        ]
    ]
)


print("\nCAUSE DISTRIBUTION")

display(
    cause_summary
)


print("\nCONTRIBUTING-FACTOR DISTRIBUTION")

display(
    factor_summary
)


print("\nRECOMMENDED-ACTION DISTRIBUTION")

display(
    action_summary
)


print("\nRECORDS REQUIRING REVIEW")

display(
    results_df.loc[
        results_df[
            "REVIEW_REQUIRED"
        ].fillna(True),

        [
            "SOURCE_RECORD_ID",

            "NARRATIVE",

            "PRIMARY_CAUSE_CATEGORY",

            "PRIMARY_CONTRIBUTING_FACTOR",

            "RECOMMENDED_ACTION_CATEGORY",

            "AUTOMATIC_REPAIR_NOTES",

            "REVIEWER_NOTE",

            "PROCESSING_STATUS",

            "PROCESSING_ERROR",
        ],
    ]
)
````

### Testing 100 narritives across 5 injury severity groups

````python
# =============================================================================
# MSHA HUGGING FACE 100-NARRATIVE SEVERITY-STRATIFIED ANALYSIS
#
# Prerequisites:
#   1. Run the package installation block from the prior version.
#   2. Restart the Azure ML notebook kernel.
#   3. Reload the pandas DataFrame named `accidents`.
#
# Model:
#   Qwen/Qwen2.5-7B-Instruct
#
# Final report:
#   MSHA_HF_100_Narrative_Severity_Report.pdf
#
# Supporting restart file:
#   MSHA_HF_100_Severity_Checkpoint.jsonl
#
# This code does not create Excel reports.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import sys
import os
import re
import gc
import json
import time
import random
import subprocess
import textwrap
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import (
    BaseModel,
    Field,
    model_validator,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from matplotlib.backends.backend_pdf import PdfPages


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

TEXT_COLUMN = "NARRATIVE"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

TOTAL_SAMPLE_SIZE = 100

TARGET_PER_SEVERITY_GROUP = 20

RANDOM_STATE = 42

PROMPT_VERSION = "MSHA_HF_CAUSE_FACTOR_ACTION_V2.0"

MAX_NEW_TOKENS = 500

MAX_RETRIES = 3

LOW_CONFIDENCE_THRESHOLD = 0.70

TRUNCATION_THRESHOLD = 383

RUN_FUNCTIONAL_TEST = True

REPROCESS_COMPLETED = False

OUTPUT_PDF = "MSHA_HF_100_Narrative_Severity_Report.pdf"

CHECKPOINT_JSONL = "MSHA_HF_100_Severity_Checkpoint.jsonl"


# =============================================================================
# 3. VERIFY ENVIRONMENT
# =============================================================================

print("=" * 90)
print("RUNTIME ENVIRONMENT VERIFICATION")
print("=" * 90)

print("Python executable:", sys.executable)
print("Python:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)
print("Pydantic:", pydantic.__version__)
print("Pydantic Core:", pydantic_core.__version__)


pydantic_major_minor = tuple(
    int(part)
    for part in pydantic.__version__.split(".")[:2]
)

if pydantic_major_minor != (2, 12):

    raise RuntimeError(
        "Pydantic 2.12.x is required in this Azure ML environment. "
        f"Loaded version: {pydantic.__version__}"
    )


if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. "
        "Attach the notebook to an Azure ML GPU compute instance."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0).total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3


print("\nGPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


dependency_check = subprocess.run(
    [
        sys.executable,
        "-m",
        "pip",
        "check",
    ],
    capture_output=True,
    text=True,
)

print("\nDependency check:")

print(
    dependency_check.stdout
    or dependency_check.stderr
    or "No dependency conflicts reported."
)

print(
    "\nA pyarrow warning from azureml-dataset-runtime may be ignored "
    "for this code because the Azure ML dataset runtime is not used."
)


# =============================================================================
# 4. TAXONOMIES
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "CAUGHT_IN_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "GROUND_OR_ROOF_FAILURE",
    "ELECTRICAL_CONTACT",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "CHEMICAL_OR_DUST_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NOT_STATED",
]


ACTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# Expected factor-action relationships.
FACTOR_ACTION_ALIGNMENT = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_SECUREMENT_OR_RESTRAINT": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    },

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_GUARDING_OR_BARRIER": {
        "INSTALL_OR_IMPROVE_GUARDING",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "ENERGY_ISOLATION_DEFICIENCY": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    },

    "COMMUNICATION_OR_COORDINATION": {
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "WORKSPACE_OR_ACCESS_LIMITATION": {
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "ENGINEERING_REVIEW_REQUIRED",
    },
}


# =============================================================================
# 5. PYDANTIC RESPONSE SCHEMA
# =============================================================================

class IncidentClassification(BaseModel):

    primary_cause_category: str

    primary_contributing_factor: str

    recommended_action_category: str

    cause_evidence_text: str = ""

    contributing_factor_evidence_text: str = ""

    action_rationale: str = ""

    cause_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    action_confidence: float = Field(
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    factor_support_level: str = "LOW"

    action_support_level: str = "LOW"

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_categories(self):

        if self.primary_cause_category not in CAUSE_CATEGORIES:

            raise ValueError(
                "Invalid primary cause category: "
                f"{self.primary_cause_category}"
            )

        if self.primary_contributing_factor not in FACTOR_CATEGORIES:

            raise ValueError(
                "Invalid contributing-factor category: "
                f"{self.primary_contributing_factor}"
            )

        if self.recommended_action_category not in ACTION_CATEGORIES:

            raise ValueError(
                "Invalid recommended-action category: "
                f"{self.recommended_action_category}"
            )

        valid_support_levels = {
            "HIGH",
            "MEDIUM",
            "LOW",
        }

        for field_name in [
            "cause_support_level",
            "factor_support_level",
            "action_support_level",
        ]:

            value = getattr(
                self,
                field_name,
            )

            if value not in valid_support_levels:

                raise ValueError(
                    f"{field_name} must be HIGH, MEDIUM, or LOW."
                )

        return self


# =============================================================================
# 6. PREPARE DATA
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"Column {TEXT_COLUMN!r} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"Column {SEVERITY_COLUMN!r} was not found."
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


df = df[
    df[TEXT_COLUMN].str.len() > 0
].copy()


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN].str.len()
)


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_MAY_BE_TRUNCATED"] = (
    df["_CHARACTER_COUNT"]
    >= TRUNCATION_THRESHOLD
)


# =============================================================================
# 7. DEFINE INJURY-SEVERITY GROUPS
# =============================================================================

def assign_severity_group(
    degree_injury,
):
    """
    Convert the MSHA DEGREE_INJURY text into five analytical severity groups.
    """

    value = str(
        degree_injury
    ).strip().upper()

    if "FATAL" in value:

        return "1_FATALITY"

    if any(
        phrase in value
        for phrase in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "2_PERMANENT_DISABILITY"

    if any(
        phrase in value
        for phrase in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "3_DAYS_AWAY"

    if any(
        phrase in value
        for phrase in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "4_RESTRICTED_OR_TRANSFERRED"

    return "5_NO_LOST_TIME_OR_OTHER"


df["SEVERITY_GROUP"] = (
    df[SEVERITY_COLUMN]
    .map(assign_severity_group)
)


SEVERITY_GROUP_ORDER = [
    "1_FATALITY",
    "2_PERMANENT_DISABILITY",
    "3_DAYS_AWAY",
    "4_RESTRICTED_OR_TRANSFERRED",
    "5_NO_LOST_TIME_OR_OTHER",
]


# =============================================================================
# 8. CREATE THE 100-RECORD STRATIFIED SAMPLE
# =============================================================================

def create_severity_sample(
    data,
    severity_groups,
    target_per_group,
    total_sample_size,
    random_state,
):
    """
    Select up to target_per_group records from each severity group.

    If a severity group contains fewer than the target, fill the remaining
    sample positions from all remaining eligible records.
    """

    selected_frames = []

    selected_indexes = set()

    sampling_rows = []

    for group_number, severity_group in enumerate(
        severity_groups
    ):

        group_frame = data[
            data["SEVERITY_GROUP"]
            == severity_group
        ]

        group_available = len(
            group_frame
        )

        group_requested = min(
            target_per_group,
            group_available,
        )

        if group_requested > 0:

            sampled_group = (
                group_frame.sample(
                    n=group_requested,
                    random_state=(
                        random_state
                        + group_number
                    ),
                )
            )

            selected_frames.append(
                sampled_group
            )

            selected_indexes.update(
                sampled_group.index.tolist()
            )

        sampling_rows.append(
            {
                "SEVERITY_GROUP":
                    severity_group,

                "AVAILABLE_RECORDS":
                    group_available,

                "INITIAL_TARGET":
                    target_per_group,

                "INITIAL_SELECTED":
                    group_requested,
            }
        )

    if selected_frames:

        sample = pd.concat(
            selected_frames,
            axis=0,
        )

    else:

        sample = data.iloc[0:0].copy()

    remaining_needed = (
        total_sample_size
        - len(sample)
    )

    if remaining_needed > 0:

        remaining_pool = data.drop(
            index=list(selected_indexes),
            errors="ignore",
        )

        additional_count = min(
            remaining_needed,
            len(remaining_pool),
        )

        if additional_count > 0:

            additional_sample = (
                remaining_pool.sample(
                    n=additional_count,
                    random_state=(
                        random_state
                        + 1000
                    ),
                )
            )

            sample = pd.concat(
                [
                    sample,
                    additional_sample,
                ],
                axis=0,
            )

    sample = (
        sample.sample(
            frac=1,
            random_state=random_state,
        )
        .reset_index(drop=False)
        .rename(
            columns={
                "index": "_ORIGINAL_INDEX"
            }
        )
    )

    sampling_summary = pd.DataFrame(
        sampling_rows
    )

    final_counts = (
        sample[
            "SEVERITY_GROUP"
        ]
        .value_counts()
    )

    sampling_summary[
        "FINAL_SELECTED"
    ] = (
        sampling_summary[
            "SEVERITY_GROUP"
        ]
        .map(final_counts)
        .fillna(0)
        .astype(int)
    )

    return sample, sampling_summary


sample_df, sampling_summary = (
    create_severity_sample(
        data=df,

        severity_groups=(
            SEVERITY_GROUP_ORDER
        ),

        target_per_group=(
            TARGET_PER_SEVERITY_GROUP
        ),

        total_sample_size=(
            TOTAL_SAMPLE_SIZE
        ),

        random_state=RANDOM_STATE,
    )
)


print("\n" + "=" * 90)
print("SEVERITY-STRATIFIED SAMPLE")
print("=" * 90)

print(
    sampling_summary.to_string(
        index=False
    )
)

print(
    f"\nTotal records selected: "
    f"{len(sample_df):,}"
)


# =============================================================================
# 9. LOAD MODEL
# =============================================================================

print("\n" + "=" * 90)
print("LOADING QWEN2.5-7B-INSTRUCT")
print("=" * 90)


quantization_config = (
    BitsAndBytesConfig(
        load_in_4bit=True,

        bnb_4bit_quant_type="nf4",

        bnb_4bit_use_double_quant=True,

        bnb_4bit_compute_dtype=torch.float16,
    )
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,

    use_fast=True,

    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = (
        tokenizer.eos_token
    )


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,

    quantization_config=(
        quantization_config
    ),

    device_map="auto",

    dtype=torch.float16,

    low_cpu_mem_usage=True,

    trust_remote_code=False,
)


model.eval()


print("Model loaded successfully.")
print("Model device:", model.device)


# =============================================================================
# 10. CLASSIFICATION PROMPT
# =============================================================================

SYSTEM_PROMPT = f"""
You classify short MSHA accident narratives into three separate fields.

This is reported-event classification, not verified root-cause analysis.

Use only information stated in the narrative.

FIELD 1 - PRIMARY CAUSE CATEGORY:
The physical event, movement, contact, release, body-motion event, or
energy-transfer mechanism.

Allowed cause values:
{json.dumps(CAUSE_CATEGORIES, indent=2)}

FIELD 2 - PRIMARY CONTRIBUTING FACTOR:
The most important stated material condition, body position, work method,
equipment condition, ground condition, environment, access limitation, or
coordination issue.

Allowed factor values:
{json.dumps(FACTOR_CATEGORIES, indent=2)}

FIELD 3 - RECOMMENDED ACTION CATEGORY:
A preliminary control category directly supported by the narrative.

Allowed action values:
{json.dumps(ACTION_CATEGORIES, indent=2)}

IMPORTANT RULES:

1. Cause, factor, and action are different concepts.

2. Use POOR_HOUSEKEEPING_OR_SURFACE_CONDITION only when the narrative
   describes grease, mud, water, ice, debris, uneven footing, loose material
   on a travel surface, or another condition affecting footing or access.

3. Do not select housekeeping merely because the words slip or slipped occur.

4. A hand slipping from a handle or strap is not automatically a
   housekeeping event.

5. Use STRUCK_AGAINST_STATIONARY_OBJECT when the worker moves into or strikes
   a fixed object.

6. Use LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT when control of a hand tool,
   control device, strap, handle, machine component, or equipment is lost.

7. Use GROUND_OR_ROOF_FAILURE for falling roof, rib, highwall, face, or
   uncontrolled ground movement.

8. Use UNSTABLE_GROUND_OR_ROOF_CONDITION when unstable rock, roof, rib,
   highwall, or ground is the contributing condition.

9. Use LINE_OF_FIRE_BODY_POSITION when the person is positioned in the path
   of moving material, falling ground, tools, vehicles, or released energy.

10. Do not recommend equipment repair unless the narrative states or strongly
    describes equipment failure, breakage, leakage, damage, or malfunction.

11. If a bucket or machine is involved but no failure is stated, do not infer
    equipment failure.

12. Copy cause and factor evidence verbatim from the narrative.

13. Evidence for the factor must explain the factor, not merely repeat an
    unrelated sentence.

14. If there is no stated factor, use NOT_STATED.

15. If the physical event cannot be determined, use
    INSUFFICIENT_INFORMATION.

16. Do not infer missing training, PPE, procedures, guarding, supervision,
    or management failure.

17. Support levels:
    HIGH = category and direct evidence are explicitly stated.
    MEDIUM = category is strongly implied but is not stated directly.
    LOW = evidence is incomplete, multiple interpretations are plausible,
          or the narrative is truncated.

18. Confidence values are model self-assessments and are not calibrated
    probabilities.

EXAMPLES:

Example A:
Narrative describes mud trapping a leg while the worker turns.
Cause = AWKWARD_MOVEMENT_OR_BODY_MOTION
Factor = POOR_HOUSEKEEPING_OR_SURFACE_CONDITION
Action = IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING

Example B:
A loose rock falls and strikes a worker.
Cause = GROUND_OR_ROOF_FAILURE
Factor = UNSTABLE_GROUND_OR_ROOF_CONDITION
Action = IMPROVE_GROUND_CONTROL_OR_SCALING

Example C:
A hand slips from a strap and strikes a fixed door.
Cause = STRUCK_AGAINST_STATIONARY_OBJECT
Factor = PROCEDURE_OR_WORK_METHOD or NOT_STATED
Action = REVISE_PROCEDURE_OR_JOB_PLAN or INVESTIGATION_REQUIRED

Example D:
A hydraulic line bursts and hydraulic fluid contacts a hot component.
Cause = PRESSURE_OR_STORED_ENERGY_RELEASE
Factor = EQUIPMENT_CONDITION_OR_FAILURE
Action = REPAIR_REPLACE_OR_INSPECT_EQUIPMENT

Return only one valid JSON object with these exact keys:

{{
  "primary_cause_category": "one allowed cause value",
  "primary_contributing_factor": "one allowed factor value",
  "recommended_action_category": "one allowed action value",
  "cause_evidence_text": "short verbatim excerpt or empty string",
  "contributing_factor_evidence_text": "short verbatim excerpt or empty string",
  "action_rationale": "one short sentence",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "action_confidence": 0.0,
  "cause_support_level": "HIGH, MEDIUM, or LOW",
  "factor_support_level": "HIGH, MEDIUM, or LOW",
  "action_support_level": "HIGH, MEDIUM, or LOW",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 11. JSON AND EVIDENCE UTILITIES
# =============================================================================

def extract_json_object(text):

    text = str(text).strip()

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace found."
        )

    depth = 0
    in_string = False
    escaped = False

    for position in range(
        start,
        len(text),
    ):

        character = text[position]

        if escaped:

            escaped = False
            continue

        if character == "\\" and in_string:

            escaped = True
            continue

        if character == '"':

            in_string = not in_string
            continue

        if in_string:
            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    position + 1
                ]

    raise ValueError(
        "Complete JSON object not found."
    )


def normalize_category(value):

    return (
        str(value)
        .strip()
        .upper()
        .replace("-", "_")
        .replace(" ", "_")
    )


def normalize_response_dictionary(
    response_dictionary,
):

    cleaned = dict(
        response_dictionary
    )

    for field_name in [
        "primary_cause_category",
        "primary_contributing_factor",
        "recommended_action_category",
        "cause_support_level",
        "factor_support_level",
        "action_support_level",
    ]:

        if field_name in cleaned:

            cleaned[field_name] = (
                normalize_category(
                    cleaned[field_name]
                )
            )

    return cleaned


def normalize_text(value):

    return re.sub(
        r"\s+",
        " ",
        str(value),
    ).strip().lower()


def evidence_is_present(
    evidence,
    narrative,
):

    normalized_evidence = (
        normalize_text(evidence)
    )

    if not normalized_evidence:
        return True

    return (
        normalized_evidence
        in normalize_text(narrative)
    )


def factor_action_aligned(
    factor,
    action,
):

    allowed_actions = (
        FACTOR_ACTION_ALIGNMENT.get(
            factor
        )
    )

    if allowed_actions is None:
        return True

    return action in allowed_actions


# =============================================================================
# 12. MODEL GENERATION
# =============================================================================

def generate_response(messages):

    model_inputs = (
        tokenizer.apply_chat_template(
            messages,

            add_generation_prompt=True,

            tokenize=True,

            return_dict=True,

            return_tensors="pt",
        )
    )

    model_inputs = {
        key: value.to(model.device)
        for key, value
        in model_inputs.items()
    }

    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        generated_output = model.generate(
            **model_inputs,

            max_new_tokens=(
                MAX_NEW_TOKENS
            ),

            do_sample=False,

            repetition_penalty=1.03,

            pad_token_id=(
                tokenizer.pad_token_id
            ),

            eos_token_id=(
                tokenizer.eos_token_id
            ),

            use_cache=True,
        )

    generated_ids = generated_output[
        0,
        input_length:
    ]

    response_text = tokenizer.decode(
        generated_ids,

        skip_special_tokens=True,
    )

    del model_inputs
    del generated_output
    del generated_ids

    return response_text


# =============================================================================
# 13. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    may_be_truncated,
):

    truncation_note = (
        "The stored narrative may be truncated at the source-field limit."
        if may_be_truncated
        else
        "The narrative is not flagged as truncated."
    )

    user_prompt = f"""
DATA QUALITY NOTE:
{truncation_note}

NARRATIVE:
{narrative}

Classify the event mechanism, separate contributing factor, and preliminary
action category.

Return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_response = ""

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role": "system",
                "content": SYSTEM_PROMPT,
            },
            {
                "role": "user",
                "content": user_prompt,
            },
        ]

        if attempt_number > 1:

            messages.append(
                {
                    "role": "user",

                    "content": f"""
The prior output failed validation.

VALIDATION ERROR:
{prior_error}

PRIOR OUTPUT:
{prior_response}

Correct the output.

Use a cause only from the cause list, a factor only from the factor list,
and an action only from the action list.

Do not return an empty category.

Return only valid JSON.
""".strip(),
                }
            )

        try:

            raw_response = (
                generate_response(
                    messages
                )
            )

            last_response = (
                raw_response
            )

            json_text = (
                extract_json_object(
                    raw_response
                )
            )

            response_dictionary = (
                json.loads(
                    json_text
                )
            )

            response_dictionary = (
                normalize_response_dictionary(
                    response_dictionary
                )
            )

            result = (
                IncidentClassification
                .model_validate(
                    response_dictionary
                )
            )

            cause_evidence_valid = (
                evidence_is_present(
                    result.cause_evidence_text,
                    narrative,
                )
            )

            factor_evidence_valid = (
                evidence_is_present(
                    result
                    .contributing_factor_evidence_text,
                    narrative,
                )
            )

            alignment_valid = (
                factor_action_aligned(
                    result
                    .primary_contributing_factor,

                    result
                    .recommended_action_category,
                )
            )

            minimum_confidence = min(
                result.cause_confidence,
                result.factor_confidence,
                result.action_confidence,
            )

            low_support = any(
                level == "LOW"
                for level in [
                    result.cause_support_level,
                    result.factor_support_level,
                    result.action_support_level,
                ]
            )

            review_required = bool(
                may_be_truncated
                or minimum_confidence
                < LOW_CONFIDENCE_THRESHOLD
                or low_support
                or not cause_evidence_valid
                or not factor_evidence_valid
                or not alignment_valid
                or result.primary_cause_category
                == "INSUFFICIENT_INFORMATION"
                or result.primary_contributing_factor
                == "NOT_STATED"
                or result.recommended_action_category
                in {
                    "ENGINEERING_REVIEW_REQUIRED",
                    "INVESTIGATION_REQUIRED",
                    "INSUFFICIENT_INFORMATION",
                }
                or bool(
                    result.reviewer_note.strip()
                )
            )

            return {
                **result.model_dump(),

                "CAUSE_EVIDENCE_VALID":
                    cause_evidence_valid,

                "FACTOR_EVIDENCE_VALID":
                    factor_evidence_valid,

                "FACTOR_ACTION_ALIGNMENT_VALID":
                    alignment_valid,

                "REVIEW_REQUIRED":
                    review_required,

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = (
                last_response
            )

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:300]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        "primary_cause_category":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NOT_STATED",

        "recommended_action_category":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "contributing_factor_evidence_text":
            "",

        "action_rationale":
            (
                "The model response could not be validated."
            ),

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "action_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "factor_support_level":
            "LOW",

        "action_support_level":
            "LOW",

        "insufficient_information_reason":
            "Generation or validation failed.",

        "reviewer_note":
            "",

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_VALID":
            False,

        "FACTOR_ACTION_ALIGNMENT_VALID":
            False,

        "REVIEW_REQUIRED":
            True,

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "RAW_MODEL_RESPONSE":
            last_response,
    }


# =============================================================================
# 14. FUNCTIONAL TEST
# =============================================================================

if RUN_FUNCTIONAL_TEST:

    print("\n" + "=" * 90)
    print("RUNNING FUNCTIONAL TEST")
    print("=" * 90)

    test_narrative = (
        "The employee was preparing to load roof bolts. "
        "The edge of a scoop bucket was resting on one roof bolt. "
        "When the retaining bands were cut, the roof bolt moved upward "
        "and struck the employee."
    )

    test_result = classify_narrative(
        narrative=test_narrative,

        may_be_truncated=False,
    )

    print(
        json.dumps(
            test_result,
            indent=2,
            default=str,
        )
    )

    if (
        test_result[
            "PROCESSING_STATUS"
        ]
        != "SUCCESS"
    ):

        raise RuntimeError(
            "The functional test failed. "
            "Review PROCESSING_ERROR and RAW_MODEL_RESPONSE."
        )

    print(
        "\nFunctional test completed successfully."
    )


# =============================================================================
# 15. CHECKPOINT FUNCTIONS
# =============================================================================

def load_checkpoint(path):

    checkpoint_file = Path(path)

    if not checkpoint_file.exists():

        return {}

    records = {}

    with checkpoint_file.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:
                continue

            try:

                record = json.loads(
                    line
                )

                records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record

            except Exception:
                continue

    return records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


existing_checkpoint = (
    load_checkpoint(
        CHECKPOINT_JSONL
    )
)


print(
    "\nExisting checkpoint records:",
    len(existing_checkpoint),
)


# =============================================================================
# 16. PROCESS THE 100 NARRATIVES
# =============================================================================

context_columns = [
    column
    for column in [
        "ACCIDENT_DT",
        "CAL_YR",
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
        "DAYS_RESTRICT",
        "DAYS_LOST",
    ]
    if column in sample_df.columns
]


records = []


for row_number, row in sample_df.iterrows():

    source_record_id = str(
        row[id_column]
    )

    if (
        not REPROCESS_COMPLETED
        and source_record_id
        in existing_checkpoint
    ):

        print(
            f"[{row_number + 1}/{len(sample_df)}] "
            f"Checkpoint: {source_record_id}"
        )

        records.append(
            existing_checkpoint[
                source_record_id
            ]
        )

        continue

    narrative = str(
        row[TEXT_COLUMN]
    )

    may_be_truncated = bool(
        row["_MAY_BE_TRUNCATED"]
    )

    print(
        f"[{row_number + 1}/{len(sample_df)}] "
        f"{row['SEVERITY_GROUP']} - "
        f"{source_record_id}"
    )

    start_time = time.time()

    classification = classify_narrative(
        narrative=narrative,

        may_be_truncated=(
            may_be_truncated
        ),
    )

    record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "SOURCE_INDEX":
            row.get(
                "_ORIGINAL_INDEX",
                row_number,
            ),

        "NARRATIVE":
            narrative,

        "SEVERITY_GROUP":
            row["SEVERITY_GROUP"],

        "NARRATIVE_CHARACTER_COUNT":
            int(
                row["_CHARACTER_COUNT"]
            ),

        "NARRATIVE_WORD_COUNT":
            int(
                row["_WORD_COUNT"]
            ),

        "NARRATIVE_MAY_BE_TRUNCATED":
            may_be_truncated,

        **{
            column: row[column]
            for column in context_columns
        },

        **classification,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    records.append(
        record
    )

    append_checkpoint(
        record,
        CHECKPOINT_JSONL,
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 17. CREATE FINAL RESULTS DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    records
)


results_df = results_df.rename(
    columns={
        "primary_cause_category":
            "PRIMARY_CAUSE_CATEGORY",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "recommended_action_category":
            "RECOMMENDED_ACTION_CATEGORY",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "contributing_factor_evidence_text":
            "CONTRIBUTING_FACTOR_EVIDENCE_TEXT",

        "action_rationale":
            "ACTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "action_confidence":
            "ACTION_CONFIDENCE",

        "cause_support_level":
            "CAUSE_SUPPORT_LEVEL",

        "factor_support_level":
            "FACTOR_SUPPORT_LEVEL",

        "action_support_level":
            "ACTION_SUPPORT_LEVEL",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


results_df["SEVERITY_GROUP"] = pd.Categorical(
    results_df["SEVERITY_GROUP"],

    categories=SEVERITY_GROUP_ORDER,

    ordered=True,
)


results_df = results_df.sort_values(
    [
        "SEVERITY_GROUP",
        "SOURCE_RECORD_ID",
    ]
).reset_index(drop=True)


# =============================================================================
# 18. ANALYTICAL TABLES
# =============================================================================

severity_counts = (
    results_df[
        "SEVERITY_GROUP"
    ]
    .value_counts(
        sort=False
    )
    .rename_axis(
        "SEVERITY_GROUP"
    )
    .reset_index(
        name="COUNT"
    )
)


cause_summary = (
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ]
    .value_counts()
    .rename_axis(
        "PRIMARY_CAUSE_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


factor_summary = (
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    .value_counts()
    .rename_axis(
        "PRIMARY_CONTRIBUTING_FACTOR"
    )
    .reset_index(
        name="COUNT"
    )
)


action_summary = (
    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ]
    .value_counts()
    .rename_axis(
        "RECOMMENDED_ACTION_CATEGORY"
    )
    .reset_index(
        name="COUNT"
    )
)


cause_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


factor_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


action_by_severity = pd.crosstab(
    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


cause_by_factor = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
)


cause_by_action = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE_CATEGORY"
    ],

    results_df[
        "RECOMMENDED_ACTION_CATEGORY"
    ],
)


quality_summary = pd.DataFrame(
    [
        {
            "MEASURE":
                "Records processed",

            "VALUE":
                len(results_df),
        },
        {
            "MEASURE":
                "Successful classifications",

            "VALUE":
                int(
                    results_df[
                        "PROCESSING_STATUS"
                    ]
                    .eq("SUCCESS")
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Failed classifications",

            "VALUE":
                int(
                    results_df[
                        "PROCESSING_STATUS"
                    ]
                    .eq("FAILED")
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Records requiring review",

            "VALUE":
                int(
                    results_df[
                        "REVIEW_REQUIRED"
                    ]
                    .fillna(True)
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Truncated or near-limit narratives",

            "VALUE":
                int(
                    results_df[
                        "NARRATIVE_MAY_BE_TRUNCATED"
                    ]
                    .fillna(False)
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Invalid cause evidence",

            "VALUE":
                int(
                    (
                        ~results_df[
                            "CAUSE_EVIDENCE_VALID"
                        ]
                        .fillna(False)
                    )
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Invalid factor evidence",

            "VALUE":
                int(
                    (
                        ~results_df[
                            "FACTOR_EVIDENCE_VALID"
                        ]
                        .fillna(False)
                    )
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Factor-action alignment flags",

            "VALUE":
                int(
                    (
                        ~results_df[
                            "FACTOR_ACTION_ALIGNMENT_VALID"
                        ]
                        .fillna(False)
                    )
                    .sum()
                ),
        },
        {
            "MEASURE":
                "Mean processing seconds",

            "VALUE":
                round(
                    float(
                        results_df[
                            "PROCESSING_SECONDS"
                        ].mean()
                    ),
                    2,
                ),
        },
    ]
)


# =============================================================================
# 19. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi": 110,
        "font.size": 9,
        "axes.grid": False,
    }
)


def create_page(
    title=None,
    figsize=(8.5, 11),
):

    figure = plt.figure(
        figsize=figsize
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.965,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(figure):

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def wrap_text_lines(
    lines,
    width=108,
):

    output = []

    for line in lines:

        if not line:

            output.append("")

        else:

            output.extend(
                textwrap.wrap(
                    str(line),
                    width=width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    return output


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=56,
    width=108,
):

    wrapped = wrap_text_lines(
        lines,
        width=width,
    )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(wrapped),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - Page {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.925,
            "\n".join(chunk),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.2,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=24,
):

    if dataframe is None or dataframe.empty:

        add_text_pages(
            title,
            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    printable.columns = [
        str(column)
        for column in printable.columns
    ]

    for column in printable.columns:

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    ""
                    if pd.isna(value)
                    else str(value)
            )
        )

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=column_wrap_width,
                            break_long_words=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(printable),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        page_title = (
            title
            if start == 0
            else title + " (continued)"
        )

        figure = create_page(
            page_title
        )

        axis = figure.add_axes(
            [
                0.025,
                0.055,
                0.95,
                0.865,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,

            colLabels=chunk.columns,

            cellLoc="left",

            colLoc="left",

            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.45,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAAAAA"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_bar_chart(
    title,
    dataframe,
    category_column,
    count_column="COUNT",
    color="#2E5E8C",
):

    plot_data = (
        dataframe.copy()
        .sort_values(
            count_column,
            ascending=True,
        )
    )

    figure = plt.figure(
        figsize=(8.5, 7)
    )

    axis = figure.add_subplot(
        111
    )

    axis.barh(
        plot_data[
            category_column
        ],

        plot_data[
            count_column
        ],

        color=color,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        "Number of sampled narratives"
    )

    axis.grid(
        axis="x",
        alpha=0.25,
    )

    figure.tight_layout()

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_heatmap(
    title,
    dataframe,
    x_label,
    y_label,
):

    if dataframe.empty:

        return

    figure_width = max(
        8.5,
        0.75 * len(
            dataframe.columns
        )
        + 4,
    )

    figure_height = max(
        6.5,
        0.35 * len(
            dataframe.index
        )
        + 3,
    )

    figure = plt.figure(
        figsize=(
            figure_width,
            figure_height,
        )
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                dataframe.columns
            )
        )
    )

    axis.set_xticklabels(
        dataframe.columns,
        rotation=45,
        ha="right",
    )

    axis.set_yticks(
        range(
            len(
                dataframe.index
            )
        )
    )

    axis.set_yticklabels(
        dataframe.index
    )

    axis.set_xlabel(
        x_label
    )

    axis.set_ylabel(
        y_label
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value:

                axis.text(
                    column_number,
                    row_number,
                    str(value),
                    ha="center",
                    va="center",
                    fontsize=8,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Count",
    )

    figure.tight_layout()

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_record_detail(
    row,
    record_number,
    total_records,
):

    lines = [
        f"Record {record_number} of {total_records}",
        "",
        f"Source record ID: {row.get('SOURCE_RECORD_ID', '')}",
        f"Severity group: {row.get('SEVERITY_GROUP', '')}",
        f"Degree of injury: {row.get('DEGREE_INJURY', '')}",
        f"MSHA classification: {row.get('CLASSIFICATION', '')}",
        f"MSHA accident type: {row.get('ACCIDENT_TYPE', '')}",
        f"Coal/metal indicator: {row.get('COAL_METAL_IND', '')}",
        f"Narrative characters: {row.get('NARRATIVE_CHARACTER_COUNT', '')}",
        f"Narrative may be truncated: {row.get('NARRATIVE_MAY_BE_TRUNCATED', '')}",
        "",
        "FULL NARRATIVE",
        "-" * 80,
        str(
            row.get(
                "NARRATIVE",
                "",
            )
        ),
        "",
        "LLM CLASSIFICATION",
        "-" * 80,
        f"Primary cause: {row.get('PRIMARY_CAUSE_CATEGORY', '')}",
        f"Cause support: {row.get('CAUSE_SUPPORT_LEVEL', '')}",
        f"Cause confidence: {row.get('CAUSE_CONFIDENCE', '')}",
        f"Cause evidence valid: {row.get('CAUSE_EVIDENCE_VALID', '')}",
        f"Cause evidence: {row.get('CAUSE_EVIDENCE_TEXT', '')}",
        "",
        f"Primary contributing factor: {row.get('PRIMARY_CONTRIBUTING_FACTOR', '')}",
        f"Factor support: {row.get('FACTOR_SUPPORT_LEVEL', '')}",
        f"Factor confidence: {row.get('FACTOR_CONFIDENCE', '')}",
        f"Factor evidence valid: {row.get('FACTOR_EVIDENCE_VALID', '')}",
        f"Factor evidence: {row.get('CONTRIBUTING_FACTOR_EVIDENCE_TEXT', '')}",
        "",
        f"Recommended action: {row.get('RECOMMENDED_ACTION_CATEGORY', '')}",
        f"Action support: {row.get('ACTION_SUPPORT_LEVEL', '')}",
        f"Action confidence: {row.get('ACTION_CONFIDENCE', '')}",
        f"Action rationale: {row.get('ACTION_RATIONALE', '')}",
        "",
        "QUALITY AND REVIEW",
        "-" * 80,
        f"Factor-action alignment valid: {row.get('FACTOR_ACTION_ALIGNMENT_VALID', '')}",
        f"Review required: {row.get('REVIEW_REQUIRED', '')}",
        f"Reviewer note: {row.get('REVIEWER_NOTE', '')}",
        f"Insufficient information reason: {row.get('INSUFFICIENT_INFORMATION_REASON', '')}",
        f"Processing status: {row.get('PROCESSING_STATUS', '')}",
        f"Processing error: {row.get('PROCESSING_ERROR', '')}",
        f"Processing seconds: {row.get('PROCESSING_SECONDS', '')}",
    ]

    add_text_pages(
        (
            "Detailed Classification - "
            f"{row.get('SOURCE_RECORD_ID', '')}"
        ),

        lines,

        font_size=8.0,

        lines_per_page=52,

        width=106,
    )


# =============================================================================
# 20. BUILD THE SINGLE PDF REPORT
# =============================================================================

# Cover page.
figure = create_page()

figure.text(
    0.5,
    0.77,
    "MSHA Narrative Cause, Factor, and Action Analysis",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.71,
    "100-Narrative Severity-Stratified Hugging Face Pilot",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.66,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=10,
    color="#666666",
)

cover_text = [
    f"Model: {MODEL_ID}",
    f"GPU: {GPU_NAME}",
    f"Prompt version: {PROMPT_VERSION}",
    f"Records classified: {len(results_df):,}",
    "",
    "Purpose:",
    "Convert narrative text into one reported causal mechanism, one",
    "contributing-factor category, and one preliminary action category.",
    "",
    "Important limitation:",
    "The extracted categories are model-generated decision-support labels.",
    "They are not verified root-cause findings or approved corrective actions.",
    "Records marked for review require qualified human evaluation.",
]

figure.text(
    0.12,
    0.56,
    "\n".join(
        cover_text
    ),
    va="top",
    ha="left",
    fontsize=9.5,
    family="monospace",
)

save_page(
    figure
)


# Executive summary.
top_cause = (
    cause_summary.iloc[0]
    if not cause_summary.empty
    else None
)

top_factor = (
    factor_summary.iloc[0]
    if not factor_summary.empty
    else None
)

top_action = (
    action_summary.iloc[0]
    if not action_summary.empty
    else None
)


executive_lines = [
    "This report evaluates a 100-record sample selected across five injury",
    "severity groups. The model assigned one cause, one contributing factor,",
    "and one preliminary action category to each narrative.",
    "",
    f"Successful classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum())}",
    f"Failed classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('FAILED').sum())}",
    f"Records requiring review: "
    f"{int(results_df['REVIEW_REQUIRED'].fillna(True).sum())}",
    f"Narratives near the source-field limit: "
    f"{int(results_df['NARRATIVE_MAY_BE_TRUNCATED'].fillna(False).sum())}",
]

if top_cause is not None:

    executive_lines.extend(
        [
            "",
            (
                "Most frequently assigned cause: "
                f"{top_cause['PRIMARY_CAUSE_CATEGORY']} "
                f"({top_cause['COUNT']} records)"
            ),
        ]
    )

if top_factor is not None:

    executive_lines.append(
        (
            "Most frequently assigned factor: "
            f"{top_factor['PRIMARY_CONTRIBUTING_FACTOR']} "
            f"({top_factor['COUNT']} records)"
        )
    )

if top_action is not None:

    executive_lines.append(
        (
            "Most frequently assigned action: "
            f"{top_action['RECOMMENDED_ACTION_CATEGORY']} "
            f"({top_action['COUNT']} records)"
        )
    )

executive_lines.extend(
    [
        "",
        "These frequencies describe the sampled records and model assignments.",
        "They should not be generalized to the complete MSHA dataset until the",
        "sample classifications have been manually reviewed and validated.",
    ]
)


add_text_pages(
    "Executive Summary",
    executive_lines,
    font_size=9,
)


# Methodology.
add_text_pages(
    "Methodology and Interpretation",
    [
        "Sampling:",
        "The code attempted to select 20 narratives from each of five injury",
        "severity groups. If a group contained fewer than 20 eligible records,",
        "remaining sample positions were filled from the unselected dataset.",
        "",
        "Classification:",
        "Qwen2.5-7B-Instruct was loaded locally in 4-bit mode on the Azure ML",
        "GPU. The model was restricted to controlled cause, factor, and action",
        "taxonomies.",
        "",
        "Evidence:",
        "The cause and factor evidence fields were required to contain literal",
        "narrative excerpts. Automatic evidence validation confirms only that",
        "the words occur in the narrative; it does not prove that the excerpt",
        "semantically supports the assigned category.",
        "",
        "Review flags:",
        "A record was marked for review when it was truncated, low confidence,",
        "low support, missing a factor, failed evidence validation, had an",
        "unusual factor-action relationship, required investigation, or",
        "contained a model reviewer note.",
        "",
        "Corrective actions:",
        "Recommended actions are preliminary categories for analytical and",
        "research purposes. They require review by qualified safety, technical,",
        "engineering, and operational personnel.",
    ],
    font_size=8.8,
)


# Sample design.
add_table_pages(
    "Severity-Stratified Sample Design",
    sampling_summary,
    rows_per_page=10,
    font_size=7.5,
    column_wrap_width=28,
)


add_table_pages(
    "Final Sample Counts by Severity",
    severity_counts,
    rows_per_page=10,
    font_size=7.5,
    column_wrap_width=30,
)


# Quality summary.
add_table_pages(
    "Processing and Quality Summary",
    quality_summary,
    rows_per_page=15,
    font_size=7.5,
    column_wrap_width=32,
)


# Summary tables.
add_table_pages(
    "Primary Cause Distribution",
    cause_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=34,
)


add_bar_chart(
    "Primary Cause Distribution",
    cause_summary,
    "PRIMARY_CAUSE_CATEGORY",
)


add_table_pages(
    "Primary Contributing-Factor Distribution",
    factor_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=34,
)


add_bar_chart(
    "Primary Contributing-Factor Distribution",
    factor_summary,
    "PRIMARY_CONTRIBUTING_FACTOR",
    color="#4E7D55",
)


add_table_pages(
    "Recommended-Action Distribution",
    action_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=34,
)


add_bar_chart(
    "Recommended-Action Distribution",
    action_summary,
    "RECOMMENDED_ACTION_CATEGORY",
    color="#B46A3C",
)


# Severity cross-tabulations.
add_table_pages(
    "Cause by Injury-Severity Group",
    cause_by_severity.reset_index(),
    rows_per_page=25,
    font_size=6.5,
    column_wrap_width=20,
)


add_heatmap(
    "Cause by Injury-Severity Group",
    cause_by_severity,
    "Injury-severity group",
    "Cause category",
)


add_table_pages(
    "Contributing Factor by Injury-Severity Group",
    factor_by_severity.reset_index(),
    rows_per_page=25,
    font_size=6.5,
    column_wrap_width=20,
)


add_heatmap(
    "Contributing Factor by Injury-Severity Group",
    factor_by_severity,
    "Injury-severity group",
    "Contributing-factor category",
)


add_table_pages(
    "Recommended Action by Injury-Severity Group",
    action_by_severity.reset_index(),
    rows_per_page=25,
    font_size=6.5,
    column_wrap_width=20,
)


add_heatmap(
    "Recommended Action by Injury-Severity Group",
    action_by_severity,
    "Injury-severity group",
    "Recommended-action category",
)


# Cause-factor and cause-action tables.
add_table_pages(
    "Cause by Contributing Factor",
    cause_by_factor.reset_index(),
    rows_per_page=20,
    font_size=5.8,
    column_wrap_width=18,
)


add_heatmap(
    "Cause by Contributing Factor",
    cause_by_factor,
    "Contributing-factor category",
    "Cause category",
)


add_table_pages(
    "Cause by Recommended Action",
    cause_by_action.reset_index(),
    rows_per_page=20,
    font_size=5.8,
    column_wrap_width=18,
)


add_heatmap(
    "Cause by Recommended Action",
    cause_by_action,
    "Recommended-action category",
    "Cause category",
)


# Records requiring review.
review_records = results_df[
    results_df[
        "REVIEW_REQUIRED"
    ].fillna(True)
].copy()


review_summary_columns = [
    "SOURCE_RECORD_ID",
    "SEVERITY_GROUP",
    "PRIMARY_CAUSE_CATEGORY",
    "PRIMARY_CONTRIBUTING_FACTOR",
    "RECOMMENDED_ACTION_CATEGORY",
    "CAUSE_SUPPORT_LEVEL",
    "FACTOR_SUPPORT_LEVEL",
    "ACTION_SUPPORT_LEVEL",
    "FACTOR_ACTION_ALIGNMENT_VALID",
    "PROCESSING_STATUS",
]


add_table_pages(
    "Records Requiring Human Review",
    review_records[
        review_summary_columns
    ],
    rows_per_page=15,
    font_size=5.9,
    column_wrap_width=18,
)


# Compact output for all records.
all_record_summary_columns = [
    "SOURCE_RECORD_ID",
    "SEVERITY_GROUP",
    "PRIMARY_CAUSE_CATEGORY",
    "PRIMARY_CONTRIBUTING_FACTOR",
    "RECOMMENDED_ACTION_CATEGORY",
    "REVIEW_REQUIRED",
    "PROCESSING_STATUS",
]


add_table_pages(
    "All Record Classifications",
    results_df[
        all_record_summary_columns
    ],
    rows_per_page=18,
    font_size=5.9,
    column_wrap_width=18,
)


# Detailed output for every record.
for record_number, (
    row_index,
    record_row,
) in enumerate(
    results_df.iterrows(),
    start=1,
):

    add_record_detail(
        row=record_row,

        record_number=record_number,

        total_records=len(
            results_df
        ),
    )


# Conclusions and recommendations.
add_text_pages(
    "Conclusions and Recommended Next Steps",
    [
        "1. The 100-record run demonstrates whether the local Hugging Face",
        "   pipeline can generate consistent categorical fields across multiple",
        "   injury-severity levels.",
        "",
        "2. Distribution tables describe model assignments within the sample.",
        "   They should not be treated as population estimates until sampling",
        "   and classification performance have been validated.",
        "",
        "3. Every record marked REVIEW_REQUIRED should be manually evaluated.",
        "   A sample of records not marked for review should also be checked to",
        "   detect confident but incorrect classifications.",
        "",
        "4. Model confidence is not a calibrated probability. Human review,",
        "   evidence relevance, and factor-action alignment provide stronger",
        "   quality controls.",
        "",
        "5. After review, calculate cause accuracy, factor accuracy, action",
        "   appropriateness, macro-F1, confusion matrices, and error rates by",
        "   severity group.",
        "",
        "6. Review repeated use of OTHER, NOT_STATED, and INSUFFICIENT_INFORMATION",
        "   to identify missing taxonomy categories.",
        "",
        "7. Before processing the complete dataset, retain checkpointing, shorten",
        "   the prompt if possible, and consider optimized batch inference.",
        "",
        "8. Corrective-action categories remain preliminary analytical outputs.",
        "   Final actions require qualified operational, engineering, and safety",
        "   review.",
    ],
    font_size=8.8,
)


# PDF metadata.
metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "MSHA 100-Narrative Severity-Stratified "
    "Cause, Factor, and Action Analysis"
)

metadata[
    "Author"
] = "Python and Qwen2.5-7B-Instruct"

metadata[
    "Subject"
] = (
    "LLM extraction of reported causes, contributing factors, "
    "and recommended-action categories"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 21. FINAL CONSOLE SUMMARY
# =============================================================================

successful_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq("SUCCESS")
    .sum()
)


failed_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq("FAILED")
    .sum()
)


review_count = int(
    results_df[
        "REVIEW_REQUIRED"
    ]
    .fillna(True)
    .sum()
)


print("\n" + "=" * 90)
print("100-NARRATIVE SEVERITY ANALYSIS COMPLETE")
print("=" * 90)

print("Model:", MODEL_ID)

print("GPU:", GPU_NAME)

print(
    "Records processed:",
    len(results_df),
)

print(
    "Successful classifications:",
    successful_count,
)

print(
    "Failed classifications:",
    failed_count,
)

print(
    "Records requiring review:",
    review_count,
)

print(
    "PDF report:",
    os.path.abspath(
        OUTPUT_PDF
    ),
)

print(
    "Checkpoint:",
    os.path.abspath(
        CHECKPOINT_JSONL
    ),
)

print("=" * 90)


# Display a concise notebook summary.
display(
    sampling_summary
)


display(
    quality_summary
)


display(
    cause_summary
)


display(
    factor_summary
)


display(
    action_summary
)
````

## Module 3 Discussion
### TF-IDF plus MiniBatch K-means clustering

````python
# =============================================================================
# MSHA NARRATIVE CLUSTERING PROOF OF CONCEPT - UPDATED VERSION
#
# METHOD:
#   TF-IDF vocabulary
#       -> Truncated SVD / latent semantic representation
#       -> MiniBatch K-means
#       -> Cluster interpretation
#       -> PDF report
#
# PREREQUISITE:
#   A pandas DataFrame named `accidents` is already loaded.
#
# PRIMARY OUTPUT:
#   MSHA_Narrative_Clustering_Proof_of_Concept_Updated.pdf
#
# SUPPORTING OUTPUT:
#   MSHA_Narrative_Cluster_Assignments_Updated.csv
#
# This is an exploratory clustering exercise. It does not replace the
# cause, contributing-factor, and recommended-action analysis.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import math
import textwrap
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_pdf import PdfPages
from matplotlib.ticker import MaxNLocator

from sklearn.feature_extraction.text import (
    TfidfVectorizer,
    ENGLISH_STOP_WORDS,
)

from sklearn.decomposition import TruncatedSVD
from sklearn.cluster import MiniBatchKMeans

from sklearn.metrics import (
    silhouette_score,
    silhouette_samples,
)

from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline


warnings.filterwarnings("ignore")


# =============================================================================
# 2. USER SETTINGS
# =============================================================================

TEXT_COLUMN = "NARRATIVE"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

# Balanced proof-of-concept sample.
TOTAL_SAMPLE_SIZE = 2000

TARGET_PER_SEVERITY_GROUP = 400

RANDOM_STATE = 42

# Nine clusters were selected because the earlier 9- and 15-cluster
# silhouette scores were effectively the same, while nine clusters are
# easier to interpret.
FINAL_CLUSTER_COUNT = 9

# Candidate solutions are still evaluated for comparison in the report.
CLUSTER_COUNTS_TO_TEST = [
    5,
    6,
    7,
    8,
    9,
    10,
    12,
    15,
]

# Vocabulary rules.
MIN_DOCUMENT_FREQUENCY = 3

MAX_DOCUMENT_FREQUENCY = 0.70

MAX_VOCABULARY_SIZE = 8000

NGRAM_RANGE = (1, 2)

# Latent semantic dimensions.
SVD_COMPONENTS = 75

# Report contents.
TOP_TERMS_PER_CLUSTER = 15

REPRESENTATIVE_NARRATIVES_PER_CLUSTER = 5

BORDERLINE_NARRATIVES_PER_CLUSTER = 2

OUTPUT_PDF = (
    "MSHA_Narrative_Clustering_Proof_of_Concept_Updated.pdf"
)

OUTPUT_CSV = (
    "MSHA_Narrative_Cluster_Assignments_Updated.csv"
)


# =============================================================================
# 3. VALIDATE THE DATAFRAME
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"Column {TEXT_COLUMN!r} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"Column {SEVERITY_COLUMN!r} was not found."
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


# Exclude blank and extremely short narratives.
df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN]
    .str.len()
)


df = df[
    df["_WORD_COUNT"] >= 8
].copy()


print("=" * 90)
print("MSHA NARRATIVE CLUSTERING - UPDATED PROOF OF CONCEPT")
print("=" * 90)

print(
    f"Eligible narratives after filtering: "
    f"{len(df):,}"
)


# =============================================================================
# 4. ASSIGN INJURY-SEVERITY GROUPS
# =============================================================================

def assign_severity_group(
    degree_injury,
):

    value = (
        str(degree_injury)
        .strip()
        .upper()
    )

    if "FATAL" in value:

        return "Fatality"

    if any(
        phrase in value
        for phrase in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "Permanent disability"

    if any(
        phrase in value
        for phrase in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "Days away"

    if any(
        phrase in value
        for phrase in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "Restricted or transferred"

    return "No lost time or other"


df["SEVERITY_GROUP"] = (
    df[SEVERITY_COLUMN]
    .map(assign_severity_group)
)


SEVERITY_ORDER = [
    "Fatality",
    "Permanent disability",
    "Days away",
    "Restricted or transferred",
    "No lost time or other",
]


# =============================================================================
# 5. CREATE A BALANCED SAMPLE
# =============================================================================

def create_balanced_sample(
    data,
    severity_order,
    target_per_group,
    total_sample_size,
    random_state,
):

    selected_frames = []

    selected_indexes = set()

    sample_records = []

    for group_number, severity_group in enumerate(
        severity_order
    ):

        group_data = data[
            data["SEVERITY_GROUP"]
            == severity_group
        ]

        available_count = len(
            group_data
        )

        selected_count = min(
            target_per_group,
            available_count,
        )

        if selected_count > 0:

            sampled_group = (
                group_data.sample(
                    n=selected_count,
                    random_state=(
                        random_state
                        + group_number
                    ),
                )
            )

            selected_frames.append(
                sampled_group
            )

            selected_indexes.update(
                sampled_group.index.tolist()
            )

        sample_records.append(
            {
                "Severity group":
                    severity_group,

                "Available records":
                    available_count,

                "Target":
                    target_per_group,

                "Selected":
                    selected_count,
            }
        )

    if selected_frames:

        sample = pd.concat(
            selected_frames,
            axis=0,
        )

    else:

        sample = data.iloc[
            0:0
        ].copy()

    remaining_needed = (
        total_sample_size
        - len(sample)
    )

    if remaining_needed > 0:

        remaining_pool = data.drop(
            index=list(
                selected_indexes
            ),
            errors="ignore",
        )

        additional_count = min(
            remaining_needed,
            len(remaining_pool),
        )

        if additional_count > 0:

            additional_sample = (
                remaining_pool.sample(
                    n=additional_count,
                    random_state=(
                        random_state
                        + 1000
                    ),
                )
            )

            sample = pd.concat(
                [
                    sample,
                    additional_sample,
                ],
                axis=0,
            )

    sample = (
        sample.sample(
            frac=1,
            random_state=random_state,
        )
        .reset_index(drop=False)
        .rename(
            columns={
                "index":
                    "_ORIGINAL_INDEX"
            }
        )
    )

    sampling_summary = pd.DataFrame(
        sample_records
    )

    final_counts = (
        sample[
            "SEVERITY_GROUP"
        ]
        .value_counts()
    )

    sampling_summary[
        "Final selected"
    ] = (
        sampling_summary[
            "Severity group"
        ]
        .map(final_counts)
        .fillna(0)
        .astype(int)
    )

    return sample, sampling_summary


sample_df, sampling_summary = (
    create_balanced_sample(
        data=df,

        severity_order=(
            SEVERITY_ORDER
        ),

        target_per_group=(
            TARGET_PER_SEVERITY_GROUP
        ),

        total_sample_size=(
            TOTAL_SAMPLE_SIZE
        ),

        random_state=RANDOM_STATE,
    )
)


print(
    f"Narratives selected: "
    f"{len(sample_df):,}"
)

print("\nSeverity sample:")

print(
    sampling_summary.to_string(
        index=False
    )
)


# =============================================================================
# 6. CONSERVATIVE NARRATIVE NORMALIZATION
#
# The original narrative remains unchanged.
# Only the CLEAN_NARRATIVE field is used for vectorization.
# =============================================================================

def normalize_narrative(
    narrative,
):

    text = str(
        narrative
    ).lower()

    # Employee references.
    text = re.sub(
        r"\bie\b",
        " injured_employee ",
        text,
    )

    text = re.sub(
        r"\bee\b",
        " employee ",
        text,
    )

    # Retain meaningful equivalent concepts.
    replacement_patterns = {
        r"\broof bolts?\b":
            " roof_bolt ",

        r"\broof bolters?\b":
            " roof_bolter ",

        r"\bbundle straps?\b":
            " material_restraint ",

        r"\bretaining bands?\b":
            " material_restraint ",

        r"\bhauling trucks?\b":
            " haul_truck ",

        r"\bhaul trucks?\b":
            " haul_truck ",

        r"\bdump trucks?\b":
            " dump_truck ",

        r"\bfork lifts?\b":
            " forklift ",

        r"\bcaught between\b":
            " caught_between ",

        r"\bpulled into\b":
            " pulled_into ",

        r"\bline of fire\b":
            " line_of_fire ",

        r"\bstored energy\b":
            " stored_energy ",

        r"\bhearing losses\b":
            " hearing_loss ",

        r"\bhearing loss\b":
            " hearing_loss ",

        r"\broof falls\b":
            " roof_fall ",

        r"\broof fall\b":
            " roof_fall ",

        r"\bhigh wall\b":
            " highwall ",

        r"\bfront end loader\b":
            " front_end_loader ",

        r"\bfront-end loader\b":
            " front_end_loader ",
    }

    for pattern, replacement in (
        replacement_patterns.items()
    ):

        text = re.sub(
            pattern,
            replacement,
            text,
        )

    # Remove dates.
    text = re.sub(
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",
        " ",
        text,
    )

    # Remove remaining numbers while preserving normalized concepts.
    text = re.sub(
        r"\b\d+\b",
        " ",
        text,
    )

    # Retain letters, underscores, apostrophes, and spaces.
    text = re.sub(
        r"[^a-z_'\s]",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


sample_df[
    "CLEAN_NARRATIVE"
] = (
    sample_df[
        TEXT_COLUMN
    ]
    .map(normalize_narrative)
)


# =============================================================================
# 7. IMPROVED STOP-WORD LIST
#
# Standard English words plus common MSHA narrative words are removed.
# These words were responsible for unclear cluster names such as
# "the / to / was" and "he / his / the."
# =============================================================================

DOMAIN_STOP_WORDS = {
    "accident",
    "accidents",
    "approximately",
    "area",
    "areas",
    "company",
    "continued",
    "day",
    "days",
    "employee",
    "employees",
    "injured",
    "injured_employee",
    "injuries",
    "injury",
    "incident",
    "incidents",
    "mine",
    "miner",
    "miners",
    "mining",
    "occurred",
    "operator",
    "person",
    "personnel",
    "reported",
    "resulted",
    "resulting",
    "said",
    "shift",
    "stated",
    "time",
    "victim",
    "work",
    "worker",
    "workers",
    "working",
    "received",
    "sustained",
    "taken",
    "hospital",
    "medical",
    "doctor",
    "approximately",
    "following",
    "prior",
    "later",
    "initially",
    "currently",
    "unknown",
    "nec",
    "nc",
}


CUSTOM_STOP_WORDS = sorted(
    set(
        ENGLISH_STOP_WORDS
    )
    .union(
        DOMAIN_STOP_WORDS
    )
)


# Terms excluded only from cluster naming.
# These terms may remain in the vector representation if not already removed.
LABEL_EXCLUSION_TERMS = {
    "employee",
    "injured employee",
    "worker",
    "miner",
    "victim",
    "person",
    "personnel",
    "reported",
    "incident",
    "accident",
    "mine",
    "mining",
    "work",
    "working",
    "occurred",
    "resulted",
    "resulting",
    "received",
    "sustained",
    "approximately",
    "unknown",
    "injury",
    "injuries",
    "medical",
    "hospital",
    "doctor",
    "shift",
}


# =============================================================================
# 8. TF-IDF VECTORIZATION
#
# Goldilocks rules:
#   - term must appear in at least three narratives;
#   - term cannot occur in more than 70% of narratives;
#   - retain one- and two-word terms;
#   - remove English and domain boilerplate.
# =============================================================================

vectorizer = TfidfVectorizer(
    stop_words=(
        CUSTOM_STOP_WORDS
    ),

    min_df=(
        MIN_DOCUMENT_FREQUENCY
    ),

    max_df=(
        MAX_DOCUMENT_FREQUENCY
    ),

    max_features=(
        MAX_VOCABULARY_SIZE
    ),

    ngram_range=(
        NGRAM_RANGE
    ),

    sublinear_tf=True,

    norm="l2",

    token_pattern=(
        r"(?u)\b[a-zA-Z_]"
        r"[a-zA-Z_']+\b"
    ),
)


tfidf_matrix = (
    vectorizer.fit_transform(
        sample_df[
            "CLEAN_NARRATIVE"
        ]
    )
)


feature_names = np.array(
    vectorizer
    .get_feature_names_out()
)


print(
    f"\nVocabulary terms retained: "
    f"{len(feature_names):,}"
)


# =============================================================================
# 9. LATENT SEMANTIC REPRESENTATION
# =============================================================================

maximum_components = max(
    2,
    min(
        SVD_COMPONENTS,
        tfidf_matrix.shape[0] - 1,
        tfidf_matrix.shape[1] - 1,
    ),
)


svd_pipeline = make_pipeline(
    TruncatedSVD(
        n_components=(
            maximum_components
        ),

        random_state=(
            RANDOM_STATE
        ),
    ),

    Normalizer(
        copy=False
    ),
)


semantic_matrix = (
    svd_pipeline.fit_transform(
        tfidf_matrix
    )
)


explained_variance = float(
    svd_pipeline[
        "truncatedsvd"
    ]
    .explained_variance_ratio_
    .sum()
)


# =============================================================================
# 10. EVALUATE CANDIDATE CLUSTER COUNTS
# =============================================================================

cluster_test_records = []

fitted_models = {}


for cluster_count in (
    CLUSTER_COUNTS_TO_TEST
):

    if cluster_count >= len(
        sample_df
    ):

        continue

    cluster_model = (
        MiniBatchKMeans(
            n_clusters=cluster_count,

            random_state=(
                RANDOM_STATE
            ),

            batch_size=256,

            n_init=20,

            max_iter=400,

            reassignment_ratio=0.01,
        )
    )

    labels = cluster_model.fit_predict(
        semantic_matrix
    )

    if len(
        np.unique(labels)
    ) < 2:

        continue

    silhouette = silhouette_score(
        semantic_matrix,

        labels,

        metric="euclidean",

        sample_size=min(
            1500,
            len(sample_df),
        ),

        random_state=(
            RANDOM_STATE
        ),
    )

    cluster_sizes = (
        pd.Series(labels)
        .value_counts()
    )

    cluster_test_records.append(
        {
            "Number of clusters":
                cluster_count,

            "Silhouette score":
                float(
                    silhouette
                ),

            "Smallest cluster":
                int(
                    cluster_sizes.min()
                ),

            "Largest cluster":
                int(
                    cluster_sizes.max()
                ),

            "Median cluster size":
                float(
                    cluster_sizes.median()
                ),
        }
    )

    fitted_models[
        cluster_count
    ] = {
        "model":
            cluster_model,

        "labels":
            labels,

        "silhouette":
            float(
                silhouette
            ),
    }


cluster_tests_df = (
    pd.DataFrame(
        cluster_test_records
    )
    .sort_values(
        "Number of clusters"
    )
    .reset_index(
        drop=True
    )
)


if (
    FINAL_CLUSTER_COUNT
    not in fitted_models
):

    raise RuntimeError(
        f"The requested {FINAL_CLUSTER_COUNT}-cluster solution "
        "was not generated."
    )


best_tested_row = (
    cluster_tests_df
    .sort_values(
        "Silhouette score",
        ascending=False,
    )
    .iloc[0]
)


HIGHEST_TESTED_CLUSTER_COUNT = int(
    best_tested_row[
        "Number of clusters"
    ]
)


HIGHEST_TESTED_SILHOUETTE = float(
    best_tested_row[
        "Silhouette score"
    ]
)


final_model = (
    fitted_models[
        FINAL_CLUSTER_COUNT
    ][
        "model"
    ]
)


final_labels = (
    fitted_models[
        FINAL_CLUSTER_COUNT
    ][
        "labels"
    ]
)


FINAL_SILHOUETTE = float(
    fitted_models[
        FINAL_CLUSTER_COUNT
    ][
        "silhouette"
    ]
)


sample_df[
    "CLUSTER_ID"
] = final_labels


# =============================================================================
# 11. DOCUMENT-LEVEL SILHOUETTE AND DISTANCE MEASURES
# =============================================================================

sample_silhouettes = silhouette_samples(
    semantic_matrix,

    final_labels,

    metric="euclidean",
)


sample_df[
    "SILHOUETTE_VALUE"
] = sample_silhouettes


distance_matrix = (
    final_model.transform(
        semantic_matrix
    )
)


sample_df[
    "DISTANCE_TO_CLUSTER_CENTER"
] = [
    distance_matrix[
        row_number,
        cluster_label,
    ]
    for row_number, cluster_label
    in enumerate(
        final_labels
    )
]


sample_df[
    "CLUSTER_REPRESENTATIVENESS"
] = (
    1.0
    / (
        1.0
        + sample_df[
            "DISTANCE_TO_CLUSTER_CENTER"
        ]
    )
)


sample_df[
    "BORDERLINE_CLUSTER_ASSIGNMENT"
] = (
    sample_df[
        "SILHOUETTE_VALUE"
    ]
    < 0
)


# =============================================================================
# 12. EXTRACT CLUSTER TERMS
# =============================================================================

cluster_term_records = []

cluster_term_lists = {}


for cluster_id in sorted(
    sample_df[
        "CLUSTER_ID"
    ].unique()
):

    cluster_mask = (
        sample_df[
            "CLUSTER_ID"
        ].to_numpy()
        == cluster_id
    )

    cluster_tfidf = (
        tfidf_matrix[
            cluster_mask
        ]
    )

    mean_scores = np.asarray(
        cluster_tfidf.mean(
            axis=0
        )
    ).ravel()

    top_indexes = (
        mean_scores
        .argsort()[
            ::-1
        ]
    )

    selected_terms = []

    selected_term_records = []

    for feature_index in top_indexes:

        if mean_scores[
            feature_index
        ] <= 0:

            continue

        raw_term = str(
            feature_names[
                feature_index
            ]
        )

        readable_term = (
            raw_term.replace(
                "_",
                " ",
            )
            .strip()
        )

        if not readable_term:

            continue

        selected_terms.append(
            readable_term
        )

        selected_term_records.append(
            {
                "Cluster ID":
                    int(
                        cluster_id
                    ),

                "Term":
                    readable_term,

                "Mean TF-IDF":
                    float(
                        mean_scores[
                            feature_index
                        ]
                    ),
            }
        )

        if len(
            selected_terms
        ) >= TOP_TERMS_PER_CLUSTER:

            break

    cluster_term_lists[
        cluster_id
    ] = selected_terms

    for term_rank, record in enumerate(
        selected_term_records,
        start=1,
    ):

        record[
            "Term rank"
        ] = term_rank

        cluster_term_records.append(
            record
        )


cluster_terms_df = pd.DataFrame(
    cluster_term_records
)


# =============================================================================
# 13. GENERATE IMPROVED CLUSTER LABELS
#
# Preference order:
#   1. meaningful two-word phrases;
#   2. distinctive one-word terms;
#   3. avoid duplicate words and generic labels.
# =============================================================================

def term_is_usable_for_label(
    term,
):

    normalized = str(
        term
    ).strip().lower()

    if not normalized:

        return False

    if normalized in (
        LABEL_EXCLUSION_TERMS
    ):

        return False

    words = normalized.split()

    if all(
        word in CUSTOM_STOP_WORDS
        for word in words
    ):

        return False

    return True


def terms_are_redundant(
    candidate,
    selected_terms,
):

    candidate_words = set(
        candidate.split()
    )

    for selected in selected_terms:

        selected_words = set(
            selected.split()
        )

        if candidate_words == selected_words:

            return True

        if (
            candidate_words
            .issubset(
                selected_words
            )
            or selected_words
            .issubset(
                candidate_words
            )
        ):

            return True

    return False


def create_cluster_label(
    terms,
    cluster_id,
):

    usable_terms = [
        term
        for term in terms
        if term_is_usable_for_label(
            term
        )
    ]

    selected = []

    # Prefer meaningful multiword phrases.
    for term in usable_terms:

        if len(
            term.split()
        ) < 2:

            continue

        if not terms_are_redundant(
            term,
            selected,
        ):

            selected.append(
                term
            )

        if len(selected) >= 2:

            break

    # Fill remaining positions with distinctive single terms.
    for term in usable_terms:

        if not terms_are_redundant(
            term,
            selected,
        ):

            selected.append(
                term
            )

        if len(selected) >= 3:

            break

    if not selected:

        return f"Cluster {cluster_id}"

    return " | ".join(
        selected[
            :3
        ]
    )


cluster_names = {
    cluster_id:
        create_cluster_label(
            cluster_term_lists[
                cluster_id
            ],

            cluster_id,
        )

    for cluster_id in (
        cluster_term_lists
    )
}


sample_df[
    "CLUSTER_NAME"
] = (
    sample_df[
        "CLUSTER_ID"
    ]
    .map(
        cluster_names
    )
)


cluster_terms_df[
    "Cluster name"
] = (
    cluster_terms_df[
        "Cluster ID"
    ]
    .map(
        cluster_names
    )
)


# =============================================================================
# 14. CREATE CLUSTER SUMMARIES
# =============================================================================

cluster_summary = (
    sample_df.groupby(
        [
            "CLUSTER_ID",
            "CLUSTER_NAME",
        ],
        observed=True,
    )
    .agg(
        NARRATIVES=(
            id_column,
            "count",
        ),

        MEDIAN_WORD_COUNT=(
            "_WORD_COUNT",
            "median",
        ),

        MEAN_SILHOUETTE=(
            "SILHOUETTE_VALUE",
            "mean",
        ),

        NEGATIVE_SILHOUETTE_COUNT=(
            "BORDERLINE_CLUSTER_ASSIGNMENT",
            "sum",
        ),

        MEAN_REPRESENTATIVENESS=(
            "CLUSTER_REPRESENTATIVENESS",
            "mean",
        ),
    )
    .reset_index()
)


cluster_summary[
    "NEGATIVE_SILHOUETTE_PCT"
] = (
    cluster_summary[
        "NEGATIVE_SILHOUETTE_COUNT"
    ]
    / cluster_summary[
        "NARRATIVES"
    ]
)


cluster_summary = cluster_summary.sort_values(
    "NARRATIVES",
    ascending=False,
).reset_index(drop=True)


severity_by_cluster = pd.crosstab(
    sample_df[
        "CLUSTER_NAME"
    ],

    sample_df[
        "SEVERITY_GROUP"
    ],
)


severity_pct_by_cluster = pd.crosstab(
    sample_df[
        "CLUSTER_NAME"
    ],

    sample_df[
        "SEVERITY_GROUP"
    ],

    normalize="index",
) * 100


# =============================================================================
# 15. POST-HOC COMPARISON WITH EXISTING FIELDS
#
# Existing MSHA fields were not used to form the clusters.
# =============================================================================

POSTHOC_FIELDS = [
    column
    for column in [
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
    ]
    if column in sample_df.columns
]


posthoc_records = []


for cluster_id in sorted(
    sample_df[
        "CLUSTER_ID"
    ].unique()
):

    cluster_records = sample_df[
        sample_df[
            "CLUSTER_ID"
        ]
        == cluster_id
    ]

    for field in POSTHOC_FIELDS:

        field_counts = (
            cluster_records[
                field
            ]
            .astype("string")
            .fillna(
                "<MISSING>"
            )
            .value_counts()
            .head(3)
        )

        for rank, (
            field_value,
            field_count,
        ) in enumerate(
            field_counts.items(),
            start=1,
        ):

            posthoc_records.append(
                {
                    "Cluster ID":
                        int(
                            cluster_id
                        ),

                    "Cluster name":
                        cluster_names[
                            cluster_id
                        ],

                    "Existing field":
                        field,

                    "Rank":
                        rank,

                    "Existing value":
                        str(
                            field_value
                        ),

                    "Count":
                        int(
                            field_count
                        ),

                    "Cluster percent":
                        (
                            100.0
                            * field_count
                            / len(
                                cluster_records
                            )
                        ),
                }
            )


posthoc_summary_df = pd.DataFrame(
    posthoc_records
)


# =============================================================================
# 16. REPRESENTATIVE AND BORDERLINE NARRATIVES
# =============================================================================

representative_records = []

borderline_records = []


for cluster_id in sorted(
    sample_df[
        "CLUSTER_ID"
    ].unique()
):

    cluster_records = (
        sample_df[
            sample_df[
                "CLUSTER_ID"
            ]
            == cluster_id
        ]
        .copy()
    )

    # Most representative records are closest to the cluster center.
    representatives = (
        cluster_records
        .sort_values(
            [
                "DISTANCE_TO_CLUSTER_CENTER",
                "SILHOUETTE_VALUE",
            ],
            ascending=[
                True,
                False,
            ],
        )
        .head(
            REPRESENTATIVE_NARRATIVES_PER_CLUSTER
        )
    )

    # Borderline records have the lowest document-level silhouette values.
    borderline = (
        cluster_records
        .sort_values(
            [
                "SILHOUETTE_VALUE",
                "DISTANCE_TO_CLUSTER_CENTER",
            ],
            ascending=[
                True,
                False,
            ],
        )
        .head(
            BORDERLINE_NARRATIVES_PER_CLUSTER
        )
    )

    for rank, (
        row_index,
        row,
    ) in enumerate(
        representatives.iterrows(),
        start=1,
    ):

        representative_records.append(
            {
                "Cluster ID":
                    int(
                        cluster_id
                    ),

                "Cluster name":
                    cluster_names[
                        cluster_id
                    ],

                "Rank":
                    rank,

                "Record ID":
                    str(
                        row[
                            id_column
                        ]
                    ),

                "Severity group":
                    row[
                        "SEVERITY_GROUP"
                    ],

                "Distance":
                    float(
                        row[
                            "DISTANCE_TO_CLUSTER_CENTER"
                        ]
                    ),

                "Silhouette":
                    float(
                        row[
                            "SILHOUETTE_VALUE"
                        ]
                    ),

                "Narrative":
                    str(
                        row[
                            TEXT_COLUMN
                        ]
                    ),
            }
        )

    for rank, (
        row_index,
        row,
    ) in enumerate(
        borderline.iterrows(),
        start=1,
    ):

        borderline_records.append(
            {
                "Cluster ID":
                    int(
                        cluster_id
                    ),

                "Cluster name":
                    cluster_names[
                        cluster_id
                    ],

                "Rank":
                    rank,

                "Record ID":
                    str(
                        row[
                            id_column
                        ]
                    ),

                "Severity group":
                    row[
                        "SEVERITY_GROUP"
                    ],

                "Distance":
                    float(
                        row[
                            "DISTANCE_TO_CLUSTER_CENTER"
                        ]
                    ),

                "Silhouette":
                    float(
                        row[
                            "SILHOUETTE_VALUE"
                        ]
                    ),

                "Narrative":
                    str(
                        row[
                            TEXT_COLUMN
                        ]
                    ),
            }
        )


representative_df = pd.DataFrame(
    representative_records
)


borderline_df = pd.DataFrame(
    borderline_records
)


# =============================================================================
# 17. TWO-DIMENSIONAL VISUAL REPRESENTATION
#
# Clustering uses the 75-dimensional semantic representation.
# The two-dimensional coordinates are used only for visualization.
# =============================================================================

visual_svd = TruncatedSVD(
    n_components=2,

    random_state=RANDOM_STATE,
)


visual_coordinates = (
    visual_svd.fit_transform(
        tfidf_matrix
    )
)


sample_df[
    "VISUAL_X"
] = visual_coordinates[
    :,
    0
]


sample_df[
    "VISUAL_Y"
] = visual_coordinates[
    :,
    1
]


visual_cluster_centers = (
    sample_df.groupby(
        "CLUSTER_ID"
    )[
        [
            "VISUAL_X",
            "VISUAL_Y",
        ]
    ]
    .mean()
)


# =============================================================================
# 18. CANDIDATE VOCABULARY
#
# Exclude boilerplate and rank useful corpus-level words and phrases.
# =============================================================================

global_tfidf_scores = np.asarray(
    tfidf_matrix.mean(
        axis=0
    )
).ravel()


global_document_counts = np.asarray(
    (
        tfidf_matrix > 0
    ).sum(
        axis=0
    )
).ravel()


candidate_records = []


for feature_index, feature_name in enumerate(
    feature_names
):

    readable_term = (
        str(
            feature_name
        )
        .replace(
            "_",
            " ",
        )
        .strip()
    )

    if not term_is_usable_for_label(
        readable_term
    ):

        continue

    candidate_records.append(
        {
            "Term or phrase":
                readable_term,

            "Documents":
                int(
                    global_document_counts[
                        feature_index
                    ]
                ),

            "Document prevalence":
                float(
                    global_document_counts[
                        feature_index
                    ]
                    / len(
                        sample_df
                    )
                ),

            "Mean TF-IDF":
                float(
                    global_tfidf_scores[
                        feature_index
                    ]
                ),

            "Phrase length":
                len(
                    readable_term.split()
                ),
        }
    )


vocabulary_candidates = (
    pd.DataFrame(
        candidate_records
    )
)


# Give a modest preference to multiword phrases.
vocabulary_candidates[
    "VOCABULARY_SCORE"
] = (
    vocabulary_candidates[
        "Mean TF-IDF"
    ]
    * (
        1
        + 0.30
        * (
            vocabulary_candidates[
                "Phrase length"
            ]
            - 1
        )
    )
)


vocabulary_candidates = (
    vocabulary_candidates
    .sort_values(
        [
            "VOCABULARY_SCORE",
            "Documents",
        ],
        ascending=[
            False,
            False,
        ],
    )
    .head(60)
    .reset_index(drop=True)
)


# =============================================================================
# 19. SAVE DOCUMENT-LEVEL CLUSTER ASSIGNMENTS
# =============================================================================

output_columns = [
    column
    for column in [
        id_column,
        TEXT_COLUMN,
        SEVERITY_COLUMN,
        "SEVERITY_GROUP",
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "CLUSTER_ID",
        "CLUSTER_NAME",
        "SILHOUETTE_VALUE",
        "DISTANCE_TO_CLUSTER_CENTER",
        "CLUSTER_REPRESENTATIVENESS",
        "BORDERLINE_CLUSTER_ASSIGNMENT",
    ]
    if column in sample_df.columns
]


sample_df[
    output_columns
].to_csv(
    OUTPUT_CSV,

    index=False,

    encoding="utf-8-sig",
)


# =============================================================================
# 20. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


PAGE_WIDTH = 8.5

PAGE_HEIGHT = 11


def create_page(
    title=None,
    landscape=False,
):

    figsize = (
        (11, 8.5)
        if landscape
        else (8.5, 11)
    )

    figure = plt.figure(
        figsize=figsize
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def add_page_number(
    figure,
    page_number,
):

    figure.text(
        0.5,
        0.018,
        f"Page {page_number}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )


page_counter = {
    "value":
        0
}


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    add_page_number(
        figure,
        page_counter[
            "value"
        ],
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def wrap_lines(
    lines,
    width=108,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append("")

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(line),
                    width=width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    return wrapped


def add_text_pages(
    title,
    lines,
    font_size=8.6,
    lines_per_page=55,
    wrap_width=106,
):

    wrapped_lines = wrap_lines(
        lines,
        width=wrap_width,
    )

    chunks = [
        wrapped_lines[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(
                wrapped_lines
            ),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=22,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    printable.columns = [
        str(column)
        .replace(
            "_",
            " "
        )
        .title()
        for column in printable.columns
    ]

    for column in printable.columns:

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    ""
                    if pd.isna(value)
                    else str(value)
            )
        )

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=(
                                column_wrap_width
                            ),
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(printable),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        page_title = (
            title
            if start == 0
            else title
            + " - Continued"
        )

        figure = create_page(
            page_title,
            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=(
                chunk.values
            ),

            colLabels=(
                chunk.columns
            ),

            cellLoc="left",

            colLoc="left",

            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.45,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_heatmap(
    title,
    dataframe,
    x_label,
    y_label,
    percent=False,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(11, 8)
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                dataframe.columns
            )
        )
    )

    axis.set_xticklabels(
        dataframe.columns,
        rotation=30,
        ha="right",
    )

    axis.set_yticks(
        range(
            len(
                dataframe.index
            )
        )
    )

    axis.set_yticklabels(
        [
            textwrap.shorten(
                str(value),
                width=56,
                placeholder="...",
            )
            for value in (
                dataframe.index
            )
        ]
    )

    axis.set_xlabel(
        x_label
    )

    axis.set_ylabel(
        y_label
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = matrix[
                row_number,
                column_number,
            ]

            if value > 0:

                label = (
                    f"{value:.0f}%"
                    if percent
                    else str(
                        int(value)
                    )
                )

                axis.text(
                    column_number,
                    row_number,
                    label,
                    ha="center",
                    va="center",
                    fontsize=8,
                )

    figure.colorbar(
        image,
        ax=axis,
        label=(
            "Percent"
            if percent
            else "Narrative count"
        ),
    )

    figure.tight_layout()

    save_page(
        figure
    )


# =============================================================================
# 21. BUILD THE PDF REPORT
# =============================================================================

# -----------------------------------------------------------------------------
# Cover
# -----------------------------------------------------------------------------

figure = create_page()

figure.text(
    0.5,
    0.78,
    "MSHA Narrative Clustering Proof of Concept",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.72,
    "Updated Vocabulary and Nine-Cluster Analysis",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.675,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Narratives analyzed: {len(sample_df):,}",
    f"Vocabulary terms retained: {len(feature_names):,}",
    f"Clusters used: {FINAL_CLUSTER_COUNT}",
    f"Nine-cluster silhouette score: {FINAL_SILHOUETTE:.4f}",
    f"Highest tested silhouette score: {HIGHEST_TESTED_SILHOUETTE:.4f}",
    f"Highest-scoring tested cluster count: {HIGHEST_TESTED_CLUSTER_COUNT}",
    f"Latent semantic dimensions: {maximum_components}",
    f"Explained variance: {explained_variance:.3f}",
    "",
    "Purpose:",
    "Test whether meaningful narrative groups can be discovered using",
    "terms and phrases that are prevalent but not overly common.",
    "",
    "Interpretation:",
    "Clusters are exploratory text-similarity groups. They are not",
    "verified causes, official classifications, or final safety topics.",
]

figure.text(
    0.12,
    0.57,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=9.2,
    family="monospace",
    linespacing=1.25,
)

save_page(
    figure
)


# -----------------------------------------------------------------------------
# Executive summary
# -----------------------------------------------------------------------------

largest_cluster = (
    cluster_summary.iloc[0]
)


highest_cohesion_cluster = (
    cluster_summary
    .sort_values(
        "MEAN_SILHOUETTE",
        ascending=False,
    )
    .iloc[0]
)


borderline_count = int(
    sample_df[
        "BORDERLINE_CLUSTER_ASSIGNMENT"
    ].sum()
)


add_text_pages(
    "Executive Summary",
    [
        "This proof-of-concept analysis grouped 2,000 MSHA accident narratives",
        "using only the narrative text. Existing MSHA classification, accident",
        "type, activity, injury source, and injury severity fields were not used",
        "to create the clusters.",
        "",
        f"A nine-cluster solution was retained for interpretation. Its silhouette",
        f"score was {FINAL_SILHOUETTE:.4f}. The highest tested score was",
        f"{HIGHEST_TESTED_SILHOUETTE:.4f} using "
        f"{HIGHEST_TESTED_CLUSTER_COUNT} clusters.",
        "",
        "The difference between the nine-cluster and highest-scoring solutions",
        "was small. The nine-cluster solution was therefore selected as a",
        "simpler exploratory representation.",
        "",
        f"The largest cluster contained {int(largest_cluster['NARRATIVES'])}",
        f"narratives and was labeled '{largest_cluster['CLUSTER_NAME']}'.",
        "",
        f"The cluster with the highest mean document-level silhouette was",
        f"'{highest_cohesion_cluster['CLUSTER_NAME']}' with a mean value of",
        f"{highest_cohesion_cluster['MEAN_SILHOUETTE']:.3f}.",
        "",
        f"{borderline_count} narratives had negative silhouette values, meaning",
        "those narratives were closer to another cluster than to portions of",
        "their assigned cluster.",
        "",
        "The results should be treated as evidence that recurring vocabulary",
        "patterns exist, not as proof that the narratives form nine definitive",
        "accident categories.",
    ],
    font_size=8.8,
)


# -----------------------------------------------------------------------------
# Key interpretation
# -----------------------------------------------------------------------------

add_text_pages(
    "How to Interpret This Report",
    [
        "Cluster:",
        "A group of narratives with similar TF-IDF and latent text features.",
        "",
        "Cluster label:",
        "A provisional name generated from the highest-weighted meaningful",
        "terms and phrases. Generic employee references and common English",
        "words were excluded from the label.",
        "",
        "Representative narrative:",
        "A narrative located relatively close to its assigned cluster center.",
        "",
        "Borderline narrative:",
        "A narrative with a low document-level silhouette value. Borderline",
        "records may contain several topics or may fit another cluster nearly",
        "as well.",
        "",
        "Silhouette score:",
        "A measure of within-cluster similarity relative to separation from",
        "other clusters. Higher values indicate clearer separation. Values near",
        "zero indicate overlap, while negative values indicate questionable",
        "assignments.",
        "",
        "Existing fields:",
        "Official MSHA fields are shown only after clustering. They help assess",
        "whether each narrative cluster adds information or mainly recreates an",
        "existing category.",
    ],
    font_size=8.7,
)


# -----------------------------------------------------------------------------
# Methodology
# -----------------------------------------------------------------------------

add_text_pages(
    "Methodology",
    [
        "1. A balanced sample was selected across five injury-severity groups.",
        "",
        "2. Common abbreviations and selected equivalent phrases were normalized.",
        "   Aggressive stemming and phrase destruction were avoided.",
        "",
        "3. Standard English stop words and common MSHA narrative boilerplate",
        "   were removed.",
        "",
        f"4. Terms had to occur in at least {MIN_DOCUMENT_FREQUENCY} narratives",
        f"   and could not occur in more than "
        f"{MAX_DOCUMENT_FREQUENCY:.0%} of the sample.",
        "",
        "5. Both individual words and two-word phrases were retained.",
        "",
        "6. TF-IDF assigned higher weight to terms that helped distinguish one",
        "   narrative from others.",
        "",
        f"7. Truncated SVD reduced the vocabulary to {maximum_components} latent",
        "   text dimensions.",
        "",
        "8. MiniBatch K-means grouped narratives in the reduced text space.",
        "",
        "9. Several cluster counts were evaluated. Nine clusters were retained",
        "   because the nine- and fifteen-cluster silhouette scores were nearly",
        "   identical in the initial test, while nine clusters were easier to",
        "   interpret.",
        "",
        "10. Existing MSHA fields were examined only after clustering.",
    ],
    font_size=8.6,
)


# -----------------------------------------------------------------------------
# Sampling
# -----------------------------------------------------------------------------

add_table_pages(
    "Severity-Stratified Sample",
    sampling_summary,
    rows_per_page=10,
    font_size=7.6,
    column_wrap_width=28,
)


# -----------------------------------------------------------------------------
# Cluster-count evaluation
# -----------------------------------------------------------------------------

cluster_tests_display = (
    cluster_tests_df.copy()
)


cluster_tests_display[
    "Silhouette score"
] = (
    cluster_tests_display[
        "Silhouette score"
    ]
    .map(
        lambda value:
            f"{value:.4f}"
    )
)


add_table_pages(
    "Cluster Count Evaluation",
    cluster_tests_display,
    rows_per_page=20,
    font_size=7.3,
    column_wrap_width=25,
)


figure = plt.figure(
    figsize=(8.5, 6)
)

axis = figure.add_subplot(
    111
)

axis.plot(
    cluster_tests_df[
        "Number of clusters"
    ],

    cluster_tests_df[
        "Silhouette score"
    ],

    marker="o",

    linewidth=1.8,
)

axis.axvline(
    FINAL_CLUSTER_COUNT,

    linestyle="--",

    color="#C0392B",

    label=(
        f"Selected solution: "
        f"{FINAL_CLUSTER_COUNT}"
    ),
)

axis.scatter(
    [
        FINAL_CLUSTER_COUNT
    ],

    [
        FINAL_SILHOUETTE
    ],

    color="#C0392B",

    s=55,

    zorder=5,
)

axis.set_title(
    "Cluster Count Comparison",
    fontsize=13,
    fontweight="bold",
)

axis.set_xlabel(
    "Number of clusters"
)

axis.set_ylabel(
    "Silhouette score"
)

axis.xaxis.set_major_locator(
    MaxNLocator(
        integer=True
    )
)

axis.grid(
    alpha=0.25
)

axis.legend()

figure.tight_layout()

save_page(
    figure
)


# -----------------------------------------------------------------------------
# Improved two-dimensional visualization
# -----------------------------------------------------------------------------

figure = plt.figure(
    figsize=(11, 8)
)

axis = figure.add_subplot(
    111
)

color_map = plt.get_cmap(
    "tab10",
    FINAL_CLUSTER_COUNT,
)


for cluster_id in sorted(
    sample_df[
        "CLUSTER_ID"
    ].unique()
):

    cluster_points = sample_df[
        sample_df[
            "CLUSTER_ID"
        ]
        == cluster_id
    ]

    axis.scatter(
        cluster_points[
            "VISUAL_X"
        ],

        cluster_points[
            "VISUAL_Y"
        ],

        s=20,

        alpha=0.48,

        color=color_map(
            cluster_id
        ),

        label=(
            f"{cluster_id}: "
            f"{textwrap.shorten(cluster_names[cluster_id], width=58)}"
        ),
    )

    center_x = visual_cluster_centers.loc[
        cluster_id,
        "VISUAL_X",
    ]

    center_y = visual_cluster_centers.loc[
        cluster_id,
        "VISUAL_Y",
    ]

    axis.scatter(
        center_x,
        center_y,
        s=180,
        color=color_map(
            cluster_id
        ),
        edgecolor="black",
        linewidth=1.3,
        marker="o",
        zorder=6,
    )

    axis.text(
        center_x,
        center_y,
        str(
            cluster_id
        ),
        ha="center",
        va="center",
        fontsize=9,
        fontweight="bold",
        color="white",
        zorder=7,
    )


axis.set_title(
    "Two-Dimensional View of Narrative Clusters",
    fontsize=13,
    fontweight="bold",
)

axis.set_xlabel(
    "Latent text dimension 1"
)

axis.set_ylabel(
    "Latent text dimension 2"
)

axis.grid(
    alpha=0.20
)

axis.legend(
    loc="upper center",

    bbox_to_anchor=(
        0.5,
        -0.11,
    ),

    ncol=2,

    fontsize=7.2,

    frameon=False,
)

figure.tight_layout()

save_page(
    figure
)


# -----------------------------------------------------------------------------
# Cluster quality
# -----------------------------------------------------------------------------

cluster_summary_display = (
    cluster_summary.copy()
)


cluster_summary_display[
    "MEAN_SILHOUETTE"
] = (
    cluster_summary_display[
        "MEAN_SILHOUETTE"
    ]
    .map(
        lambda value:
            f"{value:.3f}"
    )
)


cluster_summary_display[
    "NEGATIVE_SILHOUETTE_PCT"
] = (
    cluster_summary_display[
        "NEGATIVE_SILHOUETTE_PCT"
    ]
    .map(
        lambda value:
            f"{value:.1%}"
    )
)


cluster_summary_display[
    "MEAN_REPRESENTATIVENESS"
] = (
    cluster_summary_display[
        "MEAN_REPRESENTATIVENESS"
    ]
    .map(
        lambda value:
            f"{value:.3f}"
    )
)


add_table_pages(
    "Cluster Size and Quality Summary",
    cluster_summary_display,
    rows_per_page=15,
    font_size=6.6,
    column_wrap_width=27,
    landscape=True,
)


# Cluster-size chart.
figure = plt.figure(
    figsize=(10, 6.5)
)

axis = figure.add_subplot(
    111
)

plot_summary = (
    cluster_summary
    .sort_values(
        "NARRATIVES",
        ascending=True,
    )
)

axis.barh(
    [
        f"{cluster_id}: "
        f"{textwrap.shorten(cluster_name, width=44)}"
        for cluster_id, cluster_name
        in zip(
            plot_summary[
                "CLUSTER_ID"
            ],

            plot_summary[
                "CLUSTER_NAME"
            ],
        )
    ],

    plot_summary[
        "NARRATIVES"
    ],

    color="#2E6F9E",
)

axis.set_title(
    "Narratives per Cluster",
    fontsize=13,
    fontweight="bold",
)

axis.set_xlabel(
    "Narrative count"
)

axis.grid(
    axis="x",
    alpha=0.25,
)

figure.tight_layout()

save_page(
    figure
)


# Mean silhouette chart by cluster.
figure = plt.figure(
    figsize=(10, 6.5)
)

axis = figure.add_subplot(
    111
)

quality_plot = (
    cluster_summary
    .sort_values(
        "MEAN_SILHOUETTE",
        ascending=True,
    )
)

bar_colors = [
    (
        "#4E8B57"
        if value >= 0
        else "#C0392B"
    )
    for value in (
        quality_plot[
            "MEAN_SILHOUETTE"
        ]
    )
]

axis.barh(
    [
        f"{cluster_id}: "
        f"{textwrap.shorten(cluster_name, width=44)}"
        for cluster_id, cluster_name
        in zip(
            quality_plot[
                "CLUSTER_ID"
            ],

            quality_plot[
                "CLUSTER_NAME"
            ],
        )
    ],

    quality_plot[
        "MEAN_SILHOUETTE"
    ],

    color=bar_colors,
)

axis.axvline(
    0,
    color="black",
    linewidth=0.8,
)

axis.set_title(
    "Mean Document-Level Silhouette by Cluster",
    fontsize=13,
    fontweight="bold",
)

axis.set_xlabel(
    "Mean silhouette value"
)

axis.grid(
    axis="x",
    alpha=0.25,
)

figure.tight_layout()

save_page(
    figure
)


# -----------------------------------------------------------------------------
# Severity composition
# -----------------------------------------------------------------------------

add_heatmap(
    "Narrative Count by Cluster and Injury Severity",
    severity_by_cluster,
    "Injury-severity group",
    "Narrative cluster",
    percent=False,
)


add_heatmap(
    "Injury-Severity Distribution within Each Cluster",
    severity_pct_by_cluster,
    "Injury-severity group",
    "Narrative cluster",
    percent=True,
)


# -----------------------------------------------------------------------------
# Top terms by cluster
# -----------------------------------------------------------------------------

cluster_terms_display = (
    cluster_terms_df.copy()
)


cluster_terms_display[
    "Mean TF-IDF"
] = (
    cluster_terms_display[
        "Mean TF-IDF"
    ]
    .map(
        lambda value:
            f"{value:.4f}"
    )
)


cluster_terms_display = (
    cluster_terms_display[
        [
            "Cluster ID",
            "Cluster name",
            "Term rank",
            "Term",
            "Mean TF-IDF",
        ]
    ]
)


add_table_pages(
    "Top Terms and Phrases by Cluster",
    cluster_terms_display,
    rows_per_page=24,
    font_size=6.5,
    column_wrap_width=28,
    landscape=True,
)


# -----------------------------------------------------------------------------
# Post-hoc existing field comparison
# -----------------------------------------------------------------------------

add_text_pages(
    "Post-Hoc Comparison with Existing MSHA Fields",
    [
        "Existing MSHA fields were not used to form the clusters.",
        "",
        "The following pages show the three most common existing values within",
        "each cluster. This helps determine whether a text-derived cluster",
        "simply recreates an existing classification or connects records from",
        "several official categories.",
        "",
        "A cluster spanning several classifications may reveal a cross-cutting",
        "event, equipment, injury, or prevention pattern.",
    ],
    font_size=8.8,
)


posthoc_display = (
    posthoc_summary_df.copy()
)


posthoc_display[
    "Cluster percent"
] = (
    posthoc_display[
        "Cluster percent"
    ]
    .map(
        lambda value:
            f"{value:.1f}%"
    )
)


add_table_pages(
    "Most Common Existing Values within Each Cluster",
    posthoc_display,
    rows_per_page=22,
    font_size=6.1,
    column_wrap_width=28,
    landscape=True,
)


# -----------------------------------------------------------------------------
# Cluster interpretation pages
# -----------------------------------------------------------------------------

for cluster_id in sorted(
    sample_df[
        "CLUSTER_ID"
    ].unique()
):

    cluster_row = (
        cluster_summary[
            cluster_summary[
                "CLUSTER_ID"
            ]
            == cluster_id
        ]
        .iloc[0]
    )

    cluster_top_terms = (
        cluster_terms_df[
            cluster_terms_df[
                "Cluster ID"
            ]
            == cluster_id
        ]
        .sort_values(
            "Term rank"
        )
    )

    representatives = (
        representative_df[
            representative_df[
                "Cluster ID"
            ]
            == cluster_id
        ]
        .sort_values(
            "Rank"
        )
    )

    borderline = (
        borderline_df[
            borderline_df[
                "Cluster ID"
            ]
            == cluster_id
        ]
        .sort_values(
            "Rank"
        )
    )

    lines = [
        f"Cluster ID: {cluster_id}",
        f"Provisional label: {cluster_names[cluster_id]}",
        f"Narratives: {int(cluster_row['NARRATIVES'])}",
        f"Mean document silhouette: {cluster_row['MEAN_SILHOUETTE']:.3f}",
        (
            "Negative-silhouette narratives: "
            f"{int(cluster_row['NEGATIVE_SILHOUETTE_COUNT'])} "
            f"({cluster_row['NEGATIVE_SILHOUETTE_PCT']:.1%})"
        ),
        "",
        "TOP TERMS AND PHRASES",
        "-" * 80,
        ", ".join(
            cluster_top_terms[
                "Term"
            ]
            .head(
                TOP_TERMS_PER_CLUSTER
            )
            .tolist()
        ),
        "",
        "REPRESENTATIVE NARRATIVES",
        "-" * 80,
    ]

    for (
        row_index,
        row,
    ) in representatives.iterrows():

        lines.extend(
            [
                "",
                (
                    f"Representative {row['Rank']} | "
                    f"Record {row['Record ID']} | "
                    f"{row['Severity group']} | "
                    f"Silhouette {row['Silhouette']:.3f} | "
                    f"Distance {row['Distance']:.3f}"
                ),
                str(
                    row[
                        "Narrative"
                    ]
                ),
            ]
        )

    lines.extend(
        [
            "",
            "BORDERLINE NARRATIVES",
            "-" * 80,
        ]
    )

    for (
        row_index,
        row,
    ) in borderline.iterrows():

        lines.extend(
            [
                "",
                (
                    f"Borderline {row['Rank']} | "
                    f"Record {row['Record ID']} | "
                    f"{row['Severity group']} | "
                    f"Silhouette {row['Silhouette']:.3f} | "
                    f"Distance {row['Distance']:.3f}"
                ),
                str(
                    row[
                        "Narrative"
                    ]
                ),
            ]
        )

    lines.extend(
        [
            "",
            "INTERPRETATION NOTE",
            "-" * 80,
            (
                "The provisional label is generated from cluster vocabulary. "
                "The representative narratives should be used to decide "
                "whether the cluster reflects one coherent event pattern, "
                "several related patterns, or a mixed group."
            ),
        ]
    )

    add_text_pages(
        (
            f"Cluster {cluster_id}: "
            f"{cluster_names[cluster_id]}"
        ),

        lines,

        font_size=7.9,

        lines_per_page=52,

        wrap_width=104,
    )


# -----------------------------------------------------------------------------
# Improved vocabulary candidate table
# -----------------------------------------------------------------------------

vocabulary_display = (
    vocabulary_candidates.copy()
)


vocabulary_display[
    "Document prevalence"
] = (
    vocabulary_display[
        "Document prevalence"
    ]
    .map(
        lambda value:
            f"{value:.1%}"
    )
)


vocabulary_display[
    "Mean TF-IDF"
] = (
    vocabulary_display[
        "Mean TF-IDF"
    ]
    .map(
        lambda value:
            f"{value:.4f}"
    )
)


vocabulary_display[
    "VOCABULARY_SCORE"
] = (
    vocabulary_display[
        "VOCABULARY_SCORE"
    ]
    .map(
        lambda value:
            f"{value:.4f}"
    )
)


vocabulary_display = vocabulary_display[
    [
        "Term or phrase",
        "Documents",
        "Document prevalence",
        "Mean TF-IDF",
        "Phrase length",
        "VOCABULARY_SCORE",
    ]
]


add_text_pages(
    "Candidate Corpus Vocabulary",
    [
        "The candidate vocabulary excludes standard English stop words and",
        "common narrative boilerplate.",
        "",
        "Terms must occur in at least three documents, but not in more than",
        "70% of the sampled narratives.",
        "",
        "The vocabulary score gives a modest preference to two-word phrases",
        "because phrases often retain more operational context than isolated",
        "words.",
        "",
        "These terms are candidates for interpreting clusters. They are not",
        "automatically accepted as the final project vocabulary.",
    ],
    font_size=8.8,
)


add_table_pages(
    "Top Candidate Vocabulary Terms and Phrases",
    vocabulary_display,
    rows_per_page=25,
    font_size=6.8,
    column_wrap_width=27,
    landscape=True,
)


# -----------------------------------------------------------------------------
# Conclusions
# -----------------------------------------------------------------------------

weak_cluster_count = int(
    (
        cluster_summary[
            "MEAN_SILHOUETTE"
        ]
        <= 0
    ).sum()
)


add_text_pages(
    "Conclusions and Recommendations",
    [
        "1. The updated preprocessing reduces the influence of common English",
        "   words and generic employee references.",
        "",
        "2. Nine clusters provide a simpler exploratory structure than the",
        "   earlier fifteen-cluster solution while retaining a nearly equivalent",
        "   overall silhouette score.",
        "",
        "3. The clusters should be interpreted through top terms, representative",
        "   narratives, post-hoc MSHA fields, and severity distributions.",
        "",
        f"4. {weak_cluster_count} clusters had a mean silhouette value at or",
        "   below zero. Those clusters should be considered mixed or weakly",
        "   defined.",
        "",
        f"5. {borderline_count} individual narratives had negative silhouette",
        "   values. These narratives may contain multiple topics or fit another",
        "   cluster nearly as well.",
        "",
        "6. Clear groups such as hearing loss, roof falls, truck incidents,",
        "   hand/finger injuries, and ankle-twisting incidents provide evidence",
        "   that narrative text contains discoverable recurring patterns.",
        "",
        "7. Clusters based primarily on injury outcome or reporting language may",
        "   be less useful than clusters organized around event mechanisms,",
        "   hazards, equipment, or work activities.",
        "",
        "8. Existing MSHA categories were not used to create the clusters. The",
        "   post-hoc comparison should be used to identify clusters that connect",
        "   records across several official classifications.",
        "",
        "9. This remains a supporting proof of concept. The main final-project",
        "   direction remains the extraction of reported causes, contributing",
        "   factors, and preliminary corrective-action categories.",
    ],
    font_size=8.7,
)


# -----------------------------------------------------------------------------
# Technical appendix
# -----------------------------------------------------------------------------

add_text_pages(
    "Technical Appendix",
    [
        f"Input narratives: {len(sample_df):,}",
        f"TF-IDF vocabulary size: {len(feature_names):,}",
        f"Minimum document frequency: {MIN_DOCUMENT_FREQUENCY}",
        f"Maximum document frequency: {MAX_DOCUMENT_FREQUENCY:.0%}",
        f"N-gram range: {NGRAM_RANGE}",
        f"Maximum vocabulary size: {MAX_VOCABULARY_SIZE:,}",
        f"SVD components: {maximum_components}",
        f"SVD explained variance: {explained_variance:.4f}",
        f"Selected cluster count: {FINAL_CLUSTER_COUNT}",
        f"Selected silhouette score: {FINAL_SILHOUETTE:.4f}",
        f"Highest tested score: {HIGHEST_TESTED_SILHOUETTE:.4f}",
        f"Highest-scoring cluster count: {HIGHEST_TESTED_CLUSTER_COUNT}",
        f"Random state: {RANDOM_STATE}",
        "",
        f"PDF output: {os.path.abspath(OUTPUT_PDF)}",
        f"Cluster assignment CSV: {os.path.abspath(OUTPUT_CSV)}",
    ],
    font_size=8.5,
)


# =============================================================================
# 22. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "Updated MSHA Narrative Clustering "
    "Proof of Concept"
)

metadata[
    "Author"
] = (
    "Python TF-IDF, Truncated SVD, "
    "and MiniBatch K-means Analysis"
)

metadata[
    "Subject"
] = (
    "Exploratory vocabulary development "
    "and clustering of MSHA narratives"
)

metadata[
    "Keywords"
] = (
    "MSHA, TF-IDF, clustering, K-means, "
    "narrative analysis, vocabulary"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 23. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("UPDATED CLUSTERING PROOF OF CONCEPT COMPLETE")
print("=" * 90)

print(
    f"Narratives analyzed: "
    f"{len(sample_df):,}"
)

print(
    f"Vocabulary terms retained: "
    f"{len(feature_names):,}"
)

print(
    f"Selected clusters: "
    f"{FINAL_CLUSTER_COUNT}"
)

print(
    f"Selected silhouette score: "
    f"{FINAL_SILHOUETTE:.4f}"
)

print(
    f"Highest tested silhouette score: "
    f"{HIGHEST_TESTED_SILHOUETTE:.4f}"
)

print(
    f"Highest-scoring tested cluster count: "
    f"{HIGHEST_TESTED_CLUSTER_COUNT}"
)

print(
    f"Explained variance: "
    f"{explained_variance:.4f}"
)

print(
    f"Borderline narratives: "
    f"{borderline_count:,}"
)

print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print(
    f"Cluster assignment CSV: "
    f"{os.path.abspath(OUTPUT_CSV)}"
)

print("=" * 90)


# =============================================================================
# 24. NOTEBOOK DISPLAYS
# =============================================================================

display(
    sampling_summary
)


display(
    cluster_tests_df
)


display(
    cluster_summary
)


display(
    cluster_terms_df[
        [
            "Cluster ID",
            "Cluster name",
            "Term rank",
            "Term",
            "Mean TF-IDF",
        ]
    ]
    .head(
        FINAL_CLUSTER_COUNT
        * TOP_TERMS_PER_CLUSTER
    )
)


display(
    vocabulary_candidates.head(
        30
    )
)
````

## Module #4 Discussion- LDA

````python
# =============================================================================
# MSHA FINGER PINCH/CRUSH CLUSTER VS. LDA TOPIC
# CORRECTED FOCUSED COMPARISON
#
# CORRECTION:
#   VISUAL_X and VISUAL_Y are created in sample_df before agreement_df,
#   kmeans_only_df, lda_only_df, and background_df are created.
#
# INPUT:
#   A pandas DataFrame named `accidents` is already loaded.
#
# MODEL FITTING:
#   All 2,000 severity-balanced narratives are used to fit:
#     - shared vocabulary;
#     - TF-IDF / SVD / K-means model;
#     - LDA topic model.
#
# REPORTING:
#   The PDF focuses on:
#     - the selected finger pinch/crush K-means cluster;
#     - its closest LDA topic;
#     - Agreement records;
#     - K-means-only records;
#     - LDA-only candidate records;
#     - representative and borderline examples.
#
# FINAL OUTPUT:
#   MSHA_Finger_Cluster_Focused_KMeans_LDA_Report_Corrected.pdf
#
# No CSV or Excel output is created.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import math
import textwrap
import warnings
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_pdf import PdfPages

from sklearn.feature_extraction.text import (
    CountVectorizer,
    TfidfTransformer,
    ENGLISH_STOP_WORDS,
)

from sklearn.decomposition import (
    TruncatedSVD,
    LatentDirichletAllocation,
)

from sklearn.cluster import MiniBatchKMeans

from sklearn.metrics import (
    silhouette_score,
    silhouette_samples,
)

from sklearn.preprocessing import Normalizer
from sklearn.pipeline import make_pipeline


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

TEXT_COLUMN = "NARRATIVE"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

TOTAL_SAMPLE_SIZE = 2000

TARGET_PER_SEVERITY_GROUP = 400

RANDOM_STATE = 42

KMEANS_CLUSTER_COUNT = 9

LDA_TOPIC_COUNT = 9

SVD_COMPONENTS = 75

MIN_DOCUMENT_FREQUENCY = 3

MAX_DOCUMENT_FREQUENCY = 0.70

MAX_VOCABULARY_SIZE = 8000

NGRAM_RANGE = (1, 2)

LDA_MAX_ITERATIONS = 30

LDA_BATCH_SIZE = 128

# A narrative is associated with the matched LDA topic when:
#   1. the matched topic is the dominant topic; OR
#   2. the matched topic proportion is at least 0.50.
LDA_TOPIC_PROPORTION_THRESHOLD = 0.50

TOP_TERMS_TO_DISPLAY = 15

AGREEMENT_DOCUMENTS_TO_DISPLAY = 8

KMEANS_ONLY_DOCUMENTS_TO_DISPLAY = 8

LDA_ONLY_DOCUMENTS_TO_DISPLAY = 10

BORDERLINE_DOCUMENTS_TO_DISPLAY = 6

OUTPUT_PDF = (
    "MSHA_Finger_Cluster_Focused_KMeans_LDA_Report_Corrected.pdf"
)


# =============================================================================
# 3. VALIDATE INPUT DATA
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"Column {TEXT_COLUMN!r} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"Column {SEVERITY_COLUMN!r} was not found."
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN]
    .str.len()
)


df = df[
    df["_WORD_COUNT"] >= 8
].copy()


print("=" * 90)
print("FOCUSED K-MEANS FINGER CLUSTER VS. LDA TOPIC ANALYSIS")
print("=" * 90)

print(
    f"Eligible narratives: "
    f"{len(df):,}"
)


# =============================================================================
# 4. ASSIGN INJURY-SEVERITY GROUPS
# =============================================================================

def assign_severity_group(
    degree_injury,
):

    value = (
        str(degree_injury)
        .strip()
        .upper()
    )

    if "FATAL" in value:

        return "Fatality"

    if any(
        phrase in value
        for phrase in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "Permanent disability"

    if any(
        phrase in value
        for phrase in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "Days away"

    if any(
        phrase in value
        for phrase in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "Restricted or transferred"

    return "No lost time or other"


df["SEVERITY_GROUP"] = (
    df[SEVERITY_COLUMN]
    .map(assign_severity_group)
)


SEVERITY_ORDER = [
    "Fatality",
    "Permanent disability",
    "Days away",
    "Restricted or transferred",
    "No lost time or other",
]


# =============================================================================
# 5. CREATE THE BALANCED 2,000-NARRATIVE SAMPLE
# =============================================================================

def create_balanced_sample(
    data,
    severity_order,
    target_per_group,
    total_sample_size,
    random_state,
):

    selected_frames = []

    selected_indexes = set()

    sampling_records = []

    for group_number, severity_group in enumerate(
        severity_order
    ):

        group_data = data[
            data["SEVERITY_GROUP"]
            == severity_group
        ]

        available_count = len(
            group_data
        )

        selected_count = min(
            target_per_group,
            available_count,
        )

        if selected_count > 0:

            sampled_group = group_data.sample(
                n=selected_count,

                random_state=(
                    random_state
                    + group_number
                ),
            )

            selected_frames.append(
                sampled_group
            )

            selected_indexes.update(
                sampled_group.index.tolist()
            )

        sampling_records.append(
            {
                "Severity group":
                    severity_group,

                "Available records":
                    available_count,

                "Target":
                    target_per_group,

                "Initially selected":
                    selected_count,
            }
        )

    if selected_frames:

        sample = pd.concat(
            selected_frames,
            axis=0,
        )

    else:

        sample = data.iloc[
            0:0
        ].copy()

    remaining_needed = (
        total_sample_size
        - len(sample)
    )

    if remaining_needed > 0:

        remaining_pool = data.drop(
            index=list(
                selected_indexes
            ),
            errors="ignore",
        )

        additional_count = min(
            remaining_needed,
            len(remaining_pool),
        )

        if additional_count > 0:

            additional_sample = (
                remaining_pool.sample(
                    n=additional_count,

                    random_state=(
                        random_state
                        + 1000
                    ),
                )
            )

            sample = pd.concat(
                [
                    sample,
                    additional_sample,
                ],
                axis=0,
            )

    sample = (
        sample.sample(
            frac=1,

            random_state=random_state,
        )
        .reset_index(drop=False)
        .rename(
            columns={
                "index":
                    "_ORIGINAL_INDEX"
            }
        )
    )

    sampling_summary = pd.DataFrame(
        sampling_records
    )

    final_counts = (
        sample[
            "SEVERITY_GROUP"
        ]
        .value_counts()
    )

    sampling_summary[
        "Final selected"
    ] = (
        sampling_summary[
            "Severity group"
        ]
        .map(final_counts)
        .fillna(0)
        .astype(int)
    )

    return sample, sampling_summary


sample_df, sampling_summary = (
    create_balanced_sample(
        data=df,

        severity_order=SEVERITY_ORDER,

        target_per_group=(
            TARGET_PER_SEVERITY_GROUP
        ),

        total_sample_size=(
            TOTAL_SAMPLE_SIZE
        ),

        random_state=RANDOM_STATE,
    )
)


print(
    f"Narratives used for model fitting: "
    f"{len(sample_df):,}"
)


# =============================================================================
# 6. SYNCHRONIZED TEXT NORMALIZATION
# =============================================================================

def normalize_narrative(
    narrative,
):

    text = str(
        narrative
    ).lower()

    text = re.sub(
        r"\bie\b",
        " injured_employee ",
        text,
    )

    text = re.sub(
        r"\bee\b",
        " employee ",
        text,
    )

    replacements = {
        r"\broof bolts?\b":
            " roof_bolt ",

        r"\broof bolters?\b":
            " roof_bolter ",

        r"\bbundle straps?\b":
            " material_restraint ",

        r"\bretaining bands?\b":
            " material_restraint ",

        r"\bhauling trucks?\b":
            " haul_truck ",

        r"\bhaul trucks?\b":
            " haul_truck ",

        r"\bdump trucks?\b":
            " dump_truck ",

        r"\bfork lifts?\b":
            " forklift ",

        r"\bcaught between\b":
            " caught_between ",

        r"\bpulled into\b":
            " pulled_into ",

        r"\bline of fire\b":
            " line_of_fire ",

        r"\bstored energy\b":
            " stored_energy ",

        r"\bhearing losses\b":
            " hearing_loss ",

        r"\bhearing loss\b":
            " hearing_loss ",

        r"\broof falls\b":
            " roof_fall ",

        r"\broof fall\b":
            " roof_fall ",

        r"\bhigh wall\b":
            " highwall ",

        r"\bfront end loader\b":
            " front_end_loader ",

        r"\bfront-end loader\b":
            " front_end_loader ",
    }

    for pattern, replacement in (
        replacements.items()
    ):

        text = re.sub(
            pattern,
            replacement,
            text,
        )

    text = re.sub(
        r"\b\d{1,2}[/-]\d{1,2}[/-]\d{2,4}\b",
        " ",
        text,
    )

    text = re.sub(
        r"\b\d+\b",
        " ",
        text,
    )

    text = re.sub(
        r"[^a-z_'\s]",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


sample_df[
    "CLEAN_NARRATIVE"
] = (
    sample_df[
        TEXT_COLUMN
    ]
    .map(normalize_narrative)
)


# =============================================================================
# 7. SYNCHRONIZED STOP WORDS
# =============================================================================

DOMAIN_STOP_WORDS = {
    "accident",
    "accidents",
    "approximately",
    "area",
    "areas",
    "company",
    "continued",
    "day",
    "days",
    "employee",
    "employees",
    "injured",
    "injured_employee",
    "injuries",
    "injury",
    "incident",
    "incidents",
    "mine",
    "miner",
    "miners",
    "mining",
    "occurred",
    "operator",
    "person",
    "personnel",
    "reported",
    "resulted",
    "resulting",
    "said",
    "shift",
    "stated",
    "time",
    "victim",
    "work",
    "worker",
    "workers",
    "working",
    "received",
    "sustained",
    "taken",
    "hospital",
    "medical",
    "doctor",
    "following",
    "prior",
    "later",
    "initially",
    "currently",
    "unknown",
    "nec",
    "nc",
}


CUSTOM_STOP_WORDS = sorted(
    set(
        ENGLISH_STOP_WORDS
    )
    .union(
        DOMAIN_STOP_WORDS
    )
)


# =============================================================================
# 8. CREATE ONE SHARED VOCABULARY
# =============================================================================

count_vectorizer = CountVectorizer(
    stop_words=CUSTOM_STOP_WORDS,

    min_df=(
        MIN_DOCUMENT_FREQUENCY
    ),

    max_df=(
        MAX_DOCUMENT_FREQUENCY
    ),

    max_features=(
        MAX_VOCABULARY_SIZE
    ),

    ngram_range=(
        NGRAM_RANGE
    ),

    token_pattern=(
        r"(?u)\b[a-zA-Z_]"
        r"[a-zA-Z_']+\b"
    ),
)


count_matrix = (
    count_vectorizer.fit_transform(
        sample_df[
            "CLEAN_NARRATIVE"
        ]
    )
)


feature_names = np.array(
    count_vectorizer
    .get_feature_names_out()
)


print(
    f"Shared vocabulary terms: "
    f"{len(feature_names):,}"
)


# =============================================================================
# 9. K-MEANS PATH
# =============================================================================

tfidf_transformer = TfidfTransformer(
    sublinear_tf=True,

    norm="l2",
)


tfidf_matrix = (
    tfidf_transformer.fit_transform(
        count_matrix
    )
)


maximum_components = max(
    2,

    min(
        SVD_COMPONENTS,

        tfidf_matrix.shape[0] - 1,

        tfidf_matrix.shape[1] - 1,
    ),
)


svd_pipeline = make_pipeline(
    TruncatedSVD(
        n_components=(
            maximum_components
        ),

        random_state=(
            RANDOM_STATE
        ),
    ),

    Normalizer(
        copy=False
    ),
)


semantic_matrix = (
    svd_pipeline.fit_transform(
        tfidf_matrix
    )
)


explained_variance = float(
    svd_pipeline[
        "truncatedsvd"
    ]
    .explained_variance_ratio_
    .sum()
)


kmeans_model = MiniBatchKMeans(
    n_clusters=(
        KMEANS_CLUSTER_COUNT
    ),

    random_state=(
        RANDOM_STATE
    ),

    batch_size=256,

    n_init=20,

    max_iter=400,

    reassignment_ratio=0.01,
)


kmeans_labels = (
    kmeans_model.fit_predict(
        semantic_matrix
    )
)


sample_df[
    "KMEANS_CLUSTER_ID"
] = (
    kmeans_labels
)


KMEANS_SILHOUETTE = silhouette_score(
    semantic_matrix,

    kmeans_labels,

    metric="euclidean",

    sample_size=min(
        1500,
        len(sample_df),
    ),

    random_state=RANDOM_STATE,
)


sample_df[
    "KMEANS_DOCUMENT_SILHOUETTE"
] = silhouette_samples(
    semantic_matrix,

    kmeans_labels,

    metric="euclidean",
)


kmeans_distance_matrix = (
    kmeans_model.transform(
        semantic_matrix
    )
)


sample_df[
    "DISTANCE_TO_KMEANS_CENTER"
] = [
    kmeans_distance_matrix[
        row_number,
        cluster_label,
    ]

    for row_number, cluster_label
    in enumerate(
        kmeans_labels
    )
]


# =============================================================================
# 10. EXTRACT K-MEANS CLUSTER TERMS
# =============================================================================

kmeans_cluster_terms = {}

kmeans_term_records = []


for cluster_id in sorted(
    sample_df[
        "KMEANS_CLUSTER_ID"
    ].unique()
):

    cluster_mask = (
        sample_df[
            "KMEANS_CLUSTER_ID"
        ].to_numpy()
        == cluster_id
    )

    mean_scores = np.asarray(
        tfidf_matrix[
            cluster_mask
        ].mean(
            axis=0
        )
    ).ravel()

    ranked_indexes = (
        mean_scores
        .argsort()[
            ::-1
        ]
    )

    selected_terms = []

    for feature_index in ranked_indexes:

        if mean_scores[
            feature_index
        ] <= 0:

            continue

        readable_term = (
            str(
                feature_names[
                    feature_index
                ]
            )
            .replace(
                "_",
                " ",
            )
        )

        selected_terms.append(
            readable_term
        )

        kmeans_term_records.append(
            {
                "Cluster ID":
                    int(cluster_id),

                "Term rank":
                    len(
                        selected_terms
                    ),

                "Term":
                    readable_term,

                "Mean TF-IDF":
                    float(
                        mean_scores[
                            feature_index
                        ]
                    ),
            }
        )

        if len(
            selected_terms
        ) >= TOP_TERMS_TO_DISPLAY:

            break

    kmeans_cluster_terms[
        cluster_id
    ] = selected_terms


kmeans_terms_df = pd.DataFrame(
    kmeans_term_records
)


# =============================================================================
# 11. SELECT THE FINGER PINCH/CRUSH K-MEANS CLUSTER
# =============================================================================

FINGER_TERM_WEIGHTS = {
    "index finger": 6.0,
    "middle finger": 6.0,
    "ring finger": 6.0,
    "little finger": 6.0,
    "finger tip": 6.0,
    "fingertip": 6.0,
    "finger": 4.0,
    "fingers": 4.0,
    "amputation": 5.0,
    "amputated": 5.0,
    "pinched": 4.0,
    "pinch": 4.0,
    "caught between": 4.0,
    "caught_between": 4.0,
    "crushed": 4.0,
    "crush": 4.0,
    "degloved": 5.0,
    "hand": 1.0,
}


def finger_event_indicator(
    narrative,
):

    text = str(
        narrative
    ).lower()

    finger_patterns = [
        r"\bfinger\b",
        r"\bfingers\b",
        r"\bfingertip\b",
        r"\bfinger tip\b",
        r"\bindex finger\b",
        r"\bmiddle finger\b",
        r"\bring finger\b",
        r"\blittle finger\b",
        r"\bamputat",
        r"\bdeglov",
    ]

    pinch_patterns = [
        r"\bpinch",
        r"\bcaught",
        r"\bcrush",
        r"\bjammed",
        r"\bsmashed",
        r"\btrapped",
    ]

    has_finger = any(
        re.search(
            pattern,
            text,
        )

        for pattern in finger_patterns
    )

    has_pinch = any(
        re.search(
            pattern,
            text,
        )

        for pattern in pinch_patterns
    )

    return int(
        has_finger
        or (
            "hand"
            in text
            and has_pinch
        )
    )


sample_df[
    "FINGER_EVENT_INDICATOR"
] = (
    sample_df[
        TEXT_COLUMN
    ]
    .map(
        finger_event_indicator
    )
)


finger_cluster_score_records = []


for cluster_id in sorted(
    kmeans_cluster_terms
):

    vocabulary_score = 0.0

    for rank, term in enumerate(
        kmeans_cluster_terms[
            cluster_id
        ],

        start=1,
    ):

        normalized_term = (
            str(term)
            .lower()
            .strip()
        )

        rank_weight = (
            1.0
            / math.sqrt(
                rank
            )
        )

        for target_term, weight in (
            FINGER_TERM_WEIGHTS.items()
        ):

            if target_term in normalized_term:

                vocabulary_score += (
                    weight
                    * rank_weight
                )

    cluster_records = sample_df[
        sample_df[
            "KMEANS_CLUSTER_ID"
        ]
        == cluster_id
    ]

    finger_event_rate = float(
        cluster_records[
            "FINGER_EVENT_INDICATOR"
        ].mean()
    )

    combined_score = (
        vocabulary_score
        + 15.0
        * finger_event_rate
    )

    finger_cluster_score_records.append(
        {
            "Cluster ID":
                int(cluster_id),

            "Vocabulary score":
                vocabulary_score,

            "Finger-event rate":
                finger_event_rate,

            "Combined selection score":
                combined_score,

            "Cluster size":
                len(
                    cluster_records
                ),
        }
    )


finger_cluster_scores_df = (
    pd.DataFrame(
        finger_cluster_score_records
    )
    .sort_values(
        "Combined selection score",
        ascending=False,
    )
    .reset_index(drop=True)
)


SELECTED_FINGER_CLUSTER_ID = int(
    finger_cluster_scores_df.loc[
        0,
        "Cluster ID",
    ]
)


sample_df[
    "IN_SELECTED_KMEANS_CLUSTER"
] = (
    sample_df[
        "KMEANS_CLUSTER_ID"
    ]
    == SELECTED_FINGER_CLUSTER_ID
)


selected_kmeans_terms = (
    kmeans_terms_df[
        kmeans_terms_df[
            "Cluster ID"
        ]
        == SELECTED_FINGER_CLUSTER_ID
    ]
    .sort_values(
        "Term rank"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 12. LDA PATH
# =============================================================================

lda_model = LatentDirichletAllocation(
    n_components=(
        LDA_TOPIC_COUNT
    ),

    learning_method="batch",

    max_iter=(
        LDA_MAX_ITERATIONS
    ),

    batch_size=(
        LDA_BATCH_SIZE
    ),

    evaluate_every=-1,

    random_state=(
        RANDOM_STATE
    ),

    n_jobs=-1,
)


document_topic_matrix = (
    lda_model.fit_transform(
        count_matrix
    )
)


sample_df[
    "DOMINANT_LDA_TOPIC"
] = (
    document_topic_matrix.argmax(
        axis=1
    )
)


sample_df[
    "DOMINANT_LDA_TOPIC_PROPORTION"
] = (
    document_topic_matrix.max(
        axis=1
    )
)


# =============================================================================
# 13. EXTRACT LDA TOPIC TERMS
# =============================================================================

lda_term_records = []


for topic_id, topic_weights in enumerate(
    lda_model.components_
):

    top_indexes = (
        topic_weights
        .argsort()[
            ::-1
        ][
            :TOP_TERMS_TO_DISPLAY
        ]
    )

    for term_rank, feature_index in enumerate(
        top_indexes,
        start=1,
    ):

        readable_term = (
            str(
                feature_names[
                    feature_index
                ]
            )
            .replace(
                "_",
                " ",
            )
        )

        lda_term_records.append(
            {
                "Topic ID":
                    int(topic_id),

                "Term rank":
                    term_rank,

                "Term":
                    readable_term,

                "Topic weight":
                    float(
                        topic_weights[
                            feature_index
                        ]
                    ),
            }
        )


lda_terms_df = pd.DataFrame(
    lda_term_records
)


# =============================================================================
# 14. SELECT THE CLOSEST LDA TOPIC
# =============================================================================

cluster_mask = (
    sample_df[
        "IN_SELECTED_KMEANS_CLUSTER"
    ]
    .to_numpy()
)


topic_match_records = []


for topic_id in range(
    LDA_TOPIC_COUNT
):

    topic_proportions = (
        document_topic_matrix[
            :,
            topic_id
        ]
    )

    mean_inside = float(
        topic_proportions[
            cluster_mask
        ].mean()
    )

    mean_outside = float(
        topic_proportions[
            ~cluster_mask
        ].mean()
    )

    separation = (
        mean_inside
        - mean_outside
    )

    enrichment = (
        mean_inside
        / (
            mean_outside
            + 1e-9
        )
    )

    dominant_inside_rate = float(
        (
            sample_df.loc[
                sample_df[
                    "IN_SELECTED_KMEANS_CLUSTER"
                ],

                "DOMINANT_LDA_TOPIC",
            ]
            == topic_id
        ).mean()
    )

    combined_score = (
        0.50
        * mean_inside

        + 0.25
        * max(
            separation,
            0.0,
        )

        + 0.15
        * dominant_inside_rate

        + 0.10
        * min(
            enrichment / 5.0,
            1.0,
        )
    )

    topic_match_records.append(
        {
            "Topic ID":
                int(topic_id),

            "Mean proportion inside cluster":
                mean_inside,

            "Mean proportion outside cluster":
                mean_outside,

            "Inside-outside separation":
                separation,

            "Enrichment ratio":
                enrichment,

            "Dominant-topic rate inside cluster":
                dominant_inside_rate,

            "Combined match score":
                combined_score,
        }
    )


topic_match_df = (
    pd.DataFrame(
        topic_match_records
    )
    .sort_values(
        "Combined match score",
        ascending=False,
    )
    .reset_index(drop=True)
)


MATCHED_LDA_TOPIC_ID = int(
    topic_match_df.loc[
        0,
        "Topic ID",
    ]
)


sample_df[
    "MATCHED_LDA_TOPIC_PROPORTION"
] = (
    document_topic_matrix[
        :,
        MATCHED_LDA_TOPIC_ID
    ]
)


matched_lda_terms = (
    lda_terms_df[
        lda_terms_df[
            "Topic ID"
        ]
        == MATCHED_LDA_TOPIC_ID
    ]
    .sort_values(
        "Term rank"
    )
    .reset_index(drop=True)
)


# =============================================================================
# 15. DEFINE INDEPENDENT LDA MEMBERSHIP
# =============================================================================

sample_df[
    "MATCHED_TOPIC_IS_DOMINANT"
] = (
    sample_df[
        "DOMINANT_LDA_TOPIC"
    ]
    == MATCHED_LDA_TOPIC_ID
)


sample_df[
    "MATCHED_TOPIC_ABOVE_THRESHOLD"
] = (
    sample_df[
        "MATCHED_LDA_TOPIC_PROPORTION"
    ]
    >= LDA_TOPIC_PROPORTION_THRESHOLD
)


sample_df[
    "IN_MATCHED_LDA_TOPIC"
] = (
    sample_df[
        "MATCHED_TOPIC_IS_DOMINANT"
    ]
    |
    sample_df[
        "MATCHED_TOPIC_ABOVE_THRESHOLD"
    ]
)


# =============================================================================
# 16. ASSIGN COMPARISON GROUPS
# =============================================================================

def assign_comparison_group(
    row,
):

    in_kmeans = bool(
        row[
            "IN_SELECTED_KMEANS_CLUSTER"
        ]
    )

    in_lda = bool(
        row[
            "IN_MATCHED_LDA_TOPIC"
        ]
    )

    if in_kmeans and in_lda:

        return "Agreement"

    if in_kmeans and not in_lda:

        return "K-means only"

    if not in_kmeans and in_lda:

        return "LDA only"

    return "Background"


sample_df[
    "COMPARISON_GROUP"
] = (
    sample_df.apply(
        assign_comparison_group,
        axis=1,
    )
)


# =============================================================================
# 17. CREATE VISUALIZATION COORDINATES BEFORE FILTERED DATAFRAMES
#
# This is the critical correction.
#
# VISUAL_X and VISUAL_Y are added to sample_df first. The filtered DataFrames
# created later will therefore contain the visualization fields.
# =============================================================================

visual_svd = TruncatedSVD(
    n_components=2,

    random_state=RANDOM_STATE,
)


visual_coordinates = (
    visual_svd.fit_transform(
        tfidf_matrix
    )
)


sample_df[
    "VISUAL_X"
] = visual_coordinates[
    :,
    0
]


sample_df[
    "VISUAL_Y"
] = visual_coordinates[
    :,
    1
]


# Validate the coordinates before proceeding.
required_visual_columns = {
    "VISUAL_X",
    "VISUAL_Y",
}


missing_visual_columns = (
    required_visual_columns
    - set(
        sample_df.columns
    )
)


if missing_visual_columns:

    raise RuntimeError(
        "Visualization fields were not created: "
        f"{sorted(missing_visual_columns)}"
    )


# =============================================================================
# 18. CREATE FILTERED DATAFRAMES AFTER VISUAL FIELDS EXIST
#
# Using .copy() ensures each subset retains all current sample_df columns,
# including VISUAL_X and VISUAL_Y.
# =============================================================================

selected_cluster_df = (
    sample_df.loc[
        sample_df[
            "IN_SELECTED_KMEANS_CLUSTER"
        ]
    ]
    .copy()
)


agreement_df = (
    sample_df.loc[
        sample_df[
            "COMPARISON_GROUP"
        ]
        == "Agreement"
    ]
    .copy()
)


kmeans_only_df = (
    sample_df.loc[
        sample_df[
            "COMPARISON_GROUP"
        ]
        == "K-means only"
    ]
    .copy()
)


lda_only_df = (
    sample_df.loc[
        sample_df[
            "COMPARISON_GROUP"
        ]
        == "LDA only"
    ]
    .copy()
)


background_df = (
    sample_df.loc[
        sample_df[
            "COMPARISON_GROUP"
        ]
        == "Background"
    ]
    .copy()
)


# Confirm each subset has the visualization fields.
for subset_name, subset_df in [
    (
        "selected_cluster_df",
        selected_cluster_df,
    ),
    (
        "agreement_df",
        agreement_df,
    ),
    (
        "kmeans_only_df",
        kmeans_only_df,
    ),
    (
        "lda_only_df",
        lda_only_df,
    ),
    (
        "background_df",
        background_df,
    ),
]:

    subset_missing = (
        required_visual_columns
        - set(
            subset_df.columns
        )
    )

    if subset_missing:

        raise RuntimeError(
            f"{subset_name} is missing visualization fields: "
            f"{sorted(subset_missing)}"
        )


# =============================================================================
# 19. CALCULATE COUNTS AND QUALITY MEASURES
# =============================================================================

selected_cluster_size = len(
    selected_cluster_df
)


selected_cluster_mean_silhouette = float(
    selected_cluster_df[
        "KMEANS_DOCUMENT_SILHOUETTE"
    ].mean()
)


selected_cluster_negative_count = int(
    (
        selected_cluster_df[
            "KMEANS_DOCUMENT_SILHOUETTE"
        ]
        < 0
    ).sum()
)


selected_cluster_finger_rate = float(
    selected_cluster_df[
        "FINGER_EVENT_INDICATOR"
    ].mean()
)


AGREEMENT_COUNT = len(
    agreement_df
)


KMEANS_ONLY_COUNT = len(
    kmeans_only_df
)


LDA_ONLY_COUNT = len(
    lda_only_df
)


BACKGROUND_COUNT = len(
    background_df
)


LDA_MEMBER_COUNT = (
    AGREEMENT_COUNT
    + LDA_ONLY_COUNT
)


KMEANS_CLUSTER_CAPTURE_RATE = (
    AGREEMENT_COUNT
    / selected_cluster_size
    if selected_cluster_size
    else 0.0
)


LDA_SET_PRECISION_RELATIVE_TO_CLUSTER = (
    AGREEMENT_COUNT
    / LDA_MEMBER_COUNT
    if LDA_MEMBER_COUNT
    else 0.0
)


FOCUSED_JACCARD = (
    AGREEMENT_COUNT
    / (
        AGREEMENT_COUNT
        + KMEANS_ONLY_COUNT
        + LDA_ONLY_COUNT
    )
    if (
        AGREEMENT_COUNT
        + KMEANS_ONLY_COUNT
        + LDA_ONLY_COUNT
    )
    else 0.0
)


# =============================================================================
# 20. FOCUSED GROUP SUMMARY
# =============================================================================

focused_group_summary_records = []


for comparison_group, group_df in [
    (
        "Agreement",
        agreement_df,
    ),
    (
        "K-means only",
        kmeans_only_df,
    ),
    (
        "LDA only",
        lda_only_df,
    ),
]:

    if group_df.empty:

        finger_rate = 0.0

        mean_topic_proportion = 0.0

        mean_silhouette = np.nan

    else:

        finger_rate = float(
            group_df[
                "FINGER_EVENT_INDICATOR"
            ].mean()
        )

        mean_topic_proportion = float(
            group_df[
                "MATCHED_LDA_TOPIC_PROPORTION"
            ].mean()
        )

        mean_silhouette = float(
            group_df[
                "KMEANS_DOCUMENT_SILHOUETTE"
            ].mean()
        )

    focused_group_summary_records.append(
        {
            "Comparison group":
                comparison_group,

            "Narratives":
                len(group_df),

            "Finger-event rate":
                finger_rate,

            "Mean matched-topic proportion":
                mean_topic_proportion,

            "Mean K-means silhouette":
                mean_silhouette,
        }
    )


focused_group_summary = pd.DataFrame(
    focused_group_summary_records
)


# =============================================================================
# 21. TOP-TERM OVERLAP
# =============================================================================

kmeans_top_term_set = set(
    selected_kmeans_terms[
        "Term"
    ]
    .head(
        TOP_TERMS_TO_DISPLAY
    )
    .str.lower()
)


lda_top_term_set = set(
    matched_lda_terms[
        "Term"
    ]
    .head(
        TOP_TERMS_TO_DISPLAY
    )
    .str.lower()
)


shared_top_terms = sorted(
    kmeans_top_term_set
    .intersection(
        lda_top_term_set
    )
)


kmeans_unique_top_terms = sorted(
    kmeans_top_term_set
    - lda_top_term_set
)


lda_unique_top_terms = sorted(
    lda_top_term_set
    - kmeans_top_term_set
)


TERM_JACCARD = (
    len(
        shared_top_terms
    )
    / len(
        kmeans_top_term_set
        .union(
            lda_top_term_set
        )
    )
    if (
        kmeans_top_term_set
        .union(
            lda_top_term_set
        )
    )
    else 0.0
)


# =============================================================================
# 22. SELECT REPRESENTATIVE DOCUMENTS
# =============================================================================

agreement_examples = (
    agreement_df
    .sort_values(
        [
            "DISTANCE_TO_KMEANS_CENTER",
            "MATCHED_LDA_TOPIC_PROPORTION",
        ],

        ascending=[
            True,
            False,
        ],
    )
)


kmeans_only_examples = (
    kmeans_only_df
    .sort_values(
        [
            "DISTANCE_TO_KMEANS_CENTER",
            "MATCHED_LDA_TOPIC_PROPORTION",
        ],

        ascending=[
            True,
            True,
        ],
    )
)


lda_only_examples = (
    lda_only_df
    .sort_values(
        [
            "MATCHED_LDA_TOPIC_PROPORTION",
            "FINGER_EVENT_INDICATOR",
        ],

        ascending=[
            False,
            False,
        ],
    )
)


borderline_kmeans_examples = (
    selected_cluster_df
    .sort_values(
        [
            "KMEANS_DOCUMENT_SILHOUETTE",
            "DISTANCE_TO_KMEANS_CENTER",
        ],

        ascending=[
            True,
            False,
        ],
    )
)


# =============================================================================
# 23. POST-HOC EXISTING FIELD SUMMARIES
# =============================================================================

POSTHOC_FIELDS = [
    column
    for column in [
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
    ]
    if column in sample_df.columns
]


posthoc_records = []


for comparison_group, group_df in [
    (
        "Agreement",
        agreement_df,
    ),
    (
        "K-means only",
        kmeans_only_df,
    ),
    (
        "LDA only",
        lda_only_df,
    ),
]:

    if group_df.empty:

        continue

    for field in POSTHOC_FIELDS:

        top_values = (
            group_df[
                field
            ]
            .astype("string")
            .fillna(
                "<MISSING>"
            )
            .value_counts()
            .head(4)
        )

        for rank, (
            field_value,
            field_count,
        ) in enumerate(
            top_values.items(),
            start=1,
        ):

            posthoc_records.append(
                {
                    "Comparison group":
                        comparison_group,

                    "Existing field":
                        field,

                    "Rank":
                        rank,

                    "Existing value":
                        str(
                            field_value
                        ),

                    "Count":
                        int(
                            field_count
                        ),

                    "Group percent":
                        (
                            100.0
                            * field_count
                            / len(
                                group_df
                            )
                        ),
                }
            )


posthoc_summary_df = pd.DataFrame(
    posthoc_records
)


# =============================================================================
# 24. SEVERITY DISTRIBUTIONS
# =============================================================================

focused_mask = (
    sample_df[
        "COMPARISON_GROUP"
    ]
    .isin(
        [
            "Agreement",
            "K-means only",
            "LDA only",
        ]
    )
)


focused_severity_pct = pd.crosstab(
    sample_df.loc[
        focused_mask,
        "COMPARISON_GROUP",
    ],

    sample_df.loc[
        focused_mask,
        "SEVERITY_GROUP",
    ],

    normalize="index",
) * 100


# =============================================================================
# 25. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def wrap_lines(
    lines,
    width=106,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append("")

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(line),

                    width=width,

                    break_long_words=False,

                    break_on_hyphens=False,
                )
                or [""]
            )

    return wrapped


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    width=106,
):

    wrapped = wrap_lines(
        lines,
        width=width,
    )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]

        for start in range(
            0,
            len(wrapped),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=23,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,

            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    printable.columns = [
        str(column)
        .replace(
            "_",
            " ",
        )
        .title()

        for column in printable.columns
    ]

    for column in printable.columns:

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    ""
                    if pd.isna(value)
                    else str(value)
            )
        )

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,

                            width=(
                                column_wrap_width
                            ),

                            break_long_words=False,

                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(printable),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),

            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,

            colLabels=chunk.columns,

            cellLoc="left",

            colLoc="left",

            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.43,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_document_pages(
    title,
    dataframe,
    maximum_records,
    explanation,
):

    if dataframe.empty:

        add_text_pages(
            title,

            [
                explanation,
                "",
                "No records were identified for this section.",
            ],
        )

        return

    selected_records = dataframe.head(
        maximum_records
    )

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        row_index,
        row,
    ) in enumerate(
        selected_records.iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 82,
                (
                    f"Example {example_number} | "
                    f"Record {row[id_column]} | "
                    f"{row['SEVERITY_GROUP']}"
                ),
                (
                    f"K-means cluster: "
                    f"{row['KMEANS_CLUSTER_ID']} | "
                    f"K-means silhouette: "
                    f"{row['KMEANS_DOCUMENT_SILHOUETTE']:.3f}"
                ),
                (
                    f"Matched LDA topic proportion: "
                    f"{row['MATCHED_LDA_TOPIC_PROPORTION']:.3f} | "
                    f"Matched topic dominant: "
                    f"{row['MATCHED_TOPIC_IS_DOMINANT']}"
                ),
                (
                    f"Existing classification: "
                    f"{row.get('CLASSIFICATION', '')}"
                ),
                (
                    f"Existing accident type: "
                    f"{row.get('ACCIDENT_TYPE', '')}"
                ),
                "",
                str(
                    row[
                        TEXT_COLUMN
                    ]
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.9,
        lines_per_page=51,
        width=104,
    )


def add_heatmap(
    title,
    dataframe,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(10.5, 6.7)
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,

        aspect="auto",

        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                dataframe.columns
            )
        )
    )

    axis.set_xticklabels(
        dataframe.columns,

        rotation=30,

        ha="right",
    )

    axis.set_yticks(
        range(
            len(
                dataframe.index
            )
        )
    )

    axis.set_yticklabels(
        dataframe.index
    )

    axis.set_title(
        title,

        fontsize=13,

        fontweight="bold",
    )

    axis.set_xlabel(
        "Injury-severity group"
    )

    axis.set_ylabel(
        "Focused comparison group"
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = matrix[
                row_number,
                column_number,
            ]

            if value > 0:

                axis.text(
                    column_number,
                    row_number,
                    f"{value:.0f}%",
                    ha="center",
                    va="center",
                    fontsize=9,
                )

    figure.colorbar(
        image,

        ax=axis,

        label="Percent",
    )

    figure.tight_layout()

    save_page(
        figure
    )


# =============================================================================
# 26. BUILD PDF REPORT
# =============================================================================

# -----------------------------------------------------------------------------
# Cover
# -----------------------------------------------------------------------------

figure = create_page()

figure.text(
    0.5,
    0.79,
    "MSHA Finger Pinch/Crush Narrative Comparison",
    ha="center",
    fontsize=19,
    fontweight="bold",
)

figure.text(
    0.5,
    0.73,
    "Corrected Focused K-means and LDA Analysis",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.685,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Background corpus used for fitting: {len(sample_df):,}",
    f"Shared vocabulary terms: {len(feature_names):,}",
    f"Selected K-means cluster: {SELECTED_FINGER_CLUSTER_ID}",
    f"Selected cluster size: {selected_cluster_size}",
    f"Matched LDA topic: {MATCHED_LDA_TOPIC_ID}",
    f"LDA membership: dominant topic or proportion >= "
    f"{LDA_TOPIC_PROPORTION_THRESHOLD:.2f}",
    f"LDA topic members: {LDA_MEMBER_COUNT}",
    f"Agreement records: {AGREEMENT_COUNT}",
    f"K-means-only records: {KMEANS_ONLY_COUNT}",
    f"LDA-only candidates: {LDA_ONLY_COUNT}",
    f"Cluster capture rate: {KMEANS_CLUSTER_CAPTURE_RATE:.1%}",
    f"Focused Jaccard: {FOCUSED_JACCARD:.3f}",
    "",
    "Scope:",
    "All 2,000 narratives were used to fit the models. The report focuses",
    "only on the selected finger cluster, the matched LDA topic, and the",
    "documents for which those methods agree or disagree.",
]

figure.text(
    0.11,
    0.58,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=9.0,
    family="monospace",
    linespacing=1.24,
)

save_page(
    figure
)


# -----------------------------------------------------------------------------
# Executive summary
# -----------------------------------------------------------------------------

add_text_pages(
    "Executive Summary",
    [
        "This report uses all 2,000 narratives to fit the vocabulary, K-means",
        "clusters, and LDA topics, but the analysis is narrowed to one selected",
        "finger pinch/crush cluster and its closest LDA topic.",
        "",
        f"K-means cluster {SELECTED_FINGER_CLUSTER_ID} contained",
        f"{selected_cluster_size} narratives and had a finger-event indicator",
        f"rate of {selected_cluster_finger_rate:.1%}.",
        "",
        f"LDA topic {MATCHED_LDA_TOPIC_ID} was selected as the closest topic.",
        "LDA membership was defined independently using the dominant-topic rule",
        f"or a topic-proportion threshold of",
        f"{LDA_TOPIC_PROPORTION_THRESHOLD:.2f}.",
        "",
        f"The methods agreed on {AGREEMENT_COUNT} records. K-means retained",
        f"{KMEANS_ONLY_COUNT} records that did not meet the LDA membership rule.",
        f"LDA identified {LDA_ONLY_COUNT} records from outside the K-means",
        "cluster as possible related documents.",
        "",
        "Agreement records provide the strongest shared interpretation.",
        "LDA-only records are candidates for relevant narratives that K-means",
        "may have placed elsewhere, but these records still require qualitative",
        "review.",
    ],
    font_size=8.7,
)


# -----------------------------------------------------------------------------
# Method summary
# -----------------------------------------------------------------------------

add_text_pages(
    "Focused Comparison Method",
    [
        "1. A severity-balanced sample of 2,000 narratives was used to fit both",
        "   methods.",
        "",
        "2. One synchronized vocabulary was created using the same cleaning,",
        "   stop-word removal, document-frequency limits, and one- and two-word",
        "   phrases.",
        "",
        "3. K-means used TF-IDF and Truncated SVD and assigned each narrative",
        "   to exactly one of nine clusters.",
        "",
        "4. LDA used term counts and represented each document as a mixture of",
        "   nine topic proportions.",
        "",
        "5. The finger pinch/crush K-means cluster was selected from its",
        "   vocabulary and narrative content.",
        "",
        "6. The closest LDA topic was selected using its concentration inside",
        "   the K-means cluster, separation from outside documents, enrichment,",
        "   and dominant-topic frequency.",
        "",
        "7. LDA membership was based on a topic rule, not a forced equal-size",
        "   document set.",
        "",
        "8. The report compares Agreement, K-means-only, and LDA-only records.",
        "   The remaining corpus is used only as model-fitting background.",
    ],
    font_size=8.6,
)


# -----------------------------------------------------------------------------
# Sample
# -----------------------------------------------------------------------------

add_table_pages(
    "Model-Fitting Sample",
    sampling_summary,
    rows_per_page=10,
    font_size=7.4,
    column_wrap_width=28,
)


# -----------------------------------------------------------------------------
# Vocabulary visualization
# -----------------------------------------------------------------------------

figure, axes = plt.subplots(
    1,
    2,
    figsize=(11, 7.5),
)


kmeans_chart_df = (
    selected_kmeans_terms
    .head(
        TOP_TERMS_TO_DISPLAY
    )
    .sort_values(
        "Mean TF-IDF",
        ascending=True,
    )
)


axes[0].barh(
    kmeans_chart_df[
        "Term"
    ],

    kmeans_chart_df[
        "Mean TF-IDF"
    ],

    color="#2E6F9E",
)


axes[0].set_title(
    (
        f"K-means Cluster "
        f"{SELECTED_FINGER_CLUSTER_ID}"
    ),

    fontweight="bold",
)


axes[0].set_xlabel(
    "Mean TF-IDF"
)


axes[0].grid(
    axis="x",
    alpha=0.25,
)


lda_chart_df = (
    matched_lda_terms
    .head(
        TOP_TERMS_TO_DISPLAY
    )
    .sort_values(
        "Topic weight",
        ascending=True,
    )
)


axes[1].barh(
    lda_chart_df[
        "Term"
    ],

    lda_chart_df[
        "Topic weight"
    ],

    color="#B46A3C",
)


axes[1].set_title(
    (
        f"LDA Topic "
        f"{MATCHED_LDA_TOPIC_ID}"
    ),

    fontweight="bold",
)


axes[1].set_xlabel(
    "LDA topic-term weight"
)


axes[1].grid(
    axis="x",
    alpha=0.25,
)


figure.suptitle(
    "Vocabulary Comparison for the Selected Finger Pattern",
    fontsize=14,
    fontweight="bold",
)


figure.tight_layout(
    rect=[
        0,
        0,
        1,
        0.95,
    ]
)


save_page(
    figure
)


# -----------------------------------------------------------------------------
# Term overlap
# -----------------------------------------------------------------------------

add_text_pages(
    "Top-Term Comparison",
    [
        "SHARED TERMS",
        "-" * 82,
        (
            ", ".join(
                shared_top_terms
            )
            if shared_top_terms
            else "No exact top-term overlap."
        ),
        "",
        "K-MEANS-ONLY TERMS",
        "-" * 82,
        (
            ", ".join(
                kmeans_unique_top_terms
            )
            if kmeans_unique_top_terms
            else "None"
        ),
        "",
        "LDA-ONLY TERMS",
        "-" * 82,
        (
            ", ".join(
                lda_unique_top_terms
            )
            if lda_unique_top_terms
            else "None"
        ),
        "",
        f"Top-term Jaccard overlap: {TERM_JACCARD:.3f}",
        "",
        "K-means terms identify vocabulary distinguishing the selected cluster.",
        "LDA terms represent vocabulary associated with a probabilistic topic",
        "that may occur across several K-means clusters.",
    ],
    font_size=8.4,
)


# -----------------------------------------------------------------------------
# Topic matching
# -----------------------------------------------------------------------------

topic_match_display = (
    topic_match_df.copy()
)


for column in [
    "Mean proportion inside cluster",
    "Mean proportion outside cluster",
    "Inside-outside separation",
    "Enrichment ratio",
    "Dominant-topic rate inside cluster",
    "Combined match score",
]:

    topic_match_display[
        column
    ] = (
        topic_match_display[
            column
        ]
        .map(
            lambda value:
                f"{value:.4f}"
        )
    )


add_table_pages(
    "LDA Topic Matching Results",
    topic_match_display,
    rows_per_page=12,
    font_size=6.4,
    column_wrap_width=22,
    landscape=True,
)


# -----------------------------------------------------------------------------
# Focused counts
# -----------------------------------------------------------------------------

comparison_colors = {
    "Agreement":
        "#4E8B57",

    "K-means only":
        "#2E6F9E",

    "LDA only":
        "#D98C3F",
}


figure = plt.figure(
    figsize=(8.5, 6)
)


axis = figure.add_subplot(
    111
)


axis.bar(
    focused_group_summary[
        "Comparison group"
    ],

    focused_group_summary[
        "Narratives"
    ],

    color=[
        comparison_colors[
            group
        ]

        for group in (
            focused_group_summary[
                "Comparison group"
            ]
        )
    ],
)


axis.set_title(
    "Focused Document Comparison",
    fontsize=13,
    fontweight="bold",
)


axis.set_ylabel(
    "Narrative count"
)


axis.grid(
    axis="y",
    alpha=0.25,
)


for position, value in enumerate(
    focused_group_summary[
        "Narratives"
    ]
):

    axis.text(
        position,
        value,
        str(
            int(value)
        ),
        ha="center",
        va="bottom",
        fontsize=10,
    )


figure.tight_layout()


save_page(
    figure
)


# -----------------------------------------------------------------------------
# Quality comparison
# -----------------------------------------------------------------------------

figure, axes = plt.subplots(
    1,
    2,
    figsize=(11, 6.5),
)


axes[0].bar(
    focused_group_summary[
        "Comparison group"
    ],

    focused_group_summary[
        "Finger-event rate"
    ]
    * 100,

    color=[
        comparison_colors[
            group
        ]

        for group in (
            focused_group_summary[
                "Comparison group"
            ]
        )
    ],
)


axes[0].set_title(
    "Finger-Event Indicator Rate",
    fontweight="bold",
)


axes[0].set_ylabel(
    "Percent of narratives"
)


axes[0].set_ylim(
    0,
    105,
)


axes[0].grid(
    axis="y",
    alpha=0.25,
)


for position, value in enumerate(
    focused_group_summary[
        "Finger-event rate"
    ]
    * 100
):

    axes[0].text(
        position,
        value,
        f"{value:.0f}%",
        ha="center",
        va="bottom",
    )


axes[1].bar(
    focused_group_summary[
        "Comparison group"
    ],

    focused_group_summary[
        "Mean matched-topic proportion"
    ],

    color=[
        comparison_colors[
            group
        ]

        for group in (
            focused_group_summary[
                "Comparison group"
            ]
        )
    ],
)


axes[1].set_title(
    "Mean Matched LDA Topic Proportion",
    fontweight="bold",
)


axes[1].set_ylabel(
    "Mean topic proportion"
)


axes[1].set_ylim(
    0,
    1,
)


axes[1].grid(
    axis="y",
    alpha=0.25,
)


figure.suptitle(
    "Focused Group Quality Comparison",
    fontsize=14,
    fontweight="bold",
)


figure.tight_layout(
    rect=[
        0,
        0,
        1,
        0.94,
    ]
)


save_page(
    figure
)


# -----------------------------------------------------------------------------
# Topic-proportion boxplot
# -----------------------------------------------------------------------------

figure = plt.figure(
    figsize=(9.5, 6.5)
)


axis = figure.add_subplot(
    111
)


boxplot_data = [
    agreement_df[
        "MATCHED_LDA_TOPIC_PROPORTION"
    ].to_numpy(),

    kmeans_only_df[
        "MATCHED_LDA_TOPIC_PROPORTION"
    ].to_numpy(),

    lda_only_df[
        "MATCHED_LDA_TOPIC_PROPORTION"
    ].to_numpy(),
]


boxplot = axis.boxplot(
    boxplot_data,

    tick_labels=[
        "Agreement",
        "K-means only",
        "LDA only",
    ],

    patch_artist=True,

    showmeans=True,
)


for patch, color in zip(
    boxplot[
        "boxes"
    ],

    [
        comparison_colors[
            "Agreement"
        ],

        comparison_colors[
            "K-means only"
        ],

        comparison_colors[
            "LDA only"
        ],
    ],
):

    patch.set_facecolor(
        color
    )

    patch.set_alpha(
        0.70
    )


axis.axhline(
    LDA_TOPIC_PROPORTION_THRESHOLD,

    color="#C0392B",

    linestyle="--",

    linewidth=1.2,

    label=(
        "Topic-proportion threshold "
        f"({LDA_TOPIC_PROPORTION_THRESHOLD:.2f})"
    ),
)


axis.set_title(
    (
        f"LDA Topic {MATCHED_LDA_TOPIC_ID} "
        "Proportions by Focused Group"
    ),

    fontsize=13,

    fontweight="bold",
)


axis.set_ylabel(
    "Matched LDA topic proportion"
)


axis.set_ylim(
    0,
    1.05,
)


axis.grid(
    axis="y",
    alpha=0.25,
)


axis.legend()


figure.tight_layout()


save_page(
    figure
)


# -----------------------------------------------------------------------------
# Corrected two-dimensional visualization
# -----------------------------------------------------------------------------

figure = plt.figure(
    figsize=(11, 8)
)


axis = figure.add_subplot(
    111
)


axis.scatter(
    background_df[
        "VISUAL_X"
    ],

    background_df[
        "VISUAL_Y"
    ],

    s=12,

    alpha=0.10,

    color="#AEB6BF",

    label=(
        f"Background corpus "
        f"(n={len(background_df)})"
    ),
)


for comparison_group, group_df in [
    (
        "Agreement",
        agreement_df,
    ),
    (
        "K-means only",
        kmeans_only_df,
    ),
    (
        "LDA only",
        lda_only_df,
    ),
]:

    axis.scatter(
        group_df[
            "VISUAL_X"
        ],

        group_df[
            "VISUAL_Y"
        ],

        s=40,

        alpha=0.78,

        color=comparison_colors[
            comparison_group
        ],

        label=(
            f"{comparison_group} "
            f"(n={len(group_df)})"
        ),
    )


axis.set_title(
    "Focused Two-Dimensional Document Comparison",
    fontsize=13,
    fontweight="bold",
)


axis.set_xlabel(
    "Latent text dimension 1"
)


axis.set_ylabel(
    "Latent text dimension 2"
)


axis.grid(
    alpha=0.18,
)


axis.legend(
    loc="best"
)


figure.text(
    0.5,
    0.015,
    (
        "The light-gray documents provide model-fitting context. "
        "The colored documents are the focused analytical groups."
    ),
    ha="center",
    fontsize=8,
    color="#555555",
)


figure.tight_layout(
    rect=[
        0,
        0.04,
        1,
        1,
    ]
)


save_page(
    figure
)


# -----------------------------------------------------------------------------
# Severity distribution
# -----------------------------------------------------------------------------

add_heatmap(
    "Injury-Severity Distribution within the Focused Groups",
    focused_severity_pct,
)


# -----------------------------------------------------------------------------
# Post-hoc interpretation
# -----------------------------------------------------------------------------

if not posthoc_summary_df.empty:

    posthoc_display = (
        posthoc_summary_df.copy()
    )

    posthoc_display[
        "Group percent"
    ] = (
        posthoc_display[
            "Group percent"
        ]
        .map(
            lambda value:
                f"{value:.1f}%"
        )
    )

    add_text_pages(
        "Post-Hoc Interpretation",
        [
            "The existing MSHA fields on the following pages were not used to",
            "create the K-means cluster or LDA topic.",
            "",
            "The tables are included only to evaluate whether Agreement,",
            "K-means-only, and LDA-only records describe similar body parts,",
            "injury outcomes, activities, accident types, and classifications.",
            "",
            "The most important check is whether the LDA-only records represent",
            "finger pinch, crush, caught-between, degloving, or amputation events",
            "that K-means placed into another narrative cluster.",
        ],
        font_size=8.7,
    )

    add_table_pages(
        "Existing MSHA Fields by Focused Group",
        posthoc_display,
        rows_per_page=22,
        font_size=6.2,
        column_wrap_width=27,
        landscape=True,
    )


# -----------------------------------------------------------------------------
# Examples
# -----------------------------------------------------------------------------

add_document_pages(
    title=(
        "Agreement: Strong Shared Finger-Event Examples"
    ),

    dataframe=agreement_examples,

    maximum_records=(
        AGREEMENT_DOCUMENTS_TO_DISPLAY
    ),

    explanation=(
        "These narratives were assigned to the K-means finger cluster and "
        "also met the matched LDA topic rule. These records provide the "
        "clearest shared definition of the finger pinch/crush pattern."
    ),
)


add_document_pages(
    title=(
        "K-means Only: Cluster Records Not Captured by LDA"
    ),

    dataframe=kmeans_only_examples,

    maximum_records=(
        KMEANS_ONLY_DOCUMENTS_TO_DISPLAY
    ),

    explanation=(
        "These records are members of the K-means finger cluster but did not "
        "have the matched LDA topic as their dominant topic and did not reach "
        f"the {LDA_TOPIC_PROPORTION_THRESHOLD:.2f} threshold."
    ),
)


add_document_pages(
    title=(
        "LDA Only: Candidate Finger Events Missing from K-means"
    ),

    dataframe=lda_only_examples,

    maximum_records=(
        LDA_ONLY_DOCUMENTS_TO_DISPLAY
    ),

    explanation=(
        "These narratives were assigned to other K-means clusters but met the "
        "matched LDA topic rule. These are the primary candidates for related "
        "finger events that the first-pass K-means cluster may have missed."
    ),
)


add_document_pages(
    title=(
        "Borderline Members of the K-means Finger Cluster"
    ),

    dataframe=borderline_kmeans_examples,

    maximum_records=(
        BORDERLINE_DOCUMENTS_TO_DISPLAY
    ),

    explanation=(
        "These selected-cluster narratives have the lowest K-means silhouette "
        "values. They may fit another K-means cluster nearly as well."
    ),
)


# -----------------------------------------------------------------------------
# Conclusions
# -----------------------------------------------------------------------------

add_text_pages(
    "Conclusions and Recommended Interpretation",
    [
        "1. All 2,000 narratives were used to establish the shared vocabulary",
        "   and fit both models.",
        "",
        "2. The substantive comparison is limited to one selected K-means",
        "   finger cluster and one matched LDA topic.",
        "",
        f"3. The two methods agreed on {AGREEMENT_COUNT} narratives.",
        "   These provide the strongest shared examples of the finger-event",
        "   pattern.",
        "",
        f"4. K-means retained {KMEANS_ONLY_COUNT} records that did not meet",
        "   the LDA membership rule. These may use different event-mechanism,",
        "   equipment, or injury vocabulary.",
        "",
        f"5. LDA identified {LDA_ONLY_COUNT} candidate records outside the",
        "   K-means cluster. These are possible related narratives that K-means",
        "   placed elsewhere.",
        "",
        "6. LDA-only records should not automatically be added to the cluster.",
        "   Qualitative review is still needed to confirm that the same abstract",
        "   finger pinch/crush concept is present.",
        "",
        "7. K-means creates an exclusive cluster assignment, while LDA allows",
        "   each narrative to contain a mixture of topics. Some disagreement is",
        "   therefore expected and informative.",
        "",
        "8. The comparison demonstrates why first-pass clusters require review.",
        "   Similar incidents may use different equipment, activity, mechanism,",
        "   and injury language.",
    ],
    font_size=8.6,
)


# -----------------------------------------------------------------------------
# Technical appendix
# -----------------------------------------------------------------------------

add_text_pages(
    "Technical Appendix",
    [
        f"Model-fitting narratives: {len(sample_df):,}",
        f"Shared vocabulary terms: {len(feature_names):,}",
        f"Minimum document frequency: {MIN_DOCUMENT_FREQUENCY}",
        f"Maximum document frequency: {MAX_DOCUMENT_FREQUENCY:.0%}",
        f"N-gram range: {NGRAM_RANGE}",
        f"SVD components: {maximum_components}",
        f"SVD explained variance: {explained_variance:.4f}",
        f"K-means clusters: {KMEANS_CLUSTER_COUNT}",
        f"Overall K-means silhouette: {KMEANS_SILHOUETTE:.4f}",
        f"Selected K-means cluster: {SELECTED_FINGER_CLUSTER_ID}",
        f"Selected cluster size: {selected_cluster_size}",
        f"Selected cluster mean silhouette: "
        f"{selected_cluster_mean_silhouette:.4f}",
        f"LDA topics: {LDA_TOPIC_COUNT}",
        f"Matched LDA topic: {MATCHED_LDA_TOPIC_ID}",
        f"LDA proportion threshold: "
        f"{LDA_TOPIC_PROPORTION_THRESHOLD:.2f}",
        f"LDA topic members: {LDA_MEMBER_COUNT}",
        f"Agreement count: {AGREEMENT_COUNT}",
        f"K-means-only count: {KMEANS_ONLY_COUNT}",
        f"LDA-only count: {LDA_ONLY_COUNT}",
        f"Focused Jaccard: {FOCUSED_JACCARD:.4f}",
        f"Top-term Jaccard: {TERM_JACCARD:.4f}",
        f"Random state: {RANDOM_STATE}",
        "",
        f"PDF output: {os.path.abspath(OUTPUT_PDF)}",
    ],
    font_size=8.3,
)


# =============================================================================
# 27. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "Corrected Focused MSHA Finger Cluster K-means and LDA Comparison"
)

metadata[
    "Author"
] = (
    "Python TF-IDF, MiniBatch K-means, "
    "and Latent Dirichlet Allocation Analysis"
)

metadata[
    "Subject"
] = (
    "Focused comparison of an MSHA finger pinch/crush "
    "K-means cluster with the closest LDA topic"
)

metadata[
    "Keywords"
] = (
    "MSHA, TF-IDF, K-means, LDA, finger injury, "
    "pinch point, crush, amputation, topic modeling"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 28. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 90)
print("CORRECTED FOCUSED K-MEANS AND LDA COMPARISON COMPLETE")
print("=" * 90)

print(
    f"Background narratives used for fitting: "
    f"{len(sample_df):,}"
)

print(
    f"Selected K-means finger cluster: "
    f"{SELECTED_FINGER_CLUSTER_ID}"
)

print(
    f"K-means cluster records: "
    f"{selected_cluster_size:,}"
)

print(
    f"Matched LDA topic: "
    f"{MATCHED_LDA_TOPIC_ID}"
)

print(
    f"LDA membership rule: dominant topic or "
    f"proportion >= {LDA_TOPIC_PROPORTION_THRESHOLD:.2f}"
)

print(
    f"LDA topic members: "
    f"{LDA_MEMBER_COUNT:,}"
)

print(
    f"Agreement records: "
    f"{AGREEMENT_COUNT:,}"
)

print(
    f"K-means-only records: "
    f"{KMEANS_ONLY_COUNT:,}"
)

print(
    f"LDA-only candidate records: "
    f"{LDA_ONLY_COUNT:,}"
)

print(
    f"Cluster capture rate: "
    f"{KMEANS_CLUSTER_CAPTURE_RATE:.1%}"
)

print(
    f"Focused Jaccard: "
    f"{FOCUSED_JACCARD:.4f}"
)

print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print("=" * 90)


# =============================================================================
# 29. CONCISE NOTEBOOK OUTPUTS
# =============================================================================

display(
    selected_kmeans_terms[
        [
            "Term rank",
            "Term",
            "Mean TF-IDF",
        ]
    ]
)


display(
    matched_lda_terms[
        [
            "Term rank",
            "Term",
            "Topic weight",
        ]
    ]
)


display(
    topic_match_df
)


display(
    focused_group_summary
)
````

````python
# =============================================================================
# TF-IDF MULTICLASS LOGISTIC REGRESSION BASELINE
# PRIMARY ACCIDENT-CAUSE CLASSIFICATION
#
# PURPOSE
# -------
# Train and evaluate an interpretable text-classification baseline that predicts
# the primary accident cause from MSHA accident narratives.
#
# TRAINING LABELS
# ---------------
# High-confidence primary-cause labels generated by the frozen ontology-guided
# model in Code Block 2.
#
# INDEPENDENT BENCHMARK
# ---------------------
# Human classifications in:
#
#   MSHA_Block3_Human_Review_Workbook- Completed.xlsx
#
# METHOD COMPARISON
# -----------------
# The PDF compares:
#
#   1. TF-IDF logistic regression versus held-out Code Block 2 labels.
#   2. TF-IDF logistic regression versus independent human labels.
#   3. Qwen ontology-guided classifications versus human labels.
#   4. Exact versus exact-or-related agreement.
#   5. Logistic regression top-one versus top-two performance.
#   6. Performance by severity, time period, confidence, and human certainty.
#   7. Category-level precision, recall, and F1.
#   8. Confusion matrices and disagreement examples.
#   9. Influential TF-IDF terms for each predicted cause.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. One of the following Code Block 2 result files is present:
#
#       MSHA_Block2_Expanded_Results.csv
#       MSHA_Block2_Revised_Results.csv
#       MSHA_Block2_Unseen_Results.csv
#
# 3. The completed Block 3 workbook is present:
#
#       MSHA_Block3_Human_Review_Workbook- Completed.xlsx
#
# OUTPUTS
# -------
# 1. MSHA_TFIDF_Logistic_Regression_Comparison_Report.pdf
# 2. MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv
# 3. MSHA_TFIDF_Logistic_Regression_Heldout_Predictions.csv
# 4. MSHA_TFIDF_Logistic_Regression_Category_Metrics.csv
# 5. MSHA_TFIDF_Logistic_Regression_Top_Terms.csv
# 6. MSHA_TFIDF_Logistic_Regression_Model.joblib
# 7. MSHA_TFIDF_Logistic_Regression_Training_Summary.json
#
# IMPORTANT INTERPRETATION
# ------------------------
# Code Block 2 labels are preliminary labels generated by Qwen rather than
# fully independent ground truth. The held-out Code Block 2 evaluation measures
# how closely logistic regression reproduces the ontology-guided model.
#
# The 100 human-reviewed records provide the independent benchmark. Human
# responses marked uncertain are retained and reported separately.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import json
import math
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_pdf import PdfPages

from scipy import sparse
from scipy.stats import chi2_contingency

from sklearn.base import clone
from sklearn.compose import ColumnTransformer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support,
    f1_score,
    cohen_kappa_score,
    log_loss,
)
from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_val_score,
)
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import normalize

import joblib


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

RANDOM_STATE = 42

TEXT_COLUMN = "NARRATIVE"

ID_COLUMN_CANDIDATES = [
    "SOURCE_RECORD_ID",
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

BLOCK2_RESULT_CANDIDATES = [
    "MSHA_Block2_Expanded_Results.csv",
    "MSHA_Block2_Revised_Results.csv",
    "MSHA_Block2_Unseen_Results.csv",
]

HUMAN_REVIEW_WORKBOOK = (
    "MSHA_Block3_Human_Review_Workbook- Completed.xlsx"
)

OUTPUT_PDF = (
    "MSHA_TFIDF_Logistic_Regression_Comparison_Report.pdf"
)

HUMAN_BENCHMARK_CSV = (
    "MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv"
)

HELDOUT_PREDICTIONS_CSV = (
    "MSHA_TFIDF_Logistic_Regression_Heldout_Predictions.csv"
)

CATEGORY_METRICS_CSV = (
    "MSHA_TFIDF_Logistic_Regression_Category_Metrics.csv"
)

TOP_TERMS_CSV = (
    "MSHA_TFIDF_Logistic_Regression_Top_Terms.csv"
)

MODEL_FILE = (
    "MSHA_TFIDF_Logistic_Regression_Model.joblib"
)

TRAINING_SUMMARY_JSON = (
    "MSHA_TFIDF_Logistic_Regression_Training_Summary.json"
)


# High-confidence preliminary-label criteria.
MIN_CAUSE_CONFIDENCE = 0.80

ALLOWED_CAUSE_SUPPORT_LEVELS = {
    "HIGH",
    "MEDIUM",
}

EXCLUDE_CAUSES = {
    "",
    "INSUFFICIENT_INFORMATION",
    "OTHER",
    "UNCERTAIN",
    "NAN",
    "NONE",
}


# Holdout design for internal surrogate-model evaluation.
HELDOUT_TEST_SIZE = 0.20


# TF-IDF settings.
TFIDF_MAX_FEATURES = 25000

TFIDF_MIN_DF = 2

TFIDF_MAX_DF = 0.95

TFIDF_NGRAM_RANGE = (
    1,
    2,
)

TFIDF_SUBLINEAR_TF = True

TFIDF_STOP_WORDS = "english"


# Logistic-regression regularization values evaluated by cross-validation.
LOGISTIC_C_VALUES = [
    0.10,
    0.25,
    0.50,
    1.00,
    2.00,
    4.00,
]


# Class weighting helps reduce dominance by common classes.
USE_BALANCED_CLASS_WEIGHTS = True


# Cross-validation.
REQUESTED_CV_FOLDS = 5


# Number of top influential terms displayed per cause.
TOP_TERMS_PER_CAUSE = 15


# Minimum class size needed for the held-out split.
MIN_CLASS_COUNT_FOR_SPLIT = 5


# Maximum examples included in each PDF disagreement section.
MAX_DISAGREEMENT_EXAMPLES = 15


# Human certainty handling.
UNCERTAIN_VALUES = {
    "UNCERTAIN",
    "UNSURE",
    "NOT SURE",
    "NOT_SURE",
    "UNKNOWN",
    "CANNOT DETERMINE",
    "CANT DETERMINE",
    "N/A",
}


# =============================================================================
# 3. PRIMARY-CAUSE ONTOLOGY
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


CATEGORY_ALIASES = {
    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "LOSS_OF_CONTROL":
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",

    "UNEXPECTED_MACHINE_MOVEMENT":
        "UNEXPECTED_EQUIPMENT_MOVEMENT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "MANUAL_HANDLING":
        "MANUAL_HANDLING_OR_OVEREXERTION",

    "GROUND_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "ROOF_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "MOBILE_EQUIPMENT_EVENT":
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",

    "STORED_ENERGY_RELEASE":
        "PRESSURE_OR_STORED_ENERGY_RELEASE",

    "FIRE_OR_EXPLOSION":
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",
}


# =============================================================================
# 4. RELATED PRIMARY-CAUSE GROUPS
#
# Exact agreement remains the primary metric.
# Related-category agreement is a secondary sensitivity measure.
# =============================================================================

RELATED_CAUSE_GROUPS = [
    {
        "CAUGHT_IN_OR_BETWEEN",
        "STRUCK_BY_MOVING_OBJECT",
        "MATERIAL_RELEASE_OR_SHIFT",
    },

    {
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    },

    {
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
        "MANUAL_HANDLING_OR_OVEREXERTION",
    },

    {
        "GROUND_OR_ROOF_FAILURE",
        "MATERIAL_RELEASE_OR_SHIFT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "PRESSURE_OR_STORED_ENERGY_RELEASE",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "EQUIPMENT_CONDITION_OR_FAILURE",
    },

    {
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    },

    {
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "SLIP_TRIP_OR_FALL",
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    },
]


# Remove the factor-only category accidentally present in the stored-energy
# relationship group if it is not part of the cause ontology.
RELATED_CAUSE_GROUPS = [
    {
        value
        for value in group
        if value in set(
            CAUSE_CATEGORIES
        )
    }
    for group in RELATED_CAUSE_GROUPS
]


# =============================================================================
# 5. INPUT VALIDATION
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(
    accidents,
    pd.DataFrame,
):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


block2_results_file = next(
    (
        candidate
        for candidate in BLOCK2_RESULT_CANDIDATES
        if Path(candidate).exists()
    ),
    None,
)


if block2_results_file is None:

    raise FileNotFoundError(
        "No Code Block 2 results file was found. Expected one of: "
        + ", ".join(
            BLOCK2_RESULT_CANDIDATES
        )
    )


if not Path(
    HUMAN_REVIEW_WORKBOOK
).exists():

    raise FileNotFoundError(
        f"Human-review workbook not found: {HUMAN_REVIEW_WORKBOOK}"
    )


print("=" * 95)
print("TF-IDF MULTICLASS LOGISTIC REGRESSION BASELINE")
print("=" * 95)

print(
    "Code Block 2 results:",
    block2_results_file,
)

print(
    "Human-review workbook:",
    HUMAN_REVIEW_WORKBOOK,
)


# =============================================================================
# 6. GENERAL UTILITIES
# =============================================================================

def normalize_column_name(
    value,
):

    return re.sub(
        r"[^A-Z0-9]+",
        "_",
        str(
            value
        )
        .strip()
        .upper(),
    ).strip(
        "_"
    )


def normalize_category(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def normalize_yes_no_uncertain(
    value,
):

    normalized = normalize_column_name(
        value
    )

    if normalized in {
        "YES",
        "Y",
        "TRUE",
        "1",
        "COMPLETE",
        "COMPLETED",
    }:

        return "YES"

    if normalized in {
        "NO",
        "N",
        "FALSE",
        "0",
    }:

        return "NO"

    if normalized in {
        normalize_column_name(
            item
        )
        for item in UNCERTAIN_VALUES
    }:

        return "UNCERTAIN"

    return normalized


def first_existing_column(
    dataframe,
    candidates,
    required=False,
    description="column",
):

    normalized_lookup = {
        normalize_column_name(
            column
        ):
            column
        for column in dataframe.columns
    }

    for candidate in candidates:

        normalized_candidate = (
            normalize_column_name(
                candidate
            )
        )

        if normalized_candidate in normalized_lookup:

            return normalized_lookup[
                normalized_candidate
            ]

    if required:

        raise KeyError(
            f"Unable to find the required {description}. "
            f"Checked: {candidates}. "
            f"Available columns: {list(dataframe.columns)}"
        )

    return None


def readable_label(
    value,
):

    return (
        str(
            value
        )
        .replace(
            "_",
            " ",
        )
        .title()
    )


def safe_float(
    value,
    default=np.nan,
):

    try:

        number = float(
            value
        )

        if np.isnan(
            number
        ):

            return default

        return number

    except Exception:

        return default


def categories_related(
    category_a,
    category_b,
):

    a = normalize_category(
        category_a
    )

    b = normalize_category(
        category_b
    )

    if not a or not b:

        return False

    if a == b:

        return True

    return any(
        (
            a in group
            and b in group
        )
        for group in RELATED_CAUSE_GROUPS
    )


def is_uncertain_value(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return (
        normalized
        in {
            normalize_column_name(
                item
            )
            for item in UNCERTAIN_VALUES
        }
    )


def derive_confidence_band(
    value,
):

    confidence = safe_float(
        value
    )

    if pd.isna(
        confidence
    ):

        return "Missing"

    if confidence >= 0.80:

        return "High: 0.80-1.00"

    if confidence >= 0.60:

        return "Medium: 0.60-0.79"

    return "Low: Below 0.60"


# =============================================================================
# 7. LOAD AND STANDARDIZE CODE BLOCK 2 RESULTS
# =============================================================================

block2_raw_df = pd.read_csv(
    block2_results_file,
    low_memory=False,
)


block2_df = block2_raw_df.copy()


block2_df.columns = [
    normalize_column_name(
        column
    )
    for column in block2_df.columns
]


block2_id_column = first_existing_column(
    block2_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="Block 2 record-ID column",
)


block2_text_column = first_existing_column(
    block2_df,
    [
        "NARRATIVE",
        "TEXT",
        "ACCIDENT_NARRATIVE",
    ],
    required=True,
    description="Block 2 narrative column",
)


block2_cause_column = first_existing_column(
    block2_df,
    [
        "PRIMARY_CAUSE",
        "MODEL_PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "REVISED_CAUSE",
    ],
    required=True,
    description="Block 2 primary-cause column",
)


block2_record_type_column = first_existing_column(
    block2_df,
    [
        "RECORD_TYPE",
        "ORIGINAL_MODEL_RECORD_TYPE",
        "MODEL_RECORD_TYPE",
    ],
)


block2_processing_status_column = first_existing_column(
    block2_df,
    [
        "PROCESSING_STATUS",
        "MODEL_PROCESSING_STATUS",
    ],
)


block2_cause_evidence_valid_column = first_existing_column(
    block2_df,
    [
        "CAUSE_EVIDENCE_VALID",
        "MODEL_CAUSE_EVIDENCE_VALID",
    ],
)


block2_cause_confidence_column = first_existing_column(
    block2_df,
    [
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
        "MODEL_CAUSE_CONFIDENCE",
    ],
)


block2_cause_support_column = first_existing_column(
    block2_df,
    [
        "CAUSE_SUPPORT_LEVEL",
        "ORIGINAL_MODEL_CAUSE_SUPPORT",
        "MODEL_CAUSE_SUPPORT",
    ],
)


block2_severity_column = first_existing_column(
    block2_df,
    [
        "SEVERITY_GROUP",
        "INJURY_SEVERITY_GROUP",
    ],
)


block2_time_column = first_existing_column(
    block2_df,
    [
        "TIME_PERIOD",
        "ACCIDENT_TIME_PERIOD",
    ],
)


block2_df[
    block2_id_column
] = (
    block2_df[
        block2_id_column
    ]
    .astype(str)
)


block2_df[
    block2_text_column
] = (
    block2_df[
        block2_text_column
    ]
    .astype("string")
    .fillna("")
    .str.strip()
)


block2_df[
    "PRELIMINARY_CAUSE_LABEL"
] = (
    block2_df[
        block2_cause_column
    ]
    .map(
        normalize_category
    )
)


if block2_cause_confidence_column:

    block2_df[
        "CAUSE_CONFIDENCE_NUMERIC"
    ] = pd.to_numeric(
        block2_df[
            block2_cause_confidence_column
        ],
        errors="coerce",
    )

else:

    block2_df[
        "CAUSE_CONFIDENCE_NUMERIC"
    ] = np.nan


if block2_cause_support_column:

    block2_df[
        "CAUSE_SUPPORT_NORMALIZED"
    ] = (
        block2_df[
            block2_cause_support_column
        ]
        .map(
            normalize_category
        )
    )

else:

    block2_df[
        "CAUSE_SUPPORT_NORMALIZED"
    ] = "MISSING"


# =============================================================================
# 8. SELECT HIGH-CONFIDENCE TRAINING LABELS
# =============================================================================

training_mask = (
    block2_df[
        block2_text_column
    ]
    .str.split()
    .str.len()
    .ge(
        5
    )
)


training_mask &= (
    block2_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .isin(
        CAUSE_CATEGORIES
    )
)


training_mask &= (
    ~block2_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .isin(
        EXCLUDE_CAUSES
    )
)


if block2_processing_status_column:

    training_mask &= (
        block2_df[
            block2_processing_status_column
        ]
        .astype(str)
        .str.upper()
        .eq(
            "SUCCESS"
        )
    )


if block2_record_type_column:

    training_mask &= (
        block2_df[
            block2_record_type_column
        ]
        .map(
            normalize_category
        )
        .eq(
            "ACUTE_ACCIDENT"
        )
    )


if block2_cause_evidence_valid_column:

    evidence_values = (
        block2_df[
            block2_cause_evidence_valid_column
        ]
        .astype(str)
        .str.upper()
    )

    training_mask &= evidence_values.isin(
        [
            "TRUE",
            "1",
            "YES",
        ]
    )


if block2_cause_confidence_column:

    training_mask &= (
        block2_df[
            "CAUSE_CONFIDENCE_NUMERIC"
        ]
        .ge(
            MIN_CAUSE_CONFIDENCE
        )
    )


if block2_cause_support_column:

    training_mask &= (
        block2_df[
            "CAUSE_SUPPORT_NORMALIZED"
        ]
        .isin(
            ALLOWED_CAUSE_SUPPORT_LEVELS
        )
    )


training_candidates_df = (
    block2_df.loc[
        training_mask
    ]
    .copy()
)


# Remove exact duplicate narratives.
training_candidates_df[
    "_NORMALIZED_NARRATIVE"
] = (
    training_candidates_df[
        block2_text_column
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


training_candidates_df = (
    training_candidates_df
    .drop_duplicates(
        subset=[
            "_NORMALIZED_NARRATIVE"
        ],
        keep="first",
    )
    .reset_index(
        drop=True
    )
)


# Classes must have enough records for stratified holdout and cross-validation.
training_class_counts = (
    training_candidates_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .value_counts()
)


eligible_training_classes = (
    training_class_counts[
        training_class_counts
        >= MIN_CLASS_COUNT_FOR_SPLIT
    ]
    .index
    .tolist()
)


training_df = training_candidates_df[
    training_candidates_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .isin(
        eligible_training_classes
    )
].copy()


if training_df.empty:

    raise ValueError(
        "No eligible high-confidence Block 2 training records were found."
    )


if training_df[
    "PRELIMINARY_CAUSE_LABEL"
].nunique() < 2:

    raise ValueError(
        "At least two eligible primary-cause classes are required."
    )


print(
    "High-confidence training records:",
    f"{len(training_df):,}",
)

print(
    "Eligible training classes:",
    training_df[
        "PRELIMINARY_CAUSE_LABEL"
    ].nunique(),
)


# =============================================================================
# 9. LOAD HUMAN-REVIEW WORKBOOK
# =============================================================================

excel_file = pd.ExcelFile(
    HUMAN_REVIEW_WORKBOOK,
    engine="openpyxl",
)


sheet_names = excel_file.sheet_names


def choose_sheet(
    names,
    preferred_terms,
):

    for term in preferred_terms:

        for name in names:

            if (
                normalize_column_name(
                    term
                )
                in normalize_column_name(
                    name
                )
            ):

                return name

    return None


human_sheet = choose_sheet(
    sheet_names,
    [
        "Human Review",
        "Review",
    ],
)


model_sheet = choose_sheet(
    sheet_names,
    [
        "Model Output",
        "Model",
    ],
)


if human_sheet is None:

    raise KeyError(
        f"Unable to identify the Human Review worksheet. "
        f"Available sheets: {sheet_names}"
    )


if model_sheet is None:

    raise KeyError(
        f"Unable to identify the Model Output worksheet. "
        f"Available sheets: {sheet_names}"
    )


human_raw_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=human_sheet,
    engine="openpyxl",
)


model_raw_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=model_sheet,
    engine="openpyxl",
)


human_df = human_raw_df.copy()

model_df = model_raw_df.copy()


human_df.columns = [
    normalize_column_name(
        column
    )
    for column in human_df.columns
]


model_df.columns = [
    normalize_column_name(
        column
    )
    for column in model_df.columns
]


# =============================================================================
# 10. IDENTIFY HUMAN-REVIEW COLUMNS
# =============================================================================

human_id_column = first_existing_column(
    human_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="human-review record-ID column",
)


human_text_column = first_existing_column(
    human_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
)


human_cause_column = first_existing_column(
    human_df,
    [
        "HUMAN_PRIMARY_CAUSE",
        "HUMAN_CAUSE",
        "HUMAN_CORRECTED_CAUSE",
        "PRIMARY_CAUSE_HUMAN",
    ],
    required=True,
    description="human primary-cause column",
)


human_cause_uncertain_column = first_existing_column(
    human_df,
    [
        "HUMAN_CAUSE_UNCERTAIN",
        "CAUSE_UNCERTAIN",
        "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    ],
)


human_completion_column = first_existing_column(
    human_df,
    [
        "REVIEW_COMPLETED",
        "HUMAN_REVIEW_COMPLETED",
        "COMPLETED",
    ],
)


human_comment_column = first_existing_column(
    human_df,
    [
        "HUMAN_REVIEW_COMMENTS",
        "REVIEW_COMMENTS",
        "COMMENTS",
        "HUMAN_COMMENTS",
    ],
)


human_severity_column = first_existing_column(
    human_df,
    [
        "SEVERITY_GROUP",
        "INJURY_SEVERITY_GROUP",
    ],
)


human_time_column = first_existing_column(
    human_df,
    [
        "TIME_PERIOD",
        "ACCIDENT_TIME_PERIOD",
    ],
)


# =============================================================================
# 11. IDENTIFY MODEL-OUTPUT COLUMNS
# =============================================================================

model_id_column = first_existing_column(
    model_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="model-output record-ID column",
)


model_cause_column = first_existing_column(
    model_df,
    [
        "MODEL_PRIMARY_CAUSE",
        "PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "MODEL_CAUSE",
    ],
    required=True,
    description="model primary-cause column",
)


model_cause_confidence_column = first_existing_column(
    model_df,
    [
        "MODEL_CAUSE_CONFIDENCE",
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
    ],
)


model_cause_support_column = first_existing_column(
    model_df,
    [
        "MODEL_CAUSE_SUPPORT",
        "CAUSE_SUPPORT_LEVEL",
        "ORIGINAL_MODEL_CAUSE_SUPPORT",
    ],
)


model_cause_evidence_column = first_existing_column(
    model_df,
    [
        "MODEL_CAUSE_EVIDENCE",
        "CAUSE_EVIDENCE_TEXT",
        "ORIGINAL_MODEL_CAUSE_EVIDENCE",
    ],
)


model_df[
    model_id_column
] = (
    model_df[
        model_id_column
    ]
    .astype(str)
)


human_df[
    human_id_column
] = (
    human_df[
        human_id_column
    ]
    .astype(str)
)


# =============================================================================
# 12. BUILD THE HUMAN BENCHMARK TABLE
# =============================================================================

benchmark_df = human_df.copy()


benchmark_df[
    "SOURCE_RECORD_ID"
] = benchmark_df[
    human_id_column
].astype(str)


benchmark_df[
    "HUMAN_PRIMARY_CAUSE"
] = benchmark_df[
    human_cause_column
].map(
    normalize_category
)


if human_text_column:

    benchmark_df[
        "NARRATIVE"
    ] = (
        benchmark_df[
            human_text_column
        ]
        .astype("string")
        .fillna("")
        .str.strip()
    )

else:

    benchmark_df[
        "NARRATIVE"
    ] = ""


if human_completion_column:

    benchmark_df[
        "REVIEW_COMPLETED_NORMALIZED"
    ] = benchmark_df[
        human_completion_column
    ].map(
        normalize_yes_no_uncertain
    )

else:

    benchmark_df[
        "REVIEW_COMPLETED_NORMALIZED"
    ] = "YES"


benchmark_df[
    "HUMAN_CAUSE_UNCERTAIN"
] = benchmark_df[
    "HUMAN_PRIMARY_CAUSE"
].map(
    is_uncertain_value
)


if human_cause_uncertain_column:

    supplementary_uncertainty = (
        benchmark_df[
            human_cause_uncertain_column
        ]
        .map(
            normalize_yes_no_uncertain
        )
    )

    benchmark_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ] |= supplementary_uncertainty.eq(
        "UNCERTAIN"
    )


if human_comment_column:

    benchmark_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = (
        benchmark_df[
            human_comment_column
        ]
        .astype("string")
        .fillna("")
    )

else:

    benchmark_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = ""


# Merge saved Qwen output.
model_merge_df = pd.DataFrame(
    {
        "SOURCE_RECORD_ID":
            model_df[
                model_id_column
            ].astype(str),

        "QWEN_PRIMARY_CAUSE":
            model_df[
                model_cause_column
            ].map(
                normalize_category
            ),
    }
)


if model_cause_confidence_column:

    model_merge_df[
        "QWEN_CAUSE_CONFIDENCE"
    ] = pd.to_numeric(
        model_df[
            model_cause_confidence_column
        ],
        errors="coerce",
    )

else:

    model_merge_df[
        "QWEN_CAUSE_CONFIDENCE"
    ] = np.nan


if model_cause_support_column:

    model_merge_df[
        "QWEN_CAUSE_SUPPORT"
    ] = model_df[
        model_cause_support_column
    ].map(
        normalize_category
    )

else:

    model_merge_df[
        "QWEN_CAUSE_SUPPORT"
    ] = ""


if model_cause_evidence_column:

    model_merge_df[
        "QWEN_CAUSE_EVIDENCE"
    ] = (
        model_df[
            model_cause_evidence_column
        ]
        .astype("string")
        .fillna("")
    )

else:

    model_merge_df[
        "QWEN_CAUSE_EVIDENCE"
    ] = ""


model_merge_df = model_merge_df.drop_duplicates(
    subset=[
        "SOURCE_RECORD_ID"
    ],
    keep="last",
)


benchmark_df = benchmark_df.merge(
    model_merge_df,
    on="SOURCE_RECORD_ID",
    how="left",
)


# Fill missing narratives from the accidents DataFrame.
accidents_lookup_df = accidents.copy()


accidents_lookup_df.columns = [
    normalize_column_name(
        column
    )
    for column in accidents_lookup_df.columns
]


accidents_id_column = first_existing_column(
    accidents_lookup_df,
    [
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
        "SOURCE_RECORD_ID",
    ],
    required=True,
    description="accidents DataFrame record-ID column",
)


accidents_text_column = first_existing_column(
    accidents_lookup_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
    required=True,
    description="accidents DataFrame narrative column",
)


accidents_lookup_df[
    accidents_id_column
] = (
    accidents_lookup_df[
        accidents_id_column
    ]
    .astype(str)
)


narrative_lookup = (
    accidents_lookup_df
    .drop_duplicates(
        subset=[
            accidents_id_column
        ],
        keep="first",
    )
    .set_index(
        accidents_id_column
    )[
        accidents_text_column
    ]
    .astype("string")
    .fillna("")
    .to_dict()
)


missing_narrative_mask = (
    benchmark_df[
        "NARRATIVE"
    ]
    .str.strip()
    .eq("")
)


benchmark_df.loc[
    missing_narrative_mask,
    "NARRATIVE",
] = (
    benchmark_df.loc[
        missing_narrative_mask,
        "SOURCE_RECORD_ID",
    ]
    .map(
        narrative_lookup
    )
    .fillna("")
)


# Add severity and time data, preferring workbook fields.
if human_severity_column:

    benchmark_df[
        "SEVERITY_GROUP"
    ] = benchmark_df[
        human_severity_column
    ]

else:

    benchmark_df[
        "SEVERITY_GROUP"
    ] = "Unknown"


if human_time_column:

    benchmark_df[
        "TIME_PERIOD"
    ] = benchmark_df[
        human_time_column
    ]

else:

    benchmark_df[
        "TIME_PERIOD"
    ] = "Unknown"


# Keep completed, evaluable records.
benchmark_df = benchmark_df[
    benchmark_df[
        "REVIEW_COMPLETED_NORMALIZED"
    ]
    .isin(
        [
            "YES",
            "UNCERTAIN",
        ]
    )
].copy()


benchmark_df = benchmark_df[
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
    .notna()
].copy()


benchmark_df = benchmark_df[
    benchmark_df[
        "NARRATIVE"
    ]
    .astype(str)
    .str.strip()
    .ne("")
].copy()


print(
    "Completed human benchmark records:",
    f"{len(benchmark_df):,}",
)

print(
    "Human cause-uncertain records:",
    f"{int(benchmark_df['HUMAN_CAUSE_UNCERTAIN'].sum()):,}",
)


# =============================================================================
# 13. REMOVE BENCHMARK RECORDS FROM TRAINING DATA
# =============================================================================

benchmark_ids = set(
    benchmark_df[
        "SOURCE_RECORD_ID"
    ]
    .astype(str)
)


training_df = training_df[
    ~training_df[
        block2_id_column
    ]
    .astype(str)
    .isin(
        benchmark_ids
    )
].copy()


# Remove exact narrative overlap between training and human benchmark.
benchmark_normalized_narratives = set(
    benchmark_df[
        "NARRATIVE"
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


training_df = training_df[
    ~training_df[
        "_NORMALIZED_NARRATIVE"
    ]
    .isin(
        benchmark_normalized_narratives
    )
].copy()


# Recheck class counts after benchmark exclusion.
post_exclusion_class_counts = (
    training_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .value_counts()
)


eligible_training_classes = (
    post_exclusion_class_counts[
        post_exclusion_class_counts
        >= MIN_CLASS_COUNT_FOR_SPLIT
    ]
    .index
    .tolist()
)


training_df = training_df[
    training_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .isin(
        eligible_training_classes
    )
].copy()


if training_df[
    "PRELIMINARY_CAUSE_LABEL"
].nunique() < 2:

    raise ValueError(
        "Insufficient eligible classes remain after benchmark exclusion."
    )


# =============================================================================
# 14. TRAIN-HOLDOUT SPLIT
# =============================================================================

X = training_df[
    block2_text_column
].astype(str)


y = training_df[
    "PRELIMINARY_CAUSE_LABEL"
].astype(str)


X_train, X_holdout, y_train, y_holdout = train_test_split(
    X,
    y,
    test_size=HELDOUT_TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y,
)


print(
    "Surrogate-model training records:",
    f"{len(X_train):,}",
)

print(
    "Held-out Block 2 records:",
    f"{len(X_holdout):,}",
)


# =============================================================================
# 15. BUILD THE TF-IDF LOGISTIC-REGRESSION PIPELINE
# =============================================================================

class_weight_value = (
    "balanced"
    if USE_BALANCED_CLASS_WEIGHTS
    else None
)


def build_pipeline(
    c_value,
):

    return Pipeline(
        steps=[
            (
                "tfidf",
                TfidfVectorizer(
                    lowercase=True,
                    strip_accents="unicode",
                    stop_words=TFIDF_STOP_WORDS,
                    ngram_range=TFIDF_NGRAM_RANGE,
                    min_df=TFIDF_MIN_DF,
                    max_df=TFIDF_MAX_DF,
                    max_features=TFIDF_MAX_FEATURES,
                    sublinear_tf=TFIDF_SUBLINEAR_TF,
                    norm="l2",
                    dtype=np.float64,
                ),
            ),
            (
                "logistic",
                LogisticRegression(
                    C=c_value,
                    max_iter=3000,
                    solver="lbfgs",
                    class_weight=(
                        class_weight_value
                    ),
                    random_state=RANDOM_STATE,
                ),
            ),
        ]
    )


# =============================================================================
# 16. SELECT REGULARIZATION STRENGTH THROUGH CROSS-VALIDATION
# =============================================================================

minimum_class_size = int(
    y_train.value_counts().min()
)


cv_folds = max(
    2,
    min(
        REQUESTED_CV_FOLDS,
        minimum_class_size,
    ),
)


cv = StratifiedKFold(
    n_splits=cv_folds,
    shuffle=True,
    random_state=RANDOM_STATE,
)


cv_records = []


for c_value in LOGISTIC_C_VALUES:

    candidate_pipeline = build_pipeline(
        c_value
    )

    scores = cross_val_score(
        candidate_pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="f1_macro",
        n_jobs=-1,
    )

    cv_records.append(
        {
            "C":
                c_value,

            "Mean macro F1":
                float(
                    scores.mean()
                ),

            "Standard deviation":
                float(
                    scores.std()
                ),

            "Minimum fold":
                float(
                    scores.min()
                ),

            "Maximum fold":
                float(
                    scores.max()
                ),
        }
    )

    print(
        f"C={c_value:.2f} | "
        f"CV macro F1={scores.mean():.4f}"
    )


cv_results_df = (
    pd.DataFrame(
        cv_records
    )
    .sort_values(
        [
            "Mean macro F1",
            "C",
        ],
        ascending=[
            False,
            True,
        ],
    )
    .reset_index(
        drop=True
    )
)


best_c = float(
    cv_results_df.iloc[0][
        "C"
    ]
)


print(
    "Selected logistic-regression C:",
    best_c,
)


# =============================================================================
# 17. TRAIN THE SELECTED MODEL
# =============================================================================

selected_pipeline = build_pipeline(
    best_c
)


selected_pipeline.fit(
    X_train,
    y_train,
)


joblib.dump(
    selected_pipeline,
    MODEL_FILE,
)


# =============================================================================
# 18. HELD-OUT BLOCK 2 EVALUATION
# =============================================================================

holdout_predictions = (
    selected_pipeline.predict(
        X_holdout
    )
)


holdout_probabilities = (
    selected_pipeline.predict_proba(
        X_holdout
    )
)


model_classes = (
    selected_pipeline.named_steps[
        "logistic"
    ].classes_
)


holdout_top_two_indexes = np.argsort(
    holdout_probabilities,
    axis=1,
)[:, -2:][:, ::-1]


holdout_top_one_probabilities = (
    holdout_probabilities[
        np.arange(
            len(
                holdout_probabilities
            )
        ),
        holdout_top_two_indexes[
            :,
            0,
        ],
    ]
)


holdout_top_two_labels = [
    [
        model_classes[
            index
        ]
        for index in row_indexes
    ]
    for row_indexes in holdout_top_two_indexes
]


holdout_results_df = pd.DataFrame(
    {
        "NARRATIVE":
            X_holdout.values,

        "BLOCK2_PRELIMINARY_LABEL":
            y_holdout.values,

        "LOGISTIC_TOP1_CAUSE":
            holdout_predictions,

        "LOGISTIC_TOP1_PROBABILITY":
            holdout_top_one_probabilities,

        "LOGISTIC_TOP2_CAUSE":
            [
                labels[1]
                if len(labels) > 1
                else ""
                for labels in (
                    holdout_top_two_labels
                )
            ],
    }
)


holdout_results_df[
    "LOGISTIC_EXACT_MATCH"
] = (
    holdout_results_df[
        "LOGISTIC_TOP1_CAUSE"
    ]
    == holdout_results_df[
        "BLOCK2_PRELIMINARY_LABEL"
    ]
)


holdout_results_df[
    "LOGISTIC_TOP2_MATCH"
] = [
    true_label in predicted_labels
    for true_label, predicted_labels in zip(
        y_holdout.values,
        holdout_top_two_labels,
    )
]


holdout_results_df[
    "LOGISTIC_EXACT_OR_RELATED"
] = [
    categories_related(
        truth,
        prediction,
    )
    for truth, prediction in zip(
        holdout_results_df[
            "BLOCK2_PRELIMINARY_LABEL"
        ],
        holdout_results_df[
            "LOGISTIC_TOP1_CAUSE"
        ],
    )
]


holdout_results_df.to_csv(
    HELDOUT_PREDICTIONS_CSV,
    index=False,
    encoding="utf-8-sig",
)


# =============================================================================
# 19. HUMAN-BENCHMARK PREDICTIONS
# =============================================================================

benchmark_predictions = (
    selected_pipeline.predict(
        benchmark_df[
            "NARRATIVE"
        ]
    )
)


benchmark_probabilities = (
    selected_pipeline.predict_proba(
        benchmark_df[
            "NARRATIVE"
        ]
    )
)


benchmark_top_two_indexes = np.argsort(
    benchmark_probabilities,
    axis=1,
)[:, -2:][:, ::-1]


benchmark_top_two_labels = [
    [
        model_classes[
            index
        ]
        for index in row_indexes
    ]
    for row_indexes in benchmark_top_two_indexes
]


benchmark_df[
    "LOGISTIC_TOP1_CAUSE"
] = benchmark_predictions


benchmark_df[
    "LOGISTIC_TOP1_PROBABILITY"
] = (
    benchmark_probabilities[
        np.arange(
            len(
                benchmark_probabilities
            )
        ),
        benchmark_top_two_indexes[
            :,
            0,
        ],
    ]
)


benchmark_df[
    "LOGISTIC_TOP2_CAUSE"
] = [
    labels[1]
    if len(labels) > 1
    else ""
    for labels in (
        benchmark_top_two_labels
    )
]


benchmark_df[
    "LOGISTIC_TOP2_PROBABILITY"
] = (
    benchmark_probabilities[
        np.arange(
            len(
                benchmark_probabilities
            )
        ),
        benchmark_top_two_indexes[
            :,
            1,
        ],
    ]
)


# Exact agreement.
benchmark_df[
    "LOGISTIC_EXACT_MATCH"
] = (
    benchmark_df[
        "LOGISTIC_TOP1_CAUSE"
    ]
    == benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


benchmark_df[
    "QWEN_EXACT_MATCH"
] = (
    benchmark_df[
        "QWEN_PRIMARY_CAUSE"
    ]
    == benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


# Related agreement.
benchmark_df[
    "LOGISTIC_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_df[
            "LOGISTIC_TOP1_CAUSE"
        ],
    )
]


benchmark_df[
    "QWEN_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_df[
            "QWEN_PRIMARY_CAUSE"
        ],
    )
]


# Logistic top-two agreement.
benchmark_df[
    "LOGISTIC_TOP2_MATCH"
] = [
    human_value in predictions
    for human_value, predictions in zip(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_top_two_labels,
    )
]


benchmark_df[
    "LOGISTIC_TOP2_EXACT_OR_RELATED"
] = [
    any(
        categories_related(
            human_value,
            prediction,
        )
        for prediction in predictions
    )
    for human_value, predictions in zip(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_top_two_labels,
    )
]


benchmark_df[
    "LOGISTIC_CONFIDENCE_BAND"
] = benchmark_df[
    "LOGISTIC_TOP1_PROBABILITY"
].map(
    derive_confidence_band
)


benchmark_df[
    "QWEN_CONFIDENCE_BAND"
] = benchmark_df[
    "QWEN_CAUSE_CONFIDENCE"
].map(
    derive_confidence_band
)


benchmark_df.to_csv(
    HUMAN_BENCHMARK_CSV,
    index=False,
    encoding="utf-8-sig",
)


# =============================================================================
# 20. EVALUATION UTILITIES
# =============================================================================

def safe_metric(
    function,
    *args,
    **kwargs,
):

    try:

        return function(
            *args,
            **kwargs,
        )

    except Exception:

        return np.nan


def evaluate_predictions(
    y_true,
    y_pred,
    model_name,
):

    y_true = pd.Series(
        y_true
    ).astype(str)

    y_pred = pd.Series(
        y_pred
    ).astype(str)

    return {
        "Model":
            model_name,

        "Records":
            len(
                y_true
            ),

        "Exact accuracy":
            safe_metric(
                accuracy_score,
                y_true,
                y_pred,
            ),

        "Balanced accuracy":
            safe_metric(
                balanced_accuracy_score,
                y_true,
                y_pred,
            ),

        "Macro F1":
            safe_metric(
                f1_score,
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            ),

        "Weighted F1":
            safe_metric(
                f1_score,
                y_true,
                y_pred,
                average="weighted",
                zero_division=0,
            ),

        "Cohen's kappa":
            safe_metric(
                cohen_kappa_score,
                y_true,
                y_pred,
            ),
    }


# =============================================================================
# 21. OVERALL PERFORMANCE TABLES
# =============================================================================

holdout_metrics = evaluate_predictions(
    y_holdout,
    holdout_predictions,
    "Logistic regression vs held-out Qwen labels",
)


holdout_metrics[
    "Exact-or-related accuracy"
] = float(
    holdout_results_df[
        "LOGISTIC_EXACT_OR_RELATED"
    ].mean()
)


holdout_metrics[
    "Top-two accuracy"
] = float(
    holdout_results_df[
        "LOGISTIC_TOP2_MATCH"
    ].mean()
)


human_evaluable_df = benchmark_df[
    ~benchmark_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


logistic_human_metrics = evaluate_predictions(
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    benchmark_df[
        "LOGISTIC_TOP1_CAUSE"
    ],
    "Logistic regression vs all human labels",
)


logistic_human_metrics[
    "Exact-or-related accuracy"
] = float(
    benchmark_df[
        "LOGISTIC_EXACT_OR_RELATED"
    ].mean()
)


logistic_human_metrics[
    "Top-two accuracy"
] = float(
    benchmark_df[
        "LOGISTIC_TOP2_MATCH"
    ].mean()
)


logistic_human_metrics[
    "Top-two exact-or-related"
] = float(
    benchmark_df[
        "LOGISTIC_TOP2_EXACT_OR_RELATED"
    ].mean()
)


qwen_human_metrics = evaluate_predictions(
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    benchmark_df[
        "QWEN_PRIMARY_CAUSE"
    ],
    "Qwen ontology model vs all human labels",
)


qwen_human_metrics[
    "Exact-or-related accuracy"
] = float(
    benchmark_df[
        "QWEN_EXACT_OR_RELATED"
    ].mean()
)


if not human_evaluable_df.empty:

    logistic_certain_metrics = evaluate_predictions(
        human_evaluable_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        human_evaluable_df[
            "LOGISTIC_TOP1_CAUSE"
        ],
        "Logistic regression vs certain human labels",
    )

    logistic_certain_metrics[
        "Exact-or-related accuracy"
    ] = float(
        human_evaluable_df[
            "LOGISTIC_EXACT_OR_RELATED"
        ].mean()
    )

    logistic_certain_metrics[
        "Top-two accuracy"
    ] = float(
        human_evaluable_df[
            "LOGISTIC_TOP2_MATCH"
        ].mean()
    )

    logistic_certain_metrics[
        "Top-two exact-or-related"
    ] = float(
        human_evaluable_df[
            "LOGISTIC_TOP2_EXACT_OR_RELATED"
        ].mean()
    )

    qwen_certain_metrics = evaluate_predictions(
        human_evaluable_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        human_evaluable_df[
            "QWEN_PRIMARY_CAUSE"
        ],
        "Qwen ontology model vs certain human labels",
    )

    qwen_certain_metrics[
        "Exact-or-related accuracy"
    ] = float(
        human_evaluable_df[
            "QWEN_EXACT_OR_RELATED"
        ].mean()
    )

else:

    logistic_certain_metrics = {}

    qwen_certain_metrics = {}


overall_metrics_records = [
    holdout_metrics,
    logistic_human_metrics,
    qwen_human_metrics,
]


if logistic_certain_metrics:

    overall_metrics_records.extend(
        [
            logistic_certain_metrics,
            qwen_certain_metrics,
        ]
    )


overall_metrics_df = pd.DataFrame(
    overall_metrics_records
)


# =============================================================================
# 22. CATEGORY-LEVEL PERFORMANCE
# =============================================================================

evaluation_labels = sorted(
    set(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
    .union(
        benchmark_df[
            "LOGISTIC_TOP1_CAUSE"
        ]
    )
    .union(
        benchmark_df[
            "QWEN_PRIMARY_CAUSE"
        ]
        .dropna()
        .astype(str)
    )
)


evaluation_labels = [
    label
    for label in evaluation_labels
    if label
    and label not in {
        "NAN",
        "NONE",
        "",
    }
]


def category_metrics_dataframe(
    y_true,
    y_pred,
    labels,
    model_name,
):

    precision, recall, f1, support = (
        precision_recall_fscore_support(
            y_true,
            y_pred,
            labels=labels,
            zero_division=0,
        )
    )

    return pd.DataFrame(
        {
            "Model":
                model_name,

            "Primary cause":
                labels,

            "Precision":
                precision,

            "Recall":
                recall,

            "F1":
                f1,

            "Human support":
                support,
        }
    )


logistic_category_metrics_df = (
    category_metrics_dataframe(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_df[
            "LOGISTIC_TOP1_CAUSE"
        ],
        evaluation_labels,
        "TF-IDF logistic regression",
    )
)


qwen_category_metrics_df = (
    category_metrics_dataframe(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        benchmark_df[
            "QWEN_PRIMARY_CAUSE"
        ],
        evaluation_labels,
        "Qwen ontology model",
    )
)


category_metrics_df = pd.concat(
    [
        logistic_category_metrics_df,
        qwen_category_metrics_df,
    ],
    ignore_index=True,
)


category_metrics_df.to_csv(
    CATEGORY_METRICS_CSV,
    index=False,
    encoding="utf-8-sig",
)


# =============================================================================
# 23. PERFORMANCE BY CONFIDENCE BAND
# =============================================================================

def grouped_performance(
    dataframe,
    group_column,
    exact_column,
    related_column,
    model_name,
):

    records = []

    for group_value, group_df in dataframe.groupby(
        group_column,
        dropna=False,
    ):

        records.append(
            {
                "Model":
                    model_name,

                "Group":
                    group_value,

                "Records":
                    len(
                        group_df
                    ),

                "Exact accuracy":
                    float(
                        group_df[
                            exact_column
                        ].mean()
                    ),

                "Exact-or-related accuracy":
                    float(
                        group_df[
                            related_column
                        ].mean()
                    ),
            }
        )

    return pd.DataFrame(
        records
    )


logistic_confidence_performance_df = (
    grouped_performance(
        benchmark_df,
        "LOGISTIC_CONFIDENCE_BAND",
        "LOGISTIC_EXACT_MATCH",
        "LOGISTIC_EXACT_OR_RELATED",
        "TF-IDF logistic regression",
    )
)


qwen_confidence_performance_df = (
    grouped_performance(
        benchmark_df,
        "QWEN_CONFIDENCE_BAND",
        "QWEN_EXACT_MATCH",
        "QWEN_EXACT_OR_RELATED",
        "Qwen ontology model",
    )
)


confidence_performance_df = pd.concat(
    [
        logistic_confidence_performance_df,
        qwen_confidence_performance_df,
    ],
    ignore_index=True,
)


# =============================================================================
# 24. PERFORMANCE BY SEVERITY AND TIME PERIOD
# =============================================================================

def comparison_by_group(
    dataframe,
    group_column,
):

    records = []

    for group_value, group_df in dataframe.groupby(
        group_column,
        dropna=False,
    ):

        records.append(
            {
                group_column:
                    group_value,

                "Records":
                    len(
                        group_df
                    ),

                "Logistic exact":
                    float(
                        group_df[
                            "LOGISTIC_EXACT_MATCH"
                        ].mean()
                    ),

                "Logistic exact or related":
                    float(
                        group_df[
                            "LOGISTIC_EXACT_OR_RELATED"
                        ].mean()
                    ),

                "Logistic top two":
                    float(
                        group_df[
                            "LOGISTIC_TOP2_MATCH"
                        ].mean()
                    ),

                "Qwen exact":
                    float(
                        group_df[
                            "QWEN_EXACT_MATCH"
                        ].mean()
                    ),

                "Qwen exact or related":
                    float(
                        group_df[
                            "QWEN_EXACT_OR_RELATED"
                        ].mean()
                    ),

                "Human uncertain":
                    float(
                        group_df[
                            "HUMAN_CAUSE_UNCERTAIN"
                        ].mean()
                    ),
            }
        )

    return pd.DataFrame(
        records
    )


severity_performance_df = (
    comparison_by_group(
        benchmark_df,
        "SEVERITY_GROUP",
    )
)


time_performance_df = (
    comparison_by_group(
        benchmark_df,
        "TIME_PERIOD",
    )
)


# =============================================================================
# 25. CONFUSION MATRICES
# =============================================================================

confusion_labels = sorted(
    set(
        benchmark_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
    .union(
        benchmark_df[
            "LOGISTIC_TOP1_CAUSE"
        ]
    )
    .union(
        benchmark_df[
            "QWEN_PRIMARY_CAUSE"
        ]
        .dropna()
        .astype(str)
    )
)


confusion_labels = [
    label
    for label in confusion_labels
    if label
    and label not in {
        "NAN",
        "NONE",
        "",
    }
]


logistic_confusion_matrix = confusion_matrix(
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    benchmark_df[
        "LOGISTIC_TOP1_CAUSE"
    ],
    labels=confusion_labels,
)


qwen_confusion_matrix = confusion_matrix(
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    benchmark_df[
        "QWEN_PRIMARY_CAUSE"
    ],
    labels=confusion_labels,
)


# =============================================================================
# 26. EXTRACT INFLUENTIAL TERMS
# =============================================================================

tfidf_vectorizer = (
    selected_pipeline.named_steps[
        "tfidf"
    ]
)


logistic_model = (
    selected_pipeline.named_steps[
        "logistic"
    ]
)


feature_names = np.asarray(
    tfidf_vectorizer.get_feature_names_out()
)


top_term_records = []


for class_index, class_name in enumerate(
    logistic_model.classes_
):

    coefficients = (
        logistic_model.coef_[
            class_index
        ]
    )

    top_indexes = np.argsort(
        coefficients
    )[
        -TOP_TERMS_PER_CAUSE:
    ][
        ::-1
    ]

    for rank, feature_index in enumerate(
        top_indexes,
        start=1,
    ):

        top_term_records.append(
            {
                "Primary cause":
                    class_name,

                "Rank":
                    rank,

                "Term":
                    feature_names[
                        feature_index
                    ],

                "Coefficient":
                    float(
                        coefficients[
                            feature_index
                        ]
                    ),
            }
        )


top_terms_df = pd.DataFrame(
    top_term_records
)


top_terms_df.to_csv(
    TOP_TERMS_CSV,
    index=False,
    encoding="utf-8-sig",
)


# =============================================================================
# 27. METHOD-DIFFERENCE SUMMARIES
# =============================================================================

benchmark_df[
    "MODEL_AGREEMENT"
] = (
    benchmark_df[
        "LOGISTIC_TOP1_CAUSE"
    ]
    == benchmark_df[
        "QWEN_PRIMARY_CAUSE"
    ]
)


benchmark_df[
    "LOGISTIC_ONLY_EXACT"
] = (
    benchmark_df[
        "LOGISTIC_EXACT_MATCH"
    ]
    &
    ~benchmark_df[
        "QWEN_EXACT_MATCH"
    ]
)


benchmark_df[
    "QWEN_ONLY_EXACT"
] = (
    benchmark_df[
        "QWEN_EXACT_MATCH"
    ]
    &
    ~benchmark_df[
        "LOGISTIC_EXACT_MATCH"
    ]
)


benchmark_df[
    "BOTH_EXACT"
] = (
    benchmark_df[
        "LOGISTIC_EXACT_MATCH"
    ]
    &
    benchmark_df[
        "QWEN_EXACT_MATCH"
    ]
)


benchmark_df[
    "NEITHER_EXACT"
] = (
    ~benchmark_df[
        "LOGISTIC_EXACT_MATCH"
    ]
    &
    ~benchmark_df[
        "QWEN_EXACT_MATCH"
    ]
)


method_outcome_summary_df = pd.DataFrame(
    [
        {
            "Comparison outcome":
                "Both models exactly matched the human label",

            "Records":
                int(
                    benchmark_df[
                        "BOTH_EXACT"
                    ].sum()
                ),
        },
        {
            "Comparison outcome":
                "Only logistic regression exactly matched",

            "Records":
                int(
                    benchmark_df[
                        "LOGISTIC_ONLY_EXACT"
                    ].sum()
                ),
        },
        {
            "Comparison outcome":
                "Only Qwen exactly matched",

            "Records":
                int(
                    benchmark_df[
                        "QWEN_ONLY_EXACT"
                    ].sum()
                ),
        },
        {
            "Comparison outcome":
                "Neither model exactly matched",

            "Records":
                int(
                    benchmark_df[
                        "NEITHER_EXACT"
                    ].sum()
                ),
        },
        {
            "Comparison outcome":
                "Logistic regression and Qwen predicted the same cause",

            "Records":
                int(
                    benchmark_df[
                        "MODEL_AGREEMENT"
                    ].sum()
                ),
        },
    ]
)


# =============================================================================
# 28. TRAINING AND BENCHMARK SUMMARY
# =============================================================================

training_distribution_df = (
    training_df[
        "PRELIMINARY_CAUSE_LABEL"
    ]
    .value_counts()
    .rename_axis(
        "Primary cause"
    )
    .reset_index(
        name="Training records"
    )
)


human_distribution_df = (
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
    .value_counts()
    .rename_axis(
        "Primary cause"
    )
    .reset_index(
        name="Human benchmark records"
    )
)


training_summary = {
    "created_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "block2_results_file":
        block2_results_file,

    "human_review_workbook":
        HUMAN_REVIEW_WORKBOOK,

    "training_records_total":
        int(
            len(
                training_df
            )
        ),

    "training_records_used_for_fit":
        int(
            len(
                X_train
            )
        ),

    "heldout_records":
        int(
            len(
                X_holdout
            )
        ),

    "human_benchmark_records":
        int(
            len(
                benchmark_df
            )
        ),

    "human_certain_records":
        int(
            len(
                human_evaluable_df
            )
        ),

    "training_classes":
        sorted(
            training_df[
                "PRELIMINARY_CAUSE_LABEL"
            ]
            .unique()
            .tolist()
        ),

    "minimum_cause_confidence":
        MIN_CAUSE_CONFIDENCE,

    "allowed_support_levels":
        sorted(
            ALLOWED_CAUSE_SUPPORT_LEVELS
        ),

    "best_logistic_c":
        best_c,

    "cross_validation_folds":
        cv_folds,

    "tfidf_max_features":
        TFIDF_MAX_FEATURES,

    "tfidf_ngram_range":
        TFIDF_NGRAM_RANGE,

    "class_weight":
        class_weight_value,

    "random_state":
        RANDOM_STATE,

    "outputs": {
        "pdf":
            OUTPUT_PDF,

        "human_benchmark_csv":
            HUMAN_BENCHMARK_CSV,

        "heldout_predictions_csv":
            HELDOUT_PREDICTIONS_CSV,

        "category_metrics_csv":
            CATEGORY_METRICS_CSV,

        "top_terms_csv":
            TOP_TERMS_CSV,

        "model_file":
            MODEL_FILE,
    },
}


with open(
    TRAINING_SUMMARY_JSON,
    "w",
    encoding="utf-8",
) as output_file:

    json.dump(
        training_summary,
        output_file,
        indent=2,
        ensure_ascii=False,
        default=str,
    )


# =============================================================================
# 29. PDF HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis(
        "off"
    )

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append(
                ""
            )

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(
                        line
                    ),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(
                wrapped
            ),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(
            chunks
        ) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=28,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were available."
            ],
        )

        return

    printable = dataframe.copy()

    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[
                column
            ] = (
                printable[
                    column
                ]
                .map(
                    lambda value:
                        ""
                        if pd.isna(
                            value
                        )
                        else f"{value:.4f}"
                )
            )

    printable.columns = [
        readable_label(
            column
        )
        for column in printable.columns
    ]

    for column in printable.columns:

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    ""
                    if pd.isna(
                        value
                    )
                    else str(
                        value
                    )
            )
        )

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=column_wrap_width,
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(
            printable
        ),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis(
            "off"
        )

        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.36,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_confusion_heatmap(
    title,
    matrix,
    labels,
):

    if (
        matrix.size == 0
        or len(
            labels
        ) == 0
    ):

        return

    figure = plt.figure(
        figsize=(
            11,
            8.5,
        )
    )

    axis = figure.add_subplot(
        111
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                labels
            )
        )
    )

    axis.set_xticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        rotation=45,
        ha="right",
        fontsize=6.5,
    )

    axis.set_yticks(
        range(
            len(
                labels
            )
        )
    )

    axis.set_yticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        fontsize=6.5,
    )

    axis.set_xlabel(
        "Predicted primary cause"
    )

    axis.set_ylabel(
        "Human primary cause"
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    maximum_value = (
        matrix.max()
        if matrix.size
        else 0
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value > 0:

                color = (
                    "white"
                    if maximum_value
                    and value
                    > maximum_value * 0.55
                    else "black"
                )

                axis.text(
                    column_number,
                    row_number,
                    str(
                        value
                    ),
                    ha="center",
                    va="center",
                    fontsize=6.5,
                    color=color,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_metric_comparison_chart(
    metrics_dataframe,
):

    plot_df = metrics_dataframe[
        metrics_dataframe[
            "Model"
        ]
        .isin(
            [
                "Logistic regression vs all human labels",
                "Qwen ontology model vs all human labels",
            ]
        )
    ].copy()

    if plot_df.empty:

        return

    metric_columns = [
        "Exact accuracy",
        "Exact-or-related accuracy",
        "Macro F1",
        "Weighted F1",
        "Cohen's kappa",
    ]

    metric_columns = [
        column
        for column in metric_columns
        if column in plot_df.columns
    ]

    figure = plt.figure(
        figsize=(
            10.5,
            7,
        )
    )

    axis = figure.add_subplot(
        111
    )

    x_positions = np.arange(
        len(
            metric_columns
        )
    )

    width = 0.36

    colors = [
        "#2E6F9E",
        "#B46A3C",
    ]

    for model_number, (
        _,
        row,
    ) in enumerate(
        plot_df.iterrows()
    ):

        values = [
            row.get(
                metric,
                np.nan,
            )
            for metric in metric_columns
        ]

        axis.bar(
            x_positions
            + (
                model_number
                - 0.5
            )
            * width,
            values,
            width=width,
            label=(
                "TF-IDF Logistic Regression"
                if "Logistic" in row[
                    "Model"
                ]
                else "Qwen Ontology Model"
            ),
            color=colors[
                model_number
                % len(
                    colors
                )
            ],
        )

    axis.set_xticks(
        x_positions
    )

    axis.set_xticklabels(
        metric_columns,
        rotation=20,
        ha="right",
    )

    axis.set_ylim(
        min(
            -0.05,
            np.nanmin(
                plot_df[
                    metric_columns
                ].values
            )
            - 0.05,
        ),
        1.05,
    )

    axis.set_ylabel(
        "Metric value"
    )

    axis.set_title(
        "Independent Human-Benchmark Performance",
        fontsize=13,
        fontweight="bold",
    )

    axis.grid(
        axis="y",
        alpha=0.25,
    )

    axis.legend()

    figure.tight_layout()

    save_page(
        figure
    )


def add_cv_chart(
    dataframe,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(
            9,
            6,
        )
    )

    axis = figure.add_subplot(
        111
    )

    axis.errorbar(
        dataframe[
            "C"
        ],
        dataframe[
            "Mean macro F1"
        ],
        yerr=dataframe[
            "Standard deviation"
        ],
        marker="o",
        capsize=4,
        linewidth=2,
    )

    axis.axvline(
        best_c,
        color="#C44E52",
        linestyle="--",
        label=f"Selected C = {best_c:g}",
    )

    axis.set_xscale(
        "log"
    )

    axis.set_xlabel(
        "Inverse regularization strength, C"
    )

    axis.set_ylabel(
        "Cross-validated macro F1"
    )

    axis.set_title(
        "Logistic-Regression Regularization Selection",
        fontsize=13,
        fontweight="bold",
    )

    axis.grid(
        alpha=0.25,
    )

    axis.legend()

    figure.tight_layout()

    save_page(
        figure
    )


def add_top_terms_pages(
    dataframe,
    maximum_classes_per_page=4,
):

    classes = dataframe[
        "Primary cause"
    ].drop_duplicates().tolist()

    for start in range(
        0,
        len(
            classes
        ),
        maximum_classes_per_page,
    ):

        selected_classes = classes[
            start:
            start
            + maximum_classes_per_page
        ]

        figure, axes = plt.subplots(
            nrows=len(
                selected_classes
            ),
            ncols=1,
            figsize=(
                10,
                2.6
                * len(
                    selected_classes
                )
                + 1,
            ),
        )

        if len(
            selected_classes
        ) == 1:

            axes = [
                axes
            ]

        for axis, class_name in zip(
            axes,
            selected_classes,
        ):

            class_df = (
                dataframe[
                    dataframe[
                        "Primary cause"
                    ]
                    .eq(
                        class_name
                    )
                ]
                .sort_values(
                    "Coefficient",
                    ascending=True,
                )
            )

            axis.barh(
                class_df[
                    "Term"
                ],
                class_df[
                    "Coefficient"
                ],
                color="#4E8B57",
            )

            axis.set_title(
                readable_label(
                    class_name
                ),
                fontsize=10,
                fontweight="bold",
            )

            axis.set_xlabel(
                "Positive logistic-regression coefficient"
            )

            axis.grid(
                axis="x",
                alpha=0.2,
            )

        figure.suptitle(
            "Most Influential TF-IDF Terms by Cause",
            fontsize=14,
            fontweight="bold",
        )

        figure.tight_layout(
            rect=[
                0,
                0,
                1,
                0.96,
            ]
        )

        save_page(
            figure
        )


def add_disagreement_examples(
    title,
    dataframe,
    explanation,
    maximum_records=15,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )

        return

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        _,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 84,
                (
                    f"Example {example_number} | "
                    f"Record {row.get('SOURCE_RECORD_ID', '')} | "
                    f"{row.get('SEVERITY_GROUP', '')} | "
                    f"{row.get('TIME_PERIOD', '')}"
                ),
                (
                    "Human cause: "
                    f"{row.get('HUMAN_PRIMARY_CAUSE', '')}"
                ),
                (
                    "Logistic top-one: "
                    f"{row.get('LOGISTIC_TOP1_CAUSE', '')} "
                    f"({row.get('LOGISTIC_TOP1_PROBABILITY', np.nan):.3f})"
                ),
                (
                    "Logistic top-two alternative: "
                    f"{row.get('LOGISTIC_TOP2_CAUSE', '')} "
                    f"({row.get('LOGISTIC_TOP2_PROBABILITY', np.nan):.3f})"
                ),
                (
                    "Qwen cause: "
                    f"{row.get('QWEN_PRIMARY_CAUSE', '')} "
                    f"({row.get('QWEN_CAUSE_CONFIDENCE', np.nan):.3f})"
                ),
                (
                    "Human uncertain: "
                    f"{row.get('HUMAN_CAUSE_UNCERTAIN', False)}"
                ),
                (
                    "Human comments: "
                    f"{row.get('HUMAN_REVIEW_COMMENTS', '')}"
                ),
                "",
                "Narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.6,
        lines_per_page=48,
        wrap_width=104,
    )


# =============================================================================
# 30. GENERATE PDF REPORT
# =============================================================================

# Cover page.
figure = create_page()


figure.text(
    0.5,
    0.79,
    "TF-IDF Multiclass Logistic Regression",
    ha="center",
    fontsize=20,
    fontweight="bold",
)


figure.text(
    0.5,
    0.735,
    "Primary Accident-Cause Classification Baseline",
    ha="center",
    fontsize=13,
    color="#444444",
)


figure.text(
    0.5,
    0.695,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)


cover_lines = [
    f"Code Block 2 results: {block2_results_file}",
    f"Human benchmark workbook: {HUMAN_REVIEW_WORKBOOK}",
    "",
    f"High-confidence preliminary-label records: {len(training_df):,}",
    f"Training records used for model fitting: {len(X_train):,}",
    f"Held-out Block 2 records: {len(X_holdout):,}",
    f"Independent human benchmark records: {len(benchmark_df):,}",
    f"Human benchmark records not marked uncertain: "
    f"{len(human_evaluable_df):,}",
    f"Primary-cause classes included in training: "
    f"{training_df['PRELIMINARY_CAUSE_LABEL'].nunique():,}",
    f"Selected logistic-regression C: {best_c:g}",
    "",
    "Comparison purpose:",
    "Determine whether a transparent TF-IDF logistic-regression model can",
    "approximate the ontology-guided Qwen cause classifications and how both",
    "methods compare with the independent human review.",
]


figure.text(
    0.10,
    0.61,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=8.9,
    family="monospace",
    linespacing=1.24,
)


save_page(
    figure
)


# Executive summary.
add_text_pages(
    "Executive Summary",
    [
        "This experiment trained a multiclass logistic-regression classifier",
        "using TF-IDF representations of MSHA accident narratives.",
        "",
        "The preliminary training labels were the high-confidence primary-cause",
        "classifications from Code Block 2. These labels were generated by the",
        "frozen ontology-guided Qwen system and therefore are not independent",
        "ground truth.",
        "",
        "The classifier was first evaluated against a held-out portion of the",
        "Code Block 2 preliminary labels. This measures how closely the linear",
        "model reproduces the ontology-guided model.",
        "",
        "The principal evaluation used the 100 independently reviewed Block 3",
        "narratives. Logistic-regression predictions and saved Qwen predictions",
        "were compared with the human primary-cause labels.",
        "",
        "Exact agreement is the primary measure. Exact-or-related agreement is",
        "reported as a sensitivity measure because several short narratives can",
        "reasonably fit neighboring event-mechanism categories.",
        "",
        "Top-two logistic-regression accuracy is also reported. This measure is",
        "useful when the two highest-probability causes represent plausible",
        "alternatives for professional review.",
    ],
)


# Research design.
add_text_pages(
    "Experimental Design",
    [
        "1. Read the expanded Code Block 2 structured results.",
        "",
        "2. Retain successful acute-accident records with valid cause evidence,",
        f"   cause confidence of at least {MIN_CAUSE_CONFIDENCE:.2f}, and high",
        "   or medium cause support.",
        "",
        "3. Exclude insufficient-information, other, and uncertain labels.",
        "",
        "4. Remove exact narrative duplicates and all Block 3 benchmark records",
        "   from the training population.",
        "",
        "5. Divide the remaining records into a stratified training population",
        "   and held-out Block 2 test population.",
        "",
        "6. Vectorize narratives using TF-IDF unigrams and bigrams.",
        "",
        "7. Select the logistic-regression regularization strength using",
        "   stratified cross-validation and macro F1.",
        "",
        "8. Evaluate the selected model against held-out Code Block 2 labels.",
        "",
        "9. Evaluate both logistic regression and Qwen against the independent",
        "   human-review labels.",
        "",
        "10. Compare exact, related-category, top-two, category-level, severity,",
        "    time-period, and confidence-band results.",
    ],
)


# Training distributions.
add_table_pages(
    "Code Block 2 Training-Label Distribution",
    training_distribution_df,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=38,
)


add_table_pages(
    "Independent Human-Benchmark Distribution",
    human_distribution_df,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=38,
)


# CV results.
add_table_pages(
    "Regularization Cross-Validation Results",
    cv_results_df,
    rows_per_page=12,
    font_size=7.1,
    column_wrap_width=30,
)


add_cv_chart(
    cv_results_df
)


# Overall metrics.
add_table_pages(
    "Overall Classification Performance",
    overall_metrics_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=34,
    landscape=True,
)


add_metric_comparison_chart(
    overall_metrics_df
)


# Method outcomes.
add_table_pages(
    "Human-Benchmark Method Outcomes",
    method_outcome_summary_df,
    rows_per_page=12,
    font_size=7.2,
    column_wrap_width=50,
)


# Category performance.
add_table_pages(
    "Category-Level Human-Benchmark Performance",
    category_metrics_df,
    rows_per_page=28,
    font_size=6.7,
    column_wrap_width=29,
    landscape=True,
)


# Confusion matrices.
add_confusion_heatmap(
    "TF-IDF Logistic Regression versus Human Review",
    logistic_confusion_matrix,
    confusion_labels,
)


add_confusion_heatmap(
    "Qwen Ontology Model versus Human Review",
    qwen_confusion_matrix,
    confusion_labels,
)


# Confidence.
add_table_pages(
    "Performance by Model-Confidence Band",
    confidence_performance_df,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=36,
    landscape=True,
)


# Severity and time.
add_table_pages(
    "Human-Benchmark Performance by Injury Severity",
    severity_performance_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=30,
    landscape=True,
)


add_table_pages(
    "Human-Benchmark Performance by Time Period",
    time_performance_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=30,
    landscape=True,
)


# Influential terms.
add_text_pages(
    "Interpreting the Logistic-Regression Features",
    [
        "The positive logistic-regression coefficients identify terms that",
        "increase the relative probability of a primary-cause class.",
        "",
        "These coefficients provide transparency that is more difficult to",
        "obtain from a large language model. They can help determine whether the",
        "classifier learned meaningful event-language patterns or relied on",
        "administrative, injury-outcome, or overly specific vocabulary.",
        "",
        "A highly weighted term does not establish causal importance. It only",
        "indicates that the term was useful for separating preliminary training",
        "labels within this dataset.",
        "",
        "Terms should therefore be interpreted together with confusion matrices,",
        "representative narratives, and independent human-review performance.",
    ],
)


add_top_terms_pages(
    top_terms_df
)


# Disagreement examples.
logistic_only_correct_df = benchmark_df[
    benchmark_df[
        "LOGISTIC_ONLY_EXACT"
    ]
].copy()


qwen_only_correct_df = benchmark_df[
    benchmark_df[
        "QWEN_ONLY_EXACT"
    ]
].copy()


neither_correct_df = benchmark_df[
    benchmark_df[
        "NEITHER_EXACT"
    ]
].copy()


uncertain_cases_df = benchmark_df[
    benchmark_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


add_disagreement_examples(
    "Cases Correct Only for Logistic Regression",
    logistic_only_correct_df,
    (
        "The logistic-regression cause exactly matched the human label, while "
        "the Qwen classification did not. These examples show where a linear "
        "TF-IDF model may recognize a stable lexical pattern effectively."
    ),
    maximum_records=MAX_DISAGREEMENT_EXAMPLES,
)


add_disagreement_examples(
    "Cases Correct Only for Qwen",
    qwen_only_correct_df,
    (
        "The Qwen cause exactly matched the human label, while logistic "
        "regression did not. These examples may show where contextual sequence "
        "or event interpretation provides value beyond weighted terms."
    ),
    maximum_records=MAX_DISAGREEMENT_EXAMPLES,
)


add_disagreement_examples(
    "Cases Where Neither Model Exactly Matched",
    neither_correct_df,
    (
        "Neither model exactly matched the human-selected cause. The top-two "
        "logistic prediction and related-category results should be reviewed "
        "for narratives with plausible alternative mechanisms."
    ),
    maximum_records=MAX_DISAGREEMENT_EXAMPLES,
)


add_disagreement_examples(
    "Human-Uncertain Benchmark Cases",
    uncertain_cases_df,
    (
        "The human reviewer marked these primary-cause decisions uncertain. "
        "Differences should not be interpreted as definitive model errors "
        "without additional professional review."
    ),
    maximum_records=MAX_DISAGREEMENT_EXAMPLES,
)


# Interpretation.
add_text_pages(
    "Interpretation Framework",
    [
        "The held-out Block 2 evaluation answers:",
        "",
        "How closely can a TF-IDF logistic-regression model reproduce the",
        "high-confidence ontology-guided Qwen labels?",
        "",
        "The independent human benchmark answers:",
        "",
        "How closely do logistic regression and Qwen agree with the separate",
        "human review?",
        "",
        "A strong held-out result combined with weak human agreement would mean",
        "that logistic regression successfully learned the Qwen labeling pattern",
        "but that the preliminary labels did not generalize well to the human",
        "benchmark.",
        "",
        "If logistic regression approaches Qwen's human-benchmark performance,",
        "the simpler model may be useful for rapid, inexpensive primary-cause",
        "screening.",
        "",
        "If Qwen materially outperforms logistic regression, contextual language",
        "interpretation is providing value beyond TF-IDF term weighting.",
        "",
        "Top-two agreement is operationally useful when the classifier identifies",
        "two plausible neighboring mechanisms for a safety professional to",
        "review.",
    ],
)


# Limitations.
add_text_pages(
    "Limitations",
    [
        "1. Code Block 2 labels are model-generated preliminary labels rather",
        "   than fully independent ground truth.",
        "",
        "2. The human benchmark contains only 100 narratives and includes",
        "   uncertain classifications.",
        "",
        "3. Rare primary-cause categories may have too few records for stable",
        "   training or category-level metrics.",
        "",
        "4. TF-IDF emphasizes terms and short phrases but does not represent",
        "   complete event sequence or contextual meaning as directly as a",
        "   transformer model.",
        "",
        "5. Class weighting improves minority-class participation but may reduce",
        "   the probability calibration of the logistic-regression outputs.",
        "",
        "6. Exact-or-related agreement depends on predefined category",
        "   relationships and should not replace exact agreement.",
        "",
        "7. A high logistic-regression coefficient identifies predictive language",
        "   but does not establish root cause or corrective action.",
        "",
        "8. The experiment addresses primary accident cause only. Contributing",
        "   factors and prevention themes remain more dependent on missing",
        "   contextual information.",
    ],
)


# Business use.
add_text_pages(
    "Potential Business Application",
    [
        "The logistic-regression model provides a transparent and computationally",
        "efficient baseline for primary-cause screening.",
        "",
        "A practical tiered workflow could use logistic regression for routine",
        "high-volume classification and route uncertain records to the",
        "ontology-guided language model or professional review.",
        "",
        "Potential routing logic:",
        "",
        "1. Accept a logistic-regression classification when top-one probability",
        "   is high and substantially exceeds the second-highest probability.",
        "",
        "2. Route records with close top-one and top-two probabilities to Qwen or",
        "   human review.",
        "",
        "3. Route high-severity events and new or rare narrative patterns directly",
        "   to professional review regardless of model confidence.",
        "",
        "4. Use the influential-term tables to monitor whether predictions rely",
        "   on meaningful event language or undesirable shortcuts.",
        "",
        "The classifier should support narrative organization, similar-case",
        "retrieval, and preliminary triage. It should not replace formal accident",
        "investigation or professional safety judgment.",
    ],
)


# Technical appendix.
add_text_pages(
    "Technical Appendix",
    [
        f"Code Block 2 results: {block2_results_file}",
        f"Human-review workbook: {HUMAN_REVIEW_WORKBOOK}",
        f"Training records after filters: {len(training_df):,}",
        f"Training records used for fit: {len(X_train):,}",
        f"Internal held-out records: {len(X_holdout):,}",
        f"Human benchmark records: {len(benchmark_df):,}",
        f"Human benchmark certain records: {len(human_evaluable_df):,}",
        f"Training classes: {training_df['PRELIMINARY_CAUSE_LABEL'].nunique():,}",
        f"Minimum cause confidence: {MIN_CAUSE_CONFIDENCE:.2f}",
        f"Selected C: {best_c:g}",
        f"Cross-validation folds: {cv_folds}",
        f"TF-IDF maximum features: {TFIDF_MAX_FEATURES:,}",
        f"TF-IDF n-gram range: {TFIDF_NGRAM_RANGE}",
        f"TF-IDF minimum document frequency: {TFIDF_MIN_DF}",
        f"Class weighting: {class_weight_value}",
        f"Random state: {RANDOM_STATE}",
        "",
        f"PDF report: {os.path.abspath(OUTPUT_PDF)}",
        f"Model file: {os.path.abspath(MODEL_FILE)}",
        f"Human benchmark output: {os.path.abspath(HUMAN_BENCHMARK_CSV)}",
        f"Held-out predictions: {os.path.abspath(HELDOUT_PREDICTIONS_CSV)}",
        f"Category metrics: {os.path.abspath(CATEGORY_METRICS_CSV)}",
        f"Top terms: {os.path.abspath(TOP_TERMS_CSV)}",
        f"Training summary: {os.path.abspath(TRAINING_SUMMARY_JSON)}",
    ],
    font_size=8.2,
)


# =============================================================================
# 31. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()


metadata[
    "Title"
] = (
    "MSHA TF-IDF Multiclass Logistic Regression Comparison Report"
)


metadata[
    "Author"
] = (
    "Python scikit-learn analysis"
)


metadata[
    "Subject"
] = (
    "Comparison of TF-IDF logistic regression, ontology-guided Qwen "
    "classification, and independent human review for MSHA primary causes"
)


metadata[
    "Keywords"
] = (
    "MSHA, TF-IDF, logistic regression, natural language processing, "
    "primary accident causes, Qwen, human validation"
)


metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 32. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("TF-IDF LOGISTIC REGRESSION ANALYSIS COMPLETE")
print("=" * 95)


print(
    f"Code Block 2 training records: "
    f"{len(training_df):,}"
)


print(
    f"Training records used for fit: "
    f"{len(X_train):,}"
)


print(
    f"Held-out Block 2 records: "
    f"{len(X_holdout):,}"
)


print(
    f"Independent human benchmark records: "
    f"{len(benchmark_df):,}"
)


print(
    f"Selected regularization C: "
    f"{best_c:g}"
)


print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)


print(
    f"Saved model: "
    f"{os.path.abspath(MODEL_FILE)}"
)


print(
    f"Human benchmark predictions: "
    f"{os.path.abspath(HUMAN_BENCHMARK_CSV)}"
)


print(
    f"Held-out predictions: "
    f"{os.path.abspath(HELDOUT_PREDICTIONS_CSV)}"
)


print("=" * 95)


# =============================================================================
# 33. NOTEBOOK DISPLAYS
# =============================================================================

display(
    cv_results_df
)


display(
    overall_metrics_df
)


display(
    method_outcome_summary_df
)


display(
    category_metrics_df
)


display(
    confidence_performance_df
)


display(
    severity_performance_df
)


display(
    top_terms_df.head(
        60
    )
)
````

````python
# =============================================================================
# BLOCK 1
# MSHA EXPANDED ONTOLOGY DEVELOPMENT AND STRESS TEST
#
# PURPOSE
# -------
# Expand the earlier 100-narrative Qwen trial to approximately 1,000
# development narratives in order to:
#
#   1. Stress-test the cause, contributing-factor, and prevention ontology.
#   2. Identify overused, unused, ambiguous, or missing categories.
#   3. Evaluate schema validity, evidence quality, support levels, and
#      factor-prevention alignment.
#   4. Identify representative and questionable examples.
#   5. Produce recommendations before freezing the ontology for Block 2.
#
# INPUT
# -----
# A pandas DataFrame named `accidents` is already loaded.
#
# OUTPUTS
# -------
# 1. MSHA_Block1_Ontology_Development_Report.pdf
# 2. MSHA_Block1_Development_Checkpoint.jsonl
# 3. MSHA_Block1_Development_Record_IDs.json
#
# The ID file is important. Block 2 should read it and exclude all Block 1
# records so that the 2,000-record analysis remains unseen.
#
# IMPORTANT
# ---------
# This is a DEVELOPMENT sample, not the final unseen evaluation sample.
# Results from this block may be used to revise:
#
#   - the ontology;
#   - category definitions;
#   - prompt instructions;
#   - aliases;
#   - compatibility rules;
#   - review rules.
#
# After Block 1 is complete, those items should be frozen before Block 2.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import time
import math
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import (
    BaseModel,
    Field,
    model_validator,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from matplotlib.backends.backend_pdf import PdfPages


warnings.filterwarnings("ignore")


# =============================================================================
# 2. USER SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

TEXT_COLUMN = "NARRATIVE"

DATE_COLUMN = "ACCIDENT_DT"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

# Development sample.
TOTAL_SAMPLE_SIZE = 1000

TARGET_PER_SEVERITY_GROUP = 200

RANDOM_STATE = 42

# Narrative-quality rules.
MIN_NARRATIVE_WORDS = 8

TRUNCATION_CHARACTER_THRESHOLD = 383

# Model-generation settings.
MAX_NEW_TOKENS = 550

MAX_RETRIES = 3

LOW_CONFIDENCE_THRESHOLD = 0.70

RUN_FUNCTIONAL_TEST = True

REPROCESS_COMPLETED = False

PROMPT_VERSION = "MSHA_BLOCK1_ONTOLOGY_DEVELOPMENT_V1.0"

# Output files.
OUTPUT_PDF = (
    "MSHA_Block1_Ontology_Development_Report.pdf"
)

CHECKPOINT_JSONL = (
    "MSHA_Block1_Development_Checkpoint.jsonl"
)

DEVELOPMENT_IDS_JSON = (
    "MSHA_Block1_Development_Record_IDs.json"
)


# =============================================================================
# 3. VERIFY THE NOTEBOOK ENVIRONMENT
# =============================================================================

print("=" * 95)
print("BLOCK 1 - RUNTIME ENVIRONMENT")
print("=" * 95)

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)
print("Pydantic:", pydantic.__version__)
print("Pydantic Core:", pydantic_core.__version__)


if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. "
        "Attach the Azure ML notebook to a GPU compute instance."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0).total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3


print("\nGPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


# =============================================================================
# 4. FINAL DEVELOPMENT ONTOLOGY
#
# These categories remain DEVELOPMENT categories during Block 1.
# They may be revised after reviewing the Block 1 report.
# They should be frozen before Block 2.
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NOT_STATED",
]


PREVENTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 5. CATEGORY DEFINITIONS
#
# These definitions are inserted into the prompt and included in the report.
# =============================================================================

CAUSE_DEFINITIONS = {
    "SLIP_TRIP_OR_FALL":
        "A person loses footing, balance, or elevation and falls or nearly falls.",

    "AWKWARD_MOVEMENT_OR_BODY_MOTION":
        "An injury caused primarily by twisting, bending, reaching, turning, or another body motion without a distinct fall.",

    "STRUCK_BY_MOVING_OBJECT":
        "A moving, falling, flying, swinging, or projected object contacts the person.",

    "STRUCK_AGAINST_STATIONARY_OBJECT":
        "The person moves into or strikes a fixed object or surface.",

    "CAUGHT_IN_OR_BETWEEN":
        "A body part or person is pinched, crushed, trapped, entangled, or pulled between objects.",

    "MATERIAL_RELEASE_OR_SHIFT":
        "Material unexpectedly shifts, rolls, slides, spills, or is released without a more specific ground-failure mechanism.",

    "GROUND_OR_ROOF_FAILURE":
        "Roof, rib, highwall, face, rock, or ground fails, falls, or moves uncontrollably.",

    "UNEXPECTED_EQUIPMENT_MOVEMENT":
        "A machine or component moves unexpectedly without a stated equipment failure or loss of operator control.",

    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT":
        "Control of a tool, machine, handle, strap, load, or equipment component is lost.",

    "MANUAL_HANDLING_OR_OVEREXERTION":
        "Lifting, carrying, pushing, pulling, holding, or exerting force produces the reported event.",

    "PRESSURE_OR_STORED_ENERGY_RELEASE":
        "Pressurized fluid, compressed material, tension, spring force, hydraulic force, or stored energy is released.",

    "ELECTRICAL_CONTACT":
        "Electrical current, energized conductors, arc contact, or electrical equipment causes the event.",

    "FIRE_EXPLOSION_OR_THERMAL_EVENT":
        "Fire, ignition, explosion, flame, heat, hot material, sparks, or thermal exposure causes the event.",

    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT":
        "A vehicle or mobile-equipment collision, overturn, runaway, roadway departure, or transportation event occurs.",

    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE":
        "Chemical, dust, toxic material, fumes, heat, noise, or another occupational exposure is the main event.",

    "ERGONOMIC_OR_REPETITIVE_EXPOSURE":
        "Repeated motion, cumulative trauma, posture, or prolonged ergonomic exposure is the main event.",

    "OTHER":
        "The event is stated but does not fit another cause category.",

    "INSUFFICIENT_INFORMATION":
        "The narrative does not contain enough information to identify the event mechanism.",
}


FACTOR_DEFINITIONS = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
        "Material, a load, or a component is unstable, unsupported, loose, or improperly positioned.",

    "INADEQUATE_SECUREMENT_OR_RESTRAINT":
        "A load, component, material, or object is not adequately secured, restrained, blocked, or supported.",

    "LINE_OF_FIRE_BODY_POSITION":
        "The person is positioned in the path of moving material, equipment, tools, falling objects, or released energy.",

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
        "Grease, mud, water, ice, debris, poor footing, uneven surfaces, or access conditions affect movement or footing.",

    "UNSTABLE_GROUND_OR_ROOF_CONDITION":
        "Unstable roof, rib, face, highwall, rock, or ground is directly described.",

    "INADEQUATE_GUARDING_OR_BARRIER":
        "A missing, removed, bypassed, open, or inadequate guard or barrier is described.",

    "ENERGY_ISOLATION_DEFICIENCY":
        "Hazardous or stored energy was not fully isolated, blocked, locked out, dissipated, or controlled.",

    "EQUIPMENT_CONDITION_OR_FAILURE":
        "The narrative states equipment failure, breakage, leakage, malfunction, damage, wear, or defective operation.",

    "PROCEDURE_OR_WORK_METHOD":
        "The described task sequence, handling method, maintenance method, or work practice contributed to the event.",

    "INADEQUATE_HAZARD_RECOGNITION":
        "The narrative directly indicates that a hazard was not recognized or evaluated.",

    "COMMUNICATION_OR_COORDINATION":
        "The event involves inadequate notification, signaling, communication, coordination, or spotter use.",

    "TRAINING_OR_EXPERIENCE":
        "The narrative directly states insufficient training, experience, proficiency, or qualification.",

    "PPE_OR_PROTECTIVE_CONTROL":
        "The narrative directly identifies missing, inadequate, damaged, or incorrect protective equipment or control.",

    "ENVIRONMENTAL_CONDITION":
        "Weather, visibility, lighting, temperature, noise, or another environmental condition contributed.",

    "WORKSPACE_OR_ACCESS_LIMITATION":
        "Restricted clearance, difficult access, poor layout, or limited workspace contributed.",

    "OTHER":
        "A stated contributing factor does not fit another category.",

    "NOT_STATED":
        "No contributing factor is directly stated or strongly supported by the narrative.",
}


PREVENTION_DEFINITIONS = {
    "SECURE_OR_REPOSITION_MATERIAL":
        "Secure, block, support, restrain, or safely reposition material or components.",

    "ESTABLISH_LINE_OF_FIRE_EXCLUSION":
        "Keep people outside the path of moving material, equipment, tools, or released energy.",

    "ISOLATE_OR_CONTROL_STORED_ENERGY":
        "Lock out, block, bleed, dissipate, restrain, or otherwise control hazardous energy.",

    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT":
        "Inspect, repair, or replace equipment when failure, damage, wear, leakage, or malfunction is supported.",

    "INSTALL_OR_IMPROVE_GUARDING":
        "Install, restore, redesign, or improve guards and barriers.",

    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING":
        "Correct debris, grease, mud, water, uneven footing, poor access, or other surface conditions.",

    "IMPROVE_GROUND_CONTROL_OR_SCALING":
        "Improve examination, support, scaling, exclusion, or control of roof, rib, highwall, or ground.",

    "REVISE_PROCEDURE_OR_JOB_PLAN":
        "Revise the work method, task sequence, maintenance method, or written job plan.",

    "CONDUCT_PRE_JOB_HAZARD_REVIEW":
        "Perform a structured hazard assessment before beginning or changing the task.",

    "PROVIDE_TARGETED_TRAINING":
        "Provide task-specific training when the narrative supports a training or proficiency need.",

    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE":
        "Improve signaling, notification, coordination, visibility, or spotter practices.",

    "USE_MECHANICAL_HANDLING_ASSISTANCE":
        "Use lifting, positioning, or material-handling assistance to reduce manual exertion or exposure.",

    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT":
        "Provide or improve protective equipment when supported by the narrative.",

    "REDESIGN_WORKSPACE_OR_ACCESS":
        "Improve clearance, access, layout, reach, working position, or physical workspace.",

    "ENGINEERING_REVIEW_REQUIRED":
        "The event suggests a technical control need, but the narrative does not support one specific engineering action.",

    "INVESTIGATION_REQUIRED":
        "Additional investigation is required before assigning a specific prevention theme.",

    "INSUFFICIENT_INFORMATION":
        "The narrative does not support a defensible prevention theme.",
}


# =============================================================================
# 6. CATEGORY ALIASES FOR MINOR MODEL VARIATIONS
# =============================================================================

CATEGORY_ALIASES = {
    "UNSTATED":
        "NOT_STATED",

    "NOT_PROVIDED":
        "NOT_STATED",

    "NONE_STATED":
        "NOT_STATED",

    "NOT_SPECIFIED":
        "NOT_STATED",

    "UNKNOWN":
        "INSUFFICIENT_INFORMATION",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "FIRE_EXPLOSION_THERMAL_EVENT":
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",

    "PRESSURE_STORED_ENERGY_RELEASE":
        "PRESSURE_OR_STORED_ENERGY_RELEASE",

    "REVISE_PROCEDURE":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "PRE_JOB_HAZARD_REVIEW":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "EQUIPMENT_REPAIR_OR_INSPECTION":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "IMPROVE_HOUSEKEEPING":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "GROUND_CONTROL_REVIEW":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
}


SUPPORT_LEVEL_ALIASES = {
    "HIGH":
        "HIGH",

    "DIRECT":
        "HIGH",

    "EXPLICIT":
        "HIGH",

    "EXPLICITLY_STATED":
        "HIGH",

    "SUPPORTED":
        "HIGH",

    "MEDIUM":
        "MEDIUM",

    "MODERATE":
        "MEDIUM",

    "IMPLIED":
        "MEDIUM",

    "STRONGLY_IMPLIED":
        "MEDIUM",

    "LOW":
        "LOW",

    "WEAK":
        "LOW",

    "UNCERTAIN":
        "LOW",

    "NOT_APPLICABLE":
        "LOW",

    "NONE":
        "LOW",

    "":
        "LOW",
}


# =============================================================================
# 7. FACTOR-PREVENTION COMPATIBILITY RULES
#
# An unusual pairing is flagged for review but is not automatically replaced.
# =============================================================================

FACTOR_PREVENTION_ALIGNMENT = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_SECUREMENT_OR_RESTRAINT": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    },

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },

    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_GUARDING_OR_BARRIER": {
        "INSTALL_OR_IMPROVE_GUARDING",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "ENERGY_ISOLATION_DEFICIENCY": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    },

    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    },

    "INADEQUATE_HAZARD_RECOGNITION": {
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "COMMUNICATION_OR_COORDINATION": {
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    },

    "TRAINING_OR_EXPERIENCE": {
        "PROVIDE_TARGETED_TRAINING",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "PPE_OR_PROTECTIVE_CONTROL": {
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "ENVIRONMENTAL_CONDITION": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "WORKSPACE_OR_ACCESS_LIMITATION": {
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "NOT_STATED": {
        "INVESTIGATION_REQUIRED",
        "ENGINEERING_REVIEW_REQUIRED",
        "INSUFFICIENT_INFORMATION",
    },
}


# =============================================================================
# 8. PYDANTIC RESPONSE SCHEMA
# =============================================================================

class IncidentExtraction(BaseModel):

    primary_cause: str

    primary_contributing_factor: str

    prevention_theme: str

    cause_evidence_text: str = ""

    factor_evidence_text: str = ""

    prevention_rationale: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    prevention_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    factor_support_level: str = "LOW"

    prevention_support_level: str = "LOW"

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_ontology(self):

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                "primary_cause must be selected from CAUSE_CATEGORIES. "
                f"Received: {self.primary_cause}"
            )

        if (
            self.primary_contributing_factor
            not in FACTOR_CATEGORIES
        ):

            raise ValueError(
                "primary_contributing_factor must be selected from "
                "FACTOR_CATEGORIES. "
                f"Received: {self.primary_contributing_factor}"
            )

        if self.prevention_theme not in PREVENTION_CATEGORIES:

            raise ValueError(
                "prevention_theme must be selected from "
                "PREVENTION_CATEGORIES. "
                f"Received: {self.prevention_theme}"
            )

        for support_field in [
            "cause_support_level",
            "factor_support_level",
            "prevention_support_level",
        ]:

            support_value = getattr(
                self,
                support_field,
            )

            if support_value not in {
                "HIGH",
                "MEDIUM",
                "LOW",
            }:

                raise ValueError(
                    f"{support_field} must be HIGH, MEDIUM, or LOW. "
                    f"Received: {support_value}"
                )

        return self


# =============================================================================
# 9. PREPARE THE SOURCE DATA
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"Column {TEXT_COLUMN!r} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"Column {SEVERITY_COLUMN!r} was not found."
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


if DATE_COLUMN in df.columns:

    df[DATE_COLUMN] = pd.to_datetime(
        df[DATE_COLUMN],
        errors="coerce",
    )

    df["ACCIDENT_YEAR"] = (
        df[DATE_COLUMN]
        .dt.year
        .astype("Int64")
    )

else:

    if "CAL_YR" not in df.columns:

        raise KeyError(
            "Neither ACCIDENT_DT nor CAL_YR is available."
        )

    df["ACCIDENT_YEAR"] = pd.to_numeric(
        df["CAL_YR"],
        errors="coerce",
    ).astype("Int64")


df["_NARRATIVE_NORMALIZED_FOR_DUPLICATES"] = (
    df[TEXT_COLUMN]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN]
    .str.len()
)


df["_MAY_BE_TRUNCATED"] = (
    df["_CHARACTER_COUNT"]
    >= TRUNCATION_CHARACTER_THRESHOLD
)


# Remove blank and extremely short narratives.
df = df[
    df["_WORD_COUNT"]
    >= MIN_NARRATIVE_WORDS
].copy()


# Remove exact duplicate narratives while preserving the first record.
before_duplicate_count = len(
    df
)


df = df.drop_duplicates(
    subset=[
        "_NARRATIVE_NORMALIZED_FOR_DUPLICATES"
    ],
    keep="first",
).copy()


duplicate_records_removed = (
    before_duplicate_count
    - len(df)
)


# =============================================================================
# 10. SEVERITY GROUPS
# =============================================================================

def assign_severity_group(
    degree_injury,
):

    value = (
        str(degree_injury)
        .strip()
        .upper()
    )

    if "FATAL" in value:

        return "Fatality"

    if any(
        phrase in value
        for phrase in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "Permanent disability"

    if any(
        phrase in value
        for phrase in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "Days away"

    if any(
        phrase in value
        for phrase in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "Restricted or transferred"

    return "No lost time or other"


df["SEVERITY_GROUP"] = (
    df[SEVERITY_COLUMN]
    .map(assign_severity_group)
)


SEVERITY_ORDER = [
    "Fatality",
    "Permanent disability",
    "Days away",
    "Restricted or transferred",
    "No lost time or other",
]


# =============================================================================
# 11. BROAD RECORD-TYPE ROUTING
#
# The primary ontology is intended for acute accidents.
#
# This routing is intentionally conservative. It uses existing structured
# fields only to avoid forcing obvious illness or noninjury records into the
# acute-accident ontology.
# =============================================================================

def assign_record_type(
    row,
):

    degree = str(
        row.get(
            "DEGREE_INJURY",
            "",
        )
    ).upper()

    classification = str(
        row.get(
            "CLASSIFICATION",
            "",
        )
    ).upper()

    nature = str(
        row.get(
            "NATURE_INJURY",
            "",
        )
    ).upper()

    narrative = str(
        row.get(
            TEXT_COLUMN,
            "",
        )
    ).upper()

    illness_terms = [
        "OCCUPATIONAL ILLNESS",
        "HEARING LOSS",
        "PNEUMOCONIOSIS",
        "SILICOSIS",
        "RESPIRATORY",
        "REPEATED TRAUMA",
        "DUST DISEASE",
        "POISONING",
        "DERMATITIS",
    ]

    natural_terms = [
        "HEART ATTACK",
        "CARDIAC",
        "NATURAL CAUSE",
        "SEIZURE",
        "STROKE",
    ]

    noninjury_terms = [
        "ACCIDENT ONLY",
        "NO INJURY",
        "NO INJURIES",
        "NO PERSON INJURED",
    ]

    combined_text = " ".join(
        [
            degree,
            classification,
            nature,
            narrative,
        ]
    )

    if any(
        term in combined_text
        for term in illness_terms
    ):

        return "Occupational illness"

    if any(
        term in combined_text
        for term in natural_terms
    ):

        return "Natural or nonoccupational"

    if any(
        term in combined_text
        for term in noninjury_terms
    ):

        return "Noninjury accident"

    return "Acute accident"


df["RECORD_TYPE"] = (
    df.apply(
        assign_record_type,
        axis=1,
    )
)


# Primary development population.
eligible_df = df[
    df["RECORD_TYPE"]
    == "Acute accident"
].copy()


if eligible_df.empty:

    raise ValueError(
        "No acute-accident narratives remained after routing."
    )


# =============================================================================
# 12. CREATE BROAD TIME PERIODS
# =============================================================================

valid_years = (
    eligible_df[
        "ACCIDENT_YEAR"
    ]
    .dropna()
    .astype(int)
)


if valid_years.empty:

    eligible_df[
        "TIME_PERIOD"
    ] = "Unknown period"

    period_order = [
        "Unknown period"
    ]

else:

    unique_years = np.sort(
        valid_years.unique()
    )

    if len(unique_years) >= 3:

        year_quantiles = np.quantile(
            unique_years,
            [
                0.0,
                1.0 / 3.0,
                2.0 / 3.0,
                1.0,
            ],
        )

        period_start = int(
            unique_years.min()
        )

        first_end = int(
            math.floor(
                year_quantiles[1]
            )
        )

        second_end = int(
            math.floor(
                year_quantiles[2]
            )
        )

        period_end = int(
            unique_years.max()
        )

        if first_end >= second_end:

            first_end = int(
                unique_years[
                    len(unique_years) // 3
                ]
            )

            second_end = int(
                unique_years[
                    min(
                        2 * len(unique_years) // 3,
                        len(unique_years) - 1,
                    )
                ]
            )

        def assign_time_period(year):

            if pd.isna(year):

                return "Unknown period"

            year = int(year)

            if year <= first_end:

                return (
                    f"Early: "
                    f"{period_start}-{first_end}"
                )

            if year <= second_end:

                return (
                    f"Middle: "
                    f"{first_end + 1}-{second_end}"
                )

            return (
                f"Recent: "
                f"{second_end + 1}-{period_end}"
            )

        eligible_df[
            "TIME_PERIOD"
        ] = (
            eligible_df[
                "ACCIDENT_YEAR"
            ]
            .map(assign_time_period)
        )

        period_order = [
            f"Early: {period_start}-{first_end}",
            f"Middle: {first_end + 1}-{second_end}",
            f"Recent: {second_end + 1}-{period_end}",
            "Unknown period",
        ]

    else:

        eligible_df[
            "TIME_PERIOD"
        ] = (
            eligible_df[
                "ACCIDENT_YEAR"
            ]
            .astype("string")
            .fillna(
                "Unknown period"
            )
        )

        period_order = sorted(
            eligible_df[
                "TIME_PERIOD"
            ]
            .dropna()
            .unique()
            .tolist()
        )


# =============================================================================
# 13. SECTOR GROUP
# =============================================================================

def assign_sector(
    value,
):

    normalized = str(
        value
    ).strip().upper()

    if normalized == "C":

        return "Coal"

    if normalized == "M":

        return "Metal/nonmetal"

    return "Unknown"


if "COAL_METAL_IND" in eligible_df.columns:

    eligible_df["SECTOR_GROUP"] = (
        eligible_df[
            "COAL_METAL_IND"
        ]
        .map(assign_sector)
    )

else:

    eligible_df[
        "SECTOR_GROUP"
    ] = "Unknown"


# =============================================================================
# 14. CREATE THE 1,000-RECORD DEVELOPMENT SAMPLE
#
# The sample is balanced first by severity and then distributed across broad
# time periods within each severity group.
# =============================================================================

def stratified_development_sample(
    data,
    severity_order,
    target_per_severity,
    total_sample_size,
    random_state,
):

    selected_indexes = []

    sample_summary_records = []

    for severity_number, severity_group in enumerate(
        severity_order
    ):

        severity_data = data[
            data["SEVERITY_GROUP"]
            == severity_group
        ].copy()

        if severity_data.empty:

            sample_summary_records.append(
                {
                    "Severity group":
                        severity_group,

                    "Available":
                        0,

                    "Target":
                        target_per_severity,

                    "Selected":
                        0,
                }
            )

            continue

        available_periods = (
            severity_data[
                "TIME_PERIOD"
            ]
            .value_counts()
            .index
            .tolist()
        )

        period_target = max(
            1,
            target_per_severity
            // max(
                len(available_periods),
                1,
            )
        )

        severity_selected = []

        for period_number, period_name in enumerate(
            available_periods
        ):

            period_data = severity_data[
                severity_data[
                    "TIME_PERIOD"
                ]
                == period_name
            ]

            n_select = min(
                period_target,
                len(period_data),
            )

            if n_select > 0:

                period_sample = period_data.sample(
                    n=n_select,

                    random_state=(
                        random_state
                        + severity_number * 100
                        + period_number
                    ),
                )

                severity_selected.extend(
                    period_sample.index.tolist()
                )

        remaining_for_severity = (
            target_per_severity
            - len(
                severity_selected
            )
        )

        if remaining_for_severity > 0:

            remaining_pool = severity_data.drop(
                index=severity_selected,
                errors="ignore",
            )

            fill_count = min(
                remaining_for_severity,
                len(
                    remaining_pool
                ),
            )

            if fill_count > 0:

                fill_sample = remaining_pool.sample(
                    n=fill_count,

                    random_state=(
                        random_state
                        + severity_number * 1000
                    ),
                )

                severity_selected.extend(
                    fill_sample.index.tolist()
                )

        selected_indexes.extend(
            severity_selected
        )

        sample_summary_records.append(
            {
                "Severity group":
                    severity_group,

                "Available":
                    len(
                        severity_data
                    ),

                "Target":
                    target_per_severity,

                "Selected":
                    len(
                        severity_selected
                    ),
            }
        )

    selected_indexes = list(
        dict.fromkeys(
            selected_indexes
        )
    )

    sample = data.loc[
        selected_indexes
    ].copy()

    remaining_needed = (
        total_sample_size
        - len(sample)
    )

    if remaining_needed > 0:

        remaining_pool = data.drop(
            index=selected_indexes,
            errors="ignore",
        )

        additional_count = min(
            remaining_needed,
            len(
                remaining_pool
            ),
        )

        if additional_count > 0:

            additional_sample = (
                remaining_pool.sample(
                    n=additional_count,

                    random_state=(
                        random_state
                        + 9999
                    ),
                )
            )

            sample = pd.concat(
                [
                    sample,
                    additional_sample,
                ],
                axis=0,
            )

    if len(sample) > total_sample_size:

        sample = sample.sample(
            n=total_sample_size,

            random_state=random_state,
        )

    sample = (
        sample.sample(
            frac=1,

            random_state=random_state,
        )
        .reset_index(drop=False)
        .rename(
            columns={
                "index":
                    "_ORIGINAL_INDEX"
            }
        )
    )

    return (
        sample,
        pd.DataFrame(
            sample_summary_records
        ),
    )


development_df, sampling_summary = (
    stratified_development_sample(
        data=eligible_df,

        severity_order=SEVERITY_ORDER,

        target_per_severity=(
            TARGET_PER_SEVERITY_GROUP
        ),

        total_sample_size=(
            TOTAL_SAMPLE_SIZE
        ),

        random_state=RANDOM_STATE,
    )
)


if development_df.empty:

    raise ValueError(
        "The development sample is empty."
    )


print("\n" + "=" * 95)
print("DEVELOPMENT SAMPLE")
print("=" * 95)

print(
    f"Acute-accident narratives available: "
    f"{len(eligible_df):,}"
)

print(
    f"Development narratives selected: "
    f"{len(development_df):,}"
)

print(
    f"Exact duplicate narratives removed: "
    f"{duplicate_records_removed:,}"
)

print("\nSeverity sampling summary:")

print(
    sampling_summary.to_string(
        index=False
    )
)


# =============================================================================
# 15. SAVE DEVELOPMENT RECORD IDS FOR BLOCK 2 EXCLUSION
# =============================================================================

development_record_ids = (
    development_df[
        id_column
    ]
    .astype(str)
    .tolist()
)


development_id_payload = {
    "created_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "purpose":
        "Block 1 ontology development sample; exclude these IDs from Block 2.",

    "id_column":
        id_column,

    "record_count":
        len(
            development_record_ids
        ),

    "record_ids":
        development_record_ids,
}


with open(
    DEVELOPMENT_IDS_JSON,
    "w",
    encoding="utf-8",
) as output_file:

    json.dump(
        development_id_payload,
        output_file,
        indent=2,
        ensure_ascii=False,
    )


# =============================================================================
# 16. BUILD THE SYSTEM PROMPT
# =============================================================================

def format_definitions(
    definition_mapping,
):

    return "\n".join(
        [
            (
                f"- {category}: "
                f"{definition}"
            )

            for category, definition
            in definition_mapping.items()
        ]
    )


SYSTEM_PROMPT = f"""
You analyze short MSHA accident narratives.

This is reported-event extraction, not verified root-cause analysis.

Use only information stated in or directly supported by the narrative.

Return one primary cause, one primary contributing factor, and one
preliminary prevention theme.

CAUSE CATEGORIES AND DEFINITIONS
{format_definitions(CAUSE_DEFINITIONS)}

CONTRIBUTING-FACTOR CATEGORIES AND DEFINITIONS
{format_definitions(FACTOR_DEFINITIONS)}

PREVENTION CATEGORIES AND DEFINITIONS
{format_definitions(PREVENTION_DEFINITIONS)}

CRITICAL RULES

1. Cause, contributing factor, and prevention theme are different concepts.

2. The cause is the reported event mechanism or energy-transfer event.

3. The contributing factor is the most important stated or directly supported
   condition, body position, work method, equipment condition, ground
   condition, access limitation, environmental condition, or coordination
   issue.

4. The prevention theme is a preliminary control category supported by the
   narrative. It is not an approved corrective action.

5. Use POOR_HOUSEKEEPING_OR_SURFACE_CONDITION only when the narrative
   describes debris, grease, mud, water, ice, uneven footing, or another
   condition affecting walking, footing, or access.

6. Do not select housekeeping merely because the word slipped appears.

7. If a hand slips from a strap, tool, handle, or component, do not treat
   that as a walking-surface or housekeeping event.

8. Use CAUGHT_IN_OR_BETWEEN when a body part is pinched, crushed, trapped,
   entangled, pulled into, or caught between objects.

9. Use STRUCK_AGAINST_STATIONARY_OBJECT when the person moves into a fixed
   object.

10. Use STRUCK_BY_MOVING_OBJECT when a falling, flying, swinging, rolling,
    or projected object contacts the person.

11. Use GROUND_OR_ROOF_FAILURE for roof, rib, face, highwall, rock, or ground
    failure.

12. Use FIRE_EXPLOSION_OR_THERMAL_EVENT for flame, heat, ignition, explosion,
    hot material, or thermal injury.

13. Use ELECTRICAL_CONTACT only when electrical current, an energized
    conductor, arc contact, or electrical energy causes the event.

14. Do not recommend REPAIR_REPLACE_OR_INSPECT_EQUIPMENT unless the narrative
    states or strongly describes failure, breakage, leakage, damage, wear,
    defect, or malfunction.

15. Do not infer inadequate training, supervision, procedures, PPE, guarding,
    or communication merely because an event occurred.

16. If no contributing factor is stated or directly supported, use
    NOT_STATED.

17. If the event mechanism cannot be identified, use
    INSUFFICIENT_INFORMATION.

18. Copy evidence excerpts verbatim from the narrative whenever possible.

19. Cause evidence must support the cause.

20. Factor evidence must support the contributing factor.

21. Support levels:
    HIGH = directly and explicitly supported.
    MEDIUM = strongly implied by the event description.
    LOW = incomplete, ambiguous, or dependent on uncertain interpretation.

22. Confidence values are model self-assessments, not calibrated
    probabilities.

23. Return only one valid JSON object.

REQUIRED JSON KEYS

{{
  "primary_cause": "one allowed cause category",
  "primary_contributing_factor": "one allowed factor category",
  "prevention_theme": "one allowed prevention category",
  "cause_evidence_text": "short verbatim excerpt or empty string",
  "factor_evidence_text": "short verbatim excerpt or empty string",
  "prevention_rationale": "one short sentence",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "prevention_confidence": 0.0,
  "cause_support_level": "HIGH, MEDIUM, or LOW",
  "factor_support_level": "HIGH, MEDIUM, or LOW",
  "prevention_support_level": "HIGH, MEDIUM, or LOW",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 17. JSON AND NORMALIZATION UTILITIES
# =============================================================================

def normalize_token(
    value,
):

    return (
        str(value)
        .strip()
        .upper()
        .replace("-", "_")
        .replace(" ", "_")
        .replace("/", "_")
    )


def normalize_category(
    value,
):

    normalized = normalize_token(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def normalize_support_level(
    value,
):

    normalized = normalize_token(
        value
    )

    return SUPPORT_LEVEL_ALIASES.get(
        normalized,
        "LOW",
    )


def normalize_response_dictionary(
    response_dictionary,
):

    cleaned = dict(
        response_dictionary
    )

    for category_field in [
        "primary_cause",
        "primary_contributing_factor",
        "prevention_theme",
    ]:

        cleaned[
            category_field
        ] = normalize_category(
            cleaned.get(
                category_field,
                "",
            )
        )

    for support_field in [
        "cause_support_level",
        "factor_support_level",
        "prevention_support_level",
    ]:

        cleaned[
            support_field
        ] = normalize_support_level(
            cleaned.get(
                support_field,
                "",
            )
        )

    for confidence_field in [
        "cause_confidence",
        "factor_confidence",
        "prevention_confidence",
    ]:

        try:

            cleaned[
                confidence_field
            ] = float(
                cleaned.get(
                    confidence_field,
                    0.0,
                )
            )

        except Exception:

            cleaned[
                confidence_field
            ] = 0.0

    for text_field in [
        "cause_evidence_text",
        "factor_evidence_text",
        "prevention_rationale",
        "insufficient_information_reason",
        "reviewer_note",
    ]:

        cleaned[
            text_field
        ] = str(
            cleaned.get(
                text_field,
                "",
            )
            or ""
        ).strip()

    return cleaned


def extract_json_object(
    text,
):

    text = str(
        text
    ).strip()

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0
    in_string = False
    escaped = False

    for position in range(
        start,
        len(text),
    ):

        character = text[
            position
        ]

        if escaped:

            escaped = False
            continue

        if (
            character == "\\"
            and in_string
        ):

            escaped = True
            continue

        if character == '"':

            in_string = (
                not in_string
            )

            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    position + 1
                ]

    raise ValueError(
        "A complete JSON object was not found."
    )


def normalize_evidence_text(
    value,
):

    text = str(
        value
    ).lower()

    text = (
        text.replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def exact_evidence_present(
    evidence,
    narrative,
):

    evidence = str(
        evidence
    ).strip()

    if not evidence:

        return False

    return (
        evidence.lower()
        in str(
            narrative
        ).lower()
    )


def normalized_evidence_present(
    evidence,
    narrative,
):

    evidence_normalized = (
        normalize_evidence_text(
            evidence
        )
    )

    narrative_normalized = (
        normalize_evidence_text(
            narrative
        )
    )

    if not evidence_normalized:

        return False

    return (
        evidence_normalized
        in narrative_normalized
    )


def evidence_token_overlap(
    evidence,
    narrative,
):

    evidence_tokens = set(
        normalize_evidence_text(
            evidence
        ).split()
    )

    narrative_tokens = set(
        normalize_evidence_text(
            narrative
        ).split()
    )

    if not evidence_tokens:

        return 0.0

    return (
        len(
            evidence_tokens
            .intersection(
                narrative_tokens
            )
        )
        / len(
            evidence_tokens
        )
    )


def factor_prevention_aligned(
    factor,
    prevention,
):

    allowed_preventions = (
        FACTOR_PREVENTION_ALIGNMENT.get(
            factor
        )
    )

    if allowed_preventions is None:

        return True

    return (
        prevention
        in allowed_preventions
    )


# =============================================================================
# 18. LOAD THE LOCAL QWEN MODEL
# =============================================================================

print("\n" + "=" * 95)
print("LOADING LOCAL QWEN MODEL")
print("=" * 95)

print("Model:", MODEL_ID)


quantization_config = (
    BitsAndBytesConfig(
        load_in_4bit=True,

        bnb_4bit_quant_type="nf4",

        bnb_4bit_use_double_quant=True,

        bnb_4bit_compute_dtype=torch.float16,
    )
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,

    use_fast=True,

    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = (
        tokenizer.eos_token
    )


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,

    quantization_config=(
        quantization_config
    ),

    device_map="auto",

    dtype=torch.float16,

    low_cpu_mem_usage=True,

    trust_remote_code=False,
)


model.eval()


print("Model loaded successfully.")
print("Model device:", model.device)


# =============================================================================
# 19. MODEL GENERATION FUNCTION
# =============================================================================

def generate_model_response(
    messages,
):

    model_inputs = (
        tokenizer.apply_chat_template(
            messages,

            add_generation_prompt=True,

            tokenize=True,

            return_dict=True,

            return_tensors="pt",
        )
    )

    model_inputs = {
        key:
            value.to(
                model.device
            )

        for key, value
        in model_inputs.items()
    }

    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        generated_output = model.generate(
            **model_inputs,

            max_new_tokens=(
                MAX_NEW_TOKENS
            ),

            do_sample=False,

            repetition_penalty=1.03,

            pad_token_id=(
                tokenizer.pad_token_id
            ),

            eos_token_id=(
                tokenizer.eos_token_id
            ),

            use_cache=True,
        )

    generated_ids = generated_output[
        0,
        input_length:
    ]

    response_text = tokenizer.decode(
        generated_ids,

        skip_special_tokens=True,
    )

    del model_inputs
    del generated_output
    del generated_ids

    return response_text


# =============================================================================
# 20. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    may_be_truncated,
):

    truncation_message = (
        "The stored narrative may be truncated near the source-field limit."
        if may_be_truncated
        else
        "The stored narrative is not flagged as truncated."
    )

    user_prompt = f"""
DATA QUALITY NOTE
{truncation_message}

NARRATIVE
{narrative}

Extract the reported primary cause, primary contributing factor, and
preliminary prevention theme.

Return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_response = ""

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]

        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        f"""
The previous response failed validation.

VALIDATION ERROR
{prior_error}

PREVIOUS RESPONSE
{prior_response}

Correct only the invalid output.

Use only categories from the allowed ontology.
Use HIGH, MEDIUM, or LOW for every support level.
Return only one valid JSON object.
""".strip(),
                }
            )

        try:

            raw_response = (
                generate_model_response(
                    messages
                )
            )

            last_response = (
                raw_response
            )

            json_text = (
                extract_json_object(
                    raw_response
                )
            )

            response_dictionary = (
                json.loads(
                    json_text
                )
            )

            normalized_dictionary = (
                normalize_response_dictionary(
                    response_dictionary
                )
            )

            validated = (
                IncidentExtraction
                .model_validate(
                    normalized_dictionary
                )
            )

            cause_exact = (
                exact_evidence_present(
                    validated
                    .cause_evidence_text,

                    narrative,
                )
            )

            cause_normalized = (
                normalized_evidence_present(
                    validated
                    .cause_evidence_text,

                    narrative,
                )
            )

            cause_overlap = (
                evidence_token_overlap(
                    validated
                    .cause_evidence_text,

                    narrative,
                )
            )

            factor_exact = (
                exact_evidence_present(
                    validated
                    .factor_evidence_text,

                    narrative,
                )
            )

            factor_normalized = (
                normalized_evidence_present(
                    validated
                    .factor_evidence_text,

                    narrative,
                )
            )

            factor_overlap = (
                evidence_token_overlap(
                    validated
                    .factor_evidence_text,

                    narrative,
                )
            )

            alignment_valid = (
                factor_prevention_aligned(
                    validated
                    .primary_contributing_factor,

                    validated
                    .prevention_theme,
                )
            )

            review_reasons = []

            if may_be_truncated:

                review_reasons.append(
                    "Narrative may be truncated"
                )

            if (
                validated
                .primary_cause
                == "INSUFFICIENT_INFORMATION"
            ):

                review_reasons.append(
                    "Cause is insufficient information"
                )

            if (
                validated
                .primary_contributing_factor
                == "NOT_STATED"
            ):

                review_reasons.append(
                    "Contributing factor is not stated"
                )

            if (
                validated
                .prevention_theme
                in {
                    "ENGINEERING_REVIEW_REQUIRED",
                    "INVESTIGATION_REQUIRED",
                    "INSUFFICIENT_INFORMATION",
                }
            ):

                review_reasons.append(
                    "Prevention theme requires further review"
                )

            if not (
                cause_exact
                or cause_normalized
                or cause_overlap >= 0.85
            ):

                review_reasons.append(
                    "Cause evidence not found in narrative"
                )

            if (
                validated
                .primary_contributing_factor
                != "NOT_STATED"
                and not (
                    factor_exact
                    or factor_normalized
                    or factor_overlap >= 0.85
                )
            ):

                review_reasons.append(
                    "Factor evidence not found in narrative"
                )

            if not alignment_valid:

                review_reasons.append(
                    "Factor-prevention combination is unusual"
                )

            if min(
                validated.cause_confidence,
                validated.factor_confidence,
                validated.prevention_confidence,
            ) < LOW_CONFIDENCE_THRESHOLD:

                review_reasons.append(
                    "At least one confidence value is low"
                )

            if "LOW" in {
                validated.cause_support_level,
                validated.factor_support_level,
                validated.prevention_support_level,
            }:

                review_reasons.append(
                    "At least one support level is low"
                )

            if (
                validated
                .reviewer_note
                .strip()
            ):

                review_reasons.append(
                    "Model supplied a reviewer note"
                )

            return {
                **validated.model_dump(),

                "SCHEMA_VALID":
                    True,

                "CAUSE_EVIDENCE_EXACT":
                    cause_exact,

                "CAUSE_EVIDENCE_NORMALIZED":
                    cause_normalized,

                "CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        cause_overlap,
                        4,
                    ),

                "FACTOR_EVIDENCE_EXACT":
                    factor_exact,

                "FACTOR_EVIDENCE_NORMALIZED":
                    factor_normalized,

                "FACTOR_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        factor_overlap,
                        4,
                    ),

                "FACTOR_PREVENTION_ALIGNED":
                    alignment_valid,

                "REVIEW_REQUIRED":
                    bool(
                        review_reasons
                    ),

                "REVIEW_REASONS":
                    "; ".join(
                        review_reasons
                    ),

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "ATTEMPT_COUNT":
                    attempt_number,

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = (
                last_response
            )

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:350]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        "primary_cause":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NOT_STATED",

        "prevention_theme":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "factor_evidence_text":
            "",

        "prevention_rationale":
            "The model response could not be validated.",

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "prevention_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "factor_support_level":
            "LOW",

        "prevention_support_level":
            "LOW",

        "insufficient_information_reason":
            "Generation or validation failed.",

        "reviewer_note":
            "",

        "SCHEMA_VALID":
            False,

        "CAUSE_EVIDENCE_EXACT":
            False,

        "CAUSE_EVIDENCE_NORMALIZED":
            False,

        "CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "FACTOR_EVIDENCE_EXACT":
            False,

        "FACTOR_EVIDENCE_NORMALIZED":
            False,

        "FACTOR_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "FACTOR_PREVENTION_ALIGNED":
            False,

        "REVIEW_REQUIRED":
            True,

        "REVIEW_REASONS":
            "Model response failed validation",

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAW_MODEL_RESPONSE":
            last_response,
    }


# =============================================================================
# 21. FUNCTIONAL TEST
# =============================================================================

if RUN_FUNCTIONAL_TEST:

    print("\n" + "=" * 95)
    print("RUNNING FUNCTIONAL TEST")
    print("=" * 95)

    test_narrative = (
        "The employee was loading roof bolts. "
        "When the retaining bands were cut, one roof bolt moved upward "
        "and struck the employee in the mouth."
    )

    test_result = classify_narrative(
        narrative=test_narrative,

        may_be_truncated=False,
    )

    print(
        json.dumps(
            test_result,
            indent=2,
            default=str,
        )
    )

    if (
        test_result[
            "PROCESSING_STATUS"
        ]
        != "SUCCESS"
    ):

        raise RuntimeError(
            "The functional test failed."
        )

    print(
        "\nFunctional test passed."
    )


# =============================================================================
# 22. CHECKPOINT FUNCTIONS
# =============================================================================

def load_checkpoint(
    path,
):

    checkpoint_path = Path(
        path
    )

    if not checkpoint_path.exists():

        return {}

    records = {}

    with checkpoint_path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:

                continue

            try:

                record = json.loads(
                    line
                )

                record_id = str(
                    record[
                        "SOURCE_RECORD_ID"
                    ]
                )

                records[
                    record_id
                ] = record

            except Exception:

                continue

    return records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


existing_checkpoint = load_checkpoint(
    CHECKPOINT_JSONL
)


print(
    "\nExisting valid checkpoint records:",
    len(
        existing_checkpoint
    ),
)


# =============================================================================
# 23. PROCESS THE DEVELOPMENT SAMPLE
# =============================================================================

context_columns = [
    column
    for column in [
        "ACCIDENT_DT",
        "ACCIDENT_YEAR",
        "TIME_PERIOD",
        "SEVERITY_GROUP",
        "RECORD_TYPE",
        "SECTOR_GROUP",
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
        "DAYS_RESTRICT",
        "DAYS_LOST",
    ]
    if column in development_df.columns
]


processed_records = []


for row_number, row in (
    development_df.iterrows()
):

    source_record_id = str(
        row[
            id_column
        ]
    )

    if (
        not REPROCESS_COMPLETED
        and source_record_id
        in existing_checkpoint
    ):

        print(
            f"[{row_number + 1}/"
            f"{len(development_df)}] "
            f"Checkpoint: "
            f"{source_record_id}"
        )

        processed_records.append(
            existing_checkpoint[
                source_record_id
            ]
        )

        continue

    narrative = str(
        row[
            TEXT_COLUMN
        ]
    )

    may_be_truncated = bool(
        row[
            "_MAY_BE_TRUNCATED"
        ]
    )

    print(
        f"[{row_number + 1}/"
        f"{len(development_df)}] "
        f"{row['SEVERITY_GROUP']} | "
        f"{row['TIME_PERIOD']} | "
        f"{source_record_id}"
    )

    start_time = time.time()

    model_result = classify_narrative(
        narrative=narrative,

        may_be_truncated=(
            may_be_truncated
        ),
    )

    record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "SOURCE_INDEX":
            row.get(
                "_ORIGINAL_INDEX",
                row_number,
            ),

        "NARRATIVE":
            narrative,

        "NARRATIVE_WORD_COUNT":
            int(
                row[
                    "_WORD_COUNT"
                ]
            ),

        "NARRATIVE_CHARACTER_COUNT":
            int(
                row[
                    "_CHARACTER_COUNT"
                ]
            ),

        "NARRATIVE_MAY_BE_TRUNCATED":
            may_be_truncated,

        **{
            column:
                (
                    None
                    if pd.isna(
                        row[
                            column
                        ]
                    )
                    else row[
                        column
                    ]
                )

            for column
            in context_columns
        },

        **model_result,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    processed_records.append(
        record
    )

    append_checkpoint(
        record,
        CHECKPOINT_JSONL,
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 24. BUILD RESULTS DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    processed_records
)


results_df = results_df.rename(
    columns={
        "primary_cause":
            "PRIMARY_CAUSE",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "prevention_theme":
            "PREVENTION_THEME",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "factor_evidence_text":
            "FACTOR_EVIDENCE_TEXT",

        "prevention_rationale":
            "PREVENTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "prevention_confidence":
            "PREVENTION_CONFIDENCE",

        "cause_support_level":
            "CAUSE_SUPPORT_LEVEL",

        "factor_support_level":
            "FACTOR_SUPPORT_LEVEL",

        "prevention_support_level":
            "PREVENTION_SUPPORT_LEVEL",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


for boolean_column in [
    "SCHEMA_VALID",
    "CAUSE_EVIDENCE_EXACT",
    "CAUSE_EVIDENCE_NORMALIZED",
    "FACTOR_EVIDENCE_EXACT",
    "FACTOR_EVIDENCE_NORMALIZED",
    "FACTOR_PREVENTION_ALIGNED",
    "REVIEW_REQUIRED",
]:

    if boolean_column in results_df.columns:

        results_df[
            boolean_column
        ] = (
            results_df[
                boolean_column
            ]
            .fillna(False)
            .astype(bool)
        )


# =============================================================================
# 25. ANALYTICAL SUMMARIES
# =============================================================================

cause_summary = (
    results_df[
        "PRIMARY_CAUSE"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Primary cause"
    )
    .reset_index(
        name="Count"
    )
)


factor_summary = (
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Primary contributing factor"
    )
    .reset_index(
        name="Count"
    )
)


prevention_summary = (
    results_df[
        "PREVENTION_THEME"
    ]
    .value_counts(
        dropna=False
    )
    .rename_axis(
        "Prevention theme"
    )
    .reset_index(
        name="Count"
    )
)


cause_summary[
    "Percent"
] = (
    100
    * cause_summary[
        "Count"
    ]
    / len(
        results_df
    )
)


factor_summary[
    "Percent"
] = (
    100
    * factor_summary[
        "Count"
    ]
    / len(
        results_df
    )
)


prevention_summary[
    "Percent"
] = (
    100
    * prevention_summary[
        "Count"
    ]
    / len(
        results_df
    )
)


cause_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


factor_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


prevention_by_severity = pd.crosstab(
    results_df[
        "PREVENTION_THEME"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


cause_by_period = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE"
    ],

    results_df[
        "TIME_PERIOD"
    ],
)


factor_by_period = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "TIME_PERIOD"
    ],
)


cause_factor_table = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE"
    ],

    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
)


factor_prevention_table = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "PREVENTION_THEME"
    ],
)


review_reason_counts = (
    results_df[
        "REVIEW_REASONS"
    ]
    .fillna("")
    .str.split(
        "; "
    )
    .explode()
)


review_reason_summary = (
    review_reason_counts[
        review_reason_counts
        .str.len()
        > 0
    ]
    .value_counts()
    .rename_axis(
        "Review reason"
    )
    .reset_index(
        name="Count"
    )
)


quality_summary = pd.DataFrame(
    [
        {
            "Measure":
                "Development narratives",

            "Value":
                len(
                    results_df
                ),
        },
        {
            "Measure":
                "Successful classifications",

            "Value":
                int(
                    results_df[
                        "PROCESSING_STATUS"
                    ]
                    .eq(
                        "SUCCESS"
                    )
                    .sum()
                ),
        },
        {
            "Measure":
                "Failed classifications",

            "Value":
                int(
                    results_df[
                        "PROCESSING_STATUS"
                    ]
                    .eq(
                        "FAILED"
                    )
                    .sum()
                ),
        },
        {
            "Measure":
                "Schema-valid results",

            "Value":
                int(
                    results_df[
                        "SCHEMA_VALID"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Records requiring review",

            "Value":
                int(
                    results_df[
                        "REVIEW_REQUIRED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Cause evidence exact",

            "Value":
                int(
                    results_df[
                        "CAUSE_EVIDENCE_EXACT"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Cause evidence normalized",

            "Value":
                int(
                    results_df[
                        "CAUSE_EVIDENCE_NORMALIZED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor evidence exact",

            "Value":
                int(
                    results_df[
                        "FACTOR_EVIDENCE_EXACT"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor evidence normalized",

            "Value":
                int(
                    results_df[
                        "FACTOR_EVIDENCE_NORMALIZED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor-prevention alignment flags",

            "Value":
                int(
                    (
                        ~results_df[
                            "FACTOR_PREVENTION_ALIGNED"
                        ]
                    ).sum()
                ),
        },
        {
            "Measure":
                "NOT_STATED factors",

            "Value":
                int(
                    results_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "NOT_STATED"
                    )
                    .sum()
                ),
        },
        {
            "Measure":
                "Insufficient-information causes",

            "Value":
                int(
                    results_df[
                        "PRIMARY_CAUSE"
                    ]
                    .eq(
                        "INSUFFICIENT_INFORMATION"
                    )
                    .sum()
                ),
        },
        {
            "Measure":
                "Truncated or near-limit narratives",

            "Value":
                int(
                    results_df[
                        "NARRATIVE_MAY_BE_TRUNCATED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Mean processing seconds",

            "Value":
                round(
                    float(
                        results_df[
                            "PROCESSING_SECONDS"
                        ].mean()
                    ),
                    2,
                ),
        },
    ]
)


# =============================================================================
# 26. IDENTIFY UNUSED, RARE, AND DOMINANT CATEGORIES
# =============================================================================

def ontology_usage_table(
    allowed_categories,
    summary_dataframe,
    category_column,
):

    counts = dict(
        zip(
            summary_dataframe[
                category_column
            ].astype(str),

            summary_dataframe[
                "Count"
            ].astype(int),
        )
    )

    records = []

    for category in (
        allowed_categories
    ):

        count = int(
            counts.get(
                category,
                0,
            )
        )

        percent = (
            100
            * count
            / len(
                results_df
            )
        )

        if count == 0:

            status = "Unused"

        elif count < 5:

            status = "Very rare"

        elif percent >= 30:

            status = "Potentially overused"

        else:

            status = "Used"

        records.append(
            {
                "Category":
                    category,

                "Count":
                    count,

                "Percent":
                    percent,

                "Usage assessment":
                    status,
            }
        )

    return pd.DataFrame(
        records
    )


cause_usage = ontology_usage_table(
    CAUSE_CATEGORIES,
    cause_summary,
    "Primary cause",
)


factor_usage = ontology_usage_table(
    FACTOR_CATEGORIES,
    factor_summary,
    "Primary contributing factor",
)


prevention_usage = ontology_usage_table(
    PREVENTION_CATEGORIES,
    prevention_summary,
    "Prevention theme",
)


# =============================================================================
# 27. DEVELOPMENT RECOMMENDATIONS
# =============================================================================

recommendations = []


review_rate = float(
    results_df[
        "REVIEW_REQUIRED"
    ].mean()
)


not_stated_rate = float(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    .eq(
        "NOT_STATED"
    )
    .mean()
)


alignment_flag_rate = float(
    (
        ~results_df[
            "FACTOR_PREVENTION_ALIGNED"
        ]
    ).mean()
)


failed_rate = float(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq(
        "FAILED"
    )
    .mean()
)


if failed_rate > 0.05:

    recommendations.append(
        "The schema-failure rate exceeds 5%. Strengthen alias normalization "
        "and targeted retry instructions before freezing the pipeline."
    )

else:

    recommendations.append(
        "The schema-failure rate is at or below 5%. Retain the current "
        "normalization and retry approach unless individual failures reveal "
        "a repeated formatting problem."
    )


if review_rate > 0.50:

    recommendations.append(
        "More than half of the development records require review. Examine "
        "the most common review reasons and revise deterministic review rules "
        "or ontology definitions before Block 2."
    )

else:

    recommendations.append(
        "The review-required rate is below 50%. Continue reviewing the flagged "
        "records to confirm that the rules identify difficult cases."
    )


if not_stated_rate > 0.25:

    recommendations.append(
        "NOT_STATED is assigned to more than 25% of contributing factors. "
        "Review whether the prompt is too conservative or whether the "
        "narratives genuinely omit factor information."
    )


if alignment_flag_rate > 0.15:

    recommendations.append(
        "More than 15% of records have unusual factor-prevention combinations. "
        "Review the compatibility map and determine whether the ontology or "
        "action-selection prompt requires refinement."
    )


for usage_name, usage_df in [
    (
        "cause",
        cause_usage,
    ),
    (
        "factor",
        factor_usage,
    ),
    (
        "prevention",
        prevention_usage,
    ),
]:

    unused_categories = (
        usage_df.loc[
            usage_df[
                "Usage assessment"
            ]
            == "Unused",

            "Category",
        ]
        .tolist()
    )

    overused_categories = (
        usage_df.loc[
            usage_df[
                "Usage assessment"
            ]
            == "Potentially overused",

            "Category",
        ]
        .tolist()
    )

    if unused_categories:

        recommendations.append(
            f"Unused {usage_name} categories: "
            + ", ".join(
                unused_categories
            )
            + ". Determine whether these categories are needed, too narrowly "
            "defined, or absent from the development sample."
        )

    if overused_categories:

        recommendations.append(
            f"Potentially overused {usage_name} categories: "
            + ", ".join(
                overused_categories
            )
            + ". Review representative examples for category drift."
        )


recommendations.append(
    "Do not freeze the ontology solely from frequency counts. Review "
    "representative high-support records, low-support records, alignment "
    "flags, and fallback-category records before Block 2."
)


recommendations.append(
    "After the Block 1 review is complete, freeze the ontology, system prompt, "
    "alias mapping, and compatibility rules. Block 2 should use a separate "
    "2,000-record sample that excludes every ID in the saved development-ID "
    "file."
)


# =============================================================================
# 28. SELECT REPRESENTATIVE AND QUESTIONABLE EXAMPLES
# =============================================================================

representative_examples = (
    results_df[
        (
            results_df[
                "PROCESSING_STATUS"
            ]
            == "SUCCESS"
        )
        &
        (
            results_df[
                "REVIEW_REQUIRED"
            ]
            == False
        )
    ]
    .sort_values(
        [
            "CAUSE_CONFIDENCE",
            "FACTOR_CONFIDENCE",
            "PREVENTION_CONFIDENCE",
        ],
        ascending=False,
    )
    .head(15)
)


review_examples = (
    results_df[
        results_df[
            "REVIEW_REQUIRED"
        ]
    ]
    .sort_values(
        [
            "PROCESSING_STATUS",
            "CAUSE_CONFIDENCE",
            "FACTOR_CONFIDENCE",
        ],
        ascending=[
            True,
            True,
            True,
        ],
    )
    .head(20)
)


alignment_examples = (
    results_df[
        ~results_df[
            "FACTOR_PREVENTION_ALIGNED"
        ]
    ]
    .head(15)
)


not_stated_examples = (
    results_df[
        results_df[
            "PRIMARY_CONTRIBUTING_FACTOR"
        ]
        == "NOT_STATED"
    ]
    .head(15)
)


# =============================================================================
# 29. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped_lines = []

    for line in lines:

        if not line:

            wrapped_lines.append("")

        else:

            wrapped_lines.extend(
                textwrap.wrap(
                    str(line),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    chunks = [
        wrapped_lines[
            start:
            start + lines_per_page
        ]

        for start in range(
            0,
            len(
                wrapped_lines
            ),
            lines_per_page,
        )
    ] or [[""]]

    for chunk_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - {chunk_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=24,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[
                column
            ] = (
                printable[
                    column
                ]
                .map(
                    lambda value:
                        (
                            ""
                            if pd.isna(
                                value
                            )
                            else f"{value:.4f}"
                        )
                )
            )

    printable.columns = [
        str(column)
        .replace(
            "_",
            " ",
        )
        .title()

        for column in printable.columns
    ]

    for column in printable.columns:

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    ""
                    if pd.isna(
                        value
                    )
                    else str(
                        value
                    )
            )
        )

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=(
                                column_wrap_width
                            ),
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(
            printable
        ),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.42,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_bar_chart(
    title,
    dataframe,
    category_column,
    count_column="Count",
    color="#2E6F9E",
    top_n=20,
):

    plot_df = (
        dataframe
        .head(
            top_n
        )
        .sort_values(
            count_column,
            ascending=True,
        )
    )

    figure = plt.figure(
        figsize=(10, 7)
    )

    axis = figure.add_subplot(
        111
    )

    axis.barh(
        plot_df[
            category_column
        ],

        plot_df[
            count_column
        ],

        color=color,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        "Narrative count"
    )

    axis.grid(
        axis="x",
        alpha=0.25,
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_heatmap(
    title,
    dataframe,
    x_label,
    y_label,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(11, 8)
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                dataframe.columns
            )
        )
    )

    axis.set_xticklabels(
        dataframe.columns,
        rotation=30,
        ha="right",
    )

    axis.set_yticks(
        range(
            len(
                dataframe.index
            )
        )
    )

    axis.set_yticklabels(
        [
            str(value)
            .replace(
                "_",
                " ",
            )
            .title()

            for value in (
                dataframe.index
            )
        ],
        fontsize=7.5,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        x_label
    )

    axis.set_ylabel(
        y_label
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value > 0:

                axis.text(
                    column_number,
                    row_number,
                    str(value),
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=12,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )

        return

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        row_index,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 82,
                (
                    f"Example {example_number} | "
                    f"Record "
                    f"{row['SOURCE_RECORD_ID']} | "
                    f"{row.get('SEVERITY_GROUP', '')} | "
                    f"{row.get('TIME_PERIOD', '')}"
                ),
                (
                    f"Cause: "
                    f"{row.get('PRIMARY_CAUSE', '')} | "
                    f"Support: "
                    f"{row.get('CAUSE_SUPPORT_LEVEL', '')}"
                ),
                (
                    f"Factor: "
                    f"{row.get('PRIMARY_CONTRIBUTING_FACTOR', '')} | "
                    f"Support: "
                    f"{row.get('FACTOR_SUPPORT_LEVEL', '')}"
                ),
                (
                    f"Prevention: "
                    f"{row.get('PREVENTION_THEME', '')} | "
                    f"Support: "
                    f"{row.get('PREVENTION_SUPPORT_LEVEL', '')}"
                ),
                (
                    f"Review required: "
                    f"{row.get('REVIEW_REQUIRED', '')}"
                ),
                (
                    f"Review reasons: "
                    f"{row.get('REVIEW_REASONS', '')}"
                ),
                "",
                "Narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "Cause evidence: "
                    f"{row.get('CAUSE_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Factor evidence: "
                    f"{row.get('FACTOR_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Prevention rationale: "
                    f"{row.get('PREVENTION_RATIONALE', '')}"
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.8,
        lines_per_page=50,
        wrap_width=104,
    )


# =============================================================================
# 30. BUILD PDF REPORT
# =============================================================================

# Cover.
figure = create_page()

figure.text(
    0.5,
    0.79,
    "MSHA Expanded Ontology Development",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.73,
    "Block 1 - 1,000-Narrative Stress Test",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.685,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Model: {MODEL_ID}",
    f"GPU: {GPU_NAME}",
    f"Development narratives: {len(results_df):,}",
    f"Cause categories: {len(CAUSE_CATEGORIES)}",
    f"Factor categories: {len(FACTOR_CATEGORIES)}",
    f"Prevention categories: {len(PREVENTION_CATEGORIES)}",
    f"Successful classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum())}",
    f"Records requiring review: "
    f"{int(results_df['REVIEW_REQUIRED'].sum())}",
    "",
    "Purpose:",
    "Stress-test the draft ontology, prompt, compatibility rules, evidence",
    "checks, and review logic before freezing the system for the unseen",
    "2,000-record analysis.",
    "",
    "Important:",
    "This is a development dataset. Its results may be used to improve the",
    "ontology and prompt. Block 2 must exclude every Block 1 record ID.",
]

figure.text(
    0.11,
    0.57,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=9.1,
    family="monospace",
    linespacing=1.24,
)

save_page(
    figure
)


# Executive summary.
add_text_pages(
    "Executive Summary",
    [
        "This report expands the initial 100-narrative Qwen trial to a",
        f"{len(results_df):,}-narrative development sample.",
        "",
        "The development sample was selected across five injury-severity groups",
        "and broad time periods. Only records routed as acute accidents were",
        "included in the primary ontology stress test.",
        "",
        "The model assigned one primary cause, one primary contributing factor,",
        "and one preliminary prevention theme to each narrative. It also",
        "returned supporting evidence, confidence values, support levels, and",
        "review notes.",
        "",
        "The report evaluates whether the draft ontology is sufficiently broad,",
        "whether any categories are overused or unused, how often the model",
        "returns insufficient information, whether evidence excerpts can be",
        "located in the narrative, and whether contributing factors align with",
        "their proposed prevention themes.",
        "",
        "The results are intended to support final ontology and prompt decisions.",
        "They are not a final test of model accuracy because these development",
        "records may influence the final design.",
    ],
    font_size=8.8,
)


# Methodology.
add_text_pages(
    "Development Methodology",
    [
        "1. The existing accidents DataFrame was used directly.",
        "",
        "2. Blank, extremely short, and exact duplicate narratives were removed.",
        "",
        "3. Records were routed into acute accidents, occupational illnesses,",
        "   noninjury accidents, and natural or nonoccupational events.",
        "",
        "4. The primary development sample included acute-accident narratives.",
        "",
        "5. Sampling was balanced across five severity groups and distributed",
        "   across broad time periods where records were available.",
        "",
        "6. Qwen2.5-7B-Instruct was loaded locally in four-bit mode.",
        "",
        "7. The model was restricted to the draft cause, factor, and prevention",
        "   ontology.",
        "",
        "8. Responses were normalized and validated with Pydantic.",
        "",
        "9. Supporting evidence was checked using exact, normalized, and",
        "   token-overlap methods.",
        "",
        "10. Factor-prevention combinations were checked against deterministic",
        "    compatibility rules.",
        "",
        "11. Records with low support, missing evidence, unusual relationships,",
        "    truncation, or fallback categories were flagged for review.",
        "",
        "12. Development record IDs were saved separately so Block 2 can exclude",
        "    every record used in this ontology-development stage.",
    ],
    font_size=8.6,
)


# Sampling summaries.
add_table_pages(
    "Severity Sampling Summary",
    sampling_summary,
    rows_per_page=10,
    font_size=7.5,
    column_wrap_width=30,
)


sample_by_period = pd.crosstab(
    development_df[
        "SEVERITY_GROUP"
    ],

    development_df[
        "TIME_PERIOD"
    ],
).reset_index()


add_table_pages(
    "Development Sample by Severity and Time Period",
    sample_by_period,
    rows_per_page=12,
    font_size=7.2,
    column_wrap_width=24,
    landscape=True,
)


# Quality summary.
add_table_pages(
    "Processing and Quality Summary",
    quality_summary,
    rows_per_page=20,
    font_size=7.4,
    column_wrap_width=38,
)


# Category distributions.
add_table_pages(
    "Primary Cause Distribution",
    cause_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=35,
)


add_bar_chart(
    "Primary Cause Distribution",
    cause_summary,
    "Primary cause",
    color="#2E6F9E",
)


add_table_pages(
    "Contributing-Factor Distribution",
    factor_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=35,
)


add_bar_chart(
    "Contributing-Factor Distribution",
    factor_summary,
    "Primary contributing factor",
    color="#4E8B57",
)


add_table_pages(
    "Prevention-Theme Distribution",
    prevention_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=35,
)


add_bar_chart(
    "Prevention-Theme Distribution",
    prevention_summary,
    "Prevention theme",
    color="#B46A3C",
)


# Ontology use.
add_table_pages(
    "Cause Ontology Usage",
    cause_usage,
    rows_per_page=24,
    font_size=7.0,
    column_wrap_width=35,
)


add_table_pages(
    "Factor Ontology Usage",
    factor_usage,
    rows_per_page=24,
    font_size=7.0,
    column_wrap_width=35,
)


add_table_pages(
    "Prevention Ontology Usage",
    prevention_usage,
    rows_per_page=24,
    font_size=7.0,
    column_wrap_width=35,
)


# Severity and time views.
add_heatmap(
    "Primary Cause by Injury Severity",
    cause_by_severity,
    "Injury-severity group",
    "Primary cause",
)


add_heatmap(
    "Contributing Factor by Injury Severity",
    factor_by_severity,
    "Injury-severity group",
    "Contributing factor",
)


add_heatmap(
    "Prevention Theme by Injury Severity",
    prevention_by_severity,
    "Injury-severity group",
    "Prevention theme",
)


add_heatmap(
    "Primary Cause by Time Period",
    cause_by_period,
    "Time period",
    "Primary cause",
)


add_heatmap(
    "Contributing Factor by Time Period",
    factor_by_period,
    "Time period",
    "Contributing factor",
)


# Relationships.
add_heatmap(
    "Cause and Contributing-Factor Combinations",
    cause_factor_table,
    "Contributing factor",
    "Primary cause",
)


add_heatmap(
    "Factor and Prevention-Theme Combinations",
    factor_prevention_table,
    "Prevention theme",
    "Contributing factor",
)


# Review reasons.
add_table_pages(
    "Automated Review Reasons",
    review_reason_summary,
    rows_per_page=25,
    font_size=7.2,
    column_wrap_width=45,
)


# Recommendations.
add_text_pages(
    "Ontology Development Recommendations",
    [
        line
        for recommendation_number, recommendation
        in enumerate(
            recommendations,
            start=1,
        )
        for line in [
            (
                f"{recommendation_number}. "
                f"{recommendation}"
            ),
            "",
        ]
    ],
    font_size=8.5,
)


# Representative and questionable examples.
add_example_pages(
    "Representative High-Support Results",
    representative_examples,
    (
        "These records completed successfully, were not automatically flagged "
        "for review, and had comparatively strong confidence values. They "
        "should be inspected to confirm that apparently clean results are "
        "substantively correct."
    ),
    maximum_records=12,
)


add_example_pages(
    "Records Requiring Review",
    review_examples,
    (
        "These records were automatically flagged because of missing evidence, "
        "low support, fallback categories, truncation, schema problems, or "
        "unusual factor-prevention relationships."
    ),
    maximum_records=15,
)


add_example_pages(
    "Factor-Prevention Alignment Flags",
    alignment_examples,
    (
        "These records contain factor-prevention combinations outside the "
        "current compatibility map. Review whether the model chose an "
        "inappropriate prevention theme or whether the compatibility map is "
        "too restrictive."
    ),
    maximum_records=12,
)


add_example_pages(
    "NOT_STATED Contributing-Factor Examples",
    not_stated_examples,
    (
        "These records were assigned NOT_STATED as the contributing factor. "
        "Review whether the narratives genuinely omit factor information or "
        "whether the prompt is abstaining too often."
    ),
    maximum_records=12,
)


# Final interpretation.
add_text_pages(
    "Block 1 Conclusions and Decision Criteria",
    [
        "Block 1 is complete when the development report has been reviewed and",
        "the project has made final decisions regarding:",
        "",
        "1. Cause categories and definitions.",
        "",
        "2. Contributing-factor categories and definitions.",
        "",
        "3. Prevention-theme categories and definitions.",
        "",
        "4. Category aliases and schema-repair rules.",
        "",
        "5. Evidence requirements.",
        "",
        "6. Factor-prevention compatibility rules.",
        "",
        "7. Automated human-review conditions.",
        "",
        "8. Treatment of occupational illnesses, noninjury reports, and natural",
        "   or nonoccupational events.",
        "",
        "After those decisions are made, the ontology and prompt should be",
        "frozen. Block 2 should then select 2,000 unseen narratives while",
        "excluding all IDs stored in the Block 1 development-ID file.",
        "",
        "Block 1 frequency distributions should not be interpreted as population",
        "prevalence because the development sample is balanced by severity.",
        "",
        "Final accuracy will be measured later using the independently reviewed",
        "100-record sample from Block 2.",
    ],
    font_size=8.7,
)


# Technical appendix.
add_text_pages(
    "Technical Appendix",
    [
        f"Model: {MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"GPU: {GPU_NAME}",
        f"Development records: {len(results_df):,}",
        f"Cause categories: {len(CAUSE_CATEGORIES)}",
        f"Factor categories: {len(FACTOR_CATEGORIES)}",
        f"Prevention categories: {len(PREVENTION_CATEGORIES)}",
        f"Minimum narrative words: {MIN_NARRATIVE_WORDS}",
        f"Truncation threshold: {TRUNCATION_CHARACTER_THRESHOLD}",
        f"Maximum retries: {MAX_RETRIES}",
        f"Maximum new tokens: {MAX_NEW_TOKENS}",
        f"Random state: {RANDOM_STATE}",
        "",
        f"PDF report: {os.path.abspath(OUTPUT_PDF)}",
        f"Checkpoint: {os.path.abspath(CHECKPOINT_JSONL)}",
        f"Development IDs: {os.path.abspath(DEVELOPMENT_IDS_JSON)}",
    ],
    font_size=8.3,
)


# =============================================================================
# 31. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "MSHA Block 1 Expanded Ontology Development Report"
)

metadata[
    "Author"
] = (
    "Python and Qwen2.5-7B-Instruct"
)

metadata[
    "Subject"
] = (
    "Development stress test of an ontology for MSHA accident causes, "
    "contributing factors, and prevention themes"
)

metadata[
    "Keywords"
] = (
    "MSHA, NLP, Qwen, ontology, cause extraction, "
    "contributing factors, prevention themes"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 32. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 1 ONTOLOGY DEVELOPMENT COMPLETE")
print("=" * 95)

print(
    f"Development narratives: "
    f"{len(results_df):,}"
)

print(
    f"Successful classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum()):,}"
)

print(
    f"Failed classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('FAILED').sum()):,}"
)

print(
    f"Records requiring review: "
    f"{int(results_df['REVIEW_REQUIRED'].sum()):,}"
)

print(
    f"NOT_STATED factors: "
    f"{int(results_df['PRIMARY_CONTRIBUTING_FACTOR'].eq('NOT_STATED').sum()):,}"
)

print(
    f"Alignment flags: "
    f"{int((~results_df['FACTOR_PREVENTION_ALIGNED']).sum()):,}"
)

print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print(
    f"Checkpoint: "
    f"{os.path.abspath(CHECKPOINT_JSONL)}"
)

print(
    f"Development ID file for Block 2 exclusion: "
    f"{os.path.abspath(DEVELOPMENT_IDS_JSON)}"
)

print("=" * 95)


# =============================================================================
# 33. CONCISE NOTEBOOK DISPLAYS
# =============================================================================

display(
    sampling_summary
)


display(
    quality_summary
)


display(
    cause_summary
)


display(
    factor_summary
)


display(
    prevention_summary
)


display(
    review_reason_summary
)
````

````python
# =============================================================================
# BLOCK 1A
# MSHA ONTOLOGY AND PROMPT REFINEMENT
#
# PURPOSE
# -------
# Rerun the SAME 1,000 development narratives used in Block 1 after making
# targeted ontology, prompt, routing, evidence, and prevention-rule revisions.
#
# This block compares Block 1a results with the original Block 1 results to
# determine whether the extraction system improved before it is frozen for
# Block 2.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. The Block 1 development-record ID file:
#       MSHA_Block1_Development_Record_IDs.json
#
# 3. The original Block 1 checkpoint:
#       MSHA_Block1_Development_Checkpoint.jsonl
#
# OUTPUTS
# -------
# 1. MSHA_Block1a_Refinement_Comparison_Report.pdf
# 2. MSHA_Block1a_Refined_Checkpoint.jsonl
#
# Block 1a does NOT create a new sample. It reuses the same record IDs so the
# Block 1 versus Block 1a comparison is direct.
#
# IMPORTANT
# ---------
# Block 1a is still development work. The ontology and prompt should be frozen
# only after reviewing the Block 1a comparison report.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import time
import math
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import (
    BaseModel,
    Field,
    model_validator,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from matplotlib.backends.backend_pdf import PdfPages


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

TEXT_COLUMN = "NARRATIVE"

DATE_COLUMN = "ACCIDENT_DT"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

BLOCK1_IDS_JSON = (
    "MSHA_Block1_Development_Record_IDs.json"
)

BLOCK1_CHECKPOINT_JSONL = (
    "MSHA_Block1_Development_Checkpoint.jsonl"
)

BLOCK1A_CHECKPOINT_JSONL = (
    "MSHA_Block1a_Refined_Checkpoint.jsonl"
)

OUTPUT_PDF = (
    "MSHA_Block1a_Refinement_Comparison_Report.pdf"
)

PROMPT_VERSION = (
    "MSHA_BLOCK1A_ONTOLOGY_REFINEMENT_V2.0"
)

MAX_NEW_TOKENS = 600

MAX_RETRIES = 3

LOW_CONFIDENCE_THRESHOLD = 0.70

TRUNCATION_CHARACTER_THRESHOLD = 383

RUN_FUNCTIONAL_TEST = True

REPROCESS_COMPLETED = False


# =============================================================================
# 3. RUNTIME CHECKS
# =============================================================================

print("=" * 95)
print("BLOCK 1A - RUNTIME ENVIRONMENT")
print("=" * 95)

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)
print("Pydantic:", pydantic.__version__)
print("Pydantic Core:", pydantic_core.__version__)


if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. "
        "Attach the notebook to a GPU compute instance."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0).total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3


print("\nGPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


# =============================================================================
# 4. REFINED RECORD-TYPE ONTOLOGY
# =============================================================================

RECORD_TYPE_CATEGORIES = [
    "ACUTE_ACCIDENT",
    "OCCUPATIONAL_EXPOSURE_OR_ILLNESS",
    "NONWORK_RELATED_OR_NATURAL_EVENT",
    "NONINJURY_REPORT",
    "INSUFFICIENT_EVENT_INFORMATION",
]


# =============================================================================
# 5. REFINED CAUSE ONTOLOGY
#
# Changes:
#   - Added FOREIGN_BODY_OR_PARTICLE_CONTACT.
#   - Added HEAT_OR_ENVIRONMENTAL_EXPOSURE.
#   - Clarified that acute strains and hernias belong under
#     MANUAL_HANDLING_OR_OVEREXERTION.
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


CAUSE_DEFINITIONS = {
    "SLIP_TRIP_OR_FALL":
        "A person loses footing, balance, or elevation and falls, trips, "
        "slips, or nearly falls.",

    "AWKWARD_MOVEMENT_OR_BODY_MOTION":
        "A twisting, bending, reaching, stepping, turning, or other acute "
        "body motion causes injury without a fall or substantial manual load.",

    "STRUCK_BY_MOVING_OBJECT":
        "A moving, falling, flying, swinging, rolling, or projected object "
        "contacts the person. Use this for larger objects, not small foreign "
        "particles entering an eye.",

    "STRUCK_AGAINST_STATIONARY_OBJECT":
        "The person moves into or against a stationary object or surface.",

    "FOREIGN_BODY_OR_PARTICLE_CONTACT":
        "A small particle, fragment, chip, dust particle, or foreign body "
        "enters or contacts the eye, skin, or body.",

    "CAUGHT_IN_OR_BETWEEN":
        "A body part or person is pinched, crushed, trapped, entangled, "
        "pulled into, or caught between objects.",

    "MATERIAL_RELEASE_OR_SHIFT":
        "Bulk or loose material unexpectedly shifts, rolls, slides, spills, "
        "or releases without a more specific roof or ground failure.",

    "GROUND_OR_ROOF_FAILURE":
        "Roof, rib, highwall, face, rock, slate, or ground fails, falls, "
        "breaks loose, or moves uncontrollably.",

    "UNEXPECTED_EQUIPMENT_MOVEMENT":
        "A machine, door, component, attachment, or equipment part moves "
        "unexpectedly without an explicitly stated failure.",

    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT":
        "Control of a tool, hose, strap, handle, machine, load, or equipment "
        "component is lost during use or handling.",

    "MANUAL_HANDLING_OR_OVEREXERTION":
        "Lifting, carrying, pushing, pulling, shoveling, holding, moving "
        "material, or exerting force causes an acute strain, pain, sprain, "
        "or hernia.",

    "PRESSURE_OR_STORED_ENERGY_RELEASE":
        "Pressurized fluid, compressed material, spring force, hydraulic "
        "force, tension, or other stored energy is released.",

    "ELECTRICAL_CONTACT":
        "Electrical current, energized conductors, electrical arc, or "
        "electrical energy directly causes the event.",

    "FIRE_EXPLOSION_OR_THERMAL_EVENT":
        "Fire, ignition, explosion, flame, hot material, sparks, or direct "
        "thermal contact causes the event.",

    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT":
        "A collision, overturn, runaway, roadway departure, backing event, "
        "or transportation event involving mobile equipment occurs.",

    "HEAT_OR_ENVIRONMENTAL_EXPOSURE":
        "Heat stress, heat exhaustion, dehydration, cold, weather, poor "
        "visibility, or another acute environmental exposure causes illness "
        "or injury.",

    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE":
        "Chemical, fumes, toxic material, dust, gas, or another acute "
        "occupational substance exposure causes the event.",

    "ERGONOMIC_OR_REPETITIVE_EXPOSURE":
        "Repeated motion, cumulative trauma, prolonged posture, vibration, "
        "or chronic ergonomic exposure causes the condition.",

    "OTHER":
        "The work-related event is stated but does not fit another category.",

    "INSUFFICIENT_INFORMATION":
        "The narrative does not provide enough information to identify a "
        "work-related event mechanism.",
}


# =============================================================================
# 6. REFINED FACTOR ONTOLOGY
#
# Changes:
#   - Tightened unstable-material and housekeeping definitions.
#   - Clarified securement, line of fire, and procedure boundaries.
#   - Added NO_SEPARATE_FACTOR_STATED as a clearer replacement for NOT_STATED.
# =============================================================================

FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NO_SEPARATE_FACTOR_STATED",
]


FACTOR_DEFINITIONS = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
        "Use only when the narrative explicitly describes material, a load, "
        "or a component as loose, unstable, unsupported, shifting, rolling, "
        "falling, sliding, or improperly positioned. Do not use merely "
        "because an object moved or contacted a person.",

    "INADEQUATE_SECUREMENT_OR_RESTRAINT":
        "Use only when blocking, fastening, bracing, chocking, strapping, "
        "support, restraint, or securement is explicitly missing, removed, "
        "inadequate, or failed.",

    "LINE_OF_FIRE_BODY_POSITION":
        "The person or body part is positioned in the path of moving "
        "equipment, material, tools, falling objects, pinch points, or "
        "released energy.",

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
        "Use only when a walking, standing, access, or working surface is "
        "explicitly affected by ice, water, mud, grease, debris, loose rock, "
        "uneven footing, holes, damaged walking surface, or a similar hazard. "
        "Do not use when a tool, hose, load, material, or hand slips.",

    "UNSTABLE_GROUND_OR_ROOF_CONDITION":
        "The narrative explicitly describes unstable, loose, unsupported, "
        "cracked, deteriorated, or failed roof, rib, face, highwall, rock, "
        "or ground.",

    "INADEQUATE_GUARDING_OR_BARRIER":
        "A guard or barrier is missing, removed, bypassed, open, damaged, "
        "inadequate, or defeated.",

    "ENERGY_ISOLATION_DEFICIENCY":
        "Hazardous energy was not locked out, blocked, bled, restrained, "
        "dissipated, disconnected, or otherwise controlled.",

    "EQUIPMENT_CONDITION_OR_FAILURE":
        "The narrative explicitly describes failure, breakage, leakage, "
        "malfunction, defect, damage, excessive wear, or unexpected operation "
        "caused by equipment condition.",

    "PROCEDURE_OR_WORK_METHOD":
        "The task method, work sequence, handling approach, maintenance "
        "method, or operating practice created the exposure, and no more "
        "specific physical factor is directly supported.",

    "INADEQUATE_HAZARD_RECOGNITION":
        "The narrative explicitly states or strongly shows that a hazard was "
        "not recognized, evaluated, or understood before exposure.",

    "COMMUNICATION_OR_COORDINATION":
        "Inadequate signaling, notification, communication, coordination, "
        "visibility between workers, or spotter use contributed.",

    "TRAINING_OR_EXPERIENCE":
        "The narrative explicitly states inadequate training, proficiency, "
        "qualification, task experience, or knowledge.",

    "PPE_OR_PROTECTIVE_CONTROL":
        "Missing, inadequate, damaged, incorrect, or unused protective "
        "equipment is explicitly described as relevant.",

    "ENVIRONMENTAL_CONDITION":
        "Heat, cold, weather, lighting, visibility, noise, or another "
        "environmental condition contributed.",

    "WORKSPACE_OR_ACCESS_LIMITATION":
        "Restricted clearance, difficult access, poor layout, excessive "
        "reach, awkward access, or limited workspace contributed.",

    "OTHER":
        "A directly stated contributing factor does not fit another category.",

    "NO_SEPARATE_FACTOR_STATED":
        "The narrative establishes what happened but does not separately "
        "state or directly support why the event occurred.",
}


# =============================================================================
# 7. REFINED PREVENTION ONTOLOGY
# =============================================================================

PREVENTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "CONTROL_ENVIRONMENTAL_EXPOSURE",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


PREVENTION_DEFINITIONS = {
    "SECURE_OR_REPOSITION_MATERIAL":
        "Secure, block, support, restrain, stabilize, or safely reposition "
        "material when instability, positioning, or inadequate securement is "
        "directly supported.",

    "ESTABLISH_LINE_OF_FIRE_EXCLUSION":
        "Keep people and body parts outside the path of moving equipment, "
        "falling objects, pinch points, tools, loads, or released energy.",

    "ISOLATE_OR_CONTROL_STORED_ENERGY":
        "Lock out, block, bleed, dissipate, disconnect, restrain, or otherwise "
        "control hazardous energy.",

    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT":
        "Inspect, repair, or replace equipment only when failure, damage, "
        "wear, leakage, defect, or malfunction is supported.",

    "INSTALL_OR_IMPROVE_GUARDING":
        "Install, restore, redesign, or improve a guard or barrier.",

    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING":
        "Correct debris, grease, mud, water, ice, loose material, damaged "
        "walking surfaces, uneven footing, or access conditions.",

    "IMPROVE_GROUND_CONTROL_OR_SCALING":
        "Improve examination, support, scaling, exclusion, or control of "
        "roof, rib, highwall, face, rock, or ground.",

    "REVISE_PROCEDURE_OR_JOB_PLAN":
        "Revise the work method, task sequence, maintenance approach, or "
        "written job plan.",

    "CONDUCT_PRE_JOB_HAZARD_REVIEW":
        "Perform a structured hazard assessment when the narrative supports "
        "an unrecognized or changing hazard.",

    "PROVIDE_TARGETED_TRAINING":
        "Provide task-specific training only when a training, proficiency, "
        "qualification, or experience need is stated or strongly supported.",

    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE":
        "Improve notification, signaling, coordination, visibility, or "
        "spotter practices.",

    "USE_MECHANICAL_HANDLING_ASSISTANCE":
        "Use lifting, positioning, transfer, or mechanical assistance to "
        "reduce manual handling or overexertion exposure.",

    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT":
        "Provide or improve protective equipment when PPE or a protective "
        "control is directly relevant.",

    "CONTROL_ENVIRONMENTAL_EXPOSURE":
        "Use hydration, work-rest controls, cooling, ventilation, shelter, "
        "monitoring, or other controls for heat, cold, weather, or acute "
        "environmental exposure.",

    "REDESIGN_WORKSPACE_OR_ACCESS":
        "Improve layout, clearance, access, reach, working position, or "
        "physical workspace.",

    "ENGINEERING_REVIEW_REQUIRED":
        "A technical control appears necessary, but the narrative does not "
        "support one specific engineering recommendation.",

    "INVESTIGATION_REQUIRED":
        "The narrative does not state a separate factor, or more information "
        "is needed before selecting a defensible prevention theme.",

    "INSUFFICIENT_INFORMATION":
        "The narrative does not support a defensible prevention theme.",
}


# =============================================================================
# 8. CATEGORY ALIASES
# =============================================================================

CATEGORY_ALIASES = {
    "UNSTATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_PROVIDED":
        "NO_SEPARATE_FACTOR_STATED",

    "NONE_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_SPECIFIED":
        "NO_SEPARATE_FACTOR_STATED",

    "UNKNOWN":
        "INSUFFICIENT_INFORMATION",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "REVISE_PROCEDURE":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "PRE_JOB_HAZARD_REVIEW":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "EQUIPMENT_REPAIR_OR_INSPECTION":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "IMPROVE_HOUSEKEEPING":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "GROUND_CONTROL_REVIEW":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
}


SUPPORT_LEVEL_ALIASES = {
    "HIGH": "HIGH",
    "DIRECT": "HIGH",
    "EXPLICIT": "HIGH",
    "EXPLICITLY_STATED": "HIGH",
    "SUPPORTED": "HIGH",
    "MEDIUM": "MEDIUM",
    "MODERATE": "MEDIUM",
    "IMPLIED": "MEDIUM",
    "STRONGLY_IMPLIED": "MEDIUM",
    "LOW": "LOW",
    "WEAK": "LOW",
    "UNCERTAIN": "LOW",
    "NONE": "LOW",
    "": "LOW",
}


# =============================================================================
# 9. PRIMARY FACTOR-TO-PREVENTION MAP
#
# The model is instructed to use this preferred mapping. Some alternatives
# remain valid where supporting evidence warrants them.
# =============================================================================

PREFERRED_PREVENTION_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
        "SECURE_OR_REPOSITION_MATERIAL",

    "INADEQUATE_SECUREMENT_OR_RESTRAINT":
        "SECURE_OR_REPOSITION_MATERIAL",

    "LINE_OF_FIRE_BODY_POSITION":
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "UNSTABLE_GROUND_OR_ROOF_CONDITION":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",

    "INADEQUATE_GUARDING_OR_BARRIER":
        "INSTALL_OR_IMPROVE_GUARDING",

    "ENERGY_ISOLATION_DEFICIENCY":
        "ISOLATE_OR_CONTROL_STORED_ENERGY",

    "EQUIPMENT_CONDITION_OR_FAILURE":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "PROCEDURE_OR_WORK_METHOD":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "INADEQUATE_HAZARD_RECOGNITION":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "COMMUNICATION_OR_COORDINATION":
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",

    "TRAINING_OR_EXPERIENCE":
        "PROVIDE_TARGETED_TRAINING",

    "PPE_OR_PROTECTIVE_CONTROL":
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",

    "ENVIRONMENTAL_CONDITION":
        "CONTROL_ENVIRONMENTAL_EXPOSURE",

    "WORKSPACE_OR_ACCESS_LIMITATION":
        "REDESIGN_WORKSPACE_OR_ACCESS",

    "NO_SEPARATE_FACTOR_STATED":
        "INVESTIGATION_REQUIRED",
}


ALLOWED_PREVENTIONS_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_SECUREMENT_OR_RESTRAINT": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    },

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },

    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_GUARDING_OR_BARRIER": {
        "INSTALL_OR_IMPROVE_GUARDING",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "ENERGY_ISOLATION_DEFICIENCY": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    },

    "INADEQUATE_HAZARD_RECOGNITION": {
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "COMMUNICATION_OR_COORDINATION": {
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "TRAINING_OR_EXPERIENCE": {
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "PPE_OR_PROTECTIVE_CONTROL": {
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "ENVIRONMENTAL_CONDITION": {
        "CONTROL_ENVIRONMENTAL_EXPOSURE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "WORKSPACE_OR_ACCESS_LIMITATION": {
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "OTHER": {
        "ENGINEERING_REVIEW_REQUIRED",
        "INVESTIGATION_REQUIRED",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "NO_SEPARATE_FACTOR_STATED": {
        "INVESTIGATION_REQUIRED",
        "INSUFFICIENT_INFORMATION",
    },
}


# =============================================================================
# 10. PYDANTIC RESPONSE SCHEMA
# =============================================================================

class IncidentExtraction(BaseModel):

    record_type: str

    primary_cause: str

    primary_contributing_factor: str

    prevention_theme: str

    cause_evidence_text: str = ""

    factor_evidence_text: str = ""

    prevention_rationale: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    prevention_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    factor_support_level: str = "LOW"

    prevention_support_level: str = "LOW"

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_response(self):

        if self.record_type not in RECORD_TYPE_CATEGORIES:

            raise ValueError(
                f"Invalid record_type: {self.record_type}"
            )

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                f"Invalid primary_cause: {self.primary_cause}"
            )

        if (
            self.primary_contributing_factor
            not in FACTOR_CATEGORIES
        ):

            raise ValueError(
                "Invalid primary_contributing_factor: "
                f"{self.primary_contributing_factor}"
            )

        if self.prevention_theme not in PREVENTION_CATEGORIES:

            raise ValueError(
                f"Invalid prevention_theme: {self.prevention_theme}"
            )

        for field_name in [
            "cause_support_level",
            "factor_support_level",
            "prevention_support_level",
        ]:

            if getattr(
                self,
                field_name,
            ) not in {
                "HIGH",
                "MEDIUM",
                "LOW",
            }:

                raise ValueError(
                    f"{field_name} must be HIGH, MEDIUM, or LOW."
                )

        return self


# =============================================================================
# 11. LOAD THE SAME BLOCK 1 RECORD IDS
# =============================================================================

if not Path(
    BLOCK1_IDS_JSON
).exists():

    raise FileNotFoundError(
        f"Required Block 1 ID file not found: {BLOCK1_IDS_JSON}"
    )


with open(
    BLOCK1_IDS_JSON,
    "r",
    encoding="utf-8",
) as input_file:

    block1_id_payload = json.load(
        input_file
    )


BLOCK1_ID_COLUMN = block1_id_payload[
    "id_column"
]


BLOCK1_RECORD_IDS = [
    str(value)

    for value in block1_id_payload[
        "record_ids"
    ]
]


if not BLOCK1_RECORD_IDS:

    raise ValueError(
        "The Block 1 development ID file contains no record IDs."
    )


print(
    f"\nBlock 1 development IDs loaded: "
    f"{len(BLOCK1_RECORD_IDS):,}"
)


# =============================================================================
# 12. PREPARE THE ACCIDENT DATAFRAME
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(
    accidents,
    pd.DataFrame,
):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


df = accidents.copy()


df.columns = (
    df.columns.astype(str)
    .str.strip()
    .str.upper()
)


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    if BLOCK1_ID_COLUMN in df.columns:

        id_column = BLOCK1_ID_COLUMN

    else:

        raise KeyError(
            "No recognized record-ID column was found."
        )


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"{TEXT_COLUMN} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"{SEVERITY_COLUMN} was not found."
    )


df[id_column] = (
    df[id_column]
    .astype(str)
)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


if DATE_COLUMN in df.columns:

    df[DATE_COLUMN] = pd.to_datetime(
        df[DATE_COLUMN],
        errors="coerce",
    )

    df["ACCIDENT_YEAR"] = (
        df[DATE_COLUMN]
        .dt.year
        .astype("Int64")
    )

elif "CAL_YR" in df.columns:

    df["ACCIDENT_YEAR"] = pd.to_numeric(
        df["CAL_YR"],
        errors="coerce",
    ).astype("Int64")

else:

    df["ACCIDENT_YEAR"] = pd.Series(
        [pd.NA] * len(df),
        dtype="Int64",
    )


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN]
    .str.len()
)


df["_MAY_BE_TRUNCATED"] = (
    df["_CHARACTER_COUNT"]
    >= TRUNCATION_CHARACTER_THRESHOLD
)


development_df = df[
    df[id_column]
    .isin(
        BLOCK1_RECORD_IDS
    )
].copy()


found_ids = set(
    development_df[
        id_column
    ].astype(str)
)


missing_ids = sorted(
    set(
        BLOCK1_RECORD_IDS
    )
    - found_ids
)


if missing_ids:

    print(
        f"Warning: {len(missing_ids)} Block 1 record IDs "
        "were not found in the current accidents DataFrame."
    )


development_df = (
    development_df
    .drop_duplicates(
        subset=[
            id_column
        ],
        keep="first",
    )
    .copy()
)


if development_df.empty:

    raise ValueError(
        "None of the Block 1 development records were found."
    )


# =============================================================================
# 13. RECREATE SEVERITY AND TIME-PERIOD FIELDS
# =============================================================================

def assign_severity_group(
    degree_injury,
):

    value = str(
        degree_injury
    ).upper()

    if "FATAL" in value:

        return "Fatality"

    if any(
        term in value
        for term in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "Permanent disability"

    if any(
        term in value
        for term in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "Days away"

    if any(
        term in value
        for term in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "Restricted or transferred"

    return "No lost time or other"


development_df[
    "SEVERITY_GROUP"
] = (
    development_df[
        SEVERITY_COLUMN
    ]
    .map(
        assign_severity_group
    )
)


def assign_time_period(
    year,
):

    if pd.isna(year):

        return "Unknown period"

    year = int(year)

    if year <= 2008:

        return "Early: 2000-2008"

    if year <= 2017:

        return "Middle: 2009-2017"

    return "Recent: 2018-2026"


development_df[
    "TIME_PERIOD"
] = (
    development_df[
        "ACCIDENT_YEAR"
    ]
    .map(
        assign_time_period
    )
)


# =============================================================================
# 14. DETERMINISTIC PRE-ROUTING HINTS
#
# These are hints to the model and later validation. The model still returns
# the final record type.
# =============================================================================

def preliminary_record_type_hint(
    row,
):

    narrative = str(
        row.get(
            TEXT_COLUMN,
            "",
        )
    ).upper()

    classification = str(
        row.get(
            "CLASSIFICATION",
            "",
        )
    ).upper()

    nature = str(
        row.get(
            "NATURE_INJURY",
            "",
        )
    ).upper()

    degree = str(
        row.get(
            "DEGREE_INJURY",
            "",
        )
    ).upper()

    combined = " ".join(
        [
            narrative,
            classification,
            nature,
            degree,
        ]
    )

    nonwork_patterns = [
        "NOT DUE TO WORK",
        "NOT WORK RELATED",
        "CANNOT BE CONNECTED TO THE JOB",
        "NON-OCCUPATIONAL",
        "NATURAL CAUSE",
        "CAMPING ON PROPERTY",
        "NOT A MINER",
        "JURISDICTION",
    ]

    natural_patterns = [
        "HEART ATTACK",
        "CARDIAC",
        "STROKE",
        "SEIZURE",
        "DIABETES",
        "BLOOD SUGAR",
        "PRE-EXISTING CONDITION",
    ]

    illness_patterns = [
        "HEARING LOSS",
        "SILICOSIS",
        "PNEUMOCONIOSIS",
        "DUST DISEASE",
        "REPEATED TRAUMA",
        "OCCUPATIONAL ILLNESS",
    ]

    noninjury_patterns = [
        "ACCIDENT ONLY",
        "NO INJURY",
        "NO INJURIES",
        "NO PERSON INJURED",
    ]

    if any(
        pattern in combined
        for pattern in nonwork_patterns
    ):

        return "NONWORK_RELATED_OR_NATURAL_EVENT"

    if any(
        pattern in combined
        for pattern in natural_patterns
    ):

        return "NONWORK_RELATED_OR_NATURAL_EVENT"

    if any(
        pattern in combined
        for pattern in illness_patterns
    ):

        return "OCCUPATIONAL_EXPOSURE_OR_ILLNESS"

    if any(
        pattern in combined
        for pattern in noninjury_patterns
    ):

        return "NONINJURY_REPORT"

    if len(
        narrative.split()
    ) < 8:

        return "INSUFFICIENT_EVENT_INFORMATION"

    return "ACUTE_ACCIDENT"


development_df[
    "PRELIMINARY_RECORD_TYPE_HINT"
] = (
    development_df.apply(
        preliminary_record_type_hint,
        axis=1,
    )
)


# =============================================================================
# 15. SYSTEM PROMPT
# =============================================================================

def format_definitions(
    mapping,
):

    return "\n".join(
        [
            f"- {key}: {value}"

            for key, value
            in mapping.items()
        ]
    )


preferred_mapping_text = "\n".join(
    [
        f"- {factor} -> {prevention}"

        for factor, prevention
        in PREFERRED_PREVENTION_BY_FACTOR.items()
    ]
)


SYSTEM_PROMPT = f"""
You analyze short MSHA accident narratives.

This task extracts the reported event mechanism, the most defensible
contributing factor, and a preliminary prevention theme.

This is not a verified root-cause investigation.

RECORD TYPES
- ACUTE_ACCIDENT: A specific work-related acute event is described.
- OCCUPATIONAL_EXPOSURE_OR_ILLNESS: Chronic disease, cumulative trauma, or
  occupational illness is reported.
- NONWORK_RELATED_OR_NATURAL_EVENT: The narrative says the condition was not
  work related, describes a natural medical event, or describes a person not
  engaged in mining work.
- NONINJURY_REPORT: A reportable accident occurred without personal injury.
- INSUFFICIENT_EVENT_INFORMATION: The narrative does not adequately describe
  a work-related event.

CAUSE CATEGORIES
{format_definitions(CAUSE_DEFINITIONS)}

CONTRIBUTING-FACTOR CATEGORIES
{format_definitions(FACTOR_DEFINITIONS)}

PREVENTION CATEGORIES
{format_definitions(PREVENTION_DEFINITIONS)}

PREFERRED FACTOR-TO-PREVENTION MAPPING
{preferred_mapping_text}

CRITICAL DECISION RULES

1. First determine record_type.

2. If record_type is NONWORK_RELATED_OR_NATURAL_EVENT:
   - primary_cause = INSUFFICIENT_INFORMATION
   - primary_contributing_factor = NO_SEPARATE_FACTOR_STATED
   - prevention_theme = INSUFFICIENT_INFORMATION
   - use LOW support levels
   - explain the routing briefly.

3. If record_type is NONINJURY_REPORT, classify the reported event only when
   the mechanism is clear.

4. The primary cause describes WHAT HAPPENED.

5. The contributing factor describes WHY OR UNDER WHAT CONDITION the event
   occurred.

6. Do not use the event description itself as a separate contributing factor
   unless the narrative directly supports a condition or work method.

7. Use NO_SEPARATE_FACTOR_STATED when the narrative establishes the event but
   does not separately state or directly support why it happened.

8. Do not infer poor housekeeping merely because the word "slipped" appears.

9. POOR_HOUSEKEEPING_OR_SURFACE_CONDITION requires an explicit walking,
   standing, access, or working-surface hazard such as ice, water, mud,
   grease, debris, loose rock, uneven footing, a hole, or damaged surface.

10. A hand, tool, hose, load, strap, or component slipping is not
    housekeeping.

11. UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL requires explicit evidence
    that material was loose, unsupported, unstable, shifting, rolling,
    sliding, falling, or improperly positioned.

12. Do not classify every movable object as unstable material.

13. INADEQUATE_SECUREMENT_OR_RESTRAINT requires explicit evidence that
    fastening, blocking, support, restraint, chocking, bracing, or securement
    was missing, removed, inadequate, or failed.

14. LINE_OF_FIRE_BODY_POSITION applies when a person or body part is placed
    in a pinch point or in the path of moving equipment, loads, falling
    objects, tools, or released energy.

15. A pinched finger does not automatically prove unstable material. If the
    narrative identifies only the pinch point or body position, prefer
    LINE_OF_FIRE_BODY_POSITION. If no separate factor is supported, use
    NO_SEPARATE_FACTOR_STATED.

16. Use PROCEDURE_OR_WORK_METHOD when the task method or sequence created the
    exposure and no more specific factor is supported.

17. Use FOREIGN_BODY_OR_PARTICLE_CONTACT for chips, fragments, dust, or a
    small foreign object entering an eye. Do not use ELECTRICAL_CONTACT for
    metal particles unless electrical energy caused the event.

18. Use MANUAL_HANDLING_OR_OVEREXERTION for acute strain, back pain, groin
    pain, sprain, or hernia associated with lifting, pushing, pulling,
    carrying, bending while handling material, or exerting force.

19. Use HEAT_OR_ENVIRONMENTAL_EXPOSURE for acute heat illness, heat stress,
    dehydration caused by work conditions, cold, or weather exposure.

20. Use EQUIPMENT_CONDITION_OR_FAILURE only when failure, damage, breakage,
    wear, defect, leakage, or malfunction is explicitly described.

21. Use TRAINING_OR_EXPERIENCE only when inadequate training, proficiency,
    qualification, or experience is stated or strongly supported.

22. Use PPE_OR_PROTECTIVE_CONTROL only when PPE or a protective control is
    directly relevant. A foreign object in an eye does not by itself prove
    that eye protection was missing.

23. Select the prevention theme from the contributing factor.

24. Do not recommend SECURE_OR_REPOSITION_MATERIAL unless material
    instability, positioning, support, restraint, or securement is supported.

25. If the factor is LINE_OF_FIRE_BODY_POSITION, normally select
    ESTABLISH_LINE_OF_FIRE_EXCLUSION.

26. If the factor is POOR_HOUSEKEEPING_OR_SURFACE_CONDITION, normally select
    IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING.

27. If the factor is UNSTABLE_GROUND_OR_ROOF_CONDITION, normally select
    IMPROVE_GROUND_CONTROL_OR_SCALING.

28. If the factor is PROCEDURE_OR_WORK_METHOD, normally select
    REVISE_PROCEDURE_OR_JOB_PLAN.

29. If the cause is MANUAL_HANDLING_OR_OVEREXERTION and the factor supports
    handling or exertion, consider USE_MECHANICAL_HANDLING_ASSISTANCE or
    REVISE_PROCEDURE_OR_JOB_PLAN.

30. If the factor is ENVIRONMENTAL_CONDITION, normally select
    CONTROL_ENVIRONMENTAL_EXPOSURE.

31. If the factor is NO_SEPARATE_FACTOR_STATED, select
    INVESTIGATION_REQUIRED. Do not invent training, PPE, equipment repair,
    housekeeping, material securement, or a pre-job review.

32. Copy cause and factor evidence verbatim whenever possible.

33. Cause evidence must support the event mechanism.

34. Factor evidence must support the contributing factor, not merely the
    injury outcome.

35. If primary_contributing_factor is NO_SEPARATE_FACTOR_STATED, return an
    empty factor_evidence_text.

36. Confidence is a self-assessment, not a calibrated probability.

37. Support levels:
    HIGH = directly and explicitly supported.
    MEDIUM = strongly implied.
    LOW = incomplete, ambiguous, or dependent on uncertain interpretation.

38. Return only one valid JSON object.

REQUIRED JSON

{{
  "record_type": "one allowed record type",
  "primary_cause": "one allowed cause category",
  "primary_contributing_factor": "one allowed factor category",
  "prevention_theme": "one allowed prevention category",
  "cause_evidence_text": "short verbatim excerpt or empty string",
  "factor_evidence_text": "short verbatim excerpt or empty string",
  "prevention_rationale": "one short sentence grounded in the narrative",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "prevention_confidence": 0.0,
  "cause_support_level": "HIGH, MEDIUM, or LOW",
  "factor_support_level": "HIGH, MEDIUM, or LOW",
  "prevention_support_level": "HIGH, MEDIUM, or LOW",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 16. NORMALIZATION AND JSON UTILITIES
# =============================================================================

def normalize_token(
    value,
):

    return (
        str(value)
        .strip()
        .upper()
        .replace("-", "_")
        .replace(" ", "_")
        .replace("/", "_")
    )


def normalize_category(
    value,
):

    normalized = normalize_token(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def normalize_support(
    value,
):

    normalized = normalize_token(
        value
    )

    return SUPPORT_LEVEL_ALIASES.get(
        normalized,
        "LOW",
    )


def normalize_response_dictionary(
    response_dictionary,
):

    cleaned = dict(
        response_dictionary
    )

    cleaned["record_type"] = normalize_token(
        cleaned.get(
            "record_type",
            "",
        )
    )

    for field_name in [
        "primary_cause",
        "primary_contributing_factor",
        "prevention_theme",
    ]:

        cleaned[field_name] = normalize_category(
            cleaned.get(
                field_name,
                "",
            )
        )

    for field_name in [
        "cause_support_level",
        "factor_support_level",
        "prevention_support_level",
    ]:

        cleaned[field_name] = normalize_support(
            cleaned.get(
                field_name,
                "",
            )
        )

    for field_name in [
        "cause_confidence",
        "factor_confidence",
        "prevention_confidence",
    ]:

        try:

            cleaned[field_name] = float(
                cleaned.get(
                    field_name,
                    0.0,
                )
            )

        except Exception:

            cleaned[field_name] = 0.0

    for field_name in [
        "cause_evidence_text",
        "factor_evidence_text",
        "prevention_rationale",
        "insufficient_information_reason",
        "reviewer_note",
    ]:

        cleaned[field_name] = str(
            cleaned.get(
                field_name,
                "",
            )
            or ""
        ).strip()

    return cleaned


def extract_json_object(
    text,
):

    text = str(
        text
    ).strip()

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0
    in_string = False
    escaped = False

    for position in range(
        start,
        len(text),
    ):

        character = text[position]

        if escaped:

            escaped = False
            continue

        if (
            character == "\\"
            and in_string
        ):

            escaped = True
            continue

        if character == '"':

            in_string = not in_string
            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    position + 1
                ]

    raise ValueError(
        "A complete JSON object was not found."
    )


def normalize_evidence_text(
    value,
):

    text = str(
        value
    ).lower()

    text = (
        text.replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def exact_evidence_present(
    evidence,
    narrative,
):

    evidence = str(
        evidence
    ).strip()

    if not evidence:

        return False

    return (
        evidence.lower()
        in str(
            narrative
        ).lower()
    )


def normalized_evidence_present(
    evidence,
    narrative,
):

    evidence_normalized = normalize_evidence_text(
        evidence
    )

    narrative_normalized = normalize_evidence_text(
        narrative
    )

    if not evidence_normalized:

        return False

    return (
        evidence_normalized
        in narrative_normalized
    )


def evidence_token_overlap(
    evidence,
    narrative,
):

    evidence_tokens = set(
        normalize_evidence_text(
            evidence
        ).split()
    )

    narrative_tokens = set(
        normalize_evidence_text(
            narrative
        ).split()
    )

    if not evidence_tokens:

        return 0.0

    return (
        len(
            evidence_tokens
            .intersection(
                narrative_tokens
            )
        )
        / len(
            evidence_tokens
        )
    )


def factor_prevention_aligned(
    factor,
    prevention,
):

    allowed = ALLOWED_PREVENTIONS_BY_FACTOR.get(
        factor
    )

    if allowed is None:

        return True

    return prevention in allowed


def preferred_prevention_used(
    factor,
    prevention,
):

    preferred = PREFERRED_PREVENTION_BY_FACTOR.get(
        factor
    )

    if preferred is None:

        return True

    return prevention == preferred


# =============================================================================
# 17. DETERMINISTIC POST-PROCESSING RULES
#
# These rules do not invent new substantive classifications. They enforce
# relationships that were explicitly defined before model execution.
# =============================================================================

def apply_deterministic_consistency_rules(
    validated,
):

    output = validated.model_dump()

    repairs = []

    factor = output[
        "primary_contributing_factor"
    ]

    prevention = output[
        "prevention_theme"
    ]

    record_type = output[
        "record_type"
    ]

    if (
        record_type
        == "NONWORK_RELATED_OR_NATURAL_EVENT"
    ):

        output[
            "primary_cause"
        ] = "INSUFFICIENT_INFORMATION"

        output[
            "primary_contributing_factor"
        ] = "NO_SEPARATE_FACTOR_STATED"

        output[
            "prevention_theme"
        ] = "INSUFFICIENT_INFORMATION"

        output[
            "factor_evidence_text"
        ] = ""

        output[
            "cause_support_level"
        ] = "LOW"

        output[
            "factor_support_level"
        ] = "LOW"

        output[
            "prevention_support_level"
        ] = "LOW"

        repairs.append(
            "Applied nonwork/natural-event routing"
        )

    elif (
        factor
        == "NO_SEPARATE_FACTOR_STATED"
    ):

        output[
            "factor_evidence_text"
        ] = ""

        if prevention not in {
            "INVESTIGATION_REQUIRED",
            "INSUFFICIENT_INFORMATION",
        }:

            output[
                "prevention_theme"
            ] = "INVESTIGATION_REQUIRED"

            output[
                "prevention_support_level"
            ] = "LOW"

            output[
                "prevention_confidence"
            ] = min(
                float(
                    output[
                        "prevention_confidence"
                    ]
                ),
                0.60,
            )

            repairs.append(
                "Mapped no separate factor to investigation"
            )

    return output, repairs


# =============================================================================
# 18. LOAD MODEL
# =============================================================================

print("\n" + "=" * 95)
print("LOADING LOCAL QWEN MODEL")
print("=" * 95)


quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,
    use_fast=True,
    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = tokenizer.eos_token


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=quantization_config,
    device_map="auto",
    dtype=torch.float16,
    low_cpu_mem_usage=True,
    trust_remote_code=False,
)


model.eval()


print("Model loaded.")
print("Model device:", model.device)


# =============================================================================
# 19. GENERATION FUNCTION
# =============================================================================

def generate_model_response(
    messages,
):

    model_inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )

    model_inputs = {
        key:
            value.to(
                model.device
            )

        for key, value
        in model_inputs.items()
    }

    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        generated = model.generate(
            **model_inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.03,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )

    generated_ids = generated[
        0,
        input_length:
    ]

    response = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )

    del model_inputs
    del generated
    del generated_ids

    return response


# =============================================================================
# 20. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    may_be_truncated,
    preliminary_record_type,
):

    user_prompt = f"""
PRELIMINARY ROUTING HINT
{preliminary_record_type}

The routing hint is only a rule-based hint. Review the narrative and return the
most defensible record_type.

DATA QUALITY
Narrative may be truncated: {may_be_truncated}

NARRATIVE
{narrative}

Return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_response = ""

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]

        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        f"""
The previous response failed validation.

ERROR
{prior_error}

PREVIOUS RESPONSE
{prior_response}

Correct the response. Use only allowed categories and return only one JSON
object.
""".strip(),
                }
            )

        try:

            raw_response = generate_model_response(
                messages
            )

            last_response = raw_response

            json_text = extract_json_object(
                raw_response
            )

            response_dictionary = json.loads(
                json_text
            )

            normalized_dictionary = (
                normalize_response_dictionary(
                    response_dictionary
                )
            )

            validated = (
                IncidentExtraction
                .model_validate(
                    normalized_dictionary
                )
            )

            consistent_output, repairs = (
                apply_deterministic_consistency_rules(
                    validated
                )
            )

            validated = (
                IncidentExtraction
                .model_validate(
                    consistent_output
                )
            )

            cause_exact = exact_evidence_present(
                validated.cause_evidence_text,
                narrative,
            )

            cause_normalized = (
                normalized_evidence_present(
                    validated.cause_evidence_text,
                    narrative,
                )
            )

            cause_overlap = evidence_token_overlap(
                validated.cause_evidence_text,
                narrative,
            )

            factor_exact = exact_evidence_present(
                validated.factor_evidence_text,
                narrative,
            )

            factor_normalized = (
                normalized_evidence_present(
                    validated.factor_evidence_text,
                    narrative,
                )
            )

            factor_overlap = evidence_token_overlap(
                validated.factor_evidence_text,
                narrative,
            )

            alignment_valid = factor_prevention_aligned(
                validated.primary_contributing_factor,
                validated.prevention_theme,
            )

            preferred_used = preferred_prevention_used(
                validated.primary_contributing_factor,
                validated.prevention_theme,
            )

            review_reasons = []

            if may_be_truncated:

                review_reasons.append(
                    "Narrative may be truncated"
                )

            if (
                validated.record_type
                != "ACUTE_ACCIDENT"
            ):

                review_reasons.append(
                    "Record is outside primary acute-accident population"
                )

            if (
                validated.primary_cause
                == "INSUFFICIENT_INFORMATION"
            ):

                review_reasons.append(
                    "Cause is insufficient information"
                )

            if (
                validated.primary_contributing_factor
                == "NO_SEPARATE_FACTOR_STATED"
            ):

                review_reasons.append(
                    "No separate contributing factor stated"
                )

            cause_evidence_valid = (
                cause_exact
                or cause_normalized
                or cause_overlap >= 0.85
            )

            factor_evidence_valid = (
                factor_exact
                or factor_normalized
                or factor_overlap >= 0.85
            )

            if (
                validated.primary_cause
                != "INSUFFICIENT_INFORMATION"
                and not cause_evidence_valid
            ):

                review_reasons.append(
                    "Cause evidence not found in narrative"
                )

            if (
                validated.primary_contributing_factor
                != "NO_SEPARATE_FACTOR_STATED"
                and not factor_evidence_valid
            ):

                review_reasons.append(
                    "Factor evidence not found in narrative"
                )

            if not alignment_valid:

                review_reasons.append(
                    "Factor-prevention combination is outside allowed map"
                )

            if (
                alignment_valid
                and not preferred_used
            ):

                review_reasons.append(
                    "Alternative prevention selected instead of preferred mapping"
                )

            if min(
                validated.cause_confidence,
                validated.factor_confidence,
                validated.prevention_confidence,
            ) < LOW_CONFIDENCE_THRESHOLD:

                review_reasons.append(
                    "At least one confidence value is low"
                )

            if "LOW" in {
                validated.cause_support_level,
                validated.factor_support_level,
                validated.prevention_support_level,
            }:

                review_reasons.append(
                    "At least one support level is low"
                )

            return {
                **validated.model_dump(),

                "SCHEMA_VALID":
                    True,

                "CAUSE_EVIDENCE_EXACT":
                    cause_exact,

                "CAUSE_EVIDENCE_NORMALIZED":
                    cause_normalized,

                "CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        cause_overlap,
                        4,
                    ),

                "CAUSE_EVIDENCE_VALID":
                    cause_evidence_valid,

                "FACTOR_EVIDENCE_EXACT":
                    factor_exact,

                "FACTOR_EVIDENCE_NORMALIZED":
                    factor_normalized,

                "FACTOR_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        factor_overlap,
                        4,
                    ),

                "FACTOR_EVIDENCE_VALID":
                    (
                        factor_evidence_valid
                        if (
                            validated
                            .primary_contributing_factor
                            != "NO_SEPARATE_FACTOR_STATED"
                        )
                        else True
                    ),

                "FACTOR_PREVENTION_ALIGNED":
                    alignment_valid,

                "PREFERRED_PREVENTION_USED":
                    preferred_used,

                "AUTOMATIC_REPAIR_APPLIED":
                    bool(
                        repairs
                    ),

                "AUTOMATIC_REPAIR_NOTES":
                    "; ".join(
                        repairs
                    ),

                "REVIEW_REQUIRED":
                    bool(
                        review_reasons
                    ),

                "REVIEW_REASONS":
                    "; ".join(
                        review_reasons
                    ),

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "ATTEMPT_COUNT":
                    attempt_number,

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = last_response

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:350]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        "record_type":
            preliminary_record_type,

        "primary_cause":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NO_SEPARATE_FACTOR_STATED",

        "prevention_theme":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "factor_evidence_text":
            "",

        "prevention_rationale":
            "The response could not be validated.",

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "prevention_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "factor_support_level":
            "LOW",

        "prevention_support_level":
            "LOW",

        "insufficient_information_reason":
            "Generation or schema validation failed.",

        "reviewer_note":
            "",

        "SCHEMA_VALID":
            False,

        "CAUSE_EVIDENCE_EXACT":
            False,

        "CAUSE_EVIDENCE_NORMALIZED":
            False,

        "CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_EXACT":
            False,

        "FACTOR_EVIDENCE_NORMALIZED":
            False,

        "FACTOR_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "FACTOR_EVIDENCE_VALID":
            False,

        "FACTOR_PREVENTION_ALIGNED":
            False,

        "PREFERRED_PREVENTION_USED":
            False,

        "AUTOMATIC_REPAIR_APPLIED":
            False,

        "AUTOMATIC_REPAIR_NOTES":
            "",

        "REVIEW_REQUIRED":
            True,

        "REVIEW_REASONS":
            "Model response failed validation",

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAW_MODEL_RESPONSE":
            last_response,
    }


# =============================================================================
# 21. FUNCTIONAL TESTS
# =============================================================================

if RUN_FUNCTIONAL_TEST:

    print("\n" + "=" * 95)
    print("RUNNING BLOCK 1A FUNCTIONAL TESTS")
    print("=" * 95)

    functional_tests = [
        {
            "name":
                "Finger pinch should not default to unstable material",

            "narrative":
                "The employee placed his hand between the gate and roller "
                "while closing the gate and pinched his right index finger.",

            "hint":
                "ACUTE_ACCIDENT",
        },
        {
            "name":
                "Foreign body in eye",

            "narrative":
                "While shoveling dirt, a small foreign object entered the "
                "employee's left eye.",

            "hint":
                "ACUTE_ACCIDENT",
        },
        {
            "name":
                "Heat exposure",

            "narrative":
                "After working outside in high heat, the employee developed "
                "cramps and was treated for heat exhaustion.",

            "hint":
                "ACUTE_ACCIDENT",
        },
        {
            "name":
                "Manual handling hernia",

            "narrative":
                "The employee felt sharp groin pain while lifting a heavy "
                "cable spool and was later diagnosed with a hernia.",

            "hint":
                "ACUTE_ACCIDENT",
        },
        {
            "name":
                "Nonwork event",

            "narrative":
                "Emergency room personnel determined that the eye condition "
                "was not work related and could not be connected to the job.",

            "hint":
                "NONWORK_RELATED_OR_NATURAL_EVENT",
        },
    ]

    for test in functional_tests:

        print(
            f"\nTest: {test['name']}"
        )

        test_result = classify_narrative(
            narrative=test[
                "narrative"
            ],

            may_be_truncated=False,

            preliminary_record_type=test[
                "hint"
            ],
        )

        print(
            json.dumps(
                {
                    key:
                        value

                    for key, value
                    in test_result.items()

                    if key not in {
                        "RAW_MODEL_RESPONSE"
                    }
                },
                indent=2,
                default=str,
            )
        )

        if (
            test_result[
                "PROCESSING_STATUS"
            ]
            != "SUCCESS"
        ):

            raise RuntimeError(
                f"Functional test failed: {test['name']}"
            )


# =============================================================================
# 22. CHECKPOINT UTILITIES
# =============================================================================

def load_jsonl_checkpoint(
    path,
):

    path = Path(
        path
    )

    if not path.exists():

        return {}

    records = {}

    with path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:

                continue

            try:

                record = json.loads(
                    line
                )

                records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record

            except Exception:

                continue

    return records


def append_jsonl(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


block1a_checkpoint = load_jsonl_checkpoint(
    BLOCK1A_CHECKPOINT_JSONL
)


block1_checkpoint = load_jsonl_checkpoint(
    BLOCK1_CHECKPOINT_JSONL
)


print(
    "\nExisting Block 1a checkpoint records:",
    len(
        block1a_checkpoint
    ),
)

print(
    "Original Block 1 checkpoint records:",
    len(
        block1_checkpoint
    ),
)


# =============================================================================
# 23. PROCESS SAME DEVELOPMENT RECORDS
# =============================================================================

context_columns = [
    column
    for column in [
        "ACCIDENT_DT",
        "ACCIDENT_YEAR",
        "TIME_PERIOD",
        "SEVERITY_GROUP",
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
        "DAYS_RESTRICT",
        "DAYS_LOST",
        "PRELIMINARY_RECORD_TYPE_HINT",
    ]
    if column in development_df.columns
]


processed_records = []


development_df = (
    development_df
    .set_index(
        id_column,
        drop=False,
    )
    .loc[
        [
            record_id

            for record_id
            in BLOCK1_RECORD_IDS

            if record_id
            in found_ids
        ]
    ]
    .reset_index(
        drop=True
    )
)


for row_number, row in development_df.iterrows():

    source_record_id = str(
        row[
            id_column
        ]
    )

    if (
        not REPROCESS_COMPLETED
        and source_record_id
        in block1a_checkpoint
    ):

        print(
            f"[{row_number + 1}/"
            f"{len(development_df)}] "
            f"Checkpoint: {source_record_id}"
        )

        processed_records.append(
            block1a_checkpoint[
                source_record_id
            ]
        )

        continue

    print(
        f"[{row_number + 1}/"
        f"{len(development_df)}] "
        f"{row['SEVERITY_GROUP']} | "
        f"{row['TIME_PERIOD']} | "
        f"{source_record_id}"
    )

    start_time = time.time()

    model_result = classify_narrative(
        narrative=str(
            row[
                TEXT_COLUMN
            ]
        ),

        may_be_truncated=bool(
            row[
                "_MAY_BE_TRUNCATED"
            ]
        ),

        preliminary_record_type=str(
            row[
                "PRELIMINARY_RECORD_TYPE_HINT"
            ]
        ),
    )

    record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "NARRATIVE":
            str(
                row[
                    TEXT_COLUMN
                ]
            ),

        "NARRATIVE_WORD_COUNT":
            int(
                row[
                    "_WORD_COUNT"
                ]
            ),

        "NARRATIVE_CHARACTER_COUNT":
            int(
                row[
                    "_CHARACTER_COUNT"
                ]
            ),

        "NARRATIVE_MAY_BE_TRUNCATED":
            bool(
                row[
                    "_MAY_BE_TRUNCATED"
                ]
            ),

        **{
            column:
                (
                    None
                    if pd.isna(
                        row[
                            column
                        ]
                    )
                    else row[
                        column
                    ]
                )

            for column
            in context_columns
        },

        **model_result,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    processed_records.append(
        record
    )

    append_jsonl(
        record,
        BLOCK1A_CHECKPOINT_JSONL,
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 24. RESULTS DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    processed_records
)


results_df = results_df.rename(
    columns={
        "record_type":
            "RECORD_TYPE",

        "primary_cause":
            "PRIMARY_CAUSE",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "prevention_theme":
            "PREVENTION_THEME",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "factor_evidence_text":
            "FACTOR_EVIDENCE_TEXT",

        "prevention_rationale":
            "PREVENTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "prevention_confidence":
            "PREVENTION_CONFIDENCE",

        "cause_support_level":
            "CAUSE_SUPPORT_LEVEL",

        "factor_support_level":
            "FACTOR_SUPPORT_LEVEL",

        "prevention_support_level":
            "PREVENTION_SUPPORT_LEVEL",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


boolean_columns = [
    "SCHEMA_VALID",
    "CAUSE_EVIDENCE_EXACT",
    "CAUSE_EVIDENCE_NORMALIZED",
    "CAUSE_EVIDENCE_VALID",
    "FACTOR_EVIDENCE_EXACT",
    "FACTOR_EVIDENCE_NORMALIZED",
    "FACTOR_EVIDENCE_VALID",
    "FACTOR_PREVENTION_ALIGNED",
    "PREFERRED_PREVENTION_USED",
    "AUTOMATIC_REPAIR_APPLIED",
    "REVIEW_REQUIRED",
]


for column in boolean_columns:

    if column in results_df.columns:

        results_df[column] = (
            results_df[
                column
            ]
            .fillna(False)
            .astype(bool)
        )


# =============================================================================
# 25. LOAD AND NORMALIZE BLOCK 1 RESULTS
# =============================================================================

block1_df = pd.DataFrame(
    list(
        block1_checkpoint.values()
    )
)


if not block1_df.empty:

    block1_df = block1_df.rename(
        columns={
            "primary_cause":
                "PRIMARY_CAUSE",

            "primary_contributing_factor":
                "PRIMARY_CONTRIBUTING_FACTOR",

            "prevention_theme":
                "PREVENTION_THEME",

            "cause_support_level":
                "CAUSE_SUPPORT_LEVEL",

            "factor_support_level":
                "FACTOR_SUPPORT_LEVEL",

            "prevention_support_level":
                "PREVENTION_SUPPORT_LEVEL",
        }
    )


for column in [
    "SOURCE_RECORD_ID",
    "PRIMARY_CAUSE",
    "PRIMARY_CONTRIBUTING_FACTOR",
    "PREVENTION_THEME",
]:

    if column not in block1_df.columns:

        block1_df[column] = ""


block1_df[
    "SOURCE_RECORD_ID"
] = (
    block1_df[
        "SOURCE_RECORD_ID"
    ]
    .astype(str)
)


results_df[
    "SOURCE_RECORD_ID"
] = (
    results_df[
        "SOURCE_RECORD_ID"
    ]
    .astype(str)
)


comparison_df = results_df.merge(
    block1_df[
        [
            "SOURCE_RECORD_ID",
            "PRIMARY_CAUSE",
            "PRIMARY_CONTRIBUTING_FACTOR",
            "PREVENTION_THEME",
            "REVIEW_REQUIRED",
            "FACTOR_PREVENTION_ALIGNED",
            "PROCESSING_STATUS",
        ]
    ].rename(
        columns={
            "PRIMARY_CAUSE":
                "BLOCK1_CAUSE",

            "PRIMARY_CONTRIBUTING_FACTOR":
                "BLOCK1_FACTOR",

            "PREVENTION_THEME":
                "BLOCK1_PREVENTION",

            "REVIEW_REQUIRED":
                "BLOCK1_REVIEW_REQUIRED",

            "FACTOR_PREVENTION_ALIGNED":
                "BLOCK1_ALIGNMENT",

            "PROCESSING_STATUS":
                "BLOCK1_STATUS",
        }
    ),

    on="SOURCE_RECORD_ID",

    how="left",
)


comparison_df[
    "CAUSE_CHANGED"
] = (
    comparison_df[
        "PRIMARY_CAUSE"
    ]
    != comparison_df[
        "BLOCK1_CAUSE"
    ]
)


comparison_df[
    "FACTOR_CHANGED"
] = (
    comparison_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ]
    != comparison_df[
        "BLOCK1_FACTOR"
    ]
)


comparison_df[
    "PREVENTION_CHANGED"
] = (
    comparison_df[
        "PREVENTION_THEME"
    ]
    != comparison_df[
        "BLOCK1_PREVENTION"
    ]
)


# =============================================================================
# 26. SUMMARIES
# =============================================================================

def category_summary(
    dataframe,
    column,
    display_name,
):

    output = (
        dataframe[
            column
        ]
        .value_counts(
            dropna=False
        )
        .rename_axis(
            display_name
        )
        .reset_index(
            name="Count"
        )
    )

    output[
        "Percent"
    ] = (
        100
        * output[
            "Count"
        ]
        / len(
            dataframe
        )
    )

    return output


cause_summary = category_summary(
    results_df,
    "PRIMARY_CAUSE",
    "Primary cause",
)


factor_summary = category_summary(
    results_df,
    "PRIMARY_CONTRIBUTING_FACTOR",
    "Primary contributing factor",
)


prevention_summary = category_summary(
    results_df,
    "PREVENTION_THEME",
    "Prevention theme",
)


record_type_summary = category_summary(
    results_df,
    "RECORD_TYPE",
    "Record type",
)


def safe_rate(
    dataframe,
    column,
    desired_value=True,
):

    if (
        column not in dataframe.columns
        or len(dataframe) == 0
    ):

        return np.nan

    return float(
        dataframe[
            column
        ]
        .eq(
            desired_value
        )
        .mean()
    )


comparison_metrics = pd.DataFrame(
    [
        {
            "Measure":
                "Successful classifications",

            "Block 1":
                (
                    safe_rate(
                        block1_df,
                        "PROCESSING_STATUS",
                        "SUCCESS",
                    )
                ),

            "Block 1a":
                (
                    safe_rate(
                        results_df,
                        "PROCESSING_STATUS",
                        "SUCCESS",
                    )
                ),
        },
        {
            "Measure":
                "Review-required rate",

            "Block 1":
                safe_rate(
                    block1_df,
                    "REVIEW_REQUIRED",
                    True,
                ),

            "Block 1a":
                safe_rate(
                    results_df,
                    "REVIEW_REQUIRED",
                    True,
                ),
        },
        {
            "Measure":
                "Factor-prevention alignment rate",

            "Block 1":
                safe_rate(
                    block1_df,
                    "FACTOR_PREVENTION_ALIGNED",
                    True,
                ),

            "Block 1a":
                safe_rate(
                    results_df,
                    "FACTOR_PREVENTION_ALIGNED",
                    True,
                ),
        },
        {
            "Measure":
                "Cause evidence-valid rate",

            "Block 1":
                float(
                    (
                        block1_df.get(
                            "CAUSE_EVIDENCE_EXACT",
                            False
                        )
                        |
                        block1_df.get(
                            "CAUSE_EVIDENCE_NORMALIZED",
                            False
                        )
                    ).mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                safe_rate(
                    results_df,
                    "CAUSE_EVIDENCE_VALID",
                    True,
                ),
        },
        {
            "Measure":
                "Factor evidence-valid rate",

            "Block 1":
                float(
                    (
                        block1_df.get(
                            "FACTOR_EVIDENCE_EXACT",
                            False
                        )
                        |
                        block1_df.get(
                            "FACTOR_EVIDENCE_NORMALIZED",
                            False
                        )
                    ).mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                safe_rate(
                    results_df,
                    "FACTOR_EVIDENCE_VALID",
                    True,
                ),
        },
        {
            "Measure":
                "No separate factor rate",

            "Block 1":
                float(
                    block1_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .isin(
                        [
                            "NOT_STATED",
                            "NO_SEPARATE_FACTOR_STATED",
                        ]
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "NO_SEPARATE_FACTOR_STATED"
                    )
                    .mean()
                ),
        },
        {
            "Measure":
                "Unstable-material factor rate",

            "Block 1":
                float(
                    block1_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL"
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL"
                    )
                    .mean()
                ),
        },
        {
            "Measure":
                "Housekeeping factor rate",

            "Block 1":
                float(
                    block1_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION"
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PRIMARY_CONTRIBUTING_FACTOR"
                    ]
                    .eq(
                        "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION"
                    )
                    .mean()
                ),
        },
        {
            "Measure":
                "Secure-material prevention rate",

            "Block 1":
                float(
                    block1_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "SECURE_OR_REPOSITION_MATERIAL"
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "SECURE_OR_REPOSITION_MATERIAL"
                    )
                    .mean()
                ),
        },
        {
            "Measure":
                "Line-of-fire prevention rate",

            "Block 1":
                float(
                    block1_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "ESTABLISH_LINE_OF_FIRE_EXCLUSION"
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "ESTABLISH_LINE_OF_FIRE_EXCLUSION"
                    )
                    .mean()
                ),
        },
        {
            "Measure":
                "Mechanical handling prevention rate",

            "Block 1":
                float(
                    block1_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "USE_MECHANICAL_HANDLING_ASSISTANCE"
                    )
                    .mean()
                )
                if not block1_df.empty
                else np.nan,

            "Block 1a":
                float(
                    results_df[
                        "PREVENTION_THEME"
                    ]
                    .eq(
                        "USE_MECHANICAL_HANDLING_ASSISTANCE"
                    )
                    .mean()
                ),
        },
    ]
)


comparison_metrics[
    "Change"
] = (
    comparison_metrics[
        "Block 1a"
    ]
    - comparison_metrics[
        "Block 1"
    ]
)


# =============================================================================
# 27. REVIEW REASONS
# =============================================================================

review_reason_series = (
    results_df[
        "REVIEW_REASONS"
    ]
    .fillna("")
    .str.split(
        "; "
    )
    .explode()
)


review_reason_summary = (
    review_reason_series[
        review_reason_series
        .str.len()
        > 0
    ]
    .value_counts()
    .rename_axis(
        "Review reason"
    )
    .reset_index(
        name="Count"
    )
)


# =============================================================================
# 28. CROSS-TABLES
# =============================================================================

cause_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CAUSE"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


factor_by_severity = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


prevention_by_severity = pd.crosstab(
    results_df[
        "PREVENTION_THEME"
    ],

    results_df[
        "SEVERITY_GROUP"
    ],
)


factor_prevention_table = pd.crosstab(
    results_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],

    results_df[
        "PREVENTION_THEME"
    ],
)


# =============================================================================
# 29. EXAMPLES FOR REVIEW
# =============================================================================

improved_alignment_examples = (
    comparison_df[
        (
            comparison_df[
                "BLOCK1_ALIGNMENT"
            ]
            == False
        )
        &
        (
            comparison_df[
                "FACTOR_PREVENTION_ALIGNED"
            ]
            == True
        )
    ]
    .head(15)
)


changed_factor_examples = (
    comparison_df[
        comparison_df[
            "FACTOR_CHANGED"
        ]
    ]
    .head(15)
)


changed_prevention_examples = (
    comparison_df[
        comparison_df[
            "PREVENTION_CHANGED"
        ]
    ]
    .head(15)
)


remaining_review_examples = (
    comparison_df[
        comparison_df[
            "REVIEW_REQUIRED"
        ]
    ]
    .sort_values(
        [
            "PROCESSING_STATUS",
            "CAUSE_CONFIDENCE",
            "FACTOR_CONFIDENCE",
        ],
        ascending=[
            True,
            True,
            True,
        ],
    )
    .head(15)
)


# =============================================================================
# 30. PDF HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append("")

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(line),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]

        for start in range(
            0,
            len(wrapped),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=24,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[column] = (
                printable[column]
                .map(
                    lambda value:
                        (
                            ""
                            if pd.isna(value)
                            else f"{value:.4f}"
                        )
                )
            )

    printable.columns = [
        str(column)
        .replace(
            "_",
            " ",
        )
        .title()

        for column in printable.columns
    ]

    for column in printable.columns:

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    ""
                    if pd.isna(value)
                    else str(value)
            )
        )

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=column_wrap_width,
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(printable),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.42,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_bar_chart(
    title,
    dataframe,
    category_column,
    color,
    top_n=20,
):

    plot_df = (
        dataframe
        .head(
            top_n
        )
        .sort_values(
            "Count",
            ascending=True,
        )
    )

    figure = plt.figure(
        figsize=(10, 7)
    )

    axis = figure.add_subplot(
        111
    )

    axis.barh(
        plot_df[
            category_column
        ],
        plot_df[
            "Count"
        ],
        color=color,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        "Narrative count"
    )

    axis.grid(
        axis="x",
        alpha=0.25,
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_heatmap(
    title,
    dataframe,
    x_label,
    y_label,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(11, 8)
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(dataframe.columns)
        )
    )

    axis.set_xticklabels(
        dataframe.columns,
        rotation=30,
        ha="right",
    )

    axis.set_yticks(
        range(
            len(dataframe.index)
        )
    )

    axis.set_yticklabels(
        [
            str(value)
            .replace(
                "_",
                " ",
            )
            .title()

            for value
            in dataframe.index
        ],
        fontsize=7.2,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        x_label
    )

    axis.set_ylabel(
        y_label
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value > 0:

                axis.text(
                    column_number,
                    row_number,
                    str(value),
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_comparison_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=12,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )

        return

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        row_index,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 82,
                (
                    f"Example {example_number} | "
                    f"Record {row['SOURCE_RECORD_ID']} | "
                    f"{row.get('SEVERITY_GROUP', '')}"
                ),
                (
                    f"Block 1 cause: "
                    f"{row.get('BLOCK1_CAUSE', '')}"
                ),
                (
                    f"Block 1a cause: "
                    f"{row.get('PRIMARY_CAUSE', '')}"
                ),
                (
                    f"Block 1 factor: "
                    f"{row.get('BLOCK1_FACTOR', '')}"
                ),
                (
                    f"Block 1a factor: "
                    f"{row.get('PRIMARY_CONTRIBUTING_FACTOR', '')}"
                ),
                (
                    f"Block 1 prevention: "
                    f"{row.get('BLOCK1_PREVENTION', '')}"
                ),
                (
                    f"Block 1a prevention: "
                    f"{row.get('PREVENTION_THEME', '')}"
                ),
                (
                    f"Block 1a review reasons: "
                    f"{row.get('REVIEW_REASONS', '')}"
                ),
                "",
                "Narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "Block 1a cause evidence: "
                    f"{row.get('CAUSE_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Block 1a factor evidence: "
                    f"{row.get('FACTOR_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Block 1a prevention rationale: "
                    f"{row.get('PREVENTION_RATIONALE', '')}"
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.7,
        lines_per_page=49,
        wrap_width=104,
    )


# =============================================================================
# 31. BUILD COMPARISON PDF
# =============================================================================

# Cover.
figure = create_page()

figure.text(
    0.5,
    0.79,
    "MSHA Block 1a Ontology Refinement",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.73,
    "Block 1 Versus Block 1a Development Comparison",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.685,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Model: {MODEL_ID}",
    f"Prompt version: {PROMPT_VERSION}",
    f"Development records reprocessed: {len(results_df):,}",
    f"Block 1 successful results: "
    f"{int(block1_df['PROCESSING_STATUS'].eq('SUCCESS').sum())}",
    f"Block 1a successful results: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum())}",
    f"Block 1a review-required records: "
    f"{int(results_df['REVIEW_REQUIRED'].sum())}",
    "",
    "Purpose:",
    "Evaluate whether stricter category definitions, improved routing,",
    "evidence requirements, and factor-based prevention mapping reduce",
    "category drift before the ontology is frozen for Block 2.",
]

figure.text(
    0.11,
    0.57,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=9.1,
    family="monospace",
    linespacing=1.24,
)

save_page(
    figure
)


# Executive summary.
block1_review_rate = (
    float(
        block1_df[
            "REVIEW_REQUIRED"
        ].mean()
    )
    if not block1_df.empty
    else np.nan
)


block1a_review_rate = float(
    results_df[
        "REVIEW_REQUIRED"
    ].mean()
)


block1_alignment_rate = (
    float(
        block1_df[
            "FACTOR_PREVENTION_ALIGNED"
        ].mean()
    )
    if not block1_df.empty
    else np.nan
)


block1a_alignment_rate = float(
    results_df[
        "FACTOR_PREVENTION_ALIGNED"
    ].mean()
)


add_text_pages(
    "Executive Summary",
    [
        "Block 1a reprocessed the same development records used in Block 1.",
        "No new development sample was selected.",
        "",
        "The refinement tightened the definitions of unstable material,",
        "housekeeping, securement, line of fire, work method, and no-separate-",
        "factor cases.",
        "",
        "The cause ontology was expanded to identify foreign-body or particle",
        "contact and acute heat or environmental exposure. Acute strains and",
        "hernias were explicitly included under manual handling or",
        "overexertion.",
        "",
        "The prevention theme is now selected from the contributing factor using",
        "a preferred factor-to-prevention mapping. When no separate factor is",
        "stated, the system uses INVESTIGATION_REQUIRED rather than inventing",
        "training, PPE, material securement, or procedure improvements.",
        "",
        f"Block 1 review-required rate: {block1_review_rate:.1%}",
        f"Block 1a review-required rate: {block1a_review_rate:.1%}",
        "",
        f"Block 1 factor-prevention alignment rate: "
        f"{block1_alignment_rate:.1%}",
        f"Block 1a factor-prevention alignment rate: "
        f"{block1a_alignment_rate:.1%}",
        "",
        "The remaining sections show where labels changed and whether the",
        "revised system reduced the category concentrations observed in",
        "Block 1.",
    ],
    font_size=8.7,
)


# Revision summary.
add_text_pages(
    "Block 1a Revisions",
    [
        "1. Added record-type classification within the model output.",
        "",
        "2. Added FOREIGN_BODY_OR_PARTICLE_CONTACT.",
        "",
        "3. Added HEAT_OR_ENVIRONMENTAL_EXPOSURE.",
        "",
        "4. Explicitly routed acute hernias and strains to",
        "   MANUAL_HANDLING_OR_OVEREXERTION when supported.",
        "",
        "5. Restricted POOR_HOUSEKEEPING_OR_SURFACE_CONDITION to explicit",
        "   walking, standing, access, or working-surface hazards.",
        "",
        "6. Restricted UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL to explicitly",
        "   unstable, loose, unsupported, rolling, shifting, or falling material.",
        "",
        "7. Activated INADEQUATE_SECUREMENT_OR_RESTRAINT when missing or failed",
        "   blocking, fastening, support, restraint, or securement is stated.",
        "",
        "8. Emphasized LINE_OF_FIRE_BODY_POSITION for pinch points and exposure",
        "   to moving or falling objects.",
        "",
        "9. Replaced generic NOT_STATED with NO_SEPARATE_FACTOR_STATED.",
        "",
        "10. Required NO_SEPARATE_FACTOR_STATED to map to",
        "    INVESTIGATION_REQUIRED or INSUFFICIENT_INFORMATION.",
        "",
        "11. Added deterministic preferred factor-to-prevention mappings.",
        "",
        "12. Added evidence checks that distinguish event evidence from factor",
        "    evidence.",
        "",
        "13. Improved routing of nonwork, natural, illness, noninjury, and",
        "    insufficient-information records.",
    ],
    font_size=8.6,
)


# Metric comparison.
comparison_metrics_display = (
    comparison_metrics.copy()
)


for column in [
    "Block 1",
    "Block 1a",
    "Change",
]:

    comparison_metrics_display[column] = (
        comparison_metrics_display[
            column
        ]
        .map(
            lambda value:
                (
                    ""
                    if pd.isna(value)
                    else f"{value:.4f}"
                )
        )
    )


add_table_pages(
    "Block 1 Versus Block 1a Metrics",
    comparison_metrics_display,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=37,
)


# Comparison visual.
plot_metrics = comparison_metrics[
    comparison_metrics[
        "Measure"
    ]
    .isin(
        [
            "Review-required rate",
            "Factor-prevention alignment rate",
            "Cause evidence-valid rate",
            "Factor evidence-valid rate",
            "Unstable-material factor rate",
            "Housekeeping factor rate",
            "Secure-material prevention rate",
            "Line-of-fire prevention rate",
            "Mechanical handling prevention rate",
        ]
    )
].copy()


figure = plt.figure(
    figsize=(11, 7.5)
)

axis = figure.add_subplot(
    111
)

positions = np.arange(
    len(plot_metrics)
)

width = 0.36

axis.barh(
    positions - width / 2,
    plot_metrics[
        "Block 1"
    ],
    height=width,
    label="Block 1",
    color="#9AA4AE",
)

axis.barh(
    positions + width / 2,
    plot_metrics[
        "Block 1a"
    ],
    height=width,
    label="Block 1a",
    color="#2E6F9E",
)

axis.set_yticks(
    positions
)

axis.set_yticklabels(
    plot_metrics[
        "Measure"
    ]
)

axis.set_xlabel(
    "Rate"
)

axis.set_xlim(
    0,
    max(
        1.0,
        float(
            np.nanmax(
                plot_metrics[
                    [
                        "Block 1",
                        "Block 1a",
                    ]
                ].values
            )
        )
        + 0.05,
    ),
)

axis.set_title(
    "Block 1 and Block 1a Comparison",
    fontsize=13,
    fontweight="bold",
)

axis.grid(
    axis="x",
    alpha=0.25,
)

axis.legend()

figure.tight_layout()

save_page(
    figure
)


# Record types.
add_table_pages(
    "Block 1a Record-Type Distribution",
    record_type_summary,
    rows_per_page=12,
    font_size=7.5,
    column_wrap_width=38,
)


# Category distributions.
add_table_pages(
    "Block 1a Primary Cause Distribution",
    cause_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=38,
)


add_bar_chart(
    "Block 1a Primary Cause Distribution",
    cause_summary,
    "Primary cause",
    "#2E6F9E",
)


add_table_pages(
    "Block 1a Contributing-Factor Distribution",
    factor_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=38,
)


add_bar_chart(
    "Block 1a Contributing-Factor Distribution",
    factor_summary,
    "Primary contributing factor",
    "#4E8B57",
)


add_table_pages(
    "Block 1a Prevention-Theme Distribution",
    prevention_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=38,
)


add_bar_chart(
    "Block 1a Prevention-Theme Distribution",
    prevention_summary,
    "Prevention theme",
    "#B46A3C",
)


# Review reasons.
add_table_pages(
    "Block 1a Automated Review Reasons",
    review_reason_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=47,
)


# Heatmaps.
add_heatmap(
    "Block 1a Cause by Injury Severity",
    cause_by_severity,
    "Injury-severity group",
    "Primary cause",
)


add_heatmap(
    "Block 1a Factor by Injury Severity",
    factor_by_severity,
    "Injury-severity group",
    "Contributing factor",
)


add_heatmap(
    "Block 1a Prevention Theme by Injury Severity",
    prevention_by_severity,
    "Injury-severity group",
    "Prevention theme",
)


add_heatmap(
    "Block 1a Factor and Prevention Relationships",
    factor_prevention_table,
    "Prevention theme",
    "Contributing factor",
)


# Change counts.
change_summary = pd.DataFrame(
    [
        {
            "Measure":
                "Cause labels changed",

            "Count":
                int(
                    comparison_df[
                        "CAUSE_CHANGED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor labels changed",

            "Count":
                int(
                    comparison_df[
                        "FACTOR_CHANGED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Prevention labels changed",

            "Count":
                int(
                    comparison_df[
                        "PREVENTION_CHANGED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Block 1 alignment failures corrected",

            "Count":
                len(
                    improved_alignment_examples
                ),
        },
        {
            "Measure":
                "Block 1a automatic consistency repairs",

            "Count":
                int(
                    results_df[
                        "AUTOMATIC_REPAIR_APPLIED"
                    ].sum()
                ),
        },
    ]
)


add_table_pages(
    "Label and Consistency Changes",
    change_summary,
    rows_per_page=12,
    font_size=7.5,
    column_wrap_width=43,
)


# Example pages.
add_comparison_example_pages(
    "Examples with Revised Contributing Factors",
    changed_factor_examples,
    (
        "These examples show narratives for which the contributing-factor "
        "label changed between Block 1 and Block 1a. Review whether the new "
        "label is better supported by a separate factor-evidence excerpt."
    ),
    maximum_records=12,
)


add_comparison_example_pages(
    "Examples with Revised Prevention Themes",
    changed_prevention_examples,
    (
        "These examples show changes in prevention-theme selection after "
        "introducing the preferred factor-to-prevention mapping."
    ),
    maximum_records=12,
)


add_comparison_example_pages(
    "Corrected Factor-Prevention Alignment Examples",
    improved_alignment_examples,
    (
        "These records were outside the Block 1 compatibility map but are "
        "aligned under Block 1a. Confirm that the change reflects an improved "
        "factor or prevention decision rather than a mechanical remapping."
    ),
    maximum_records=12,
)


add_comparison_example_pages(
    "Remaining Block 1a Review Cases",
    remaining_review_examples,
    (
        "These records still require review because of missing evidence, low "
        "support, nonacute routing, unusual prevention selection, or failed "
        "validation."
    ),
    maximum_records=12,
)


# Freeze decision.
acceptance_lines = [
    "Recommended checks before freezing the system for Block 2:",
    "",
    "1. Successful and schema-valid responses should remain high.",
    "",
    "2. Review-required records should decline from the Block 1 level without",
    "   hiding genuinely ambiguous records.",
    "",
    "3. Factor-prevention alignment should improve substantially.",
    "",
    "4. SECURE_OR_REPOSITION_MATERIAL should no longer dominate unrelated",
    "   factors.",
    "",
    "5. ESTABLISH_LINE_OF_FIRE_EXCLUSION should be used for supported line-of-",
    "   fire and pinch-point cases.",
    "",
    "6. USE_MECHANICAL_HANDLING_ASSISTANCE or REVISE_PROCEDURE_OR_JOB_PLAN",
    "   should appear for supported manual-handling events.",
    "",
    "7. Housekeeping should be limited to explicit walking, standing, access,",
    "   or working-surface hazards.",
    "",
    "8. Foreign-body, heat-exposure, acute strain, hernia, nonwork, and natural",
    "   events should route consistently.",
    "",
    "9. High-support examples should contain distinct evidence for both the",
    "   event mechanism and the contributing factor.",
    "",
    "10. The ontology, prompt, aliases, routing rules, compatibility map, and",
    "    review rules should be frozen before Block 2.",
]


add_text_pages(
    "Block 1a Freeze Decision Criteria",
    acceptance_lines,
    font_size=8.6,
)


# Technical appendix.
add_text_pages(
    "Technical Appendix",
    [
        f"Model: {MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"GPU: {GPU_NAME}",
        f"Block 1 development IDs requested: {len(BLOCK1_RECORD_IDS):,}",
        f"Block 1a records found: {len(results_df):,}",
        f"Missing development IDs: {len(missing_ids):,}",
        f"Cause categories: {len(CAUSE_CATEGORIES)}",
        f"Factor categories: {len(FACTOR_CATEGORIES)}",
        f"Prevention categories: {len(PREVENTION_CATEGORIES)}",
        f"Maximum retries: {MAX_RETRIES}",
        f"Maximum new tokens: {MAX_NEW_TOKENS}",
        "",
        f"Original Block 1 checkpoint: "
        f"{os.path.abspath(BLOCK1_CHECKPOINT_JSONL)}",
        f"Block 1a checkpoint: "
        f"{os.path.abspath(BLOCK1A_CHECKPOINT_JSONL)}",
        f"Block 1a comparison PDF: "
        f"{os.path.abspath(OUTPUT_PDF)}",
    ],
    font_size=8.3,
)


# =============================================================================
# 32. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "MSHA Block 1a Ontology Refinement Comparison Report"
)

metadata[
    "Author"
] = (
    "Python and Qwen2.5-7B-Instruct"
)

metadata[
    "Subject"
] = (
    "Comparison of the original and refined MSHA cause, contributing-factor, "
    "and prevention-theme extraction systems"
)

metadata[
    "Keywords"
] = (
    "MSHA, NLP, Qwen, ontology refinement, cause extraction, "
    "contributing factors, prevention themes"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 33. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 1A REFINEMENT COMPLETE")
print("=" * 95)

print(
    f"Development records reprocessed: "
    f"{len(results_df):,}"
)

print(
    f"Successful classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum()):,}"
)

print(
    f"Failed classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('FAILED').sum()):,}"
)

print(
    f"Block 1 review-required rate: "
    f"{block1_review_rate:.1%}"
)

print(
    f"Block 1a review-required rate: "
    f"{block1a_review_rate:.1%}"
)

print(
    f"Block 1 alignment rate: "
    f"{block1_alignment_rate:.1%}"
)

print(
    f"Block 1a alignment rate: "
    f"{block1a_alignment_rate:.1%}"
)

print(
    f"Cause labels changed: "
    f"{int(comparison_df['CAUSE_CHANGED'].sum()):,}"
)

print(
    f"Factor labels changed: "
    f"{int(comparison_df['FACTOR_CHANGED'].sum()):,}"
)

print(
    f"Prevention labels changed: "
    f"{int(comparison_df['PREVENTION_CHANGED'].sum()):,}"
)

print(
    f"Comparison PDF: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print(
    f"Block 1a checkpoint: "
    f"{os.path.abspath(BLOCK1A_CHECKPOINT_JSONL)}"
)

print("=" * 95)


# =============================================================================
# 34. CONCISE NOTEBOOK DISPLAYS
# =============================================================================

display(
    comparison_metrics
)


display(
    record_type_summary
)


display(
    cause_summary
)


display(
    factor_summary
)


display(
    prevention_summary
)


display(
    review_reason_summary
)
````

````python
# =============================================================================
# BLOCK 1C
# FOCUSED FINAL ACCEPTANCE TEST BEFORE BLOCK 2
#
# PURPOSE
# -------
# Test the specific corrections recommended after Block 1a without rerunning
# the complete 1,000-record development set.
#
# Block 1c tests:
#   1. Whether STRUCK_BY_MOVING_OBJECT still absorbs caught-between,
#      loss-of-control, and unexpected-equipment-movement narratives.
#   2. Whether the revised shorter prompt reduces JSON/schema failures.
#   3. Whether valid model fields can be retained when one output field needs
#      normalization or deterministic repair.
#   4. Whether critical and advisory review flags are reported separately.
#   5. Whether the total number of corrected alignments is calculated before
#      limiting displayed examples.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. Block 1a checkpoint:
#       MSHA_Block1a_Refined_Checkpoint.jsonl
#
# 3. Original Block 1 checkpoint, used only for the corrected total-alignment
#    comparison:
#       MSHA_Block1_Development_Checkpoint.jsonl
#
# OUTPUT
# ------
#   MSHA_Block1c_Focused_Acceptance_Test_Report.pdf
#
# No CSV, Excel, JSON, or additional checkpoint file is produced.
#
# TEST DESIGN
# -----------
# Approximately 180 unique narratives are selected from Block 1a:
#
#   - up to 55 model-validation failures;
#   - 30 struck-by-moving-object cases;
#   - 25 caught-in-or-between cases;
#   - 25 loss-of-control cases;
#   - 20 unexpected-equipment-movement cases;
#   - 20 manual-handling/overexertion cases;
#   - 20 no-separate-factor cases;
#   - additional random controls if needed.
#
# Because groups overlap, the final sample is filled to the target with a
# random control sample from the remaining Block 1a records.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import time
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import (
    BaseModel,
    Field,
    model_validator,
)

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from matplotlib.backends.backend_pdf import PdfPages


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

BLOCK1_CHECKPOINT_JSONL = (
    "MSHA_Block1_Development_Checkpoint.jsonl"
)

BLOCK1A_CHECKPOINT_JSONL = (
    "MSHA_Block1a_Refined_Checkpoint.jsonl"
)

OUTPUT_PDF = (
    "MSHA_Block1c_Focused_Acceptance_Test_Report.pdf"
)

TEXT_COLUMN = "NARRATIVE"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

PROMPT_VERSION = (
    "MSHA_BLOCK1C_FOCUSED_ACCEPTANCE_V3.0"
)

RANDOM_STATE = 42

TARGET_TEST_SIZE = 180

MAX_NEW_TOKENS = 750

MAX_RETRIES = 4

LOW_CONFIDENCE_THRESHOLD = 0.70

TRUNCATION_CHARACTER_THRESHOLD = 383

RUN_FUNCTIONAL_TESTS = True

# Number of records requested from each targeted group.
SELECTION_TARGETS = {
    "BLOCK1A_FAILURE":
        55,

    "STRUCK_BY_MOVING_OBJECT":
        30,

    "CAUGHT_IN_OR_BETWEEN":
        25,

    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT":
        25,

    "UNEXPECTED_EQUIPMENT_MOVEMENT":
        20,

    "MANUAL_HANDLING_OR_OVEREXERTION":
        20,

    "NO_SEPARATE_FACTOR_STATED":
        20,
}


# =============================================================================
# 3. RUNTIME CHECKS
# =============================================================================

print("=" * 95)
print("BLOCK 1C - FOCUSED FINAL ACCEPTANCE TEST")
print("=" * 95)

print("Python executable:", sys.executable)
print("Python version:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Accelerate:", accelerate.__version__)
print("BitsAndBytes:", bitsandbytes.__version__)
print("Pydantic:", pydantic.__version__)
print("Pydantic Core:", pydantic_core.__version__)


if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. "
        "Attach the notebook to a GPU compute instance."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0).total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3


print("\nGPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


# =============================================================================
# 4. FINAL CANDIDATE ONTOLOGY
#
# This retains the Block 1a categories. Block 1c changes only:
#   - cause-selection hierarchy;
#   - concise prompt wording;
#   - response repair;
#   - review-flag organization;
#   - focused acceptance testing.
# =============================================================================

RECORD_TYPE_CATEGORIES = [
    "ACUTE_ACCIDENT",
    "OCCUPATIONAL_EXPOSURE_OR_ILLNESS",
    "NONWORK_RELATED_OR_NATURAL_EVENT",
    "NONINJURY_REPORT",
    "INSUFFICIENT_EVENT_INFORMATION",
]


CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NO_SEPARATE_FACTOR_STATED",
]


PREVENTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "CONTROL_ENVIRONMENTAL_EXPOSURE",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 5. FACTOR-TO-PREVENTION RELATIONSHIPS
# =============================================================================

PREFERRED_PREVENTION_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
        "SECURE_OR_REPOSITION_MATERIAL",

    "INADEQUATE_SECUREMENT_OR_RESTRAINT":
        "SECURE_OR_REPOSITION_MATERIAL",

    "LINE_OF_FIRE_BODY_POSITION":
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "UNSTABLE_GROUND_OR_ROOF_CONDITION":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",

    "INADEQUATE_GUARDING_OR_BARRIER":
        "INSTALL_OR_IMPROVE_GUARDING",

    "ENERGY_ISOLATION_DEFICIENCY":
        "ISOLATE_OR_CONTROL_STORED_ENERGY",

    "EQUIPMENT_CONDITION_OR_FAILURE":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "PROCEDURE_OR_WORK_METHOD":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "INADEQUATE_HAZARD_RECOGNITION":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "COMMUNICATION_OR_COORDINATION":
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",

    "TRAINING_OR_EXPERIENCE":
        "PROVIDE_TARGETED_TRAINING",

    "PPE_OR_PROTECTIVE_CONTROL":
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",

    "ENVIRONMENTAL_CONDITION":
        "CONTROL_ENVIRONMENTAL_EXPOSURE",

    "WORKSPACE_OR_ACCESS_LIMITATION":
        "REDESIGN_WORKSPACE_OR_ACCESS",

    "NO_SEPARATE_FACTOR_STATED":
        "INVESTIGATION_REQUIRED",
}


ALLOWED_PREVENTIONS_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_SECUREMENT_OR_RESTRAINT": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    },

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },

    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_GUARDING_OR_BARRIER": {
        "INSTALL_OR_IMPROVE_GUARDING",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "ENERGY_ISOLATION_DEFICIENCY": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    },

    "INADEQUATE_HAZARD_RECOGNITION": {
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "COMMUNICATION_OR_COORDINATION": {
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "TRAINING_OR_EXPERIENCE": {
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "PPE_OR_PROTECTIVE_CONTROL": {
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "ENVIRONMENTAL_CONDITION": {
        "CONTROL_ENVIRONMENTAL_EXPOSURE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "WORKSPACE_OR_ACCESS_LIMITATION": {
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "OTHER": {
        "ENGINEERING_REVIEW_REQUIRED",
        "INVESTIGATION_REQUIRED",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "NO_SEPARATE_FACTOR_STATED": {
        "INVESTIGATION_REQUIRED",
        "INSUFFICIENT_INFORMATION",
    },
}


# =============================================================================
# 6. CATEGORY ALIASES
# =============================================================================

CATEGORY_ALIASES = {
    "UNSTATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NONE_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_PROVIDED":
        "NO_SEPARATE_FACTOR_STATED",

    "NO_FACTOR_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "LOSS_OF_CONTROL":
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",

    "UNEXPECTED_MACHINE_MOVEMENT":
        "UNEXPECTED_EQUIPMENT_MOVEMENT",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "REVISE_PROCEDURE":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "PRE_JOB_HAZARD_REVIEW":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "EQUIPMENT_REPAIR_OR_INSPECTION":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "IMPROVE_HOUSEKEEPING":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "GROUND_CONTROL_REVIEW":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",

    "UNKNOWN":
        "INSUFFICIENT_INFORMATION",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",
}


SUPPORT_LEVEL_ALIASES = {
    "HIGH":
        "HIGH",

    "DIRECT":
        "HIGH",

    "EXPLICIT":
        "HIGH",

    "MEDIUM":
        "MEDIUM",

    "MODERATE":
        "MEDIUM",

    "IMPLIED":
        "MEDIUM",

    "LOW":
        "LOW",

    "WEAK":
        "LOW",

    "UNCERTAIN":
        "LOW",

    "NONE":
        "LOW",

    "":
        "LOW",
}


# =============================================================================
# 7. RESPONSE SCHEMA
# =============================================================================

class IncidentExtraction(BaseModel):

    record_type: str

    primary_cause: str

    primary_contributing_factor: str

    prevention_theme: str

    cause_evidence_text: str = ""

    factor_evidence_text: str = ""

    prevention_rationale: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    prevention_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    factor_support_level: str = "LOW"

    prevention_support_level: str = "LOW"

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_output(self):

        if self.record_type not in RECORD_TYPE_CATEGORIES:

            raise ValueError(
                f"Invalid record_type: {self.record_type}"
            )

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                f"Invalid primary_cause: {self.primary_cause}"
            )

        if (
            self.primary_contributing_factor
            not in FACTOR_CATEGORIES
        ):

            raise ValueError(
                "Invalid primary_contributing_factor: "
                f"{self.primary_contributing_factor}"
            )

        if self.prevention_theme not in PREVENTION_CATEGORIES:

            raise ValueError(
                f"Invalid prevention_theme: {self.prevention_theme}"
            )

        for field_name in [
            "cause_support_level",
            "factor_support_level",
            "prevention_support_level",
        ]:

            if getattr(
                self,
                field_name,
            ) not in {
                "HIGH",
                "MEDIUM",
                "LOW",
            }:

                raise ValueError(
                    f"{field_name} must be HIGH, MEDIUM, or LOW."
                )

        return self


# =============================================================================
# 8. LOAD CHECKPOINTS
# =============================================================================

def load_jsonl(
    path,
):

    path = Path(
        path
    )

    if not path.exists():

        raise FileNotFoundError(
            f"Required checkpoint was not found: {path}"
        )

    records = {}

    with path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:

                continue

            try:

                record = json.loads(
                    line
                )

                record_id = str(
                    record[
                        "SOURCE_RECORD_ID"
                    ]
                )

                records[
                    record_id
                ] = record

            except Exception:

                continue

    return records


block1_records = load_jsonl(
    BLOCK1_CHECKPOINT_JSONL
)


block1a_records = load_jsonl(
    BLOCK1A_CHECKPOINT_JSONL
)


block1_df = pd.DataFrame(
    list(
        block1_records.values()
    )
)


block1a_df = pd.DataFrame(
    list(
        block1a_records.values()
    )
)


if block1a_df.empty:

    raise ValueError(
        "The Block 1a checkpoint contains no records."
    )


block1_df[
    "SOURCE_RECORD_ID"
] = (
    block1_df[
        "SOURCE_RECORD_ID"
    ]
    .astype(str)
)


block1a_df[
    "SOURCE_RECORD_ID"
] = (
    block1a_df[
        "SOURCE_RECORD_ID"
    ]
    .astype(str)
)


print(
    f"\nBlock 1 records loaded: "
    f"{len(block1_df):,}"
)

print(
    f"Block 1a records loaded: "
    f"{len(block1a_df):,}"
)


# =============================================================================
# 9. STANDARDIZE BLOCK 1A COLUMN NAMES
# =============================================================================

BLOCK1A_RENAME = {
    "record_type":
        "RECORD_TYPE",

    "primary_cause":
        "PRIMARY_CAUSE",

    "primary_contributing_factor":
        "PRIMARY_CONTRIBUTING_FACTOR",

    "prevention_theme":
        "PREVENTION_THEME",

    "cause_evidence_text":
        "CAUSE_EVIDENCE_TEXT",

    "factor_evidence_text":
        "FACTOR_EVIDENCE_TEXT",

    "prevention_rationale":
        "PREVENTION_RATIONALE",

    "cause_confidence":
        "CAUSE_CONFIDENCE",

    "factor_confidence":
        "FACTOR_CONFIDENCE",

    "prevention_confidence":
        "PREVENTION_CONFIDENCE",

    "cause_support_level":
        "CAUSE_SUPPORT_LEVEL",

    "factor_support_level":
        "FACTOR_SUPPORT_LEVEL",

    "prevention_support_level":
        "PREVENTION_SUPPORT_LEVEL",

    "insufficient_information_reason":
        "INSUFFICIENT_INFORMATION_REASON",

    "reviewer_note":
        "REVIEWER_NOTE",
}


block1a_df = block1a_df.rename(
    columns=BLOCK1A_RENAME
)


block1_df = block1_df.rename(
    columns={
        "primary_cause":
            "PRIMARY_CAUSE",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "prevention_theme":
            "PREVENTION_THEME",
    }
)


# =============================================================================
# 10. SELECT THE FOCUSED TEST SAMPLE
# =============================================================================

selected_ids = set()

selection_records = []


def add_targeted_sample(
    dataframe,
    mask,
    group_name,
    requested_count,
    random_state,
):

    eligible = dataframe.loc[
        mask
        & ~dataframe[
            "SOURCE_RECORD_ID"
        ].isin(
            selected_ids
        )
    ].copy()

    selected_count = min(
        requested_count,
        len(eligible),
    )

    if selected_count > 0:

        chosen = eligible.sample(
            n=selected_count,
            random_state=random_state,
        )

        selected_ids.update(
            chosen[
                "SOURCE_RECORD_ID"
            ].astype(str)
        )

    selection_records.append(
        {
            "Target group":
                group_name,

            "Available before deduplication":
                int(
                    mask.sum()
                ),

            "Requested":
                requested_count,

            "Newly selected":
                selected_count,
        }
    )


# Failed Block 1a records.
add_targeted_sample(
    dataframe=block1a_df,

    mask=(
        block1a_df[
            "PROCESSING_STATUS"
        ]
        .ne(
            "SUCCESS"
        )
    ),

    group_name="Block 1a validation failures",

    requested_count=(
        SELECTION_TARGETS[
            "BLOCK1A_FAILURE"
        ]
    ),

    random_state=(
        RANDOM_STATE + 1
    ),
)


cause_groups = [
    "STRUCK_BY_MOVING_OBJECT",
    "CAUGHT_IN_OR_BETWEEN",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
]


for group_number, cause_name in enumerate(
    cause_groups,
    start=1,
):

    add_targeted_sample(
        dataframe=block1a_df,

        mask=(
            block1a_df[
                "PRIMARY_CAUSE"
            ]
            .eq(
                cause_name
            )
        ),

        group_name=cause_name,

        requested_count=(
            SELECTION_TARGETS[
                cause_name
            ]
        ),

        random_state=(
            RANDOM_STATE
            + 10
            + group_number
        ),
    )


add_targeted_sample(
    dataframe=block1a_df,

    mask=(
        block1a_df[
            "PRIMARY_CONTRIBUTING_FACTOR"
        ]
        .eq(
            "NO_SEPARATE_FACTOR_STATED"
        )
    ),

    group_name="NO_SEPARATE_FACTOR_STATED",

    requested_count=(
        SELECTION_TARGETS[
            "NO_SEPARATE_FACTOR_STATED"
        ]
    ),

    random_state=(
        RANDOM_STATE + 30
    ),
)


remaining_needed = (
    TARGET_TEST_SIZE
    - len(
        selected_ids
    )
)


if remaining_needed > 0:

    remaining_pool = block1a_df[
        ~block1a_df[
            "SOURCE_RECORD_ID"
        ]
        .isin(
            selected_ids
        )
    ].copy()

    fill_count = min(
        remaining_needed,
        len(
            remaining_pool
        ),
    )

    if fill_count > 0:

        fill_sample = remaining_pool.sample(
            n=fill_count,
            random_state=(
                RANDOM_STATE + 999
            ),
        )

        selected_ids.update(
            fill_sample[
                "SOURCE_RECORD_ID"
            ].astype(str)
        )

    selection_records.append(
        {
            "Target group":
                "Random control fill",

            "Available before deduplication":
                len(
                    remaining_pool
                ),

            "Requested":
                remaining_needed,

            "Newly selected":
                fill_count,
        }
    )


test_df = (
    block1a_df[
        block1a_df[
            "SOURCE_RECORD_ID"
        ]
        .isin(
            selected_ids
        )
    ]
    .copy()
)


test_df = (
    test_df.sample(
        frac=1,
        random_state=RANDOM_STATE,
    )
    .reset_index(
        drop=True
    )
)


selection_summary = pd.DataFrame(
    selection_records
)


print(
    f"\nFocused Block 1c test sample: "
    f"{len(test_df):,}"
)


# =============================================================================
# 11. CAUSE DECISION HIERARCHY
#
# The hierarchy is included in the model prompt and is also used in the PDF.
# =============================================================================

CAUSE_DECISION_HIERARCHY = """
CAUSE-SELECTION HIERARCHY

Apply the first definition that clearly matches the immediate event:

1. CAUGHT_IN_OR_BETWEEN
   Use when the person or body part is trapped, pinched, crushed, compressed,
   entangled, pulled into, or held between:
   - two objects;
   - moving and stationary equipment;
   - meshing or rotating parts;
   - material and a fixed surface.

   Pinch, crush, entanglement, and compression take priority over
   STRUCK_BY_MOVING_OBJECT.

2. VEHICLE_OR_MOBILE_EQUIPMENT_EVENT
   Use for collision, overturn, runaway, roadway departure, backing, or
   transportation events involving mobile equipment.

3. GROUND_OR_ROOF_FAILURE
   Use when roof, rib, face, highwall, slate, or ground failure is the
   immediate event.

4. PRESSURE_OR_STORED_ENERGY_RELEASE
   Use when released hydraulic, pneumatic, spring, tension, electrical, or
   other stored energy is the immediate event.

5. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT
   Use when the event begins because the worker loses control of a tool, hose,
   handle, strap, load, or equipment component. The loss of control must be
   explicitly described.

6. UNEXPECTED_EQUIPMENT_MOVEMENT
   Use when equipment or a component moves unexpectedly without an explicit
   loss of control by the worker and without an explicit mechanical failure.

7. STRUCK_BY_MOVING_OBJECT
   Use when a moving, falling, flying, rolling, or swinging object contacts
   the person but does not trap or compress the person between objects.

8. STRUCK_AGAINST_STATIONARY_OBJECT
   Use when the person's own movement causes contact with a stationary object.

9. SLIP_TRIP_OR_FALL
   Use when the person loses footing, balance, or elevation.

10. AWKWARD_MOVEMENT_OR_BODY_MOTION
    Use when an acute twist, reach, bend, turn, or step causes injury without
    a fall or substantial manual load.

11. MANUAL_HANDLING_OR_OVEREXERTION
    Use for acute strains, pain, sprains, and hernias associated with lifting,
    pushing, pulling, carrying, shoveling, or exerting force.

Do not use STRUCK_BY_MOVING_OBJECT as a general fallback when a more specific
mechanism above is directly supported.
""".strip()


# =============================================================================
# 12. SHORTENED SYSTEM PROMPT
#
# Block 1c intentionally removes repetitive explanatory text while preserving
# the important boundaries and output schema.
# =============================================================================

SYSTEM_PROMPT = f"""
You extract structured information from short MSHA accident narratives.

This is reported-event extraction, not verified root-cause analysis.

ALLOWED RECORD TYPES
{", ".join(RECORD_TYPE_CATEGORIES)}

ALLOWED CAUSES
{", ".join(CAUSE_CATEGORIES)}

ALLOWED FACTORS
{", ".join(FACTOR_CATEGORIES)}

ALLOWED PREVENTION THEMES
{", ".join(PREVENTION_CATEGORIES)}

{CAUSE_DECISION_HIERARCHY}

FACTOR RULES

- UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL requires explicit unstable,
  loose, unsupported, rolling, shifting, sliding, falling, or improperly
  positioned material.

- INADEQUATE_SECUREMENT_OR_RESTRAINT requires explicitly missing, removed,
  failed, or inadequate blocking, fastening, support, bracing, restraint,
  chocking, or securement.

- LINE_OF_FIRE_BODY_POSITION applies when a person or body part is placed in
  a pinch point or the path of moving equipment, tools, loads, falling
  objects, or released energy.

- POOR_HOUSEKEEPING_OR_SURFACE_CONDITION requires an explicit walking,
  standing, access, or working-surface hazard such as ice, water, mud, grease,
  debris, loose rock, holes, or uneven footing.

- A hand, hose, tool, strap, component, or load slipping is not housekeeping.

- PROCEDURE_OR_WORK_METHOD applies when the task method or sequence created
  the exposure and no more specific physical factor is directly supported.

- NO_SEPARATE_FACTOR_STATED applies when the narrative explains what happened
  but does not separately state or directly support why it happened.

- Factor evidence must support the factor. Injury outcome text is not factor
  evidence.

PREVENTION RULES

- Select prevention from the stated factor.
- Do not invent a factor solely to justify a prevention recommendation.
- NO_SEPARATE_FACTOR_STATED maps to INVESTIGATION_REQUIRED.
- LINE_OF_FIRE_BODY_POSITION normally maps to
  ESTABLISH_LINE_OF_FIRE_EXCLUSION.
- POOR_HOUSEKEEPING_OR_SURFACE_CONDITION normally maps to
  IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING.
- UNSTABLE_GROUND_OR_ROOF_CONDITION normally maps to
  IMPROVE_GROUND_CONTROL_OR_SCALING.
- ENERGY_ISOLATION_DEFICIENCY normally maps to
  ISOLATE_OR_CONTROL_STORED_ENERGY.
- PROCEDURE_OR_WORK_METHOD normally maps to
  REVISE_PROCEDURE_OR_JOB_PLAN.
- ENVIRONMENTAL_CONDITION normally maps to
  CONTROL_ENVIRONMENTAL_EXPOSURE.
- TRAINING requires stated training or proficiency evidence.
- PPE requires stated or directly supported PPE evidence.
- Equipment repair requires stated failure, damage, wear, leakage, defect,
  or malfunction.
- Mechanical handling assistance may be selected for supported manual-lifting,
  pushing, pulling, or load-positioning exposure.

RECORD ROUTING

For NONWORK_RELATED_OR_NATURAL_EVENT:
- primary_cause = INSUFFICIENT_INFORMATION
- primary_contributing_factor = NO_SEPARATE_FACTOR_STATED
- prevention_theme = INSUFFICIENT_INFORMATION

EVIDENCE

- Copy short evidence excerpts verbatim.
- Cause evidence must support the immediate event mechanism.
- Factor evidence must support the separate contributing factor.
- If factor = NO_SEPARATE_FACTOR_STATED, factor_evidence_text must be empty.

SUPPORT LEVELS

- HIGH = explicit.
- MEDIUM = strongly implied.
- LOW = incomplete or ambiguous.

Return only one JSON object with exactly these keys:

{{
  "record_type": "",
  "primary_cause": "",
  "primary_contributing_factor": "",
  "prevention_theme": "",
  "cause_evidence_text": "",
  "factor_evidence_text": "",
  "prevention_rationale": "",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "prevention_confidence": 0.0,
  "cause_support_level": "HIGH",
  "factor_support_level": "HIGH",
  "prevention_support_level": "HIGH",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 13. NORMALIZATION UTILITIES
# =============================================================================

def normalize_token(
    value,
):

    return (
        str(
            value
        )
        .strip()
        .upper()
        .replace(
            "-",
            "_",
        )
        .replace(
            " ",
            "_",
        )
        .replace(
            "/",
            "_",
        )
    )


def normalize_category(
    value,
):

    normalized = normalize_token(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def normalize_support(
    value,
):

    normalized = normalize_token(
        value
    )

    return SUPPORT_LEVEL_ALIASES.get(
        normalized,
        "LOW",
    )


def safe_float(
    value,
    default=0.0,
):

    try:

        output = float(
            value
        )

        return min(
            max(
                output,
                0.0,
            ),
            1.0,
        )

    except Exception:

        return default


def strip_markdown_fences(
    text,
):

    text = str(
        text
    ).strip()

    text = re.sub(
        r"^```(?:json)?",
        "",
        text,
        flags=re.IGNORECASE,
    )

    text = re.sub(
        r"```$",
        "",
        text,
    )

    return text.strip()


def extract_balanced_json(
    text,
):

    text = strip_markdown_fences(
        text
    )

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0
    in_string = False
    escaped = False

    for index in range(
        start,
        len(text),
    ):

        character = text[
            index
        ]

        if escaped:

            escaped = False
            continue

        if (
            character == "\\"
            and in_string
        ):

            escaped = True
            continue

        if character == '"':

            in_string = not in_string
            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    index + 1
                ]

    # Limited repair for output truncated immediately before closing braces.
    candidate = text[
        start:
    ].strip()

    if candidate.count("{") > candidate.count("}"):

        candidate += "}" * (
            candidate.count("{")
            - candidate.count("}")
        )

        return candidate

    raise ValueError(
        "A complete JSON object was not found."
    )


def repair_common_json_problems(
    json_text,
):

    repaired = str(
        json_text
    ).strip()

    repaired = (
        repaired.replace(
            "\u201c",
            '"',
        )
        .replace(
            "\u201d",
            '"',
        )
        .replace(
            "\u2019",
            "'",
        )
    )

    # Remove trailing commas before closing braces.
    repaired = re.sub(
        r",\s*}",
        "}",
        repaired,
    )

    # Replace Python constants outside ordinary narrative text patterns.
    repaired = re.sub(
        r":\s*None\b",
        ': ""',
        repaired,
    )

    repaired = re.sub(
        r":\s*True\b",
        ": true",
        repaired,
    )

    repaired = re.sub(
        r":\s*False\b",
        ": false",
        repaired,
    )

    return repaired


EXPECTED_KEYS = [
    "record_type",
    "primary_cause",
    "primary_contributing_factor",
    "prevention_theme",
    "cause_evidence_text",
    "factor_evidence_text",
    "prevention_rationale",
    "cause_confidence",
    "factor_confidence",
    "prevention_confidence",
    "cause_support_level",
    "factor_support_level",
    "prevention_support_level",
    "insufficient_information_reason",
    "reviewer_note",
]


DEFAULT_RESPONSE_VALUES = {
    "record_type":
        "INSUFFICIENT_EVENT_INFORMATION",

    "primary_cause":
        "INSUFFICIENT_INFORMATION",

    "primary_contributing_factor":
        "NO_SEPARATE_FACTOR_STATED",

    "prevention_theme":
        "INVESTIGATION_REQUIRED",

    "cause_evidence_text":
        "",

    "factor_evidence_text":
        "",

    "prevention_rationale":
        "",

    "cause_confidence":
        0.0,

    "factor_confidence":
        0.0,

    "prevention_confidence":
        0.0,

    "cause_support_level":
        "LOW",

    "factor_support_level":
        "LOW",

    "prevention_support_level":
        "LOW",

    "insufficient_information_reason":
        "",

    "reviewer_note":
        "",
}


def normalize_response_dictionary(
    response_dictionary,
):

    cleaned = dict(
        DEFAULT_RESPONSE_VALUES
    )

    retained_keys = []

    defaulted_keys = []

    for key in EXPECTED_KEYS:

        if key in response_dictionary:

            cleaned[
                key
            ] = response_dictionary[
                key
            ]

            retained_keys.append(
                key
            )

        else:

            defaulted_keys.append(
                key
            )

    cleaned[
        "record_type"
    ] = normalize_token(
        cleaned[
            "record_type"
        ]
    )

    for field_name in [
        "primary_cause",
        "primary_contributing_factor",
        "prevention_theme",
    ]:

        cleaned[
            field_name
        ] = normalize_category(
            cleaned[
                field_name
            ]
        )

    for field_name in [
        "cause_support_level",
        "factor_support_level",
        "prevention_support_level",
    ]:

        cleaned[
            field_name
        ] = normalize_support(
            cleaned[
                field_name
            ]
        )

    for field_name in [
        "cause_confidence",
        "factor_confidence",
        "prevention_confidence",
    ]:

        cleaned[
            field_name
        ] = safe_float(
            cleaned[
                field_name
            ]
        )

    for field_name in [
        "cause_evidence_text",
        "factor_evidence_text",
        "prevention_rationale",
        "insufficient_information_reason",
        "reviewer_note",
    ]:

        cleaned[
            field_name
        ] = str(
            cleaned[
                field_name
            ]
            or ""
        ).strip()

    return (
        cleaned,
        retained_keys,
        defaulted_keys,
    )


# =============================================================================
# 14. EVIDENCE UTILITIES
# =============================================================================

def normalize_evidence_text(
    value,
):

    text = str(
        value
    ).lower()

    text = (
        text.replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def exact_evidence_present(
    evidence,
    narrative,
):

    evidence = str(
        evidence
    ).strip()

    if not evidence:

        return False

    return (
        evidence.lower()
        in str(
            narrative
        ).lower()
    )


def normalized_evidence_present(
    evidence,
    narrative,
):

    evidence_normalized = normalize_evidence_text(
        evidence
    )

    narrative_normalized = normalize_evidence_text(
        narrative
    )

    if not evidence_normalized:

        return False

    return (
        evidence_normalized
        in narrative_normalized
    )


def evidence_token_overlap(
    evidence,
    narrative,
):

    evidence_tokens = set(
        normalize_evidence_text(
            evidence
        ).split()
    )

    narrative_tokens = set(
        normalize_evidence_text(
            narrative
        ).split()
    )

    if not evidence_tokens:

        return 0.0

    return (
        len(
            evidence_tokens
            .intersection(
                narrative_tokens
            )
        )
        / len(
            evidence_tokens
        )
    )


def evidence_is_valid(
    evidence,
    narrative,
):

    exact = exact_evidence_present(
        evidence,
        narrative,
    )

    normalized = normalized_evidence_present(
        evidence,
        narrative,
    )

    overlap = evidence_token_overlap(
        evidence,
        narrative,
    )

    valid = (
        exact
        or normalized
        or overlap >= 0.85
    )

    return (
        valid,
        exact,
        normalized,
        overlap,
    )


# =============================================================================
# 15. CONSISTENCY RULES
# =============================================================================

def factor_prevention_aligned(
    factor,
    prevention,
):

    allowed = ALLOWED_PREVENTIONS_BY_FACTOR.get(
        factor
    )

    if allowed is None:

        return True

    return prevention in allowed


def preferred_prevention_used(
    factor,
    prevention,
):

    preferred = PREFERRED_PREVENTION_BY_FACTOR.get(
        factor
    )

    if preferred is None:

        return True

    return prevention == preferred


def apply_consistency_rules(
    response_dictionary,
):

    output = dict(
        response_dictionary
    )

    repairs = []

    record_type = output[
        "record_type"
    ]

    factor = output[
        "primary_contributing_factor"
    ]

    prevention = output[
        "prevention_theme"
    ]

    if (
        record_type
        == "NONWORK_RELATED_OR_NATURAL_EVENT"
    ):

        output[
            "primary_cause"
        ] = "INSUFFICIENT_INFORMATION"

        output[
            "primary_contributing_factor"
        ] = "NO_SEPARATE_FACTOR_STATED"

        output[
            "prevention_theme"
        ] = "INSUFFICIENT_INFORMATION"

        output[
            "factor_evidence_text"
        ] = ""

        output[
            "cause_support_level"
        ] = "LOW"

        output[
            "factor_support_level"
        ] = "LOW"

        output[
            "prevention_support_level"
        ] = "LOW"

        repairs.append(
            "Applied nonwork/natural-event routing"
        )

    elif (
        factor
        == "NO_SEPARATE_FACTOR_STATED"
    ):

        output[
            "factor_evidence_text"
        ] = ""

        if prevention not in {
            "INVESTIGATION_REQUIRED",
            "INSUFFICIENT_INFORMATION",
        }:

            output[
                "prevention_theme"
            ] = "INVESTIGATION_REQUIRED"

            output[
                "prevention_support_level"
            ] = "LOW"

            output[
                "prevention_confidence"
            ] = min(
                safe_float(
                    output[
                        "prevention_confidence"
                    ]
                ),
                0.60,
            )

            repairs.append(
                "Mapped no separate factor to investigation"
            )

    return output, repairs


# =============================================================================
# 16. LOAD THE QWEN MODEL
# =============================================================================

print("\n" + "=" * 95)
print("LOADING LOCAL QWEN MODEL")
print("=" * 95)


quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,
    use_fast=True,
    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = tokenizer.eos_token


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=quantization_config,
    device_map="auto",
    dtype=torch.float16,
    low_cpu_mem_usage=True,
    trust_remote_code=False,
)


model.eval()


print("Model loaded successfully.")
print("Model device:", model.device)


# =============================================================================
# 17. GENERATION FUNCTION
# =============================================================================

def generate_model_response(
    messages,
    max_new_tokens,
):

    model_inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )

    model_inputs = {
        key:
            value.to(
                model.device
            )

        for key, value
        in model_inputs.items()
    }

    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[-1]
    )

    with torch.inference_mode():

        generated = model.generate(
            **model_inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            repetition_penalty=1.02,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )

    generated_ids = generated[
        0,
        input_length:
    ]

    response = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )

    del model_inputs
    del generated
    del generated_ids

    return response


# =============================================================================
# 18. TARGETED RETRY PROMPT
# =============================================================================

def build_retry_message(
    error,
    prior_response,
):

    error_text = str(
        error
    )

    likely_field = "unknown"

    for field_name in EXPECTED_KEYS:

        if field_name in error_text:

            likely_field = field_name
            break

    return f"""
The previous response was not a valid schema response.

LIKELY INVALID FIELD
{likely_field}

VALIDATION ERROR
{error_text}

PREVIOUS RESPONSE
{prior_response}

Return one complete JSON object only.

Use exactly these allowed values:

record_type:
{RECORD_TYPE_CATEGORIES}

primary_cause:
{CAUSE_CATEGORIES}

primary_contributing_factor:
{FACTOR_CATEGORIES}

prevention_theme:
{PREVENTION_CATEGORIES}

Support levels must be HIGH, MEDIUM, or LOW.

Do not add markdown or explanation outside the JSON object.
""".strip()


# =============================================================================
# 19. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    preliminary_record_type,
    may_be_truncated,
):

    user_prompt = f"""
PRELIMINARY ROUTING HINT
{preliminary_record_type}

NARRATIVE MAY BE TRUNCATED
{may_be_truncated}

NARRATIVE
{narrative}

Apply the cause hierarchy and return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_response = ""

    last_retained_keys = []

    last_defaulted_keys = []

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]

        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        build_retry_message(
                            error=prior_error,
                            prior_response=prior_response,
                        ),
                }
            )

        try:

            raw_response = generate_model_response(
                messages=messages,
                max_new_tokens=(
                    MAX_NEW_TOKENS
                ),
            )

            last_response = raw_response

            json_text = extract_balanced_json(
                raw_response
            )

            json_text = repair_common_json_problems(
                json_text
            )

            response_dictionary = json.loads(
                json_text
            )

            (
                normalized_dictionary,
                retained_keys,
                defaulted_keys,
            ) = normalize_response_dictionary(
                response_dictionary
            )

            last_retained_keys = retained_keys

            last_defaulted_keys = defaulted_keys

            (
                consistent_dictionary,
                consistency_repairs,
            ) = apply_consistency_rules(
                normalized_dictionary
            )

            validated = (
                IncidentExtraction
                .model_validate(
                    consistent_dictionary
                )
            )

            cause_valid, cause_exact, cause_normalized, cause_overlap = (
                evidence_is_valid(
                    evidence=(
                        validated
                        .cause_evidence_text
                    ),
                    narrative=narrative,
                )
            )

            factor_valid, factor_exact, factor_normalized, factor_overlap = (
                evidence_is_valid(
                    evidence=(
                        validated
                        .factor_evidence_text
                    ),
                    narrative=narrative,
                )
            )

            if (
                validated
                .primary_contributing_factor
                == "NO_SEPARATE_FACTOR_STATED"
            ):

                factor_valid = True

            alignment_valid = factor_prevention_aligned(
                factor=(
                    validated
                    .primary_contributing_factor
                ),
                prevention=(
                    validated
                    .prevention_theme
                ),
            )

            preferred_used = preferred_prevention_used(
                factor=(
                    validated
                    .primary_contributing_factor
                ),
                prevention=(
                    validated
                    .prevention_theme
                ),
            )

            critical_reasons = []

            advisory_reasons = []

            if (
                validated.record_type
                != "ACUTE_ACCIDENT"
            ):

                critical_reasons.append(
                    "Record routed outside primary acute-accident population"
                )

            if (
                validated.primary_cause
                == "INSUFFICIENT_INFORMATION"
            ):

                critical_reasons.append(
                    "Cause is insufficient information"
                )

            if (
                validated.primary_cause
                != "INSUFFICIENT_INFORMATION"
                and not cause_valid
            ):

                critical_reasons.append(
                    "Cause evidence is not supported by narrative"
                )

            if (
                validated
                .primary_contributing_factor
                != "NO_SEPARATE_FACTOR_STATED"
                and not factor_valid
            ):

                critical_reasons.append(
                    "Factor evidence is not supported by narrative"
                )

            if not alignment_valid:

                critical_reasons.append(
                    "Factor-prevention combination is outside allowed map"
                )

            if defaulted_keys:

                critical_reasons.append(
                    "One or more required fields were absent and defaulted"
                )

            if may_be_truncated:

                advisory_reasons.append(
                    "Narrative may be truncated"
                )

            if (
                validated
                .primary_contributing_factor
                == "NO_SEPARATE_FACTOR_STATED"
            ):

                advisory_reasons.append(
                    "No separate contributing factor stated"
                )

            if (
                alignment_valid
                and not preferred_used
            ):

                advisory_reasons.append(
                    "Allowed alternative prevention used instead of preferred mapping"
                )

            if min(
                validated.cause_confidence,
                validated.factor_confidence,
                validated.prevention_confidence,
            ) < LOW_CONFIDENCE_THRESHOLD:

                advisory_reasons.append(
                    "At least one confidence value is low"
                )

            if "LOW" in {
                validated.cause_support_level,
                validated.factor_support_level,
                validated.prevention_support_level,
            }:

                advisory_reasons.append(
                    "At least one support level is low"
                )

            all_repairs = list(
                consistency_repairs
            )

            if defaulted_keys:

                all_repairs.append(
                    "Defaulted missing fields: "
                    + ", ".join(
                        defaulted_keys
                    )
                )

            return {
                **validated.model_dump(),

                "SCHEMA_VALID":
                    True,

                "FIELDS_RETAINED_FROM_MODEL":
                    "; ".join(
                        retained_keys
                    ),

                "FIELDS_DEFAULTED":
                    "; ".join(
                        defaulted_keys
                    ),

                "CAUSE_EVIDENCE_EXACT":
                    cause_exact,

                "CAUSE_EVIDENCE_NORMALIZED":
                    cause_normalized,

                "CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        cause_overlap,
                        4,
                    ),

                "CAUSE_EVIDENCE_VALID":
                    cause_valid,

                "FACTOR_EVIDENCE_EXACT":
                    factor_exact,

                "FACTOR_EVIDENCE_NORMALIZED":
                    factor_normalized,

                "FACTOR_EVIDENCE_TOKEN_OVERLAP":
                    round(
                        factor_overlap,
                        4,
                    ),

                "FACTOR_EVIDENCE_VALID":
                    factor_valid,

                "FACTOR_PREVENTION_ALIGNED":
                    alignment_valid,

                "PREFERRED_PREVENTION_USED":
                    preferred_used,

                "AUTOMATIC_REPAIR_APPLIED":
                    bool(
                        all_repairs
                    ),

                "AUTOMATIC_REPAIR_NOTES":
                    "; ".join(
                        all_repairs
                    ),

                "CRITICAL_REVIEW_REQUIRED":
                    bool(
                        critical_reasons
                    ),

                "CRITICAL_REVIEW_REASONS":
                    "; ".join(
                        critical_reasons
                    ),

                "ADVISORY_REVIEW":
                    bool(
                        advisory_reasons
                    ),

                "ADVISORY_REVIEW_REASONS":
                    "; ".join(
                        advisory_reasons
                    ),

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "ATTEMPT_COUNT":
                    attempt_number,

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = last_response

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:300]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        "record_type":
            preliminary_record_type,

        "primary_cause":
            "INSUFFICIENT_INFORMATION",

        "primary_contributing_factor":
            "NO_SEPARATE_FACTOR_STATED",

        "prevention_theme":
            "INVESTIGATION_REQUIRED",

        "cause_evidence_text":
            "",

        "factor_evidence_text":
            "",

        "prevention_rationale":
            "The model response could not be validated.",

        "cause_confidence":
            0.0,

        "factor_confidence":
            0.0,

        "prevention_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "factor_support_level":
            "LOW",

        "prevention_support_level":
            "LOW",

        "insufficient_information_reason":
            "Generation or schema validation failed.",

        "reviewer_note":
            "",

        "SCHEMA_VALID":
            False,

        "FIELDS_RETAINED_FROM_MODEL":
            "; ".join(
                last_retained_keys
            ),

        "FIELDS_DEFAULTED":
            "; ".join(
                last_defaulted_keys
            ),

        "CAUSE_EVIDENCE_EXACT":
            False,

        "CAUSE_EVIDENCE_NORMALIZED":
            False,

        "CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_EXACT":
            False,

        "FACTOR_EVIDENCE_NORMALIZED":
            False,

        "FACTOR_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "FACTOR_EVIDENCE_VALID":
            False,

        "FACTOR_PREVENTION_ALIGNED":
            False,

        "PREFERRED_PREVENTION_USED":
            False,

        "AUTOMATIC_REPAIR_APPLIED":
            False,

        "AUTOMATIC_REPAIR_NOTES":
            "",

        "CRITICAL_REVIEW_REQUIRED":
            True,

        "CRITICAL_REVIEW_REASONS":
            "Model response failed validation",

        "ADVISORY_REVIEW":
            True,

        "ADVISORY_REVIEW_REASONS":
            "Low support fallback output",

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAW_MODEL_RESPONSE":
            last_response,
    }


# =============================================================================
# 20. FUNCTIONAL ACCEPTANCE TESTS
# =============================================================================

if RUN_FUNCTIONAL_TESTS:

    print("\n" + "=" * 95)
    print("RUNNING BLOCK 1C FUNCTIONAL ACCEPTANCE TESTS")
    print("=" * 95)

    functional_tests = [
        {
            "name":
                "Caught-between priority",

            "narrative":
                (
                    "The employee's hand was pinched between the moving "
                    "gate and the stationary frame."
                ),

            "expected_cause":
                "CAUGHT_IN_OR_BETWEEN",
        },
        {
            "name":
                "Struck-by without entrapment",

            "narrative":
                (
                    "A loose wrench fell from the platform and struck the "
                    "employee on the shoulder."
                ),

            "expected_cause":
                "STRUCK_BY_MOVING_OBJECT",
        },
        {
            "name":
                "Loss of tool control",

            "narrative":
                (
                    "The employee lost control of the sledge hammer and the "
                    "hammer head struck the employee's leg."
                ),

            "expected_cause":
                "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        },
        {
            "name":
                "Unexpected equipment movement",

            "narrative":
                (
                    "The loader door moved unexpectedly and closed on the "
                    "employee's finger."
                ),

            "expected_cause":
                "CAUGHT_IN_OR_BETWEEN",
        },
        {
            "name":
                "Manual handling hernia",

            "narrative":
                (
                    "The employee felt sharp groin pain while lifting a heavy "
                    "cable spool and was later diagnosed with a hernia."
                ),

            "expected_cause":
                "MANUAL_HANDLING_OR_OVEREXERTION",
        },
    ]

    functional_test_records = []

    for test_number, test in enumerate(
        functional_tests,
        start=1,
    ):

        print(
            f"\nFunctional test {test_number}: "
            f"{test['name']}"
        )

        result = classify_narrative(
            narrative=test[
                "narrative"
            ],

            preliminary_record_type=(
                "ACUTE_ACCIDENT"
            ),

            may_be_truncated=False,
        )

        actual_cause = result[
            "primary_cause"
        ]

        passed = (
            result[
                "PROCESSING_STATUS"
            ]
            == "SUCCESS"
            and actual_cause
            == test[
                "expected_cause"
            ]
        )

        functional_test_records.append(
            {
                "Test":
                    test[
                        "name"
                    ],

                "Expected cause":
                    test[
                        "expected_cause"
                    ],

                "Actual cause":
                    actual_cause,

                "Status":
                    (
                        "PASS"
                        if passed
                        else "REVIEW"
                    ),
            }
        )

        print(
            "Expected:",
            test[
                "expected_cause"
            ]
        )

        print(
            "Actual:",
            actual_cause
        )

        print(
            "Status:",
            (
                "PASS"
                if passed
                else "REVIEW"
            )
        )


    functional_test_df = pd.DataFrame(
        functional_test_records
    )

else:

    functional_test_df = pd.DataFrame()


# =============================================================================
# 21. PROCESS THE FOCUSED TEST SAMPLE
# =============================================================================

processed_records = []


for row_number, row in test_df.iterrows():

    source_record_id = str(
        row[
            "SOURCE_RECORD_ID"
        ]
    )

    narrative = str(
        row.get(
            "NARRATIVE",
            "",
        )
    )

    preliminary_record_type = str(
        row.get(
            "PRELIMINARY_RECORD_TYPE_HINT",
            row.get(
                "RECORD_TYPE",
                "ACUTE_ACCIDENT",
            ),
        )
    )

    if (
        preliminary_record_type
        not in RECORD_TYPE_CATEGORIES
    ):

        preliminary_record_type = (
            "ACUTE_ACCIDENT"
        )

    may_be_truncated = bool(
        row.get(
            "NARRATIVE_MAY_BE_TRUNCATED",
            len(
                narrative
            )
            >= TRUNCATION_CHARACTER_THRESHOLD,
        )
    )

    print(
        f"[{row_number + 1}/"
        f"{len(test_df)}] "
        f"{source_record_id} | "
        f"Block 1a cause: "
        f"{row.get('PRIMARY_CAUSE', '')}"
    )

    start_time = time.time()

    result = classify_narrative(
        narrative=narrative,

        preliminary_record_type=(
            preliminary_record_type
        ),

        may_be_truncated=(
            may_be_truncated
        ),
    )

    record = {
        "SOURCE_RECORD_ID":
            source_record_id,

        "NARRATIVE":
            narrative,

        "SEVERITY_GROUP":
            row.get(
                "SEVERITY_GROUP",
                "",
            ),

        "TIME_PERIOD":
            row.get(
                "TIME_PERIOD",
                "",
            ),

        "BLOCK1A_RECORD_TYPE":
            row.get(
                "RECORD_TYPE",
                "",
            ),

        "BLOCK1A_CAUSE":
            row.get(
                "PRIMARY_CAUSE",
                "",
            ),

        "BLOCK1A_FACTOR":
            row.get(
                "PRIMARY_CONTRIBUTING_FACTOR",
                "",
            ),

        "BLOCK1A_PREVENTION":
            row.get(
                "PREVENTION_THEME",
                "",
            ),

        "BLOCK1A_STATUS":
            row.get(
                "PROCESSING_STATUS",
                "",
            ),

        "BLOCK1A_REVIEW_REQUIRED":
            row.get(
                "REVIEW_REQUIRED",
                False,
            ),

        "BLOCK1A_ALIGNMENT":
            row.get(
                "FACTOR_PREVENTION_ALIGNED",
                False,
            ),

        **result,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    processed_records.append(
        record
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 22. BUILD BLOCK 1C DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    processed_records
)


results_df = results_df.rename(
    columns={
        "record_type":
            "BLOCK1C_RECORD_TYPE",

        "primary_cause":
            "BLOCK1C_CAUSE",

        "primary_contributing_factor":
            "BLOCK1C_FACTOR",

        "prevention_theme":
            "BLOCK1C_PREVENTION",

        "cause_evidence_text":
            "BLOCK1C_CAUSE_EVIDENCE",

        "factor_evidence_text":
            "BLOCK1C_FACTOR_EVIDENCE",

        "prevention_rationale":
            "BLOCK1C_PREVENTION_RATIONALE",

        "cause_confidence":
            "BLOCK1C_CAUSE_CONFIDENCE",

        "factor_confidence":
            "BLOCK1C_FACTOR_CONFIDENCE",

        "prevention_confidence":
            "BLOCK1C_PREVENTION_CONFIDENCE",

        "cause_support_level":
            "BLOCK1C_CAUSE_SUPPORT",

        "factor_support_level":
            "BLOCK1C_FACTOR_SUPPORT",

        "prevention_support_level":
            "BLOCK1C_PREVENTION_SUPPORT",
    }
)


for boolean_column in [
    "SCHEMA_VALID",
    "CAUSE_EVIDENCE_VALID",
    "FACTOR_EVIDENCE_VALID",
    "FACTOR_PREVENTION_ALIGNED",
    "PREFERRED_PREVENTION_USED",
    "AUTOMATIC_REPAIR_APPLIED",
    "CRITICAL_REVIEW_REQUIRED",
    "ADVISORY_REVIEW",
]:

    if boolean_column in results_df.columns:

        results_df[
            boolean_column
        ] = (
            results_df[
                boolean_column
            ]
            .fillna(False)
            .astype(bool)
        )


results_df[
    "CAUSE_CHANGED_FROM_BLOCK1A"
] = (
    results_df[
        "BLOCK1C_CAUSE"
    ]
    != results_df[
        "BLOCK1A_CAUSE"
    ]
)


results_df[
    "FACTOR_CHANGED_FROM_BLOCK1A"
] = (
    results_df[
        "BLOCK1C_FACTOR"
    ]
    != results_df[
        "BLOCK1A_FACTOR"
    ]
)


results_df[
    "PREVENTION_CHANGED_FROM_BLOCK1A"
] = (
    results_df[
        "BLOCK1C_PREVENTION"
    ]
    != results_df[
        "BLOCK1A_PREVENTION"
    ]
)


# =============================================================================
# 23. CORRECT TOTAL ALIGNMENT-IMPROVEMENT CALCULATION
#
# This calculation uses the entire overlapping Block 1 and Block 1a set before
# limiting examples for display.
# =============================================================================

block1_alignment_df = block1_df[
    [
        "SOURCE_RECORD_ID",
        "FACTOR_PREVENTION_ALIGNED",
    ]
].rename(
    columns={
        "FACTOR_PREVENTION_ALIGNED":
            "BLOCK1_ALIGNMENT",
    }
)


block1a_alignment_df = block1a_df[
    [
        "SOURCE_RECORD_ID",
        "FACTOR_PREVENTION_ALIGNED",
    ]
].rename(
    columns={
        "FACTOR_PREVENTION_ALIGNED":
            "BLOCK1A_ALIGNMENT",
    }
)


full_alignment_comparison = (
    block1_alignment_df.merge(
        block1a_alignment_df,
        on="SOURCE_RECORD_ID",
        how="inner",
    )
)


full_alignment_improvement_mask = (
    full_alignment_comparison[
        "BLOCK1_ALIGNMENT"
    ]
    .eq(False)
    &
    full_alignment_comparison[
        "BLOCK1A_ALIGNMENT"
    ]
    .eq(True)
)


TOTAL_ALIGNMENT_FAILURES_CORRECTED = int(
    full_alignment_improvement_mask.sum()
)


# =============================================================================
# 24. SUMMARY METRICS
# =============================================================================

def rate(
    dataframe,
    column,
    expected=True,
):

    if (
        column not in dataframe.columns
        or dataframe.empty
    ):

        return np.nan

    return float(
        dataframe[
            column
        ]
        .eq(
            expected
        )
        .mean()
    )


block1a_subset_success_rate = float(
    results_df[
        "BLOCK1A_STATUS"
    ]
    .eq(
        "SUCCESS"
    )
    .mean()
)


block1c_success_rate = float(
    results_df[
        "PROCESSING_STATUS"
    ]
    .eq(
        "SUCCESS"
    )
    .mean()
)


block1a_subset_failure_count = int(
    results_df[
        "BLOCK1A_STATUS"
    ]
    .ne(
        "SUCCESS"
    )
    .sum()
)


block1c_failure_count = int(
    results_df[
        "PROCESSING_STATUS"
    ]
    .ne(
        "SUCCESS"
    )
    .sum()
)


recovered_failure_count = int(
    (
        results_df[
            "BLOCK1A_STATUS"
        ]
        .ne(
            "SUCCESS"
        )
        &
        results_df[
            "PROCESSING_STATUS"
        ]
        .eq(
            "SUCCESS"
        )
    ).sum()
)


block1a_struck_rate = float(
    results_df[
        "BLOCK1A_CAUSE"
    ]
    .eq(
        "STRUCK_BY_MOVING_OBJECT"
    )
    .mean()
)


block1c_struck_rate = float(
    results_df[
        "BLOCK1C_CAUSE"
    ]
    .eq(
        "STRUCK_BY_MOVING_OBJECT"
    )
    .mean()
)


block1c_caught_rate = float(
    results_df[
        "BLOCK1C_CAUSE"
    ]
    .eq(
        "CAUGHT_IN_OR_BETWEEN"
    )
    .mean()
)


block1c_loss_rate = float(
    results_df[
        "BLOCK1C_CAUSE"
    ]
    .eq(
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT"
    )
    .mean()
)


block1c_unexpected_rate = float(
    results_df[
        "BLOCK1C_CAUSE"
    ]
    .eq(
        "UNEXPECTED_EQUIPMENT_MOVEMENT"
    )
    .mean()
)


summary_metrics = pd.DataFrame(
    [
        {
            "Measure":
                "Focused test records",

            "Value":
                len(
                    results_df
                ),
        },
        {
            "Measure":
                "Block 1a failures in focused test",

            "Value":
                block1a_subset_failure_count,
        },
        {
            "Measure":
                "Block 1c failures",

            "Value":
                block1c_failure_count,
        },
        {
            "Measure":
                "Block 1a failures recovered by Block 1c",

            "Value":
                recovered_failure_count,
        },
        {
            "Measure":
                "Block 1a subset success rate",

            "Value":
                block1a_subset_success_rate,
        },
        {
            "Measure":
                "Block 1c success rate",

            "Value":
                block1c_success_rate,
        },
        {
            "Measure":
                "Block 1c schema-valid rate",

            "Value":
                rate(
                    results_df,
                    "SCHEMA_VALID",
                ),
        },
        {
            "Measure":
                "Block 1c cause-evidence-valid rate",

            "Value":
                rate(
                    results_df,
                    "CAUSE_EVIDENCE_VALID",
                ),
        },
        {
            "Measure":
                "Block 1c factor-evidence-valid rate",

            "Value":
                rate(
                    results_df,
                    "FACTOR_EVIDENCE_VALID",
                ),
        },
        {
            "Measure":
                "Block 1c alignment rate",

            "Value":
                rate(
                    results_df,
                    "FACTOR_PREVENTION_ALIGNED",
                ),
        },
        {
            "Measure":
                "Block 1c critical-review rate",

            "Value":
                rate(
                    results_df,
                    "CRITICAL_REVIEW_REQUIRED",
                ),
        },
        {
            "Measure":
                "Block 1c advisory-review rate",

            "Value":
                rate(
                    results_df,
                    "ADVISORY_REVIEW",
                ),
        },
        {
            "Measure":
                "Block 1c automatic-repair rate",

            "Value":
                rate(
                    results_df,
                    "AUTOMATIC_REPAIR_APPLIED",
                ),
        },
        {
            "Measure":
                "Cause labels changed from Block 1a",

            "Value":
                int(
                    results_df[
                        "CAUSE_CHANGED_FROM_BLOCK1A"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor labels changed from Block 1a",

            "Value":
                int(
                    results_df[
                        "FACTOR_CHANGED_FROM_BLOCK1A"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Prevention labels changed from Block 1a",

            "Value":
                int(
                    results_df[
                        "PREVENTION_CHANGED_FROM_BLOCK1A"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Total Block 1 alignment failures corrected in Block 1a",

            "Value":
                TOTAL_ALIGNMENT_FAILURES_CORRECTED,
        },
    ]
)


# =============================================================================
# 25. CAUSE TRANSITION MATRIX
# =============================================================================

cause_transition = pd.crosstab(
    results_df[
        "BLOCK1A_CAUSE"
    ],

    results_df[
        "BLOCK1C_CAUSE"
    ],
)


key_cause_order = [
    "STRUCK_BY_MOVING_OBJECT",
    "CAUGHT_IN_OR_BETWEEN",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
]


key_cause_comparison_records = []


for cause_name in key_cause_order:

    key_cause_comparison_records.append(
        {
            "Cause":
                cause_name,

            "Block 1a count in focused sample":
                int(
                    results_df[
                        "BLOCK1A_CAUSE"
                    ]
                    .eq(
                        cause_name
                    )
                    .sum()
                ),

            "Block 1c count in focused sample":
                int(
                    results_df[
                        "BLOCK1C_CAUSE"
                    ]
                    .eq(
                        cause_name
                    )
                    .sum()
                ),
        }
    )


key_cause_comparison = pd.DataFrame(
    key_cause_comparison_records
)


# =============================================================================
# 26. REVIEW REASON SUMMARIES
# =============================================================================

def explode_reason_counts(
    dataframe,
    column,
    output_name,
):

    reason_series = (
        dataframe[
            column
        ]
        .fillna("")
        .str.split(
            "; "
        )
        .explode()
    )

    return (
        reason_series[
            reason_series
            .str.len()
            > 0
        ]
        .value_counts()
        .rename_axis(
            output_name
        )
        .reset_index(
            name="Count"
        )
    )


critical_reason_summary = explode_reason_counts(
    dataframe=results_df,
    column="CRITICAL_REVIEW_REASONS",
    output_name="Critical review reason",
)


advisory_reason_summary = explode_reason_counts(
    dataframe=results_df,
    column="ADVISORY_REVIEW_REASONS",
    output_name="Advisory review reason",
)


# =============================================================================
# 27. ACCEPTANCE CRITERIA
#
# These are project decision criteria. They are not model claims.
# =============================================================================

functional_pass_rate = (
    float(
        functional_test_df[
            "Status"
        ]
        .eq(
            "PASS"
        )
        .mean()
    )
    if not functional_test_df.empty
    else np.nan
)


acceptance_records = [
    {
        "Acceptance check":
            "Schema-valid rate at least 98%",

        "Observed":
            rate(
                results_df,
                "SCHEMA_VALID",
            ),

        "Pass":
            (
                rate(
                    results_df,
                    "SCHEMA_VALID",
                )
                >= 0.98
            ),
    },
    {
        "Acceptance check":
            "Processing success rate at least 98%",

        "Observed":
            block1c_success_rate,

        "Pass":
            (
                block1c_success_rate
                >= 0.98
            ),
    },
    {
        "Acceptance check":
            "Recover at least 80% of tested Block 1a failures",

        "Observed":
            (
                recovered_failure_count
                / block1a_subset_failure_count
                if block1a_subset_failure_count
                else 1.0
            ),

        "Pass":
            (
                (
                    recovered_failure_count
                    / block1a_subset_failure_count
                    if block1a_subset_failure_count
                    else 1.0
                )
                >= 0.80
            ),
    },
    {
        "Acceptance check":
            "Factor-prevention alignment at least 85%",

        "Observed":
            rate(
                results_df,
                "FACTOR_PREVENTION_ALIGNED",
            ),

        "Pass":
            (
                rate(
                    results_df,
                    "FACTOR_PREVENTION_ALIGNED",
                )
                >= 0.85
            ),
    },
    {
        "Acceptance check":
            "Cause evidence valid at least 90%",

        "Observed":
            rate(
                results_df,
                "CAUSE_EVIDENCE_VALID",
            ),

        "Pass":
            (
                rate(
                    results_df,
                    "CAUSE_EVIDENCE_VALID",
                )
                >= 0.90
            ),
    },
    {
        "Acceptance check":
            "Functional cause hierarchy tests at least 80%",

        "Observed":
            functional_pass_rate,

        "Pass":
            (
                functional_pass_rate
                >= 0.80
            ),
    },
]


acceptance_df = pd.DataFrame(
    acceptance_records
)


OVERALL_ACCEPTANCE = bool(
    acceptance_df[
        "Pass"
    ].all()
)


# =============================================================================
# 28. SELECT EXAMPLES FOR THE PDF
# =============================================================================

cause_change_examples = (
    results_df[
        results_df[
            "CAUSE_CHANGED_FROM_BLOCK1A"
        ]
    ]
    .copy()
    .head(18)
)


recovered_failure_examples = (
    results_df[
        (
            results_df[
                "BLOCK1A_STATUS"
            ]
            .ne(
                "SUCCESS"
            )
        )
        &
        (
            results_df[
                "PROCESSING_STATUS"
            ]
            .eq(
                "SUCCESS"
            )
        )
    ]
    .copy()
    .head(15)
)


critical_review_examples = (
    results_df[
        results_df[
            "CRITICAL_REVIEW_REQUIRED"
        ]
    ]
    .copy()
    .head(15)
)


# =============================================================================
# 29. PDF HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append("")

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(
                        line
                    ),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]

        for start in range(
            0,
            len(
                wrapped
            ),
            lines_per_page,
        )
    ] or [[""]]

    for chunk_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(
            chunks
        ) > 1:

            page_title += (
                f" - {chunk_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(
            page_title
        )

        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=25,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were produced."
            ],
        )

        return

    printable = dataframe.copy()

    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[
                column
            ] = (
                printable[
                    column
                ]
                .map(
                    lambda value:
                        (
                            ""
                            if pd.isna(
                                value
                            )
                            else f"{value:.4f}"
                        )
                )
            )

    printable.columns = [
        str(
            column
        )
        .replace(
            "_",
            " ",
        )
        .title()

        for column
        in printable.columns
    ]

    for column in printable.columns:

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    ""
                    if pd.isna(
                        value
                    )
                    else str(
                        value
                    )
            )
        )

        printable[
            column
        ] = (
            printable[
                column
            ]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=(
                                column_wrap_width
                            ),
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(
            printable
        ),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )

        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )

        table.auto_set_font_size(
            False
        )

        table.set_fontsize(
            font_size
        )

        table.scale(
            1,
            1.42,
        )

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(
            figure
        )


def add_heatmap(
    title,
    dataframe,
    x_label,
    y_label,
):

    if dataframe.empty:

        return

    figure = plt.figure(
        figsize=(11, 8)
    )

    axis = figure.add_subplot(
        111
    )

    matrix = dataframe.values.astype(
        float
    )

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(
            len(
                dataframe.columns
            )
        )
    )

    axis.set_xticklabels(
        [
            str(
                value
            )
            .replace(
                "_",
                " ",
            )
            .title()

            for value in (
                dataframe.columns
            )
        ],
        rotation=35,
        ha="right",
        fontsize=7,
    )

    axis.set_yticks(
        range(
            len(
                dataframe.index
            )
        )
    )

    axis.set_yticklabels(
        [
            str(
                value
            )
            .replace(
                "_",
                " ",
            )
            .title()

            for value in (
                dataframe.index
            )
        ],
        fontsize=7,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(
        x_label
    )

    axis.set_ylabel(
        y_label
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value > 0:

                axis.text(
                    column_number,
                    row_number,
                    str(
                        value
                    ),
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )

    figure.tight_layout()

    save_page(
        figure
    )


def add_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=12,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )

        return

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        row_index,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 82,
                (
                    f"Example {example_number} | "
                    f"Record {row['SOURCE_RECORD_ID']} | "
                    f"{row.get('SEVERITY_GROUP', '')}"
                ),
                (
                    f"Block 1a cause: "
                    f"{row.get('BLOCK1A_CAUSE', '')}"
                ),
                (
                    f"Block 1c cause: "
                    f"{row.get('BLOCK1C_CAUSE', '')}"
                ),
                (
                    f"Block 1a factor: "
                    f"{row.get('BLOCK1A_FACTOR', '')}"
                ),
                (
                    f"Block 1c factor: "
                    f"{row.get('BLOCK1C_FACTOR', '')}"
                ),
                (
                    f"Block 1a prevention: "
                    f"{row.get('BLOCK1A_PREVENTION', '')}"
                ),
                (
                    f"Block 1c prevention: "
                    f"{row.get('BLOCK1C_PREVENTION', '')}"
                ),
                (
                    f"Block 1c status: "
                    f"{row.get('PROCESSING_STATUS', '')}"
                ),
                (
                    f"Critical review: "
                    f"{row.get('CRITICAL_REVIEW_REASONS', '')}"
                ),
                (
                    f"Advisory review: "
                    f"{row.get('ADVISORY_REVIEW_REASONS', '')}"
                ),
                "",
                "Narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "Block 1c cause evidence: "
                    f"{row.get('BLOCK1C_CAUSE_EVIDENCE', '')}"
                ),
                (
                    "Block 1c factor evidence: "
                    f"{row.get('BLOCK1C_FACTOR_EVIDENCE', '')}"
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.7,
        lines_per_page=49,
        wrap_width=104,
    )


# =============================================================================
# 30. GENERATE PDF REPORT
# =============================================================================

# Cover page.
figure = create_page()

figure.text(
    0.5,
    0.79,
    "MSHA Block 1c Focused Acceptance Test",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.73,
    "Final Targeted Corrections Before Block 2",
    ha="center",
    fontsize=13,
    color="#444444",
)

figure.text(
    0.5,
    0.685,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Model: {MODEL_ID}",
    f"Prompt version: {PROMPT_VERSION}",
    f"Focused narratives tested: {len(results_df):,}",
    f"Block 1a failures in focused sample: "
    f"{block1a_subset_failure_count}",
    f"Block 1c failures: {block1c_failure_count}",
    f"Block 1a failures recovered: {recovered_failure_count}",
    f"Block 1c success rate: {block1c_success_rate:.1%}",
    f"Block 1c alignment rate: "
    f"{rate(results_df, 'FACTOR_PREVENTION_ALIGNED'):.1%}",
    f"Block 1c critical-review rate: "
    f"{rate(results_df, 'CRITICAL_REVIEW_REQUIRED'):.1%}",
    "",
    "Overall focused acceptance:",
    (
        "PASS - ready to freeze for Block 2"
        if OVERALL_ACCEPTANCE
        else
        "REVIEW - at least one acceptance criterion was not met"
    ),
]

figure.text(
    0.11,
    0.57,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=9.1,
    family="monospace",
    linespacing=1.24,
)

save_page(
    figure
)


# Executive summary.
add_text_pages(
    "Executive Summary",
    [
        "Block 1c is a focused acceptance test. It does not rerun the complete",
        "1,000-record development sample.",
        "",
        f"The test evaluated {len(results_df):,} targeted narratives, including",
        "Block 1a validation failures and records assigned to struck-by,",
        "caught-between, loss-of-control, unexpected-equipment-movement,",
        "manual-handling, and no-separate-factor categories.",
        "",
        "The revised cause hierarchy gives caught-between, vehicle, ground",
        "failure, stored-energy, loss-of-control, and unexpected-equipment",
        "movement priority over the broader struck-by category when their",
        "definitions are directly supported.",
        "",
        "The prompt was shortened to reduce response-formatting failures. JSON",
        "repair now handles markdown fences, trailing commas, missing closing",
        "braces, aliases, missing fields, and minor Python-style constants.",
        "",
        "Critical review flags now identify results that cannot safely enter the",
        "main statistical analysis without review. Advisory flags identify",
        "uncertainty or abstention that may still be appropriate.",
        "",
        f"Block 1a focused-sample success rate: "
        f"{block1a_subset_success_rate:.1%}",
        f"Block 1c focused-sample success rate: "
        f"{block1c_success_rate:.1%}",
        f"Validation failures recovered: {recovered_failure_count} of "
        f"{block1a_subset_failure_count}",
        "",
        (
            "The focused acceptance checks passed. The prompt and ontology can "
            "be frozen for Block 2."
            if OVERALL_ACCEPTANCE
            else
            "At least one focused acceptance check requires review before the "
            "prompt is frozen for Block 2."
        ),
    ],
    font_size=8.7,
)


# Test design.
add_text_pages(
    "Focused Test Design",
    [
        "The test sample intentionally emphasizes the issues that remained after",
        "Block 1a. It is not designed to estimate the natural frequency of",
        "causes, factors, or prevention themes.",
        "",
        "Targeted groups:",
        "",
        "1. Block 1a response-validation failures.",
        "",
        "2. STRUCK_BY_MOVING_OBJECT cases.",
        "",
        "3. CAUGHT_IN_OR_BETWEEN cases.",
        "",
        "4. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT cases.",
        "",
        "5. UNEXPECTED_EQUIPMENT_MOVEMENT cases.",
        "",
        "6. MANUAL_HANDLING_OR_OVEREXERTION cases.",
        "",
        "7. NO_SEPARATE_FACTOR_STATED cases.",
        "",
        "8. Random control records used to bring the unique focused sample to",
        f"   approximately {TARGET_TEST_SIZE} records.",
        "",
        "Overlapping targeted groups were deduplicated by SOURCE_RECORD_ID.",
    ],
    font_size=8.7,
)


add_table_pages(
    "Focused Sample Selection",
    selection_summary,
    rows_per_page=15,
    font_size=7.2,
    column_wrap_width=36,
)


# Functional tests.
add_table_pages(
    "Cause-Hierarchy Functional Tests",
    functional_test_df,
    rows_per_page=10,
    font_size=7.2,
    column_wrap_width=36,
)


# Acceptance criteria.
acceptance_display = acceptance_df.copy()

acceptance_display[
    "Pass"
] = (
    acceptance_display[
        "Pass"
    ]
    .map(
        {
            True:
                "PASS",

            False:
                "REVIEW",
        }
    )
)


add_table_pages(
    "Block 1c Acceptance Criteria",
    acceptance_display,
    rows_per_page=12,
    font_size=7.2,
    column_wrap_width=42,
)


# Summary metrics.
add_table_pages(
    "Focused Acceptance-Test Metrics",
    summary_metrics,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=45,
)


# Success comparison chart.
figure = plt.figure(
    figsize=(9, 6.5)
)

axis = figure.add_subplot(
    111
)

chart_labels = [
    "Block 1a\nfocused subset",
    "Block 1c",
]

chart_values = [
    block1a_subset_success_rate * 100,
    block1c_success_rate * 100,
]

bars = axis.bar(
    chart_labels,
    chart_values,
    color=[
        "#9AA4AE",
        "#2E6F9E",
    ],
)

axis.set_title(
    "Processing Success Rate",
    fontsize=13,
    fontweight="bold",
)

axis.set_ylabel(
    "Successful classifications (%)"
)

axis.set_ylim(
    0,
    105,
)

axis.grid(
    axis="y",
    alpha=0.25,
)

for bar, value in zip(
    bars,
    chart_values,
):

    axis.text(
        bar.get_x()
        + bar.get_width() / 2,
        value,
        f"{value:.1f}%",
        ha="center",
        va="bottom",
        fontsize=10,
    )

figure.tight_layout()

save_page(
    figure
)


# Critical/advisory review chart.
review_chart_df = pd.DataFrame(
    {
        "Review type": [
            "Critical review",
            "Advisory review",
        ],

        "Rate": [
            rate(
                results_df,
                "CRITICAL_REVIEW_REQUIRED",
            )
            * 100,

            rate(
                results_df,
                "ADVISORY_REVIEW",
            )
            * 100,
        ],
    }
)


figure = plt.figure(
    figsize=(9, 6.5)
)

axis = figure.add_subplot(
    111
)

bars = axis.bar(
    review_chart_df[
        "Review type"
    ],
    review_chart_df[
        "Rate"
    ],
    color=[
        "#C44E52",
        "#D9A441",
    ],
)

axis.set_title(
    "Critical and Advisory Review Rates",
    fontsize=13,
    fontweight="bold",
)

axis.set_ylabel(
    "Focused records (%)"
)

axis.set_ylim(
    0,
    105,
)

axis.grid(
    axis="y",
    alpha=0.25,
)

for bar, value in zip(
    bars,
    review_chart_df[
        "Rate"
    ],
):

    axis.text(
        bar.get_x()
        + bar.get_width() / 2,
        value,
        f"{value:.1f}%",
        ha="center",
        va="bottom",
        fontsize=10,
    )

figure.tight_layout()

save_page(
    figure
)


# Cause comparison.
add_table_pages(
    "Key Cause Counts in the Focused Sample",
    key_cause_comparison,
    rows_per_page=12,
    font_size=7.2,
    column_wrap_width=38,
)


figure = plt.figure(
    figsize=(11, 7)
)

axis = figure.add_subplot(
    111
)

positions = np.arange(
    len(
        key_cause_comparison
    )
)

width = 0.36

axis.barh(
    positions - width / 2,
    key_cause_comparison[
        "Block 1a count in focused sample"
    ],
    height=width,
    label="Block 1a",
    color="#9AA4AE",
)

axis.barh(
    positions + width / 2,
    key_cause_comparison[
        "Block 1c count in focused sample"
    ],
    height=width,
    label="Block 1c",
    color="#2E6F9E",
)

axis.set_yticks(
    positions
)

axis.set_yticklabels(
    [
        value.replace(
            "_",
            " ",
        ).title()

        for value in (
            key_cause_comparison[
                "Cause"
            ]
        )
    ]
)

axis.set_xlabel(
    "Narrative count"
)

axis.set_title(
    "Cause Assignments Before and After the Decision Hierarchy",
    fontsize=13,
    fontweight="bold",
)

axis.grid(
    axis="x",
    alpha=0.25,
)

axis.legend()

figure.tight_layout()

save_page(
    figure
)


# Cause transition matrix.
add_heatmap(
    "Block 1a to Block 1c Cause Transitions",
    cause_transition,
    "Block 1c cause",
    "Block 1a cause",
)


# Review reason pages.
add_table_pages(
    "Critical Review Reasons",
    critical_reason_summary,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=50,
)


add_table_pages(
    "Advisory Review Reasons",
    advisory_reason_summary,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=50,
)


# Recovered failures.
add_example_pages(
    "Recovered Block 1a Validation Failures",
    recovered_failure_examples,
    (
        "These narratives failed Block 1a response validation but produced "
        "schema-valid Block 1c outputs after prompt shortening and improved "
        "JSON repair."
    ),
    maximum_records=12,
)


# Cause changes.
add_example_pages(
    "Cause Labels Changed by the Decision Hierarchy",
    cause_change_examples,
    (
        "These records changed cause labels after the specific mechanism "
        "hierarchy was added. Review whether caught-between, loss-of-control, "
        "unexpected movement, or struck-by is now the better description."
    ),
    maximum_records=15,
)


# Remaining critical reviews.
add_example_pages(
    "Remaining Critical Review Cases",
    critical_review_examples,
    (
        "These focused records still contain a validation, evidence, routing, "
        "or factor-prevention issue that would require review before inclusion "
        "in final statistical analysis."
    ),
    maximum_records=12,
)


# Final recommendation page.
add_text_pages(
    "Block 1c Decision for Block 2",
    [
        (
            "OVERALL RESULT: PASS"
            if OVERALL_ACCEPTANCE
            else
            "OVERALL RESULT: REVIEW REQUIRED"
        ),
        "",
        "The focused test verifies the final corrections without rerunning the",
        "complete 1,000-record development sample.",
        "",
        "The system is ready to freeze for Block 2 when:",
        "",
        "1. Response and schema-valid rates meet the acceptance thresholds.",
        "",
        "2. Most tested Block 1a failures are recovered.",
        "",
        "3. The cause hierarchy functional tests distinguish caught-between,",
        "   struck-by, loss-of-control, unexpected movement, and manual",
        "   handling.",
        "",
        "4. Factor-prevention alignment remains strong.",
        "",
        "5. Cause evidence validity remains strong.",
        "",
        "6. The remaining critical-review examples represent genuinely",
        "   ambiguous or out-of-scope narratives rather than repeated prompt",
        "   errors.",
        "",
        "Advisory flags should not automatically exclude a result. For example,",
        "NO_SEPARATE_FACTOR_STATED may be the correct conservative response",
        "when the narrative explains what happened but not why.",
        "",
        "Block 2 should use this Block 1c ontology, concise prompt, cause",
        "hierarchy, JSON repair, factor-prevention mapping, record routing, and",
        "critical-versus-advisory review structure without further tuning on",
        "the unseen Block 2 narratives.",
    ],
    font_size=8.7,
)


# Technical appendix.
add_text_pages(
    "Technical Appendix",
    [
        f"Model: {MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"GPU: {GPU_NAME}",
        f"Block 1 records loaded: {len(block1_df):,}",
        f"Block 1a records loaded: {len(block1a_df):,}",
        f"Focused Block 1c records: {len(results_df):,}",
        f"Maximum retries: {MAX_RETRIES}",
        f"Maximum new tokens: {MAX_NEW_TOKENS}",
        f"Low-confidence threshold: {LOW_CONFIDENCE_THRESHOLD:.2f}",
        f"Random state: {RANDOM_STATE}",
        "",
        "Corrected reporting calculation:",
        (
            "Total Block 1 alignment failures corrected in Block 1a: "
            f"{TOTAL_ALIGNMENT_FAILURES_CORRECTED}"
        ),
        "",
        f"Output PDF: {os.path.abspath(OUTPUT_PDF)}",
    ],
    font_size=8.4,
)


# =============================================================================
# 31. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata[
    "Title"
] = (
    "MSHA Block 1c Focused Acceptance Test Report"
)

metadata[
    "Author"
] = (
    "Python and Qwen2.5-7B-Instruct"
)

metadata[
    "Subject"
] = (
    "Focused acceptance testing of the final MSHA cause, contributing-factor, "
    "and prevention-theme extraction system before unseen Block 2 analysis"
)

metadata[
    "Keywords"
] = (
    "MSHA, NLP, Qwen, ontology, cause hierarchy, validation, "
    "contributing factors, prevention themes"
)

metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 32. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 1C FOCUSED ACCEPTANCE TEST COMPLETE")
print("=" * 95)

print(
    f"Focused narratives tested: "
    f"{len(results_df):,}"
)

print(
    f"Block 1a failures in focused sample: "
    f"{block1a_subset_failure_count:,}"
)

print(
    f"Block 1c failures: "
    f"{block1c_failure_count:,}"
)

print(
    f"Block 1a failures recovered: "
    f"{recovered_failure_count:,}"
)

print(
    f"Block 1c success rate: "
    f"{block1c_success_rate:.1%}"
)

print(
    f"Block 1c schema-valid rate: "
    f"{rate(results_df, 'SCHEMA_VALID'):.1%}"
)

print(
    f"Block 1c alignment rate: "
    f"{rate(results_df, 'FACTOR_PREVENTION_ALIGNED'):.1%}"
)

print(
    f"Block 1c critical-review rate: "
    f"{rate(results_df, 'CRITICAL_REVIEW_REQUIRED'):.1%}"
)

print(
    f"Block 1c advisory-review rate: "
    f"{rate(results_df, 'ADVISORY_REVIEW'):.1%}"
)

print(
    f"Total Block 1 alignment failures corrected in Block 1a: "
    f"{TOTAL_ALIGNMENT_FAILURES_CORRECTED:,}"
)

print(
    "Overall acceptance:",
    (
        "PASS"
        if OVERALL_ACCEPTANCE
        else "REVIEW REQUIRED"
    ),
)

print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print("=" * 95)


# =============================================================================
# 33. CONCISE NOTEBOOK DISPLAYS
# =============================================================================

display(
    selection_summary
)


display(
    functional_test_df
)


display(
    acceptance_df
)


display(
    summary_metrics
)


display(
    key_cause_comparison
)


display(
    critical_reason_summary
)


display(
    advisory_reason_summary
)
````

````python
# =============================================================================
# BLOCK 2 - STANDALONE VERSION
# MSHA ANALYSIS OF 2,000 UNSEEN ACCIDENT NARRATIVES
#
# PURPOSE
# -------
# Apply the frozen Block 1c ontology and model architecture to 2,000 unseen
# MSHA narratives and analyze:
#
#   1. Reported primary causes.
#   2. Supported contributing factors.
#   3. Preliminary prevention themes.
#   4. Differences across five injury-severity classes.
#   5. Changes across early, middle, and recent time periods.
#   6. Cause-factor and factor-prevention relationships.
#   7. Processing quality, abstention, review flags, and sensitivity.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. The Block 1 development-ID file is in the working directory:
#
#       MSHA_Block1_Development_Record_IDs.json
#
# OUTPUTS
# -------
# 1. MSHA_Block2_Unseen_Narrative_Analysis_Report.pdf
# 2. MSHA_Block2_Unseen_Results.csv
# 3. MSHA_Block2_Unseen_Checkpoint.jsonl
# 4. MSHA_Block2_Unseen_Record_IDs.json
#
# IMPORTANT
# ---------
# This code is fully standalone after `accidents` is loaded. Block 1c does
# not need to be run first.
#
# The sample is balanced across five severity groups and three broad time
# periods. Sample percentages are suitable for comparisons within the
# analytical sample but are not estimates of natural MSHA prevalence.
#
# Temporal results represent changes in the share of analyzed narratives,
# not exposure-adjusted accident or injury rates.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import time
import math
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import pydantic
import pydantic_core
import transformers
import accelerate
import bitsandbytes

from pydantic import BaseModel, Field, model_validator

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from matplotlib.backends.backend_pdf import PdfPages
from scipy.stats import chi2_contingency


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

PROMPT_VERSION = "MSHA_BLOCK1C_FROZEN_FOR_BLOCK2_V3.0"

TEXT_COLUMN = "NARRATIVE"

DATE_COLUMN = "ACCIDENT_DT"

SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

TOTAL_SAMPLE_SIZE = 2000

TARGET_PER_SEVERITY = 400

RANDOM_STATE = 84

MIN_NARRATIVE_WORDS = 8

TRUNCATION_CHARACTER_THRESHOLD = 383

MAX_NEW_TOKENS = 750

MAX_RETRIES = 4

LOW_CONFIDENCE_THRESHOLD = 0.70

REPROCESS_COMPLETED = False

TOP_CATEGORIES_TO_PLOT = 12

MIN_CATEGORY_COUNT_FOR_TREND = 10

EXAMPLES_PER_SECTION = 10

BLOCK1_IDS_JSON = (
    "MSHA_Block1_Development_Record_IDs.json"
)

BLOCK2_CHECKPOINT_JSONL = (
    "MSHA_Block2_Unseen_Checkpoint.jsonl"
)

BLOCK2_IDS_JSON = (
    "MSHA_Block2_Unseen_Record_IDs.json"
)

BLOCK2_RESULTS_CSV = (
    "MSHA_Block2_Unseen_Results.csv"
)

OUTPUT_PDF = (
    "MSHA_Block2_Unseen_Narrative_Analysis_Report.pdf"
)


SEVERITY_ORDER = [
    "Fatality",
    "Permanent disability",
    "Days away",
    "Restricted or transferred",
    "No lost time or other",
]


EARLY_PERIOD = "Early: 2000-2008"

MIDDLE_PERIOD = "Middle: 2009-2017"

RECENT_PERIOD = "Recent: 2018-2026"


TIME_PERIOD_ORDER = [
    EARLY_PERIOD,
    MIDDLE_PERIOD,
    RECENT_PERIOD,
]


# =============================================================================
# 3. ENVIRONMENT VALIDATION
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(accidents, pd.DataFrame):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


if not Path(BLOCK1_IDS_JSON).exists():

    raise FileNotFoundError(
        f"Required file was not found: {BLOCK1_IDS_JSON}"
    )


if not torch.cuda.is_available():

    raise RuntimeError(
        "No NVIDIA GPU is visible to PyTorch. Attach the Azure ML "
        "notebook to GPU compute before running Block 2."
    )


GPU_NAME = torch.cuda.get_device_name(0)

GPU_TOTAL_GB = (
    torch.cuda.get_device_properties(0).total_memory
    / 1024**3
)

GPU_FREE_BYTES, GPU_TOTAL_BYTES = torch.cuda.mem_get_info(0)

GPU_FREE_GB = GPU_FREE_BYTES / 1024**3


print("=" * 95)
print("BLOCK 2 - STANDALONE UNSEEN NARRATIVE ANALYSIS")
print("=" * 95)

print("Python:", sys.version.split()[0])
print("PyTorch:", torch.__version__)
print("Transformers:", transformers.__version__)
print("Pydantic:", pydantic.__version__)
print("GPU:", GPU_NAME)
print(f"Total GPU memory: {GPU_TOTAL_GB:.2f} GB")
print(f"Free GPU memory: {GPU_FREE_GB:.2f} GB")


# =============================================================================
# 4. FROZEN RECORD-TYPE ONTOLOGY
# =============================================================================

RECORD_TYPE_CATEGORIES = [
    "ACUTE_ACCIDENT",
    "OCCUPATIONAL_EXPOSURE_OR_ILLNESS",
    "NONWORK_RELATED_OR_NATURAL_EVENT",
    "NONINJURY_REPORT",
    "INSUFFICIENT_EVENT_INFORMATION",
]


# =============================================================================
# 5. FROZEN CAUSE ONTOLOGY
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 6. FROZEN CONTRIBUTING-FACTOR ONTOLOGY
# =============================================================================

FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NO_SEPARATE_FACTOR_STATED",
]


# =============================================================================
# 7. FROZEN PREVENTION-THEME ONTOLOGY
# =============================================================================

PREVENTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "CONTROL_ENVIRONMENTAL_EXPOSURE",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]


# =============================================================================
# 8. FROZEN FACTOR-TO-PREVENTION RELATIONSHIPS
# =============================================================================

PREFERRED_PREVENTION_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":
        "SECURE_OR_REPOSITION_MATERIAL",

    "INADEQUATE_SECUREMENT_OR_RESTRAINT":
        "SECURE_OR_REPOSITION_MATERIAL",

    "LINE_OF_FIRE_BODY_POSITION":
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "UNSTABLE_GROUND_OR_ROOF_CONDITION":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",

    "INADEQUATE_GUARDING_OR_BARRIER":
        "INSTALL_OR_IMPROVE_GUARDING",

    "ENERGY_ISOLATION_DEFICIENCY":
        "ISOLATE_OR_CONTROL_STORED_ENERGY",

    "EQUIPMENT_CONDITION_OR_FAILURE":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "PROCEDURE_OR_WORK_METHOD":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "INADEQUATE_HAZARD_RECOGNITION":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "COMMUNICATION_OR_COORDINATION":
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",

    "TRAINING_OR_EXPERIENCE":
        "PROVIDE_TARGETED_TRAINING",

    "PPE_OR_PROTECTIVE_CONTROL":
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",

    "ENVIRONMENTAL_CONDITION":
        "CONTROL_ENVIRONMENTAL_EXPOSURE",

    "WORKSPACE_OR_ACCESS_LIMITATION":
        "REDESIGN_WORKSPACE_OR_ACCESS",

    "NO_SEPARATE_FACTOR_STATED":
        "INVESTIGATION_REQUIRED",
}


ALLOWED_PREVENTIONS_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_SECUREMENT_OR_RESTRAINT": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    },

    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },

    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "INADEQUATE_GUARDING_OR_BARRIER": {
        "INSTALL_OR_IMPROVE_GUARDING",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "ENERGY_ISOLATION_DEFICIENCY": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    },

    "INADEQUATE_HAZARD_RECOGNITION": {
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "COMMUNICATION_OR_COORDINATION": {
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "TRAINING_OR_EXPERIENCE": {
        "PROVIDE_TARGETED_TRAINING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "PPE_OR_PROTECTIVE_CONTROL": {
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "ENVIRONMENTAL_CONDITION": {
        "CONTROL_ENVIRONMENTAL_EXPOSURE",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "WORKSPACE_OR_ACCESS_LIMITATION": {
        "REDESIGN_WORKSPACE_OR_ACCESS",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ENGINEERING_REVIEW_REQUIRED",
    },

    "OTHER": {
        "ENGINEERING_REVIEW_REQUIRED",
        "INVESTIGATION_REQUIRED",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },

    "NO_SEPARATE_FACTOR_STATED": {
        "INVESTIGATION_REQUIRED",
        "INSUFFICIENT_INFORMATION",
    },
}


# =============================================================================
# 9. CATEGORY ALIASES
# =============================================================================

CATEGORY_ALIASES = {
    "UNSTATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NONE_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "NOT_PROVIDED":
        "NO_SEPARATE_FACTOR_STATED",

    "NO_FACTOR_STATED":
        "NO_SEPARATE_FACTOR_STATED",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "LOSS_OF_CONTROL":
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",

    "UNEXPECTED_MACHINE_MOVEMENT":
        "UNEXPECTED_EQUIPMENT_MOVEMENT",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "REVISE_PROCEDURE":
        "REVISE_PROCEDURE_OR_JOB_PLAN",

    "PRE_JOB_HAZARD_REVIEW":
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",

    "EQUIPMENT_REPAIR_OR_INSPECTION":
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",

    "IMPROVE_HOUSEKEEPING":
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",

    "GROUND_CONTROL_REVIEW":
        "IMPROVE_GROUND_CONTROL_OR_SCALING",

    "UNKNOWN":
        "INSUFFICIENT_INFORMATION",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",
}


SUPPORT_LEVEL_ALIASES = {
    "HIGH": "HIGH",
    "DIRECT": "HIGH",
    "EXPLICIT": "HIGH",
    "MEDIUM": "MEDIUM",
    "MODERATE": "MEDIUM",
    "IMPLIED": "MEDIUM",
    "LOW": "LOW",
    "WEAK": "LOW",
    "UNCERTAIN": "LOW",
    "NONE": "LOW",
    "": "LOW",
}


# =============================================================================
# 10. FROZEN RESPONSE SCHEMA
# =============================================================================

class IncidentExtraction(BaseModel):

    record_type: str

    primary_cause: str

    primary_contributing_factor: str

    prevention_theme: str

    cause_evidence_text: str = ""

    factor_evidence_text: str = ""

    prevention_rationale: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    factor_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    prevention_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    factor_support_level: str = "LOW"

    prevention_support_level: str = "LOW"

    insufficient_information_reason: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_output(self):

        if self.record_type not in RECORD_TYPE_CATEGORIES:

            raise ValueError(
                f"Invalid record_type: {self.record_type}"
            )

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                f"Invalid primary_cause: {self.primary_cause}"
            )

        if (
            self.primary_contributing_factor
            not in FACTOR_CATEGORIES
        ):

            raise ValueError(
                "Invalid primary_contributing_factor: "
                f"{self.primary_contributing_factor}"
            )

        if self.prevention_theme not in PREVENTION_CATEGORIES:

            raise ValueError(
                f"Invalid prevention_theme: {self.prevention_theme}"
            )

        for field_name in [
            "cause_support_level",
            "factor_support_level",
            "prevention_support_level",
        ]:

            if getattr(self, field_name) not in {
                "HIGH",
                "MEDIUM",
                "LOW",
            }:

                raise ValueError(
                    f"{field_name} must be HIGH, MEDIUM, or LOW."
                )

        return self


# =============================================================================
# 11. FROZEN CAUSE HIERARCHY
# =============================================================================

CAUSE_DECISION_HIERARCHY = """
CAUSE-SELECTION HIERARCHY

Apply the first category that clearly describes the immediate event:

1. CAUGHT_IN_OR_BETWEEN
   Use when the person or body part is trapped, pinched, crushed, compressed,
   entangled, pulled into, or held between two objects or equipment parts.

2. VEHICLE_OR_MOBILE_EQUIPMENT_EVENT
   Use for collision, overturn, runaway, backing, roadway departure, or
   transportation events involving mobile equipment.

3. GROUND_OR_ROOF_FAILURE
   Use when roof, rib, face, highwall, slate, rock, or ground failure is the
   immediate event.

4. PRESSURE_OR_STORED_ENERGY_RELEASE
   Use when hydraulic, pneumatic, spring, tension, pressure, or other stored
   energy is released.

5. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT
   Use when the event begins because control of a tool, hose, handle, strap,
   load, or equipment component is explicitly lost.

6. UNEXPECTED_EQUIPMENT_MOVEMENT
   Use when equipment or a component moves unexpectedly without a stated
   loss of control by the worker and without an explicit mechanical failure.

7. STRUCK_BY_MOVING_OBJECT
   Use when a moving, falling, flying, rolling, or swinging object contacts
   the person but does not trap or compress the person between objects.

8. STRUCK_AGAINST_STATIONARY_OBJECT
   Use when the person's movement causes contact with a stationary object.

9. FOREIGN_BODY_OR_PARTICLE_CONTACT
   Use when a chip, fragment, dust particle, or small foreign object enters
   or contacts the eye, skin, or body.

10. SLIP_TRIP_OR_FALL
    Use when the person loses footing, balance, or elevation.

11. AWKWARD_MOVEMENT_OR_BODY_MOTION
    Use when an acute twist, reach, bend, turn, or step causes injury without
    a fall or substantial manual load.

12. MANUAL_HANDLING_OR_OVEREXERTION
    Use for acute strains, pain, sprains, and hernias associated with lifting,
    pushing, pulling, carrying, shoveling, holding, or exerting force.

Do not use STRUCK_BY_MOVING_OBJECT as a general fallback when another
specific mechanism is directly supported.
""".strip()


# =============================================================================
# 12. FROZEN CONCISE SYSTEM PROMPT
# =============================================================================

SYSTEM_PROMPT = f"""
You extract structured information from short MSHA accident narratives.

This is reported-event extraction, not verified root-cause analysis.

ALLOWED RECORD TYPES
{", ".join(RECORD_TYPE_CATEGORIES)}

ALLOWED CAUSES
{", ".join(CAUSE_CATEGORIES)}

ALLOWED FACTORS
{", ".join(FACTOR_CATEGORIES)}

ALLOWED PREVENTION THEMES
{", ".join(PREVENTION_CATEGORIES)}

{CAUSE_DECISION_HIERARCHY}

FACTOR RULES

- UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL requires explicit unstable,
  loose, unsupported, rolling, shifting, sliding, falling, or improperly
  positioned material.

- INADEQUATE_SECUREMENT_OR_RESTRAINT requires explicitly missing, removed,
  failed, or inadequate blocking, fastening, support, bracing, restraint,
  chocking, or securement.

- LINE_OF_FIRE_BODY_POSITION applies when a person or body part is placed in
  a pinch point or the path of moving equipment, tools, loads, falling
  objects, or released energy.

- POOR_HOUSEKEEPING_OR_SURFACE_CONDITION requires an explicit walking,
  standing, access, or working-surface hazard such as ice, water, mud,
  grease, debris, loose rock, holes, or uneven footing.

- A hand, hose, tool, component, object, strap, or load slipping is not
  housekeeping.

- PROCEDURE_OR_WORK_METHOD applies when the task method or sequence created
  the exposure and no more specific physical factor is directly supported.

- NO_SEPARATE_FACTOR_STATED applies when the narrative explains what
  happened but does not separately state or directly support why it happened.

- Factor evidence must support the separate factor. Injury-outcome text is
  not factor evidence.

PREVENTION RULES

- Select prevention from the supported contributing factor.
- Do not invent a factor to justify a prevention recommendation.
- NO_SEPARATE_FACTOR_STATED maps to INVESTIGATION_REQUIRED.
- LINE_OF_FIRE_BODY_POSITION normally maps to
  ESTABLISH_LINE_OF_FIRE_EXCLUSION.
- POOR_HOUSEKEEPING_OR_SURFACE_CONDITION normally maps to
  IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING.
- UNSTABLE_GROUND_OR_ROOF_CONDITION normally maps to
  IMPROVE_GROUND_CONTROL_OR_SCALING.
- ENERGY_ISOLATION_DEFICIENCY normally maps to
  ISOLATE_OR_CONTROL_STORED_ENERGY.
- PROCEDURE_OR_WORK_METHOD normally maps to
  REVISE_PROCEDURE_OR_JOB_PLAN.
- ENVIRONMENTAL_CONDITION normally maps to
  CONTROL_ENVIRONMENTAL_EXPOSURE.
- TRAINING_OR_EXPERIENCE requires stated training or proficiency evidence.
- PPE_OR_PROTECTIVE_CONTROL requires stated or directly supported PPE evidence.
- Equipment repair requires stated failure, damage, wear, leakage, defect,
  or malfunction.
- Mechanical handling assistance may be selected for supported manual
  lifting, pushing, pulling, or load-positioning exposure.

RECORD ROUTING

For NONWORK_RELATED_OR_NATURAL_EVENT:
- primary_cause = INSUFFICIENT_INFORMATION
- primary_contributing_factor = NO_SEPARATE_FACTOR_STATED
- prevention_theme = INSUFFICIENT_INFORMATION

EVIDENCE

- Copy short evidence excerpts verbatim.
- Cause evidence must support the immediate event mechanism.
- Factor evidence must support the separate contributing factor.
- If factor = NO_SEPARATE_FACTOR_STATED, factor evidence must be empty.

SUPPORT LEVELS

- HIGH = directly and explicitly supported.
- MEDIUM = strongly implied.
- LOW = incomplete or ambiguous.

Return only one JSON object with exactly these keys:

{{
  "record_type": "",
  "primary_cause": "",
  "primary_contributing_factor": "",
  "prevention_theme": "",
  "cause_evidence_text": "",
  "factor_evidence_text": "",
  "prevention_rationale": "",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "prevention_confidence": 0.0,
  "cause_support_level": "HIGH",
  "factor_support_level": "HIGH",
  "prevention_support_level": "HIGH",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 13. NORMALIZATION AND JSON REPAIR
# =============================================================================

def normalize_token(value):

    return (
        str(value)
        .strip()
        .upper()
        .replace("-", "_")
        .replace(" ", "_")
        .replace("/", "_")
    )


def normalize_category(value):

    normalized = normalize_token(value)

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def normalize_support(value):

    normalized = normalize_token(value)

    return SUPPORT_LEVEL_ALIASES.get(
        normalized,
        "LOW",
    )


def safe_float(value, default=0.0):

    try:

        output = float(value)

        return min(
            max(output, 0.0),
            1.0,
        )

    except Exception:

        return default


def strip_markdown_fences(text):

    text = str(text).strip()

    text = re.sub(
        r"^```(?:json)?",
        "",
        text,
        flags=re.IGNORECASE,
    )

    text = re.sub(
        r"```$",
        "",
        text,
    )

    return text.strip()


def extract_balanced_json(text):

    text = strip_markdown_fences(text)

    start = text.find("{")

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0

    in_string = False

    escaped = False

    for index in range(start, len(text)):

        character = text[index]

        if escaped:

            escaped = False

            continue

        if character == "\\" and in_string:

            escaped = True

            continue

        if character == '"':

            in_string = not in_string

            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    index + 1
                ]

    candidate = text[start:].strip()

    missing_braces = (
        candidate.count("{")
        - candidate.count("}")
    )

    if missing_braces > 0:

        candidate += "}" * missing_braces

        return candidate

    raise ValueError(
        "A complete JSON object was not found."
    )


def repair_common_json_problems(json_text):

    repaired = str(json_text).strip()

    repaired = (
        repaired
        .replace("“", '"')
        .replace("”", '"')
        .replace("’", "'")
    )

    repaired = re.sub(
        r",\s*}",
        "}",
        repaired,
    )

    repaired = re.sub(
        r":\s*None\b",
        ': ""',
        repaired,
    )

    repaired = re.sub(
        r":\s*True\b",
        ": true",
        repaired,
    )

    repaired = re.sub(
        r":\s*False\b",
        ": false",
        repaired,
    )

    return repaired


EXPECTED_KEYS = [
    "record_type",
    "primary_cause",
    "primary_contributing_factor",
    "prevention_theme",
    "cause_evidence_text",
    "factor_evidence_text",
    "prevention_rationale",
    "cause_confidence",
    "factor_confidence",
    "prevention_confidence",
    "cause_support_level",
    "factor_support_level",
    "prevention_support_level",
    "insufficient_information_reason",
    "reviewer_note",
]


DEFAULT_RESPONSE = {
    "record_type":
        "INSUFFICIENT_EVENT_INFORMATION",

    "primary_cause":
        "INSUFFICIENT_INFORMATION",

    "primary_contributing_factor":
        "NO_SEPARATE_FACTOR_STATED",

    "prevention_theme":
        "INVESTIGATION_REQUIRED",

    "cause_evidence_text":
        "",

    "factor_evidence_text":
        "",

    "prevention_rationale":
        "",

    "cause_confidence":
        0.0,

    "factor_confidence":
        0.0,

    "prevention_confidence":
        0.0,

    "cause_support_level":
        "LOW",

    "factor_support_level":
        "LOW",

    "prevention_support_level":
        "LOW",

    "insufficient_information_reason":
        "",

    "reviewer_note":
        "",
}


def normalize_response_dictionary(response_dictionary):

    cleaned = dict(DEFAULT_RESPONSE)

    retained_keys = []

    defaulted_keys = []

    for key in EXPECTED_KEYS:

        if key in response_dictionary:

            cleaned[key] = response_dictionary[key]

            retained_keys.append(key)

        else:

            defaulted_keys.append(key)

    cleaned["record_type"] = normalize_token(
        cleaned["record_type"]
    )

    for field_name in [
        "primary_cause",
        "primary_contributing_factor",
        "prevention_theme",
    ]:

        cleaned[field_name] = normalize_category(
            cleaned[field_name]
        )

    for field_name in [
        "cause_support_level",
        "factor_support_level",
        "prevention_support_level",
    ]:

        cleaned[field_name] = normalize_support(
            cleaned[field_name]
        )

    for field_name in [
        "cause_confidence",
        "factor_confidence",
        "prevention_confidence",
    ]:

        cleaned[field_name] = safe_float(
            cleaned[field_name]
        )

    for field_name in [
        "cause_evidence_text",
        "factor_evidence_text",
        "prevention_rationale",
        "insufficient_information_reason",
        "reviewer_note",
    ]:

        cleaned[field_name] = str(
            cleaned[field_name] or ""
        ).strip()

    return (
        cleaned,
        retained_keys,
        defaulted_keys,
    )


# =============================================================================
# 14. EVIDENCE VALIDATION
# =============================================================================

def normalize_evidence_text(value):

    text = str(value).lower()

    text = (
        text
        .replace("“", '"')
        .replace("”", '"')
        .replace("’", "'")
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def exact_evidence_present(
    evidence,
    narrative,
):

    evidence = str(evidence).strip()

    if not evidence:

        return False

    return (
        evidence.lower()
        in str(narrative).lower()
    )


def normalized_evidence_present(
    evidence,
    narrative,
):

    evidence_normalized = normalize_evidence_text(
        evidence
    )

    narrative_normalized = normalize_evidence_text(
        narrative
    )

    if not evidence_normalized:

        return False

    return (
        evidence_normalized
        in narrative_normalized
    )


def evidence_token_overlap(
    evidence,
    narrative,
):

    evidence_tokens = set(
        normalize_evidence_text(
            evidence
        ).split()
    )

    narrative_tokens = set(
        normalize_evidence_text(
            narrative
        ).split()
    )

    if not evidence_tokens:

        return 0.0

    return (
        len(
            evidence_tokens.intersection(
                narrative_tokens
            )
        )
        / len(evidence_tokens)
    )


def evaluate_evidence(
    evidence,
    narrative,
):

    exact = exact_evidence_present(
        evidence,
        narrative,
    )

    normalized = normalized_evidence_present(
        evidence,
        narrative,
    )

    overlap = evidence_token_overlap(
        evidence,
        narrative,
    )

    valid = (
        exact
        or normalized
        or overlap >= 0.85
    )

    return (
        valid,
        exact,
        normalized,
        overlap,
    )


# =============================================================================
# 15. FACTOR-PREVENTION CHECKS
# =============================================================================

def factor_prevention_aligned(
    factor,
    prevention,
):

    allowed = ALLOWED_PREVENTIONS_BY_FACTOR.get(
        factor
    )

    if allowed is None:

        return True

    return prevention in allowed


def preferred_prevention_used(
    factor,
    prevention,
):

    preferred = PREFERRED_PREVENTION_BY_FACTOR.get(
        factor
    )

    if preferred is None:

        return True

    return prevention == preferred


# =============================================================================
# 16. DETERMINISTIC CONSISTENCY RULES
# =============================================================================

def apply_consistency_rules(
    response_dictionary,
):

    output = dict(response_dictionary)

    repairs = []

    record_type = output["record_type"]

    factor = output[
        "primary_contributing_factor"
    ]

    prevention = output[
        "prevention_theme"
    ]

    if (
        record_type
        == "NONWORK_RELATED_OR_NATURAL_EVENT"
    ):

        output["primary_cause"] = (
            "INSUFFICIENT_INFORMATION"
        )

        output[
            "primary_contributing_factor"
        ] = (
            "NO_SEPARATE_FACTOR_STATED"
        )

        output["prevention_theme"] = (
            "INSUFFICIENT_INFORMATION"
        )

        output["factor_evidence_text"] = ""

        output["cause_support_level"] = "LOW"

        output["factor_support_level"] = "LOW"

        output[
            "prevention_support_level"
        ] = "LOW"

        repairs.append(
            "Applied nonwork/natural-event routing"
        )

    elif (
        factor
        == "NO_SEPARATE_FACTOR_STATED"
    ):

        output["factor_evidence_text"] = ""

        if prevention not in {
            "INVESTIGATION_REQUIRED",
            "INSUFFICIENT_INFORMATION",
        }:

            output["prevention_theme"] = (
                "INVESTIGATION_REQUIRED"
            )

            output[
                "prevention_support_level"
            ] = "LOW"

            output[
                "prevention_confidence"
            ] = min(
                safe_float(
                    output[
                        "prevention_confidence"
                    ]
                ),
                0.60,
            )

            repairs.append(
                "Mapped no separate factor to investigation"
            )

    return output, repairs


# =============================================================================
# 17. LOAD LOCAL QWEN MODEL
# =============================================================================

print("\n" + "=" * 95)
print("LOADING LOCAL QWEN MODEL")
print("=" * 95)


quantization_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.float16,
)


tokenizer = AutoTokenizer.from_pretrained(
    MODEL_ID,
    use_fast=True,
    trust_remote_code=False,
)


if tokenizer.pad_token_id is None:

    tokenizer.pad_token = tokenizer.eos_token


model = AutoModelForCausalLM.from_pretrained(
    MODEL_ID,
    quantization_config=quantization_config,
    device_map="auto",
    dtype=torch.float16,
    low_cpu_mem_usage=True,
    trust_remote_code=False,
)


model.eval()


print("Model loaded successfully.")
print("Model device:", model.device)


# =============================================================================
# 18. MODEL GENERATION
# =============================================================================

def generate_model_response(messages):

    model_inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )

    model_inputs = {
        key:
            value.to(model.device)
        for key, value in model_inputs.items()
    }

    input_length = int(
        model_inputs["input_ids"].shape[-1]
    )

    with torch.inference_mode():

        generated = model.generate(
            **model_inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.02,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )

    generated_ids = generated[
        0,
        input_length:
    ]

    response = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )

    del model_inputs
    del generated
    del generated_ids

    return response


# =============================================================================
# 19. RETRY PROMPT
# =============================================================================

def build_retry_message(
    error,
    prior_response,
):

    error_text = str(error)

    likely_field = "unknown"

    for field_name in EXPECTED_KEYS:

        if field_name in error_text:

            likely_field = field_name

            break

    return f"""
The previous response was not valid.

LIKELY INVALID FIELD
{likely_field}

VALIDATION ERROR
{error_text}

PREVIOUS RESPONSE
{prior_response}

Return one complete JSON object only.

Allowed record types:
{RECORD_TYPE_CATEGORIES}

Allowed causes:
{CAUSE_CATEGORIES}

Allowed factors:
{FACTOR_CATEGORIES}

Allowed prevention themes:
{PREVENTION_CATEGORIES}

Support values must be HIGH, MEDIUM, or LOW.

Do not include markdown or explanation outside the JSON.
""".strip()


# =============================================================================
# 20. CLASSIFY ONE NARRATIVE
# =============================================================================

def classify_narrative(
    narrative,
    preliminary_record_type,
    may_be_truncated,
):

    user_prompt = f"""
PRELIMINARY ROUTING HINT
{preliminary_record_type}

NARRATIVE MAY BE TRUNCATED
{may_be_truncated}

NARRATIVE
{narrative}

Apply the cause hierarchy and return only the required JSON object.
""".strip()

    prior_response = ""

    prior_error = ""

    last_response = ""

    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]

        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        build_retry_message(
                            error=prior_error,
                            prior_response=prior_response,
                        ),
                }
            )

        try:

            raw_response = generate_model_response(
                messages
            )

            last_response = raw_response

            json_text = extract_balanced_json(
                raw_response
            )

            json_text = repair_common_json_problems(
                json_text
            )

            response_dictionary = json.loads(
                json_text
            )

            (
                normalized_dictionary,
                retained_keys,
                defaulted_keys,
            ) = normalize_response_dictionary(
                response_dictionary
            )

            (
                consistent_dictionary,
                consistency_repairs,
            ) = apply_consistency_rules(
                normalized_dictionary
            )

            validated = (
                IncidentExtraction.model_validate(
                    consistent_dictionary
                )
            )

            (
                cause_valid,
                cause_exact,
                cause_normalized,
                cause_overlap,
            ) = evaluate_evidence(
                validated.cause_evidence_text,
                narrative,
            )

            (
                factor_valid,
                factor_exact,
                factor_normalized,
                factor_overlap,
            ) = evaluate_evidence(
                validated.factor_evidence_text,
                narrative,
            )

            if (
                validated
                .primary_contributing_factor
                == "NO_SEPARATE_FACTOR_STATED"
            ):

                factor_valid = True

            aligned = factor_prevention_aligned(
                validated
                .primary_contributing_factor,
                validated
                .prevention_theme,
            )

            preferred = preferred_prevention_used(
                validated
                .primary_contributing_factor,
                validated
                .prevention_theme,
            )

            repair_notes = list(
                consistency_repairs
            )

            if defaulted_keys:

                repair_notes.append(
                    "Defaulted missing fields: "
                    + ", ".join(defaulted_keys)
                )

            return {
                **validated.model_dump(),

                "SCHEMA_VALID":
                    True,

                "FIELDS_RETAINED_FROM_MODEL":
                    "; ".join(retained_keys),

                "FIELDS_DEFAULTED":
                    "; ".join(defaulted_keys),

                "CAUSE_EVIDENCE_EXACT":
                    cause_exact,

                "CAUSE_EVIDENCE_NORMALIZED":
                    cause_normalized,

                "CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    round(cause_overlap, 4),

                "CAUSE_EVIDENCE_VALID":
                    cause_valid,

                "FACTOR_EVIDENCE_EXACT":
                    factor_exact,

                "FACTOR_EVIDENCE_NORMALIZED":
                    factor_normalized,

                "FACTOR_EVIDENCE_TOKEN_OVERLAP":
                    round(factor_overlap, 4),

                "FACTOR_EVIDENCE_VALID":
                    factor_valid,

                "FACTOR_PREVENTION_ALIGNED":
                    aligned,

                "PREFERRED_PREVENTION_USED":
                    preferred,

                "AUTOMATIC_REPAIR_APPLIED":
                    bool(repair_notes),

                "AUTOMATIC_REPAIR_NOTES":
                    "; ".join(repair_notes),

                "PROCESSING_STATUS":
                    "SUCCESS",

                "PROCESSING_ERROR":
                    "",

                "ATTEMPT_COUNT":
                    attempt_number,

                "RAW_MODEL_RESPONSE":
                    raw_response,
            }

        except Exception as error:

            prior_response = last_response

            prior_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )

            print(
                f"  Attempt {attempt_number} failed: "
                f"{prior_error[:300]}"
            )

            gc.collect()

            torch.cuda.empty_cache()

    return {
        **DEFAULT_RESPONSE,

        "SCHEMA_VALID":
            False,

        "FIELDS_RETAINED_FROM_MODEL":
            "",

        "FIELDS_DEFAULTED":
            "; ".join(EXPECTED_KEYS),

        "CAUSE_EVIDENCE_EXACT":
            False,

        "CAUSE_EVIDENCE_NORMALIZED":
            False,

        "CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "CAUSE_EVIDENCE_VALID":
            False,

        "FACTOR_EVIDENCE_EXACT":
            False,

        "FACTOR_EVIDENCE_NORMALIZED":
            False,

        "FACTOR_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "FACTOR_EVIDENCE_VALID":
            False,

        "FACTOR_PREVENTION_ALIGNED":
            False,

        "PREFERRED_PREVENTION_USED":
            False,

        "AUTOMATIC_REPAIR_APPLIED":
            False,

        "AUTOMATIC_REPAIR_NOTES":
            "",

        "PROCESSING_STATUS":
            "FAILED",

        "PROCESSING_ERROR":
            prior_error,

        "ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAW_MODEL_RESPONSE":
            last_response,
    }


# =============================================================================
# 21. UNSUPPORTED-FACTOR SAFEGUARD
# =============================================================================

def apply_unsupported_factor_safeguard(
    model_result,
):

    output = dict(model_result)

    factor = output.get(
        "primary_contributing_factor",
        "",
    )

    factor_valid = bool(
        output.get(
            "FACTOR_EVIDENCE_VALID",
            False,
        )
    )

    repair_notes = [
        value
        for value in str(
            output.get(
                "AUTOMATIC_REPAIR_NOTES",
                "",
            )
        ).split("; ")
        if value
    ]

    safeguard_applied = False

    if (
        factor
        != "NO_SEPARATE_FACTOR_STATED"
        and not factor_valid
    ):

        output[
            "primary_contributing_factor"
        ] = (
            "NO_SEPARATE_FACTOR_STATED"
        )

        output["factor_evidence_text"] = ""

        output["factor_support_level"] = "LOW"

        output["factor_confidence"] = min(
            safe_float(
                output.get(
                    "factor_confidence",
                    0.0,
                )
            ),
            0.60,
        )

        output["prevention_theme"] = (
            "INVESTIGATION_REQUIRED"
        )

        output[
            "prevention_support_level"
        ] = "LOW"

        output[
            "prevention_confidence"
        ] = min(
            safe_float(
                output.get(
                    "prevention_confidence",
                    0.0,
                )
            ),
            0.60,
        )

        output["FACTOR_EVIDENCE_VALID"] = True

        output[
            "FACTOR_PREVENTION_ALIGNED"
        ] = True

        output[
            "PREFERRED_PREVENTION_USED"
        ] = True

        safeguard_applied = True

        repair_notes.append(
            "Unsupported factor replaced with "
            "NO_SEPARATE_FACTOR_STATED; prevention changed to "
            "INVESTIGATION_REQUIRED"
        )

    output[
        "UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED"
    ] = safeguard_applied

    output[
        "AUTOMATIC_REPAIR_APPLIED"
    ] = bool(repair_notes)

    output[
        "AUTOMATIC_REPAIR_NOTES"
    ] = "; ".join(repair_notes)

    return output


# =============================================================================
# 22. REBUILD CRITICAL AND ADVISORY REVIEW FLAGS
# =============================================================================

def rebuild_review_flags(
    result,
    may_be_truncated,
):

    output = dict(result)

    critical_reasons = []

    advisory_reasons = []

    if (
        output.get("PROCESSING_STATUS")
        != "SUCCESS"
    ):

        critical_reasons.append(
            "Model response failed validation"
        )

    if (
        output.get("record_type")
        != "ACUTE_ACCIDENT"
    ):

        critical_reasons.append(
            "Record routed outside primary acute-accident population"
        )

    if (
        output.get("primary_cause")
        == "INSUFFICIENT_INFORMATION"
    ):

        critical_reasons.append(
            "Cause is insufficient information"
        )

    if not bool(
        output.get(
            "CAUSE_EVIDENCE_VALID",
            False,
        )
    ):

        critical_reasons.append(
            "Cause evidence is not supported by narrative"
        )

    if not bool(
        output.get(
            "FACTOR_PREVENTION_ALIGNED",
            False,
        )
    ):

        critical_reasons.append(
            "Factor-prevention combination is outside allowed map"
        )

    if str(
        output.get(
            "FIELDS_DEFAULTED",
            "",
        )
    ).strip():

        critical_reasons.append(
            "One or more required fields were absent and defaulted"
        )

    if may_be_truncated:

        advisory_reasons.append(
            "Narrative may be truncated"
        )

    if (
        output.get(
            "primary_contributing_factor"
        )
        == "NO_SEPARATE_FACTOR_STATED"
    ):

        advisory_reasons.append(
            "No separate contributing factor stated"
        )

    if (
        bool(
            output.get(
                "FACTOR_PREVENTION_ALIGNED",
                False,
            )
        )
        and not bool(
            output.get(
                "PREFERRED_PREVENTION_USED",
                False,
            )
        )
    ):

        advisory_reasons.append(
            "Allowed alternative prevention used instead of "
            "preferred mapping"
        )

    confidences = [
        safe_float(
            output.get(
                "cause_confidence",
                0.0,
            )
        ),
        safe_float(
            output.get(
                "factor_confidence",
                0.0,
            )
        ),
        safe_float(
            output.get(
                "prevention_confidence",
                0.0,
            )
        ),
    ]

    if min(confidences) < LOW_CONFIDENCE_THRESHOLD:

        advisory_reasons.append(
            "At least one confidence value is low"
        )

    if "LOW" in {
        output.get(
            "cause_support_level",
            "LOW",
        ),
        output.get(
            "factor_support_level",
            "LOW",
        ),
        output.get(
            "prevention_support_level",
            "LOW",
        ),
    }:

        advisory_reasons.append(
            "At least one support level is low"
        )

    if bool(
        output.get(
            "UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED",
            False,
        )
    ):

        advisory_reasons.append(
            "Unsupported factor conservatively replaced"
        )

    output[
        "CRITICAL_REVIEW_REQUIRED"
    ] = bool(critical_reasons)

    output[
        "CRITICAL_REVIEW_REASONS"
    ] = "; ".join(critical_reasons)

    output["ADVISORY_REVIEW"] = bool(
        advisory_reasons
    )

    output[
        "ADVISORY_REVIEW_REASONS"
    ] = "; ".join(advisory_reasons)

    return output


# =============================================================================
# 23. PREPARE ACCIDENT DATA
# =============================================================================

df = accidents.copy()


df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.upper()
)


if TEXT_COLUMN not in df.columns:

    raise KeyError(
        f"{TEXT_COLUMN!r} was not found."
    )


if SEVERITY_COLUMN not in df.columns:

    raise KeyError(
        f"{SEVERITY_COLUMN!r} was not found."
    )


id_column = next(
    (
        column
        for column in ID_COLUMN_CANDIDATES
        if column in df.columns
    ),
    None,
)


if id_column is None:

    id_column = "_SOURCE_INDEX"

    df[id_column] = df.index.astype(str)


df[id_column] = df[id_column].astype(str)


df[TEXT_COLUMN] = (
    df[TEXT_COLUMN]
    .astype("string")
    .fillna("")
    .str.strip()
)


df[SEVERITY_COLUMN] = (
    df[SEVERITY_COLUMN]
    .astype("string")
    .fillna("UNKNOWN")
    .str.strip()
)


if DATE_COLUMN in df.columns:

    df[DATE_COLUMN] = pd.to_datetime(
        df[DATE_COLUMN],
        errors="coerce",
    )

    df["ACCIDENT_YEAR"] = (
        df[DATE_COLUMN]
        .dt.year
        .astype("Int64")
    )

elif "CAL_YR" in df.columns:

    df["ACCIDENT_YEAR"] = pd.to_numeric(
        df["CAL_YR"],
        errors="coerce",
    ).astype("Int64")

else:

    raise KeyError(
        "Neither ACCIDENT_DT nor CAL_YR was found."
    )


df["_WORD_COUNT"] = (
    df[TEXT_COLUMN]
    .str.split()
    .str.len()
)


df["_CHARACTER_COUNT"] = (
    df[TEXT_COLUMN]
    .str.len()
)


df["_MAY_BE_TRUNCATED"] = (
    df["_CHARACTER_COUNT"]
    >= TRUNCATION_CHARACTER_THRESHOLD
)


df["_DUPLICATE_KEY"] = (
    df[TEXT_COLUMN]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


df = df[
    df["_WORD_COUNT"]
    >= MIN_NARRATIVE_WORDS
].copy()


df = df.drop_duplicates(
    subset=["_DUPLICATE_KEY"],
    keep="first",
).copy()


# =============================================================================
# 24. EXCLUDE BLOCK 1 DEVELOPMENT RECORDS
# =============================================================================

with open(
    BLOCK1_IDS_JSON,
    "r",
    encoding="utf-8",
) as input_file:

    block1_id_payload = json.load(input_file)


block1_record_ids = set(
    str(value)
    for value in block1_id_payload[
        "record_ids"
    ]
)


unseen_df = df[
    ~df[id_column].isin(
        block1_record_ids
    )
].copy()


if unseen_df.empty:

    raise ValueError(
        "No unseen records remained after excluding Block 1 IDs."
    )


# =============================================================================
# 25. SEVERITY, TIME, AND SECTOR FIELDS
# =============================================================================

def assign_severity_group(value):

    value = str(value).upper()

    if "FATAL" in value:

        return "Fatality"

    if any(
        term in value
        for term in [
            "PERM TOT",
            "PERMANENT TOTAL",
            "PERM PART",
            "PERMANENT PART",
        ]
    ):

        return "Permanent disability"

    if any(
        term in value
        for term in [
            "DAYS AWAY",
            "DYS AWY",
            "LOST WORKDAY",
            "LOST TIME",
        ]
    ):

        return "Days away"

    if any(
        term in value
        for term in [
            "RESTRICT",
            "TRANSFER",
            "RSTR ACT",
            "RESTRCTD ACT",
        ]
    ):

        return "Restricted or transferred"

    return "No lost time or other"


def assign_time_period(year):

    if pd.isna(year):

        return "Unknown period"

    year = int(year)

    if year <= 2008:

        return EARLY_PERIOD

    if year <= 2017:

        return MIDDLE_PERIOD

    return RECENT_PERIOD


def assign_sector(value):

    value = str(value).strip().upper()

    if value == "C":

        return "Coal"

    if value == "M":

        return "Metal/nonmetal"

    return "Unknown"


unseen_df["SEVERITY_GROUP"] = (
    unseen_df[
        SEVERITY_COLUMN
    ]
    .map(assign_severity_group)
)


unseen_df["TIME_PERIOD"] = (
    unseen_df[
        "ACCIDENT_YEAR"
    ]
    .map(assign_time_period)
)


if "COAL_METAL_IND" in unseen_df.columns:

    unseen_df["SECTOR_GROUP"] = (
        unseen_df[
            "COAL_METAL_IND"
        ]
        .map(assign_sector)
    )

else:

    unseen_df["SECTOR_GROUP"] = "Unknown"


# =============================================================================
# 26. PRELIMINARY RECORD ROUTING
# =============================================================================

def preliminary_record_type_hint(row):

    narrative = str(
        row.get(TEXT_COLUMN, "")
    ).upper()

    classification = str(
        row.get("CLASSIFICATION", "")
    ).upper()

    nature = str(
        row.get("NATURE_INJURY", "")
    ).upper()

    degree = str(
        row.get("DEGREE_INJURY", "")
    ).upper()

    combined = " ".join(
        [
            narrative,
            classification,
            nature,
            degree,
        ]
    )

    nonwork_patterns = [
        "NOT DUE TO WORK",
        "NOT WORK RELATED",
        "CANNOT BE CONNECTED TO THE JOB",
        "NON-OCCUPATIONAL",
        "NATURAL CAUSE",
        "CAMPING ON PROPERTY",
        "NOT A MINER",
    ]

    natural_patterns = [
        "HEART ATTACK",
        "CARDIAC",
        "STROKE",
        "SEIZURE",
        "DIABETES",
        "BLOOD SUGAR",
    ]

    illness_patterns = [
        "HEARING LOSS",
        "SILICOSIS",
        "PNEUMOCONIOSIS",
        "DUST DISEASE",
        "REPEATED TRAUMA",
        "OCCUPATIONAL ILLNESS",
    ]

    noninjury_patterns = [
        "ACCIDENT ONLY",
        "NO INJURY",
        "NO INJURIES",
        "NO PERSON INJURED",
    ]

    if any(
        value in combined
        for value in (
            nonwork_patterns
            + natural_patterns
        )
    ):

        return (
            "NONWORK_RELATED_OR_NATURAL_EVENT"
        )

    if any(
        value in combined
        for value in illness_patterns
    ):

        return (
            "OCCUPATIONAL_EXPOSURE_OR_ILLNESS"
        )

    if any(
        value in combined
        for value in noninjury_patterns
    ):

        return "NONINJURY_REPORT"

    return "ACUTE_ACCIDENT"


unseen_df[
    "PRELIMINARY_RECORD_TYPE_HINT"
] = unseen_df.apply(
    preliminary_record_type_hint,
    axis=1,
)


candidate_df = unseen_df[
    unseen_df[
        "PRELIMINARY_RECORD_TYPE_HINT"
    ].eq("ACUTE_ACCIDENT")
].copy()


if len(candidate_df) < TOTAL_SAMPLE_SIZE:

    raise ValueError(
        "Fewer than 2,000 candidate acute-accident narratives remain."
    )


# =============================================================================
# 27. SELECT BALANCED 2,000-RECORD SAMPLE
# =============================================================================

def create_balanced_sample(
    data,
    target_per_severity,
    total_target,
    random_state,
):

    selected_indexes = []

    sampling_records = []

    for severity_number, severity in enumerate(
        SEVERITY_ORDER
    ):

        severity_df = data[
            data[
                "SEVERITY_GROUP"
            ].eq(severity)
        ].copy()

        selected_for_severity = []

        base_target = (
            target_per_severity
            // len(TIME_PERIOD_ORDER)
        )

        remainder = (
            target_per_severity
            % len(TIME_PERIOD_ORDER)
        )

        for period_number, period in enumerate(
            TIME_PERIOD_ORDER
        ):

            period_df = severity_df[
                severity_df[
                    "TIME_PERIOD"
                ].eq(period)
            ]

            requested = (
                base_target
                + (
                    1
                    if period_number < remainder
                    else 0
                )
            )

            selected_count = min(
                requested,
                len(period_df),
            )

            if selected_count > 0:

                sampled = period_df.sample(
                    n=selected_count,
                    random_state=(
                        random_state
                        + severity_number * 100
                        + period_number
                    ),
                )

                selected_for_severity.extend(
                    sampled.index.tolist()
                )

            sampling_records.append(
                {
                    "Severity group":
                        severity,

                    "Time period":
                        period,

                    "Available":
                        len(period_df),

                    "Target":
                        requested,

                    "Initially selected":
                        selected_count,
                }
            )

        shortage = (
            target_per_severity
            - len(selected_for_severity)
        )

        if shortage > 0:

            remaining = severity_df.drop(
                index=selected_for_severity,
                errors="ignore",
            )

            fill_count = min(
                shortage,
                len(remaining),
            )

            if fill_count > 0:

                fill_sample = remaining.sample(
                    n=fill_count,
                    random_state=(
                        random_state
                        + severity_number * 1000
                    ),
                )

                selected_for_severity.extend(
                    fill_sample.index.tolist()
                )

        selected_indexes.extend(
            selected_for_severity
        )

    selected_indexes = list(
        dict.fromkeys(selected_indexes)
    )

    sample = data.loc[
        selected_indexes
    ].copy()

    if len(sample) < total_target:

        remaining = data.drop(
            index=selected_indexes,
            errors="ignore",
        )

        additional_count = min(
            total_target - len(sample),
            len(remaining),
        )

        if additional_count > 0:

            sample = pd.concat(
                [
                    sample,
                    remaining.sample(
                        n=additional_count,
                        random_state=(
                            random_state + 9999
                        ),
                    ),
                ],
                axis=0,
            )

    if len(sample) > total_target:

        sample = sample.sample(
            n=total_target,
            random_state=random_state,
        )

    sample = (
        sample.sample(
            frac=1,
            random_state=random_state,
        )
        .reset_index(drop=False)
        .rename(
            columns={
                "index":
                    "_ORIGINAL_INDEX"
            }
        )
    )

    return (
        sample,
        pd.DataFrame(sampling_records),
    )


block2_df, sampling_summary = create_balanced_sample(
    candidate_df,
    TARGET_PER_SEVERITY,
    TOTAL_SAMPLE_SIZE,
    RANDOM_STATE,
)


if len(block2_df) != TOTAL_SAMPLE_SIZE:

    raise ValueError(
        f"Expected {TOTAL_SAMPLE_SIZE} records but selected "
        f"{len(block2_df)}."
    )


overlapping_ids = set(
    block2_df[id_column].astype(str)
).intersection(
    block1_record_ids
)


if overlapping_ids:

    raise RuntimeError(
        "Block 2 contains Block 1 development records."
    )


print(
    f"Block 2 unseen narratives selected: "
    f"{len(block2_df):,}"
)


# =============================================================================
# 28. SAVE BLOCK 2 IDS
# =============================================================================

block2_id_payload = {
    "created_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "purpose":
        "Block 2 unseen analytical sample",

    "id_column":
        id_column,

    "record_count":
        len(block2_df),

    "record_ids":
        block2_df[
            id_column
        ].astype(str).tolist(),
}


with open(
    BLOCK2_IDS_JSON,
    "w",
    encoding="utf-8",
) as output_file:

    json.dump(
        block2_id_payload,
        output_file,
        indent=2,
        ensure_ascii=False,
    )


# =============================================================================
# 29. CHECKPOINT UTILITIES
# =============================================================================

def load_checkpoint(path):

    path = Path(path)

    if not path.exists():

        return {}

    records = {}

    with path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()

            if not line:

                continue

            try:

                record = json.loads(line)

                records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record

            except Exception:

                continue

    return records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


checkpoint_records = load_checkpoint(
    BLOCK2_CHECKPOINT_JSONL
)


print(
    f"Existing checkpoint records: "
    f"{len(checkpoint_records):,}"
)


# =============================================================================
# 30. PROCESS 2,000 NARRATIVES
# =============================================================================

context_columns = [
    column
    for column in [
        "ACCIDENT_DT",
        "ACCIDENT_YEAR",
        "TIME_PERIOD",
        "SEVERITY_GROUP",
        "SECTOR_GROUP",
        "CLASSIFICATION",
        "ACCIDENT_TYPE",
        "ACTIVITY",
        "INJURY_SOURCE",
        "NATURE_INJURY",
        "INJ_BODY_PART",
        "DEGREE_INJURY",
        "MINING_EQUIP",
        "OCCUPATION",
        "COAL_METAL_IND",
        "DAYS_RESTRICT",
        "DAYS_LOST",
        "PRELIMINARY_RECORD_TYPE_HINT",
    ]
    if column in block2_df.columns
]


processed_records = []


for row_number, row in block2_df.iterrows():

    record_id = str(row[id_column])

    if (
        not REPROCESS_COMPLETED
        and record_id in checkpoint_records
    ):

        print(
            f"[{row_number + 1}/"
            f"{len(block2_df)}] "
            f"Checkpoint: {record_id}"
        )

        processed_records.append(
            checkpoint_records[record_id]
        )

        continue

    narrative = str(row[TEXT_COLUMN])

    may_be_truncated = bool(
        row["_MAY_BE_TRUNCATED"]
    )

    routing_hint = str(
        row[
            "PRELIMINARY_RECORD_TYPE_HINT"
        ]
    )

    print(
        f"[{row_number + 1}/"
        f"{len(block2_df)}] "
        f"{row['SEVERITY_GROUP']} | "
        f"{row['TIME_PERIOD']} | "
        f"{record_id}"
    )

    start_time = time.time()

    model_result = classify_narrative(
        narrative=narrative,
        preliminary_record_type=routing_hint,
        may_be_truncated=may_be_truncated,
    )

    model_result = (
        apply_unsupported_factor_safeguard(
            model_result
        )
    )

    model_result = rebuild_review_flags(
        model_result,
        may_be_truncated,
    )

    record = {
        "SOURCE_RECORD_ID":
            record_id,

        "SOURCE_INDEX":
            row.get(
                "_ORIGINAL_INDEX",
                row_number,
            ),

        "NARRATIVE":
            narrative,

        "NARRATIVE_WORD_COUNT":
            int(row["_WORD_COUNT"]),

        "NARRATIVE_CHARACTER_COUNT":
            int(row["_CHARACTER_COUNT"]),

        "NARRATIVE_MAY_BE_TRUNCATED":
            may_be_truncated,

        **{
            column:
                (
                    None
                    if pd.isna(row[column])
                    else row[column]
                )
            for column in context_columns
        },

        **model_result,

        "MODEL_ID":
            MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }

    processed_records.append(record)

    append_checkpoint(
        record,
        BLOCK2_CHECKPOINT_JSONL,
    )

    gc.collect()

    torch.cuda.empty_cache()


# =============================================================================
# 31. BUILD FINAL RESULTS DATAFRAME
# =============================================================================

results_df = pd.DataFrame(
    processed_records
)


results_df = results_df.rename(
    columns={
        "record_type":
            "RECORD_TYPE",

        "primary_cause":
            "PRIMARY_CAUSE",

        "primary_contributing_factor":
            "PRIMARY_CONTRIBUTING_FACTOR",

        "prevention_theme":
            "PREVENTION_THEME",

        "cause_evidence_text":
            "CAUSE_EVIDENCE_TEXT",

        "factor_evidence_text":
            "FACTOR_EVIDENCE_TEXT",

        "prevention_rationale":
            "PREVENTION_RATIONALE",

        "cause_confidence":
            "CAUSE_CONFIDENCE",

        "factor_confidence":
            "FACTOR_CONFIDENCE",

        "prevention_confidence":
            "PREVENTION_CONFIDENCE",

        "cause_support_level":
            "CAUSE_SUPPORT_LEVEL",

        "factor_support_level":
            "FACTOR_SUPPORT_LEVEL",

        "prevention_support_level":
            "PREVENTION_SUPPORT_LEVEL",

        "insufficient_information_reason":
            "INSUFFICIENT_INFORMATION_REASON",

        "reviewer_note":
            "REVIEWER_NOTE",
    }
)


boolean_columns = [
    "SCHEMA_VALID",
    "CAUSE_EVIDENCE_VALID",
    "FACTOR_EVIDENCE_VALID",
    "FACTOR_PREVENTION_ALIGNED",
    "PREFERRED_PREVENTION_USED",
    "AUTOMATIC_REPAIR_APPLIED",
    "UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED",
    "CRITICAL_REVIEW_REQUIRED",
    "ADVISORY_REVIEW",
]


for column in boolean_columns:

    if column in results_df.columns:

        results_df[column] = (
            results_df[column]
            .fillna(False)
            .astype(bool)
        )


# =============================================================================
# 32. ANALYTICAL POPULATION STATUS
# =============================================================================

def assign_analysis_status(row):

    success = (
        row.get("PROCESSING_STATUS")
        == "SUCCESS"
    )

    acute = (
        row.get("RECORD_TYPE")
        == "ACUTE_ACCIDENT"
    )

    cause_valid = bool(
        row.get(
            "CAUSE_EVIDENCE_VALID",
            False,
        )
    )

    factor = row.get(
        "PRIMARY_CONTRIBUTING_FACTOR",
        "",
    )

    factor_valid = bool(
        row.get(
            "FACTOR_EVIDENCE_VALID",
            False,
        )
    )

    aligned = bool(
        row.get(
            "FACTOR_PREVENTION_ALIGNED",
            False,
        )
    )

    critical = bool(
        row.get(
            "CRITICAL_REVIEW_REQUIRED",
            False,
        )
    )

    if (
        success
        and acute
        and cause_valid
        and factor
        != "NO_SEPARATE_FACTOR_STATED"
        and factor_valid
        and aligned
        and not critical
    ):

        return "VALID_FOR_FULL_ANALYSIS"

    if (
        success
        and acute
        and cause_valid
        and factor
        == "NO_SEPARATE_FACTOR_STATED"
    ):

        return "VALID_FOR_CAUSE_ANALYSIS_ONLY"

    return "EXCLUDED_OR_REVIEW_REQUIRED"


results_df["ANALYSIS_STATUS"] = (
    results_df.apply(
        assign_analysis_status,
        axis=1,
    )
)


# Save structured results for later Block 3 selection.
results_df.to_csv(
    BLOCK2_RESULTS_CSV,
    index=False,
    encoding="utf-8-sig",
)


# =============================================================================
# 33. DEFINE ANALYTICAL POPULATIONS
# =============================================================================

cause_analysis_df = results_df[
    results_df[
        "ANALYSIS_STATUS"
    ].isin(
        [
            "VALID_FOR_FULL_ANALYSIS",
            "VALID_FOR_CAUSE_ANALYSIS_ONLY",
        ]
    )
].copy()


full_analysis_df = results_df[
    results_df[
        "ANALYSIS_STATUS"
    ].eq(
        "VALID_FOR_FULL_ANALYSIS"
    )
].copy()


review_df = results_df[
    results_df[
        "ANALYSIS_STATUS"
    ].eq(
        "EXCLUDED_OR_REVIEW_REQUIRED"
    )
].copy()


successful_acute_df = results_df[
    (
        results_df[
            "PROCESSING_STATUS"
        ].eq("SUCCESS")
    )
    &
    (
        results_df[
            "RECORD_TYPE"
        ].eq("ACUTE_ACCIDENT")
    )
].copy()


# =============================================================================
# 34. SUMMARY FUNCTIONS
# =============================================================================

def category_summary(
    dataframe,
    category_column,
    category_name,
):

    if dataframe.empty:

        return pd.DataFrame(
            columns=[
                category_name,
                "Count",
                "Percent",
            ]
        )

    output = (
        dataframe[
            category_column
        ]
        .value_counts(
            dropna=False
        )
        .rename_axis(category_name)
        .reset_index(
            name="Count"
        )
    )

    output["Percent"] = (
        100.0
        * output["Count"]
        / len(dataframe)
    )

    return output


def normalized_crosstab(
    dataframe,
    row_column,
    category_column,
    row_order,
):

    counts = pd.crosstab(
        dataframe[row_column],
        dataframe[category_column],
    )

    counts = counts.reindex(
        row_order,
        fill_value=0,
    )

    percentages = (
        counts.div(
            counts.sum(axis=1).replace(
                0,
                np.nan,
            ),
            axis=0,
        )
        * 100.0
    ).fillna(0.0)

    return counts, percentages


def readable_label(value):

    return (
        str(value)
        .replace("_", " ")
        .title()
    )


# =============================================================================
# 35. OVERALL CATEGORY SUMMARIES
# =============================================================================

cause_summary = category_summary(
    cause_analysis_df,
    "PRIMARY_CAUSE",
    "Primary cause",
)


factor_summary = category_summary(
    full_analysis_df,
    "PRIMARY_CONTRIBUTING_FACTOR",
    "Contributing factor",
)


prevention_summary = category_summary(
    full_analysis_df,
    "PREVENTION_THEME",
    "Prevention theme",
)


record_type_summary = category_summary(
    results_df,
    "RECORD_TYPE",
    "Record type",
)


analysis_status_summary = category_summary(
    results_df,
    "ANALYSIS_STATUS",
    "Analysis status",
)


# =============================================================================
# 36. SEVERITY AND TIME-PERIOD TABLES
# =============================================================================

cause_severity_counts, cause_severity_pct = (
    normalized_crosstab(
        cause_analysis_df,
        "SEVERITY_GROUP",
        "PRIMARY_CAUSE",
        SEVERITY_ORDER,
    )
)


factor_severity_counts, factor_severity_pct = (
    normalized_crosstab(
        full_analysis_df,
        "SEVERITY_GROUP",
        "PRIMARY_CONTRIBUTING_FACTOR",
        SEVERITY_ORDER,
    )
)


prevention_severity_counts, prevention_severity_pct = (
    normalized_crosstab(
        full_analysis_df,
        "SEVERITY_GROUP",
        "PREVENTION_THEME",
        SEVERITY_ORDER,
    )
)


cause_period_counts, cause_period_pct = (
    normalized_crosstab(
        cause_analysis_df,
        "TIME_PERIOD",
        "PRIMARY_CAUSE",
        TIME_PERIOD_ORDER,
    )
)


factor_period_counts, factor_period_pct = (
    normalized_crosstab(
        full_analysis_df,
        "TIME_PERIOD",
        "PRIMARY_CONTRIBUTING_FACTOR",
        TIME_PERIOD_ORDER,
    )
)


prevention_period_counts, prevention_period_pct = (
    normalized_crosstab(
        full_analysis_df,
        "TIME_PERIOD",
        "PREVENTION_THEME",
        TIME_PERIOD_ORDER,
    )
)


# =============================================================================
# 37. CHI-SQUARE AND CRAMÉR'S V
# =============================================================================

def cramers_v(table):

    matrix = np.asarray(
        table,
        dtype=float,
    )

    if (
        matrix.size == 0
        or matrix.sum() == 0
    ):

        return np.nan

    chi2, p_value, dof, expected = (
        chi2_contingency(matrix)
    )

    n = matrix.sum()

    rows, columns = matrix.shape

    denominator = min(
        rows - 1,
        columns - 1,
    )

    if denominator <= 0:

        return np.nan

    return math.sqrt(
        chi2
        / (
            n
            * denominator
        )
    )


def association_test(
    dataframe,
    group_column,
    category_column,
    comparison_name,
):

    table = pd.crosstab(
        dataframe[group_column],
        dataframe[category_column],
    )

    table = table.loc[
        table.sum(axis=1) > 0,
        table.sum(axis=0) > 0,
    ]

    if (
        table.shape[0] < 2
        or table.shape[1] < 2
    ):

        return {
            "Comparison":
                comparison_name,

            "Chi-square":
                np.nan,

            "Degrees of freedom":
                np.nan,

            "P-value":
                np.nan,

            "Cramer's V":
                np.nan,

            "Records":
                int(table.values.sum()),
        }

    chi2, p_value, dof, expected = (
        chi2_contingency(table)
    )

    return {
        "Comparison":
            comparison_name,

        "Chi-square":
            chi2,

        "Degrees of freedom":
            dof,

        "P-value":
            p_value,

        "Cramer's V":
            cramers_v(table.values),

        "Records":
            int(table.values.sum()),
    }


association_results = pd.DataFrame(
    [
        association_test(
            cause_analysis_df,
            "SEVERITY_GROUP",
            "PRIMARY_CAUSE",
            "Primary cause by severity",
        ),

        association_test(
            full_analysis_df,
            "SEVERITY_GROUP",
            "PRIMARY_CONTRIBUTING_FACTOR",
            "Contributing factor by severity",
        ),

        association_test(
            full_analysis_df,
            "SEVERITY_GROUP",
            "PREVENTION_THEME",
            "Prevention theme by severity",
        ),

        association_test(
            cause_analysis_df,
            "TIME_PERIOD",
            "PRIMARY_CAUSE",
            "Primary cause by time period",
        ),

        association_test(
            full_analysis_df,
            "TIME_PERIOD",
            "PRIMARY_CONTRIBUTING_FACTOR",
            "Contributing factor by time period",
        ),

        association_test(
            full_analysis_df,
            "TIME_PERIOD",
            "PREVENTION_THEME",
            "Prevention theme by time period",
        ),
    ]
)


# =============================================================================
# 38. EARLY-VERSUS-RECENT CHANGE TABLES
# =============================================================================

def early_recent_change_table(
    percentage_table,
    count_table,
    category_label,
):

    if (
        EARLY_PERIOD not in percentage_table.index
        or RECENT_PERIOD not in percentage_table.index
    ):

        return pd.DataFrame()

    records = []

    for category in percentage_table.columns:

        total_count = int(
            count_table[category].sum()
        )

        if (
            total_count
            < MIN_CATEGORY_COUNT_FOR_TREND
        ):

            continue

        early_pct = float(
            percentage_table.loc[
                EARLY_PERIOD,
                category,
            ]
        )

        recent_pct = float(
            percentage_table.loc[
                RECENT_PERIOD,
                category,
            ]
        )

        change = recent_pct - early_pct

        if change > 1.0:

            direction = "Increasing share"

        elif change < -1.0:

            direction = "Declining share"

        else:

            direction = (
                "Approximately stable"
            )

        records.append(
            {
                category_label:
                    category,

                "Total count":
                    total_count,

                "Early percent":
                    early_pct,

                "Recent percent":
                    recent_pct,

                "Percentage-point change":
                    change,

                "Direction":
                    direction,
            }
        )

    if not records:

        return pd.DataFrame()

    return (
        pd.DataFrame(records)
        .sort_values(
            "Percentage-point change",
            ascending=False,
        )
        .reset_index(drop=True)
    )


cause_change_table = early_recent_change_table(
    cause_period_pct,
    cause_period_counts,
    "Primary cause",
)


factor_change_table = early_recent_change_table(
    factor_period_pct,
    factor_period_counts,
    "Contributing factor",
)


prevention_change_table = (
    early_recent_change_table(
        prevention_period_pct,
        prevention_period_counts,
        "Prevention theme",
    )
)


# =============================================================================
# 39. RELATIONSHIP TABLES
# =============================================================================

cause_factor_counts = pd.crosstab(
    full_analysis_df["PRIMARY_CAUSE"],
    full_analysis_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
)


factor_prevention_counts = pd.crosstab(
    full_analysis_df[
        "PRIMARY_CONTRIBUTING_FACTOR"
    ],
    full_analysis_df["PREVENTION_THEME"],
)


# =============================================================================
# 40. PROCESSING QUALITY SUMMARIES
# =============================================================================

quality_summary = pd.DataFrame(
    [
        {
            "Measure":
                "Block 2 narratives",

            "Value":
                len(results_df),
        },
        {
            "Measure":
                "Successful classifications",

            "Value":
                int(
                    results_df[
                        "PROCESSING_STATUS"
                    ].eq("SUCCESS").sum()
                ),
        },
        {
            "Measure":
                "Schema-valid classifications",

            "Value":
                int(
                    results_df[
                        "SCHEMA_VALID"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Valid cause evidence",

            "Value":
                int(
                    results_df[
                        "CAUSE_EVIDENCE_VALID"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Unsupported-factor safeguards",

            "Value":
                int(
                    results_df[
                        "UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Factor-prevention aligned",

            "Value":
                int(
                    results_df[
                        "FACTOR_PREVENTION_ALIGNED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Critical review required",

            "Value":
                int(
                    results_df[
                        "CRITICAL_REVIEW_REQUIRED"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Advisory review",

            "Value":
                int(
                    results_df[
                        "ADVISORY_REVIEW"
                    ].sum()
                ),
        },
        {
            "Measure":
                "Valid for full analysis",

            "Value":
                len(full_analysis_df),
        },
        {
            "Measure":
                "Valid for cause analysis only",

            "Value":
                int(
                    results_df[
                        "ANALYSIS_STATUS"
                    ]
                    .eq(
                        "VALID_FOR_CAUSE_ANALYSIS_ONLY"
                    )
                    .sum()
                ),
        },
        {
            "Measure":
                "Excluded or review required",

            "Value":
                len(review_df),
        },
        {
            "Measure":
                "Mean processing seconds",

            "Value":
                round(
                    float(
                        results_df[
                            "PROCESSING_SECONDS"
                        ].mean()
                    ),
                    2,
                ),
        },
    ]
)


def explode_reason_counts(
    dataframe,
    column,
    output_name,
):

    reasons = (
        dataframe[column]
        .fillna("")
        .str.split("; ")
        .explode()
    )

    return (
        reasons[
            reasons.str.len() > 0
        ]
        .value_counts()
        .rename_axis(output_name)
        .reset_index(name="Count")
    )


critical_review_summary = (
    explode_reason_counts(
        results_df,
        "CRITICAL_REVIEW_REASONS",
        "Critical review reason",
    )
)


advisory_review_summary = (
    explode_reason_counts(
        results_df,
        "ADVISORY_REVIEW_REASONS",
        "Advisory review reason",
    )
)


# =============================================================================
# 41. QUALITY-SENSITIVITY ANALYSIS
# =============================================================================

def cause_distribution(
    dataframe,
    display_name,
):

    return (
        dataframe[
            "PRIMARY_CAUSE"
        ]
        .value_counts(normalize=True)
        .mul(100)
        .rename(display_name)
    )


sensitivity_table = pd.concat(
    [
        cause_distribution(
            successful_acute_df,
            "All successful acute",
        ),
        cause_distribution(
            cause_analysis_df,
            "Cause-valid population",
        ),
        cause_distribution(
            full_analysis_df,
            "Full-analysis population",
        ),
    ],
    axis=1,
).fillna(0.0)


sensitivity_table[
    "Maximum percentage spread"
] = (
    sensitivity_table.max(axis=1)
    - sensitivity_table.min(axis=1)
)


sensitivity_display = (
    sensitivity_table
    .reset_index()
    .rename(
        columns={
            "index":
                "Primary cause"
        }
    )
    .sort_values(
        "Maximum percentage spread",
        ascending=False,
    )
)


# =============================================================================
# 42. REPRESENTATIVE EXAMPLES
# =============================================================================

representative_examples = (
    full_analysis_df
    .sort_values(
        [
            "CAUSE_CONFIDENCE",
            "FACTOR_CONFIDENCE",
            "PREVENTION_CONFIDENCE",
        ],
        ascending=False,
    )
    .head(EXAMPLES_PER_SECTION)
)


cause_only_examples = (
    cause_analysis_df[
        cause_analysis_df[
            "ANALYSIS_STATUS"
        ].eq(
            "VALID_FOR_CAUSE_ANALYSIS_ONLY"
        )
    ]
    .head(EXAMPLES_PER_SECTION)
)


safeguard_examples = (
    results_df[
        results_df[
            "UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED"
        ]
    ]
    .head(EXAMPLES_PER_SECTION)
)


critical_examples = (
    review_df
    .head(EXAMPLES_PER_SECTION)
)


# =============================================================================
# 43. PDF HELPERS
# =============================================================================

pdf = PdfPages(OUTPUT_PDF)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {"value": 0}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (11, 8.5)
            if landscape
            else (8.5, 11)
        )
    )

    axis = figure.add_axes(
        [0, 0, 1, 1]
    )

    axis.axis("off")

    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )

    return figure


def save_page(figure):

    page_counter["value"] += 1

    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )

    pdf.savefig(
        figure,
        bbox_inches="tight",
    )

    plt.close(figure)


def add_text_pages(
    title,
    lines,
    font_size=8.5,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped = []

    for line in lines:

        if not line:

            wrapped.append("")

        else:

            wrapped.extend(
                textwrap.wrap(
                    str(line),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )

    chunks = [
        wrapped[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(wrapped),
            lines_per_page,
        )
    ] or [[""]]

    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title

        if len(chunks) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )

        figure = create_page(page_title)

        figure.text(
            0.055,
            0.915,
            "\n".join(chunk),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.20,
        )

        save_page(figure)


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=25,
    landscape=False,
):

    if dataframe is None or dataframe.empty:

        add_text_pages(
            title,
            ["No results were produced."],
        )

        return

    printable = dataframe.copy()

    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[column]
        ):

            printable[column] = (
                printable[column]
                .map(
                    lambda value:
                        ""
                        if pd.isna(value)
                        else f"{value:.4f}"
                )
            )

    printable.columns = [
        str(column)
        .replace("_", " ")
        .title()
        for column in printable.columns
    ]

    for column in printable.columns:

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    ""
                    if pd.isna(value)
                    else str(value)
            )
        )

        printable[column] = (
            printable[column]
            .map(
                lambda value:
                    "\n".join(
                        textwrap.wrap(
                            value,
                            width=column_wrap_width,
                            break_long_words=False,
                            break_on_hyphens=False,
                        )
                    )
                    if value
                    else ""
            )
        )

    for start in range(
        0,
        len(printable),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start + rows_per_page
        ]

        figure = create_page(
            (
                title
                if start == 0
                else title + " - Continued"
            ),
            landscape=landscape,
        )

        axis = figure.add_axes(
            [0.025, 0.065, 0.95, 0.84]
        )

        axis.axis("off")

        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )

        table.auto_set_font_size(False)

        table.set_fontsize(font_size)

        table.scale(1, 1.42)

        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor("#AAB2BD")

            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )

                cell.set_text_props(
                    weight="bold"
                )

        save_page(figure)


def add_bar_chart(
    title,
    dataframe,
    category_column,
    color,
    top_n=12,
):

    if dataframe.empty:

        return

    plot_df = (
        dataframe.head(top_n)
        .sort_values(
            "Count",
            ascending=True,
        )
    )

    figure = plt.figure(
        figsize=(10, 7)
    )

    axis = figure.add_subplot(111)

    axis.barh(
        [
            readable_label(value)
            for value in plot_df[
                category_column
            ]
        ],
        plot_df["Count"],
        color=color,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel("Narrative count")

    axis.grid(
        axis="x",
        alpha=0.25,
    )

    figure.tight_layout()

    save_page(figure)


def add_percentage_heatmap(
    title,
    percentage_table,
    top_categories,
    x_label,
    y_label,
):

    if percentage_table.empty:

        return

    columns = [
        category
        for category in top_categories
        if category
        in percentage_table.columns
    ]

    if not columns:

        return

    table = percentage_table[columns]

    figure = plt.figure(
        figsize=(11, 7.5)
    )

    axis = figure.add_subplot(111)

    matrix = table.values.astype(float)

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
        vmin=0,
    )

    axis.set_xticks(
        range(len(table.columns))
    )

    axis.set_xticklabels(
        [
            readable_label(value)
            for value in table.columns
        ],
        rotation=35,
        ha="right",
        fontsize=7,
    )

    axis.set_yticks(
        range(len(table.index))
    )

    axis.set_yticklabels(
        [
            readable_label(value)
            for value in table.index
        ],
        fontsize=8,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_xlabel(x_label)

    axis.set_ylabel(y_label)

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = matrix[
                row_number,
                column_number,
            ]

            if value >= 1.0:

                axis.text(
                    column_number,
                    row_number,
                    f"{value:.1f}%",
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Within-group percentage",
    )

    figure.tight_layout()

    save_page(figure)


def add_trend_chart(
    title,
    percentage_table,
    categories,
):

    available = [
        category
        for category in categories
        if category
        in percentage_table.columns
    ]

    if not available:

        return

    figure = plt.figure(
        figsize=(10.5, 7)
    )

    axis = figure.add_subplot(111)

    x_positions = np.arange(
        len(percentage_table.index)
    )

    for category in available:

        axis.plot(
            x_positions,
            percentage_table[category],
            marker="o",
            linewidth=2,
            label=readable_label(category),
        )

    axis.set_xticks(x_positions)

    axis.set_xticklabels(
        percentage_table.index,
        rotation=15,
        ha="right",
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    axis.set_ylabel(
        "Within-period percentage"
    )

    axis.grid(
        axis="y",
        alpha=0.25,
    )

    axis.legend(
        fontsize=7.5,
        loc="best",
    )

    figure.tight_layout()

    save_page(figure)


def add_relationship_heatmap(
    title,
    counts_table,
    top_rows=10,
    top_columns=10,
):

    if counts_table.empty:

        return

    row_names = (
        counts_table.sum(axis=1)
        .nlargest(top_rows)
        .index
    )

    column_names = (
        counts_table.sum(axis=0)
        .nlargest(top_columns)
        .index
    )

    table = counts_table.loc[
        row_names,
        column_names,
    ]

    figure = plt.figure(
        figsize=(11, 8)
    )

    axis = figure.add_subplot(111)

    matrix = table.values.astype(float)

    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )

    axis.set_xticks(
        range(len(table.columns))
    )

    axis.set_xticklabels(
        [
            readable_label(value)
            for value in table.columns
        ],
        rotation=40,
        ha="right",
        fontsize=7,
    )

    axis.set_yticks(
        range(len(table.index))
    )

    axis.set_yticklabels(
        [
            readable_label(value)
            for value in table.index
        ],
        fontsize=7,
    )

    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )

    for row_number in range(
        matrix.shape[0]
    ):

        for column_number in range(
            matrix.shape[1]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )

            if value > 0:

                axis.text(
                    column_number,
                    row_number,
                    str(value),
                    ha="center",
                    va="center",
                    fontsize=7,
                )

    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )

    figure.tight_layout()

    save_page(figure)


def add_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=10,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )

        return

    lines = [
        explanation,
        "",
    ]

    for example_number, (
        row_index,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 82,
                (
                    f"Example {example_number} | "
                    f"Record "
                    f"{row['SOURCE_RECORD_ID']} | "
                    f"{row.get('SEVERITY_GROUP', '')} | "
                    f"{row.get('TIME_PERIOD', '')}"
                ),
                (
                    f"Analysis status: "
                    f"{row.get('ANALYSIS_STATUS', '')}"
                ),
                (
                    f"Cause: "
                    f"{row.get('PRIMARY_CAUSE', '')}"
                ),
                (
                    f"Factor: "
                    f"{row.get('PRIMARY_CONTRIBUTING_FACTOR', '')}"
                ),
                (
                    f"Prevention: "
                    f"{row.get('PREVENTION_THEME', '')}"
                ),
                (
                    f"Critical review: "
                    f"{row.get('CRITICAL_REVIEW_REASONS', '')}"
                ),
                (
                    f"Advisory review: "
                    f"{row.get('ADVISORY_REVIEW_REASONS', '')}"
                ),
                "",
                "Narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "Cause evidence: "
                    f"{row.get('CAUSE_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Factor evidence: "
                    f"{row.get('FACTOR_EVIDENCE_TEXT', '')}"
                ),
                (
                    "Prevention rationale: "
                    f"{row.get('PREVENTION_RATIONALE', '')}"
                ),
                "",
            ]
        )

    add_text_pages(
        title,
        lines,
        font_size=7.7,
        lines_per_page=48,
        wrap_width=104,
    )


# =============================================================================
# 44. SELECT TOP VISUAL CATEGORIES
# =============================================================================

top_causes = (
    cause_summary[
        "Primary cause"
    ]
    .head(TOP_CATEGORIES_TO_PLOT)
    .tolist()
)


top_factors = (
    factor_summary[
        "Contributing factor"
    ]
    .head(TOP_CATEGORIES_TO_PLOT)
    .tolist()
)


top_preventions = (
    prevention_summary[
        "Prevention theme"
    ]
    .head(TOP_CATEGORIES_TO_PLOT)
    .tolist()
)


# =============================================================================
# 45. GENERATE PDF REPORT
# =============================================================================

# Cover.
figure = create_page()

figure.text(
    0.5,
    0.79,
    "MSHA Unseen Narrative Analysis",
    ha="center",
    fontsize=20,
    fontweight="bold",
)

figure.text(
    0.5,
    0.73,
    (
        "Block 2 - Causes, Contributing Factors, "
        "and Prevention Themes"
    ),
    ha="center",
    fontsize=12.5,
    color="#444444",
)

figure.text(
    0.5,
    0.685,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)

cover_lines = [
    f"Unseen narratives processed: {len(results_df):,}",
    f"Block 1 development IDs excluded: "
    f"{len(block1_record_ids):,}",
    f"Cause-analysis population: "
    f"{len(cause_analysis_df):,}",
    f"Full-analysis population: "
    f"{len(full_analysis_df):,}",
    f"Excluded/review population: "
    f"{len(review_df):,}",
    f"Unsupported factors replaced: "
    f"{int(results_df['UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED'].sum()):,}",
    f"Model: {MODEL_ID}",
    f"Prompt: {PROMPT_VERSION}",
    "",
    "Objective:",
    "Evaluate how causes, contributing factors, and preliminary prevention",
    "themes vary by injury severity and time period.",
    "",
    "Temporal results represent changes in narrative-category shares.",
    "They are not exposure-adjusted injury-incidence rates.",
]

figure.text(
    0.10,
    0.59,
    "\n".join(cover_lines),
    va="top",
    ha="left",
    fontsize=9.0,
    family="monospace",
    linespacing=1.24,
)

save_page(figure)


add_text_pages(
    "Executive Summary",
    [
        "Block 2 applies the frozen Block 1c model and ontology to 2,000",
        "narratives that were not used during ontology development.",
        "",
        "The sample contains 400 records from each of five injury-severity",
        "groups and is distributed across early, middle, and recent periods.",
        "",
        "The model assigns a record type, primary cause, contributing factor,",
        "prevention theme, evidence excerpts, confidence values, and support",
        "levels.",
        "",
        "A deterministic safeguard removes contributing factors that lack",
        "supporting narrative evidence. The factor is replaced with",
        "NO_SEPARATE_FACTOR_STATED and the prevention theme is changed to",
        "INVESTIGATION_REQUIRED.",
        "",
        "Cause analysis includes valid causes even when no separate factor is",
        "supported. Factor and prevention analyses use only records valid for",
        "full analysis.",
        "",
        "Within-group percentages are used for severity and time comparisons.",
        "Chi-square tests and Cramer's V summarize differences in category",
        "distributions but do not establish causation.",
    ],
)


sample_by_severity_period = (
    pd.crosstab(
        block2_df["SEVERITY_GROUP"],
        block2_df["TIME_PERIOD"],
    )
    .reindex(
        SEVERITY_ORDER,
        fill_value=0,
    )
    .reset_index()
)


add_table_pages(
    "Sample by Severity and Time Period",
    sample_by_severity_period,
    rows_per_page=10,
    font_size=7.4,
    column_wrap_width=28,
    landscape=True,
)


add_table_pages(
    "Processing and Quality Summary",
    quality_summary,
    rows_per_page=22,
    font_size=7.2,
    column_wrap_width=42,
)


add_table_pages(
    "Record-Type Distribution",
    record_type_summary,
    rows_per_page=12,
    font_size=7.4,
    column_wrap_width=40,
)


add_table_pages(
    "Analysis-Status Distribution",
    analysis_status_summary,
    rows_per_page=12,
    font_size=7.4,
    column_wrap_width=42,
)


add_table_pages(
    "Critical Review Reasons",
    critical_review_summary,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=50,
)


add_table_pages(
    "Advisory Review Reasons",
    advisory_review_summary,
    rows_per_page=20,
    font_size=7.2,
    column_wrap_width=50,
)


add_table_pages(
    "Primary Cause Distribution",
    cause_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=40,
)


add_bar_chart(
    "Most Common Primary Causes",
    cause_summary,
    "Primary cause",
    "#2E6F9E",
    TOP_CATEGORIES_TO_PLOT,
)


add_table_pages(
    "Supported Contributing-Factor Distribution",
    factor_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=40,
)


add_bar_chart(
    "Most Common Supported Contributing Factors",
    factor_summary,
    "Contributing factor",
    "#4E8B57",
    TOP_CATEGORIES_TO_PLOT,
)


add_table_pages(
    "Supported Prevention-Theme Distribution",
    prevention_summary,
    rows_per_page=24,
    font_size=7.1,
    column_wrap_width=40,
)


add_bar_chart(
    "Most Common Supported Prevention Themes",
    prevention_summary,
    "Prevention theme",
    "#B46A3C",
    TOP_CATEGORIES_TO_PLOT,
)


add_text_pages(
    "Severity Comparison",
    [
        "The following heatmaps show percentages within each severity group.",
        "",
        "Cause results use the cause-analysis population.",
        "",
        "Factor and prevention results use only the full-analysis population.",
        "",
        "Observed differences are associations and do not establish that a",
        "specific cause or factor produces a particular severity.",
    ],
)


add_percentage_heatmap(
    "Primary Causes by Injury Severity",
    cause_severity_pct,
    top_causes,
    "Primary cause",
    "Injury-severity group",
)


add_percentage_heatmap(
    "Supported Factors by Injury Severity",
    factor_severity_pct,
    top_factors,
    "Contributing factor",
    "Injury-severity group",
)


add_percentage_heatmap(
    "Supported Prevention Themes by Injury Severity",
    prevention_severity_pct,
    top_preventions,
    "Prevention theme",
    "Injury-severity group",
)


add_text_pages(
    "Time-Trend Interpretation",
    [
        "Time comparisons use within-period percentages.",
        "",
        "An increasing share means that a category represented a larger",
        "percentage of recent analyzed narratives than early narratives.",
        "",
        "A declining share means that a category represented a smaller",
        "percentage.",
        "",
        "These results do not account for employee-hours, operating-mine count,",
        "production, contractor activity, sector composition, or reporting",
        "changes. They must not be described as exposure-adjusted incidence",
        "rates.",
    ],
)


add_percentage_heatmap(
    "Primary Causes by Time Period",
    cause_period_pct,
    top_causes,
    "Primary cause",
    "Time period",
)


add_percentage_heatmap(
    "Supported Factors by Time Period",
    factor_period_pct,
    top_factors,
    "Contributing factor",
    "Time period",
)


add_percentage_heatmap(
    "Supported Prevention Themes by Time Period",
    prevention_period_pct,
    top_preventions,
    "Prevention theme",
    "Time period",
)


add_trend_chart(
    "Leading Primary-Cause Trends",
    cause_period_pct,
    top_causes[:8],
)


add_trend_chart(
    "Leading Contributing-Factor Trends",
    factor_period_pct,
    top_factors[:8],
)


add_trend_chart(
    "Leading Prevention-Theme Trends",
    prevention_period_pct,
    top_preventions[:8],
)


add_table_pages(
    "Early-to-Recent Cause Changes",
    cause_change_table,
    rows_per_page=24,
    font_size=6.9,
    column_wrap_width=32,
    landscape=True,
)


add_table_pages(
    "Early-to-Recent Factor Changes",
    factor_change_table,
    rows_per_page=24,
    font_size=6.9,
    column_wrap_width=32,
    landscape=True,
)


add_table_pages(
    "Early-to-Recent Prevention Changes",
    prevention_change_table,
    rows_per_page=24,
    font_size=6.9,
    column_wrap_width=32,
    landscape=True,
)


add_text_pages(
    "Statistical Comparison Method",
    [
        "Chi-square tests compare category distributions among severity groups",
        "and time periods.",
        "",
        "Cramer's V summarizes association strength from zero to one.",
        "",
        "Statistical significance should be interpreted with category counts,",
        "percentage differences, effect size, and practical importance.",
        "",
        "These tests do not establish causation and do not adjust for exposure",
        "or changing industry composition.",
    ],
)


add_table_pages(
    "Chi-Square and Cramer's V Results",
    association_results,
    rows_per_page=12,
    font_size=7.0,
    column_wrap_width=34,
    landscape=True,
)


add_relationship_heatmap(
    "Common Cause and Contributing-Factor Combinations",
    cause_factor_counts,
)


add_relationship_heatmap(
    "Common Factor and Prevention-Theme Combinations",
    factor_prevention_counts,
)


add_text_pages(
    "Quality-Sensitivity Analysis",
    [
        "Cause distributions are compared across three populations:",
        "",
        "1. All successfully processed acute-accident records.",
        "",
        "2. Cause-valid records, including conservative no-factor cases.",
        "",
        "3. Full-analysis records with supported factors and aligned prevention",
        "   themes.",
        "",
        "The maximum percentage spread shows how much a cause's estimated share",
        "changes under stricter quality filters.",
    ],
)


add_table_pages(
    "Cause Sensitivity to Quality Filters",
    sensitivity_display,
    rows_per_page=24,
    font_size=6.8,
    column_wrap_width=29,
    landscape=True,
)


add_example_pages(
    "Representative Fully Validated Records",
    representative_examples,
    (
        "These records met the full-analysis requirements with valid cause "
        "evidence, supported factor evidence, aligned prevention, and no "
        "critical review condition."
    ),
)


add_example_pages(
    "Cause-Valid Records with No Separate Factor",
    cause_only_examples,
    (
        "These narratives support the event mechanism but do not separately "
        "support why the event occurred."
    ),
)


add_example_pages(
    "Unsupported-Factor Safeguard Examples",
    safeguard_examples,
    (
        "The original factor lacked supporting narrative evidence. The factor "
        "was conservatively replaced with NO_SEPARATE_FACTOR_STATED and the "
        "prevention theme was changed to INVESTIGATION_REQUIRED."
    ),
)


add_example_pages(
    "Excluded and Critical-Review Examples",
    critical_examples,
    (
        "These records contain an unresolved processing, routing, schema, "
        "cause-evidence, or alignment condition and are excluded from primary "
        "statistical conclusions."
    ),
)


add_text_pages(
    "Interpretation and Limitations",
    [
        "1. The sample is balanced by severity. It does not represent the",
        "   natural severity distribution of the complete MSHA dataset.",
        "",
        "2. Time trends describe narrative-category shares, not exposure-",
        "   adjusted accident rates.",
        "",
        "3. The primary cause represents the reported event mechanism, not a",
        "   verified root cause.",
        "",
        "4. Prevention themes are preliminary analytical categories, not",
        "   approved corrective actions.",
        "",
        "5. Short narratives may describe what occurred without explaining why.",
        "",
        "6. NO_SEPARATE_FACTOR_STATED is a valid result when a factor is not",
        "   supported.",
        "",
        "7. Block 3 will run the frozen system on 100 additional unseen",
        "   narratives and create the independent human-review workbook.",
        "",
        "8. Block 4 will compare the completed human labels with the model",
        "   labels and generate the final validation report.",
    ],
)


add_text_pages(
    "Technical Appendix",
    [
        f"Model: {MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"Unseen records processed: {len(results_df):,}",
        f"Block 1 IDs excluded: {len(block1_record_ids):,}",
        f"Target per severity group: {TARGET_PER_SEVERITY}",
        f"Random state: {RANDOM_STATE}",
        f"Maximum retries: {MAX_RETRIES}",
        f"Maximum generated tokens: {MAX_NEW_TOKENS}",
        f"Minimum narrative words: {MIN_NARRATIVE_WORDS}",
        f"Truncation threshold: {TRUNCATION_CHARACTER_THRESHOLD}",
        "",
        f"PDF: {os.path.abspath(OUTPUT_PDF)}",
        f"Results CSV: {os.path.abspath(BLOCK2_RESULTS_CSV)}",
        f"Checkpoint: {os.path.abspath(BLOCK2_CHECKPOINT_JSONL)}",
        f"Block 2 IDs: {os.path.abspath(BLOCK2_IDS_JSON)}",
    ],
    font_size=8.3,
)


# =============================================================================
# 46. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()

metadata["Title"] = (
    "MSHA Block 2 Unseen Narrative Analysis Report"
)

metadata["Author"] = (
    "Python and Qwen2.5-7B-Instruct"
)

metadata["Subject"] = (
    "Causes, contributing factors, prevention themes, severity "
    "comparisons, and temporal changes in unseen MSHA narratives"
)

metadata["Keywords"] = (
    "MSHA, NLP, Qwen, accident narratives, causes, contributing "
    "factors, prevention themes, severity, time trends"
)

metadata["CreationDate"] = datetime.now()


pdf.close()


# =============================================================================
# 47. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 2 STANDALONE ANALYSIS COMPLETE")
print("=" * 95)

print(
    f"Unseen narratives processed: "
    f"{len(results_df):,}"
)

print(
    f"Successful classifications: "
    f"{int(results_df['PROCESSING_STATUS'].eq('SUCCESS').sum()):,}"
)

print(
    f"Schema-valid classifications: "
    f"{int(results_df['SCHEMA_VALID'].sum()):,}"
)

print(
    f"Unsupported-factor safeguards: "
    f"{int(results_df['UNSUPPORTED_FACTOR_SAFEGUARD_APPLIED'].sum()):,}"
)

print(
    f"Valid for full analysis: "
    f"{len(full_analysis_df):,}"
)

print(
    f"Valid for cause analysis only: "
    f"{int(results_df['ANALYSIS_STATUS'].eq('VALID_FOR_CAUSE_ANALYSIS_ONLY').sum()):,}"
)

print(
    f"Excluded or review required: "
    f"{len(review_df):,}"
)

print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)

print(
    f"Structured results: "
    f"{os.path.abspath(BLOCK2_RESULTS_CSV)}"
)

print(
    f"Checkpoint: "
    f"{os.path.abspath(BLOCK2_CHECKPOINT_JSONL)}"
)

print(
    f"Block 2 IDs: "
    f"{os.path.abspath(BLOCK2_IDS_JSON)}"
)

print("=" * 95)


# =============================================================================
# 48. CONCISE NOTEBOOK DISPLAYS
# =============================================================================

display(quality_summary)

display(analysis_status_summary)

display(cause_summary)

display(factor_summary)

display(prevention_summary)

display(association_results)

display(cause_change_table)

display(factor_change_table)

display(prevention_change_table)
````

````python
# BLOCK 2 - STANDALONE, INCREMENTAL, REVISED FACTOR-SUPPORT ANALYSIS
# Prerequisites: `accidents` DataFrame loaded; MSHA_Block1_Development_Record_IDs.json present; NVIDIA GPU.

import os, re, gc, sys, json, time, math, textwrap, warnings
from pathlib import Path
from datetime import datetime
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
from pydantic import BaseModel, Field, model_validator
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig
from matplotlib.backends.backend_pdf import PdfPages
from scipy.stats import chi2_contingency
warnings.filterwarnings("ignore")

# ------------------------------- SETTINGS ------------------------------------
MODEL_ID="Qwen/Qwen2.5-7B-Instruct"
PROMPT_VERSION="MSHA_BLOCK1C_FROZEN_FOR_BLOCK2_V3.2_EXPANDED_SUPPORT"
TEXT_COLUMN="NARRATIVE"; DATE_COLUMN="ACCIDENT_DT"; SEVERITY_COLUMN="DEGREE_INJURY"
ID_COLUMN_CANDIDATES=["DOCUMENT_NO","DOC_NO","ACCIDENT_ID"]
TOTAL_SAMPLE_SIZE=2000                 # Increase later; prior IDs are preserved.
TARGET_PER_SEVERITY=TOTAL_SAMPLE_SIZE//5
RANDOM_STATE=84; MIN_NARRATIVE_WORDS=8; TRUNCATION_CHARACTER_THRESHOLD=383
MAX_NEW_TOKENS=750; MAX_RETRIES=4; LOW_CONFIDENCE_THRESHOLD=.70
REPROCESS_COMPLETED=False; TOP_CATEGORIES_TO_PLOT=12; MIN_CATEGORY_COUNT_FOR_TREND=10
BLOCK1_IDS_JSON="MSHA_Block1_Development_Record_IDs.json"
PRIOR_IDS_JSON="MSHA_Block2_Unseen_Record_IDs.json"
REVISED_IDS_JSON="MSHA_Block2_Revised_Record_IDs.json"
CHECKPOINT_JSONL="MSHA_Block2_Unseen_Checkpoint.jsonl"
PRIOR_RESULTS_CSV="MSHA_Block2_Unseen_Results.csv"
REVISED_RESULTS_CSV="MSHA_Block2_Expanded_Results.csv"
OUTPUT_PDF="MSHA_Block2_Expanded_Analysis_Report.pdf"
SEVERITY_ORDER=["Fatality","Permanent disability","Days away","Restricted or transferred","No lost time or other"]
EARLY="Early: 2000-2008"; MIDDLE="Middle: 2009-2017"; RECENT="Recent: 2018-2026"
TIME_ORDER=[EARLY,MIDDLE,RECENT]

if "accidents" not in globals() or not isinstance(accidents,pd.DataFrame) or accidents.empty:
    raise ValueError("Load a nonempty pandas DataFrame named 'accidents' before running Block 2.")
if not Path(BLOCK1_IDS_JSON).exists(): raise FileNotFoundError(BLOCK1_IDS_JSON)

# ---------------------------- FROZEN ONTOLOGY --------------------------------
RECORD_TYPES=["ACUTE_ACCIDENT","OCCUPATIONAL_EXPOSURE_OR_ILLNESS","NONWORK_RELATED_OR_NATURAL_EVENT","NONINJURY_REPORT","INSUFFICIENT_EVENT_INFORMATION"]
CAUSES=["SLIP_TRIP_OR_FALL","AWKWARD_MOVEMENT_OR_BODY_MOTION","STRUCK_BY_MOVING_OBJECT","STRUCK_AGAINST_STATIONARY_OBJECT","FOREIGN_BODY_OR_PARTICLE_CONTACT","CAUGHT_IN_OR_BETWEEN","MATERIAL_RELEASE_OR_SHIFT","GROUND_OR_ROOF_FAILURE","UNEXPECTED_EQUIPMENT_MOVEMENT","LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT","MANUAL_HANDLING_OR_OVEREXERTION","PRESSURE_OR_STORED_ENERGY_RELEASE","ELECTRICAL_CONTACT","FIRE_EXPLOSION_OR_THERMAL_EVENT","VEHICLE_OR_MOBILE_EQUIPMENT_EVENT","HEAT_OR_ENVIRONMENTAL_EXPOSURE","CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE","ERGONOMIC_OR_REPETITIVE_EXPOSURE","OTHER","INSUFFICIENT_INFORMATION"]
FACTORS=["UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL","INADEQUATE_SECUREMENT_OR_RESTRAINT","LINE_OF_FIRE_BODY_POSITION","POOR_HOUSEKEEPING_OR_SURFACE_CONDITION","UNSTABLE_GROUND_OR_ROOF_CONDITION","INADEQUATE_GUARDING_OR_BARRIER","ENERGY_ISOLATION_DEFICIENCY","EQUIPMENT_CONDITION_OR_FAILURE","PROCEDURE_OR_WORK_METHOD","INADEQUATE_HAZARD_RECOGNITION","COMMUNICATION_OR_COORDINATION","TRAINING_OR_EXPERIENCE","PPE_OR_PROTECTIVE_CONTROL","ENVIRONMENTAL_CONDITION","WORKSPACE_OR_ACCESS_LIMITATION","OTHER","NO_SEPARATE_FACTOR_STATED"]
PREVENTIONS=["SECURE_OR_REPOSITION_MATERIAL","ESTABLISH_LINE_OF_FIRE_EXCLUSION","ISOLATE_OR_CONTROL_STORED_ENERGY","REPAIR_REPLACE_OR_INSPECT_EQUIPMENT","INSTALL_OR_IMPROVE_GUARDING","IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING","IMPROVE_GROUND_CONTROL_OR_SCALING","REVISE_PROCEDURE_OR_JOB_PLAN","CONDUCT_PRE_JOB_HAZARD_REVIEW","PROVIDE_TARGETED_TRAINING","IMPROVE_COMMUNICATION_OR_SPOTTER_USE","USE_MECHANICAL_HANDLING_ASSISTANCE","IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT","CONTROL_ENVIRONMENTAL_EXPOSURE","REDESIGN_WORKSPACE_OR_ACCESS","ENGINEERING_REVIEW_REQUIRED","INVESTIGATION_REQUIRED","INSUFFICIENT_INFORMATION"]
PREFERRED={
"UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":"SECURE_OR_REPOSITION_MATERIAL","INADEQUATE_SECUREMENT_OR_RESTRAINT":"SECURE_OR_REPOSITION_MATERIAL","LINE_OF_FIRE_BODY_POSITION":"ESTABLISH_LINE_OF_FIRE_EXCLUSION","POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":"IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING","UNSTABLE_GROUND_OR_ROOF_CONDITION":"IMPROVE_GROUND_CONTROL_OR_SCALING","INADEQUATE_GUARDING_OR_BARRIER":"INSTALL_OR_IMPROVE_GUARDING","ENERGY_ISOLATION_DEFICIENCY":"ISOLATE_OR_CONTROL_STORED_ENERGY","EQUIPMENT_CONDITION_OR_FAILURE":"REPAIR_REPLACE_OR_INSPECT_EQUIPMENT","PROCEDURE_OR_WORK_METHOD":"REVISE_PROCEDURE_OR_JOB_PLAN","INADEQUATE_HAZARD_RECOGNITION":"CONDUCT_PRE_JOB_HAZARD_REVIEW","COMMUNICATION_OR_COORDINATION":"IMPROVE_COMMUNICATION_OR_SPOTTER_USE","TRAINING_OR_EXPERIENCE":"PROVIDE_TARGETED_TRAINING","PPE_OR_PROTECTIVE_CONTROL":"IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT","ENVIRONMENTAL_CONDITION":"CONTROL_ENVIRONMENTAL_EXPOSURE","WORKSPACE_OR_ACCESS_LIMITATION":"REDESIGN_WORKSPACE_OR_ACCESS","NO_SEPARATE_FACTOR_STATED":"INVESTIGATION_REQUIRED"}
ALLOWED={k:{v} for k,v in PREFERRED.items()}
ALLOWED.update({
"UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":{PREFERRED["UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL"],"ESTABLISH_LINE_OF_FIRE_EXCLUSION","REVISE_PROCEDURE_OR_JOB_PLAN","ENGINEERING_REVIEW_REQUIRED"},
"LINE_OF_FIRE_BODY_POSITION":{"ESTABLISH_LINE_OF_FIRE_EXCLUSION","REVISE_PROCEDURE_OR_JOB_PLAN","CONDUCT_PRE_JOB_HAZARD_REVIEW","IMPROVE_COMMUNICATION_OR_SPOTTER_USE"},
"POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":{"IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING","REVISE_PROCEDURE_OR_JOB_PLAN","REDESIGN_WORKSPACE_OR_ACCESS"},
"UNSTABLE_GROUND_OR_ROOF_CONDITION":{"IMPROVE_GROUND_CONTROL_OR_SCALING","ESTABLISH_LINE_OF_FIRE_EXCLUSION","ENGINEERING_REVIEW_REQUIRED"},
"EQUIPMENT_CONDITION_OR_FAILURE":{"REPAIR_REPLACE_OR_INSPECT_EQUIPMENT","ISOLATE_OR_CONTROL_STORED_ENERGY","ENGINEERING_REVIEW_REQUIRED"},
"PROCEDURE_OR_WORK_METHOD":{"REVISE_PROCEDURE_OR_JOB_PLAN","CONDUCT_PRE_JOB_HAZARD_REVIEW","USE_MECHANICAL_HANDLING_ASSISTANCE","ESTABLISH_LINE_OF_FIRE_EXCLUSION"},
"OTHER":{"ENGINEERING_REVIEW_REQUIRED","INVESTIGATION_REQUIRED","REVISE_PROCEDURE_OR_JOB_PLAN"},
"NO_SEPARATE_FACTOR_STATED":{"INVESTIGATION_REQUIRED","INSUFFICIENT_INFORMATION"}})
ALIASES={"UNSTATED":"NO_SEPARATE_FACTOR_STATED","NOT_STATED":"NO_SEPARATE_FACTOR_STATED","NONE_STATED":"NO_SEPARATE_FACTOR_STATED","NO_FACTOR_STATED":"NO_SEPARATE_FACTOR_STATED","CAUGHT_BETWEEN":"CAUGHT_IN_OR_BETWEEN","CAUGHT_IN_BETWEEN":"CAUGHT_IN_OR_BETWEEN","STRUCK_BY_OBJECT":"STRUCK_BY_MOVING_OBJECT","LOSS_OF_CONTROL":"LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT","UNEXPECTED_MACHINE_MOVEMENT":"UNEXPECTED_EQUIPMENT_MOVEMENT","FOREIGN_OBJECT_CONTACT":"FOREIGN_BODY_OR_PARTICLE_CONTACT","HEAT_EXPOSURE":"HEAT_OR_ENVIRONMENTAL_EXPOSURE","SLIP_TRIP_FALL":"SLIP_TRIP_OR_FALL","REVISE_PROCEDURE":"REVISE_PROCEDURE_OR_JOB_PLAN","PRE_JOB_HAZARD_REVIEW":"CONDUCT_PRE_JOB_HAZARD_REVIEW","EQUIPMENT_REPAIR_OR_INSPECTION":"REPAIR_REPLACE_OR_INSPECT_EQUIPMENT","IMPROVE_HOUSEKEEPING":"IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING","UNKNOWN":"INSUFFICIENT_INFORMATION","INSUFFICIENT_INFO":"INSUFFICIENT_INFORMATION"}

FACTOR_TERMS={
"UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL":["unstable","unsupported","loose material","load shifted","shifted","rolling","sliding","fell over","falling over","tipped over","turned over","improperly positioned"],
"INADEQUATE_SECUREMENT_OR_RESTRAINT":["not secured","unsecured","restraint failed","strap broke","chain broke","not blocked","not braced","pin removed","safety pin","pulled out of hooks"],
"LINE_OF_FIRE_BODY_POSITION":["between","in front of","under the","underneath","in the path","pinch point","reached into","hand between","arm between","foot between","attempted to catch","tried to catch","leaning over"],
"POOR_HOUSEKEEPING_OR_SURFACE_CONDITION":["ice","icy","snow","mud","muddy","wet floor","wet surface","grease","oil on","debris","uneven surface","uneven ground","poor footing","loose gravel","slippery walkway","slippery surface","damaged step","broken step","tripping hazard"],
"UNSTABLE_GROUND_OR_ROOF_CONDITION":["roof fell","roof fall","rib fell","rib fall","rock fell","slate fell","draw rock","highwall","cave in","caved in","ground failure","loose roof","unsupported roof","roof popped","bank collapsed","embankment cave","material sliding down the face"],
"INADEQUATE_GUARDING_OR_BARRIER":["guard removed","guard missing","missing guard","unguarded","guard was off","guard was open","barrier missing","bypassed guard"],
"ENERGY_ISOLATION_DEFICIENCY":["not locked out","failed to lock out","lockout","tagout","loto","failed to bleed","not bled","pressure remained","still under pressure","energized","not deenergized","not blocked","not isolated","stored energy","engine running","without shutting off"],
"EQUIPMENT_CONDITION_OR_FAILURE":["failed","failure","broke","broken","blew off","blew out","malfunction","defective","damaged","worn","leak","leaking","not working","catastrophically failed","mechanical failure","component failed","hose failed"],
"PROCEDURE_OR_WORK_METHOD":["failed to follow procedure","procedure was not followed","improper method","work method","task sequence","using his foot","using her foot"],
"INADEQUATE_HAZARD_RECOGNITION":["did not recognize","did not realize","was unaware","unaware of","failed to identify","hazard not identified","failed to examine"],
"COMMUNICATION_OR_COORDINATION":["did not communicate","failed to communicate","no communication","not notified","failed to notify","did not signal","miscommunication","no spotter","without a spotter"],
"TRAINING_OR_EXPERIENCE":["not trained","inadequately trained","lack of training","new employee","inexperienced","lack of experience","not qualified","unqualified","task training"],
"PPE_OR_PROTECTIVE_CONTROL":["not wearing","wasn't wearing","without wearing","no safety glasses","no eye protection","no gloves","no hard hat","no seat belt","not using seat belt","no life jacket","ppe","protective equipment"],
"ENVIRONMENTAL_CONDITION":["heat","hot","temperature","degrees","heat stress","heat exhaustion","dehydration","cold","snow","ice","wind","weather","visibility","lighting","dark","fog","rain","storm"],
"WORKSPACE_OR_ACCESS_LIMITATION":["restricted clearance","limited clearance","tight space","confined space","limited access","difficult access","awkward access","poor access","restricted access","reaching over","excessive reach","limited workspace","narrow area","crowded area"]}

# -------------------------- SCHEMA AND UTILITIES ------------------------------
class IncidentExtraction(BaseModel):
    record_type:str; primary_cause:str; primary_contributing_factor:str; prevention_theme:str
    cause_evidence_text:str=""; factor_evidence_text:str=""; prevention_rationale:str=""
    cause_confidence:float=Field(default=0,ge=0,le=1); factor_confidence:float=Field(default=0,ge=0,le=1); prevention_confidence:float=Field(default=0,ge=0,le=1)
    cause_support_level:str="LOW"; factor_support_level:str="LOW"; prevention_support_level:str="LOW"
    insufficient_information_reason:str=""; reviewer_note:str=""
    @model_validator(mode="after")
    def check(self):
        if self.record_type not in RECORD_TYPES: raise ValueError("record_type")
        if self.primary_cause not in CAUSES: raise ValueError("primary_cause")
        if self.primary_contributing_factor not in FACTORS: raise ValueError("primary_contributing_factor")
        if self.prevention_theme not in PREVENTIONS: raise ValueError("prevention_theme")
        if any(getattr(self,x) not in {"HIGH","MEDIUM","LOW"} for x in ["cause_support_level","factor_support_level","prevention_support_level"]): raise ValueError("support_level")
        return self

def tok(x): return str(x).strip().upper().replace("-","_").replace(" ","_").replace("/","_")
def cat(x):
    y=tok(x); return ALIASES.get(y,y)
def sf(x,d=0.0):
    try:return min(max(float(x),0),1)
    except:return d
def norm_text(x): return re.sub(r"\s+"," ",re.sub(r"[^a-z0-9\s']"," ",str(x).lower().replace("’","'"))).strip()
def extract_json(text):
    text=re.sub(r"^```(?:json)?|```$","",str(text).strip(),flags=re.I); start=text.find("{")
    if start<0: raise ValueError("No JSON object")
    depth=0; ins=False; esc=False
    for i,ch in enumerate(text[start:],start):
        if esc: esc=False; continue
        if ch=="\\" and ins: esc=True; continue
        if ch=='"': ins=not ins; continue
        if ins: continue
        if ch=="{": depth+=1
        elif ch=="}":
            depth-=1
            if depth==0:return text[start:i+1]
    candidate=text[start:]; return candidate+"}"*max(0,candidate.count("{")-candidate.count("}"))
def repair_json(x):
    x=str(x).replace("“",'"').replace("”",'"').replace("’","'")
    x=re.sub(r",\s*}","}",x); x=re.sub(r":\s*None\b",': ""',x); x=re.sub(r":\s*True\b",": true",x); return re.sub(r":\s*False\b",": false",x)
DEFAULT={"record_type":"INSUFFICIENT_EVENT_INFORMATION","primary_cause":"INSUFFICIENT_INFORMATION","primary_contributing_factor":"NO_SEPARATE_FACTOR_STATED","prevention_theme":"INVESTIGATION_REQUIRED","cause_evidence_text":"","factor_evidence_text":"","prevention_rationale":"","cause_confidence":0,"factor_confidence":0,"prevention_confidence":0,"cause_support_level":"LOW","factor_support_level":"LOW","prevention_support_level":"LOW","insufficient_information_reason":"","reviewer_note":""}
def normalize_dict(d):
    z={**DEFAULT,**{k:d[k] for k in DEFAULT if k in d}}
    z["record_type"]=tok(z["record_type"])
    for k in ["primary_cause","primary_contributing_factor","prevention_theme"]:z[k]=cat(z[k])
    for k in ["cause_support_level","factor_support_level","prevention_support_level"]:z[k]=tok(z[k]) if tok(z[k]) in {"HIGH","MEDIUM","LOW"} else "LOW"
    for k in ["cause_confidence","factor_confidence","prevention_confidence"]:z[k]=sf(z[k])
    return z

def evidence_metrics(e,n):
    e0=str(e).strip(); nn=norm_text(n); en=norm_text(e)
    exact=bool(e0) and e0.lower() in str(n).lower(); normalized=bool(en) and en in nn
    et=set(en.split()); nt=set(nn.split()); overlap=len(et&nt)/len(et) if et else 0
    return exact or normalized or overlap>=.85,exact,normalized,overlap
def aligned(f,p):return p in ALLOWED.get(f,{p})
def phrase_hits(f,n):
    nn=norm_text(n); return [t for t in FACTOR_TERMS.get(f,[]) if norm_text(t) in nn]

CAUSE_HIERARCHY="""Use the first clearly supported immediate mechanism: CAUGHT_IN_OR_BETWEEN for trapped/pinched/crushed between objects; VEHICLE_OR_MOBILE_EQUIPMENT_EVENT for collision/overturn/runaway/backing; GROUND_OR_ROOF_FAILURE for roof/rib/highwall/ground failure; PRESSURE_OR_STORED_ENERGY_RELEASE for released pressure/tension; LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT for explicit loss of control; UNEXPECTED_EQUIPMENT_MOVEMENT for unexpected movement without worker loss of control; STRUCK_BY_MOVING_OBJECT for contact without trapping; FOREIGN_BODY_OR_PARTICLE_CONTACT for small particles; SLIP_TRIP_OR_FALL for loss of footing/balance/elevation; AWKWARD_MOVEMENT_OR_BODY_MOTION for acute motion without fall/load; MANUAL_HANDLING_OR_OVEREXERTION for acute strain/hernia from exertion. Do not use struck-by as a general fallback."""
SYSTEM_PROMPT=f"""Extract structured information from an MSHA narrative. This is reported-event extraction, not a verified root-cause analysis.
Allowed record types: {RECORD_TYPES}
Allowed causes: {CAUSES}
Allowed factors: {FACTORS}
Allowed prevention themes: {PREVENTIONS}
{CAUSE_HIERARCHY}
Factor rules: unstable material requires explicit instability; securement requires missing/failed restraint; housekeeping requires an explicit walking/working-surface hazard; line of fire requires body position in the path/pinch point; equipment failure requires stated failure/damage/malfunction; no separate factor stated applies when the event is clear but why it happened is not supported. Factor evidence must support the factor, not the injury outcome.
Prevention follows the supported factor. NO_SEPARATE_FACTOR_STATED maps to INVESTIGATION_REQUIRED. Copy short verbatim evidence when possible. HIGH=explicit, MEDIUM=strongly implied, LOW=ambiguous.
Return only one JSON object with keys: {list(DEFAULT)}"""

# ------------------------------ MODEL ----------------------------------------
# Model loads only if at least one selected record needs inference.
tokenizer=None; model=None
def load_model():
    global tokenizer,model
    if model is not None:return
    if not torch.cuda.is_available():raise RuntimeError("NVIDIA GPU required for new inference")
    q=BitsAndBytesConfig(load_in_4bit=True,bnb_4bit_quant_type="nf4",bnb_4bit_use_double_quant=True,bnb_4bit_compute_dtype=torch.float16)
    tokenizer=AutoTokenizer.from_pretrained(MODEL_ID,use_fast=True,trust_remote_code=False)
    if tokenizer.pad_token_id is None:tokenizer.pad_token=tokenizer.eos_token
    model=AutoModelForCausalLM.from_pretrained(MODEL_ID,quantization_config=q,device_map="auto",dtype=torch.float16,low_cpu_mem_usage=True,trust_remote_code=False); model.eval()
def generate(messages):
    load_model(); inputs=tokenizer.apply_chat_template(messages,add_generation_prompt=True,tokenize=True,return_dict=True,return_tensors="pt")
    inputs={k:v.to(model.device) for k,v in inputs.items()}; n=inputs["input_ids"].shape[-1]
    with torch.inference_mode():out=model.generate(**inputs,max_new_tokens=MAX_NEW_TOKENS,do_sample=False,repetition_penalty=1.02,pad_token_id=tokenizer.pad_token_id,eos_token_id=tokenizer.eos_token_id,use_cache=True)
    return tokenizer.decode(out[0,n:],skip_special_tokens=True)
def classify(narrative,hint,truncated):
    prior=""; err=""
    for attempt in range(1,MAX_RETRIES+1):
        user=f"PRELIMINARY ROUTING HINT: {hint}\nNARRATIVE MAY BE TRUNCATED: {truncated}\nNARRATIVE:\n{narrative}\nReturn only JSON."
        if attempt>1:user+=f"\nPrevious response failed: {err}\nPrevious response: {prior}\nCorrect it using only allowed values."
        try:
            raw=generate([{"role":"system","content":SYSTEM_PROMPT},{"role":"user","content":user}]); prior=raw
            val=IncidentExtraction.model_validate(normalize_dict(json.loads(repair_json(extract_json(raw)))))
            cv,ce,cn,co=evidence_metrics(val.cause_evidence_text,narrative); fv,fe,fn,fo=evidence_metrics(val.factor_evidence_text,narrative)
            if val.primary_contributing_factor=="NO_SEPARATE_FACTOR_STATED":fv=True
            return {**val.model_dump(),"SCHEMA_VALID":True,"CAUSE_EVIDENCE_VALID":cv,"CAUSE_EVIDENCE_EXACT":ce,"CAUSE_EVIDENCE_NORMALIZED":cn,"CAUSE_EVIDENCE_TOKEN_OVERLAP":round(co,4),"FACTOR_EVIDENCE_VALID":fv,"FACTOR_EVIDENCE_EXACT":fe,"FACTOR_EVIDENCE_NORMALIZED":fn,"FACTOR_EVIDENCE_TOKEN_OVERLAP":round(fo,4),"FACTOR_PREVENTION_ALIGNED":aligned(val.primary_contributing_factor,val.prevention_theme),"PROCESSING_STATUS":"SUCCESS","PROCESSING_ERROR":"","ATTEMPT_COUNT":attempt,"RAW_MODEL_RESPONSE":raw}
        except Exception as e:err=f"{type(e).__name__}: {e}"; gc.collect(); torch.cuda.empty_cache()
    return {**DEFAULT,"SCHEMA_VALID":False,"CAUSE_EVIDENCE_VALID":False,"FACTOR_EVIDENCE_VALID":False,"FACTOR_PREVENTION_ALIGNED":False,"PROCESSING_STATUS":"FAILED","PROCESSING_ERROR":err,"ATTEMPT_COUNT":MAX_RETRIES,"RAW_MODEL_RESPONSE":prior}

# ------------------------- DATA PREPARATION ----------------------------------
df=accidents.copy(); df.columns=df.columns.astype(str).str.strip().str.upper()
if TEXT_COLUMN not in df or SEVERITY_COLUMN not in df:raise KeyError("NARRATIVE and DEGREE_INJURY required")
id_col=next((x for x in ID_COLUMN_CANDIDATES if x in df),None)
if id_col is None:id_col="_SOURCE_INDEX";df[id_col]=df.index.astype(str)
df[id_col]=df[id_col].astype(str);df[TEXT_COLUMN]=df[TEXT_COLUMN].astype("string").fillna("").str.strip();df[SEVERITY_COLUMN]=df[SEVERITY_COLUMN].astype("string").fillna("UNKNOWN").str.strip()
if DATE_COLUMN in df:df[DATE_COLUMN]=pd.to_datetime(df[DATE_COLUMN],errors="coerce");df["ACCIDENT_YEAR"]=df[DATE_COLUMN].dt.year.astype("Int64")
elif "CAL_YR" in df:df["ACCIDENT_YEAR"]=pd.to_numeric(df["CAL_YR"],errors="coerce").astype("Int64")
else:raise KeyError("ACCIDENT_DT or CAL_YR required")
df["_WORD_COUNT"]=df[TEXT_COLUMN].str.split().str.len();df["_CHARACTER_COUNT"]=df[TEXT_COLUMN].str.len();df["_MAY_BE_TRUNCATED"]=df["_CHARACTER_COUNT"]>=TRUNCATION_CHARACTER_THRESHOLD
df["_DUPKEY"]=df[TEXT_COLUMN].str.lower().str.replace(r"\s+"," ",regex=True).str.strip();df=df[df["_WORD_COUNT"]>=MIN_NARRATIVE_WORDS].drop_duplicates("_DUPKEY").copy()
with open(BLOCK1_IDS_JSON,encoding="utf-8") as f:block1_ids=set(map(str,json.load(f)["record_ids"]))
df=df[~df[id_col].isin(block1_ids)].copy()
def severity(x):
    x=str(x).upper()
    if "FATAL" in x:return "Fatality"
    if any(t in x for t in ["PERM TOT","PERMANENT TOTAL","PERM PART","PERMANENT PART"]):return "Permanent disability"
    if any(t in x for t in ["DAYS AWAY","DYS AWY","LOST WORKDAY","LOST TIME"]):return "Days away"
    if any(t in x for t in ["RESTRICT","TRANSFER","RSTR ACT","RESTRCTD ACT"]):return "Restricted or transferred"
    return "No lost time or other"
def period(y):
    if pd.isna(y):return "Unknown period"
    return EARLY if int(y)<=2008 else MIDDLE if int(y)<=2017 else RECENT
def route(r):
    x=" ".join(str(r.get(c,"")).upper() for c in [TEXT_COLUMN,"CLASSIFICATION","NATURE_INJURY",SEVERITY_COLUMN])
    if any(t in x for t in ["NOT DUE TO WORK","NOT WORK RELATED","NON-OCCUPATIONAL","NATURAL CAUSE","NOT A MINER","HEART ATTACK","CARDIAC","STROKE","SEIZURE"]):return "NONWORK_RELATED_OR_NATURAL_EVENT"
    if any(t in x for t in ["HEARING LOSS","SILICOSIS","PNEUMOCONIOSIS","DUST DISEASE","OCCUPATIONAL ILLNESS"]):return "OCCUPATIONAL_EXPOSURE_OR_ILLNESS"
    return "ACUTE_ACCIDENT"
df["SEVERITY_GROUP"]=df[SEVERITY_COLUMN].map(severity);df["TIME_PERIOD"]=df["ACCIDENT_YEAR"].map(period);df["PRELIMINARY_RECORD_TYPE_HINT"]=df.apply(route,axis=1)
candidates=df[df["PRELIMINARY_RECORD_TYPE_HINT"].eq("ACUTE_ACCIDENT")].copy()

# -------------------- PRESERVE PRIOR IDS / EXPAND SAMPLE ---------------------
def read_ids(path):
    if not Path(path).exists():return []
    with open(path,encoding="utf-8") as f:return list(map(str,json.load(f).get("record_ids",[])))
prior_ids=read_ids(REVISED_IDS_JSON) or read_ids(PRIOR_IDS_JSON)
preserved=candidates[candidates[id_col].isin(prior_ids)].drop_duplicates(id_col).copy()
if len(preserved)>TOTAL_SAMPLE_SIZE:preserved=preserved.head(TOTAL_SAMPLE_SIZE)
need=TOTAL_SAMPLE_SIZE-len(preserved); selected=[preserved]
if need>0:
    pool=candidates[~candidates[id_col].isin(preserved[id_col])].copy(); pieces=[]
    base=max(1,math.ceil(need/(len(SEVERITY_ORDER)*len(TIME_ORDER))))
    for si,s in enumerate(SEVERITY_ORDER):
        for ti,t in enumerate(TIME_ORDER):
            g=pool[(pool.SEVERITY_GROUP==s)&(pool.TIME_PERIOD==t)];n=min(base,len(g))
            if n:pieces.append(g.sample(n=n,random_state=RANDOM_STATE+si*100+ti))
    add=pd.concat(pieces).drop_duplicates(id_col) if pieces else pool.iloc[0:0]
    if len(add)<need:
        remain=pool[~pool[id_col].isin(add[id_col])];add=pd.concat([add,remain.sample(n=min(need-len(add),len(remain)),random_state=RANDOM_STATE+9999)])
    selected.append(add.head(need))
block2_df=pd.concat(selected).drop_duplicates(id_col).head(TOTAL_SAMPLE_SIZE).reset_index(drop=True)
if len(block2_df)<TOTAL_SAMPLE_SIZE:raise ValueError(f"Only {len(block2_df)} eligible records available")
with open(REVISED_IDS_JSON,"w",encoding="utf-8") as f:json.dump({"created_at":datetime.now().isoformat(),"id_column":id_col,"record_count":len(block2_df),"record_ids":block2_df[id_col].astype(str).tolist()},f,indent=2)

# ----------------------- LOAD EXISTING OUTPUTS -------------------------------
def load_jsonl(path):
    out={}
    if Path(path).exists():
        with open(path,encoding="utf-8") as f:
            for line in f:
                try:r=json.loads(line);out[str(r["SOURCE_RECORD_ID"])]=r
                except:pass
    return out
existing=load_jsonl(CHECKPOINT_JSONL)
if Path(PRIOR_RESULTS_CSV).exists():
    old=pd.read_csv(PRIOR_RESULTS_CSV,low_memory=False);old["SOURCE_RECORD_ID"]=old["SOURCE_RECORD_ID"].astype(str)
    for _,r in old.iterrows():existing.setdefault(r["SOURCE_RECORD_ID"],r.to_dict())

def append_jsonl(r):
    with open(CHECKPOINT_JSONL,"a",encoding="utf-8") as f:f.write(json.dumps(r,ensure_ascii=False,default=str)+"\n")
context=[c for c in ["ACCIDENT_DT","ACCIDENT_YEAR","TIME_PERIOD","SEVERITY_GROUP","CLASSIFICATION","ACCIDENT_TYPE","ACTIVITY","INJURY_SOURCE","NATURE_INJURY","INJ_BODY_PART","DEGREE_INJURY","MINING_EQUIP","OCCUPATION","COAL_METAL_IND","DAYS_RESTRICT","DAYS_LOST","PRELIMINARY_RECORD_TYPE_HINT"] if c in block2_df]
records=[];reused=0;new_calls=0
for i,r in block2_df.iterrows():
    rid=str(r[id_col])
    if not REPROCESS_COMPLETED and rid in existing:
        records.append(existing[rid]);reused+=1;print(f"[{i+1}/{len(block2_df)}] reused {rid}");continue
    result=classify(str(r[TEXT_COLUMN]),str(r["PRELIMINARY_RECORD_TYPE_HINT"]),bool(r["_MAY_BE_TRUNCATED"]))
    rec={"SOURCE_RECORD_ID":rid,"NARRATIVE":str(r[TEXT_COLUMN]),"NARRATIVE_WORD_COUNT":int(r["_WORD_COUNT"]),"NARRATIVE_CHARACTER_COUNT":int(r["_CHARACTER_COUNT"]),"NARRATIVE_MAY_BE_TRUNCATED":bool(r["_MAY_BE_TRUNCATED"]),**{c:(None if pd.isna(r[c]) else r[c]) for c in context},**result,"MODEL_ID":MODEL_ID,"PROMPT_VERSION":PROMPT_VERSION,"PROCESSED_AT":datetime.now().isoformat(timespec="seconds")}
    records.append(rec);append_jsonl(rec);new_calls+=1;print(f"[{i+1}/{len(block2_df)}] new {rid}");gc.collect();torch.cuda.empty_cache()

# ---------------- RECOVER ORIGINAL OUTPUT / SUPPORT TIERS --------------------
def recover(row):
    base={"ORIGINAL_MODEL_RECORD_TYPE":row.get("RECORD_TYPE",row.get("record_type","")),"ORIGINAL_MODEL_CAUSE":row.get("PRIMARY_CAUSE",row.get("primary_cause","")),"ORIGINAL_MODEL_FACTOR":row.get("PRIMARY_CONTRIBUTING_FACTOR",row.get("primary_contributing_factor","")),"ORIGINAL_MODEL_PREVENTION":row.get("PREVENTION_THEME",row.get("prevention_theme","")),"ORIGINAL_MODEL_CAUSE_EVIDENCE":row.get("CAUSE_EVIDENCE_TEXT",row.get("cause_evidence_text","")),"ORIGINAL_MODEL_FACTOR_EVIDENCE":row.get("FACTOR_EVIDENCE_TEXT",row.get("factor_evidence_text","")),"ORIGINAL_MODEL_PREVENTION_RATIONALE":row.get("PREVENTION_RATIONALE",row.get("prevention_rationale","")),"ORIGINAL_MODEL_FACTOR_CONFIDENCE":sf(row.get("FACTOR_CONFIDENCE",row.get("factor_confidence",0))),"ORIGINAL_MODEL_PREVENTION_CONFIDENCE":sf(row.get("PREVENTION_CONFIDENCE",row.get("prevention_confidence",0))),"ORIGINAL_MODEL_OUTPUT_RECOVERED":False}
    raw=str(row.get("RAW_MODEL_RESPONSE","") or "").strip()
    if raw:
        try:
            v=IncidentExtraction.model_validate(normalize_dict(json.loads(repair_json(extract_json(raw)))))
            base.update({"ORIGINAL_MODEL_RECORD_TYPE":v.record_type,"ORIGINAL_MODEL_CAUSE":v.primary_cause,"ORIGINAL_MODEL_FACTOR":v.primary_contributing_factor,"ORIGINAL_MODEL_PREVENTION":v.prevention_theme,"ORIGINAL_MODEL_CAUSE_EVIDENCE":v.cause_evidence_text,"ORIGINAL_MODEL_FACTOR_EVIDENCE":v.factor_evidence_text,"ORIGINAL_MODEL_PREVENTION_RATIONALE":v.prevention_rationale,"ORIGINAL_MODEL_FACTOR_CONFIDENCE":v.factor_confidence,"ORIGINAL_MODEL_PREVENTION_CONFIDENCE":v.prevention_confidence,"ORIGINAL_MODEL_OUTPUT_RECOVERED":True})
        except:pass
    return base

def revise(row):
    """
    Less-stringent, auditable factor-support logic.

    DIRECTLY_SUPPORTED:
        Original factor is not abstention and model evidence is lexically
        traceable to the narrative. Category cue terms strengthen this tier
        but are not mandatory.

    CONTEXTUALLY_SUPPORTED:
        Direct lexical matching fails, but at least one of the following holds:
        1. category-specific narrative cue is present;
        2. model factor confidence >= 0.65 and support is HIGH or MEDIUM;
        3. model factor confidence >= 0.75 and a nonempty factor-evidence
           statement was returned.

    NO_SUPPORTED_FACTOR:
        Model abstained, or no direct/contextual support rule is met.

    This intentionally analyzes more records than the strict Block 2 rule,
    while preserving support tiers for sensitivity analysis and human review.
    """
    rec = recover(row)
    factor = rec["ORIGINAL_MODEL_FACTOR"]
    evidence = rec["ORIGINAL_MODEL_FACTOR_EVIDENCE"]
    narrative = str(row.get("NARRATIVE", ""))
    confidence = sf(rec.get("ORIGINAL_MODEL_FACTOR_CONFIDENCE", 0.0))

    # Recover model support level from either raw or standardized saved fields.
    model_support = tok(
        row.get(
            "FACTOR_SUPPORT_LEVEL",
            row.get("factor_support_level", "LOW"),
        )
    )
    if model_support not in {"HIGH", "MEDIUM", "LOW"}:
        model_support = "LOW"

    lexical_valid, evidence_exact, evidence_normalized, evidence_overlap = (
        evidence_metrics(evidence, narrative)
    )
    narrative_hits = phrase_hits(factor, narrative)
    evidence_hits = phrase_hits(factor, evidence)
    nonempty_evidence = bool(str(evidence).strip())

    if factor == "NO_SEPARATE_FACTOR_STATED":
        tier = "NO_SUPPORTED_FACTOR"
        method = "MODEL_ABSTAINED"

    elif lexical_valid:
        # Less stringent than the previous version: a traceable evidence excerpt
        # is sufficient for direct support. Category terms are recorded but are
        # not required because the ontology wording may be expressed differently.
        tier = "DIRECTLY_SUPPORTED"
        method = (
            "TRACEABLE_MODEL_EVIDENCE_WITH_CATEGORY_CUE"
            if evidence_hits
            else "TRACEABLE_MODEL_EVIDENCE"
        )

    elif narrative_hits:
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "CATEGORY_SPECIFIC_NARRATIVE_CUE"

    elif confidence >= 0.65 and model_support in {"HIGH", "MEDIUM"}:
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "MODEL_CONFIDENCE_AND_SUPPORT"

    elif confidence >= 0.75 and nonempty_evidence:
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "HIGH_CONFIDENCE_WITH_EXPLANATORY_EVIDENCE"

    else:
        tier = "NO_SUPPORTED_FACTOR"
        method = "INSUFFICIENT_FACTOR_SUPPORT"

    if tier == "NO_SUPPORTED_FACTOR":
        revised_factor = "NO_SEPARATE_FACTOR_STATED"
        revised_prevention = "INVESTIGATION_REQUIRED"
        revised_rationale = (
            "Additional investigation is required before selecting a specific "
            "corrective action."
        )
        prevention_method = "UNSUPPORTED_FACTOR_ABSTENTION"

    else:
        revised_factor = factor
        original_prevention = rec["ORIGINAL_MODEL_PREVENTION"]

        if aligned(factor, original_prevention):
            revised_prevention = original_prevention
            revised_rationale = rec["ORIGINAL_MODEL_PREVENTION_RATIONALE"]
            prevention_method = "ORIGINAL_ALLOWED_PREVENTION"
        else:
            revised_prevention = PREFERRED.get(factor, "INVESTIGATION_REQUIRED")
            revised_rationale = (
                "The prevention theme was assigned from the approved "
                "factor-to-prevention map."
            )
            prevention_method = "PREFERRED_MAPPING_REPAIR"

    cause_valid = bool(row.get("CAUSE_EVIDENCE_VALID", False))
    success = row.get("PROCESSING_STATUS", "") == "SUCCESS"
    acute = rec["ORIGINAL_MODEL_RECORD_TYPE"] == "ACUTE_ACCIDENT"

    if not (success and acute and cause_valid):
        status = "EXCLUDED_OR_REVIEW_REQUIRED"
    elif tier == "DIRECTLY_SUPPORTED":
        status = "VALID_DIRECT_FACTOR_ANALYSIS"
    elif tier == "CONTEXTUALLY_SUPPORTED":
        status = "VALID_CONTEXTUAL_FACTOR_ANALYSIS"
    else:
        status = "VALID_FOR_CAUSE_ANALYSIS_ONLY"

    return {
        **row,
        **rec,
        "ORIGINAL_MODEL_FACTOR_SUPPORT": model_support,
        "FACTOR_SUPPORT_TIER": tier,
        "FACTOR_SUPPORT_METHOD": method,
        "FACTOR_CONTEXT_TERMS": "; ".join(narrative_hits),
        "FACTOR_EVIDENCE_CATEGORY_TERMS": "; ".join(evidence_hits),
        "ORIGINAL_FACTOR_EVIDENCE_LEXICALLY_VALID": lexical_valid,
        "ORIGINAL_FACTOR_EVIDENCE_EXACT": evidence_exact,
        "ORIGINAL_FACTOR_EVIDENCE_NORMALIZED": evidence_normalized,
        "ORIGINAL_FACTOR_EVIDENCE_TOKEN_OVERLAP": round(evidence_overlap, 4),
        "REVISED_FACTOR": revised_factor,
        "REVISED_PREVENTION": revised_prevention,
        "REVISED_PREVENTION_RATIONALE": revised_rationale,
        "REVISED_PREVENTION_METHOD": prevention_method,
        "REVISED_FACTOR_PREVENTION_ALIGNED": aligned(
            revised_factor, revised_prevention
        ),
        "REVISED_ANALYSIS_STATUS": status,
    }

results=pd.DataFrame([revise(r) for r in records]);results.to_csv(REVISED_RESULTS_CSV,index=False,encoding="utf-8-sig")
cause_df=results[results.REVISED_ANALYSIS_STATUS.isin(["VALID_DIRECT_FACTOR_ANALYSIS","VALID_CONTEXTUAL_FACTOR_ANALYSIS","VALID_FOR_CAUSE_ANALYSIS_ONLY"])].copy()
direct_df=results[results.REVISED_ANALYSIS_STATUS.eq("VALID_DIRECT_FACTOR_ANALYSIS")].copy()
context_df=results[results.REVISED_ANALYSIS_STATUS.eq("VALID_CONTEXTUAL_FACTOR_ANALYSIS")].copy()
expanded_df=results[results.REVISED_ANALYSIS_STATUS.isin(["VALID_DIRECT_FACTOR_ANALYSIS","VALID_CONTEXTUAL_FACTOR_ANALYSIS"])].copy()
cause_only_df=results[results.REVISED_ANALYSIS_STATUS.eq("VALID_FOR_CAUSE_ANALYSIS_ONLY")].copy()
excluded_df=results[results.REVISED_ANALYSIS_STATUS.eq("EXCLUDED_OR_REVIEW_REQUIRED")].copy()

# ----------------------------- ANALYSIS ---------------------------------------
def summary(d,col,name):
    if d.empty:return pd.DataFrame(columns=[name,"Count","Percent"])
    z=d[col].value_counts().rename_axis(name).reset_index(name="Count");z["Percent"]=100*z.Count/len(d);return z
def pcttab(d,row,col,order):
    c=pd.crosstab(d[row],d[col]).reindex(order,fill_value=0);return c,c.div(c.sum(1).replace(0,np.nan),axis=0).mul(100).fillna(0)
def assoc(d,g,c,name):
    t=pd.crosstab(d[g],d[c]);t=t.loc[t.sum(1)>0,t.sum(0)>0]
    if min(t.shape)<2:return {"Comparison":name,"P-value":np.nan,"Cramer's V":np.nan,"Records":int(t.values.sum())}
    chi,p,dof,e=chi2_contingency(t);v=math.sqrt(chi/(t.values.sum()*min(t.shape[0]-1,t.shape[1]-1)));return {"Comparison":name,"Chi-square":chi,"Degrees of freedom":dof,"P-value":p,"Cramer's V":v,"Records":int(t.values.sum())}
def changes(p,c,label):
    rows=[]
    for x in p.columns:
        if c[x].sum()>=MIN_CATEGORY_COUNT_FOR_TREND:
            a=float(p.loc[EARLY,x]);b=float(p.loc[RECENT,x]);rows.append({label:x,"Total count":int(c[x].sum()),"Early percent":a,"Recent percent":b,"Percentage-point change":b-a})
    return pd.DataFrame(rows).sort_values("Percentage-point change",ascending=False) if rows else pd.DataFrame()

cause_sum=summary(cause_df,"ORIGINAL_MODEL_CAUSE","Primary cause");factor_sum=summary(expanded_df,"REVISED_FACTOR","Contributing factor");prev_sum=summary(expanded_df,"REVISED_PREVENTION","Prevention theme");tier_sum=summary(results,"FACTOR_SUPPORT_TIER","Factor support tier");status_sum=summary(results,"REVISED_ANALYSIS_STATUS","Analysis status")
ccs,csp=pcttab(cause_df,"SEVERITY_GROUP","ORIGINAL_MODEL_CAUSE",SEVERITY_ORDER);fcs,fsp=pcttab(expanded_df,"SEVERITY_GROUP","REVISED_FACTOR",SEVERITY_ORDER);pcs,psp=pcttab(expanded_df,"SEVERITY_GROUP","REVISED_PREVENTION",SEVERITY_ORDER)
cct,ctp=pcttab(cause_df,"TIME_PERIOD","ORIGINAL_MODEL_CAUSE",TIME_ORDER);fct,ftp=pcttab(expanded_df,"TIME_PERIOD","REVISED_FACTOR",TIME_ORDER);pct,ptp=pcttab(expanded_df,"TIME_PERIOD","REVISED_PREVENTION",TIME_ORDER)
association=pd.DataFrame([assoc(cause_df,"SEVERITY_GROUP","ORIGINAL_MODEL_CAUSE","Cause by severity"),assoc(expanded_df,"SEVERITY_GROUP","REVISED_FACTOR","Factor by severity"),assoc(expanded_df,"SEVERITY_GROUP","REVISED_PREVENTION","Prevention by severity"),assoc(cause_df,"TIME_PERIOD","ORIGINAL_MODEL_CAUSE","Cause by time"),assoc(expanded_df,"TIME_PERIOD","REVISED_FACTOR","Factor by time"),assoc(expanded_df,"TIME_PERIOD","REVISED_PREVENTION","Prevention by time")])
cause_change=changes(ctp,cct,"Primary cause");factor_change=changes(ftp,fct,"Contributing factor");prev_change=changes(ptp,pct,"Prevention theme")
cf=pd.crosstab(expanded_df.ORIGINAL_MODEL_CAUSE,expanded_df.REVISED_FACTOR);fp=pd.crosstab(expanded_df.REVISED_FACTOR,expanded_df.REVISED_PREVENTION)
quality=pd.DataFrame([("Selected narratives",len(results)),("Existing records reused",reused),("New Qwen calls",new_calls),("Cause-analysis population",len(cause_df)),("Directly supported factors",len(direct_df)),("Contextually supported factors",len(context_df)),("Expanded factor-analysis population",len(expanded_df)),("Cause-only population",len(cause_only_df)),("Excluded/review population",len(excluded_df))],columns=["Measure","Value"])

# Recovery and prevention-repair diagnostics.
recovery_count = int(results["ORIGINAL_MODEL_OUTPUT_RECOVERED"].fillna(False).sum())
prevention_method_sum = summary(
    results,
    "REVISED_PREVENTION_METHOD",
    "Prevention assignment method",
)
support_method_sum = summary(
    results,
    "FACTOR_SUPPORT_METHOD",
    "Factor support method",
)
quality = pd.concat(
    [
        quality,
        pd.DataFrame(
            [
                ("Original outputs recovered from raw response", recovery_count),
                ("Original outputs using saved-field fallback", len(results)-recovery_count),
                ("Prevention mapping repairs", int((results["REVISED_PREVENTION_METHOD"]=="PREFERRED_MAPPING_REPAIR").sum())),
                ("Unsupported-factor abstentions", int((results["REVISED_PREVENTION_METHOD"]=="UNSUPPORTED_FACTOR_ABSTENTION").sum())),
            ],
            columns=["Measure", "Value"],
        ),
    ],
    ignore_index=True,
)

# Strict direct-only versus expanded direct-plus-contextual sensitivity.
def dist_series(data, column, label):
    if data.empty:
        return pd.Series(dtype=float, name=label)
    return data[column].value_counts(normalize=True).mul(100).rename(label)

factor_sensitivity = pd.concat(
    [
        dist_series(direct_df, "REVISED_FACTOR", "Direct support only"),
        dist_series(expanded_df, "REVISED_FACTOR", "Direct plus contextual"),
    ],
    axis=1,
).fillna(0.0)
factor_sensitivity["Percentage-point difference"] = (
    factor_sensitivity["Direct plus contextual"]
    - factor_sensitivity["Direct support only"]
)
factor_sensitivity_display = (
    factor_sensitivity.reset_index()
    .rename(columns={"index": "Contributing factor"})
    .sort_values("Percentage-point difference", ascending=False)
)

prevention_sensitivity = pd.concat(
    [
        dist_series(direct_df, "REVISED_PREVENTION", "Direct support only"),
        dist_series(expanded_df, "REVISED_PREVENTION", "Direct plus contextual"),
    ],
    axis=1,
).fillna(0.0)
prevention_sensitivity["Percentage-point difference"] = (
    prevention_sensitivity["Direct plus contextual"]
    - prevention_sensitivity["Direct support only"]
)
prevention_sensitivity_display = (
    prevention_sensitivity.reset_index()
    .rename(columns={"index": "Prevention theme"})
    .sort_values("Percentage-point difference", ascending=False)
)


# ------------------------------ PDF ------------------------------------------
pdf=PdfPages(OUTPUT_PDF);page=[0]
def save(fig):page[0]+=1;fig.text(.5,.018,f"Page {page[0]}",ha="center",fontsize=7);pdf.savefig(fig,bbox_inches="tight");plt.close(fig)
def textpage(title,lines):
    wrapped=[]
    for line in lines:wrapped+=textwrap.wrap(str(line),100,break_long_words=False) or [""]
    for start in range(0,len(wrapped),52):
        fig=plt.figure(figsize=(8.5,11));fig.text(.5,.96,title,ha="center",va="top",fontsize=14,fontweight="bold");fig.text(.055,.915,"\n".join(wrapped[start:start+52]),va="top",family="monospace",fontsize=8.4,linespacing=1.2);save(fig)
def tablepage(title,d,rows=22):
    if d.empty:textpage(title,["No results."]);return
    z=d.copy()
    for c in z:
        if pd.api.types.is_float_dtype(z[c]):z[c]=z[c].map(lambda x:"" if pd.isna(x) else f"{x:.4f}")
        z[c]=z[c].astype(str).map(lambda x:"\n".join(textwrap.wrap(x,26,break_long_words=False)))
    for s in range(0,len(z),rows):
        fig=plt.figure(figsize=(11,8.5));ax=fig.add_axes([.02,.06,.96,.86]);ax.axis("off");fig.suptitle(title if s==0 else title+" - Continued",fontweight="bold");tb=ax.table(cellText=z.iloc[s:s+rows].values,colLabels=z.columns,cellLoc="left",loc="upper left");tb.auto_set_font_size(False);tb.set_fontsize(7);tb.scale(1,1.35);save(fig)
def heat(title,p,top):
    cols=[x for x in top if x in p]
    if not cols:return
    q=p[cols];fig,ax=plt.subplots(figsize=(11,7));im=ax.imshow(q.values,aspect="auto",cmap="Blues");ax.set_xticks(range(len(cols)));ax.set_xticklabels([x.replace("_"," ").title() for x in cols],rotation=35,ha="right",fontsize=7);ax.set_yticks(range(len(q.index)));ax.set_yticklabels(q.index,fontsize=8);ax.set_title(title,fontweight="bold")
    for i in range(q.shape[0]):
        for j in range(q.shape[1]):
            if q.iloc[i,j]>=1:ax.text(j,i,f"{q.iloc[i,j]:.1f}%",ha="center",va="center",fontsize=7)
    fig.colorbar(im,ax=ax,label="Within-group percentage");fig.tight_layout();save(fig)
def relheat(title,t):
    if t.empty:return
    r=t.sum(1).nlargest(10).index;c=t.sum(0).nlargest(10).index;q=t.loc[r,c];fig,ax=plt.subplots(figsize=(11,8));im=ax.imshow(q.values,aspect="auto",cmap="Blues");ax.set_xticks(range(len(c)));ax.set_xticklabels([x.replace("_"," ").title() for x in c],rotation=40,ha="right",fontsize=7);ax.set_yticks(range(len(r)));ax.set_yticklabels([x.replace("_"," ").title() for x in r],fontsize=7);ax.set_title(title,fontweight="bold")
    for i in range(q.shape[0]):
        for j in range(q.shape[1]):
            if q.iloc[i,j]:ax.text(j,i,str(q.iloc[i,j]),ha="center",va="center",fontsize=7)
    fig.colorbar(im,ax=ax);fig.tight_layout();save(fig)

textpage("MSHA Block 2 Revised Expanded Analysis",[f"Selected unseen narratives: {len(results):,}",f"Existing records reused: {reused:,}",f"New model calls: {new_calls:,}",f"Cause-analysis population: {len(cause_df):,}",f"Directly supported factors: {len(direct_df):,}",f"Contextually supported factors: {len(context_df):,}",f"Expanded factor-analysis population: {len(expanded_df):,}",f"Cause-only population: {len(cause_only_df):,}","","Main change: factor evidence is now tiered as directly supported, contextually supported, or unsupported. Expanded factor and prevention analyses use direct plus contextual support. Direct-only results remain the strict sensitivity population."])
tablepage("Processing and Analytical Populations",quality);tablepage("Factor-Support Tier Distribution",tier_sum);tablepage("Revised Analysis Status",status_sum);tablepage("Factor Support Methods",support_method_sum);tablepage("Prevention Assignment Methods",prevention_method_sum)
textpage("Factor-Evidence Method",["DIRECTLY_SUPPORTED: the model evidence is found in the narrative and includes category-specific factor terminology.","","CONTEXTUALLY_SUPPORTED: strict evidence matching fails, but the source narrative contains transparent, category-specific contextual cues.","","NO_SUPPORTED_FACTOR: no defensible separate factor is retained; prevention is INVESTIGATION_REQUIRED.","","The contextual tier is exploratory and will be tested in the 100-record human review."])
tablepage("Primary Cause Distribution",cause_sum);tablepage("Expanded Contributing-Factor Distribution",factor_sum);tablepage("Expanded Prevention-Theme Distribution",prev_sum);tablepage("Factor Sensitivity: Direct versus Expanded",factor_sensitivity_display);tablepage("Prevention Sensitivity: Direct versus Expanded",prevention_sensitivity_display)
heat("Primary Causes by Severity",csp,cause_sum["Primary cause"].head(TOP_CATEGORIES_TO_PLOT).tolist());heat("Expanded Factors by Severity",fsp,factor_sum["Contributing factor"].head(TOP_CATEGORIES_TO_PLOT).tolist());heat("Expanded Prevention Themes by Severity",psp,prev_sum["Prevention theme"].head(TOP_CATEGORIES_TO_PLOT).tolist())
heat("Primary Causes by Time Period",ctp,cause_sum["Primary cause"].head(TOP_CATEGORIES_TO_PLOT).tolist());heat("Expanded Factors by Time Period",ftp,factor_sum["Contributing factor"].head(TOP_CATEGORIES_TO_PLOT).tolist());heat("Expanded Prevention Themes by Time Period",ptp,prev_sum["Prevention theme"].head(TOP_CATEGORIES_TO_PLOT).tolist())
tablepage("Early-to-Recent Cause Changes",cause_change);tablepage("Early-to-Recent Factor Changes",factor_change);tablepage("Early-to-Recent Prevention Changes",prev_change);tablepage("Association Tests",association)
relheat("Cause and Revised Factor Relationships",cf);relheat("Revised Factor and Prevention Relationships",fp)
textpage("Interpretation and Limitations",["Cause results use valid acute accident mechanisms.","","Expanded factor and prevention results include direct and contextual evidence tiers. They should be compared with the strict direct-support subset and validated in Blocks 3 and 4.","","Category-specific phrase rules are transparent but imperfect. They can recover plausible factors missed by verbatim matching, but they may also admit false positives.","","Time-period changes are narrative-share changes, not exposure-adjusted incidence rates.","","The sample is balanced by severity and does not represent the natural severity distribution."])
meta=pdf.infodict();meta["Title"]="MSHA Block 2 Revised Expanded Analysis";meta["CreationDate"]=datetime.now();pdf.close()

print("="*90)
print("REVISED BLOCK 2 COMPLETE")
print(f"Selected: {len(results):,} | Reused: {reused:,} | New calls: {new_calls:,}")
print(f"Cause analysis: {len(cause_df):,}")
print(f"Direct factors: {len(direct_df):,}")
print(f"Contextual factors: {len(context_df):,}")
print(f"Expanded factor analysis: {len(expanded_df):,}")
print(f"Results: {os.path.abspath(REVISED_RESULTS_CSV)}")
print(f"Report: {os.path.abspath(OUTPUT_PDF)}")
print("="*90)

display(quality);display(tier_sum);display(cause_sum);display(factor_sum);display(prev_sum);display(association)
````

````python
# =============================================================================
# BLOCK 3 - STANDALONE HUMAN-VALIDATION SAMPLE
# RUN FROZEN MODEL ON 100 ADDITIONAL UNSEEN MSHA NARRATIVES
# CREATE INDEPENDENT HUMAN-REVIEW WORKBOOK
#
# REQUIRED INPUTS
# ---------------
# 1. pandas DataFrame named `accidents` already loaded.
# 2. MSHA_Block1_Development_Record_IDs.json
# 3. MSHA_Block2_Unseen_Record_IDs.json OR MSHA_Block2_Revised_Record_IDs.json
#
# OUTPUTS
# -------
# 1. MSHA_Block3_Human_Review_Workbook.xlsx
# 2. MSHA_Block3_Model_Results.csv
# 3. MSHA_Block3_Checkpoint.jsonl
# 4. MSHA_Block3_Record_IDs.json
#
# DESIGN
# ------
# - Select exactly 100 additional narratives not used in Block 1 or Block 2.
# - Target 20 narratives per severity group.
# - Distribute records across early, middle, and recent periods where available.
# - Run the frozen Block 1c ontology/model.
# - Retain original model labels and expanded factor-support labels.
# - Create an independent human-review worksheet with ontology dropdowns.
# - Put model results on a separate worksheet to reduce anchoring bias.
# - Block 4 will read the completed workbook and create the validation report.
# =============================================================================

import os
import re
import gc
import sys
import json
import time
import math
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import torch

from pydantic import BaseModel, Field, model_validator
from transformers import AutoTokenizer, AutoModelForCausalLM, BitsAndBytesConfig

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Border, Side, Alignment
from openpyxl.worksheet.datavalidation import DataValidation
from openpyxl.formatting.rule import FormulaRule
from openpyxl.utils import get_column_letter
from openpyxl.worksheet.table import Table, TableStyleInfo

warnings.filterwarnings("ignore")


# =============================================================================
# 1. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"
PROMPT_VERSION = "MSHA_BLOCK1C_FROZEN_FOR_BLOCK3_V1.0"

TEXT_COLUMN = "NARRATIVE"
DATE_COLUMN = "ACCIDENT_DT"
SEVERITY_COLUMN = "DEGREE_INJURY"

ID_COLUMN_CANDIDATES = [
    "DOCUMENT_NO",
    "DOC_NO",
    "ACCIDENT_ID",
]

TOTAL_REVIEW_SIZE = 100
TARGET_PER_SEVERITY = 20
RANDOM_STATE = 126
MIN_NARRATIVE_WORDS = 8
TRUNCATION_CHARACTER_THRESHOLD = 383
MAX_NEW_TOKENS = 750
MAX_RETRIES = 4
LOW_CONFIDENCE_THRESHOLD = 0.70
REPROCESS_COMPLETED = False

CONTEXTUAL_FACTOR_CONFIDENCE_THRESHOLD = 0.65
HIGH_CONFIDENCE_FACTOR_THRESHOLD = 0.75

BLOCK1_IDS_JSON = "MSHA_Block1_Development_Record_IDs.json"
BLOCK2_ID_CANDIDATES = [
    "MSHA_Block2_Expanded_Record_IDs.json",
    "MSHA_Block2_Revised_Record_IDs.json",
    "MSHA_Block2_Unseen_Record_IDs.json",
]

BLOCK3_IDS_JSON = "MSHA_Block3_Record_IDs.json"
BLOCK3_CHECKPOINT_JSONL = "MSHA_Block3_Checkpoint.jsonl"
BLOCK3_RESULTS_CSV = "MSHA_Block3_Model_Results.csv"
BLOCK3_REVIEW_XLSX = "MSHA_Block3_Human_Review_Workbook.xlsx"

SEVERITY_ORDER = [
    "Fatality",
    "Permanent disability",
    "Days away",
    "Restricted or transferred",
    "No lost time or other",
]

EARLY_PERIOD = "Early: 2000-2008"
MIDDLE_PERIOD = "Middle: 2009-2017"
RECENT_PERIOD = "Recent: 2018-2026"
TIME_PERIOD_ORDER = [EARLY_PERIOD, MIDDLE_PERIOD, RECENT_PERIOD]


# =============================================================================
# 2. ENVIRONMENT VALIDATION
# =============================================================================

if "accidents" not in globals():
    raise NameError("A pandas DataFrame named 'accidents' must already be loaded.")

if not isinstance(accidents, pd.DataFrame) or accidents.empty:
    raise ValueError("'accidents' must be a nonempty pandas DataFrame.")

if not Path(BLOCK1_IDS_JSON).exists():
    raise FileNotFoundError(BLOCK1_IDS_JSON)

block2_ids_path = next(
    (Path(path) for path in BLOCK2_ID_CANDIDATES if Path(path).exists()),
    None,
)

if block2_ids_path is None:
    raise FileNotFoundError(
        "No Block 2 record-ID file was found. Expected one of: "
        + ", ".join(BLOCK2_ID_CANDIDATES)
    )

print("=" * 95)
print("BLOCK 3 - 100-RECORD HUMAN-VALIDATION SAMPLE")
print("=" * 95)
print("Block 2 exclusion file:", block2_ids_path)


# =============================================================================
# 3. FROZEN ONTOLOGIES
# =============================================================================

RECORD_TYPE_CATEGORIES = [
    "ACUTE_ACCIDENT",
    "OCCUPATIONAL_EXPOSURE_OR_ILLNESS",
    "NONWORK_RELATED_OR_NATURAL_EVENT",
    "NONINJURY_REPORT",
    "INSUFFICIENT_EVENT_INFORMATION",
]

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]

FACTOR_CATEGORIES = [
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    "LINE_OF_FIRE_BODY_POSITION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION",
    "INADEQUATE_GUARDING_OR_BARRIER",
    "ENERGY_ISOLATION_DEFICIENCY",
    "EQUIPMENT_CONDITION_OR_FAILURE",
    "PROCEDURE_OR_WORK_METHOD",
    "INADEQUATE_HAZARD_RECOGNITION",
    "COMMUNICATION_OR_COORDINATION",
    "TRAINING_OR_EXPERIENCE",
    "PPE_OR_PROTECTIVE_CONTROL",
    "ENVIRONMENTAL_CONDITION",
    "WORKSPACE_OR_ACCESS_LIMITATION",
    "OTHER",
    "NO_SEPARATE_FACTOR_STATED",
]

PREVENTION_CATEGORIES = [
    "SECURE_OR_REPOSITION_MATERIAL",
    "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "INSTALL_OR_IMPROVE_GUARDING",
    "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "REVISE_PROCEDURE_OR_JOB_PLAN",
    "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "PROVIDE_TARGETED_TRAINING",
    "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "USE_MECHANICAL_HANDLING_ASSISTANCE",
    "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "CONTROL_ENVIRONMENTAL_EXPOSURE",
    "REDESIGN_WORKSPACE_OR_ACCESS",
    "ENGINEERING_REVIEW_REQUIRED",
    "INVESTIGATION_REQUIRED",
    "INSUFFICIENT_INFORMATION",
]

PREFERRED_PREVENTION_BY_FACTOR = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": "SECURE_OR_REPOSITION_MATERIAL",
    "INADEQUATE_SECUREMENT_OR_RESTRAINT": "SECURE_OR_REPOSITION_MATERIAL",
    "LINE_OF_FIRE_BODY_POSITION": "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
    "UNSTABLE_GROUND_OR_ROOF_CONDITION": "IMPROVE_GROUND_CONTROL_OR_SCALING",
    "INADEQUATE_GUARDING_OR_BARRIER": "INSTALL_OR_IMPROVE_GUARDING",
    "ENERGY_ISOLATION_DEFICIENCY": "ISOLATE_OR_CONTROL_STORED_ENERGY",
    "EQUIPMENT_CONDITION_OR_FAILURE": "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
    "PROCEDURE_OR_WORK_METHOD": "REVISE_PROCEDURE_OR_JOB_PLAN",
    "INADEQUATE_HAZARD_RECOGNITION": "CONDUCT_PRE_JOB_HAZARD_REVIEW",
    "COMMUNICATION_OR_COORDINATION": "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    "TRAINING_OR_EXPERIENCE": "PROVIDE_TARGETED_TRAINING",
    "PPE_OR_PROTECTIVE_CONTROL": "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
    "ENVIRONMENTAL_CONDITION": "CONTROL_ENVIRONMENTAL_EXPOSURE",
    "WORKSPACE_OR_ACCESS_LIMITATION": "REDESIGN_WORKSPACE_OR_ACCESS",
    "NO_SEPARATE_FACTOR_STATED": "INVESTIGATION_REQUIRED",
}

ALLOWED_PREVENTIONS_BY_FACTOR = {
    factor: {prevention}
    for factor, prevention in PREFERRED_PREVENTION_BY_FACTOR.items()
}

ALLOWED_PREVENTIONS_BY_FACTOR.update({
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "ENGINEERING_REVIEW_REQUIRED",
    },
    "LINE_OF_FIRE_BODY_POSITION": {
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    },
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },
    "UNSTABLE_GROUND_OR_ROOF_CONDITION": {
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "ENGINEERING_REVIEW_REQUIRED",
    },
    "EQUIPMENT_CONDITION_OR_FAILURE": {
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "ENGINEERING_REVIEW_REQUIRED",
    },
    "PROCEDURE_OR_WORK_METHOD": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
    },
    "OTHER": {
        "ENGINEERING_REVIEW_REQUIRED",
        "INVESTIGATION_REQUIRED",
        "REVISE_PROCEDURE_OR_JOB_PLAN",
    },
    "NO_SEPARATE_FACTOR_STATED": {
        "INVESTIGATION_REQUIRED",
        "INSUFFICIENT_INFORMATION",
    },
})

CATEGORY_ALIASES = {
    "UNSTATED": "NO_SEPARATE_FACTOR_STATED",
    "NOT_STATED": "NO_SEPARATE_FACTOR_STATED",
    "NONE_STATED": "NO_SEPARATE_FACTOR_STATED",
    "NO_FACTOR_STATED": "NO_SEPARATE_FACTOR_STATED",
    "CAUGHT_BETWEEN": "CAUGHT_IN_OR_BETWEEN",
    "CAUGHT_IN_BETWEEN": "CAUGHT_IN_OR_BETWEEN",
    "STRUCK_BY_OBJECT": "STRUCK_BY_MOVING_OBJECT",
    "LOSS_OF_CONTROL": "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "UNEXPECTED_MACHINE_MOVEMENT": "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "FOREIGN_OBJECT_CONTACT": "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "HEAT_EXPOSURE": "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "SLIP_TRIP_FALL": "SLIP_TRIP_OR_FALL",
    "UNKNOWN": "INSUFFICIENT_INFORMATION",
    "INSUFFICIENT_INFO": "INSUFFICIENT_INFORMATION",
}

SUPPORT_LEVEL_ALIASES = {
    "HIGH": "HIGH",
    "DIRECT": "HIGH",
    "EXPLICIT": "HIGH",
    "MEDIUM": "MEDIUM",
    "MODERATE": "MEDIUM",
    "IMPLIED": "MEDIUM",
    "LOW": "LOW",
    "WEAK": "LOW",
    "UNCERTAIN": "LOW",
    "NONE": "LOW",
    "": "LOW",
}

FACTOR_EVIDENCE_TERMS = {
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": [
        "unstable", "unsupported", "load shifted", "material shifted",
        "rolling", "sliding", "fell over", "tipped over",
        "improperly positioned",
    ],
    "INADEQUATE_SECUREMENT_OR_RESTRAINT": [
        "not secured", "unsecured", "restraint failed", "strap broke",
        "chain broke", "not blocked", "not braced", "pin removed",
    ],
    "LINE_OF_FIRE_BODY_POSITION": [
        "pinch point", "in the path", "reached into", "hand between",
        "arm between", "foot between", "attempted to catch",
        "tried to catch", "under suspended",
    ],
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": [
        "ice", "icy", "snow", "mud", "wet floor", "wet surface",
        "grease", "oil on", "debris", "uneven surface", "uneven ground",
        "poor footing", "loose gravel", "slippery", "broken step",
        "tripping hazard",
    ],
    "UNSTABLE_GROUND_OR_ROOF_CONDITION": [
        "roof fell", "roof fall", "rib fell", "rib fall", "rock fell",
        "slate fell", "highwall", "cave in", "ground failure",
        "unsupported roof", "bank collapsed",
    ],
    "INADEQUATE_GUARDING_OR_BARRIER": [
        "guard removed", "guard missing", "missing guard", "unguarded",
        "guard was off", "barrier missing", "bypassed guard",
    ],
    "ENERGY_ISOLATION_DEFICIENCY": [
        "not locked out", "failed to lock out", "lockout", "tagout", "loto",
        "failed to bleed", "pressure remained", "still under pressure",
        "energized", "not deenergized", "not blocked", "not isolated",
        "stored energy", "engine running",
    ],
    "EQUIPMENT_CONDITION_OR_FAILURE": [
        "equipment failed", "component failed", "mechanical failure",
        "malfunction", "defective", "damaged", "worn", "leaking",
        "not working", "hose failed", "pin failed", "cable broke",
    ],
    "PROCEDURE_OR_WORK_METHOD": [
        "failed to follow procedure", "procedure was not followed",
        "improper method", "unsafe method", "work method", "task sequence",
        "using his foot", "using her foot",
    ],
    "INADEQUATE_HAZARD_RECOGNITION": [
        "did not recognize", "did not realize", "was unaware",
        "failed to identify", "hazard not identified", "failed to examine",
    ],
    "COMMUNICATION_OR_COORDINATION": [
        "did not communicate", "failed to communicate", "no communication",
        "not notified", "did not signal", "miscommunication", "no spotter",
        "without a spotter",
    ],
    "TRAINING_OR_EXPERIENCE": [
        "not trained", "inadequately trained", "lack of training",
        "new employee", "inexperienced", "not qualified", "task training",
    ],
    "PPE_OR_PROTECTIVE_CONTROL": [
        "not wearing", "no safety glasses", "no eye protection", "no gloves",
        "no hard hat", "no seat belt", "not using seat belt",
        "no life jacket", "ppe", "protective equipment",
    ],
    "ENVIRONMENTAL_CONDITION": [
        "heat", "hot", "temperature", "heat stress", "heat exhaustion",
        "dehydration", "cold", "snow", "ice", "wind", "weather",
        "poor visibility", "lighting", "fog", "rain", "storm",
    ],
    "WORKSPACE_OR_ACCESS_LIMITATION": [
        "restricted clearance", "limited clearance", "tight space",
        "confined space", "limited access", "difficult access",
        "awkward access", "reaching over", "limited workspace",
    ],
}


# =============================================================================
# 4. RESPONSE SCHEMA AND PROMPT
# =============================================================================

class IncidentExtraction(BaseModel):
    record_type: str
    primary_cause: str
    primary_contributing_factor: str
    prevention_theme: str
    cause_evidence_text: str = ""
    factor_evidence_text: str = ""
    prevention_rationale: str = ""
    cause_confidence: float = Field(default=0.0, ge=0.0, le=1.0)
    factor_confidence: float = Field(default=0.0, ge=0.0, le=1.0)
    prevention_confidence: float = Field(default=0.0, ge=0.0, le=1.0)
    cause_support_level: str = "LOW"
    factor_support_level: str = "LOW"
    prevention_support_level: str = "LOW"
    insufficient_information_reason: str = ""
    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_output(self):
        if self.record_type not in RECORD_TYPE_CATEGORIES:
            raise ValueError("Invalid record_type")
        if self.primary_cause not in CAUSE_CATEGORIES:
            raise ValueError("Invalid primary_cause")
        if self.primary_contributing_factor not in FACTOR_CATEGORIES:
            raise ValueError("Invalid primary_contributing_factor")
        if self.prevention_theme not in PREVENTION_CATEGORIES:
            raise ValueError("Invalid prevention_theme")
        for field_name in [
            "cause_support_level",
            "factor_support_level",
            "prevention_support_level",
        ]:
            if getattr(self, field_name) not in {"HIGH", "MEDIUM", "LOW"}:
                raise ValueError("Invalid support level")
        return self


CAUSE_DECISION_HIERARCHY = """
Apply the first category that clearly describes the immediate event:
1. CAUGHT_IN_OR_BETWEEN for trapped, pinched, crushed, compressed, or entangled.
2. VEHICLE_OR_MOBILE_EQUIPMENT_EVENT for collision, overturn, runaway, or backing.
3. GROUND_OR_ROOF_FAILURE for roof, rib, face, highwall, or ground failure.
4. PRESSURE_OR_STORED_ENERGY_RELEASE for released pressure, tension, or stored energy.
5. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT when loss of control is explicit.
6. UNEXPECTED_EQUIPMENT_MOVEMENT for unexpected movement without explicit loss of control.
7. STRUCK_BY_MOVING_OBJECT for contact without trapping or compression.
8. STRUCK_AGAINST_STATIONARY_OBJECT when the person's movement causes contact.
9. FOREIGN_BODY_OR_PARTICLE_CONTACT for chips, fragments, dust, or particles.
10. SLIP_TRIP_OR_FALL for loss of footing, balance, or elevation.
11. AWKWARD_MOVEMENT_OR_BODY_MOTION for an acute motion without a fall or load.
12. MANUAL_HANDLING_OR_OVEREXERTION for acute injury from lifting, pushing, pulling,
    carrying, shoveling, holding, or exerting force.
Do not use STRUCK_BY_MOVING_OBJECT as a general fallback.
""".strip()

SYSTEM_PROMPT = f"""
You extract structured information from short MSHA accident narratives.
This is reported-event extraction, not verified root-cause analysis.

ALLOWED RECORD TYPES
{", ".join(RECORD_TYPE_CATEGORIES)}

ALLOWED CAUSES
{", ".join(CAUSE_CATEGORIES)}

ALLOWED FACTORS
{", ".join(FACTOR_CATEGORIES)}

ALLOWED PREVENTION THEMES
{", ".join(PREVENTION_CATEGORIES)}

{CAUSE_DECISION_HIERARCHY}

FACTOR RULES
- Unstable material requires explicit instability, movement, or improper positioning.
- Securement requires missing, removed, failed, or inadequate restraint.
- Line of fire requires body position in a pinch point or path of movement or energy.
- Housekeeping requires an explicit walking, standing, access, or surface hazard.
- Equipment failure requires stated failure, damage, wear, leakage, or malfunction.
- NO_SEPARATE_FACTOR_STATED applies when what happened is clear but why is not supported.
- Factor evidence must support the separate factor, not merely the injury outcome.

PREVENTION RULES
- Select prevention from the supported contributing factor.
- Do not invent a factor to justify prevention.
- NO_SEPARATE_FACTOR_STATED maps to INVESTIGATION_REQUIRED.

EVIDENCE
- Copy short evidence excerpts verbatim when possible.
- If factor is NO_SEPARATE_FACTOR_STATED, factor evidence must be empty.

SUPPORT
- HIGH = direct and explicit.
- MEDIUM = strongly implied.
- LOW = incomplete or ambiguous.

Return only one JSON object with exactly these keys:
{{
  "record_type": "",
  "primary_cause": "",
  "primary_contributing_factor": "",
  "prevention_theme": "",
  "cause_evidence_text": "",
  "factor_evidence_text": "",
  "prevention_rationale": "",
  "cause_confidence": 0.0,
  "factor_confidence": 0.0,
  "prevention_confidence": 0.0,
  "cause_support_level": "HIGH",
  "factor_support_level": "HIGH",
  "prevention_support_level": "HIGH",
  "insufficient_information_reason": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 5. NORMALIZATION, JSON REPAIR, AND EVIDENCE UTILITIES
# =============================================================================

def normalize_token(value):
    return (
        str(value).strip().upper()
        .replace("-", "_")
        .replace(" ", "_")
        .replace("/", "_")
    )


def normalize_category(value):
    normalized = normalize_token(value)
    return CATEGORY_ALIASES.get(normalized, normalized)


def normalize_support(value):
    normalized = normalize_token(value)
    return SUPPORT_LEVEL_ALIASES.get(normalized, "LOW")


def safe_float(value, default=0.0):
    try:
        return min(max(float(value), 0.0), 1.0)
    except Exception:
        return default


def strip_markdown_fences(text):
    text = str(text).strip()
    text = re.sub(r"^```(?:json)?", "", text, flags=re.IGNORECASE)
    text = re.sub(r"```$", "", text)
    return text.strip()


def extract_balanced_json(text):
    text = strip_markdown_fences(text)
    start = text.find("{")
    if start < 0:
        raise ValueError("No opening JSON brace was found.")
    depth = 0
    in_string = False
    escaped = False
    for index in range(start, len(text)):
        character = text[index]
        if escaped:
            escaped = False
            continue
        if character == "\\" and in_string:
            escaped = True
            continue
        if character == '"':
            in_string = not in_string
            continue
        if in_string:
            continue
        if character == "{":
            depth += 1
        elif character == "}":
            depth -= 1
            if depth == 0:
                return text[start:index + 1]
    candidate = text[start:].strip()
    missing = candidate.count("{") - candidate.count("}")
    if missing > 0:
        return candidate + "}" * missing
    raise ValueError("Complete JSON object was not found.")


def repair_common_json_problems(json_text):
    repaired = str(json_text).strip()
    repaired = repaired.replace("“", '"').replace("”", '"').replace("’", "'")
    repaired = re.sub(r",\s*}", "}", repaired)
    repaired = re.sub(r":\s*None\b", ': ""', repaired)
    repaired = re.sub(r":\s*True\b", ": true", repaired)
    repaired = re.sub(r":\s*False\b", ": false", repaired)
    return repaired


DEFAULT_RESPONSE = {
    "record_type": "INSUFFICIENT_EVENT_INFORMATION",
    "primary_cause": "INSUFFICIENT_INFORMATION",
    "primary_contributing_factor": "NO_SEPARATE_FACTOR_STATED",
    "prevention_theme": "INVESTIGATION_REQUIRED",
    "cause_evidence_text": "",
    "factor_evidence_text": "",
    "prevention_rationale": "",
    "cause_confidence": 0.0,
    "factor_confidence": 0.0,
    "prevention_confidence": 0.0,
    "cause_support_level": "LOW",
    "factor_support_level": "LOW",
    "prevention_support_level": "LOW",
    "insufficient_information_reason": "",
    "reviewer_note": "",
}

EXPECTED_KEYS = list(DEFAULT_RESPONSE.keys())


def normalize_response_dictionary(response_dictionary):
    cleaned = dict(DEFAULT_RESPONSE)
    retained = []
    defaulted = []
    for key in EXPECTED_KEYS:
        if key in response_dictionary:
            cleaned[key] = response_dictionary[key]
            retained.append(key)
        else:
            defaulted.append(key)
    cleaned["record_type"] = normalize_token(cleaned["record_type"])
    for name in ["primary_cause", "primary_contributing_factor", "prevention_theme"]:
        cleaned[name] = normalize_category(cleaned[name])
    for name in ["cause_support_level", "factor_support_level", "prevention_support_level"]:
        cleaned[name] = normalize_support(cleaned[name])
    for name in ["cause_confidence", "factor_confidence", "prevention_confidence"]:
        cleaned[name] = safe_float(cleaned[name])
    for name in [
        "cause_evidence_text", "factor_evidence_text", "prevention_rationale",
        "insufficient_information_reason", "reviewer_note",
    ]:
        cleaned[name] = str(cleaned[name] or "").strip()
    return cleaned, retained, defaulted


def normalize_evidence_text(value):
    text = str(value).lower().replace("’", "'")
    text = re.sub(r"[^a-z0-9\s']", " ", text)
    return re.sub(r"\s+", " ", text).strip()


def evaluate_evidence(evidence, narrative):
    evidence_text = str(evidence).strip()
    exact = bool(evidence_text) and evidence_text.lower() in str(narrative).lower()
    evidence_norm = normalize_evidence_text(evidence)
    narrative_norm = normalize_evidence_text(narrative)
    normalized = bool(evidence_norm) and evidence_norm in narrative_norm
    evidence_tokens = set(evidence_norm.split())
    narrative_tokens = set(narrative_norm.split())
    overlap = (
        len(evidence_tokens.intersection(narrative_tokens)) / len(evidence_tokens)
        if evidence_tokens else 0.0
    )
    valid = exact or normalized or overlap >= 0.85
    return valid, exact, normalized, overlap


def normalized_phrase_present(phrase, narrative):
    phrase_norm = normalize_evidence_text(phrase)
    narrative_norm = normalize_evidence_text(narrative)
    return bool(phrase_norm) and phrase_norm in narrative_norm


def find_factor_context_terms(factor, narrative):
    return [
        term for term in FACTOR_EVIDENCE_TERMS.get(factor, [])
        if normalized_phrase_present(term, narrative)
    ]


def factor_prevention_aligned(factor, prevention):
    allowed = ALLOWED_PREVENTIONS_BY_FACTOR.get(factor)
    return True if allowed is None else prevention in allowed


# =============================================================================
# 6. LAZY MODEL LOADING AND CLASSIFICATION
# =============================================================================

tokenizer = None
model = None


def load_model_if_needed():
    global tokenizer, model
    if model is not None:
        return
    if not torch.cuda.is_available():
        raise RuntimeError("An NVIDIA GPU is required for records not in the checkpoint.")

    quantization_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )

    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_ID,
        use_fast=True,
        trust_remote_code=False,
    )

    if tokenizer.pad_token_id is None:
        tokenizer.pad_token = tokenizer.eos_token

    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        quantization_config=quantization_config,
        device_map="auto",
        dtype=torch.float16,
        low_cpu_mem_usage=True,
        trust_remote_code=False,
    )
    model.eval()
    print("Model loaded:", MODEL_ID)


def generate_model_response(messages):
    load_model_if_needed()
    inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )
    inputs = {key: value.to(model.device) for key, value in inputs.items()}
    input_length = int(inputs["input_ids"].shape[-1])
    with torch.inference_mode():
        generated = model.generate(
            **inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.02,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )
    response = tokenizer.decode(generated[0, input_length:], skip_special_tokens=True)
    del inputs, generated
    return response


def classify_narrative(narrative, preliminary_record_type, may_be_truncated):
    prior_response = ""
    prior_error = ""

    for attempt in range(1, MAX_RETRIES + 1):
        user_prompt = f"""
PRELIMINARY ROUTING HINT
{preliminary_record_type}

NARRATIVE MAY BE TRUNCATED
{may_be_truncated}

NARRATIVE
{narrative}

Apply the cause hierarchy and return only the JSON object.
""".strip()

        if attempt > 1:
            user_prompt += f"""

The previous response failed validation.
ERROR
{prior_error}

PREVIOUS RESPONSE
{prior_response}

Return one corrected JSON object only.
"""

        try:
            raw_response = generate_model_response([
                {"role": "system", "content": SYSTEM_PROMPT},
                {"role": "user", "content": user_prompt},
            ])
            prior_response = raw_response
            json_text = repair_common_json_problems(extract_balanced_json(raw_response))
            response_dictionary = json.loads(json_text)
            normalized, retained, defaulted = normalize_response_dictionary(response_dictionary)
            validated = IncidentExtraction.model_validate(normalized)

            cause_valid, cause_exact, cause_normalized, cause_overlap = evaluate_evidence(
                validated.cause_evidence_text,
                narrative,
            )
            factor_valid, factor_exact, factor_normalized, factor_overlap = evaluate_evidence(
                validated.factor_evidence_text,
                narrative,
            )
            if validated.primary_contributing_factor == "NO_SEPARATE_FACTOR_STATED":
                factor_valid = True

            return {
                **validated.model_dump(),
                "SCHEMA_VALID": True,
                "FIELDS_RETAINED_FROM_MODEL": "; ".join(retained),
                "FIELDS_DEFAULTED": "; ".join(defaulted),
                "CAUSE_EVIDENCE_VALID": cause_valid,
                "CAUSE_EVIDENCE_EXACT": cause_exact,
                "CAUSE_EVIDENCE_NORMALIZED": cause_normalized,
                "CAUSE_EVIDENCE_TOKEN_OVERLAP": round(cause_overlap, 4),
                "FACTOR_EVIDENCE_VALID_STRICT": factor_valid,
                "FACTOR_EVIDENCE_EXACT": factor_exact,
                "FACTOR_EVIDENCE_NORMALIZED": factor_normalized,
                "FACTOR_EVIDENCE_TOKEN_OVERLAP": round(factor_overlap, 4),
                "FACTOR_PREVENTION_ALIGNED_ORIGINAL": factor_prevention_aligned(
                    validated.primary_contributing_factor,
                    validated.prevention_theme,
                ),
                "PROCESSING_STATUS": "SUCCESS",
                "PROCESSING_ERROR": "",
                "ATTEMPT_COUNT": attempt,
                "RAW_MODEL_RESPONSE": raw_response,
            }

        except Exception as error:
            prior_error = f"{type(error).__name__}: {error}"
            print(f"  Attempt {attempt} failed: {prior_error[:250]}")
            gc.collect()
            if torch.cuda.is_available():
                torch.cuda.empty_cache()

    return {
        **DEFAULT_RESPONSE,
        "SCHEMA_VALID": False,
        "FIELDS_RETAINED_FROM_MODEL": "",
        "FIELDS_DEFAULTED": "; ".join(EXPECTED_KEYS),
        "CAUSE_EVIDENCE_VALID": False,
        "FACTOR_EVIDENCE_VALID_STRICT": False,
        "FACTOR_PREVENTION_ALIGNED_ORIGINAL": False,
        "PROCESSING_STATUS": "FAILED",
        "PROCESSING_ERROR": prior_error,
        "ATTEMPT_COUNT": MAX_RETRIES,
        "RAW_MODEL_RESPONSE": prior_response,
    }


# =============================================================================
# 7. EXPANDED FACTOR-SUPPORT LOGIC USED IN BLOCK 2
# =============================================================================

def apply_expanded_factor_logic(model_result, narrative):
    output = dict(model_result)
    original_factor = output.get("primary_contributing_factor", "")
    original_prevention = output.get("prevention_theme", "")
    factor_evidence = output.get("factor_evidence_text", "")
    factor_confidence = safe_float(output.get("factor_confidence", 0.0))
    factor_support = normalize_support(output.get("factor_support_level", "LOW"))

    strict_valid, evidence_exact, evidence_normalized, evidence_overlap = evaluate_evidence(
        factor_evidence,
        narrative,
    )
    context_terms = find_factor_context_terms(original_factor, narrative)
    evidence_context_terms = find_factor_context_terms(original_factor, factor_evidence)
    nonempty_evidence = bool(str(factor_evidence or "").strip())

    if original_factor == "NO_SEPARATE_FACTOR_STATED":
        tier = "NO_SUPPORTED_FACTOR"
        method = "MODEL_ABSTAINED"
    elif strict_valid:
        tier = "DIRECTLY_SUPPORTED"
        method = (
            "TRACEABLE_EVIDENCE_WITH_CATEGORY_CUE"
            if evidence_context_terms
            else "TRACEABLE_MODEL_EVIDENCE"
        )
    elif context_terms:
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "CATEGORY_SPECIFIC_NARRATIVE_CUE"
    elif (
        factor_support in {"HIGH", "MEDIUM"}
        and factor_confidence >= CONTEXTUAL_FACTOR_CONFIDENCE_THRESHOLD
    ):
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "MODEL_CONFIDENCE_AND_SUPPORT"
    elif factor_confidence >= HIGH_CONFIDENCE_FACTOR_THRESHOLD and nonempty_evidence:
        tier = "CONTEXTUALLY_SUPPORTED"
        method = "HIGH_CONFIDENCE_WITH_EXPLANATORY_EVIDENCE"
    else:
        tier = "NO_SUPPORTED_FACTOR"
        method = "INSUFFICIENT_FACTOR_SUPPORT"

    if tier in {"DIRECTLY_SUPPORTED", "CONTEXTUALLY_SUPPORTED"}:
        revised_factor = original_factor
        if factor_prevention_aligned(original_factor, original_prevention):
            revised_prevention = original_prevention
            revised_rationale = output.get("prevention_rationale", "")
            prevention_method = "ORIGINAL_ALLOWED_PREVENTION"
        else:
            revised_prevention = PREFERRED_PREVENTION_BY_FACTOR.get(
                original_factor,
                "INVESTIGATION_REQUIRED",
            )
            revised_rationale = (
                "The prevention theme was assigned from the approved "
                "factor-to-prevention map."
            )
            prevention_method = "PREFERRED_MAPPING_REPAIR"
    else:
        revised_factor = "NO_SEPARATE_FACTOR_STATED"
        revised_prevention = "INVESTIGATION_REQUIRED"
        revised_rationale = (
            "Additional investigation is required before selecting a specific "
            "corrective action."
        )
        prevention_method = "UNSUPPORTED_FACTOR_ABSTENTION"

    output.update({
        "ORIGINAL_MODEL_FACTOR": original_factor,
        "ORIGINAL_MODEL_PREVENTION": original_prevention,
        "ORIGINAL_MODEL_FACTOR_EVIDENCE": factor_evidence,
        "ORIGINAL_MODEL_PREVENTION_RATIONALE": output.get("prevention_rationale", ""),
        "FACTOR_SUPPORT_TIER": tier,
        "FACTOR_SUPPORT_METHOD": method,
        "FACTOR_CONTEXT_TERMS": "; ".join(context_terms),
        "FACTOR_EVIDENCE_CONTEXT_TERMS": "; ".join(evidence_context_terms),
        "ORIGINAL_FACTOR_EVIDENCE_VALID": strict_valid,
        "ORIGINAL_FACTOR_EVIDENCE_EXACT": evidence_exact,
        "ORIGINAL_FACTOR_EVIDENCE_NORMALIZED": evidence_normalized,
        "ORIGINAL_FACTOR_EVIDENCE_TOKEN_OVERLAP": round(evidence_overlap, 4),
        "REVISED_FACTOR": revised_factor,
        "REVISED_PREVENTION": revised_prevention,
        "REVISED_PREVENTION_RATIONALE": revised_rationale,
        "REVISED_PREVENTION_METHOD": prevention_method,
        "REVISED_FACTOR_PREVENTION_ALIGNED": factor_prevention_aligned(
            revised_factor,
            revised_prevention,
        ),
    })

    return output


# =============================================================================
# 8. PREPARE DATA AND EXCLUDE BLOCK 1 + BLOCK 2
# =============================================================================

df = accidents.copy()
df.columns = df.columns.astype(str).str.strip().str.upper()

if TEXT_COLUMN not in df.columns or SEVERITY_COLUMN not in df.columns:
    raise KeyError("NARRATIVE and DEGREE_INJURY columns are required.")

id_column = next((column for column in ID_COLUMN_CANDIDATES if column in df.columns), None)
if id_column is None:
    id_column = "_SOURCE_INDEX"
    df[id_column] = df.index.astype(str)

df[id_column] = df[id_column].astype(str)
df[TEXT_COLUMN] = df[TEXT_COLUMN].astype("string").fillna("").str.strip()
df[SEVERITY_COLUMN] = df[SEVERITY_COLUMN].astype("string").fillna("UNKNOWN").str.strip()

if DATE_COLUMN in df.columns:
    df[DATE_COLUMN] = pd.to_datetime(df[DATE_COLUMN], errors="coerce")
    df["ACCIDENT_YEAR"] = df[DATE_COLUMN].dt.year.astype("Int64")
elif "CAL_YR" in df.columns:
    df["ACCIDENT_YEAR"] = pd.to_numeric(df["CAL_YR"], errors="coerce").astype("Int64")
else:
    raise KeyError("ACCIDENT_DT or CAL_YR is required.")

df["_WORD_COUNT"] = df[TEXT_COLUMN].str.split().str.len()
df["_CHARACTER_COUNT"] = df[TEXT_COLUMN].str.len()
df["_MAY_BE_TRUNCATED"] = df["_CHARACTER_COUNT"] >= TRUNCATION_CHARACTER_THRESHOLD
df["_DUPLICATE_KEY"] = (
    df[TEXT_COLUMN].str.lower().str.replace(r"\s+", " ", regex=True).str.strip()
)
df = df[df["_WORD_COUNT"] >= MIN_NARRATIVE_WORDS].copy()
df = df.drop_duplicates(subset=["_DUPLICATE_KEY"], keep="first").copy()

with open(BLOCK1_IDS_JSON, "r", encoding="utf-8") as input_file:
    block1_record_ids = set(map(str, json.load(input_file)["record_ids"]))

with open(block2_ids_path, "r", encoding="utf-8") as input_file:
    block2_record_ids = set(map(str, json.load(input_file)["record_ids"]))

excluded_ids = block1_record_ids.union(block2_record_ids)
unseen_df = df[~df[id_column].isin(excluded_ids)].copy()

if unseen_df.empty:
    raise ValueError("No records remain after excluding Block 1 and Block 2 IDs.")


def assign_severity_group(value):
    value = str(value).upper()
    if "FATAL" in value:
        return "Fatality"
    if any(term in value for term in ["PERM TOT", "PERMANENT TOTAL", "PERM PART", "PERMANENT PART"]):
        return "Permanent disability"
    if any(term in value for term in ["DAYS AWAY", "DYS AWY", "LOST WORKDAY", "LOST TIME"]):
        return "Days away"
    if any(term in value for term in ["RESTRICT", "TRANSFER", "RSTR ACT", "RESTRCTD ACT"]):
        return "Restricted or transferred"
    return "No lost time or other"


def assign_time_period(year):
    if pd.isna(year):
        return "Unknown period"
    year = int(year)
    if year <= 2008:
        return EARLY_PERIOD
    if year <= 2017:
        return MIDDLE_PERIOD
    return RECENT_PERIOD


def assign_sector(value):
    value = str(value).strip().upper()
    if value == "C":
        return "Coal"
    if value == "M":
        return "Metal/nonmetal"
    return "Unknown"


def preliminary_record_type_hint(row):
    combined = " ".join(
        str(row.get(column, "")).upper()
        for column in [TEXT_COLUMN, "CLASSIFICATION", "NATURE_INJURY", SEVERITY_COLUMN]
    )
    if any(pattern in combined for pattern in [
        "NOT DUE TO WORK", "NOT WORK RELATED", "NON-OCCUPATIONAL",
        "NATURAL CAUSE", "NOT A MINER", "HEART ATTACK", "CARDIAC",
        "STROKE", "SEIZURE",
    ]):
        return "NONWORK_RELATED_OR_NATURAL_EVENT"
    if any(pattern in combined for pattern in [
        "HEARING LOSS", "SILICOSIS", "PNEUMOCONIOSIS", "DUST DISEASE",
        "OCCUPATIONAL ILLNESS",
    ]):
        return "OCCUPATIONAL_EXPOSURE_OR_ILLNESS"
    return "ACUTE_ACCIDENT"


unseen_df["SEVERITY_GROUP"] = unseen_df[SEVERITY_COLUMN].map(assign_severity_group)
unseen_df["TIME_PERIOD"] = unseen_df["ACCIDENT_YEAR"].map(assign_time_period)
unseen_df["SECTOR_GROUP"] = (
    unseen_df["COAL_METAL_IND"].map(assign_sector)
    if "COAL_METAL_IND" in unseen_df.columns
    else "Unknown"
)
unseen_df["PRELIMINARY_RECORD_TYPE_HINT"] = unseen_df.apply(
    preliminary_record_type_hint,
    axis=1,
)

candidate_df = unseen_df[
    unseen_df["PRELIMINARY_RECORD_TYPE_HINT"].eq("ACUTE_ACCIDENT")
].copy()

if len(candidate_df) < TOTAL_REVIEW_SIZE:
    raise ValueError("Fewer than 100 eligible unseen acute-accident narratives remain.")


# =============================================================================
# 9. SELECT 100 NEW NARRATIVES
# =============================================================================

def create_review_sample(data):
    selected_indexes = []
    sampling_records = []

    for severity_number, severity_group in enumerate(SEVERITY_ORDER):
        severity_data = data[data["SEVERITY_GROUP"].eq(severity_group)].copy()
        selected_for_severity = []
        base_period_target = TARGET_PER_SEVERITY // len(TIME_PERIOD_ORDER)
        period_remainder = TARGET_PER_SEVERITY % len(TIME_PERIOD_ORDER)

        for period_number, period_name in enumerate(TIME_PERIOD_ORDER):
            requested = base_period_target + (1 if period_number < period_remainder else 0)
            period_data = severity_data[severity_data["TIME_PERIOD"].eq(period_name)]
            selected_count = min(requested, len(period_data))

            if selected_count > 0:
                chosen = period_data.sample(
                    n=selected_count,
                    random_state=RANDOM_STATE + severity_number * 100 + period_number,
                )
                selected_for_severity.extend(chosen.index.tolist())

            sampling_records.append({
                "Severity group": severity_group,
                "Time period": period_name,
                "Available": len(period_data),
                "Requested": requested,
                "Selected": selected_count,
            })

        shortage = TARGET_PER_SEVERITY - len(selected_for_severity)
        if shortage > 0:
            remaining = severity_data.drop(index=selected_for_severity, errors="ignore")
            fill_count = min(shortage, len(remaining))
            if fill_count > 0:
                fill = remaining.sample(
                    n=fill_count,
                    random_state=RANDOM_STATE + severity_number * 1000,
                )
                selected_for_severity.extend(fill.index.tolist())

        selected_indexes.extend(selected_for_severity)

    selected_indexes = list(dict.fromkeys(selected_indexes))
    sample = data.loc[selected_indexes].copy()

    if len(sample) < TOTAL_REVIEW_SIZE:
        remaining = data.drop(index=selected_indexes, errors="ignore")
        fill_count = min(TOTAL_REVIEW_SIZE - len(sample), len(remaining))
        if fill_count > 0:
            sample = pd.concat([
                sample,
                remaining.sample(n=fill_count, random_state=RANDOM_STATE + 9999),
            ])

    if len(sample) > TOTAL_REVIEW_SIZE:
        sample = sample.sample(n=TOTAL_REVIEW_SIZE, random_state=RANDOM_STATE)

    sample = sample.sample(frac=1, random_state=RANDOM_STATE).reset_index(drop=True)
    return sample, pd.DataFrame(sampling_records)


block3_df, sampling_summary = create_review_sample(candidate_df)

if len(block3_df) != TOTAL_REVIEW_SIZE:
    raise ValueError(f"Expected 100 records but selected {len(block3_df)}.")

if set(block3_df[id_column]).intersection(excluded_ids):
    raise RuntimeError("Block 3 contains records previously used in Block 1 or Block 2.")

with open(BLOCK3_IDS_JSON, "w", encoding="utf-8") as output_file:
    json.dump({
        "created_at": datetime.now().isoformat(timespec="seconds"),
        "purpose": "Block 3 independent 100-record human-validation sample",
        "id_column": id_column,
        "record_count": len(block3_df),
        "record_ids": block3_df[id_column].astype(str).tolist(),
    }, output_file, indent=2, ensure_ascii=False)

print("Block 3 records selected:", len(block3_df))
print(pd.crosstab(block3_df["SEVERITY_GROUP"], block3_df["TIME_PERIOD"]))


# =============================================================================
# 10. CHECKPOINT UTILITIES AND PROCESSING
# =============================================================================

def load_checkpoint(path):
    records = {}
    path = Path(path)
    if not path.exists():
        return records
    with path.open("r", encoding="utf-8") as input_file:
        for line in input_file:
            try:
                record = json.loads(line.strip())
                records[str(record["SOURCE_RECORD_ID"])] = record
            except Exception:
                continue
    return records


def append_checkpoint(record, path):
    with open(path, "a", encoding="utf-8") as output_file:
        output_file.write(json.dumps(record, ensure_ascii=False, default=str) + "\n")


checkpoint_records = load_checkpoint(BLOCK3_CHECKPOINT_JSONL)
processed_records = []
new_model_calls = 0
reused_records = 0

context_columns = [
    column for column in [
        "ACCIDENT_DT", "ACCIDENT_YEAR", "TIME_PERIOD", "SEVERITY_GROUP",
        "SECTOR_GROUP", "CLASSIFICATION", "ACCIDENT_TYPE", "ACTIVITY",
        "INJURY_SOURCE", "NATURE_INJURY", "INJ_BODY_PART", "DEGREE_INJURY",
        "MINING_EQUIP", "OCCUPATION", "COAL_METAL_IND", "DAYS_RESTRICT",
        "DAYS_LOST", "PRELIMINARY_RECORD_TYPE_HINT",
    ]
    if column in block3_df.columns
]

for row_number, row in block3_df.iterrows():
    record_id = str(row[id_column])

    if not REPROCESS_COMPLETED and record_id in checkpoint_records:
        print(f"[{row_number + 1}/100] Checkpoint: {record_id}")
        processed_records.append(checkpoint_records[record_id])
        reused_records += 1
        continue

    narrative = str(row[TEXT_COLUMN])
    start_time = time.time()
    print(f"[{row_number + 1}/100] New inference: {record_id}")

    model_result = classify_narrative(
        narrative=narrative,
        preliminary_record_type=str(row["PRELIMINARY_RECORD_TYPE_HINT"]),
        may_be_truncated=bool(row["_MAY_BE_TRUNCATED"]),
    )
    model_result = apply_expanded_factor_logic(model_result, narrative)

    record = {
        "SOURCE_RECORD_ID": record_id,
        "NARRATIVE": narrative,
        "NARRATIVE_WORD_COUNT": int(row["_WORD_COUNT"]),
        "NARRATIVE_CHARACTER_COUNT": int(row["_CHARACTER_COUNT"]),
        "NARRATIVE_MAY_BE_TRUNCATED": bool(row["_MAY_BE_TRUNCATED"]),
        **{
            column: None if pd.isna(row[column]) else row[column]
            for column in context_columns
        },
        **model_result,
        "MODEL_ID": MODEL_ID,
        "PROMPT_VERSION": PROMPT_VERSION,
        "PROCESSING_SECONDS": round(time.time() - start_time, 3),
        "PROCESSED_AT": datetime.now().isoformat(timespec="seconds"),
    }

    processed_records.append(record)
    append_checkpoint(record, BLOCK3_CHECKPOINT_JSONL)
    new_model_calls += 1
    gc.collect()
    if torch.cuda.is_available():
        torch.cuda.empty_cache()


# =============================================================================
# 11. SAVE MODEL RESULTS
# =============================================================================

results_df = pd.DataFrame(processed_records)
results_df["SOURCE_RECORD_ID"] = results_df["SOURCE_RECORD_ID"].astype(str)
results_df.to_csv(BLOCK3_RESULTS_CSV, index=False, encoding="utf-8-sig")

print("Existing Block 3 records reused:", reused_records)
print("New Block 3 model calls:", new_model_calls)


# =============================================================================
# 12. CREATE HUMAN-REVIEW WORKBOOK
# =============================================================================

wb = Workbook()
ws_review = wb.active
ws_review.title = "Human Review"
ws_model = wb.create_sheet("Model Output")
ws_ontology = wb.create_sheet("Ontology Definitions")
ws_instructions = wb.create_sheet("Instructions")
ws_lists = wb.create_sheet("Lists")

# Theme colors.
NAVY = "1F4E78"
BLUE = "D9EAF7"
LIGHT_BLUE = "EAF3F8"
GREEN = "E2F0D9"
YELLOW = "FFF2CC"
RED = "FCE4D6"
GRAY = "E7E6E6"
WHITE = "FFFFFF"
DARK_GRAY = "666666"
THIN_GRAY = Side(style="thin", color="B7B7B7")

header_fill = PatternFill("solid", fgColor=NAVY)
subheader_fill = PatternFill("solid", fgColor=BLUE)
human_fill = PatternFill("solid", fgColor=GREEN)
input_fill = PatternFill("solid", fgColor=YELLOW)
model_fill = PatternFill("solid", fgColor=LIGHT_BLUE)
warning_fill = PatternFill("solid", fgColor=RED)
header_font = Font(color=WHITE, bold=True)
section_font = Font(color=NAVY, bold=True, size=12)

# Instructions sheet first content.
instruction_rows = [
    ["Block 3 Human Review Instructions"],
    ["Purpose", "Create an independent human reference label for 100 narratives that were not used in Block 1 or Block 2."],
    ["Review order", "Complete the Human Review worksheet before opening the Model Output worksheet whenever practical."],
    ["Primary cause", "Select the immediate accident mechanism described by the narrative."],
    ["Contributing factor", "Select a separate condition explaining why the event occurred. Use NO_SEPARATE_FACTOR_STATED when the narrative describes what happened but not why."],
    ["Prevention theme", "Select the most appropriate prevention category supported by the human-assigned factor. Use INVESTIGATION_REQUIRED when no separate factor is supported."],
    ["Evidence support", "Use YES when the narrative supports the selected label, NO when it does not, and UNCERTAIN when the wording is ambiguous."],
    ["Overall acceptable", "Judge whether the model's combined cause, factor, and prevention output is reasonable after completing the independent labels."],
    ["Review completed", "Set to YES only after all required human fields have been reviewed."],
    ["Do not edit", "Do not modify SOURCE_RECORD_ID values or the Model Output worksheet."],
    ["Block 4", "Block 4 will compare the completed human labels with the model results and generate the final validation PDF."],
]

for row in instruction_rows:
    ws_instructions.append(row)

ws_instructions["A1"].font = Font(size=16, bold=True, color=NAVY)
ws_instructions.merge_cells("A1:B1")
ws_instructions.column_dimensions["A"].width = 24
ws_instructions.column_dimensions["B"].width = 110
for row in range(2, ws_instructions.max_row + 1):
    ws_instructions.cell(row, 1).font = Font(bold=True, color=NAVY)
    ws_instructions.cell(row, 2).alignment = Alignment(wrap_text=True, vertical="top")
    ws_instructions.row_dimensions[row].height = 38
ws_instructions.sheet_view.showGridLines = False

# Lists sheet for validation.
list_columns = {
    "A": ("RECORD_TYPES", RECORD_TYPE_CATEGORIES),
    "B": ("CAUSES", CAUSE_CATEGORIES),
    "C": ("FACTORS", FACTOR_CATEGORIES),
    "D": ("PREVENTIONS", PREVENTION_CATEGORIES),
    "E": ("YES_NO_UNCERTAIN", ["YES", "NO", "UNCERTAIN"]),
    "F": ("YES_NO", ["YES", "NO"]),
}

for column_letter, (header, values) in list_columns.items():
    ws_lists[f"{column_letter}1"] = header
    ws_lists[f"{column_letter}1"].font = Font(bold=True)
    for index, value in enumerate(values, start=2):
        ws_lists[f"{column_letter}{index}"] = value

ws_lists.sheet_state = "hidden"

# Human Review sheet.
human_headers = [
    "REVIEW_NUMBER",
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "SEVERITY_GROUP",
    "TIME_PERIOD",
    "SECTOR_GROUP",
    "HUMAN_RECORD_TYPE",
    "HUMAN_PRIMARY_CAUSE",
    "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    "HUMAN_CONTRIBUTING_FACTOR",
    "HUMAN_FACTOR_EVIDENCE_SUPPORTED",
    "HUMAN_PREVENTION_THEME",
    "HUMAN_PREVENTION_APPROPRIATE",
    "HUMAN_OVERALL_ACCEPTABLE",
    "HUMAN_REVIEW_COMMENTS",
    "REVIEW_COMPLETED",
]

ws_review.append(human_headers)

review_data = results_df.sort_values(
    ["SEVERITY_GROUP", "TIME_PERIOD", "SOURCE_RECORD_ID"],
    kind="stable",
).reset_index(drop=True)

for index, row in review_data.iterrows():
    ws_review.append([
        index + 1,
        row.get("SOURCE_RECORD_ID", ""),
        row.get("NARRATIVE", ""),
        row.get("SEVERITY_GROUP", ""),
        row.get("TIME_PERIOD", ""),
        row.get("SECTOR_GROUP", ""),
        "",
        "",
        "",
        "",
        "",
        "",
        "",
        "",
        "",
        "NO",
    ])

for cell in ws_review[1]:
    cell.fill = header_fill
    cell.font = header_font
    cell.alignment = Alignment(wrap_text=True, horizontal="center", vertical="center")
    cell.border = Border(bottom=THIN_GRAY)

ws_review.freeze_panes = "A2"
ws_review.auto_filter.ref = f"A1:P{ws_review.max_row}"
ws_review.sheet_view.showGridLines = False
ws_review.row_dimensions[1].height = 45

human_widths = {
    "A": 12, "B": 20, "C": 85, "D": 24, "E": 24, "F": 18,
    "G": 34, "H": 42, "I": 22, "J": 46, "K": 22,
    "L": 46, "M": 22, "N": 22, "O": 55, "P": 20,
}
for column, width in human_widths.items():
    ws_review.column_dimensions[column].width = width

for row_number in range(2, ws_review.max_row + 1):
    ws_review.row_dimensions[row_number].height = 90
    for column_number in range(1, 17):
        cell = ws_review.cell(row_number, column_number)
        cell.alignment = Alignment(wrap_text=True, vertical="top")
        cell.border = Border(bottom=THIN_GRAY)
        if column_number >= 7:
            cell.fill = input_fill
        else:
            cell.fill = PatternFill("solid", fgColor=WHITE)

# Validation lists.
def add_list_validation(sheet, cell_range, formula):
    validation = DataValidation(type="list", formula1=formula, allow_blank=True)
    validation.error = "Select a value from the dropdown list."
    validation.errorTitle = "Invalid selection"
    validation.prompt = "Use the dropdown to select an ontology value."
    validation.promptTitle = "Human review field"
    validation.showErrorMessage = True
    validation.showInputMessage = True
    sheet.add_data_validation(validation)
    validation.add(cell_range)

last_review_row = ws_review.max_row
add_list_validation(ws_review, f"G2:G{last_review_row}", f"=Lists!$A$2:$A${len(RECORD_TYPE_CATEGORIES)+1}")
add_list_validation(ws_review, f"H2:H{last_review_row}", f"=Lists!$B$2:$B${len(CAUSE_CATEGORIES)+1}")
add_list_validation(ws_review, f"I2:I{last_review_row}", "=Lists!$E$2:$E$4")
add_list_validation(ws_review, f"J2:J{last_review_row}", f"=Lists!$C$2:$C${len(FACTOR_CATEGORIES)+1}")
add_list_validation(ws_review, f"K2:K{last_review_row}", "=Lists!$E$2:$E$4")
add_list_validation(ws_review, f"L2:L{last_review_row}", f"=Lists!$D$2:$D${len(PREVENTION_CATEGORIES)+1}")
add_list_validation(ws_review, f"M2:N{last_review_row}", "=Lists!$E$2:$E$4")
add_list_validation(ws_review, f"P2:P{last_review_row}", "=Lists!$F$2:$F$3")

# Conditional formatting for incomplete reviews.
ws_review.conditional_formatting.add(
    f"A2:P{last_review_row}",
    FormulaRule(formula=["$P2=\"NO\""], fill=warning_fill),
)
ws_review.conditional_formatting.add(
    f"A2:P{last_review_row}",
    FormulaRule(formula=["$P2=\"YES\""], fill=PatternFill("solid", fgColor="E2F0D9")),
)

# Add table.
human_table = Table(displayName="HumanReviewTable", ref=f"A1:P{last_review_row}")
human_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium2",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False,
)
ws_review.add_table(human_table)

# Model Output sheet.
model_headers = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "SEVERITY_GROUP",
    "TIME_PERIOD",
    "SECTOR_GROUP",
    "MODEL_RECORD_TYPE",
    "MODEL_PRIMARY_CAUSE",
    "MODEL_CAUSE_EVIDENCE",
    "MODEL_CAUSE_CONFIDENCE",
    "MODEL_CAUSE_SUPPORT",
    "MODEL_ORIGINAL_FACTOR",
    "MODEL_FACTOR_EVIDENCE",
    "MODEL_FACTOR_CONFIDENCE",
    "MODEL_FACTOR_SUPPORT",
    "MODEL_FACTOR_SUPPORT_TIER",
    "MODEL_FACTOR_SUPPORT_METHOD",
    "MODEL_FACTOR_CONTEXT_TERMS",
    "MODEL_REVISED_FACTOR",
    "MODEL_ORIGINAL_PREVENTION",
    "MODEL_REVISED_PREVENTION",
    "MODEL_PREVENTION_RATIONALE",
    "MODEL_PREVENTION_CONFIDENCE",
    "MODEL_PREVENTION_SUPPORT",
    "MODEL_PREVENTION_METHOD",
    "MODEL_SCHEMA_VALID",
    "MODEL_CAUSE_EVIDENCE_VALID",
    "MODEL_STRICT_FACTOR_EVIDENCE_VALID",
    "MODEL_PROCESSING_STATUS",
    "MODEL_ATTEMPT_COUNT",
]

ws_model.append(model_headers)

for _, row in review_data.iterrows():
    ws_model.append([
        row.get("SOURCE_RECORD_ID", ""),
        row.get("NARRATIVE", ""),
        row.get("SEVERITY_GROUP", ""),
        row.get("TIME_PERIOD", ""),
        row.get("SECTOR_GROUP", ""),
        row.get("record_type", ""),
        row.get("primary_cause", ""),
        row.get("cause_evidence_text", ""),
        row.get("cause_confidence", ""),
        row.get("cause_support_level", ""),
        row.get("ORIGINAL_MODEL_FACTOR", row.get("primary_contributing_factor", "")),
        row.get("factor_evidence_text", ""),
        row.get("factor_confidence", ""),
        row.get("factor_support_level", ""),
        row.get("FACTOR_SUPPORT_TIER", ""),
        row.get("FACTOR_SUPPORT_METHOD", ""),
        row.get("FACTOR_CONTEXT_TERMS", ""),
        row.get("REVISED_FACTOR", ""),
        row.get("ORIGINAL_MODEL_PREVENTION", row.get("prevention_theme", "")),
        row.get("REVISED_PREVENTION", ""),
        row.get("REVISED_PREVENTION_RATIONALE", ""),
        row.get("prevention_confidence", ""),
        row.get("prevention_support_level", ""),
        row.get("REVISED_PREVENTION_METHOD", ""),
        row.get("SCHEMA_VALID", ""),
        row.get("CAUSE_EVIDENCE_VALID", ""),
        row.get("FACTOR_EVIDENCE_VALID_STRICT", ""),
        row.get("PROCESSING_STATUS", ""),
        row.get("ATTEMPT_COUNT", ""),
    ])

for cell in ws_model[1]:
    cell.fill = header_fill
    cell.font = header_font
    cell.alignment = Alignment(wrap_text=True, horizontal="center", vertical="center")

ws_model.freeze_panes = "A2"
ws_model.auto_filter.ref = f"A1:AC{ws_model.max_row}"
ws_model.sheet_view.showGridLines = False
ws_model.row_dimensions[1].height = 60

for column_number in range(1, 30):
    column_letter = get_column_letter(column_number)
    ws_model.column_dimensions[column_letter].width = 20

for column_letter in ["B", "H", "L", "Q", "U"]:
    ws_model.column_dimensions[column_letter].width = 65

for row_number in range(2, ws_model.max_row + 1):
    ws_model.row_dimensions[row_number].height = 75
    for column_number in range(1, 30):
        ws_model.cell(row_number, column_number).alignment = Alignment(
            wrap_text=True,
            vertical="top",
        )

model_table = Table(displayName="ModelOutputTable", ref=f"A1:AC{ws_model.max_row}")
model_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium9",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False,
)
ws_model.add_table(model_table)

# Ontology Definitions sheet.
definition_rows = [
    ["Ontology", "Category", "Review guidance"],
]

for category in RECORD_TYPE_CATEGORIES:
    definition_rows.append(["Record type", category, "Select the record population described by the narrative."])

cause_guidance = {
    "CAUGHT_IN_OR_BETWEEN": "Trapped, pinched, crushed, compressed, or entangled between objects or parts.",
    "STRUCK_BY_MOVING_OBJECT": "Moving or falling object contacts the person without trapping or compression.",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT": "The event begins with explicit loss of control of a tool, load, hose, handle, or component.",
    "UNEXPECTED_EQUIPMENT_MOVEMENT": "Equipment moves unexpectedly without explicit worker loss of control.",
    "SLIP_TRIP_OR_FALL": "Loss of footing, balance, or elevation.",
    "MANUAL_HANDLING_OR_OVEREXERTION": "Acute injury from lifting, pushing, pulling, carrying, or exerting force.",
}
for category in CAUSE_CATEGORIES:
    definition_rows.append([
        "Primary cause",
        category,
        cause_guidance.get(category, "Use the immediate event mechanism directly supported by the narrative."),
    ])

factor_guidance = {
    "NO_SEPARATE_FACTOR_STATED": "Use when what happened is clear but the narrative does not support why it happened.",
    "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION": "Requires an explicit walking, standing, access, or surface hazard.",
    "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL": "Requires explicit instability, movement, or improper positioning.",
    "LINE_OF_FIRE_BODY_POSITION": "Requires body position in a pinch point or path of movement or released energy.",
    "EQUIPMENT_CONDITION_OR_FAILURE": "Requires stated failure, damage, wear, leakage, defect, or malfunction.",
}
for category in FACTOR_CATEGORIES:
    definition_rows.append([
        "Contributing factor",
        category,
        factor_guidance.get(category, "Select only when a separate contributing condition is supported."),
    ])

for category in PREVENTION_CATEGORIES:
    definition_rows.append([
        "Prevention theme",
        category,
        "Select the prevention category that follows from the human-assigned contributing factor.",
    ])

for row in definition_rows:
    ws_ontology.append(row)

for cell in ws_ontology[1]:
    cell.fill = header_fill
    cell.font = header_font
    cell.alignment = Alignment(horizontal="center", vertical="center")

ws_ontology.freeze_panes = "A2"
ws_ontology.auto_filter.ref = f"A1:C{ws_ontology.max_row}"
ws_ontology.column_dimensions["A"].width = 24
ws_ontology.column_dimensions["B"].width = 52
ws_ontology.column_dimensions["C"].width = 95
ws_ontology.sheet_view.showGridLines = False

for row_number in range(2, ws_ontology.max_row + 1):
    ws_ontology.row_dimensions[row_number].height = 36
    for column_number in range(1, 4):
        ws_ontology.cell(row_number, column_number).alignment = Alignment(
            wrap_text=True,
            vertical="top",
        )

ontology_table = Table(
    displayName="OntologyTable",
    ref=f"A1:C{ws_ontology.max_row}",
)
ontology_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium4",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False,
)
ws_ontology.add_table(ontology_table)

# Make Instructions the first sheet, followed by Human Review.
wb._sheets = [ws_instructions, ws_review, ws_model, ws_ontology, ws_lists]
wb.active = 0

# Workbook properties.
wb.properties.title = "MSHA Block 3 Human Review Workbook"
wb.properties.subject = "Independent human validation of 100 unseen MSHA accident narratives"
wb.properties.creator = "Python and Qwen2.5-7B-Instruct"
wb.properties.description = (
    "Human-review workbook containing 100 narratives not used in Block 1 or Block 2, "
    "with model output isolated on a separate worksheet."
)

wb.save(BLOCK3_REVIEW_XLSX)


# =============================================================================
# 13. FINAL SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 3 COMPLETE")
print("=" * 95)
print("Validation narratives:", len(results_df))
print("Existing checkpoint records reused:", reused_records)
print("New model calls:", new_model_calls)
print("Workbook:", os.path.abspath(BLOCK3_REVIEW_XLSX))
print("Model results:", os.path.abspath(BLOCK3_RESULTS_CSV))
print("Checkpoint:", os.path.abspath(BLOCK3_CHECKPOINT_JSONL))
print("Record IDs:", os.path.abspath(BLOCK3_IDS_JSON))
print("=" * 95)

print("\nFactor-support tiers:")
print(results_df["FACTOR_SUPPORT_TIER"].value_counts(dropna=False))

print("\nSample by severity and period:")
print(pd.crosstab(results_df["SEVERITY_GROUP"], results_df["TIME_PERIOD"]))
````

## Create human review comparison report

````python
# =============================================================================
# BLOCK 4 - HUMAN VERSUS MODEL VALIDATION REPORT
#
# INPUT
# -----
# MSHA_Block3_Human_Review_Workbook- Completed.xlsx
#
# OUTPUTS
# -------
# 1. MSHA_Block4_Human_Validation_Report.pdf
# 2. MSHA_Block4_Human_Model_Comparison.csv
# 3. MSHA_Block4_Disagreements_for_Review.xlsx
#
# PURPOSE
# -------
# Compare the completed 100-record human review against the Block 3 model
# output. The report distinguishes:
#
#   - Exact category agreement
#   - Reasonably related category agreement
#   - Material disagreement
#   - Human labels marked uncertain
#   - Blank or incomplete human fields
#
# The related-category analysis is intentionally secondary. Exact agreement
# remains the primary metric. Related agreement acknowledges that short MSHA
# narratives may reasonably support more than one nearby ontology category.
# =============================================================================

import os
import re
import math
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from matplotlib.backends.backend_pdf import PdfPages
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    precision_recall_fscore_support,
    cohen_kappa_score,
)

from openpyxl import Workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.worksheet.table import Table, TableStyleInfo
from openpyxl.utils import get_column_letter

warnings.filterwarnings("ignore")


# =============================================================================
# 1. SETTINGS
# =============================================================================

INPUT_XLSX = "MSHA_Block3_Human_Review_Workbook- Completed.xlsx"
OUTPUT_PDF = "MSHA_Block4_Human_Validation_Report.pdf"
OUTPUT_COMPARISON_CSV = "MSHA_Block4_Human_Model_Comparison.csv"
OUTPUT_DISAGREEMENTS_XLSX = "MSHA_Block4_Disagreements_for_Review.xlsx"

HUMAN_SHEET = "Human Review"
MODEL_SHEET = "Model Output"
EXPECTED_RECORDS = 100
MAX_EXAMPLES_PER_SECTION = 12
TOP_CONFUSIONS = 20

YES_VALUES = {"YES", "Y", "TRUE", "1"}
NO_VALUES = {"NO", "N", "FALSE", "0"}
UNCERTAIN_VALUES = {"UNCERTAIN", "UNSURE", "MAYBE", "UNKNOWN", "?"}


# =============================================================================
# 2. RELATED-CATEGORY GROUPS
#
# Exact matching is always reported separately. These groups are used only for
# a secondary "exact or related" analysis.
# =============================================================================

CAUSE_RELATION_GROUPS = {
    "ENTRAPMENT_OR_OBJECT_CONTACT": {
        "CAUGHT_IN_OR_BETWEEN",
        "STRUCK_BY_MOVING_OBJECT",
        "MATERIAL_RELEASE_OR_SHIFT",
    },
    "EQUIPMENT_MOVEMENT_OR_CONTROL": {
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
        "STRUCK_BY_MOVING_OBJECT",
    },
    "BODY_MOTION_OR_EXERTION": {
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
        "MANUAL_HANDLING_OR_OVEREXERTION",
        "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    },
    "FALL_OR_SURFACE_CONTACT": {
        "SLIP_TRIP_OR_FALL",
        "STRUCK_AGAINST_STATIONARY_OBJECT",
    },
    "ENERGY_OR_EQUIPMENT_EVENT": {
        "PRESSURE_OR_STORED_ENERGY_RELEASE",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        "ELECTRICAL_CONTACT",
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    },
    "GROUND_OR_MOVING_MATERIAL": {
        "GROUND_OR_ROOF_FAILURE",
        "MATERIAL_RELEASE_OR_SHIFT",
        "STRUCK_BY_MOVING_OBJECT",
    },
    "EXPOSURE": {
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
        "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
        "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    },
}

FACTOR_RELATION_GROUPS = {
    "MATERIAL_STABILITY_OR_SECUREMENT": {
        "UNSTABLE_OR_IMPROPERLY_POSITIONED_MATERIAL",
        "INADEQUATE_SECUREMENT_OR_RESTRAINT",
    },
    "POSITION_OR_WORK_METHOD": {
        "LINE_OF_FIRE_BODY_POSITION",
        "PROCEDURE_OR_WORK_METHOD",
        "INADEQUATE_HAZARD_RECOGNITION",
        "COMMUNICATION_OR_COORDINATION",
    },
    "PHYSICAL_WORK_ENVIRONMENT": {
        "POOR_HOUSEKEEPING_OR_SURFACE_CONDITION",
        "ENVIRONMENTAL_CONDITION",
        "WORKSPACE_OR_ACCESS_LIMITATION",
    },
    "EQUIPMENT_OR_ENERGY_CONTROL": {
        "ENERGY_ISOLATION_DEFICIENCY",
        "EQUIPMENT_CONDITION_OR_FAILURE",
        "INADEQUATE_GUARDING_OR_BARRIER",
    },
    "ADMINISTRATIVE_OR_PERSONAL_CONTROL": {
        "TRAINING_OR_EXPERIENCE",
        "PPE_OR_PROTECTIVE_CONTROL",
        "INADEQUATE_HAZARD_RECOGNITION",
        "PROCEDURE_OR_WORK_METHOD",
    },
}

PREVENTION_RELATION_GROUPS = {
    "MATERIAL_AND_LINE_OF_FIRE_CONTROL": {
        "SECURE_OR_REPOSITION_MATERIAL",
        "ESTABLISH_LINE_OF_FIRE_EXCLUSION",
        "USE_MECHANICAL_HANDLING_ASSISTANCE",
    },
    "ENERGY_EQUIPMENT_AND_GUARDING": {
        "ISOLATE_OR_CONTROL_STORED_ENERGY",
        "REPAIR_REPLACE_OR_INSPECT_EQUIPMENT",
        "INSTALL_OR_IMPROVE_GUARDING",
        "ENGINEERING_REVIEW_REQUIRED",
    },
    "WORK_PLANNING_AND_HAZARD_REVIEW": {
        "REVISE_PROCEDURE_OR_JOB_PLAN",
        "CONDUCT_PRE_JOB_HAZARD_REVIEW",
        "PROVIDE_TARGETED_TRAINING",
        "IMPROVE_COMMUNICATION_OR_SPOTTER_USE",
    },
    "WORKPLACE_CONDITION_CONTROL": {
        "IMPROVE_WORK_SURFACE_OR_HOUSEKEEPING",
        "IMPROVE_GROUND_CONTROL_OR_SCALING",
        "CONTROL_ENVIRONMENTAL_EXPOSURE",
        "REDESIGN_WORKSPACE_OR_ACCESS",
    },
    "PROTECTIVE_CONTROL": {
        "IMPROVE_PPE_OR_PROTECTIVE_EQUIPMENT",
        "PROVIDE_TARGETED_TRAINING",
    },
    "INSUFFICIENT_BASIS": {
        "INVESTIGATION_REQUIRED",
        "INSUFFICIENT_INFORMATION",
        "ENGINEERING_REVIEW_REQUIRED",
    },
}


# =============================================================================
# 3. VALIDATE AND LOAD WORKBOOK
# =============================================================================

if not Path(INPUT_XLSX).exists():
    raise FileNotFoundError(
        f"Required completed workbook was not found: {INPUT_XLSX}"
    )

xls = pd.ExcelFile(INPUT_XLSX, engine="openpyxl")

missing_sheets = [
    sheet for sheet in [HUMAN_SHEET, MODEL_SHEET]
    if sheet not in xls.sheet_names
]

if missing_sheets:
    raise ValueError(
        "The completed workbook is missing required worksheets: "
        + ", ".join(missing_sheets)
    )

human_df = pd.read_excel(INPUT_XLSX, sheet_name=HUMAN_SHEET, engine="openpyxl")
model_df = pd.read_excel(INPUT_XLSX, sheet_name=MODEL_SHEET, engine="openpyxl")

human_df.columns = human_df.columns.astype(str).str.strip().str.upper()
model_df.columns = model_df.columns.astype(str).str.strip().str.upper()

required_human_columns = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "SEVERITY_GROUP",
    "TIME_PERIOD",
    "HUMAN_RECORD_TYPE",
    "HUMAN_PRIMARY_CAUSE",
    "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    "HUMAN_CONTRIBUTING_FACTOR",
    "HUMAN_FACTOR_EVIDENCE_SUPPORTED",
    "HUMAN_PREVENTION_THEME",
    "HUMAN_PREVENTION_APPROPRIATE",
    "HUMAN_OVERALL_ACCEPTABLE",
    "HUMAN_REVIEW_COMMENTS",
    "REVIEW_COMPLETED",
]

required_model_columns = [
    "SOURCE_RECORD_ID",
    "MODEL_RECORD_TYPE",
    "MODEL_PRIMARY_CAUSE",
    "MODEL_REVISED_FACTOR",
    "MODEL_REVISED_PREVENTION",
    "MODEL_FACTOR_SUPPORT_TIER",
    "MODEL_FACTOR_SUPPORT_METHOD",
    "MODEL_CAUSE_CONFIDENCE",
    "MODEL_FACTOR_CONFIDENCE",
    "MODEL_PREVENTION_CONFIDENCE",
]

missing_human_columns = [
    column for column in required_human_columns
    if column not in human_df.columns
]
missing_model_columns = [
    column for column in required_model_columns
    if column not in model_df.columns
]

if missing_human_columns:
    raise ValueError(
        "Human Review worksheet is missing columns: "
        + ", ".join(missing_human_columns)
    )

if missing_model_columns:
    raise ValueError(
        "Model Output worksheet is missing columns: "
        + ", ".join(missing_model_columns)
    )

human_df["SOURCE_RECORD_ID"] = human_df["SOURCE_RECORD_ID"].astype(str).str.strip()
model_df["SOURCE_RECORD_ID"] = model_df["SOURCE_RECORD_ID"].astype(str).str.strip()

if human_df["SOURCE_RECORD_ID"].duplicated().any():
    duplicated = human_df.loc[
        human_df["SOURCE_RECORD_ID"].duplicated(False),
        "SOURCE_RECORD_ID",
    ].tolist()
    raise ValueError(f"Duplicate record IDs in Human Review: {duplicated}")

if model_df["SOURCE_RECORD_ID"].duplicated().any():
    duplicated = model_df.loc[
        model_df["SOURCE_RECORD_ID"].duplicated(False),
        "SOURCE_RECORD_ID",
    ].tolist()
    raise ValueError(f"Duplicate record IDs in Model Output: {duplicated}")

comparison_df = human_df.merge(
    model_df,
    on="SOURCE_RECORD_ID",
    how="outer",
    suffixes=("_HUMAN_SHEET", "_MODEL_SHEET"),
    indicator=True,
)

unmatched = comparison_df[comparison_df["_merge"] != "both"]
if not unmatched.empty:
    raise ValueError(
        "Human and model worksheets do not contain identical record IDs. "
        f"Unmatched rows: {len(unmatched)}"
    )

comparison_df = comparison_df.drop(columns=["_merge"])

# Prefer human-sheet context fields after merge.
for base_column in ["NARRATIVE", "SEVERITY_GROUP", "TIME_PERIOD", "SECTOR_GROUP"]:
    human_name = f"{base_column}_HUMAN_SHEET"
    model_name = f"{base_column}_MODEL_SHEET"
    if human_name in comparison_df.columns:
        comparison_df[base_column] = comparison_df[human_name]
    elif base_column in comparison_df.columns:
        pass
    elif model_name in comparison_df.columns:
        comparison_df[base_column] = comparison_df[model_name]


# =============================================================================
# 4. NORMALIZATION AND REVIEW-STATUS RULES
# =============================================================================

def normalize_category(value):
    if pd.isna(value):
        return ""
    return (
        str(value).strip().upper()
        .replace("-", "_")
        .replace(" ", "_")
        .replace("/", "_")
    )


def normalize_review_value(value):
    if pd.isna(value):
        return ""
    value = str(value).strip().upper()
    if value in YES_VALUES:
        return "YES"
    if value in NO_VALUES:
        return "NO"
    if value in UNCERTAIN_VALUES:
        return "UNCERTAIN"
    return value


category_columns = [
    "HUMAN_RECORD_TYPE",
    "HUMAN_PRIMARY_CAUSE",
    "HUMAN_CONTRIBUTING_FACTOR",
    "HUMAN_PREVENTION_THEME",
    "MODEL_RECORD_TYPE",
    "MODEL_PRIMARY_CAUSE",
    "MODEL_REVISED_FACTOR",
    "MODEL_REVISED_PREVENTION",
    "MODEL_FACTOR_SUPPORT_TIER",
    "MODEL_FACTOR_SUPPORT_METHOD",
]

for column in category_columns:
    comparison_df[column] = comparison_df[column].map(normalize_category)

review_value_columns = [
    "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    "HUMAN_FACTOR_EVIDENCE_SUPPORTED",
    "HUMAN_PREVENTION_APPROPRIATE",
    "HUMAN_OVERALL_ACCEPTABLE",
    "REVIEW_COMPLETED",
]

for column in review_value_columns:
    comparison_df[column] = comparison_df[column].map(normalize_review_value)

for column in [
    "MODEL_CAUSE_CONFIDENCE",
    "MODEL_FACTOR_CONFIDENCE",
    "MODEL_PREVENTION_CONFIDENCE",
]:
    comparison_df[column] = pd.to_numeric(comparison_df[column], errors="coerce")

comparison_df["REVIEW_COMPLETE_FLAG"] = comparison_df["REVIEW_COMPLETED"].eq("YES")
comparison_df["CAUSE_HUMAN_UNCERTAIN"] = comparison_df[
    "HUMAN_CAUSE_EVIDENCE_SUPPORTED"
].eq("UNCERTAIN")
comparison_df["FACTOR_HUMAN_UNCERTAIN"] = comparison_df[
    "HUMAN_FACTOR_EVIDENCE_SUPPORTED"
].eq("UNCERTAIN")
comparison_df["PREVENTION_HUMAN_UNCERTAIN"] = comparison_df[
    "HUMAN_PREVENTION_APPROPRIATE"
].eq("UNCERTAIN")
comparison_df["ANY_HUMAN_UNCERTAIN"] = comparison_df[
    [
        "CAUSE_HUMAN_UNCERTAIN",
        "FACTOR_HUMAN_UNCERTAIN",
        "PREVENTION_HUMAN_UNCERTAIN",
    ]
].any(axis=1)

comparison_df["CAUSE_EVALUABLE"] = (
    comparison_df["REVIEW_COMPLETE_FLAG"]
    & comparison_df["HUMAN_PRIMARY_CAUSE"].ne("")
)
comparison_df["FACTOR_EVALUABLE"] = (
    comparison_df["REVIEW_COMPLETE_FLAG"]
    & comparison_df["HUMAN_CONTRIBUTING_FACTOR"].ne("")
)
comparison_df["PREVENTION_EVALUABLE"] = (
    comparison_df["REVIEW_COMPLETE_FLAG"]
    & comparison_df["HUMAN_PREVENTION_THEME"].ne("")
)


# =============================================================================
# 5. EXACT AND RELATED AGREEMENT
# =============================================================================

def categories_related(human_value, model_value, relation_groups):
    if not human_value or not model_value:
        return False
    if human_value == model_value:
        return True
    return any(
        human_value in categories and model_value in categories
        for categories in relation_groups.values()
    )


def relation_group_names(human_value, model_value, relation_groups):
    groups = [
        group_name
        for group_name, categories in relation_groups.items()
        if human_value in categories and model_value in categories
    ]
    return "; ".join(groups)


comparison_df["RECORD_TYPE_EXACT"] = (
    comparison_df["HUMAN_RECORD_TYPE"] == comparison_df["MODEL_RECORD_TYPE"]
) & comparison_df["HUMAN_RECORD_TYPE"].ne("")

comparison_df["CAUSE_EXACT"] = (
    comparison_df["HUMAN_PRIMARY_CAUSE"] == comparison_df["MODEL_PRIMARY_CAUSE"]
) & comparison_df["CAUSE_EVALUABLE"]
comparison_df["CAUSE_RELATED"] = comparison_df.apply(
    lambda row: categories_related(
        row["HUMAN_PRIMARY_CAUSE"],
        row["MODEL_PRIMARY_CAUSE"],
        CAUSE_RELATION_GROUPS,
    ) if row["CAUSE_EVALUABLE"] else False,
    axis=1,
)
comparison_df["CAUSE_RELATED_GROUP"] = comparison_df.apply(
    lambda row: relation_group_names(
        row["HUMAN_PRIMARY_CAUSE"],
        row["MODEL_PRIMARY_CAUSE"],
        CAUSE_RELATION_GROUPS,
    ),
    axis=1,
)

comparison_df["FACTOR_EXACT"] = (
    comparison_df["HUMAN_CONTRIBUTING_FACTOR"]
    == comparison_df["MODEL_REVISED_FACTOR"]
) & comparison_df["FACTOR_EVALUABLE"]
comparison_df["FACTOR_RELATED"] = comparison_df.apply(
    lambda row: categories_related(
        row["HUMAN_CONTRIBUTING_FACTOR"],
        row["MODEL_REVISED_FACTOR"],
        FACTOR_RELATION_GROUPS,
    ) if row["FACTOR_EVALUABLE"] else False,
    axis=1,
)
comparison_df["FACTOR_RELATED_GROUP"] = comparison_df.apply(
    lambda row: relation_group_names(
        row["HUMAN_CONTRIBUTING_FACTOR"],
        row["MODEL_REVISED_FACTOR"],
        FACTOR_RELATION_GROUPS,
    ),
    axis=1,
)

comparison_df["PREVENTION_EXACT"] = (
    comparison_df["HUMAN_PREVENTION_THEME"]
    == comparison_df["MODEL_REVISED_PREVENTION"]
) & comparison_df["PREVENTION_EVALUABLE"]
comparison_df["PREVENTION_RELATED"] = comparison_df.apply(
    lambda row: categories_related(
        row["HUMAN_PREVENTION_THEME"],
        row["MODEL_REVISED_PREVENTION"],
        PREVENTION_RELATION_GROUPS,
    ) if row["PREVENTION_EVALUABLE"] else False,
    axis=1,
)
comparison_df["PREVENTION_RELATED_GROUP"] = comparison_df.apply(
    lambda row: relation_group_names(
        row["HUMAN_PREVENTION_THEME"],
        row["MODEL_REVISED_PREVENTION"],
        PREVENTION_RELATION_GROUPS,
    ),
    axis=1,
)


def comparison_status(exact, related, evaluable, uncertain):
    if not evaluable:
        return "NOT_EVALUABLE"
    if exact:
        return "EXACT_MATCH_UNCERTAIN" if uncertain else "EXACT_MATCH"
    if related:
        return "RELATED_MATCH_UNCERTAIN" if uncertain else "RELATED_MATCH"
    return "DISAGREEMENT_UNCERTAIN" if uncertain else "MATERIAL_DISAGREEMENT"


comparison_df["CAUSE_COMPARISON_STATUS"] = comparison_df.apply(
    lambda row: comparison_status(
        row["CAUSE_EXACT"],
        row["CAUSE_RELATED"],
        row["CAUSE_EVALUABLE"],
        row["CAUSE_HUMAN_UNCERTAIN"],
    ),
    axis=1,
)
comparison_df["FACTOR_COMPARISON_STATUS"] = comparison_df.apply(
    lambda row: comparison_status(
        row["FACTOR_EXACT"],
        row["FACTOR_RELATED"],
        row["FACTOR_EVALUABLE"],
        row["FACTOR_HUMAN_UNCERTAIN"],
    ),
    axis=1,
)
comparison_df["PREVENTION_COMPARISON_STATUS"] = comparison_df.apply(
    lambda row: comparison_status(
        row["PREVENTION_EXACT"],
        row["PREVENTION_RELATED"],
        row["PREVENTION_EVALUABLE"],
        row["PREVENTION_HUMAN_UNCERTAIN"],
    ),
    axis=1,
)

comparison_df["ALL_THREE_EXACT"] = (
    comparison_df["CAUSE_EXACT"]
    & comparison_df["FACTOR_EXACT"]
    & comparison_df["PREVENTION_EXACT"]
)
comparison_df["ALL_THREE_EXACT_OR_RELATED"] = (
    comparison_df["CAUSE_RELATED"]
    & comparison_df["FACTOR_RELATED"]
    & comparison_df["PREVENTION_RELATED"]
)
comparison_df["CAUSE_AND_FACTOR_EXACT"] = (
    comparison_df["CAUSE_EXACT"] & comparison_df["FACTOR_EXACT"]
)


def confidence_band(value):
    if pd.isna(value):
        return "Missing"
    if value >= 0.80:
        return "High (>=0.80)"
    if value >= 0.65:
        return "Medium (0.65-0.79)"
    return "Low (<0.65)"


comparison_df["CAUSE_CONFIDENCE_BAND"] = comparison_df["MODEL_CAUSE_CONFIDENCE"].map(confidence_band)
comparison_df["FACTOR_CONFIDENCE_BAND"] = comparison_df["MODEL_FACTOR_CONFIDENCE"].map(confidence_band)
comparison_df["PREVENTION_CONFIDENCE_BAND"] = comparison_df["MODEL_PREVENTION_CONFIDENCE"].map(confidence_band)

comparison_df.to_csv(OUTPUT_COMPARISON_CSV, index=False, encoding="utf-8-sig")


# =============================================================================
# 6. METRIC FUNCTIONS
# =============================================================================

def safe_rate(numerator, denominator):
    return 100.0 * numerator / denominator if denominator else np.nan


def field_summary(
    dataframe,
    field_name,
    evaluable_column,
    exact_column,
    related_column,
    uncertain_column,
):
    evaluable = dataframe[dataframe[evaluable_column]].copy()
    certain = evaluable[~evaluable[uncertain_column]].copy()
    return {
        "Field": field_name,
        "Completed/evaluable": len(evaluable),
        "Human uncertain": int(evaluable[uncertain_column].sum()),
        "Exact matches": int(evaluable[exact_column].sum()),
        "Exact agreement %": safe_rate(evaluable[exact_column].sum(), len(evaluable)),
        "Exact or related matches": int(evaluable[related_column].sum()),
        "Exact or related %": safe_rate(evaluable[related_column].sum(), len(evaluable)),
        "Certain-only records": len(certain),
        "Certain-only exact %": safe_rate(certain[exact_column].sum(), len(certain)),
        "Certain-only exact or related %": safe_rate(certain[related_column].sum(), len(certain)),
    }


agreement_summary = pd.DataFrame([
    field_summary(
        comparison_df, "Primary cause", "CAUSE_EVALUABLE", "CAUSE_EXACT",
        "CAUSE_RELATED", "CAUSE_HUMAN_UNCERTAIN",
    ),
    field_summary(
        comparison_df, "Contributing factor", "FACTOR_EVALUABLE", "FACTOR_EXACT",
        "FACTOR_RELATED", "FACTOR_HUMAN_UNCERTAIN",
    ),
    field_summary(
        comparison_df, "Prevention theme", "PREVENTION_EVALUABLE", "PREVENTION_EXACT",
        "PREVENTION_RELATED", "PREVENTION_HUMAN_UNCERTAIN",
    ),
])

completion_summary = pd.DataFrame([
    {"Measure": "Workbook rows", "Value": len(comparison_df)},
    {"Measure": "Review completed = YES", "Value": int(comparison_df["REVIEW_COMPLETE_FLAG"].sum())},
    {"Measure": "Rows with any uncertainty", "Value": int(comparison_df["ANY_HUMAN_UNCERTAIN"].sum())},
    {"Measure": "Cause uncertainty", "Value": int(comparison_df["CAUSE_HUMAN_UNCERTAIN"].sum())},
    {"Measure": "Factor uncertainty", "Value": int(comparison_df["FACTOR_HUMAN_UNCERTAIN"].sum())},
    {"Measure": "Prevention uncertainty", "Value": int(comparison_df["PREVENTION_HUMAN_UNCERTAIN"].sum())},
    {"Measure": "Human overall acceptable = YES", "Value": int(comparison_df["HUMAN_OVERALL_ACCEPTABLE"].eq("YES").sum())},
    {"Measure": "Human overall acceptable = NO", "Value": int(comparison_df["HUMAN_OVERALL_ACCEPTABLE"].eq("NO").sum())},
    {"Measure": "Human overall acceptable = UNCERTAIN", "Value": int(comparison_df["HUMAN_OVERALL_ACCEPTABLE"].eq("UNCERTAIN").sum())},
])


def classification_metrics(dataframe, human_column, model_column, evaluable_column, label):
    data = dataframe[dataframe[evaluable_column]].copy()
    if data.empty:
        return pd.DataFrame(), pd.DataFrame(), np.nan

    labels = sorted(set(data[human_column]).union(set(data[model_column])))
    report = classification_report(
        data[human_column],
        data[model_column],
        labels=labels,
        output_dict=True,
        zero_division=0,
    )
    category_rows = []
    for category in labels:
        values = report.get(category, {})
        category_rows.append({
            "Category": category,
            "Precision": values.get("precision", 0.0),
            "Recall": values.get("recall", 0.0),
            "F1": values.get("f1-score", 0.0),
            "Human support": int(values.get("support", 0)),
            "Model count": int((data[model_column] == category).sum()),
        })

    aggregate_rows = []
    for average in ["macro avg", "weighted avg"]:
        values = report.get(average, {})
        aggregate_rows.append({
            "Metric": f"{label} {average}",
            "Precision": values.get("precision", np.nan),
            "Recall": values.get("recall", np.nan),
            "F1": values.get("f1-score", np.nan),
            "Support": int(values.get("support", 0)),
        })

    kappa = cohen_kappa_score(data[human_column], data[model_column])
    return pd.DataFrame(category_rows), pd.DataFrame(aggregate_rows), kappa


cause_category_metrics, cause_aggregate_metrics, cause_kappa = classification_metrics(
    comparison_df,
    "HUMAN_PRIMARY_CAUSE",
    "MODEL_PRIMARY_CAUSE",
    "CAUSE_EVALUABLE",
    "Cause",
)
factor_category_metrics, factor_aggregate_metrics, factor_kappa = classification_metrics(
    comparison_df,
    "HUMAN_CONTRIBUTING_FACTOR",
    "MODEL_REVISED_FACTOR",
    "FACTOR_EVALUABLE",
    "Factor",
)
prevention_category_metrics, prevention_aggregate_metrics, prevention_kappa = classification_metrics(
    comparison_df,
    "HUMAN_PREVENTION_THEME",
    "MODEL_REVISED_PREVENTION",
    "PREVENTION_EVALUABLE",
    "Prevention",
)

aggregate_metrics = pd.concat(
    [cause_aggregate_metrics, factor_aggregate_metrics, prevention_aggregate_metrics],
    ignore_index=True,
)
aggregate_metrics["Cohen's kappa"] = [
    cause_kappa, cause_kappa,
    factor_kappa, factor_kappa,
    prevention_kappa, prevention_kappa,
]


# =============================================================================
# 7. CONFUSIONS, SUBGROUPS, AND RELATED-CATEGORY RESULTS
# =============================================================================

def confusion_pairs(dataframe, human_column, model_column, evaluable_column, exact_column, name):
    data = dataframe[dataframe[evaluable_column] & ~dataframe[exact_column]].copy()
    if data.empty:
        return pd.DataFrame(columns=["Field", "Human category", "Model category", "Count"])
    output = (
        data.groupby([human_column, model_column])
        .size()
        .reset_index(name="Count")
        .sort_values("Count", ascending=False)
    )
    output.insert(0, "Field", name)
    return output.rename(columns={
        human_column: "Human category",
        model_column: "Model category",
    })


confusion_pairs_df = pd.concat([
    confusion_pairs(
        comparison_df, "HUMAN_PRIMARY_CAUSE", "MODEL_PRIMARY_CAUSE",
        "CAUSE_EVALUABLE", "CAUSE_EXACT", "Primary cause",
    ),
    confusion_pairs(
        comparison_df, "HUMAN_CONTRIBUTING_FACTOR", "MODEL_REVISED_FACTOR",
        "FACTOR_EVALUABLE", "FACTOR_EXACT", "Contributing factor",
    ),
    confusion_pairs(
        comparison_df, "HUMAN_PREVENTION_THEME", "MODEL_REVISED_PREVENTION",
        "PREVENTION_EVALUABLE", "PREVENTION_EXACT", "Prevention theme",
    ),
], ignore_index=True)


def subgroup_agreement(dataframe, group_column, evaluable_column, exact_column, related_column, label):
    records = []
    for group_value, group in dataframe.groupby(group_column, dropna=False):
        evaluable = group[group[evaluable_column]]
        records.append({
            "Field": label,
            "Group": group_value,
            "Evaluable": len(evaluable),
            "Exact %": safe_rate(evaluable[exact_column].sum(), len(evaluable)),
            "Exact or related %": safe_rate(evaluable[related_column].sum(), len(evaluable)),
        })
    return pd.DataFrame(records)


severity_agreement = pd.concat([
    subgroup_agreement(comparison_df, "SEVERITY_GROUP", "CAUSE_EVALUABLE", "CAUSE_EXACT", "CAUSE_RELATED", "Cause"),
    subgroup_agreement(comparison_df, "SEVERITY_GROUP", "FACTOR_EVALUABLE", "FACTOR_EXACT", "FACTOR_RELATED", "Factor"),
    subgroup_agreement(comparison_df, "SEVERITY_GROUP", "PREVENTION_EVALUABLE", "PREVENTION_EXACT", "PREVENTION_RELATED", "Prevention"),
], ignore_index=True)

period_agreement = pd.concat([
    subgroup_agreement(comparison_df, "TIME_PERIOD", "CAUSE_EVALUABLE", "CAUSE_EXACT", "CAUSE_RELATED", "Cause"),
    subgroup_agreement(comparison_df, "TIME_PERIOD", "FACTOR_EVALUABLE", "FACTOR_EXACT", "FACTOR_RELATED", "Factor"),
    subgroup_agreement(comparison_df, "TIME_PERIOD", "PREVENTION_EVALUABLE", "PREVENTION_EXACT", "PREVENTION_RELATED", "Prevention"),
], ignore_index=True)

factor_tier_agreement = pd.concat([
    subgroup_agreement(
        comparison_df, "MODEL_FACTOR_SUPPORT_TIER", "FACTOR_EVALUABLE",
        "FACTOR_EXACT", "FACTOR_RELATED", "Factor",
    ),
    subgroup_agreement(
        comparison_df, "MODEL_FACTOR_SUPPORT_TIER", "PREVENTION_EVALUABLE",
        "PREVENTION_EXACT", "PREVENTION_RELATED", "Prevention",
    ),
], ignore_index=True)

confidence_agreement = pd.concat([
    subgroup_agreement(
        comparison_df, "CAUSE_CONFIDENCE_BAND", "CAUSE_EVALUABLE",
        "CAUSE_EXACT", "CAUSE_RELATED", "Cause",
    ),
    subgroup_agreement(
        comparison_df, "FACTOR_CONFIDENCE_BAND", "FACTOR_EVALUABLE",
        "FACTOR_EXACT", "FACTOR_RELATED", "Factor",
    ),
    subgroup_agreement(
        comparison_df, "PREVENTION_CONFIDENCE_BAND", "PREVENTION_EVALUABLE",
        "PREVENTION_EXACT", "PREVENTION_RELATED", "Prevention",
    ),
], ignore_index=True)

related_match_summary = pd.DataFrame([
    {
        "Field": "Primary cause",
        "Exact": int(comparison_df["CAUSE_EXACT"].sum()),
        "Related but not exact": int((comparison_df["CAUSE_RELATED"] & ~comparison_df["CAUSE_EXACT"]).sum()),
        "Material disagreement": int((comparison_df["CAUSE_EVALUABLE"] & ~comparison_df["CAUSE_RELATED"]).sum()),
    },
    {
        "Field": "Contributing factor",
        "Exact": int(comparison_df["FACTOR_EXACT"].sum()),
        "Related but not exact": int((comparison_df["FACTOR_RELATED"] & ~comparison_df["FACTOR_EXACT"]).sum()),
        "Material disagreement": int((comparison_df["FACTOR_EVALUABLE"] & ~comparison_df["FACTOR_RELATED"]).sum()),
    },
    {
        "Field": "Prevention theme",
        "Exact": int(comparison_df["PREVENTION_EXACT"].sum()),
        "Related but not exact": int((comparison_df["PREVENTION_RELATED"] & ~comparison_df["PREVENTION_EXACT"]).sum()),
        "Material disagreement": int((comparison_df["PREVENTION_EVALUABLE"] & ~comparison_df["PREVENTION_RELATED"]).sum()),
    },
])

combined_summary = pd.DataFrame([
    {
        "Measure": "Cause and factor both exact",
        "Count": int(comparison_df["CAUSE_AND_FACTOR_EXACT"].sum()),
        "Percent of reviewed records": safe_rate(comparison_df["CAUSE_AND_FACTOR_EXACT"].sum(), len(comparison_df)),
    },
    {
        "Measure": "Cause, factor, and prevention all exact",
        "Count": int(comparison_df["ALL_THREE_EXACT"].sum()),
        "Percent of reviewed records": safe_rate(comparison_df["ALL_THREE_EXACT"].sum(), len(comparison_df)),
    },
    {
        "Measure": "Cause, factor, and prevention exact or related",
        "Count": int(comparison_df["ALL_THREE_EXACT_OR_RELATED"].sum()),
        "Percent of reviewed records": safe_rate(comparison_df["ALL_THREE_EXACT_OR_RELATED"].sum(), len(comparison_df)),
    },
    {
        "Measure": "Human overall acceptable = YES",
        "Count": int(comparison_df["HUMAN_OVERALL_ACCEPTABLE"].eq("YES").sum()),
        "Percent of reviewed records": safe_rate(comparison_df["HUMAN_OVERALL_ACCEPTABLE"].eq("YES").sum(), len(comparison_df)),
    },
])


# =============================================================================
# 8. DISAGREEMENT WORKBOOK FOR OPTIONAL HUMAN UPDATES
# =============================================================================

review_priority = comparison_df[
    comparison_df["CAUSE_COMPARISON_STATUS"].str.contains("DISAGREEMENT")
    | comparison_df["FACTOR_COMPARISON_STATUS"].str.contains("DISAGREEMENT")
    | comparison_df["PREVENTION_COMPARISON_STATUS"].str.contains("DISAGREEMENT")
    | comparison_df["ANY_HUMAN_UNCERTAIN"]
].copy()

review_priority["REVIEW_PRIORITY"] = np.select(
    [
        review_priority["FACTOR_COMPARISON_STATUS"].eq("MATERIAL_DISAGREEMENT")
        | review_priority["PREVENTION_COMPARISON_STATUS"].eq("MATERIAL_DISAGREEMENT"),
        review_priority["CAUSE_COMPARISON_STATUS"].eq("MATERIAL_DISAGREEMENT"),
        review_priority["ANY_HUMAN_UNCERTAIN"],
    ],
    ["HIGH", "MEDIUM", "UNCERTAIN"],
    default="LOW",
)

priority_columns = [
    "REVIEW_PRIORITY",
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "SEVERITY_GROUP",
    "TIME_PERIOD",
    "HUMAN_PRIMARY_CAUSE",
    "MODEL_PRIMARY_CAUSE",
    "CAUSE_COMPARISON_STATUS",
    "CAUSE_RELATED_GROUP",
    "HUMAN_CONTRIBUTING_FACTOR",
    "MODEL_REVISED_FACTOR",
    "FACTOR_COMPARISON_STATUS",
    "FACTOR_RELATED_GROUP",
    "MODEL_FACTOR_SUPPORT_TIER",
    "MODEL_FACTOR_SUPPORT_METHOD",
    "HUMAN_PREVENTION_THEME",
    "MODEL_REVISED_PREVENTION",
    "PREVENTION_COMPARISON_STATUS",
    "PREVENTION_RELATED_GROUP",
    "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    "HUMAN_FACTOR_EVIDENCE_SUPPORTED",
    "HUMAN_PREVENTION_APPROPRIATE",
    "HUMAN_OVERALL_ACCEPTABLE",
    "HUMAN_REVIEW_COMMENTS",
]

priority_export = review_priority[priority_columns].copy()

wb = Workbook()
ws = wb.active
ws.title = "Differences to Review"
ws.append(priority_columns)
for row in priority_export.itertuples(index=False, name=None):
    ws.append(list(row))

header_fill = PatternFill("solid", fgColor="1F4E78")
header_font = Font(color="FFFFFF", bold=True)
for cell in ws[1]:
    cell.fill = header_fill
    cell.font = header_font
    cell.alignment = Alignment(wrap_text=True, horizontal="center")

ws.freeze_panes = "A2"
ws.auto_filter.ref = f"A1:X{ws.max_row}"
ws.sheet_view.showGridLines = False

for column_index in range(1, ws.max_column + 1):
    letter = get_column_letter(column_index)
    width = 22
    if ws.cell(1, column_index).value in {"NARRATIVE", "HUMAN_REVIEW_COMMENTS"}:
        width = 70
    ws.column_dimensions[letter].width = width

for row_index in range(2, ws.max_row + 1):
    ws.row_dimensions[row_index].height = 80
    for column_index in range(1, ws.max_column + 1):
        ws.cell(row_index, column_index).alignment = Alignment(wrap_text=True, vertical="top")

review_table = Table(displayName="DifferencesReviewTable", ref=f"A1:X{ws.max_row}")
review_table.tableStyleInfo = TableStyleInfo(
    name="TableStyleMedium2",
    showFirstColumn=False,
    showLastColumn=False,
    showRowStripes=True,
    showColumnStripes=False,
)
ws.add_table(review_table)
wb.save(OUTPUT_DISAGREEMENTS_XLSX)


# =============================================================================
# 9. PDF HELPERS
# =============================================================================

pdf = PdfPages(OUTPUT_PDF)
plt.rcParams.update({"figure.dpi": 120, "font.size": 9, "axes.grid": False})
page_counter = {"value": 0}


def readable(value):
    return str(value).replace("_", " ").title()


def create_page(title=None, landscape=False):
    figure = plt.figure(figsize=(11, 8.5) if landscape else (8.5, 11))
    axis = figure.add_axes([0, 0, 1, 1])
    axis.axis("off")
    if title:
        figure.text(0.5, 0.96, title, ha="center", va="top", fontsize=14, fontweight="bold")
    return figure


def save_page(figure):
    page_counter["value"] += 1
    figure.text(
        0.5, 0.018, f"Page {page_counter['value']}",
        ha="center", va="bottom", fontsize=7.5, color="#666666",
    )
    pdf.savefig(figure, bbox_inches="tight")
    plt.close(figure)


def add_text_pages(title, lines, font_size=8.4, lines_per_page=52, wrap_width=104):
    wrapped = []
    for line in lines:
        wrapped.extend(
            textwrap.wrap(str(line), width=wrap_width, break_long_words=False, break_on_hyphens=False)
            if line else [""]
        )
    chunks = [wrapped[i:i + lines_per_page] for i in range(0, len(wrapped), lines_per_page)] or [[""]]
    for number, chunk in enumerate(chunks, start=1):
        page_title = title if len(chunks) == 1 else f"{title} - {number} of {len(chunks)}"
        figure = create_page(page_title)
        figure.text(
            0.055, 0.915, "\n".join(chunk),
            va="top", ha="left", fontsize=font_size,
            family="monospace", linespacing=1.2,
        )
        save_page(figure)


def add_table_pages(title, dataframe, rows_per_page=20, font_size=6.8, landscape=True, wrap_width=25):
    if dataframe is None or dataframe.empty:
        add_text_pages(title, ["No results were available."])
        return
    printable = dataframe.copy()
    for column in printable.columns:
        if pd.api.types.is_float_dtype(printable[column]):
            printable[column] = printable[column].map(
                lambda value: "" if pd.isna(value) else f"{value:.3f}"
            )
        printable[column] = printable[column].map(
            lambda value: "" if pd.isna(value) else "\n".join(
                textwrap.wrap(str(value), width=wrap_width, break_long_words=False)
            )
        )
    for start in range(0, len(printable), rows_per_page):
        chunk = printable.iloc[start:start + rows_per_page]
        figure = create_page(title if start == 0 else title + " - Continued", landscape=landscape)
        axis = figure.add_axes([0.025, 0.06, 0.95, 0.84])
        axis.axis("off")
        table = axis.table(
            cellText=chunk.values,
            colLabels=[readable(column) for column in chunk.columns],
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )
        table.auto_set_font_size(False)
        table.set_fontsize(font_size)
        table.scale(1, 1.36)
        for (row_number, column_number), cell in table.get_celld().items():
            cell.set_edgecolor("#AAB2BD")
            if row_number == 0:
                cell.set_facecolor("#D9EAF7")
                cell.set_text_props(weight="bold")
        save_page(figure)


def add_agreement_bar_chart():
    figure, axis = plt.subplots(figsize=(10, 6.5))
    positions = np.arange(len(agreement_summary))
    width = 0.34
    axis.bar(
        positions - width / 2,
        agreement_summary["Exact agreement %"],
        width,
        label="Exact",
        color="#2E6F9E",
    )
    axis.bar(
        positions + width / 2,
        agreement_summary["Exact or related %"],
        width,
        label="Exact or related",
        color="#76A85A",
    )
    axis.set_xticks(positions)
    axis.set_xticklabels(agreement_summary["Field"])
    axis.set_ylim(0, 100)
    axis.set_ylabel("Agreement percent")
    axis.set_title("Human-Model Agreement", fontweight="bold")
    axis.legend()
    axis.grid(axis="y", alpha=0.25)
    for position, value in zip(positions - width / 2, agreement_summary["Exact agreement %"]):
        if not pd.isna(value):
            axis.text(position, value + 1.5, f"{value:.1f}%", ha="center", fontsize=8)
    for position, value in zip(positions + width / 2, agreement_summary["Exact or related %"]):
        if not pd.isna(value):
            axis.text(position, value + 1.5, f"{value:.1f}%", ha="center", fontsize=8)
    figure.tight_layout()
    save_page(figure)


def add_confusion_heatmap(title, dataframe, human_column, model_column, evaluable_column, top_n=12):
    data = dataframe[dataframe[evaluable_column]].copy()
    if data.empty:
        return
    top_labels = list(
        pd.concat([data[human_column], data[model_column]])
        .value_counts()
        .head(top_n)
        .index
    )
    matrix = pd.crosstab(data[human_column], data[model_column]).reindex(
        index=top_labels,
        columns=top_labels,
        fill_value=0,
    )
    figure, axis = plt.subplots(figsize=(10.5, 8))
    image = axis.imshow(matrix.values, cmap="Blues", aspect="auto")
    axis.set_xticks(range(len(matrix.columns)))
    axis.set_xticklabels([readable(value) for value in matrix.columns], rotation=40, ha="right", fontsize=7)
    axis.set_yticks(range(len(matrix.index)))
    axis.set_yticklabels([readable(value) for value in matrix.index], fontsize=7)
    axis.set_xlabel("Model category")
    axis.set_ylabel("Human category")
    axis.set_title(title, fontweight="bold")
    for row_number in range(matrix.shape[0]):
        for column_number in range(matrix.shape[1]):
            value = int(matrix.iloc[row_number, column_number])
            if value > 0:
                axis.text(column_number, row_number, str(value), ha="center", va="center", fontsize=7)
    figure.colorbar(image, ax=axis, label="Record count")
    figure.tight_layout()
    save_page(figure)


def add_difference_examples(title, dataframe, field, maximum_records=MAX_EXAMPLES_PER_SECTION):
    if dataframe.empty:
        add_text_pages(title, ["No records were available."])
        return
    lines = []
    configurations = {
        "cause": (
            "HUMAN_PRIMARY_CAUSE", "MODEL_PRIMARY_CAUSE",
            "CAUSE_COMPARISON_STATUS", "CAUSE_RELATED_GROUP",
            "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
        ),
        "factor": (
            "HUMAN_CONTRIBUTING_FACTOR", "MODEL_REVISED_FACTOR",
            "FACTOR_COMPARISON_STATUS", "FACTOR_RELATED_GROUP",
            "HUMAN_FACTOR_EVIDENCE_SUPPORTED",
        ),
        "prevention": (
            "HUMAN_PREVENTION_THEME", "MODEL_REVISED_PREVENTION",
            "PREVENTION_COMPARISON_STATUS", "PREVENTION_RELATED_GROUP",
            "HUMAN_PREVENTION_APPROPRIATE",
        ),
    }
    human_column, model_column, status_column, related_column, uncertainty_column = configurations[field]
    for example_number, (_, row) in enumerate(dataframe.head(maximum_records).iterrows(), start=1):
        lines.extend([
            "=" * 86,
            f"Example {example_number} | Record {row['SOURCE_RECORD_ID']} | {row.get('SEVERITY_GROUP', '')} | {row.get('TIME_PERIOD', '')}",
            f"Status: {row.get(status_column, '')}",
            f"Human category: {row.get(human_column, '')}",
            f"Model category: {row.get(model_column, '')}",
            f"Related-category group: {row.get(related_column, '')}",
            f"Human support assessment: {row.get(uncertainty_column, '')}",
            f"Model confidence: {row.get('MODEL_' + ('CAUSE' if field == 'cause' else 'FACTOR' if field == 'factor' else 'PREVENTION') + '_CONFIDENCE', '')}",
            f"Model factor support tier: {row.get('MODEL_FACTOR_SUPPORT_TIER', '')}",
            "Narrative:",
            str(row.get("NARRATIVE", "")),
            "Human comments:",
            str(row.get("HUMAN_REVIEW_COMMENTS", "")),
            "",
        ])
    add_text_pages(title, lines, font_size=7.5, lines_per_page=48)


# =============================================================================
# 10. GENERATE PDF REPORT
# =============================================================================

figure = create_page()
figure.text(0.5, 0.79, "MSHA Human-Model Validation", ha="center", fontsize=21, fontweight="bold")
figure.text(
    0.5, 0.735,
    "Block 4 - Independent Review of 100 Unseen Accident Narratives",
    ha="center", fontsize=12.5, color="#444444",
)
figure.text(
    0.5, 0.69,
    datetime.now().strftime("Generated %B %d, %Y at %H:%M"),
    ha="center", fontsize=9, color="#666666",
)
cover_lines = [
    f"Workbook records: {len(comparison_df):,}",
    f"Completed reviews: {int(comparison_df['REVIEW_COMPLETE_FLAG'].sum()):,}",
    f"Rows with at least one uncertain human response: {int(comparison_df['ANY_HUMAN_UNCERTAIN'].sum()):,}",
    "",
    "Exact agreement is the primary validation measure.",
    "Exact-or-related agreement is a secondary sensitivity measure recognizing",
    "that short narratives may reasonably fit nearby ontology categories.",
    "",
    "Human uncertainty is reported separately and is not silently treated as",
    "model error or model correctness.",
]
figure.text(
    0.10, 0.59, "\n".join(cover_lines),
    va="top", ha="left", fontsize=9.2, family="monospace", linespacing=1.3,
)
save_page(figure)

add_text_pages(
    "Executive Summary and Interpretation Rules",
    [
        "This report compares the independent human labels in the completed Block 3 workbook with the frozen model output for the same 100 narratives.",
        "",
        "The human review is treated as the reference comparison, not as an infallible gold standard. The reviewer explicitly noted that contributing factors and corrective actions were often difficult to determine from the short narratives and that some causes could reasonably fit multiple categories.",
        "",
        "The report therefore provides two agreement measures. Exact agreement requires identical ontology labels. Exact-or-related agreement also recognizes predefined neighboring categories that describe similar mechanisms, conditions, or prevention approaches.",
        "",
        "UNCERTAIN responses remain visible throughout the report. Certain-only agreement rates exclude records marked uncertain for that field, while all-evaluable rates retain them.",
        "",
        "The disagreement examples are intended to support optional reviewer reconsideration. They do not automatically mean that the human or model label is wrong.",
    ],
)

add_table_pages("Review Completion and Uncertainty", completion_summary, rows_per_page=15, landscape=False)
add_table_pages("Agreement Summary", agreement_summary, rows_per_page=10, landscape=True, wrap_width=28)
add_agreement_bar_chart()
add_table_pages("Combined Agreement Results", combined_summary, rows_per_page=10, landscape=False)
add_table_pages("Aggregate Precision, Recall, F1, and Kappa", aggregate_metrics, rows_per_page=12, landscape=True)
add_table_pages("Exact, Related, and Material Disagreement", related_match_summary, rows_per_page=10, landscape=False)

add_text_pages(
    "Related-Category Sensitivity Method",
    [
        "Related matching is a secondary analysis and does not replace exact agreement.",
        "",
        "Cause examples include caught-in/between versus struck-by or material release; loss of control versus unexpected equipment movement; and awkward motion versus manual handling or repetitive exposure.",
        "",
        "Factor examples include unstable material versus inadequate securement; line-of-fire versus work method or hazard recognition; and equipment condition versus energy isolation or guarding.",
        "",
        "Prevention examples include secure/reposition material versus line-of-fire exclusion; equipment repair versus energy isolation or guarding; and procedure revision versus hazard review, training, or communication.",
        "",
        "A related match means the two answers are conceptually adjacent within the frozen ontology. It does not establish that both labels are equally correct for the specific narrative.",
    ],
)

add_confusion_heatmap(
    "Primary-Cause Confusion Matrix",
    comparison_df,
    "HUMAN_PRIMARY_CAUSE",
    "MODEL_PRIMARY_CAUSE",
    "CAUSE_EVALUABLE",
)
add_confusion_heatmap(
    "Contributing-Factor Confusion Matrix",
    comparison_df,
    "HUMAN_CONTRIBUTING_FACTOR",
    "MODEL_REVISED_FACTOR",
    "FACTOR_EVALUABLE",
)
add_confusion_heatmap(
    "Prevention-Theme Confusion Matrix",
    comparison_df,
    "HUMAN_PREVENTION_THEME",
    "MODEL_REVISED_PREVENTION",
    "PREVENTION_EVALUABLE",
)

add_table_pages("Most Common Human-Model Differences", confusion_pairs_df.head(TOP_CONFUSIONS), rows_per_page=20, landscape=True, wrap_width=32)
add_table_pages("Cause Category Metrics", cause_category_metrics, rows_per_page=22, landscape=True)
add_table_pages("Factor Category Metrics", factor_category_metrics, rows_per_page=22, landscape=True)
add_table_pages("Prevention Category Metrics", prevention_category_metrics, rows_per_page=22, landscape=True)
add_table_pages("Agreement by Severity", severity_agreement, rows_per_page=20, landscape=True)
add_table_pages("Agreement by Time Period", period_agreement, rows_per_page=20, landscape=True)
add_table_pages("Agreement by Factor-Support Tier", factor_tier_agreement, rows_per_page=15, landscape=True)
add_table_pages("Agreement by Model Confidence Band", confidence_agreement, rows_per_page=20, landscape=True)

cause_differences = comparison_df[
    comparison_df["CAUSE_EVALUABLE"] & ~comparison_df["CAUSE_EXACT"]
].copy()
factor_differences = comparison_df[
    comparison_df["FACTOR_EVALUABLE"] & ~comparison_df["FACTOR_EXACT"]
].copy()
prevention_differences = comparison_df[
    comparison_df["PREVENTION_EVALUABLE"] & ~comparison_df["PREVENTION_EXACT"]
].copy()

# Put material disagreements first, then uncertainty, then related matches.
status_rank = {
    "MATERIAL_DISAGREEMENT": 0,
    "DISAGREEMENT_UNCERTAIN": 1,
    "RELATED_MATCH": 2,
    "RELATED_MATCH_UNCERTAIN": 3,
}
for frame, column in [
    (cause_differences, "CAUSE_COMPARISON_STATUS"),
    (factor_differences, "FACTOR_COMPARISON_STATUS"),
    (prevention_differences, "PREVENTION_COMPARISON_STATUS"),
]:
    frame["_STATUS_RANK"] = frame[column].map(status_rank).fillna(9)
    frame.sort_values(["_STATUS_RANK", "SOURCE_RECORD_ID"], inplace=True)

add_difference_examples("Cause Differences for Review", cause_differences, "cause")
add_difference_examples("Contributing-Factor Differences for Review", factor_differences, "factor")
add_difference_examples("Prevention-Theme Differences for Review", prevention_differences, "prevention")

add_text_pages(
    "Conclusions and Appropriate Use",
    [
        "Primary-cause performance should be interpreted from exact agreement, exact-or-related agreement, category-level F1, and the cause confusion matrix together.",
        "",
        "Contributing-factor and prevention performance should be interpreted more cautiously because many narratives describe what happened without providing enough detail to establish why it happened or which corrective action is best.",
        "",
        "The factor-support-tier analysis tests whether directly supported model factors perform better than contextually supported factors. Confidence-band results test whether the model's numerical confidence corresponds to higher human agreement.",
        "",
        "Records marked uncertain should not be used as definitive evidence for or against a category. They are retained as a transparent part of the validation set and may be updated after reviewing the disagreement examples.",
        "",
        "If the human workbook is revised, rerun this Block 4 script. The report and comparison outputs will be regenerated from the updated labels without rerunning the language model.",
    ],
)

add_text_pages(
    "Technical Appendix",
    [
        f"Input workbook: {os.path.abspath(INPUT_XLSX)}",
        f"Records compared: {len(comparison_df):,}",
        f"Expected records: {EXPECTED_RECORDS}",
        f"Completed reviews: {int(comparison_df['REVIEW_COMPLETE_FLAG'].sum()):,}",
        f"Rows with uncertainty: {int(comparison_df['ANY_HUMAN_UNCERTAIN'].sum()):,}",
        f"Comparison CSV: {os.path.abspath(OUTPUT_COMPARISON_CSV)}",
        f"Differences workbook: {os.path.abspath(OUTPUT_DISAGREEMENTS_XLSX)}",
        f"PDF report: {os.path.abspath(OUTPUT_PDF)}",
        "",
        "Exact agreement is primary. Related agreement depends on the explicit related-category groups defined at the beginning of this script.",
    ],
)

metadata = pdf.infodict()
metadata["Title"] = "MSHA Block 4 Human-Model Validation Report"
metadata["Author"] = "Python"
metadata["Subject"] = "Human review comparison for 100 unseen MSHA accident narratives"
metadata["Keywords"] = "MSHA, human review, validation, NLP, ontology, causes, contributing factors, prevention"
metadata["CreationDate"] = datetime.now()
pdf.close()


# =============================================================================
# 11. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("BLOCK 4 HUMAN VALIDATION COMPLETE")
print("=" * 95)
print("Records compared:", len(comparison_df))
print("Completed reviews:", int(comparison_df["REVIEW_COMPLETE_FLAG"].sum()))
print("Rows with uncertainty:", int(comparison_df["ANY_HUMAN_UNCERTAIN"].sum()))
print("\nAgreement summary:")
print(agreement_summary.to_string(index=False))
print("\nPDF report:", os.path.abspath(OUTPUT_PDF))
print("Comparison CSV:", os.path.abspath(OUTPUT_COMPARISON_CSV))
print("Differences workbook:", os.path.abspath(OUTPUT_DISAGREEMENTS_XLSX))
print("=" * 95)
````

````python
# =============================================================================
# STANDALONE HYBRID RAG PRIMARY-CAUSE CLASSIFICATION EXPERIMENT
#
# PURPOSE
# -------
# Test whether retrieval of similar, labeled MSHA accident narratives improves
# primary-cause classification compared with:
#
#   1. Narrative-only Qwen classification.
#   2. TF-IDF multiclass logistic regression, when its saved output is present.
#   3. Independent human primary-cause classifications.
#
# RAG REFERENCE LIBRARY
# ---------------------
# The retrieval library combines:
#
#   A. Human-reviewed Block 3 narratives, used through leave-one-out retrieval.
#      The target record is always excluded from its own retrieved examples.
#
#   B. High-confidence, cause-valid Code Block 2 narratives.
#
# RETRIEVAL METHOD
# ----------------
# Hybrid retrieval combines:
#
#   1. Semantic similarity from Sentence-Transformers embeddings.
#   2. TF-IDF cosine similarity.
#
# GENERATOR
# ---------
# Qwen/Qwen2.5-7B-Instruct, run locally with four-bit quantization.
#
# IMPORTANT CONTROL
# -----------------
# Retrieved examples are references for interpreting terminology and ontology
# boundaries. The model is instructed not to transfer facts from a retrieved
# narrative to the target narrative.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. Completed human review workbook:
#
#       MSHA_Block3_Human_Review_Workbook- Completed.xlsx
#
# 3. One Code Block 2 structured result file:
#
#       MSHA_Block2_Expanded_Results.csv
#       MSHA_Block2_Revised_Results.csv
#       MSHA_Block2_Unseen_Results.csv
#
# OPTIONAL INPUT
# --------------
# Earlier logistic-regression benchmark output:
#
#       MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv
#
# OUTPUTS
# -------
# 1. MSHA_RAG_Primary_Cause_Comparison_Report.pdf
# 2. MSHA_RAG_Primary_Cause_Results.csv
# 3. MSHA_RAG_Primary_Cause_Checkpoint.jsonl
# 4. MSHA_RAG_Retrieval_Details.csv
# 5. MSHA_RAG_Primary_Cause_Summary.json
#
# INSTALLATION NOTE
# -----------------
# The environment must contain:
#
#   sentence-transformers
#   transformers
#   accelerate
#   bitsandbytes
#   scikit-learn
#   openpyxl
#
# The Sentence-Transformers embedding model and Qwen model must be cached
# locally or downloadable from Hugging Face when first used.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import math
import time
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import transformers

from pydantic import BaseModel, Field, model_validator

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from sentence_transformers import SentenceTransformer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_recall_fscore_support,
    f1_score,
    cohen_kappa_score,
    confusion_matrix,
)

from matplotlib.backends.backend_pdf import PdfPages


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

EMBEDDING_MODEL_ID = "sentence-transformers/all-MiniLM-L6-v2"

PROMPT_VERSION = "MSHA_RAG_PRIMARY_CAUSE_V1.0"

HUMAN_REVIEW_WORKBOOK = (
    "MSHA_Block3_Human_Review_Workbook- Completed.xlsx"
)

BLOCK2_RESULT_CANDIDATES = [
    "MSHA_Block2_Expanded_Results.csv",
    "MSHA_Block2_Revised_Results.csv",
    "MSHA_Block2_Unseen_Results.csv",
]

LOGISTIC_RESULT_CANDIDATES = [
    "MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv",
]

OUTPUT_PDF = (
    "MSHA_RAG_Primary_Cause_Comparison_Report.pdf"
)

OUTPUT_RESULTS_CSV = (
    "MSHA_RAG_Primary_Cause_Results.csv"
)

OUTPUT_RETRIEVAL_CSV = (
    "MSHA_RAG_Retrieval_Details.csv"
)

CHECKPOINT_JSONL = (
    "MSHA_RAG_Primary_Cause_Checkpoint.jsonl"
)

OUTPUT_SUMMARY_JSON = (
    "MSHA_RAG_Primary_Cause_Summary.json"
)


RANDOM_STATE = 42

REPROCESS_COMPLETED = False

MAX_RETRIES = 4

MAX_NEW_TOKENS = 350

TOP_K_RETRIEVAL = 5

MAX_HUMAN_EXAMPLES_PER_QUERY = 2

MAX_EXAMPLES_PER_CAUSE_PER_QUERY = 2


SEMANTIC_WEIGHT = 0.70

LEXICAL_WEIGHT = 0.30


MIN_RETRIEVAL_SCORE = 0.15

MIN_BLOCK2_CAUSE_CONFIDENCE = 0.80

ALLOWED_BLOCK2_SUPPORT = {
    "HIGH",
    "MEDIUM",
}


TFIDF_MAX_FEATURES = 30000

TFIDF_NGRAM_RANGE = (
    1,
    2,
)

TFIDF_MIN_DF = 1

TFIDF_MAX_DF = 0.98


TOP_DISAGREEMENT_EXAMPLES = 15

TOP_RETRIEVAL_EXAMPLES = 15


# =============================================================================
# 3. FROZEN PRIMARY-CAUSE ONTOLOGY
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


CATEGORY_ALIASES = {
    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "LOSS_OF_CONTROL":
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",

    "UNEXPECTED_MACHINE_MOVEMENT":
        "UNEXPECTED_EQUIPMENT_MOVEMENT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "MANUAL_HANDLING":
        "MANUAL_HANDLING_OR_OVEREXERTION",

    "GROUND_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "ROOF_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "MOBILE_EQUIPMENT_EVENT":
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",

    "STORED_ENERGY_RELEASE":
        "PRESSURE_OR_STORED_ENERGY_RELEASE",

    "FIRE_OR_EXPLOSION":
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",
}


# =============================================================================
# 4. RELATED-CAUSE GROUPS
#
# Exact agreement remains the primary metric.
# Exact-or-related agreement is a secondary sensitivity measure.
# =============================================================================

RELATED_CAUSE_GROUPS = [
    {
        "CAUGHT_IN_OR_BETWEEN",
        "STRUCK_BY_MOVING_OBJECT",
        "MATERIAL_RELEASE_OR_SHIFT",
    },

    {
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    },

    {
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
        "MANUAL_HANDLING_OR_OVEREXERTION",
    },

    {
        "GROUND_OR_ROOF_FAILURE",
        "MATERIAL_RELEASE_OR_SHIFT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "PRESSURE_OR_STORED_ENERGY_RELEASE",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
    },

    {
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    },

    {
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "SLIP_TRIP_OR_FALL",
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    },
]


# =============================================================================
# 5. RAG RESPONSE SCHEMA
# =============================================================================

class RAGCauseExtraction(BaseModel):

    primary_cause: str

    cause_evidence_text: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    retrieval_used: bool = True

    retrieval_explanation: str = ""

    alternative_cause: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_output(self):

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                f"Invalid primary cause: {self.primary_cause}"
            )

        if (
            self.alternative_cause
            and self.alternative_cause
            not in CAUSE_CATEGORIES
        ):

            raise ValueError(
                f"Invalid alternative cause: "
                f"{self.alternative_cause}"
            )

        if self.cause_support_level not in {
            "HIGH",
            "MEDIUM",
            "LOW",
        }:

            raise ValueError(
                "cause_support_level must be HIGH, MEDIUM, or LOW."
            )

        return self


# =============================================================================
# 6. INPUT VALIDATION
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(
    accidents,
    pd.DataFrame,
):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


if not Path(
    HUMAN_REVIEW_WORKBOOK
).exists():

    raise FileNotFoundError(
        f"Required completed workbook not found: "
        f"{HUMAN_REVIEW_WORKBOOK}"
    )


block2_results_file = next(
    (
        path
        for path in BLOCK2_RESULT_CANDIDATES
        if Path(
            path
        ).exists()
    ),
    None,
)


if block2_results_file is None:

    raise FileNotFoundError(
        "No Code Block 2 results file was found. "
        "Expected one of: "
        + ", ".join(
            BLOCK2_RESULT_CANDIDATES
        )
    )


logistic_results_file = next(
    (
        path
        for path in LOGISTIC_RESULT_CANDIDATES
        if Path(
            path
        ).exists()
    ),
    None,
)


print("=" * 95)
print("HYBRID RAG PRIMARY-CAUSE CLASSIFICATION")
print("=" * 95)

print(
    "Code Block 2 results:",
    block2_results_file,
)

print(
    "Human-review workbook:",
    HUMAN_REVIEW_WORKBOOK,
)

print(
    "Logistic-regression comparison:",
    logistic_results_file
    if logistic_results_file
    else "Not found; logistic comparison will be omitted.",
)


# =============================================================================
# 7. GENERAL UTILITIES
# =============================================================================

def normalize_column_name(
    value,
):

    return re.sub(
        r"[^A-Z0-9]+",
        "_",
        str(
            value
        )
        .strip()
        .upper(),
    ).strip(
        "_"
    )


def normalize_category(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def readable_label(
    value,
):

    return (
        str(
            value
        )
        .replace(
            "_",
            " ",
        )
        .title()
    )


def safe_float(
    value,
    default=np.nan,
):

    try:

        output = float(
            value
        )

        if np.isnan(
            output
        ):

            return default

        return output

    except Exception:

        return default


def first_existing_column(
    dataframe,
    candidates,
    required=False,
    description="column",
):

    lookup = {
        normalize_column_name(
            column
        ):
            column
        for column in dataframe.columns
    }

    for candidate in candidates:

        normalized_candidate = normalize_column_name(
            candidate
        )

        if normalized_candidate in lookup:

            return lookup[
                normalized_candidate
            ]

    if required:

        raise KeyError(
            f"Required {description} was not found. "
            f"Checked {candidates}. "
            f"Available columns: {list(dataframe.columns)}"
        )

    return None


def choose_sheet(
    sheet_names,
    preferred_names,
):

    for preferred in preferred_names:

        preferred_normalized = normalize_column_name(
            preferred
        )

        for sheet_name in sheet_names:

            if (
                preferred_normalized
                in normalize_column_name(
                    sheet_name
                )
            ):

                return sheet_name

    return None


def normalize_boolean(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return normalized in {
        "TRUE",
        "YES",
        "Y",
        "1",
    }


def is_uncertain(
    value,
):

    return normalize_column_name(
        value
    ) in {
        "UNCERTAIN",
        "UNSURE",
        "UNKNOWN",
        "NOT_SURE",
        "CANNOT_DETERMINE",
        "CANT_DETERMINE",
    }


def categories_related(
    human_value,
    prediction,
):

    human_value = normalize_category(
        human_value
    )

    prediction = normalize_category(
        prediction
    )

    if (
        not human_value
        or not prediction
    ):

        return False

    if human_value == prediction:

        return True

    return any(
        (
            human_value in group
            and prediction in group
        )
        for group in RELATED_CAUSE_GROUPS
    )


def normalize_evidence_text(
    value,
):

    text = str(
        value
    ).lower()

    text = (
        text
        .replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def evaluate_evidence(
    evidence,
    narrative,
):

    evidence_text = str(
        evidence
        or ""
    ).strip()

    if not evidence_text:

        return {
            "valid":
                False,

            "exact":
                False,

            "normalized":
                False,

            "token_overlap":
                0.0,
        }

    exact = (
        evidence_text.lower()
        in str(
            narrative
        ).lower()
    )

    normalized_evidence = normalize_evidence_text(
        evidence
    )

    normalized_narrative = normalize_evidence_text(
        narrative
    )

    normalized_match = (
        bool(
            normalized_evidence
        )
        and normalized_evidence
        in normalized_narrative
    )

    evidence_tokens = set(
        normalized_evidence.split()
    )

    narrative_tokens = set(
        normalized_narrative.split()
    )

    overlap = (
        len(
            evidence_tokens.intersection(
                narrative_tokens
            )
        )
        / len(
            evidence_tokens
        )
        if evidence_tokens
        else 0.0
    )

    valid = (
        exact
        or normalized_match
        or overlap >= 0.85
    )

    return {
        "valid":
            valid,

        "exact":
            exact,

        "normalized":
            normalized_match,

        "token_overlap":
            round(
                overlap,
                4,
            ),
    }


# =============================================================================
# 8. JSON PARSING UTILITIES
# =============================================================================

def strip_markdown_fences(
    text,
):

    text = str(
        text
    ).strip()

    text = re.sub(
        r"^```(?:json)?",
        "",
        text,
        flags=re.IGNORECASE,
    )

    text = re.sub(
        r"```$",
        "",
        text,
    )

    return text.strip()


def extract_balanced_json(
    text,
):

    text = strip_markdown_fences(
        text
    )

    start = text.find(
        "{"
    )

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0

    in_string = False

    escaped = False

    for index in range(
        start,
        len(
            text
        ),
    ):

        character = text[
            index
        ]

        if escaped:

            escaped = False

            continue

        if (
            character == "\\"
            and in_string
        ):

            escaped = True

            continue

        if character == '"':

            in_string = not in_string

            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    index + 1
                ]

    candidate = text[
        start:
    ].strip()

    missing_braces = (
        candidate.count(
            "{"
        )
        - candidate.count(
            "}"
        )
    )

    if missing_braces > 0:

        candidate += (
            "}"
            * missing_braces
        )

        return candidate

    raise ValueError(
        "A complete JSON object was not found."
    )


def repair_json(
    value,
):

    output = str(
        value
    ).strip()

    output = (
        output
        .replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    output = re.sub(
        r",\s*}",
        "}",
        output,
    )

    output = re.sub(
        r":\s*None\b",
        ': ""',
        output,
    )

    output = re.sub(
        r":\s*True\b",
        ": true",
        output,
    )

    output = re.sub(
        r":\s*False\b",
        ": false",
        output,
    )

    return output


# =============================================================================
# 9. LOAD HUMAN-REVIEW WORKBOOK
# =============================================================================

excel_file = pd.ExcelFile(
    HUMAN_REVIEW_WORKBOOK,
    engine="openpyxl",
)


human_sheet = choose_sheet(
    excel_file.sheet_names,
    [
        "Human Review",
        "Review",
    ],
)


model_sheet = choose_sheet(
    excel_file.sheet_names,
    [
        "Model Output",
        "Model",
    ],
)


if human_sheet is None:

    raise KeyError(
        "The Human Review worksheet could not be identified."
    )


if model_sheet is None:

    raise KeyError(
        "The Model Output worksheet could not be identified."
    )


human_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=human_sheet,
    engine="openpyxl",
)


original_qwen_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=model_sheet,
    engine="openpyxl",
)


human_df.columns = [
    normalize_column_name(
        column
    )
    for column in human_df.columns
]


original_qwen_df.columns = [
    normalize_column_name(
        column
    )
    for column in original_qwen_df.columns
]


human_id_column = first_existing_column(
    human_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="human-review record ID",
)


human_text_column = first_existing_column(
    human_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
)


human_cause_column = first_existing_column(
    human_df,
    [
        "HUMAN_PRIMARY_CAUSE",
        "HUMAN_CAUSE",
        "PRIMARY_CAUSE_HUMAN",
    ],
    required=True,
    description="human primary-cause field",
)


human_cause_evidence_column = first_existing_column(
    human_df,
    [
        "HUMAN_CAUSE_EVIDENCE",
        "HUMAN_CAUSE_EVIDENCE_TEXT",
    ],
)


human_uncertainty_column = first_existing_column(
    human_df,
    [
        "HUMAN_CAUSE_UNCERTAIN",
        "CAUSE_UNCERTAIN",
        "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    ],
)


human_comment_column = first_existing_column(
    human_df,
    [
        "HUMAN_REVIEW_COMMENTS",
        "REVIEW_COMMENTS",
        "COMMENTS",
    ],
)


human_severity_column = first_existing_column(
    human_df,
    [
        "SEVERITY_GROUP",
        "INJURY_SEVERITY_GROUP",
    ],
)


human_time_column = first_existing_column(
    human_df,
    [
        "TIME_PERIOD",
        "ACCIDENT_TIME_PERIOD",
    ],
)


human_completion_column = first_existing_column(
    human_df,
    [
        "REVIEW_COMPLETED",
        "HUMAN_REVIEW_COMPLETED",
        "COMPLETED",
    ],
)


qwen_id_column = first_existing_column(
    original_qwen_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="original Qwen record ID",
)


qwen_cause_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_PRIMARY_CAUSE",
        "PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "MODEL_CAUSE",
    ],
    required=True,
    description="original Qwen primary cause",
)


qwen_confidence_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_CAUSE_CONFIDENCE",
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
    ],
)


qwen_evidence_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_CAUSE_EVIDENCE",
        "CAUSE_EVIDENCE_TEXT",
        "ORIGINAL_MODEL_CAUSE_EVIDENCE",
    ],
)


human_df[
    "SOURCE_RECORD_ID"
] = human_df[
    human_id_column
].astype(
    str
)


human_df[
    "HUMAN_PRIMARY_CAUSE"
] = human_df[
    human_cause_column
].map(
    normalize_category
)


if human_text_column:

    human_df[
        "NARRATIVE"
    ] = (
        human_df[
            human_text_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
        .str.strip()
    )

else:

    human_df[
        "NARRATIVE"
    ] = ""


human_df[
    "HUMAN_CAUSE_UNCERTAIN"
] = human_df[
    "HUMAN_PRIMARY_CAUSE"
].map(
    is_uncertain
)


if human_uncertainty_column:

    uncertainty_values = human_df[
        human_uncertainty_column
    ].map(
        normalize_column_name
    )

    human_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ] |= uncertainty_values.isin(
        {
            "UNCERTAIN",
            "UNSURE",
            "UNKNOWN",
            "NOT_SURE",
        }
    )


if human_cause_evidence_column:

    human_df[
        "HUMAN_CAUSE_EVIDENCE"
    ] = (
        human_df[
            human_cause_evidence_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
    )

else:

    human_df[
        "HUMAN_CAUSE_EVIDENCE"
    ] = ""


if human_comment_column:

    human_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = (
        human_df[
            human_comment_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
    )

else:

    human_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = ""


human_df[
    "SEVERITY_GROUP"
] = (
    human_df[
        human_severity_column
    ]
    if human_severity_column
    else "Unknown"
)


human_df[
    "TIME_PERIOD"
] = (
    human_df[
        human_time_column
    ]
    if human_time_column
    else "Unknown"
)


if human_completion_column:

    completion_mask = human_df[
        human_completion_column
    ].map(
        normalize_column_name
    ).isin(
        {
            "YES",
            "Y",
            "TRUE",
            "1",
            "COMPLETE",
            "COMPLETED",
        }
    )

    human_df = human_df[
        completion_mask
    ].copy()


original_qwen_merge_df = pd.DataFrame(
    {
        "SOURCE_RECORD_ID":
            original_qwen_df[
                qwen_id_column
            ].astype(
                str
            ),

        "ORIGINAL_QWEN_CAUSE":
            original_qwen_df[
                qwen_cause_column
            ].map(
                normalize_category
            ),
    }
)


original_qwen_merge_df[
    "ORIGINAL_QWEN_CONFIDENCE"
] = (
    pd.to_numeric(
        original_qwen_df[
            qwen_confidence_column
        ],
        errors="coerce",
    )
    if qwen_confidence_column
    else np.nan
)


original_qwen_merge_df[
    "ORIGINAL_QWEN_EVIDENCE"
] = (
    original_qwen_df[
        qwen_evidence_column
    ].astype(
        "string"
    ).fillna(
        ""
    )
    if qwen_evidence_column
    else ""
)


original_qwen_merge_df = (
    original_qwen_merge_df
    .drop_duplicates(
        subset=[
            "SOURCE_RECORD_ID"
        ],
        keep="last",
    )
)


benchmark_df = human_df.merge(
    original_qwen_merge_df,
    on="SOURCE_RECORD_ID",
    how="left",
)


# =============================================================================
# 10. FILL NARRATIVES FROM THE ACCIDENTS DATAFRAME IF NEEDED
# =============================================================================

accidents_df = accidents.copy()


accidents_df.columns = [
    normalize_column_name(
        column
    )
    for column in accidents_df.columns
]


accidents_id_column = first_existing_column(
    accidents_df,
    [
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
        "SOURCE_RECORD_ID",
    ],
    required=True,
    description="accidents DataFrame record ID",
)


accidents_text_column = first_existing_column(
    accidents_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
    required=True,
    description="accidents DataFrame narrative",
)


accidents_df[
    accidents_id_column
] = accidents_df[
    accidents_id_column
].astype(
    str
)


narrative_lookup = (
    accidents_df
    .drop_duplicates(
        subset=[
            accidents_id_column
        ],
        keep="first",
    )
    .set_index(
        accidents_id_column
    )[
        accidents_text_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .to_dict()
)


missing_narrative_mask = benchmark_df[
    "NARRATIVE"
].astype(
    str
).str.strip().eq(
    ""
)


benchmark_df.loc[
    missing_narrative_mask,
    "NARRATIVE",
] = (
    benchmark_df.loc[
        missing_narrative_mask,
        "SOURCE_RECORD_ID",
    ]
    .map(
        narrative_lookup
    )
    .fillna(
        ""
    )
)


benchmark_df = benchmark_df[
    benchmark_df[
        "NARRATIVE"
    ]
    .astype(
        str
    )
    .str.strip()
    .ne(
        ""
    )
].copy()


benchmark_df = benchmark_df[
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
    .isin(
        CAUSE_CATEGORIES
    )
].copy()


benchmark_df = benchmark_df.reset_index(
    drop=True
)


print(
    "Independent benchmark records:",
    f"{len(benchmark_df):,}",
)


# =============================================================================
# 11. LOAD HIGH-CONFIDENCE CODE BLOCK 2 REFERENCES
# =============================================================================

block2_df = pd.read_csv(
    block2_results_file,
    low_memory=False,
)


block2_df.columns = [
    normalize_column_name(
        column
    )
    for column in block2_df.columns
]


block2_id_column = first_existing_column(
    block2_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="Block 2 record ID",
)


block2_text_column = first_existing_column(
    block2_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
    required=True,
    description="Block 2 narrative",
)


block2_cause_column = first_existing_column(
    block2_df,
    [
        "PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "MODEL_PRIMARY_CAUSE",
        "REVISED_CAUSE",
    ],
    required=True,
    description="Block 2 primary cause",
)


block2_confidence_column = first_existing_column(
    block2_df,
    [
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
        "MODEL_CAUSE_CONFIDENCE",
    ],
)


block2_support_column = first_existing_column(
    block2_df,
    [
        "CAUSE_SUPPORT_LEVEL",
        "ORIGINAL_MODEL_CAUSE_SUPPORT",
        "MODEL_CAUSE_SUPPORT",
    ],
)


block2_evidence_column = first_existing_column(
    block2_df,
    [
        "CAUSE_EVIDENCE_TEXT",
        "ORIGINAL_MODEL_CAUSE_EVIDENCE",
        "MODEL_CAUSE_EVIDENCE",
    ],
)


block2_evidence_valid_column = first_existing_column(
    block2_df,
    [
        "CAUSE_EVIDENCE_VALID",
        "MODEL_CAUSE_EVIDENCE_VALID",
    ],
)


block2_record_type_column = first_existing_column(
    block2_df,
    [
        "RECORD_TYPE",
        "ORIGINAL_MODEL_RECORD_TYPE",
        "MODEL_RECORD_TYPE",
    ],
)


block2_status_column = first_existing_column(
    block2_df,
    [
        "PROCESSING_STATUS",
        "MODEL_PROCESSING_STATUS",
    ],
)


block2_df[
    "SOURCE_RECORD_ID"
] = block2_df[
    block2_id_column
].astype(
    str
)


block2_df[
    "NARRATIVE"
] = (
    block2_df[
        block2_text_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .str.strip()
)


block2_df[
    "REFERENCE_CAUSE"
] = block2_df[
    block2_cause_column
].map(
    normalize_category
)


block2_df[
    "REFERENCE_CONFIDENCE"
] = (
    pd.to_numeric(
        block2_df[
            block2_confidence_column
        ],
        errors="coerce",
    )
    if block2_confidence_column
    else np.nan
)


block2_df[
    "REFERENCE_SUPPORT"
] = (
    block2_df[
        block2_support_column
    ].map(
        normalize_category
    )
    if block2_support_column
    else "MISSING"
)


block2_df[
    "REFERENCE_EVIDENCE"
] = (
    block2_df[
        block2_evidence_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    if block2_evidence_column
    else ""
)


block2_mask = (
    block2_df[
        "NARRATIVE"
    ]
    .str.split()
    .str.len()
    .ge(
        5
    )
)


block2_mask &= block2_df[
    "REFERENCE_CAUSE"
].isin(
    CAUSE_CATEGORIES
)


block2_mask &= ~block2_df[
    "REFERENCE_CAUSE"
].isin(
    {
        "OTHER",
        "INSUFFICIENT_INFORMATION",
    }
)


if block2_status_column:

    block2_mask &= (
        block2_df[
            block2_status_column
        ]
        .astype(
            str
        )
        .str.upper()
        .eq(
            "SUCCESS"
        )
    )


if block2_record_type_column:

    block2_mask &= (
        block2_df[
            block2_record_type_column
        ]
        .map(
            normalize_category
        )
        .eq(
            "ACUTE_ACCIDENT"
        )
    )


if block2_confidence_column:

    block2_mask &= block2_df[
        "REFERENCE_CONFIDENCE"
    ].ge(
        MIN_BLOCK2_CAUSE_CONFIDENCE
    )


if block2_support_column:

    block2_mask &= block2_df[
        "REFERENCE_SUPPORT"
    ].isin(
        ALLOWED_BLOCK2_SUPPORT
    )


if block2_evidence_valid_column:

    block2_mask &= block2_df[
        block2_evidence_valid_column
    ].map(
        normalize_boolean
    )


block2_library_df = block2_df[
    block2_mask
].copy()


block2_library_df[
    "_NORMALIZED_NARRATIVE"
] = (
    block2_library_df[
        "NARRATIVE"
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


block2_library_df = (
    block2_library_df
    .drop_duplicates(
        subset=[
            "_NORMALIZED_NARRATIVE"
        ],
        keep="first",
    )
    .reset_index(
        drop=True
    )
)


block2_library_df[
    "REFERENCE_SOURCE"
] = "QWEN_HIGH_CONFIDENCE"


block2_library_df[
    "REFERENCE_REVIEW_STATUS"
] = "MODEL_HIGH_CONFIDENCE"


print(
    "High-confidence Block 2 references:",
    f"{len(block2_library_df):,}",
)


# =============================================================================
# 12. BUILD HUMAN-REVIEWED REFERENCE LIBRARY
#
# These records are used through leave-one-out retrieval.
# Each target is excluded from its own retrieved examples.
# =============================================================================

human_library_df = pd.DataFrame(
    {
        "SOURCE_RECORD_ID":
            benchmark_df[
                "SOURCE_RECORD_ID"
            ].astype(
                str
            ),

        "NARRATIVE":
            benchmark_df[
                "NARRATIVE"
            ].astype(
                str
            ),

        "REFERENCE_CAUSE":
            benchmark_df[
                "HUMAN_PRIMARY_CAUSE"
            ].map(
                normalize_category
            ),

        "REFERENCE_EVIDENCE":
            benchmark_df[
                "HUMAN_CAUSE_EVIDENCE"
            ].astype(
                str
            ),

        "REFERENCE_CONFIDENCE":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                0.50,
                1.00,
            ),

        "REFERENCE_SUPPORT":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                "UNCERTAIN",
                "HUMAN_REVIEWED",
            ),

        "REFERENCE_SOURCE":
            "HUMAN_REVIEWED",

        "REFERENCE_REVIEW_STATUS":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                "HUMAN_UNCERTAIN",
                "HUMAN_CERTAIN",
            ),
    }
)


human_library_df = human_library_df[
    human_library_df[
        "REFERENCE_CAUSE"
    ]
    .isin(
        CAUSE_CATEGORIES
    )
].copy()


# =============================================================================
# 13. COMBINE THE HYBRID REFERENCE LIBRARY
# =============================================================================

reference_columns = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "REFERENCE_CAUSE",
    "REFERENCE_EVIDENCE",
    "REFERENCE_CONFIDENCE",
    "REFERENCE_SUPPORT",
    "REFERENCE_SOURCE",
    "REFERENCE_REVIEW_STATUS",
]


reference_library_df = pd.concat(
    [
        block2_library_df[
            reference_columns
        ],
        human_library_df[
            reference_columns
        ],
    ],
    ignore_index=True,
)


reference_library_df[
    "SOURCE_RECORD_ID"
] = reference_library_df[
    "SOURCE_RECORD_ID"
].astype(
    str
)


reference_library_df[
    "NARRATIVE"
] = (
    reference_library_df[
        "NARRATIVE"
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .str.strip()
)


reference_library_df[
    "_NORMALIZED_NARRATIVE"
] = (
    reference_library_df[
        "NARRATIVE"
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


# Prefer human-reviewed records when identical narratives appear in both sources.
reference_library_df[
    "_SOURCE_PRIORITY"
] = reference_library_df[
    "REFERENCE_SOURCE"
].map(
    {
        "HUMAN_REVIEWED":
            0,

        "QWEN_HIGH_CONFIDENCE":
            1,
    }
).fillna(
    2
)


reference_library_df = (
    reference_library_df
    .sort_values(
        "_SOURCE_PRIORITY"
    )
    .drop_duplicates(
        subset=[
            "_NORMALIZED_NARRATIVE"
        ],
        keep="first",
    )
    .reset_index(
        drop=True
    )
)


print(
    "Combined hybrid reference library:",
    f"{len(reference_library_df):,}",
)


print(
    reference_library_df[
        "REFERENCE_SOURCE"
    ]
    .value_counts()
)


# =============================================================================
# 14. BUILD RETRIEVAL INDEXES
# =============================================================================

print("\n" + "=" * 95)
print("BUILDING RETRIEVAL INDEXES")
print("=" * 95)


embedding_model = SentenceTransformer(
    EMBEDDING_MODEL_ID
)


reference_embeddings = embedding_model.encode(
    reference_library_df[
        "NARRATIVE"
    ].tolist(),
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True,
)


benchmark_embeddings = embedding_model.encode(
    benchmark_df[
        "NARRATIVE"
    ].tolist(),
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True,
)


tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode",
    stop_words="english",
    ngram_range=TFIDF_NGRAM_RANGE,
    min_df=TFIDF_MIN_DF,
    max_df=TFIDF_MAX_DF,
    max_features=TFIDF_MAX_FEATURES,
    sublinear_tf=True,
    norm="l2",
)


combined_texts_for_tfidf = (
    reference_library_df[
        "NARRATIVE"
    ].tolist()
    + benchmark_df[
        "NARRATIVE"
    ].tolist()
)


combined_tfidf_matrix = tfidf_vectorizer.fit_transform(
    combined_texts_for_tfidf
)


reference_tfidf_matrix = combined_tfidf_matrix[
    :
    len(
        reference_library_df
    )
]


benchmark_tfidf_matrix = combined_tfidf_matrix[
    len(
        reference_library_df
    ):
]


# =============================================================================
# 15. HYBRID RETRIEVAL
# =============================================================================

def retrieve_examples(
    benchmark_index,
    target_record_id,
    target_narrative,
):

    semantic_scores = np.dot(
        reference_embeddings,
        benchmark_embeddings[
            benchmark_index
        ],
    )


    lexical_scores = cosine_similarity(
        benchmark_tfidf_matrix[
            benchmark_index
        ],
        reference_tfidf_matrix,
    ).ravel()


    hybrid_scores = (
        SEMANTIC_WEIGHT
        * semantic_scores
        + LEXICAL_WEIGHT
        * lexical_scores
    )


    candidates = reference_library_df.copy()


    candidates[
        "SEMANTIC_SCORE"
    ] = semantic_scores


    candidates[
        "LEXICAL_SCORE"
    ] = lexical_scores


    candidates[
        "HYBRID_SCORE"
    ] = hybrid_scores


    target_normalized = normalize_evidence_text(
        target_narrative
    )


    # Exclude the target by source ID.
    candidates = candidates[
        candidates[
            "SOURCE_RECORD_ID"
        ]
        .astype(
            str
        )
        .ne(
            str(
                target_record_id
            )
        )
    ].copy()


    # Exclude duplicate target text even if the identifier differs.
    candidates = candidates[
        candidates[
            "NARRATIVE"
        ]
        .map(
            normalize_evidence_text
        )
        .ne(
            target_normalized
        )
    ].copy()


    candidates = candidates[
        candidates[
            "HYBRID_SCORE"
        ]
        .ge(
            MIN_RETRIEVAL_SCORE
        )
    ].copy()


    candidates = candidates.sort_values(
        [
            "HYBRID_SCORE",
            "REFERENCE_SOURCE",
        ],
        ascending=[
            False,
            True,
        ],
    )


    selected_rows = []

    human_count = 0

    cause_counts = {}


    for _, candidate in candidates.iterrows():

        cause_value = candidate[
            "REFERENCE_CAUSE"
        ]


        current_cause_count = cause_counts.get(
            cause_value,
            0,
        )


        if (
            current_cause_count
            >= MAX_EXAMPLES_PER_CAUSE_PER_QUERY
        ):

            continue


        if (
            candidate[
                "REFERENCE_SOURCE"
            ]
            == "HUMAN_REVIEWED"
        ):

            if (
                human_count
                >= MAX_HUMAN_EXAMPLES_PER_QUERY
            ):

                continue

            human_count += 1


        selected_rows.append(
            candidate
        )


        cause_counts[
            cause_value
        ] = (
            current_cause_count
            + 1
        )


        if len(
            selected_rows
        ) >= TOP_K_RETRIEVAL:

            break


    if not selected_rows:

        return pd.DataFrame(
            columns=candidates.columns
        )


    output = pd.DataFrame(
        selected_rows
    ).reset_index(
        drop=True
    )


    output[
        "RETRIEVAL_RANK"
    ] = np.arange(
        1,
        len(
            output
        )
        + 1,
    )


    return output


# =============================================================================
# 16. RAG SYSTEM PROMPT
# =============================================================================

CAUSE_HIERARCHY = """
CAUSE-SELECTION HIERARCHY

Select the first category clearly supported by the TARGET NARRATIVE:

1. CAUGHT_IN_OR_BETWEEN
   Trapped, pinched, crushed, compressed, entangled, or held between objects.

2. VEHICLE_OR_MOBILE_EQUIPMENT_EVENT
   Collision, overturn, runaway, backing, or transportation event.

3. GROUND_OR_ROOF_FAILURE
   Roof, rib, face, highwall, slate, rock, or ground failure.

4. PRESSURE_OR_STORED_ENERGY_RELEASE
   Hydraulic, pneumatic, spring, tension, pressure, or stored-energy release.

5. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT
   Explicit loss of control of a tool, hose, handle, load, or component.

6. UNEXPECTED_EQUIPMENT_MOVEMENT
   Unexpected movement without explicit loss of control.

7. STRUCK_BY_MOVING_OBJECT
   Moving, falling, flying, rolling, or swinging object contact without
   trapping or compression.

8. STRUCK_AGAINST_STATIONARY_OBJECT
   The person's movement causes contact with a stationary object.

9. FOREIGN_BODY_OR_PARTICLE_CONTACT
   Chip, fragment, particle, or small foreign object contact.

10. SLIP_TRIP_OR_FALL
    Loss of footing, balance, or elevation.

11. AWKWARD_MOVEMENT_OR_BODY_MOTION
    Acute twist, reach, bend, turn, or step without a fall or substantial load.

12. MANUAL_HANDLING_OR_OVEREXERTION
    Acute injury from lifting, pushing, pulling, carrying, shoveling, holding,
    or exerting force.

Do not use STRUCK_BY_MOVING_OBJECT as a general fallback when another
specific mechanism is clearly supported.
""".strip()


RAG_SYSTEM_PROMPT = f"""
You classify the PRIMARY ACCIDENT CAUSE in short MSHA accident narratives.

This is reported-event classification, not verified root-cause analysis.

ALLOWED PRIMARY CAUSES
{", ".join(CAUSE_CATEGORIES)}

{CAUSE_HIERARCHY}

You will receive:

1. A TARGET NARRATIVE that must be classified.
2. Retrieved historical examples with accepted or high-confidence cause labels.

RETRIEVAL RULES

- Retrieved examples are references for terminology and category boundaries.
- The TARGET NARRATIVE is the only source of facts about the target event.
- Never copy an event detail, cause, or evidence excerpt merely because it
  appears in a retrieved example.
- A retrieved example may be similar but still have a different cause.
- Human-reviewed examples are stronger references than model-generated examples.
- Examples marked HUMAN_UNCERTAIN are weak references.
- If retrieved examples disagree, classify from the TARGET NARRATIVE.
- Cause evidence must be copied from the TARGET NARRATIVE, not from a retrieved
  example.
- Use INSUFFICIENT_INFORMATION when the target does not support an event
  mechanism.

CONFIDENCE

- HIGH confidence generally corresponds to 0.80-1.00.
- MEDIUM confidence generally corresponds to 0.60-0.79.
- LOW confidence is below 0.60.
- Confidence is an analytical estimate, not a calibrated probability.

Return only one JSON object with exactly these keys:

{{
  "primary_cause": "",
  "cause_evidence_text": "",
  "cause_confidence": 0.0,
  "cause_support_level": "HIGH",
  "retrieval_used": true,
  "retrieval_explanation": "",
  "alternative_cause": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 17. MODEL LOADING
# =============================================================================

tokenizer = None

model = None


def load_qwen_model():

    global tokenizer
    global model


    if (
        tokenizer is not None
        and model is not None
    ):

        return


    if not torch.cuda.is_available():

        raise RuntimeError(
            "No NVIDIA GPU is available. "
            "Qwen inference requires compatible GPU compute."
        )


    print("\n" + "=" * 95)
    print("LOADING LOCAL QWEN MODEL")
    print("=" * 95)


    quantization_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )


    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_ID,
        use_fast=True,
        trust_remote_code=False,
    )


    if tokenizer.pad_token_id is None:

        tokenizer.pad_token = tokenizer.eos_token


    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        quantization_config=quantization_config,
        device_map="auto",
        dtype=torch.float16,
        low_cpu_mem_usage=True,
        trust_remote_code=False,
    )


    model.eval()


    print(
        "Model loaded:",
        MODEL_ID,
    )


# =============================================================================
# 18. MODEL GENERATION
# =============================================================================

def generate_response(
    messages,
):

    load_qwen_model()


    model_inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )


    model_inputs = {
        key:
            value.to(
                model.device
            )
        for key, value in model_inputs.items()
    }


    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[
            -1
        ]
    )


    with torch.inference_mode():

        generated = model.generate(
            **model_inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.02,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )


    generated_ids = generated[
        0,
        input_length:
    ]


    response_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )


    del model_inputs
    del generated
    del generated_ids


    return response_text


# =============================================================================
# 19. BUILD THE RAG USER PROMPT
# =============================================================================

def format_retrieved_examples(
    retrieved_df,
):

    if retrieved_df.empty:

        return (
            "No retrieved example exceeded the minimum similarity threshold."
        )


    sections = []


    for _, row in retrieved_df.iterrows():

        sections.append(
            "\n".join(
                [
                    (
                        f"RETRIEVED EXAMPLE "
                        f"{int(row['RETRIEVAL_RANK'])}"
                    ),
                    (
                        f"Source quality: "
                        f"{row['REFERENCE_REVIEW_STATUS']}"
                    ),
                    (
                        f"Hybrid similarity: "
                        f"{row['HYBRID_SCORE']:.4f}"
                    ),
                    (
                        f"Accepted cause: "
                        f"{row['REFERENCE_CAUSE']}"
                    ),
                    (
                        f"Historical narrative: "
                        f"{row['NARRATIVE']}"
                    ),
                    (
                        f"Historical cause evidence: "
                        f"{row['REFERENCE_EVIDENCE']}"
                    ),
                ]
            )
        )


    return "\n\n".join(
        sections
    )


def build_user_prompt(
    target_narrative,
    retrieved_df,
):

    retrieved_text = format_retrieved_examples(
        retrieved_df
    )


    return f"""
TARGET NARRATIVE
{target_narrative}

RETRIEVED HISTORICAL EXAMPLES
{retrieved_text}

Classify only the TARGET NARRATIVE.

Use the retrieved examples to understand terminology and category boundaries,
but do not transfer facts from a historical example to the target.

Copy the cause evidence from the TARGET NARRATIVE.

Return only the required JSON object.
""".strip()


# =============================================================================
# 20. CLASSIFY ONE TARGET WITH RAG
# =============================================================================

def classify_with_rag(
    target_narrative,
    retrieved_df,
):

    user_prompt = build_user_prompt(
        target_narrative,
        retrieved_df,
    )


    last_response = ""

    last_error = ""


    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    RAG_SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]


        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        f"""
The previous response was invalid.

VALIDATION ERROR
{last_error}

PREVIOUS RESPONSE
{last_response}

Return one corrected JSON object only.
Use only the allowed primary-cause categories.
Do not include Markdown or any text outside the JSON object.
""".strip(),
                }
            )


        try:

            raw_response = generate_response(
                messages
            )


            last_response = raw_response


            json_text = extract_balanced_json(
                raw_response
            )


            json_text = repair_json(
                json_text
            )


            parsed_dictionary = json.loads(
                json_text
            )


            parsed_dictionary[
                "primary_cause"
            ] = normalize_category(
                parsed_dictionary.get(
                    "primary_cause",
                    "",
                )
            )


            alternative_value = parsed_dictionary.get(
                "alternative_cause",
                "",
            )


            if alternative_value:

                parsed_dictionary[
                    "alternative_cause"
                ] = normalize_category(
                    alternative_value
                )


            parsed_dictionary[
                "cause_support_level"
            ] = normalize_category(
                parsed_dictionary.get(
                    "cause_support_level",
                    "LOW",
                )
            )


            parsed_dictionary[
                "retrieval_used"
            ] = bool(
                parsed_dictionary.get(
                    "retrieval_used",
                    True,
                )
            )


            validated = RAGCauseExtraction.model_validate(
                parsed_dictionary
            )


            evidence_result = evaluate_evidence(
                validated.cause_evidence_text,
                target_narrative,
            )


            return {
                **validated.model_dump(),

                "RAG_SCHEMA_VALID":
                    True,

                "RAG_CAUSE_EVIDENCE_VALID":
                    evidence_result[
                        "valid"
                    ],

                "RAG_CAUSE_EVIDENCE_EXACT":
                    evidence_result[
                        "exact"
                    ],

                "RAG_CAUSE_EVIDENCE_NORMALIZED":
                    evidence_result[
                        "normalized"
                    ],

                "RAG_CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    evidence_result[
                        "token_overlap"
                    ],

                "RAG_PROCESSING_STATUS":
                    "SUCCESS",

                "RAG_PROCESSING_ERROR":
                    "",

                "RAG_ATTEMPT_COUNT":
                    attempt_number,

                "RAG_RAW_RESPONSE":
                    raw_response,
            }


        except Exception as error:

            last_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )


            print(
                f"  Attempt {attempt_number} failed: "
                f"{last_error[:250]}"
            )


            gc.collect()


            if torch.cuda.is_available():

                torch.cuda.empty_cache()


    return {
        "primary_cause":
            "INSUFFICIENT_INFORMATION",

        "cause_evidence_text":
            "",

        "cause_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "retrieval_used":
            not retrieved_df.empty,

        "retrieval_explanation":
            "",

        "alternative_cause":
            "",

        "reviewer_note":
            "",

        "RAG_SCHEMA_VALID":
            False,

        "RAG_CAUSE_EVIDENCE_VALID":
            False,

        "RAG_CAUSE_EVIDENCE_EXACT":
            False,

        "RAG_CAUSE_EVIDENCE_NORMALIZED":
            False,

        "RAG_CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "RAG_PROCESSING_STATUS":
            "FAILED",

        "RAG_PROCESSING_ERROR":
            last_error,

        "RAG_ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAG_RAW_RESPONSE":
            last_response,
    }


# =============================================================================
# 21. CHECKPOINT UTILITIES
# =============================================================================

def load_checkpoint(
    path,
):

    path = Path(
        path
    )


    if not path.exists():

        return {}


    records = {}


    with path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()


            if not line:

                continue


            try:

                record = json.loads(
                    line
                )


                records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record


            except Exception:

                continue


    return records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


checkpoint_records = load_checkpoint(
    CHECKPOINT_JSONL
)


print(
    "Existing RAG checkpoint records:",
    f"{len(checkpoint_records):,}",
)


# =============================================================================
# 22. RUN RAG ON THE HUMAN-REVIEW BENCHMARK
# =============================================================================

rag_records = []

retrieval_detail_records = []


for benchmark_index, row in benchmark_df.iterrows():

    record_id = str(
        row[
            "SOURCE_RECORD_ID"
        ]
    )


    if (
        not REPROCESS_COMPLETED
        and record_id
        in checkpoint_records
    ):

        print(
            f"[{benchmark_index + 1}/"
            f"{len(benchmark_df)}] "
            f"Checkpoint: {record_id}"
        )


        rag_records.append(
            checkpoint_records[
                record_id
            ]
        )


        continue


    target_narrative = str(
        row[
            "NARRATIVE"
        ]
    )


    retrieved_df = retrieve_examples(
        benchmark_index=benchmark_index,
        target_record_id=record_id,
        target_narrative=target_narrative,
    )


    for _, retrieved_row in retrieved_df.iterrows():

        retrieval_detail_records.append(
            {
                "TARGET_RECORD_ID":
                    record_id,

                "TARGET_HUMAN_CAUSE":
                    row[
                        "HUMAN_PRIMARY_CAUSE"
                    ],

                "RETRIEVAL_RANK":
                    int(
                        retrieved_row[
                            "RETRIEVAL_RANK"
                        ]
                    ),

                "REFERENCE_RECORD_ID":
                    retrieved_row[
                        "SOURCE_RECORD_ID"
                    ],

                "REFERENCE_SOURCE":
                    retrieved_row[
                        "REFERENCE_SOURCE"
                    ],

                "REFERENCE_REVIEW_STATUS":
                    retrieved_row[
                        "REFERENCE_REVIEW_STATUS"
                    ],

                "REFERENCE_CAUSE":
                    retrieved_row[
                        "REFERENCE_CAUSE"
                    ],

                "SEMANTIC_SCORE":
                    float(
                        retrieved_row[
                            "SEMANTIC_SCORE"
                        ]
                    ),

                "LEXICAL_SCORE":
                    float(
                        retrieved_row[
                            "LEXICAL_SCORE"
                        ]
                    ),

                "HYBRID_SCORE":
                    float(
                        retrieved_row[
                            "HYBRID_SCORE"
                        ]
                    ),

                "REFERENCE_NARRATIVE":
                    retrieved_row[
                        "NARRATIVE"
                    ],
            }
        )


    print(
        f"[{benchmark_index + 1}/"
        f"{len(benchmark_df)}] "
        f"RAG inference | "
        f"{record_id} | "
        f"{len(retrieved_df)} examples"
    )


    start_time = time.time()


    rag_result = classify_with_rag(
        target_narrative=target_narrative,
        retrieved_df=retrieved_df,
    )


    top_reference_cause = (
        retrieved_df.iloc[
            0
        ][
            "REFERENCE_CAUSE"
        ]
        if not retrieved_df.empty
        else ""
    )


    top_reference_score = (
        float(
            retrieved_df.iloc[
                0
            ][
                "HYBRID_SCORE"
            ]
        )
        if not retrieved_df.empty
        else np.nan
    )


    retrieved_causes = (
        "; ".join(
            retrieved_df[
                "REFERENCE_CAUSE"
            ].astype(
                str
            ).tolist()
        )
        if not retrieved_df.empty
        else ""
    )


    retrieved_sources = (
        "; ".join(
            retrieved_df[
                "REFERENCE_SOURCE"
            ].astype(
                str
            ).tolist()
        )
        if not retrieved_df.empty
        else ""
    )


    record = {
        "SOURCE_RECORD_ID":
            record_id,

        "NARRATIVE":
            target_narrative,

        "SEVERITY_GROUP":
            row.get(
                "SEVERITY_GROUP",
                "Unknown",
            ),

        "TIME_PERIOD":
            row.get(
                "TIME_PERIOD",
                "Unknown",
            ),

        "HUMAN_PRIMARY_CAUSE":
            row[
                "HUMAN_PRIMARY_CAUSE"
            ],

        "HUMAN_CAUSE_UNCERTAIN":
            bool(
                row[
                    "HUMAN_CAUSE_UNCERTAIN"
                ]
            ),

        "HUMAN_REVIEW_COMMENTS":
            row.get(
                "HUMAN_REVIEW_COMMENTS",
                "",
            ),

        "ORIGINAL_QWEN_CAUSE":
            row.get(
                "ORIGINAL_QWEN_CAUSE",
                "",
            ),

        "ORIGINAL_QWEN_CONFIDENCE":
            row.get(
                "ORIGINAL_QWEN_CONFIDENCE",
                np.nan,
            ),

        "ORIGINAL_QWEN_EVIDENCE":
            row.get(
                "ORIGINAL_QWEN_EVIDENCE",
                "",
            ),

        "RETRIEVED_EXAMPLE_COUNT":
            len(
                retrieved_df
            ),

        "TOP_RETRIEVED_CAUSE":
            top_reference_cause,

        "TOP_RETRIEVAL_SCORE":
            top_reference_score,

        "RETRIEVED_CAUSES":
            retrieved_causes,

        "RETRIEVED_SOURCES":
            retrieved_sources,

        **rag_result,

        "MODEL_ID":
            MODEL_ID,

        "EMBEDDING_MODEL_ID":
            EMBEDDING_MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "RAG_PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }


    rag_records.append(
        record
    )


    append_checkpoint(
        record,
        CHECKPOINT_JSONL,
    )


    gc.collect()


    if torch.cuda.is_available():

        torch.cuda.empty_cache()


# =============================================================================
# 23. BUILD FINAL RAG RESULTS
# =============================================================================

results_df = pd.DataFrame(
    rag_records
)


results_df[
    "RAG_PRIMARY_CAUSE"
] = results_df[
    "primary_cause"
].map(
    normalize_category
)


results_df[
    "RAG_ALTERNATIVE_CAUSE"
] = results_df[
    "alternative_cause"
].map(
    normalize_category
)


results_df[
    "RAG_EXACT_MATCH"
] = (
    results_df[
        "RAG_PRIMARY_CAUSE"
    ]
    == results_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


results_df[
    "RAG_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_PRIMARY_CAUSE"
        ],
    )
]


results_df[
    "RAG_TOP2_EXACT_MATCH"
] = (
    results_df[
        "RAG_EXACT_MATCH"
    ]
    |
    (
        results_df[
            "RAG_ALTERNATIVE_CAUSE"
        ]
        == results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
)


results_df[
    "RAG_TOP2_EXACT_OR_RELATED"
] = [
    (
        categories_related(
            human_value,
            primary_value,
        )
        or categories_related(
            human_value,
            alternative_value,
        )
    )
    for (
        human_value,
        primary_value,
        alternative_value,
    ) in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_ALTERNATIVE_CAUSE"
        ],
    )
]


results_df[
    "ORIGINAL_QWEN_EXACT_MATCH"
] = (
    results_df[
        "ORIGINAL_QWEN_CAUSE"
    ]
    == results_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


results_df[
    "ORIGINAL_QWEN_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "ORIGINAL_QWEN_CAUSE"
        ],
    )
]


results_df[
    "RAG_CHANGED_QWEN_CAUSE"
] = (
    results_df[
        "RAG_PRIMARY_CAUSE"
    ]
    != results_df[
        "ORIGINAL_QWEN_CAUSE"
    ]
)


results_df[
    "RAG_IMPROVED_EXACT"
] = (
    results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    ~results_df[
        "ORIGINAL_QWEN_EXACT_MATCH"
    ]
)


results_df[
    "RAG_REDUCED_EXACT"
] = (
    ~results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    results_df[
        "ORIGINAL_QWEN_EXACT_MATCH"
    ]
)


# =============================================================================
# 24. MERGE OPTIONAL LOGISTIC-REGRESSION RESULTS
# =============================================================================

if logistic_results_file:

    logistic_df = pd.read_csv(
        logistic_results_file,
        low_memory=False,
    )


    logistic_df.columns = [
        normalize_column_name(
            column
        )
        for column in logistic_df.columns
    ]


    logistic_id_column = first_existing_column(
        logistic_df,
        [
            "SOURCE_RECORD_ID",
            "DOCUMENT_NO",
            "DOC_NO",
            "ACCIDENT_ID",
        ],
        required=True,
        description="logistic result ID",
    )


    logistic_cause_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP1_CAUSE",
            "PREDICTED_PRIMARY_CAUSE",
        ],
        required=True,
        description="logistic cause",
    )


    logistic_probability_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP1_PROBABILITY",
            "PREDICTED_PROBABILITY",
        ],
    )


    logistic_top2_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP2_CAUSE",
        ],
    )


    logistic_merge_df = pd.DataFrame(
        {
            "SOURCE_RECORD_ID":
                logistic_df[
                    logistic_id_column
                ].astype(
                    str
                ),

            "LOGISTIC_CAUSE":
                logistic_df[
                    logistic_cause_column
                ].map(
                    normalize_category
                ),
        }
    )


    logistic_merge_df[
        "LOGISTIC_PROBABILITY"
    ] = (
        pd.to_numeric(
            logistic_df[
                logistic_probability_column
            ],
            errors="coerce",
        )
        if logistic_probability_column
        else np.nan
    )


    logistic_merge_df[
        "LOGISTIC_TOP2_CAUSE"
    ] = (
        logistic_df[
            logistic_top2_column
        ].map(
            normalize_category
        )
        if logistic_top2_column
        else ""
    )


    logistic_merge_df = (
        logistic_merge_df
        .drop_duplicates(
            subset=[
                "SOURCE_RECORD_ID"
            ],
            keep="last",
        )
    )


    results_df = results_df.merge(
        logistic_merge_df,
        on="SOURCE_RECORD_ID",
        how="left",
    )


    results_df[
        "LOGISTIC_EXACT_MATCH"
    ] = (
        results_df[
            "LOGISTIC_CAUSE"
        ]
        == results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )


    results_df[
        "LOGISTIC_EXACT_OR_RELATED"
    ] = [
        categories_related(
            human_value,
            prediction,
        )
        for human_value, prediction in zip(
            results_df[
                "HUMAN_PRIMARY_CAUSE"
            ],
            results_df[
                "LOGISTIC_CAUSE"
            ],
        )
    ]


    results_df[
        "LOGISTIC_TOP2_EXACT"
    ] = (
        results_df[
            "LOGISTIC_EXACT_MATCH"
        ]
        |
        (
            results_df[
                "LOGISTIC_TOP2_CAUSE"
            ]
            == results_df[
                "HUMAN_PRIMARY_CAUSE"
            ]
        )
    )


    results_df[
        "LOGISTIC_TOP2_EXACT_OR_RELATED"
    ] = [
        (
            categories_related(
                human_value,
                top1_value,
            )
            or categories_related(
                human_value,
                top2_value,
            )
        )
        for (
            human_value,
            top1_value,
            top2_value,
        ) in zip(
            results_df[
                "HUMAN_PRIMARY_CAUSE"
            ],
            results_df[
                "LOGISTIC_CAUSE"
            ],
            results_df[
                "LOGISTIC_TOP2_CAUSE"
            ],
        )
    ]


# =============================================================================
# 25. SAVE RECORD-LEVEL OUTPUTS
# =============================================================================

results_df.to_csv(
    OUTPUT_RESULTS_CSV,
    index=False,
    encoding="utf-8-sig",
)


retrieval_details_df = pd.DataFrame(
    retrieval_detail_records
)


if not retrieval_details_df.empty:

    retrieval_details_df.to_csv(
        OUTPUT_RETRIEVAL_CSV,
        index=False,
        encoding="utf-8-sig",
    )


# =============================================================================
# 26. RETRIEVAL DIAGNOSTICS
# =============================================================================

if not retrieval_details_df.empty:

    retrieval_details_df[
        "EXACT_CAUSE_RETRIEVAL"
    ] = (
        retrieval_details_df[
            "REFERENCE_CAUSE"
        ]
        == retrieval_details_df[
            "TARGET_HUMAN_CAUSE"
        ]
    )


    retrieval_details_df[
        "RELATED_CAUSE_RETRIEVAL"
    ] = [
        categories_related(
            target_value,
            reference_value,
        )
        for (
            target_value,
            reference_value,
        ) in zip(
            retrieval_details_df[
                "TARGET_HUMAN_CAUSE"
            ],
            retrieval_details_df[
                "REFERENCE_CAUSE"
            ],
        )
    ]


    retrieval_quality_by_target_df = (
        retrieval_details_df
        .groupby(
            "TARGET_RECORD_ID"
        )
        .agg(
            Retrieved_records=(
                "REFERENCE_RECORD_ID",
                "count",
            ),
            Top_score=(
                "HYBRID_SCORE",
                "max",
            ),
            Mean_score=(
                "HYBRID_SCORE",
                "mean",
            ),
            Exact_retrieval_rate=(
                "EXACT_CAUSE_RETRIEVAL",
                "mean",
            ),
            Related_retrieval_rate=(
                "RELATED_CAUSE_RETRIEVAL",
                "mean",
            ),
            Human_reference_count=(
                "REFERENCE_SOURCE",
                lambda values:
                    int(
                        (
                            values
                            == "HUMAN_REVIEWED"
                        ).sum()
                    ),
            ),
        )
        .reset_index()
    )


    results_df = results_df.merge(
        retrieval_quality_by_target_df,
        left_on="SOURCE_RECORD_ID",
        right_on="TARGET_RECORD_ID",
        how="left",
    )


else:

    retrieval_quality_by_target_df = pd.DataFrame()


# =============================================================================
# 27. PERFORMANCE UTILITIES
# =============================================================================

def evaluate_method(
    dataframe,
    prediction_column,
    model_name,
    related_column=None,
    top2_column=None,
    top2_related_column=None,
):

    valid_df = dataframe[
        dataframe[
            prediction_column
        ]
        .notna()
    ].copy()


    valid_df = valid_df[
        valid_df[
            prediction_column
        ]
        .astype(
            str
        )
        .str.strip()
        .ne(
            ""
        )
    ]


    if valid_df.empty:

        return {
            "Model":
                model_name,

            "Records":
                0,
        }


    y_true = valid_df[
        "HUMAN_PRIMARY_CAUSE"
    ]


    y_pred = valid_df[
        prediction_column
    ]


    output = {
        "Model":
            model_name,

        "Records":
            len(
                valid_df
            ),

        "Exact accuracy":
            accuracy_score(
                y_true,
                y_pred,
            ),

        "Balanced accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred,
            ),

        "Macro F1":
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            ),

        "Weighted F1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0,
            ),

        "Cohen's kappa":
            cohen_kappa_score(
                y_true,
                y_pred,
            ),
    }


    if related_column:

        output[
            "Exact-or-related accuracy"
        ] = float(
            valid_df[
                related_column
            ].mean()
        )


    if top2_column:

        output[
            "Top-two exact accuracy"
        ] = float(
            valid_df[
                top2_column
            ].mean()
        )


    if top2_related_column:

        output[
            "Top-two exact-or-related"
        ] = float(
            valid_df[
                top2_related_column
            ].mean()
        )


    return output


# =============================================================================
# 28. OVERALL METHOD COMPARISON
# =============================================================================

method_metrics = [
    evaluate_method(
        results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen",
        related_column="RAG_EXACT_OR_RELATED",
        top2_column="RAG_TOP2_EXACT_MATCH",
        top2_related_column="RAG_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen",
        related_column="ORIGINAL_QWEN_EXACT_OR_RELATED",
    ),
]


if logistic_results_file:

    method_metrics.append(
        evaluate_method(
            results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression",
            related_column="LOGISTIC_EXACT_OR_RELATED",
            top2_column="LOGISTIC_TOP2_EXACT",
            top2_related_column="LOGISTIC_TOP2_EXACT_OR_RELATED",
        )
    )


overall_metrics_df = pd.DataFrame(
    method_metrics
)


certain_results_df = results_df[
    ~results_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


certain_metrics = [
    evaluate_method(
        certain_results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen, certain human labels",
        related_column="RAG_EXACT_OR_RELATED",
        top2_column="RAG_TOP2_EXACT_MATCH",
        top2_related_column="RAG_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        certain_results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen, certain human labels",
        related_column="ORIGINAL_QWEN_EXACT_OR_RELATED",
    ),
]


if logistic_results_file:

    certain_metrics.append(
        evaluate_method(
            certain_results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression, certain human labels",
            related_column="LOGISTIC_EXACT_OR_RELATED",
            top2_column="LOGISTIC_TOP2_EXACT",
            top2_related_column="LOGISTIC_TOP2_EXACT_OR_RELATED",
        )
    )


certain_metrics_df = pd.DataFrame(
    certain_metrics
)


# =============================================================================
# 29. CATEGORY-LEVEL METRICS
# =============================================================================

evaluation_labels = sorted(
    set(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
    .union(
        results_df[
            "RAG_PRIMARY_CAUSE"
        ]
    )
    .union(
        results_df[
            "ORIGINAL_QWEN_CAUSE"
        ]
        .dropna()
        .astype(
            str
        )
    )
)


if logistic_results_file:

    evaluation_labels = sorted(
        set(
            evaluation_labels
        )
        .union(
            results_df[
                "LOGISTIC_CAUSE"
            ]
            .dropna()
            .astype(
                str
            )
        )
    )


def category_metrics(
    dataframe,
    prediction_column,
    method_name,
    labels,
):

    precision, recall, f1, support = (
        precision_recall_fscore_support(
            dataframe[
                "HUMAN_PRIMARY_CAUSE"
            ],
            dataframe[
                prediction_column
            ],
            labels=labels,
            zero_division=0,
        )
    )


    return pd.DataFrame(
        {
            "Method":
                method_name,

            "Primary cause":
                labels,

            "Precision":
                precision,

            "Recall":
                recall,

            "F1":
                f1,

            "Human support":
                support,
        }
    )


category_metrics_frames = [
    category_metrics(
        results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen",
        evaluation_labels,
    ),

    category_metrics(
        results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen",
        evaluation_labels,
    ),
]


if logistic_results_file:

    category_metrics_frames.append(
        category_metrics(
            results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression",
            evaluation_labels,
        )
    )


category_metrics_df = pd.concat(
    category_metrics_frames,
    ignore_index=True,
)


# =============================================================================
# 30. CONFUSION MATRICES
# =============================================================================

rag_confusion_matrix = confusion_matrix(
    results_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    results_df[
        "RAG_PRIMARY_CAUSE"
    ],
    labels=evaluation_labels,
)


qwen_confusion_matrix = confusion_matrix(
    results_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    results_df[
        "ORIGINAL_QWEN_CAUSE"
    ],
    labels=evaluation_labels,
)


if logistic_results_file:

    logistic_confusion_matrix = confusion_matrix(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "LOGISTIC_CAUSE"
        ],
        labels=evaluation_labels,
    )


# =============================================================================
# 31. PERFORMANCE BY RETRIEVAL QUALITY
# =============================================================================

if (
    "Top_score"
    in results_df.columns
):

    results_df[
        "RETRIEVAL_SCORE_BAND"
    ] = pd.cut(
        results_df[
            "Top_score"
        ],
        bins=[
            -np.inf,
            0.35,
            0.50,
            0.65,
            np.inf,
        ],
        labels=[
            "Low: Below 0.35",
            "Moderate: 0.35-0.49",
            "Strong: 0.50-0.64",
            "Very strong: 0.65+",
        ],
    )


    retrieval_band_records = []


    for band, group_df in results_df.groupby(
        "RETRIEVAL_SCORE_BAND",
        observed=False,
    ):

        if group_df.empty:

            continue


        retrieval_band_records.append(
            {
                "Retrieval score band":
                    str(
                        band
                    ),

                "Records":
                    len(
                        group_df
                    ),

                "RAG exact":
                    float(
                        group_df[
                            "RAG_EXACT_MATCH"
                        ].mean()
                    ),

                "RAG exact or related":
                    float(
                        group_df[
                            "RAG_EXACT_OR_RELATED"
                        ].mean()
                    ),

                "Original Qwen exact":
                    float(
                        group_df[
                            "ORIGINAL_QWEN_EXACT_MATCH"
                        ].mean()
                    ),

                "Mean exact retrieval rate":
                    float(
                        group_df[
                            "Exact_retrieval_rate"
                        ].mean()
                    ),
            }
        )


    retrieval_band_performance_df = pd.DataFrame(
        retrieval_band_records
    )


else:

    retrieval_band_performance_df = pd.DataFrame()


# =============================================================================
# 32. PERFORMANCE BY SEVERITY AND TIME
# =============================================================================

def grouped_method_comparison(
    dataframe,
    group_column,
):

    records = []


    for group_value, group_df in dataframe.groupby(
        group_column,
        dropna=False,
    ):

        record = {
            group_column:
                group_value,

            "Records":
                len(
                    group_df
                ),

            "RAG exact":
                float(
                    group_df[
                        "RAG_EXACT_MATCH"
                    ].mean()
                ),

            "RAG exact or related":
                float(
                    group_df[
                        "RAG_EXACT_OR_RELATED"
                    ].mean()
                ),

            "Original Qwen exact":
                float(
                    group_df[
                        "ORIGINAL_QWEN_EXACT_MATCH"
                    ].mean()
                ),

            "Original Qwen exact or related":
                float(
                    group_df[
                        "ORIGINAL_QWEN_EXACT_OR_RELATED"
                    ].mean()
                ),

            "Human uncertain":
                float(
                    group_df[
                        "HUMAN_CAUSE_UNCERTAIN"
                    ].mean()
                ),
        }


        if logistic_results_file:

            record[
                "Logistic exact"
            ] = float(
                group_df[
                    "LOGISTIC_EXACT_MATCH"
                ].mean()
            )


            record[
                "Logistic exact or related"
            ] = float(
                group_df[
                    "LOGISTIC_EXACT_OR_RELATED"
                ].mean()
            )


        records.append(
            record
        )


    return pd.DataFrame(
        records
    )


severity_comparison_df = grouped_method_comparison(
    results_df,
    "SEVERITY_GROUP",
)


time_comparison_df = grouped_method_comparison(
    results_df,
    "TIME_PERIOD",
)


# =============================================================================
# 33. CHANGE ANALYSIS
# =============================================================================

change_summary_df = pd.DataFrame(
    [
        {
            "Outcome":
                "RAG retained the original Qwen cause",

            "Records":
                int(
                    (
                        ~results_df[
                            "RAG_CHANGED_QWEN_CAUSE"
                        ]
                    ).sum()
                ),
        },
        {
            "Outcome":
                "RAG changed the original Qwen cause",

            "Records":
                int(
                    results_df[
                        "RAG_CHANGED_QWEN_CAUSE"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG improved an exact disagreement",

            "Records":
                int(
                    results_df[
                        "RAG_IMPROVED_EXACT"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG changed an exact Qwen match to a disagreement",

            "Records":
                int(
                    results_df[
                        "RAG_REDUCED_EXACT"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG cause evidence traceable to target narrative",

            "Records":
                int(
                    results_df[
                        "RAG_CAUSE_EVIDENCE_VALID"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG processing failures",

            "Records":
                int(
                    results_df[
                        "RAG_PROCESSING_STATUS"
                    ]
                    .ne(
                        "SUCCESS"
                    )
                    .sum()
                ),
        },
    ]
)


# =============================================================================
# 34. SAVE SUMMARY JSON
# =============================================================================

summary_payload = {
    "created_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "model_id":
        MODEL_ID,

    "embedding_model_id":
        EMBEDDING_MODEL_ID,

    "prompt_version":
        PROMPT_VERSION,

    "human_benchmark_records":
        len(
            results_df
        ),

    "human_reference_records":
        int(
            (
                reference_library_df[
                    "REFERENCE_SOURCE"
                ]
                == "HUMAN_REVIEWED"
            ).sum()
        ),

    "qwen_reference_records":
        int(
            (
                reference_library_df[
                    "REFERENCE_SOURCE"
                ]
                == "QWEN_HIGH_CONFIDENCE"
            ).sum()
        ),

    "semantic_weight":
        SEMANTIC_WEIGHT,

    "lexical_weight":
        LEXICAL_WEIGHT,

    "top_k":
        TOP_K_RETRIEVAL,

    "minimum_retrieval_score":
        MIN_RETRIEVAL_SCORE,

    "method_metrics":
        overall_metrics_df.to_dict(
            orient="records"
        ),

    "outputs": {
        "pdf":
            OUTPUT_PDF,

        "results_csv":
            OUTPUT_RESULTS_CSV,

        "retrieval_csv":
            OUTPUT_RETRIEVAL_CSV,

        "checkpoint":
            CHECKPOINT_JSONL,
    },
}


with open(
    OUTPUT_SUMMARY_JSON,
    "w",
    encoding="utf-8",
) as output_file:

    json.dump(
        summary_payload,
        output_file,
        indent=2,
        ensure_ascii=False,
        default=str,
    )


# =============================================================================
# 35. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (
                11,
                8.5,
            )
            if landscape
            else (
                8.5,
                11,
            )
        )
    )


    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )


    axis.axis(
        "off"
    )


    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )


    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1


    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )


    pdf.savefig(
        figure,
        bbox_inches="tight",
    )


    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.4,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped_lines = []


    for line in lines:

        if not line:

            wrapped_lines.append(
                ""
            )

        else:

            wrapped_lines.extend(
                textwrap.wrap(
                    str(
                        line
                    ),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )


    chunks = [
        wrapped_lines[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(
                wrapped_lines
            ),
            lines_per_page,
        )
    ] or [[""]]


    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title


        if len(
            chunks
        ) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )


        figure = create_page(
            page_title
        )


        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.19,
        )


        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=28,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were available."
            ],
        )


        return


    printable = dataframe.copy()


    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[
                column
            ] = printable[
                column
            ].map(
                lambda value:
                    ""
                    if pd.isna(
                        value
                    )
                    else f"{value:.4f}"
            )


    printable.columns = [
        readable_label(
            column
        )
        for column in printable.columns
    ]


    for column in printable.columns:

        printable[
            column
        ] = printable[
            column
        ].map(
            lambda value:
                ""
                if pd.isna(
                    value
                )
                else str(
                    value
                )
        )


        printable[
            column
        ] = printable[
            column
        ].map(
            lambda value:
                "\n".join(
                    textwrap.wrap(
                        value,
                        width=column_wrap_width,
                        break_long_words=False,
                        break_on_hyphens=False,
                    )
                )
                if value
                else ""
        )


    for start in range(
        0,
        len(
            printable
        ),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start
            + rows_per_page
        ]


        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )


        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )


        axis.axis(
            "off"
        )


        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )


        table.auto_set_font_size(
            False
        )


        table.set_fontsize(
            font_size
        )


        table.scale(
            1,
            1.38,
        )


        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )


            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )


                cell.set_text_props(
                    weight="bold"
                )


        save_page(
            figure
        )


def add_method_chart(
    metrics_df,
):

    if metrics_df.empty:

        return


    metric_columns = [
        "Exact accuracy",
        "Exact-or-related accuracy",
        "Balanced accuracy",
        "Macro F1",
        "Weighted F1",
        "Cohen's kappa",
    ]


    metric_columns = [
        column
        for column in metric_columns
        if column in metrics_df.columns
    ]


    figure = plt.figure(
        figsize=(
            10.5,
            7,
        )
    )


    axis = figure.add_subplot(
        111
    )


    x_positions = np.arange(
        len(
            metric_columns
        )
    )


    width = 0.24


    colors = [
        "#2E6F9E",
        "#B46A3C",
        "#4E8B57",
    ]


    for index, (
        _,
        row,
    ) in enumerate(
        metrics_df.iterrows()
    ):

        values = [
            row.get(
                metric,
                np.nan,
            )
            for metric in metric_columns
        ]


        axis.bar(
            x_positions
            + (
                index
                - (
                    len(
                        metrics_df
                    )
                    - 1
                )
                / 2
            )
            * width,
            values,
            width=width,
            label=row[
                "Model"
            ],
            color=colors[
                index
                % len(
                    colors
                )
            ],
        )


    axis.set_xticks(
        x_positions
    )


    axis.set_xticklabels(
        metric_columns,
        rotation=22,
        ha="right",
    )


    axis.set_ylabel(
        "Metric value"
    )


    axis.set_ylim(
        -0.05,
        1.05,
    )


    axis.set_title(
        "Primary-Cause Performance against Human Review",
        fontsize=13,
        fontweight="bold",
    )


    axis.grid(
        axis="y",
        alpha=0.25,
    )


    axis.legend(
        fontsize=8,
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_confusion_heatmap(
    title,
    matrix,
    labels,
):

    if matrix.size == 0:

        return


    figure = plt.figure(
        figsize=(
            11,
            8.5,
        )
    )


    axis = figure.add_subplot(
        111
    )


    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )


    axis.set_xticks(
        range(
            len(
                labels
            )
        )
    )


    axis.set_xticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        rotation=45,
        ha="right",
        fontsize=6.3,
    )


    axis.set_yticks(
        range(
            len(
                labels
            )
        )
    )


    axis.set_yticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        fontsize=6.3,
    )


    axis.set_xlabel(
        "Predicted primary cause"
    )


    axis.set_ylabel(
        "Human primary cause"
    )


    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )


    maximum_value = (
        matrix.max()
        if matrix.size
        else 0
    )


    for row_number in range(
        matrix.shape[
            0
        ]
    ):

        for column_number in range(
            matrix.shape[
                1
            ]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )


            if value > 0:

                color = (
                    "white"
                    if (
                        maximum_value
                        and value
                        > maximum_value
                        * 0.55
                    )
                    else "black"
                )


                axis.text(
                    column_number,
                    row_number,
                    str(
                        value
                    ),
                    ha="center",
                    va="center",
                    fontsize=6.2,
                    color=color,
                )


    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_scatter_plot(
    dataframe,
):

    if (
        dataframe.empty
        or "Top_score"
        not in dataframe.columns
    ):

        return


    plot_df = dataframe[
        dataframe[
            "Top_score"
        ]
        .notna()
    ].copy()


    if plot_df.empty:

        return


    figure = plt.figure(
        figsize=(
            10,
            6.5,
        )
    )


    axis = figure.add_subplot(
        111
    )


    colors = np.where(
        plot_df[
            "RAG_EXACT_MATCH"
        ],
        "#4E8B57",
        "#C44E52",
    )


    axis.scatter(
        plot_df[
            "Top_score"
        ],
        plot_df[
            "cause_confidence"
        ],
        c=colors,
        alpha=0.75,
        edgecolor="black",
        linewidth=0.3,
    )


    axis.set_xlabel(
        "Highest retrieved-example hybrid score"
    )


    axis.set_ylabel(
        "RAG-Qwen cause confidence"
    )


    axis.set_title(
        "Retrieval Similarity, RAG Confidence, and Exact Agreement",
        fontsize=13,
        fontweight="bold",
    )


    axis.grid(
        alpha=0.22,
    )


    axis.text(
        0.02,
        0.98,
        "Green = exact human match\nRed = disagreement",
        transform=axis.transAxes,
        va="top",
        fontsize=8,
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=15,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )


        return


    lines = [
        explanation,
        "",
    ]


    for example_number, (
        _,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 84,
                (
                    f"Example {example_number} | "
                    f"Record {row.get('SOURCE_RECORD_ID', '')} | "
                    f"{row.get('SEVERITY_GROUP', '')} | "
                    f"{row.get('TIME_PERIOD', '')}"
                ),
                (
                    f"Human cause: "
                    f"{row.get('HUMAN_PRIMARY_CAUSE', '')}"
                ),
                (
                    f"Original Qwen: "
                    f"{row.get('ORIGINAL_QWEN_CAUSE', '')}"
                ),
                (
                    f"RAG-Qwen: "
                    f"{row.get('RAG_PRIMARY_CAUSE', '')}"
                ),
                (
                    f"RAG alternative: "
                    f"{row.get('RAG_ALTERNATIVE_CAUSE', '')}"
                ),
                (
                    f"RAG confidence: "
                    f"{safe_float(row.get('cause_confidence', np.nan)):.3f}"
                ),
                (
                    f"Top retrieved cause: "
                    f"{row.get('TOP_RETRIEVED_CAUSE', '')}"
                ),
                (
                    f"Top retrieval score: "
                    f"{safe_float(row.get('TOP_RETRIEVAL_SCORE', np.nan)):.3f}"
                ),
                (
                    f"Retrieved causes: "
                    f"{row.get('RETRIEVED_CAUSES', '')}"
                ),
                (
                    f"Human uncertain: "
                    f"{row.get('HUMAN_CAUSE_UNCERTAIN', False)}"
                ),
                "",
                "Target narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "RAG evidence from target: "
                    f"{row.get('cause_evidence_text', '')}"
                ),
                (
                    "Retrieval explanation: "
                    f"{row.get('retrieval_explanation', '')}"
                ),
                (
                    "Human comments: "
                    f"{row.get('HUMAN_REVIEW_COMMENTS', '')}"
                ),
                "",
            ]
        )


    add_text_pages(
        title,
        lines,
        font_size=7.5,
        lines_per_page=47,
        wrap_width=104,
    )


# =============================================================================
# 36. GENERATE DETAILED PDF REPORT
# =============================================================================

# Cover page.
figure = create_page()


figure.text(
    0.5,
    0.79,
    "Hybrid RAG Primary-Cause Classification",
    ha="center",
    fontsize=20,
    fontweight="bold",
)


figure.text(
    0.5,
    0.735,
    "Comparison with Narrative-Only Qwen, Logistic Regression, and Human Review",
    ha="center",
    fontsize=12.5,
    color="#444444",
)


figure.text(
    0.5,
    0.69,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)


cover_lines = [
    f"Human benchmark narratives: {len(results_df):,}",
    (
        "Human-reviewed retrieval references: "
        f"{int((reference_library_df['REFERENCE_SOURCE'] == 'HUMAN_REVIEWED').sum()):,}"
    ),
    (
        "High-confidence Qwen retrieval references: "
        f"{int((reference_library_df['REFERENCE_SOURCE'] == 'QWEN_HIGH_CONFIDENCE').sum()):,}"
    ),
    f"Maximum retrieved examples per target: {TOP_K_RETRIEVAL}",
    f"Semantic retrieval weight: {SEMANTIC_WEIGHT:.2f}",
    f"TF-IDF retrieval weight: {LEXICAL_WEIGHT:.2f}",
    f"Generator: {MODEL_ID}",
    f"Embedding model: {EMBEDDING_MODEL_ID}",
    "",
    "Objective:",
    "Test whether similar reviewed and high-confidence historical narratives",
    "improve primary-cause classification beyond narrative-only Qwen.",
    "",
    "Important limitation:",
    "Human benchmark records are used through leave-one-out retrieval.",
    "The target record is excluded from its own retrieved examples.",
]


figure.text(
    0.09,
    0.60,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=8.8,
    family="monospace",
    linespacing=1.22,
)


save_page(
    figure
)


add_text_pages(
    "Executive Summary",
    [
        "This experiment evaluates retrieval-augmented generation for primary",
        "accident-cause classification.",
        "",
        "For each independently reviewed target narrative, the retriever selected",
        "similar historical examples from a hybrid reference library containing",
        "human-reviewed records and high-confidence Code Block 2 results.",
        "",
        "The target record was excluded from its own retrieval results. Identical",
        "target text was also removed before ranking references.",
        "",
        "Semantic embedding similarity and TF-IDF lexical similarity were",
        "combined into a hybrid retrieval score.",
        "",
        "The retrieved examples were supplied to the same local Qwen model used",
        "in the earlier ontology-guided analysis. The prompt required evidence",
        "to come from the target narrative and prohibited transferring facts from",
        "retrieved incidents.",
        "",
        "RAG results are compared with narrative-only Qwen, the independent human",
        "review, and the earlier logistic-regression baseline when its saved",
        "output is available.",
        "",
        "Exact agreement is the primary measure. Exact-or-related and top-two",
        "results are reported as sensitivity measures.",
    ],
)


add_text_pages(
    "RAG Experimental Design",
    [
        "1. Read the completed 100-record human-review workbook.",
        "",
        "2. Read high-confidence, cause-valid Code Block 2 results.",
        "",
        "3. Build one combined retrieval library with source-quality labels.",
        "",
        "4. Encode references and benchmark narratives using a local",
        "   Sentence-Transformers embedding model.",
        "",
        "5. Build a TF-IDF unigram and bigram representation of the same text.",
        "",
        "6. Calculate a hybrid score using 70% semantic similarity and 30%",
        "   lexical similarity.",
        "",
        "7. Retrieve up to five diverse references while limiting duplicate",
        "   cause categories and human examples.",
        "",
        "8. Exclude the target identifier and identical target text.",
        "",
        "9. Provide the target and retrieved examples to Qwen.",
        "",
        "10. Require target-narrative evidence and one ontology cause.",
        "",
        "11. Compare RAG-Qwen, narrative-only Qwen, logistic regression, and",
        "    independent human labels.",
    ],
)


reference_distribution_df = (
    reference_library_df[
        "REFERENCE_SOURCE"
    ]
    .value_counts()
    .rename_axis(
        "Reference source"
    )
    .reset_index(
        name="Records"
    )
)


cause_reference_distribution_df = (
    reference_library_df
    .groupby(
        [
            "REFERENCE_SOURCE",
            "REFERENCE_CAUSE",
        ]
    )
    .size()
    .reset_index(
        name="Records"
    )
    .sort_values(
        [
            "REFERENCE_SOURCE",
            "Records",
        ],
        ascending=[
            True,
            False,
        ],
    )
)


add_table_pages(
    "Hybrid Reference-Library Sources",
    reference_distribution_df,
    rows_per_page=10,
    font_size=7.4,
    column_wrap_width=42,
)


add_table_pages(
    "Reference-Library Cause Distribution",
    cause_reference_distribution_df,
    rows_per_page=28,
    font_size=6.9,
    column_wrap_width=36,
    landscape=True,
)


add_table_pages(
    "Overall Method Comparison",
    overall_metrics_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=34,
    landscape=True,
)


add_method_chart(
    overall_metrics_df
)


add_table_pages(
    "Comparison for Human Labels Not Marked Uncertain",
    certain_metrics_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=34,
    landscape=True,
)


add_table_pages(
    "RAG Changes Relative to Narrative-Only Qwen",
    change_summary_df,
    rows_per_page=14,
    font_size=7.1,
    column_wrap_width=52,
)


add_text_pages(
    "How to Interpret the RAG Change Results",
    [
        "RAG can help, have no effect, or reduce performance.",
        "",
        "An improvement occurs when retrieval changes an incorrect original Qwen",
        "classification into the human-selected cause.",
        "",
        "A reduction occurs when retrieval changes an exact original Qwen match",
        "into a disagreement.",
        "",
        "The number of changes should therefore be considered together with the",
        "net change in exact accuracy, not treated as improvement by itself.",
        "",
        "Retrieved examples may be relevant while still describing a different",
        "event mechanism. Strong retrieval similarity does not guarantee that",
        "the retrieved cause applies to the target.",
        "",
        "The RAG evidence field is checked against the target narrative to detect",
        "whether the model copied unsupported evidence from a reference record.",
    ],
)


add_table_pages(
    "Category-Level Primary-Cause Performance",
    category_metrics_df,
    rows_per_page=30,
    font_size=6.5,
    column_wrap_width=28,
    landscape=True,
)


add_confusion_heatmap(
    "Hybrid RAG-Qwen versus Human Review",
    rag_confusion_matrix,
    evaluation_labels,
)


add_confusion_heatmap(
    "Narrative-Only Qwen versus Human Review",
    qwen_confusion_matrix,
    evaluation_labels,
)


if logistic_results_file:

    add_confusion_heatmap(
        "TF-IDF Logistic Regression versus Human Review",
        logistic_confusion_matrix,
        evaluation_labels,
    )


add_table_pages(
    "Performance by Injury Severity",
    severity_comparison_df,
    rows_per_page=12,
    font_size=6.7,
    column_wrap_width=30,
    landscape=True,
)


add_table_pages(
    "Performance by Time Period",
    time_comparison_df,
    rows_per_page=12,
    font_size=6.7,
    column_wrap_width=30,
    landscape=True,
)


if not retrieval_band_performance_df.empty:

    add_table_pages(
        "Performance by Retrieval-Similarity Band",
        retrieval_band_performance_df,
        rows_per_page=12,
        font_size=6.9,
        column_wrap_width=35,
        landscape=True,
    )


    add_scatter_plot(
        results_df
    )


if not retrieval_quality_by_target_df.empty:

    retrieval_summary_df = pd.DataFrame(
        [
            {
                "Measure":
                    "Targets with at least one retrieved reference",

                "Value":
                    int(
                        retrieval_quality_by_target_df[
                            "Retrieved_records"
                        ]
                        .gt(
                            0
                        )
                        .sum()
                    ),
            },
            {
                "Measure":
                    "Mean top retrieval score",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Top_score"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean exact-cause retrieval rate",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Exact_retrieval_rate"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean exact-or-related retrieval rate",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Related_retrieval_rate"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean human-reviewed references per target",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Human_reference_count"
                        ].mean()
                    ),
            },
        ]
    )


    add_table_pages(
        "Retrieval-Quality Summary",
        retrieval_summary_df,
        rows_per_page=12,
        font_size=7.2,
        column_wrap_width=48,
    )


improved_examples_df = results_df[
    results_df[
        "RAG_IMPROVED_EXACT"
    ]
].copy()


reduced_examples_df = results_df[
    results_df[
        "RAG_REDUCED_EXACT"
    ]
].copy()


rag_disagreement_df = results_df[
    ~results_df[
        "RAG_EXACT_MATCH"
    ]
].copy()


uncertain_examples_df = results_df[
    results_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


add_example_pages(
    "Examples Where Retrieval Improved the Exact Classification",
    improved_examples_df,
    (
        "The original narrative-only Qwen classification differed from the "
        "human label, while RAG-Qwen exactly matched it. These examples show "
        "where similar historical narratives may have clarified terminology "
        "or a category boundary."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Examples Where Retrieval Reduced Exact Performance",
    reduced_examples_df,
    (
        "The original Qwen classification exactly matched the human label, "
        "but RAG-Qwen changed the result. These examples are important because "
        "they show how relevant-looking retrieval can introduce anchoring or "
        "category-transfer errors."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Remaining RAG-Qwen Disagreements",
    rag_disagreement_df,
    (
        "These cases did not exactly match the human primary-cause label. "
        "Review should consider the alternative cause, retrieved examples, "
        "human uncertainty, and whether the categories are closely related."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Human-Uncertain RAG Cases",
    uncertain_examples_df,
    (
        "The human reviewer marked these primary-cause decisions uncertain. "
        "Differences are retained for transparency but should not automatically "
        "be interpreted as definitive RAG errors."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_text_pages(
    "Interpretation Framework",
    [
        "The central question is whether retrieval improves classification over",
        "the same Qwen model operating on the target narrative alone.",
        "",
        "The cleanest evidence of benefit is:",
        "",
        "1. Higher exact human agreement.",
        "2. Higher macro and weighted F1.",
        "3. Improvement in difficult cause categories.",
        "4. Traceable evidence copied from the target narrative.",
        "5. More improvements than reductions among changed predictions.",
        "",
        "Retrieval quality must be evaluated separately from final classification.",
        "A reference can be semantically similar while having a different cause.",
        "",
        "RAG should be considered useful only when retrieved examples improve the",
        "target classification without causing the model to transfer unsupported",
        "facts from historical cases.",
        "",
        "If retrieval improves only exact-or-related agreement but not exact",
        "agreement, it may be helping with broad mechanism families without",
        "consistently resolving the final ontology boundary.",
    ],
)


add_text_pages(
    "Limitations",
    [
        "1. The human-reviewed reference library is small and is used through",
        "   leave-one-out retrieval rather than as a fully separate training set.",
        "",
        "2. High-confidence Code Block 2 labels were generated by Qwen and are",
        "   not independently verified ground truth.",
        "",
        "3. Retrieval may reinforce errors or dominant categories already present",
        "   in the Code Block 2 reference library.",
        "",
        "4. Similar narratives can describe different immediate mechanisms.",
        "",
        "5. The hybrid retrieval weights and minimum score are starting settings",
        "   rather than optimized values.",
        "",
        "6. Sentence embeddings and TF-IDF represent similarity differently and",
        "   may retrieve examples related by equipment, body part, or activity",
        "   rather than primary cause.",
        "",
        "7. The experiment evaluates primary cause only. Retrieval should not be",
        "   assumed to solve missing information about contributing factors or",
        "   corrective actions.",
        "",
        "8. Human labels can be uncertain, and multiple related causes may be",
        "   defensible for some short narratives.",
        "",
        "9. Exact-or-related metrics depend on predefined cause relationships and",
        "   remain secondary to exact agreement.",
    ],
)


add_text_pages(
    "Potential Business Use",
    [
        "RAG may provide value beyond classification by retrieving similar",
        "historical incidents for safety review.",
        "",
        "A practical workflow could present:",
        "",
        "1. The proposed primary cause.",
        "2. The target-narrative evidence.",
        "3. The most similar reviewed historical incidents.",
        "4. The source and review quality of each retrieved example.",
        "5. An alternative primary cause when the category boundary is unclear.",
        "",
        "This could support investigation preparation, lessons-learned searches,",
        "and review of recurring accident mechanisms.",
        "",
        "The RAG output should remain decision support. Retrieved incidents do not",
        "prove that the target event had the same conditions or root causes.",
    ],
)


add_text_pages(
    "Technical Appendix",
    [
        f"Generator model: {MODEL_ID}",
        f"Embedding model: {EMBEDDING_MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"Benchmark records: {len(results_df):,}",
        f"Hybrid reference records: {len(reference_library_df):,}",
        f"Top-K retrieval: {TOP_K_RETRIEVAL}",
        f"Semantic weight: {SEMANTIC_WEIGHT:.2f}",
        f"Lexical weight: {LEXICAL_WEIGHT:.2f}",
        f"Minimum retrieval score: {MIN_RETRIEVAL_SCORE:.2f}",
        f"Minimum Code Block 2 confidence: "
        f"{MIN_BLOCK2_CAUSE_CONFIDENCE:.2f}",
        f"Maximum generated tokens: {MAX_NEW_TOKENS}",
        f"Maximum validation attempts: {MAX_RETRIES}",
        "",
        f"PDF: {os.path.abspath(OUTPUT_PDF)}",
        f"Results CSV: {os.path.abspath(OUTPUT_RESULTS_CSV)}",
        f"Retrieval details: {os.path.abspath(OUTPUT_RETRIEVAL_CSV)}",
        f"Checkpoint: {os.path.abspath(CHECKPOINT_JSONL)}",
        f"Summary JSON: {os.path.abspath(OUTPUT_SUMMARY_JSON)}",
    ],
    font_size=8.2,
)


# =============================================================================
# 37. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()


metadata[
    "Title"
] = (
    "MSHA Hybrid RAG Primary-Cause Classification Comparison Report"
)


metadata[
    "Author"
] = (
    "Python, Sentence-Transformers, and Qwen2.5-7B-Instruct"
)


metadata[
    "Subject"
] = (
    "Comparison of hybrid retrieval-augmented generation, narrative-only "
    "Qwen, logistic regression, and independent human primary-cause labels"
)


metadata[
    "Keywords"
] = (
    "MSHA, retrieval-augmented generation, RAG, Qwen, TF-IDF, "
    "sentence embeddings, accident cause, human validation"
)


metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 38. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("HYBRID RAG PRIMARY-CAUSE ANALYSIS COMPLETE")
print("=" * 95)


print(
    f"Benchmark records: "
    f"{len(results_df):,}"
)


print(
    f"Reference-library records: "
    f"{len(reference_library_df):,}"
)


print(
    f"Successful RAG classifications: "
    f"{int(results_df['RAG_PROCESSING_STATUS'].eq('SUCCESS').sum()):,}"
)


print(
    f"RAG exact human matches: "
    f"{int(results_df['RAG_EXACT_MATCH'].sum()):,}"
)


print(
    f"RAG exact-or-related human matches: "
    f"{int(results_df['RAG_EXACT_OR_RELATED'].sum()):,}"
)


print(
    f"RAG improvements over original Qwen: "
    f"{int(results_df['RAG_IMPROVED_EXACT'].sum()):,}"
)


print(
    f"RAG reductions versus original Qwen: "
    f"{int(results_df['RAG_REDUCED_EXACT'].sum()):,}"
)


print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)


print(
    f"Results CSV: "
    f"{os.path.abspath(OUTPUT_RESULTS_CSV)}"
)


print(
    f"Retrieval details: "
    f"{os.path.abspath(OUTPUT_RETRIEVAL_CSV)}"
)


print(
    f"Checkpoint: "
    f"{os.path.abspath(CHECKPOINT_JSONL)}"
)


print("=" * 95)


# =============================================================================
# 39. NOTEBOOK DISPLAYS
# =============================================================================

display(
    overall_metrics_df
)


display(
    certain_metrics_df
)


display(
    change_summary_df
)


display(
    category_metrics_df
)


display(
    severity_comparison_df
)


if not retrieval_band_performance_df.empty:

    display(
        retrieval_band_performance_df
    )
````

## Also testing RAG Retrieval only option

````python
# =============================================================================
# STANDALONE HYBRID RAG PRIMARY-CAUSE CLASSIFICATION EXPERIMENT
#
# PURPOSE
# -------
# Test whether retrieval of similar, labeled MSHA accident narratives improves
# primary-cause classification compared with:
#
#   1. Narrative-only Qwen classification.
#   2. TF-IDF multiclass logistic regression, when its saved output is present.
#   3. Independent human primary-cause classifications.
#   4. Retrieval-only classification, using a similarity-weighted vote of the
#      same retrieved examples supplied to RAG-Qwen, with no generator.
#
# RAG REFERENCE LIBRARY
# ---------------------
# The retrieval library combines:
#
#   A. Human-reviewed Block 3 narratives, used through leave-one-out retrieval.
#      The target record is always excluded from its own retrieved examples.
#
#   B. High-confidence, cause-valid Code Block 2 narratives.
#
# RETRIEVAL METHOD
# ----------------
# Hybrid retrieval combines:
#
#   1. Semantic similarity from Sentence-Transformers embeddings.
#   2. TF-IDF cosine similarity.
#
# GENERATOR
# ---------
# Qwen/Qwen2.5-7B-Instruct, run locally with four-bit quantization.
#
# RETRIEVAL-ONLY COMPARISON
# -------------------------
# The retrieval-only option uses the exact same retrieved examples as RAG-Qwen.
# Each retrieved example votes for its accepted cause, weighted by hybrid
# similarity multiplied by reference confidence. The highest-weighted cause is
# the retrieval-only prediction. This isolates how much of the RAG result comes
# from retrieval alone versus Qwen reasoning over the target narrative.
#
# IMPORTANT CONTROL
# -----------------
# Retrieved examples are references for interpreting terminology and ontology
# boundaries. The model is instructed not to transfer facts from a retrieved
# narrative to the target narrative.
#
# REQUIRED INPUTS
# ---------------
# 1. A pandas DataFrame named `accidents` is already loaded.
#
# 2. Completed human review workbook:
#
#       MSHA_Block3_Human_Review_Workbook- Completed.xlsx
#
# 3. One Code Block 2 structured result file:
#
#       MSHA_Block2_Expanded_Results.csv
#       MSHA_Block2_Revised_Results.csv
#       MSHA_Block2_Unseen_Results.csv
#
# OPTIONAL INPUT
# --------------
# Earlier logistic-regression benchmark output:
#
#       MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv
#
# OUTPUTS
# -------
# 1. MSHA_RAG_Primary_Cause_Comparison_Report.pdf
# 2. MSHA_RAG_Primary_Cause_Results.csv
# 3. MSHA_RAG_Primary_Cause_Checkpoint.jsonl
# 4. MSHA_RAG_Retrieval_Details.csv
# 5. MSHA_RAG_Primary_Cause_Summary.json
#
# INSTALLATION NOTE
# -----------------
# The environment must contain:
#
#   sentence-transformers
#   transformers
#   accelerate
#   bitsandbytes
#   scikit-learn
#   openpyxl
#
# The Sentence-Transformers embedding model and Qwen model must be cached
# locally or downloadable from Hugging Face when first used.
# =============================================================================


# =============================================================================
# 1. IMPORTS
# =============================================================================

import os
import re
import gc
import sys
import json
import math
import time
import textwrap
import warnings
from pathlib import Path
from datetime import datetime

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import transformers

from pydantic import BaseModel, Field, model_validator

from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM,
    BitsAndBytesConfig,
)

from sentence_transformers import SentenceTransformer

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_recall_fscore_support,
    f1_score,
    cohen_kappa_score,
    confusion_matrix,
)

from matplotlib.backends.backend_pdf import PdfPages


warnings.filterwarnings("ignore")


# =============================================================================
# 2. SETTINGS
# =============================================================================

MODEL_ID = "Qwen/Qwen2.5-7B-Instruct"

EMBEDDING_MODEL_ID = "sentence-transformers/all-MiniLM-L6-v2"

PROMPT_VERSION = "MSHA_RAG_PRIMARY_CAUSE_V1.0"

HUMAN_REVIEW_WORKBOOK = (
    "MSHA_Block3_Human_Review_Workbook- Completed.xlsx"
)

BLOCK2_RESULT_CANDIDATES = [
    "MSHA_Block2_Expanded_Results.csv",
    "MSHA_Block2_Revised_Results.csv",
    "MSHA_Block2_Unseen_Results.csv",
]

LOGISTIC_RESULT_CANDIDATES = [
    "MSHA_TFIDF_Logistic_Regression_Human_Benchmark.csv",
]

OUTPUT_PDF = (
    "MSHA_RAG_Primary_Cause_Comparison_Report.pdf"
)

OUTPUT_RESULTS_CSV = (
    "MSHA_RAG_Primary_Cause_Results.csv"
)

OUTPUT_RETRIEVAL_CSV = (
    "MSHA_RAG_Retrieval_Details.csv"
)

CHECKPOINT_JSONL = (
    "MSHA_RAG_Primary_Cause_Checkpoint.jsonl"
)

OUTPUT_SUMMARY_JSON = (
    "MSHA_RAG_Primary_Cause_Summary.json"
)


RANDOM_STATE = 42

REPROCESS_COMPLETED = False

MAX_RETRIES = 4

MAX_NEW_TOKENS = 350

TOP_K_RETRIEVAL = 5

MAX_HUMAN_EXAMPLES_PER_QUERY = 2

MAX_EXAMPLES_PER_CAUSE_PER_QUERY = 2


SEMANTIC_WEIGHT = 0.70

LEXICAL_WEIGHT = 0.30


MIN_RETRIEVAL_SCORE = 0.15

MIN_BLOCK2_CAUSE_CONFIDENCE = 0.80

ALLOWED_BLOCK2_SUPPORT = {
    "HIGH",
    "MEDIUM",
}


TFIDF_MAX_FEATURES = 30000

TFIDF_NGRAM_RANGE = (
    1,
    2,
)

TFIDF_MIN_DF = 1

TFIDF_MAX_DF = 0.98


TOP_DISAGREEMENT_EXAMPLES = 15

TOP_RETRIEVAL_EXAMPLES = 15


# -----------------------------------------------------------------------------
# RETRIEVAL-ONLY COMPARISON SETTINGS
#
# The retrieval-only method uses the same retrieved examples as RAG-Qwen.
# Each example votes for its accepted cause using:
#
#   vote weight = hybrid score x reference confidence
#
# When no example passes the minimum retrieval score, the fallback cause is
# assigned so the record remains in the comparison.
# -----------------------------------------------------------------------------

RETRIEVAL_ONLY_METHOD_NAME = "Retrieval-only weighted vote"

RETRIEVAL_ONLY_USE_REFERENCE_CONFIDENCE = True

RETRIEVAL_ONLY_FALLBACK_CAUSE = "INSUFFICIENT_INFORMATION"


# =============================================================================
# 3. FROZEN PRIMARY-CAUSE ONTOLOGY
# =============================================================================

CAUSE_CATEGORIES = [
    "SLIP_TRIP_OR_FALL",
    "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    "STRUCK_BY_MOVING_OBJECT",
    "STRUCK_AGAINST_STATIONARY_OBJECT",
    "FOREIGN_BODY_OR_PARTICLE_CONTACT",
    "CAUGHT_IN_OR_BETWEEN",
    "MATERIAL_RELEASE_OR_SHIFT",
    "GROUND_OR_ROOF_FAILURE",
    "UNEXPECTED_EQUIPMENT_MOVEMENT",
    "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
    "MANUAL_HANDLING_OR_OVEREXERTION",
    "PRESSURE_OR_STORED_ENERGY_RELEASE",
    "ELECTRICAL_CONTACT",
    "FIRE_EXPLOSION_OR_THERMAL_EVENT",
    "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    "CHEMICAL_DUST_OR_OCCUPATIONAL_EXPOSURE",
    "ERGONOMIC_OR_REPETITIVE_EXPOSURE",
    "OTHER",
    "INSUFFICIENT_INFORMATION",
]


CATEGORY_ALIASES = {
    "SLIP_TRIP_FALL":
        "SLIP_TRIP_OR_FALL",

    "CAUGHT_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "CAUGHT_IN_BETWEEN":
        "CAUGHT_IN_OR_BETWEEN",

    "STRUCK_BY_OBJECT":
        "STRUCK_BY_MOVING_OBJECT",

    "FOREIGN_OBJECT_CONTACT":
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",

    "LOSS_OF_CONTROL":
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",

    "UNEXPECTED_MACHINE_MOVEMENT":
        "UNEXPECTED_EQUIPMENT_MOVEMENT",

    "HEAT_EXPOSURE":
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",

    "MANUAL_HANDLING":
        "MANUAL_HANDLING_OR_OVEREXERTION",

    "GROUND_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "ROOF_FAILURE":
        "GROUND_OR_ROOF_FAILURE",

    "MOBILE_EQUIPMENT_EVENT":
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",

    "STORED_ENERGY_RELEASE":
        "PRESSURE_OR_STORED_ENERGY_RELEASE",

    "FIRE_OR_EXPLOSION":
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",

    "INSUFFICIENT_INFO":
        "INSUFFICIENT_INFORMATION",
}


# =============================================================================
# 4. RELATED-CAUSE GROUPS
#
# Exact agreement remains the primary metric.
# Exact-or-related agreement is a secondary sensitivity measure.
# =============================================================================

RELATED_CAUSE_GROUPS = [
    {
        "CAUGHT_IN_OR_BETWEEN",
        "STRUCK_BY_MOVING_OBJECT",
        "MATERIAL_RELEASE_OR_SHIFT",
    },

    {
        "LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
        "VEHICLE_OR_MOBILE_EQUIPMENT_EVENT",
    },

    {
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
        "MANUAL_HANDLING_OR_OVEREXERTION",
    },

    {
        "GROUND_OR_ROOF_FAILURE",
        "MATERIAL_RELEASE_OR_SHIFT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "PRESSURE_OR_STORED_ENERGY_RELEASE",
        "UNEXPECTED_EQUIPMENT_MOVEMENT",
    },

    {
        "FIRE_EXPLOSION_OR_THERMAL_EVENT",
        "HEAT_OR_ENVIRONMENTAL_EXPOSURE",
    },

    {
        "FOREIGN_BODY_OR_PARTICLE_CONTACT",
        "STRUCK_BY_MOVING_OBJECT",
    },

    {
        "SLIP_TRIP_OR_FALL",
        "AWKWARD_MOVEMENT_OR_BODY_MOTION",
    },
]


# =============================================================================
# 5. RAG RESPONSE SCHEMA
# =============================================================================

class RAGCauseExtraction(BaseModel):

    primary_cause: str

    cause_evidence_text: str = ""

    cause_confidence: float = Field(
        default=0.0,
        ge=0.0,
        le=1.0,
    )

    cause_support_level: str = "LOW"

    retrieval_used: bool = True

    retrieval_explanation: str = ""

    alternative_cause: str = ""

    reviewer_note: str = ""

    @model_validator(mode="after")
    def validate_output(self):

        if self.primary_cause not in CAUSE_CATEGORIES:

            raise ValueError(
                f"Invalid primary cause: {self.primary_cause}"
            )

        if (
            self.alternative_cause
            and self.alternative_cause
            not in CAUSE_CATEGORIES
        ):

            raise ValueError(
                f"Invalid alternative cause: "
                f"{self.alternative_cause}"
            )

        if self.cause_support_level not in {
            "HIGH",
            "MEDIUM",
            "LOW",
        }:

            raise ValueError(
                "cause_support_level must be HIGH, MEDIUM, or LOW."
            )

        return self


# =============================================================================
# 6. INPUT VALIDATION
# =============================================================================

if "accidents" not in globals():

    raise NameError(
        "A pandas DataFrame named 'accidents' must already be loaded."
    )


if not isinstance(
    accidents,
    pd.DataFrame,
):

    raise TypeError(
        "'accidents' exists but is not a pandas DataFrame."
    )


if accidents.empty:

    raise ValueError(
        "The accidents DataFrame is empty."
    )


if not Path(
    HUMAN_REVIEW_WORKBOOK
).exists():

    raise FileNotFoundError(
        f"Required completed workbook not found: "
        f"{HUMAN_REVIEW_WORKBOOK}"
    )


block2_results_file = next(
    (
        path
        for path in BLOCK2_RESULT_CANDIDATES
        if Path(
            path
        ).exists()
    ),
    None,
)


if block2_results_file is None:

    raise FileNotFoundError(
        "No Code Block 2 results file was found. "
        "Expected one of: "
        + ", ".join(
            BLOCK2_RESULT_CANDIDATES
        )
    )


logistic_results_file = next(
    (
        path
        for path in LOGISTIC_RESULT_CANDIDATES
        if Path(
            path
        ).exists()
    ),
    None,
)


print("=" * 95)
print("HYBRID RAG PRIMARY-CAUSE CLASSIFICATION")
print("=" * 95)

print(
    "Code Block 2 results:",
    block2_results_file,
)

print(
    "Human-review workbook:",
    HUMAN_REVIEW_WORKBOOK,
)

print(
    "Logistic-regression comparison:",
    logistic_results_file
    if logistic_results_file
    else "Not found; logistic comparison will be omitted.",
)


# =============================================================================
# 7. GENERAL UTILITIES
# =============================================================================

def normalize_column_name(
    value,
):

    return re.sub(
        r"[^A-Z0-9]+",
        "_",
        str(
            value
        )
        .strip()
        .upper(),
    ).strip(
        "_"
    )


def normalize_category(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return CATEGORY_ALIASES.get(
        normalized,
        normalized,
    )


def readable_label(
    value,
):

    return (
        str(
            value
        )
        .replace(
            "_",
            " ",
        )
        .title()
    )


def safe_float(
    value,
    default=np.nan,
):

    try:

        output = float(
            value
        )

        if np.isnan(
            output
        ):

            return default

        return output

    except Exception:

        return default


def first_existing_column(
    dataframe,
    candidates,
    required=False,
    description="column",
):

    lookup = {
        normalize_column_name(
            column
        ):
            column
        for column in dataframe.columns
    }

    for candidate in candidates:

        normalized_candidate = normalize_column_name(
            candidate
        )

        if normalized_candidate in lookup:

            return lookup[
                normalized_candidate
            ]

    if required:

        raise KeyError(
            f"Required {description} was not found. "
            f"Checked {candidates}. "
            f"Available columns: {list(dataframe.columns)}"
        )

    return None


def choose_sheet(
    sheet_names,
    preferred_names,
):

    for preferred in preferred_names:

        preferred_normalized = normalize_column_name(
            preferred
        )

        for sheet_name in sheet_names:

            if (
                preferred_normalized
                in normalize_column_name(
                    sheet_name
                )
            ):

                return sheet_name

    return None


def normalize_boolean(
    value,
):

    normalized = normalize_column_name(
        value
    )

    return normalized in {
        "TRUE",
        "YES",
        "Y",
        "1",
    }


def is_uncertain(
    value,
):

    return normalize_column_name(
        value
    ) in {
        "UNCERTAIN",
        "UNSURE",
        "UNKNOWN",
        "NOT_SURE",
        "CANNOT_DETERMINE",
        "CANT_DETERMINE",
    }


def categories_related(
    human_value,
    prediction,
):

    human_value = normalize_category(
        human_value
    )

    prediction = normalize_category(
        prediction
    )

    if (
        not human_value
        or not prediction
    ):

        return False

    if human_value == prediction:

        return True

    return any(
        (
            human_value in group
            and prediction in group
        )
        for group in RELATED_CAUSE_GROUPS
    )


def normalize_evidence_text(
    value,
):

    text = str(
        value
    ).lower()

    text = (
        text
        .replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    text = re.sub(
        r"[^a-z0-9\s']",
        " ",
        text,
    )

    text = re.sub(
        r"\s+",
        " ",
        text,
    ).strip()

    return text


def evaluate_evidence(
    evidence,
    narrative,
):

    evidence_text = str(
        evidence
        or ""
    ).strip()

    if not evidence_text:

        return {
            "valid":
                False,

            "exact":
                False,

            "normalized":
                False,

            "token_overlap":
                0.0,
        }

    exact = (
        evidence_text.lower()
        in str(
            narrative
        ).lower()
    )

    normalized_evidence = normalize_evidence_text(
        evidence
    )

    normalized_narrative = normalize_evidence_text(
        narrative
    )

    normalized_match = (
        bool(
            normalized_evidence
        )
        and normalized_evidence
        in normalized_narrative
    )

    evidence_tokens = set(
        normalized_evidence.split()
    )

    narrative_tokens = set(
        normalized_narrative.split()
    )

    overlap = (
        len(
            evidence_tokens.intersection(
                narrative_tokens
            )
        )
        / len(
            evidence_tokens
        )
        if evidence_tokens
        else 0.0
    )

    valid = (
        exact
        or normalized_match
        or overlap >= 0.85
    )

    return {
        "valid":
            valid,

        "exact":
            exact,

        "normalized":
            normalized_match,

        "token_overlap":
            round(
                overlap,
                4,
            ),
    }


# =============================================================================
# 8. JSON PARSING UTILITIES
# =============================================================================

def strip_markdown_fences(
    text,
):

    text = str(
        text
    ).strip()

    text = re.sub(
        r"^```(?:json)?",
        "",
        text,
        flags=re.IGNORECASE,
    )

    text = re.sub(
        r"```$",
        "",
        text,
    )

    return text.strip()


def extract_balanced_json(
    text,
):

    text = strip_markdown_fences(
        text
    )

    start = text.find(
        "{"
    )

    if start < 0:

        raise ValueError(
            "No opening JSON brace was found."
        )

    depth = 0

    in_string = False

    escaped = False

    for index in range(
        start,
        len(
            text
        ),
    ):

        character = text[
            index
        ]

        if escaped:

            escaped = False

            continue

        if (
            character == "\\"
            and in_string
        ):

            escaped = True

            continue

        if character == '"':

            in_string = not in_string

            continue

        if in_string:

            continue

        if character == "{":

            depth += 1

        elif character == "}":

            depth -= 1

            if depth == 0:

                return text[
                    start:
                    index + 1
                ]

    candidate = text[
        start:
    ].strip()

    missing_braces = (
        candidate.count(
            "{"
        )
        - candidate.count(
            "}"
        )
    )

    if missing_braces > 0:

        candidate += (
            "}"
            * missing_braces
        )

        return candidate

    raise ValueError(
        "A complete JSON object was not found."
    )


def repair_json(
    value,
):

    output = str(
        value
    ).strip()

    output = (
        output
        .replace(
            "“",
            '"',
        )
        .replace(
            "”",
            '"',
        )
        .replace(
            "’",
            "'",
        )
    )

    output = re.sub(
        r",\s*}",
        "}",
        output,
    )

    output = re.sub(
        r":\s*None\b",
        ': ""',
        output,
    )

    output = re.sub(
        r":\s*True\b",
        ": true",
        output,
    )

    output = re.sub(
        r":\s*False\b",
        ": false",
        output,
    )

    return output


# =============================================================================
# 9. LOAD HUMAN-REVIEW WORKBOOK
# =============================================================================

excel_file = pd.ExcelFile(
    HUMAN_REVIEW_WORKBOOK,
    engine="openpyxl",
)


human_sheet = choose_sheet(
    excel_file.sheet_names,
    [
        "Human Review",
        "Review",
    ],
)


model_sheet = choose_sheet(
    excel_file.sheet_names,
    [
        "Model Output",
        "Model",
    ],
)


if human_sheet is None:

    raise KeyError(
        "The Human Review worksheet could not be identified."
    )


if model_sheet is None:

    raise KeyError(
        "The Model Output worksheet could not be identified."
    )


human_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=human_sheet,
    engine="openpyxl",
)


original_qwen_df = pd.read_excel(
    HUMAN_REVIEW_WORKBOOK,
    sheet_name=model_sheet,
    engine="openpyxl",
)


human_df.columns = [
    normalize_column_name(
        column
    )
    for column in human_df.columns
]


original_qwen_df.columns = [
    normalize_column_name(
        column
    )
    for column in original_qwen_df.columns
]


human_id_column = first_existing_column(
    human_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="human-review record ID",
)


human_text_column = first_existing_column(
    human_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
)


human_cause_column = first_existing_column(
    human_df,
    [
        "HUMAN_PRIMARY_CAUSE",
        "HUMAN_CAUSE",
        "PRIMARY_CAUSE_HUMAN",
    ],
    required=True,
    description="human primary-cause field",
)


human_cause_evidence_column = first_existing_column(
    human_df,
    [
        "HUMAN_CAUSE_EVIDENCE",
        "HUMAN_CAUSE_EVIDENCE_TEXT",
    ],
)


human_uncertainty_column = first_existing_column(
    human_df,
    [
        "HUMAN_CAUSE_UNCERTAIN",
        "CAUSE_UNCERTAIN",
        "HUMAN_CAUSE_EVIDENCE_SUPPORTED",
    ],
)


human_comment_column = first_existing_column(
    human_df,
    [
        "HUMAN_REVIEW_COMMENTS",
        "REVIEW_COMMENTS",
        "COMMENTS",
    ],
)


human_severity_column = first_existing_column(
    human_df,
    [
        "SEVERITY_GROUP",
        "INJURY_SEVERITY_GROUP",
    ],
)


human_time_column = first_existing_column(
    human_df,
    [
        "TIME_PERIOD",
        "ACCIDENT_TIME_PERIOD",
    ],
)


human_completion_column = first_existing_column(
    human_df,
    [
        "REVIEW_COMPLETED",
        "HUMAN_REVIEW_COMPLETED",
        "COMPLETED",
    ],
)


qwen_id_column = first_existing_column(
    original_qwen_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="original Qwen record ID",
)


qwen_cause_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_PRIMARY_CAUSE",
        "PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "MODEL_CAUSE",
    ],
    required=True,
    description="original Qwen primary cause",
)


qwen_confidence_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_CAUSE_CONFIDENCE",
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
    ],
)


qwen_evidence_column = first_existing_column(
    original_qwen_df,
    [
        "MODEL_CAUSE_EVIDENCE",
        "CAUSE_EVIDENCE_TEXT",
        "ORIGINAL_MODEL_CAUSE_EVIDENCE",
    ],
)


human_df[
    "SOURCE_RECORD_ID"
] = human_df[
    human_id_column
].astype(
    str
)


human_df[
    "HUMAN_PRIMARY_CAUSE"
] = human_df[
    human_cause_column
].map(
    normalize_category
)


if human_text_column:

    human_df[
        "NARRATIVE"
    ] = (
        human_df[
            human_text_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
        .str.strip()
    )

else:

    human_df[
        "NARRATIVE"
    ] = ""


human_df[
    "HUMAN_CAUSE_UNCERTAIN"
] = human_df[
    "HUMAN_PRIMARY_CAUSE"
].map(
    is_uncertain
)


if human_uncertainty_column:

    uncertainty_values = human_df[
        human_uncertainty_column
    ].map(
        normalize_column_name
    )

    human_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ] |= uncertainty_values.isin(
        {
            "UNCERTAIN",
            "UNSURE",
            "UNKNOWN",
            "NOT_SURE",
        }
    )


if human_cause_evidence_column:

    human_df[
        "HUMAN_CAUSE_EVIDENCE"
    ] = (
        human_df[
            human_cause_evidence_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
    )

else:

    human_df[
        "HUMAN_CAUSE_EVIDENCE"
    ] = ""


if human_comment_column:

    human_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = (
        human_df[
            human_comment_column
        ]
        .astype(
            "string"
        )
        .fillna(
            ""
        )
    )

else:

    human_df[
        "HUMAN_REVIEW_COMMENTS"
    ] = ""


human_df[
    "SEVERITY_GROUP"
] = (
    human_df[
        human_severity_column
    ]
    if human_severity_column
    else "Unknown"
)


human_df[
    "TIME_PERIOD"
] = (
    human_df[
        human_time_column
    ]
    if human_time_column
    else "Unknown"
)


if human_completion_column:

    completion_mask = human_df[
        human_completion_column
    ].map(
        normalize_column_name
    ).isin(
        {
            "YES",
            "Y",
            "TRUE",
            "1",
            "COMPLETE",
            "COMPLETED",
        }
    )

    human_df = human_df[
        completion_mask
    ].copy()


original_qwen_merge_df = pd.DataFrame(
    {
        "SOURCE_RECORD_ID":
            original_qwen_df[
                qwen_id_column
            ].astype(
                str
            ),

        "ORIGINAL_QWEN_CAUSE":
            original_qwen_df[
                qwen_cause_column
            ].map(
                normalize_category
            ),
    }
)


original_qwen_merge_df[
    "ORIGINAL_QWEN_CONFIDENCE"
] = (
    pd.to_numeric(
        original_qwen_df[
            qwen_confidence_column
        ],
        errors="coerce",
    )
    if qwen_confidence_column
    else np.nan
)


original_qwen_merge_df[
    "ORIGINAL_QWEN_EVIDENCE"
] = (
    original_qwen_df[
        qwen_evidence_column
    ].astype(
        "string"
    ).fillna(
        ""
    )
    if qwen_evidence_column
    else ""
)


original_qwen_merge_df = (
    original_qwen_merge_df
    .drop_duplicates(
        subset=[
            "SOURCE_RECORD_ID"
        ],
        keep="last",
    )
)


benchmark_df = human_df.merge(
    original_qwen_merge_df,
    on="SOURCE_RECORD_ID",
    how="left",
)


# =============================================================================
# 10. FILL NARRATIVES FROM THE ACCIDENTS DATAFRAME IF NEEDED
# =============================================================================

accidents_df = accidents.copy()


accidents_df.columns = [
    normalize_column_name(
        column
    )
    for column in accidents_df.columns
]


accidents_id_column = first_existing_column(
    accidents_df,
    [
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
        "SOURCE_RECORD_ID",
    ],
    required=True,
    description="accidents DataFrame record ID",
)


accidents_text_column = first_existing_column(
    accidents_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
    required=True,
    description="accidents DataFrame narrative",
)


accidents_df[
    accidents_id_column
] = accidents_df[
    accidents_id_column
].astype(
    str
)


narrative_lookup = (
    accidents_df
    .drop_duplicates(
        subset=[
            accidents_id_column
        ],
        keep="first",
    )
    .set_index(
        accidents_id_column
    )[
        accidents_text_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .to_dict()
)


missing_narrative_mask = benchmark_df[
    "NARRATIVE"
].astype(
    str
).str.strip().eq(
    ""
)


benchmark_df.loc[
    missing_narrative_mask,
    "NARRATIVE",
] = (
    benchmark_df.loc[
        missing_narrative_mask,
        "SOURCE_RECORD_ID",
    ]
    .map(
        narrative_lookup
    )
    .fillna(
        ""
    )
)


benchmark_df = benchmark_df[
    benchmark_df[
        "NARRATIVE"
    ]
    .astype(
        str
    )
    .str.strip()
    .ne(
        ""
    )
].copy()


benchmark_df = benchmark_df[
    benchmark_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
    .isin(
        CAUSE_CATEGORIES
    )
].copy()


benchmark_df = benchmark_df.reset_index(
    drop=True
)


print(
    "Independent benchmark records:",
    f"{len(benchmark_df):,}",
)


# =============================================================================
# 11. LOAD HIGH-CONFIDENCE CODE BLOCK 2 REFERENCES
# =============================================================================

block2_df = pd.read_csv(
    block2_results_file,
    low_memory=False,
)


block2_df.columns = [
    normalize_column_name(
        column
    )
    for column in block2_df.columns
]


block2_id_column = first_existing_column(
    block2_df,
    [
        "SOURCE_RECORD_ID",
        "DOCUMENT_NO",
        "DOC_NO",
        "ACCIDENT_ID",
    ],
    required=True,
    description="Block 2 record ID",
)


block2_text_column = first_existing_column(
    block2_df,
    [
        "NARRATIVE",
        "ACCIDENT_NARRATIVE",
        "TEXT",
    ],
    required=True,
    description="Block 2 narrative",
)


block2_cause_column = first_existing_column(
    block2_df,
    [
        "PRIMARY_CAUSE",
        "ORIGINAL_MODEL_CAUSE",
        "MODEL_PRIMARY_CAUSE",
        "REVISED_CAUSE",
    ],
    required=True,
    description="Block 2 primary cause",
)


block2_confidence_column = first_existing_column(
    block2_df,
    [
        "CAUSE_CONFIDENCE",
        "ORIGINAL_MODEL_CAUSE_CONFIDENCE",
        "MODEL_CAUSE_CONFIDENCE",
    ],
)


block2_support_column = first_existing_column(
    block2_df,
    [
        "CAUSE_SUPPORT_LEVEL",
        "ORIGINAL_MODEL_CAUSE_SUPPORT",
        "MODEL_CAUSE_SUPPORT",
    ],
)


block2_evidence_column = first_existing_column(
    block2_df,
    [
        "CAUSE_EVIDENCE_TEXT",
        "ORIGINAL_MODEL_CAUSE_EVIDENCE",
        "MODEL_CAUSE_EVIDENCE",
    ],
)


block2_evidence_valid_column = first_existing_column(
    block2_df,
    [
        "CAUSE_EVIDENCE_VALID",
        "MODEL_CAUSE_EVIDENCE_VALID",
    ],
)


block2_record_type_column = first_existing_column(
    block2_df,
    [
        "RECORD_TYPE",
        "ORIGINAL_MODEL_RECORD_TYPE",
        "MODEL_RECORD_TYPE",
    ],
)


block2_status_column = first_existing_column(
    block2_df,
    [
        "PROCESSING_STATUS",
        "MODEL_PROCESSING_STATUS",
    ],
)


block2_df[
    "SOURCE_RECORD_ID"
] = block2_df[
    block2_id_column
].astype(
    str
)


block2_df[
    "NARRATIVE"
] = (
    block2_df[
        block2_text_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .str.strip()
)


block2_df[
    "REFERENCE_CAUSE"
] = block2_df[
    block2_cause_column
].map(
    normalize_category
)


block2_df[
    "REFERENCE_CONFIDENCE"
] = (
    pd.to_numeric(
        block2_df[
            block2_confidence_column
        ],
        errors="coerce",
    )
    if block2_confidence_column
    else np.nan
)


block2_df[
    "REFERENCE_SUPPORT"
] = (
    block2_df[
        block2_support_column
    ].map(
        normalize_category
    )
    if block2_support_column
    else "MISSING"
)


block2_df[
    "REFERENCE_EVIDENCE"
] = (
    block2_df[
        block2_evidence_column
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    if block2_evidence_column
    else ""
)


block2_mask = (
    block2_df[
        "NARRATIVE"
    ]
    .str.split()
    .str.len()
    .ge(
        5
    )
)


block2_mask &= block2_df[
    "REFERENCE_CAUSE"
].isin(
    CAUSE_CATEGORIES
)


block2_mask &= ~block2_df[
    "REFERENCE_CAUSE"
].isin(
    {
        "OTHER",
        "INSUFFICIENT_INFORMATION",
    }
)


if block2_status_column:

    block2_mask &= (
        block2_df[
            block2_status_column
        ]
        .astype(
            str
        )
        .str.upper()
        .eq(
            "SUCCESS"
        )
    )


if block2_record_type_column:

    block2_mask &= (
        block2_df[
            block2_record_type_column
        ]
        .map(
            normalize_category
        )
        .eq(
            "ACUTE_ACCIDENT"
        )
    )


if block2_confidence_column:

    block2_mask &= block2_df[
        "REFERENCE_CONFIDENCE"
    ].ge(
        MIN_BLOCK2_CAUSE_CONFIDENCE
    )


if block2_support_column:

    block2_mask &= block2_df[
        "REFERENCE_SUPPORT"
    ].isin(
        ALLOWED_BLOCK2_SUPPORT
    )


if block2_evidence_valid_column:

    block2_mask &= block2_df[
        block2_evidence_valid_column
    ].map(
        normalize_boolean
    )


block2_library_df = block2_df[
    block2_mask
].copy()


block2_library_df[
    "_NORMALIZED_NARRATIVE"
] = (
    block2_library_df[
        "NARRATIVE"
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


block2_library_df = (
    block2_library_df
    .drop_duplicates(
        subset=[
            "_NORMALIZED_NARRATIVE"
        ],
        keep="first",
    )
    .reset_index(
        drop=True
    )
)


block2_library_df[
    "REFERENCE_SOURCE"
] = "QWEN_HIGH_CONFIDENCE"


block2_library_df[
    "REFERENCE_REVIEW_STATUS"
] = "MODEL_HIGH_CONFIDENCE"


print(
    "High-confidence Block 2 references:",
    f"{len(block2_library_df):,}",
)


# =============================================================================
# 12. BUILD HUMAN-REVIEWED REFERENCE LIBRARY
#
# These records are used through leave-one-out retrieval.
# Each target is excluded from its own retrieved examples.
# =============================================================================

human_library_df = pd.DataFrame(
    {
        "SOURCE_RECORD_ID":
            benchmark_df[
                "SOURCE_RECORD_ID"
            ].astype(
                str
            ),

        "NARRATIVE":
            benchmark_df[
                "NARRATIVE"
            ].astype(
                str
            ),

        "REFERENCE_CAUSE":
            benchmark_df[
                "HUMAN_PRIMARY_CAUSE"
            ].map(
                normalize_category
            ),

        "REFERENCE_EVIDENCE":
            benchmark_df[
                "HUMAN_CAUSE_EVIDENCE"
            ].astype(
                str
            ),

        "REFERENCE_CONFIDENCE":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                0.50,
                1.00,
            ),

        "REFERENCE_SUPPORT":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                "UNCERTAIN",
                "HUMAN_REVIEWED",
            ),

        "REFERENCE_SOURCE":
            "HUMAN_REVIEWED",

        "REFERENCE_REVIEW_STATUS":
            np.where(
                benchmark_df[
                    "HUMAN_CAUSE_UNCERTAIN"
                ],
                "HUMAN_UNCERTAIN",
                "HUMAN_CERTAIN",
            ),
    }
)


human_library_df = human_library_df[
    human_library_df[
        "REFERENCE_CAUSE"
    ]
    .isin(
        CAUSE_CATEGORIES
    )
].copy()


# =============================================================================
# 13. COMBINE THE HYBRID REFERENCE LIBRARY
# =============================================================================

reference_columns = [
    "SOURCE_RECORD_ID",
    "NARRATIVE",
    "REFERENCE_CAUSE",
    "REFERENCE_EVIDENCE",
    "REFERENCE_CONFIDENCE",
    "REFERENCE_SUPPORT",
    "REFERENCE_SOURCE",
    "REFERENCE_REVIEW_STATUS",
]


reference_library_df = pd.concat(
    [
        block2_library_df[
            reference_columns
        ],
        human_library_df[
            reference_columns
        ],
    ],
    ignore_index=True,
)


reference_library_df[
    "SOURCE_RECORD_ID"
] = reference_library_df[
    "SOURCE_RECORD_ID"
].astype(
    str
)


reference_library_df[
    "NARRATIVE"
] = (
    reference_library_df[
        "NARRATIVE"
    ]
    .astype(
        "string"
    )
    .fillna(
        ""
    )
    .str.strip()
)


reference_library_df[
    "_NORMALIZED_NARRATIVE"
] = (
    reference_library_df[
        "NARRATIVE"
    ]
    .str.lower()
    .str.replace(
        r"\s+",
        " ",
        regex=True,
    )
    .str.strip()
)


# Prefer human-reviewed records when identical narratives appear in both sources.
reference_library_df[
    "_SOURCE_PRIORITY"
] = reference_library_df[
    "REFERENCE_SOURCE"
].map(
    {
        "HUMAN_REVIEWED":
            0,

        "QWEN_HIGH_CONFIDENCE":
            1,
    }
).fillna(
    2
)


reference_library_df = (
    reference_library_df
    .sort_values(
        "_SOURCE_PRIORITY"
    )
    .drop_duplicates(
        subset=[
            "_NORMALIZED_NARRATIVE"
        ],
        keep="first",
    )
    .reset_index(
        drop=True
    )
)


print(
    "Combined hybrid reference library:",
    f"{len(reference_library_df):,}",
)


print(
    reference_library_df[
        "REFERENCE_SOURCE"
    ]
    .value_counts()
)


# =============================================================================
# 14. BUILD RETRIEVAL INDEXES
# =============================================================================

print("\n" + "=" * 95)
print("BUILDING RETRIEVAL INDEXES")
print("=" * 95)


embedding_model = SentenceTransformer(
    EMBEDDING_MODEL_ID
)


reference_embeddings = embedding_model.encode(
    reference_library_df[
        "NARRATIVE"
    ].tolist(),
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True,
)


benchmark_embeddings = embedding_model.encode(
    benchmark_df[
        "NARRATIVE"
    ].tolist(),
    batch_size=64,
    show_progress_bar=True,
    normalize_embeddings=True,
)


tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    strip_accents="unicode",
    stop_words="english",
    ngram_range=TFIDF_NGRAM_RANGE,
    min_df=TFIDF_MIN_DF,
    max_df=TFIDF_MAX_DF,
    max_features=TFIDF_MAX_FEATURES,
    sublinear_tf=True,
    norm="l2",
)


combined_texts_for_tfidf = (
    reference_library_df[
        "NARRATIVE"
    ].tolist()
    + benchmark_df[
        "NARRATIVE"
    ].tolist()
)


combined_tfidf_matrix = tfidf_vectorizer.fit_transform(
    combined_texts_for_tfidf
)


reference_tfidf_matrix = combined_tfidf_matrix[
    :
    len(
        reference_library_df
    )
]


benchmark_tfidf_matrix = combined_tfidf_matrix[
    len(
        reference_library_df
    ):
]


# =============================================================================
# 15. HYBRID RETRIEVAL
# =============================================================================

def retrieve_examples(
    benchmark_index,
    target_record_id,
    target_narrative,
):

    semantic_scores = np.dot(
        reference_embeddings,
        benchmark_embeddings[
            benchmark_index
        ],
    )


    lexical_scores = cosine_similarity(
        benchmark_tfidf_matrix[
            benchmark_index
        ],
        reference_tfidf_matrix,
    ).ravel()


    hybrid_scores = (
        SEMANTIC_WEIGHT
        * semantic_scores
        + LEXICAL_WEIGHT
        * lexical_scores
    )


    candidates = reference_library_df.copy()


    candidates[
        "SEMANTIC_SCORE"
    ] = semantic_scores


    candidates[
        "LEXICAL_SCORE"
    ] = lexical_scores


    candidates[
        "HYBRID_SCORE"
    ] = hybrid_scores


    target_normalized = normalize_evidence_text(
        target_narrative
    )


    # Exclude the target by source ID.
    candidates = candidates[
        candidates[
            "SOURCE_RECORD_ID"
        ]
        .astype(
            str
        )
        .ne(
            str(
                target_record_id
            )
        )
    ].copy()


    # Exclude duplicate target text even if the identifier differs.
    candidates = candidates[
        candidates[
            "NARRATIVE"
        ]
        .map(
            normalize_evidence_text
        )
        .ne(
            target_normalized
        )
    ].copy()


    candidates = candidates[
        candidates[
            "HYBRID_SCORE"
        ]
        .ge(
            MIN_RETRIEVAL_SCORE
        )
    ].copy()


    candidates = candidates.sort_values(
        [
            "HYBRID_SCORE",
            "REFERENCE_SOURCE",
        ],
        ascending=[
            False,
            True,
        ],
    )


    selected_rows = []

    human_count = 0

    cause_counts = {}


    for _, candidate in candidates.iterrows():

        cause_value = candidate[
            "REFERENCE_CAUSE"
        ]


        current_cause_count = cause_counts.get(
            cause_value,
            0,
        )


        if (
            current_cause_count
            >= MAX_EXAMPLES_PER_CAUSE_PER_QUERY
        ):

            continue


        if (
            candidate[
                "REFERENCE_SOURCE"
            ]
            == "HUMAN_REVIEWED"
        ):

            if (
                human_count
                >= MAX_HUMAN_EXAMPLES_PER_QUERY
            ):

                continue

            human_count += 1


        selected_rows.append(
            candidate
        )


        cause_counts[
            cause_value
        ] = (
            current_cause_count
            + 1
        )


        if len(
            selected_rows
        ) >= TOP_K_RETRIEVAL:

            break


    if not selected_rows:

        return pd.DataFrame(
            columns=candidates.columns
        )


    output = pd.DataFrame(
        selected_rows
    ).reset_index(
        drop=True
    )


    output[
        "RETRIEVAL_RANK"
    ] = np.arange(
        1,
        len(
            output
        )
        + 1,
    )


    return output


# =============================================================================
# 15A. RETRIEVAL-ONLY CLASSIFIER
#
# Uses the same retrieved examples as RAG-Qwen, but no generator.
# Each retrieved example votes for its accepted cause:
#
#   vote weight = hybrid score x reference confidence
#
# Ties are broken by the highest single hybrid score for the cause.
# =============================================================================

def classify_retrieval_only(
    retrieved_df,
):

    if retrieved_df.empty:

        return {
            "RETRIEVAL_ONLY_CAUSE":
                RETRIEVAL_ONLY_FALLBACK_CAUSE,

            "RETRIEVAL_ONLY_TOP2_CAUSE":
                "",

            "RETRIEVAL_ONLY_VOTE_SHARE":
                0.0,

            "RETRIEVAL_ONLY_TOP2_VOTE_SHARE":
                0.0,

            "RETRIEVAL_ONLY_SUPPORTING_EXAMPLES":
                0,

            "RETRIEVAL_ONLY_STATUS":
                "NO_RETRIEVED_EXAMPLES",
        }


    vote_df = retrieved_df[
        [
            "REFERENCE_CAUSE",
            "HYBRID_SCORE",
            "REFERENCE_CONFIDENCE",
        ]
    ].copy()


    vote_df[
        "HYBRID_SCORE"
    ] = (
        pd.to_numeric(
            vote_df[
                "HYBRID_SCORE"
            ],
            errors="coerce",
        )
        .fillna(
            0.0
        )
        .clip(
            lower=0.0
        )
    )


    vote_df[
        "REFERENCE_CONFIDENCE"
    ] = (
        pd.to_numeric(
            vote_df[
                "REFERENCE_CONFIDENCE"
            ],
            errors="coerce",
        )
        .fillna(
            1.0
        )
        .clip(
            lower=0.0,
            upper=1.0,
        )
    )


    if RETRIEVAL_ONLY_USE_REFERENCE_CONFIDENCE:

        vote_df[
            "VOTE_WEIGHT"
        ] = (
            vote_df[
                "HYBRID_SCORE"
            ]
            * vote_df[
                "REFERENCE_CONFIDENCE"
            ]
        )

    else:

        vote_df[
            "VOTE_WEIGHT"
        ] = vote_df[
            "HYBRID_SCORE"
        ]


    cause_votes_df = (
        vote_df
        .groupby(
            "REFERENCE_CAUSE"
        )
        .agg(
            Total_weight=(
                "VOTE_WEIGHT",
                "sum",
            ),
            Best_score=(
                "HYBRID_SCORE",
                "max",
            ),
            Example_count=(
                "REFERENCE_CAUSE",
                "size",
            ),
        )
        .reset_index()
        .sort_values(
            [
                "Total_weight",
                "Best_score",
            ],
            ascending=[
                False,
                False,
            ],
        )
        .reset_index(
            drop=True
        )
    )


    total_weight = float(
        cause_votes_df[
            "Total_weight"
        ].sum()
    )


    top1_row = cause_votes_df.iloc[
        0
    ]


    top2_row = (
        cause_votes_df.iloc[
            1
        ]
        if len(
            cause_votes_df
        ) > 1
        else None
    )


    return {
        "RETRIEVAL_ONLY_CAUSE":
            normalize_category(
                top1_row[
                    "REFERENCE_CAUSE"
                ]
            ),

        "RETRIEVAL_ONLY_TOP2_CAUSE":
            (
                normalize_category(
                    top2_row[
                        "REFERENCE_CAUSE"
                    ]
                )
                if top2_row is not None
                else ""
            ),

        "RETRIEVAL_ONLY_VOTE_SHARE":
            (
                round(
                    float(
                        top1_row[
                            "Total_weight"
                        ]
                    )
                    / total_weight,
                    4,
                )
                if total_weight > 0
                else 0.0
            ),

        "RETRIEVAL_ONLY_TOP2_VOTE_SHARE":
            (
                round(
                    float(
                        top2_row[
                            "Total_weight"
                        ]
                    )
                    / total_weight,
                    4,
                )
                if (
                    top2_row is not None
                    and total_weight > 0
                )
                else 0.0
            ),

        "RETRIEVAL_ONLY_SUPPORTING_EXAMPLES":
            int(
                top1_row[
                    "Example_count"
                ]
            ),

        "RETRIEVAL_ONLY_STATUS":
            "SUCCESS",
    }


# =============================================================================
# 16. RAG SYSTEM PROMPT
# =============================================================================

CAUSE_HIERARCHY = """
CAUSE-SELECTION HIERARCHY

Select the first category clearly supported by the TARGET NARRATIVE:

1. CAUGHT_IN_OR_BETWEEN
   Trapped, pinched, crushed, compressed, entangled, or held between objects.

2. VEHICLE_OR_MOBILE_EQUIPMENT_EVENT
   Collision, overturn, runaway, backing, or transportation event.

3. GROUND_OR_ROOF_FAILURE
   Roof, rib, face, highwall, slate, rock, or ground failure.

4. PRESSURE_OR_STORED_ENERGY_RELEASE
   Hydraulic, pneumatic, spring, tension, pressure, or stored-energy release.

5. LOSS_OF_CONTROL_OF_TOOL_OR_EQUIPMENT
   Explicit loss of control of a tool, hose, handle, load, or component.

6. UNEXPECTED_EQUIPMENT_MOVEMENT
   Unexpected movement without explicit loss of control.

7. STRUCK_BY_MOVING_OBJECT
   Moving, falling, flying, rolling, or swinging object contact without
   trapping or compression.

8. STRUCK_AGAINST_STATIONARY_OBJECT
   The person's movement causes contact with a stationary object.

9. FOREIGN_BODY_OR_PARTICLE_CONTACT
   Chip, fragment, particle, or small foreign object contact.

10. SLIP_TRIP_OR_FALL
    Loss of footing, balance, or elevation.

11. AWKWARD_MOVEMENT_OR_BODY_MOTION
    Acute twist, reach, bend, turn, or step without a fall or substantial load.

12. MANUAL_HANDLING_OR_OVEREXERTION
    Acute injury from lifting, pushing, pulling, carrying, shoveling, holding,
    or exerting force.

Do not use STRUCK_BY_MOVING_OBJECT as a general fallback when another
specific mechanism is clearly supported.
""".strip()


RAG_SYSTEM_PROMPT = f"""
You classify the PRIMARY ACCIDENT CAUSE in short MSHA accident narratives.

This is reported-event classification, not verified root-cause analysis.

ALLOWED PRIMARY CAUSES
{", ".join(CAUSE_CATEGORIES)}

{CAUSE_HIERARCHY}

You will receive:

1. A TARGET NARRATIVE that must be classified.
2. Retrieved historical examples with accepted or high-confidence cause labels.

RETRIEVAL RULES

- Retrieved examples are references for terminology and category boundaries.
- The TARGET NARRATIVE is the only source of facts about the target event.
- Never copy an event detail, cause, or evidence excerpt merely because it
  appears in a retrieved example.
- A retrieved example may be similar but still have a different cause.
- Human-reviewed examples are stronger references than model-generated examples.
- Examples marked HUMAN_UNCERTAIN are weak references.
- If retrieved examples disagree, classify from the TARGET NARRATIVE.
- Cause evidence must be copied from the TARGET NARRATIVE, not from a retrieved
  example.
- Use INSUFFICIENT_INFORMATION when the target does not support an event
  mechanism.

CONFIDENCE

- HIGH confidence generally corresponds to 0.80-1.00.
- MEDIUM confidence generally corresponds to 0.60-0.79.
- LOW confidence is below 0.60.
- Confidence is an analytical estimate, not a calibrated probability.

Return only one JSON object with exactly these keys:

{{
  "primary_cause": "",
  "cause_evidence_text": "",
  "cause_confidence": 0.0,
  "cause_support_level": "HIGH",
  "retrieval_used": true,
  "retrieval_explanation": "",
  "alternative_cause": "",
  "reviewer_note": ""
}}
""".strip()


# =============================================================================
# 17. MODEL LOADING
# =============================================================================

tokenizer = None

model = None


def load_qwen_model():

    global tokenizer
    global model


    if (
        tokenizer is not None
        and model is not None
    ):

        return


    if not torch.cuda.is_available():

        raise RuntimeError(
            "No NVIDIA GPU is available. "
            "Qwen inference requires compatible GPU compute."
        )


    print("\n" + "=" * 95)
    print("LOADING LOCAL QWEN MODEL")
    print("=" * 95)


    quantization_config = BitsAndBytesConfig(
        load_in_4bit=True,
        bnb_4bit_quant_type="nf4",
        bnb_4bit_use_double_quant=True,
        bnb_4bit_compute_dtype=torch.float16,
    )


    tokenizer = AutoTokenizer.from_pretrained(
        MODEL_ID,
        use_fast=True,
        trust_remote_code=False,
    )


    if tokenizer.pad_token_id is None:

        tokenizer.pad_token = tokenizer.eos_token


    model = AutoModelForCausalLM.from_pretrained(
        MODEL_ID,
        quantization_config=quantization_config,
        device_map="auto",
        dtype=torch.float16,
        low_cpu_mem_usage=True,
        trust_remote_code=False,
    )


    model.eval()


    print(
        "Model loaded:",
        MODEL_ID,
    )


# =============================================================================
# 18. MODEL GENERATION
# =============================================================================

def generate_response(
    messages,
):

    load_qwen_model()


    model_inputs = tokenizer.apply_chat_template(
        messages,
        add_generation_prompt=True,
        tokenize=True,
        return_dict=True,
        return_tensors="pt",
    )


    model_inputs = {
        key:
            value.to(
                model.device
            )
        for key, value in model_inputs.items()
    }


    input_length = int(
        model_inputs[
            "input_ids"
        ].shape[
            -1
        ]
    )


    with torch.inference_mode():

        generated = model.generate(
            **model_inputs,
            max_new_tokens=MAX_NEW_TOKENS,
            do_sample=False,
            repetition_penalty=1.02,
            pad_token_id=tokenizer.pad_token_id,
            eos_token_id=tokenizer.eos_token_id,
            use_cache=True,
        )


    generated_ids = generated[
        0,
        input_length:
    ]


    response_text = tokenizer.decode(
        generated_ids,
        skip_special_tokens=True,
    )


    del model_inputs
    del generated
    del generated_ids


    return response_text


# =============================================================================
# 19. BUILD THE RAG USER PROMPT
# =============================================================================

def format_retrieved_examples(
    retrieved_df,
):

    if retrieved_df.empty:

        return (
            "No retrieved example exceeded the minimum similarity threshold."
        )


    sections = []


    for _, row in retrieved_df.iterrows():

        sections.append(
            "\n".join(
                [
                    (
                        f"RETRIEVED EXAMPLE "
                        f"{int(row['RETRIEVAL_RANK'])}"
                    ),
                    (
                        f"Source quality: "
                        f"{row['REFERENCE_REVIEW_STATUS']}"
                    ),
                    (
                        f"Hybrid similarity: "
                        f"{row['HYBRID_SCORE']:.4f}"
                    ),
                    (
                        f"Accepted cause: "
                        f"{row['REFERENCE_CAUSE']}"
                    ),
                    (
                        f"Historical narrative: "
                        f"{row['NARRATIVE']}"
                    ),
                    (
                        f"Historical cause evidence: "
                        f"{row['REFERENCE_EVIDENCE']}"
                    ),
                ]
            )
        )


    return "\n\n".join(
        sections
    )


def build_user_prompt(
    target_narrative,
    retrieved_df,
):

    retrieved_text = format_retrieved_examples(
        retrieved_df
    )


    return f"""
TARGET NARRATIVE
{target_narrative}

RETRIEVED HISTORICAL EXAMPLES
{retrieved_text}

Classify only the TARGET NARRATIVE.

Use the retrieved examples to understand terminology and category boundaries,
but do not transfer facts from a historical example to the target.

Copy the cause evidence from the TARGET NARRATIVE.

Return only the required JSON object.
""".strip()


# =============================================================================
# 20. CLASSIFY ONE TARGET WITH RAG
# =============================================================================

def classify_with_rag(
    target_narrative,
    retrieved_df,
):

    user_prompt = build_user_prompt(
        target_narrative,
        retrieved_df,
    )


    last_response = ""

    last_error = ""


    for attempt_number in range(
        1,
        MAX_RETRIES + 1,
    ):

        messages = [
            {
                "role":
                    "system",

                "content":
                    RAG_SYSTEM_PROMPT,
            },
            {
                "role":
                    "user",

                "content":
                    user_prompt,
            },
        ]


        if attempt_number > 1:

            messages.append(
                {
                    "role":
                        "user",

                    "content":
                        f"""
The previous response was invalid.

VALIDATION ERROR
{last_error}

PREVIOUS RESPONSE
{last_response}

Return one corrected JSON object only.
Use only the allowed primary-cause categories.
Do not include Markdown or any text outside the JSON object.
""".strip(),
                }
            )


        try:

            raw_response = generate_response(
                messages
            )


            last_response = raw_response


            json_text = extract_balanced_json(
                raw_response
            )


            json_text = repair_json(
                json_text
            )


            parsed_dictionary = json.loads(
                json_text
            )


            parsed_dictionary[
                "primary_cause"
            ] = normalize_category(
                parsed_dictionary.get(
                    "primary_cause",
                    "",
                )
            )


            alternative_value = parsed_dictionary.get(
                "alternative_cause",
                "",
            )


            if alternative_value:

                parsed_dictionary[
                    "alternative_cause"
                ] = normalize_category(
                    alternative_value
                )


            parsed_dictionary[
                "cause_support_level"
            ] = normalize_category(
                parsed_dictionary.get(
                    "cause_support_level",
                    "LOW",
                )
            )


            parsed_dictionary[
                "retrieval_used"
            ] = bool(
                parsed_dictionary.get(
                    "retrieval_used",
                    True,
                )
            )


            validated = RAGCauseExtraction.model_validate(
                parsed_dictionary
            )


            evidence_result = evaluate_evidence(
                validated.cause_evidence_text,
                target_narrative,
            )


            return {
                **validated.model_dump(),

                "RAG_SCHEMA_VALID":
                    True,

                "RAG_CAUSE_EVIDENCE_VALID":
                    evidence_result[
                        "valid"
                    ],

                "RAG_CAUSE_EVIDENCE_EXACT":
                    evidence_result[
                        "exact"
                    ],

                "RAG_CAUSE_EVIDENCE_NORMALIZED":
                    evidence_result[
                        "normalized"
                    ],

                "RAG_CAUSE_EVIDENCE_TOKEN_OVERLAP":
                    evidence_result[
                        "token_overlap"
                    ],

                "RAG_PROCESSING_STATUS":
                    "SUCCESS",

                "RAG_PROCESSING_ERROR":
                    "",

                "RAG_ATTEMPT_COUNT":
                    attempt_number,

                "RAG_RAW_RESPONSE":
                    raw_response,
            }


        except Exception as error:

            last_error = (
                f"{type(error).__name__}: "
                f"{str(error)}"
            )


            print(
                f"  Attempt {attempt_number} failed: "
                f"{last_error[:250]}"
            )


            gc.collect()


            if torch.cuda.is_available():

                torch.cuda.empty_cache()


    return {
        "primary_cause":
            "INSUFFICIENT_INFORMATION",

        "cause_evidence_text":
            "",

        "cause_confidence":
            0.0,

        "cause_support_level":
            "LOW",

        "retrieval_used":
            not retrieved_df.empty,

        "retrieval_explanation":
            "",

        "alternative_cause":
            "",

        "reviewer_note":
            "",

        "RAG_SCHEMA_VALID":
            False,

        "RAG_CAUSE_EVIDENCE_VALID":
            False,

        "RAG_CAUSE_EVIDENCE_EXACT":
            False,

        "RAG_CAUSE_EVIDENCE_NORMALIZED":
            False,

        "RAG_CAUSE_EVIDENCE_TOKEN_OVERLAP":
            0.0,

        "RAG_PROCESSING_STATUS":
            "FAILED",

        "RAG_PROCESSING_ERROR":
            last_error,

        "RAG_ATTEMPT_COUNT":
            MAX_RETRIES,

        "RAG_RAW_RESPONSE":
            last_response,
    }


# =============================================================================
# 21. CHECKPOINT UTILITIES
# =============================================================================

def load_checkpoint(
    path,
):

    path = Path(
        path
    )


    if not path.exists():

        return {}


    records = {}


    with path.open(
        "r",
        encoding="utf-8",
    ) as input_file:

        for line in input_file:

            line = line.strip()


            if not line:

                continue


            try:

                record = json.loads(
                    line
                )


                records[
                    str(
                        record[
                            "SOURCE_RECORD_ID"
                        ]
                    )
                ] = record


            except Exception:

                continue


    return records


def append_checkpoint(
    record,
    path,
):

    with open(
        path,
        "a",
        encoding="utf-8",
    ) as output_file:

        output_file.write(
            json.dumps(
                record,
                ensure_ascii=False,
                default=str,
            )
            + "\n"
        )


checkpoint_records = load_checkpoint(
    CHECKPOINT_JSONL
)


print(
    "Existing RAG checkpoint records:",
    f"{len(checkpoint_records):,}",
)


# =============================================================================
# 22. RUN RAG ON THE HUMAN-REVIEW BENCHMARK
# =============================================================================

rag_records = []

retrieval_detail_records = []


for benchmark_index, row in benchmark_df.iterrows():

    record_id = str(
        row[
            "SOURCE_RECORD_ID"
        ]
    )


    if (
        not REPROCESS_COMPLETED
        and record_id
        in checkpoint_records
    ):

        print(
            f"[{benchmark_index + 1}/"
            f"{len(benchmark_df)}] "
            f"Checkpoint: {record_id}"
        )


        rag_records.append(
            checkpoint_records[
                record_id
            ]
        )


        continue


    target_narrative = str(
        row[
            "NARRATIVE"
        ]
    )


    retrieved_df = retrieve_examples(
        benchmark_index=benchmark_index,
        target_record_id=record_id,
        target_narrative=target_narrative,
    )


    for _, retrieved_row in retrieved_df.iterrows():

        retrieval_detail_records.append(
            {
                "TARGET_RECORD_ID":
                    record_id,

                "TARGET_HUMAN_CAUSE":
                    row[
                        "HUMAN_PRIMARY_CAUSE"
                    ],

                "RETRIEVAL_RANK":
                    int(
                        retrieved_row[
                            "RETRIEVAL_RANK"
                        ]
                    ),

                "REFERENCE_RECORD_ID":
                    retrieved_row[
                        "SOURCE_RECORD_ID"
                    ],

                "REFERENCE_SOURCE":
                    retrieved_row[
                        "REFERENCE_SOURCE"
                    ],

                "REFERENCE_REVIEW_STATUS":
                    retrieved_row[
                        "REFERENCE_REVIEW_STATUS"
                    ],

                "REFERENCE_CAUSE":
                    retrieved_row[
                        "REFERENCE_CAUSE"
                    ],

                "SEMANTIC_SCORE":
                    float(
                        retrieved_row[
                            "SEMANTIC_SCORE"
                        ]
                    ),

                "LEXICAL_SCORE":
                    float(
                        retrieved_row[
                            "LEXICAL_SCORE"
                        ]
                    ),

                "HYBRID_SCORE":
                    float(
                        retrieved_row[
                            "HYBRID_SCORE"
                        ]
                    ),

                "REFERENCE_NARRATIVE":
                    retrieved_row[
                        "NARRATIVE"
                    ],
            }
        )


    print(
        f"[{benchmark_index + 1}/"
        f"{len(benchmark_df)}] "
        f"RAG inference | "
        f"{record_id} | "
        f"{len(retrieved_df)} examples"
    )


    start_time = time.time()


    rag_result = classify_with_rag(
        target_narrative=target_narrative,
        retrieved_df=retrieved_df,
    )


    top_reference_cause = (
        retrieved_df.iloc[
            0
        ][
            "REFERENCE_CAUSE"
        ]
        if not retrieved_df.empty
        else ""
    )


    top_reference_score = (
        float(
            retrieved_df.iloc[
                0
            ][
                "HYBRID_SCORE"
            ]
        )
        if not retrieved_df.empty
        else np.nan
    )


    retrieved_causes = (
        "; ".join(
            retrieved_df[
                "REFERENCE_CAUSE"
            ].astype(
                str
            ).tolist()
        )
        if not retrieved_df.empty
        else ""
    )


    retrieved_sources = (
        "; ".join(
            retrieved_df[
                "REFERENCE_SOURCE"
            ].astype(
                str
            ).tolist()
        )
        if not retrieved_df.empty
        else ""
    )


    record = {
        "SOURCE_RECORD_ID":
            record_id,

        "NARRATIVE":
            target_narrative,

        "SEVERITY_GROUP":
            row.get(
                "SEVERITY_GROUP",
                "Unknown",
            ),

        "TIME_PERIOD":
            row.get(
                "TIME_PERIOD",
                "Unknown",
            ),

        "HUMAN_PRIMARY_CAUSE":
            row[
                "HUMAN_PRIMARY_CAUSE"
            ],

        "HUMAN_CAUSE_UNCERTAIN":
            bool(
                row[
                    "HUMAN_CAUSE_UNCERTAIN"
                ]
            ),

        "HUMAN_REVIEW_COMMENTS":
            row.get(
                "HUMAN_REVIEW_COMMENTS",
                "",
            ),

        "ORIGINAL_QWEN_CAUSE":
            row.get(
                "ORIGINAL_QWEN_CAUSE",
                "",
            ),

        "ORIGINAL_QWEN_CONFIDENCE":
            row.get(
                "ORIGINAL_QWEN_CONFIDENCE",
                np.nan,
            ),

        "ORIGINAL_QWEN_EVIDENCE":
            row.get(
                "ORIGINAL_QWEN_EVIDENCE",
                "",
            ),

        "RETRIEVED_EXAMPLE_COUNT":
            len(
                retrieved_df
            ),

        "TOP_RETRIEVED_CAUSE":
            top_reference_cause,

        "TOP_RETRIEVAL_SCORE":
            top_reference_score,

        "RETRIEVED_CAUSES":
            retrieved_causes,

        "RETRIEVED_SOURCES":
            retrieved_sources,

        **rag_result,

        "MODEL_ID":
            MODEL_ID,

        "EMBEDDING_MODEL_ID":
            EMBEDDING_MODEL_ID,

        "PROMPT_VERSION":
            PROMPT_VERSION,

        "RAG_PROCESSING_SECONDS":
            round(
                time.time()
                - start_time,
                3,
            ),

        "PROCESSED_AT":
            datetime.now().isoformat(
                timespec="seconds"
            ),
    }


    rag_records.append(
        record
    )


    append_checkpoint(
        record,
        CHECKPOINT_JSONL,
    )


    gc.collect()


    if torch.cuda.is_available():

        torch.cuda.empty_cache()


# =============================================================================
# 22A. RUN RETRIEVAL-ONLY CLASSIFICATION ON THE SAME BENCHMARK
#
# Retrieval is deterministic, so this reproduces the same retrieved examples
# supplied to RAG-Qwen, including records loaded from the checkpoint.
# No generator is used.
# =============================================================================

print("\n" + "=" * 95)
print("RUNNING RETRIEVAL-ONLY COMPARISON")
print("=" * 95)


retrieval_only_records = []


for benchmark_index, row in benchmark_df.iterrows():

    record_id = str(
        row[
            "SOURCE_RECORD_ID"
        ]
    )


    retrieved_df = retrieve_examples(
        benchmark_index=benchmark_index,
        target_record_id=record_id,
        target_narrative=str(
            row[
                "NARRATIVE"
            ]
        ),
    )


    retrieval_only_records.append(
        {
            "SOURCE_RECORD_ID":
                record_id,

            **classify_retrieval_only(
                retrieved_df
            ),
        }
    )


retrieval_only_df = (
    pd.DataFrame(
        retrieval_only_records
    )
    .drop_duplicates(
        subset=[
            "SOURCE_RECORD_ID"
        ],
        keep="last",
    )
)


print(
    "Retrieval-only classifications:",
    f"{len(retrieval_only_df):,}",
)


# =============================================================================
# 23. BUILD FINAL RAG RESULTS
# =============================================================================

results_df = pd.DataFrame(
    rag_records
)


results_df[
    "RAG_PRIMARY_CAUSE"
] = results_df[
    "primary_cause"
].map(
    normalize_category
)


results_df[
    "RAG_ALTERNATIVE_CAUSE"
] = results_df[
    "alternative_cause"
].map(
    normalize_category
)


results_df[
    "RAG_EXACT_MATCH"
] = (
    results_df[
        "RAG_PRIMARY_CAUSE"
    ]
    == results_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


results_df[
    "RAG_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_PRIMARY_CAUSE"
        ],
    )
]


results_df[
    "RAG_TOP2_EXACT_MATCH"
] = (
    results_df[
        "RAG_EXACT_MATCH"
    ]
    |
    (
        results_df[
            "RAG_ALTERNATIVE_CAUSE"
        ]
        == results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
)


results_df[
    "RAG_TOP2_EXACT_OR_RELATED"
] = [
    (
        categories_related(
            human_value,
            primary_value,
        )
        or categories_related(
            human_value,
            alternative_value,
        )
    )
    for (
        human_value,
        primary_value,
        alternative_value,
    ) in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_PRIMARY_CAUSE"
        ],
        results_df[
            "RAG_ALTERNATIVE_CAUSE"
        ],
    )
]


results_df[
    "ORIGINAL_QWEN_EXACT_MATCH"
] = (
    results_df[
        "ORIGINAL_QWEN_CAUSE"
    ]
    == results_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


results_df[
    "ORIGINAL_QWEN_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "ORIGINAL_QWEN_CAUSE"
        ],
    )
]


results_df[
    "RAG_CHANGED_QWEN_CAUSE"
] = (
    results_df[
        "RAG_PRIMARY_CAUSE"
    ]
    != results_df[
        "ORIGINAL_QWEN_CAUSE"
    ]
)


results_df[
    "RAG_IMPROVED_EXACT"
] = (
    results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    ~results_df[
        "ORIGINAL_QWEN_EXACT_MATCH"
    ]
)


results_df[
    "RAG_REDUCED_EXACT"
] = (
    ~results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    results_df[
        "ORIGINAL_QWEN_EXACT_MATCH"
    ]
)


# =============================================================================
# 23A. MERGE RETRIEVAL-ONLY RESULTS
# =============================================================================

results_df = results_df.merge(
    retrieval_only_df,
    on="SOURCE_RECORD_ID",
    how="left",
)


results_df[
    "RETRIEVAL_ONLY_CAUSE"
] = (
    results_df[
        "RETRIEVAL_ONLY_CAUSE"
    ]
    .fillna(
        RETRIEVAL_ONLY_FALLBACK_CAUSE
    )
    .map(
        normalize_category
    )
)


results_df[
    "RETRIEVAL_ONLY_TOP2_CAUSE"
] = (
    results_df[
        "RETRIEVAL_ONLY_TOP2_CAUSE"
    ]
    .fillna(
        ""
    )
    .astype(
        str
    )
)


results_df[
    "RETRIEVAL_ONLY_EXACT_MATCH"
] = (
    results_df[
        "RETRIEVAL_ONLY_CAUSE"
    ]
    == results_df[
        "HUMAN_PRIMARY_CAUSE"
    ]
)


results_df[
    "RETRIEVAL_ONLY_EXACT_OR_RELATED"
] = [
    categories_related(
        human_value,
        prediction,
    )
    for human_value, prediction in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RETRIEVAL_ONLY_CAUSE"
        ],
    )
]


results_df[
    "RETRIEVAL_ONLY_TOP2_EXACT_MATCH"
] = (
    results_df[
        "RETRIEVAL_ONLY_EXACT_MATCH"
    ]
    |
    (
        results_df[
            "RETRIEVAL_ONLY_TOP2_CAUSE"
        ]
        == results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
)


results_df[
    "RETRIEVAL_ONLY_TOP2_EXACT_OR_RELATED"
] = [
    (
        categories_related(
            human_value,
            top1_value,
        )
        or (
            bool(
                top2_value
            )
            and categories_related(
                human_value,
                top2_value,
            )
        )
    )
    for (
        human_value,
        top1_value,
        top2_value,
    ) in zip(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "RETRIEVAL_ONLY_CAUSE"
        ],
        results_df[
            "RETRIEVAL_ONLY_TOP2_CAUSE"
        ],
    )
]


results_df[
    "RAG_AGREES_WITH_RETRIEVAL_ONLY"
] = (
    results_df[
        "RAG_PRIMARY_CAUSE"
    ]
    == results_df[
        "RETRIEVAL_ONLY_CAUSE"
    ]
)


results_df[
    "RAG_IMPROVED_OVER_RETRIEVAL_ONLY"
] = (
    results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    ~results_df[
        "RETRIEVAL_ONLY_EXACT_MATCH"
    ]
)


results_df[
    "RAG_REDUCED_VS_RETRIEVAL_ONLY"
] = (
    ~results_df[
        "RAG_EXACT_MATCH"
    ]
    &
    results_df[
        "RETRIEVAL_ONLY_EXACT_MATCH"
    ]
)


# =============================================================================
# 24. MERGE OPTIONAL LOGISTIC-REGRESSION RESULTS
# =============================================================================

if logistic_results_file:

    logistic_df = pd.read_csv(
        logistic_results_file,
        low_memory=False,
    )


    logistic_df.columns = [
        normalize_column_name(
            column
        )
        for column in logistic_df.columns
    ]


    logistic_id_column = first_existing_column(
        logistic_df,
        [
            "SOURCE_RECORD_ID",
            "DOCUMENT_NO",
            "DOC_NO",
            "ACCIDENT_ID",
        ],
        required=True,
        description="logistic result ID",
    )


    logistic_cause_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP1_CAUSE",
            "PREDICTED_PRIMARY_CAUSE",
        ],
        required=True,
        description="logistic cause",
    )


    logistic_probability_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP1_PROBABILITY",
            "PREDICTED_PROBABILITY",
        ],
    )


    logistic_top2_column = first_existing_column(
        logistic_df,
        [
            "LOGISTIC_TOP2_CAUSE",
        ],
    )


    logistic_merge_df = pd.DataFrame(
        {
            "SOURCE_RECORD_ID":
                logistic_df[
                    logistic_id_column
                ].astype(
                    str
                ),

            "LOGISTIC_CAUSE":
                logistic_df[
                    logistic_cause_column
                ].map(
                    normalize_category
                ),
        }
    )


    logistic_merge_df[
        "LOGISTIC_PROBABILITY"
    ] = (
        pd.to_numeric(
            logistic_df[
                logistic_probability_column
            ],
            errors="coerce",
        )
        if logistic_probability_column
        else np.nan
    )


    logistic_merge_df[
        "LOGISTIC_TOP2_CAUSE"
    ] = (
        logistic_df[
            logistic_top2_column
        ].map(
            normalize_category
        )
        if logistic_top2_column
        else ""
    )


    logistic_merge_df = (
        logistic_merge_df
        .drop_duplicates(
            subset=[
                "SOURCE_RECORD_ID"
            ],
            keep="last",
        )
    )


    results_df = results_df.merge(
        logistic_merge_df,
        on="SOURCE_RECORD_ID",
        how="left",
    )


    results_df[
        "LOGISTIC_EXACT_MATCH"
    ] = (
        results_df[
            "LOGISTIC_CAUSE"
        ]
        == results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )


    results_df[
        "LOGISTIC_EXACT_OR_RELATED"
    ] = [
        categories_related(
            human_value,
            prediction,
        )
        for human_value, prediction in zip(
            results_df[
                "HUMAN_PRIMARY_CAUSE"
            ],
            results_df[
                "LOGISTIC_CAUSE"
            ],
        )
    ]


    results_df[
        "LOGISTIC_TOP2_EXACT"
    ] = (
        results_df[
            "LOGISTIC_EXACT_MATCH"
        ]
        |
        (
            results_df[
                "LOGISTIC_TOP2_CAUSE"
            ]
            == results_df[
                "HUMAN_PRIMARY_CAUSE"
            ]
        )
    )


    results_df[
        "LOGISTIC_TOP2_EXACT_OR_RELATED"
    ] = [
        (
            categories_related(
                human_value,
                top1_value,
            )
            or categories_related(
                human_value,
                top2_value,
            )
        )
        for (
            human_value,
            top1_value,
            top2_value,
        ) in zip(
            results_df[
                "HUMAN_PRIMARY_CAUSE"
            ],
            results_df[
                "LOGISTIC_CAUSE"
            ],
            results_df[
                "LOGISTIC_TOP2_CAUSE"
            ],
        )
    ]


# =============================================================================
# 25. SAVE RECORD-LEVEL OUTPUTS
# =============================================================================

results_df.to_csv(
    OUTPUT_RESULTS_CSV,
    index=False,
    encoding="utf-8-sig",
)


retrieval_details_df = pd.DataFrame(
    retrieval_detail_records
)


if not retrieval_details_df.empty:

    retrieval_details_df.to_csv(
        OUTPUT_RETRIEVAL_CSV,
        index=False,
        encoding="utf-8-sig",
    )


# =============================================================================
# 26. RETRIEVAL DIAGNOSTICS
# =============================================================================

if not retrieval_details_df.empty:

    retrieval_details_df[
        "EXACT_CAUSE_RETRIEVAL"
    ] = (
        retrieval_details_df[
            "REFERENCE_CAUSE"
        ]
        == retrieval_details_df[
            "TARGET_HUMAN_CAUSE"
        ]
    )


    retrieval_details_df[
        "RELATED_CAUSE_RETRIEVAL"
    ] = [
        categories_related(
            target_value,
            reference_value,
        )
        for (
            target_value,
            reference_value,
        ) in zip(
            retrieval_details_df[
                "TARGET_HUMAN_CAUSE"
            ],
            retrieval_details_df[
                "REFERENCE_CAUSE"
            ],
        )
    ]


    retrieval_quality_by_target_df = (
        retrieval_details_df
        .groupby(
            "TARGET_RECORD_ID"
        )
        .agg(
            Retrieved_records=(
                "REFERENCE_RECORD_ID",
                "count",
            ),
            Top_score=(
                "HYBRID_SCORE",
                "max",
            ),
            Mean_score=(
                "HYBRID_SCORE",
                "mean",
            ),
            Exact_retrieval_rate=(
                "EXACT_CAUSE_RETRIEVAL",
                "mean",
            ),
            Related_retrieval_rate=(
                "RELATED_CAUSE_RETRIEVAL",
                "mean",
            ),
            Human_reference_count=(
                "REFERENCE_SOURCE",
                lambda values:
                    int(
                        (
                            values
                            == "HUMAN_REVIEWED"
                        ).sum()
                    ),
            ),
        )
        .reset_index()
    )


    results_df = results_df.merge(
        retrieval_quality_by_target_df,
        left_on="SOURCE_RECORD_ID",
        right_on="TARGET_RECORD_ID",
        how="left",
    )


else:

    retrieval_quality_by_target_df = pd.DataFrame()


# =============================================================================
# 27. PERFORMANCE UTILITIES
# =============================================================================

def evaluate_method(
    dataframe,
    prediction_column,
    model_name,
    related_column=None,
    top2_column=None,
    top2_related_column=None,
):

    valid_df = dataframe[
        dataframe[
            prediction_column
        ]
        .notna()
    ].copy()


    valid_df = valid_df[
        valid_df[
            prediction_column
        ]
        .astype(
            str
        )
        .str.strip()
        .ne(
            ""
        )
    ]


    if valid_df.empty:

        return {
            "Model":
                model_name,

            "Records":
                0,
        }


    y_true = valid_df[
        "HUMAN_PRIMARY_CAUSE"
    ]


    y_pred = valid_df[
        prediction_column
    ]


    output = {
        "Model":
            model_name,

        "Records":
            len(
                valid_df
            ),

        "Exact accuracy":
            accuracy_score(
                y_true,
                y_pred,
            ),

        "Balanced accuracy":
            balanced_accuracy_score(
                y_true,
                y_pred,
            ),

        "Macro F1":
            f1_score(
                y_true,
                y_pred,
                average="macro",
                zero_division=0,
            ),

        "Weighted F1":
            f1_score(
                y_true,
                y_pred,
                average="weighted",
                zero_division=0,
            ),

        "Cohen's kappa":
            cohen_kappa_score(
                y_true,
                y_pred,
            ),
    }


    if related_column:

        output[
            "Exact-or-related accuracy"
        ] = float(
            valid_df[
                related_column
            ].mean()
        )


    if top2_column:

        output[
            "Top-two exact accuracy"
        ] = float(
            valid_df[
                top2_column
            ].mean()
        )


    if top2_related_column:

        output[
            "Top-two exact-or-related"
        ] = float(
            valid_df[
                top2_related_column
            ].mean()
        )


    return output


# =============================================================================
# 28. OVERALL METHOD COMPARISON
# =============================================================================

method_metrics = [
    evaluate_method(
        results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen",
        related_column="RAG_EXACT_OR_RELATED",
        top2_column="RAG_TOP2_EXACT_MATCH",
        top2_related_column="RAG_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        results_df,
        "RETRIEVAL_ONLY_CAUSE",
        RETRIEVAL_ONLY_METHOD_NAME,
        related_column="RETRIEVAL_ONLY_EXACT_OR_RELATED",
        top2_column="RETRIEVAL_ONLY_TOP2_EXACT_MATCH",
        top2_related_column="RETRIEVAL_ONLY_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen",
        related_column="ORIGINAL_QWEN_EXACT_OR_RELATED",
    ),
]


if logistic_results_file:

    method_metrics.append(
        evaluate_method(
            results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression",
            related_column="LOGISTIC_EXACT_OR_RELATED",
            top2_column="LOGISTIC_TOP2_EXACT",
            top2_related_column="LOGISTIC_TOP2_EXACT_OR_RELATED",
        )
    )


overall_metrics_df = pd.DataFrame(
    method_metrics
)


certain_results_df = results_df[
    ~results_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


certain_metrics = [
    evaluate_method(
        certain_results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen, certain human labels",
        related_column="RAG_EXACT_OR_RELATED",
        top2_column="RAG_TOP2_EXACT_MATCH",
        top2_related_column="RAG_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        certain_results_df,
        "RETRIEVAL_ONLY_CAUSE",
        f"{RETRIEVAL_ONLY_METHOD_NAME}, certain human labels",
        related_column="RETRIEVAL_ONLY_EXACT_OR_RELATED",
        top2_column="RETRIEVAL_ONLY_TOP2_EXACT_MATCH",
        top2_related_column="RETRIEVAL_ONLY_TOP2_EXACT_OR_RELATED",
    ),

    evaluate_method(
        certain_results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen, certain human labels",
        related_column="ORIGINAL_QWEN_EXACT_OR_RELATED",
    ),
]


if logistic_results_file:

    certain_metrics.append(
        evaluate_method(
            certain_results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression, certain human labels",
            related_column="LOGISTIC_EXACT_OR_RELATED",
            top2_column="LOGISTIC_TOP2_EXACT",
            top2_related_column="LOGISTIC_TOP2_EXACT_OR_RELATED",
        )
    )


certain_metrics_df = pd.DataFrame(
    certain_metrics
)


# =============================================================================
# 29. CATEGORY-LEVEL METRICS
# =============================================================================

evaluation_labels = sorted(
    set(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ]
    )
    .union(
        results_df[
            "RAG_PRIMARY_CAUSE"
        ]
    )
    .union(
        results_df[
            "RETRIEVAL_ONLY_CAUSE"
        ]
        .dropna()
        .astype(
            str
        )
    )
    .union(
        results_df[
            "ORIGINAL_QWEN_CAUSE"
        ]
        .dropna()
        .astype(
            str
        )
    )
)


if logistic_results_file:

    evaluation_labels = sorted(
        set(
            evaluation_labels
        )
        .union(
            results_df[
                "LOGISTIC_CAUSE"
            ]
            .dropna()
            .astype(
                str
            )
        )
    )


def category_metrics(
    dataframe,
    prediction_column,
    method_name,
    labels,
):

    precision, recall, f1, support = (
        precision_recall_fscore_support(
            dataframe[
                "HUMAN_PRIMARY_CAUSE"
            ],
            dataframe[
                prediction_column
            ],
            labels=labels,
            zero_division=0,
        )
    )


    return pd.DataFrame(
        {
            "Method":
                method_name,

            "Primary cause":
                labels,

            "Precision":
                precision,

            "Recall":
                recall,

            "F1":
                f1,

            "Human support":
                support,
        }
    )


category_metrics_frames = [
    category_metrics(
        results_df,
        "RAG_PRIMARY_CAUSE",
        "Hybrid RAG-Qwen",
        evaluation_labels,
    ),

    category_metrics(
        results_df,
        "RETRIEVAL_ONLY_CAUSE",
        RETRIEVAL_ONLY_METHOD_NAME,
        evaluation_labels,
    ),

    category_metrics(
        results_df,
        "ORIGINAL_QWEN_CAUSE",
        "Narrative-only Qwen",
        evaluation_labels,
    ),
]


if logistic_results_file:

    category_metrics_frames.append(
        category_metrics(
            results_df,
            "LOGISTIC_CAUSE",
            "TF-IDF logistic regression",
            evaluation_labels,
        )
    )


category_metrics_df = pd.concat(
    category_metrics_frames,
    ignore_index=True,
)


# =============================================================================
# 30. CONFUSION MATRICES
# =============================================================================

rag_confusion_matrix = confusion_matrix(
    results_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    results_df[
        "RAG_PRIMARY_CAUSE"
    ],
    labels=evaluation_labels,
)


retrieval_only_confusion_matrix = confusion_matrix(
    results_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    results_df[
        "RETRIEVAL_ONLY_CAUSE"
    ],
    labels=evaluation_labels,
)


qwen_confusion_matrix = confusion_matrix(
    results_df[
        "HUMAN_PRIMARY_CAUSE"
    ],
    results_df[
        "ORIGINAL_QWEN_CAUSE"
    ],
    labels=evaluation_labels,
)


if logistic_results_file:

    logistic_confusion_matrix = confusion_matrix(
        results_df[
            "HUMAN_PRIMARY_CAUSE"
        ],
        results_df[
            "LOGISTIC_CAUSE"
        ],
        labels=evaluation_labels,
    )


# =============================================================================
# 31. PERFORMANCE BY RETRIEVAL QUALITY
# =============================================================================

if (
    "Top_score"
    in results_df.columns
):

    results_df[
        "RETRIEVAL_SCORE_BAND"
    ] = pd.cut(
        results_df[
            "Top_score"
        ],
        bins=[
            -np.inf,
            0.35,
            0.50,
            0.65,
            np.inf,
        ],
        labels=[
            "Low: Below 0.35",
            "Moderate: 0.35-0.49",
            "Strong: 0.50-0.64",
            "Very strong: 0.65+",
        ],
    )


    retrieval_band_records = []


    for band, group_df in results_df.groupby(
        "RETRIEVAL_SCORE_BAND",
        observed=False,
    ):

        if group_df.empty:

            continue


        retrieval_band_records.append(
            {
                "Retrieval score band":
                    str(
                        band
                    ),

                "Records":
                    len(
                        group_df
                    ),

                "RAG exact":
                    float(
                        group_df[
                            "RAG_EXACT_MATCH"
                        ].mean()
                    ),

                "RAG exact or related":
                    float(
                        group_df[
                            "RAG_EXACT_OR_RELATED"
                        ].mean()
                    ),

                "Retrieval-only exact":
                    float(
                        group_df[
                            "RETRIEVAL_ONLY_EXACT_MATCH"
                        ].mean()
                    ),

                "Original Qwen exact":
                    float(
                        group_df[
                            "ORIGINAL_QWEN_EXACT_MATCH"
                        ].mean()
                    ),

                "Mean exact retrieval rate":
                    float(
                        group_df[
                            "Exact_retrieval_rate"
                        ].mean()
                    ),
            }
        )


    retrieval_band_performance_df = pd.DataFrame(
        retrieval_band_records
    )


else:

    retrieval_band_performance_df = pd.DataFrame()


# =============================================================================
# 32. PERFORMANCE BY SEVERITY AND TIME
# =============================================================================

def grouped_method_comparison(
    dataframe,
    group_column,
):

    records = []


    for group_value, group_df in dataframe.groupby(
        group_column,
        dropna=False,
    ):

        record = {
            group_column:
                group_value,

            "Records":
                len(
                    group_df
                ),

            "RAG exact":
                float(
                    group_df[
                        "RAG_EXACT_MATCH"
                    ].mean()
                ),

            "RAG exact or related":
                float(
                    group_df[
                        "RAG_EXACT_OR_RELATED"
                    ].mean()
                ),

            "Retrieval-only exact":
                float(
                    group_df[
                        "RETRIEVAL_ONLY_EXACT_MATCH"
                    ].mean()
                ),

            "Retrieval-only exact or related":
                float(
                    group_df[
                        "RETRIEVAL_ONLY_EXACT_OR_RELATED"
                    ].mean()
                ),

            "Original Qwen exact":
                float(
                    group_df[
                        "ORIGINAL_QWEN_EXACT_MATCH"
                    ].mean()
                ),

            "Original Qwen exact or related":
                float(
                    group_df[
                        "ORIGINAL_QWEN_EXACT_OR_RELATED"
                    ].mean()
                ),

            "Human uncertain":
                float(
                    group_df[
                        "HUMAN_CAUSE_UNCERTAIN"
                    ].mean()
                ),
        }


        if logistic_results_file:

            record[
                "Logistic exact"
            ] = float(
                group_df[
                    "LOGISTIC_EXACT_MATCH"
                ].mean()
            )


            record[
                "Logistic exact or related"
            ] = float(
                group_df[
                    "LOGISTIC_EXACT_OR_RELATED"
                ].mean()
            )


        records.append(
            record
        )


    return pd.DataFrame(
        records
    )


severity_comparison_df = grouped_method_comparison(
    results_df,
    "SEVERITY_GROUP",
)


time_comparison_df = grouped_method_comparison(
    results_df,
    "TIME_PERIOD",
)


# =============================================================================
# 33. CHANGE ANALYSIS
# =============================================================================

change_summary_df = pd.DataFrame(
    [
        {
            "Outcome":
                "RAG retained the original Qwen cause",

            "Records":
                int(
                    (
                        ~results_df[
                            "RAG_CHANGED_QWEN_CAUSE"
                        ]
                    ).sum()
                ),
        },
        {
            "Outcome":
                "RAG changed the original Qwen cause",

            "Records":
                int(
                    results_df[
                        "RAG_CHANGED_QWEN_CAUSE"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG improved an exact disagreement",

            "Records":
                int(
                    results_df[
                        "RAG_IMPROVED_EXACT"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG changed an exact Qwen match to a disagreement",

            "Records":
                int(
                    results_df[
                        "RAG_REDUCED_EXACT"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG cause evidence traceable to target narrative",

            "Records":
                int(
                    results_df[
                        "RAG_CAUSE_EVIDENCE_VALID"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG processing failures",

            "Records":
                int(
                    results_df[
                        "RAG_PROCESSING_STATUS"
                    ]
                    .ne(
                        "SUCCESS"
                    )
                    .sum()
                ),
        },
    ]
)


# =============================================================================
# 33A. RAG-QWEN VERSUS RETRIEVAL-ONLY ANALYSIS
#
# Shows whether the generator adds value beyond the retrieved labels.
# =============================================================================

retrieval_only_comparison_df = pd.DataFrame(
    [
        {
            "Outcome":
                "Retrieval-only exact human matches",

            "Records":
                int(
                    results_df[
                        "RETRIEVAL_ONLY_EXACT_MATCH"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG-Qwen exact human matches",

            "Records":
                int(
                    results_df[
                        "RAG_EXACT_MATCH"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG-Qwen agreed with the retrieval-only cause",

            "Records":
                int(
                    results_df[
                        "RAG_AGREES_WITH_RETRIEVAL_ONLY"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "RAG-Qwen selected a different cause than retrieval-only",

            "Records":
                int(
                    (
                        ~results_df[
                            "RAG_AGREES_WITH_RETRIEVAL_ONLY"
                        ]
                    ).sum()
                ),
        },
        {
            "Outcome":
                "RAG-Qwen correct where retrieval-only was not",

            "Records":
                int(
                    results_df[
                        "RAG_IMPROVED_OVER_RETRIEVAL_ONLY"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "Retrieval-only correct where RAG-Qwen was not",

            "Records":
                int(
                    results_df[
                        "RAG_REDUCED_VS_RETRIEVAL_ONLY"
                    ].sum()
                ),
        },
        {
            "Outcome":
                "Targets with no retrieved example (fallback cause used)",

            "Records":
                int(
                    results_df[
                        "RETRIEVAL_ONLY_STATUS"
                    ]
                    .ne(
                        "SUCCESS"
                    )
                    .sum()
                ),
        },
    ]
)


# =============================================================================
# 34. SAVE SUMMARY JSON
# =============================================================================

summary_payload = {
    "created_at":
        datetime.now().isoformat(
            timespec="seconds"
        ),

    "model_id":
        MODEL_ID,

    "embedding_model_id":
        EMBEDDING_MODEL_ID,

    "prompt_version":
        PROMPT_VERSION,

    "human_benchmark_records":
        len(
            results_df
        ),

    "human_reference_records":
        int(
            (
                reference_library_df[
                    "REFERENCE_SOURCE"
                ]
                == "HUMAN_REVIEWED"
            ).sum()
        ),

    "qwen_reference_records":
        int(
            (
                reference_library_df[
                    "REFERENCE_SOURCE"
                ]
                == "QWEN_HIGH_CONFIDENCE"
            ).sum()
        ),

    "semantic_weight":
        SEMANTIC_WEIGHT,

    "lexical_weight":
        LEXICAL_WEIGHT,

    "top_k":
        TOP_K_RETRIEVAL,

    "minimum_retrieval_score":
        MIN_RETRIEVAL_SCORE,

    "retrieval_only_method":
        RETRIEVAL_ONLY_METHOD_NAME,

    "retrieval_only_uses_reference_confidence":
        RETRIEVAL_ONLY_USE_REFERENCE_CONFIDENCE,

    "retrieval_only_comparison":
        retrieval_only_comparison_df.to_dict(
            orient="records"
        ),

    "method_metrics":
        overall_metrics_df.to_dict(
            orient="records"
        ),

    "outputs": {
        "pdf":
            OUTPUT_PDF,

        "results_csv":
            OUTPUT_RESULTS_CSV,

        "retrieval_csv":
            OUTPUT_RETRIEVAL_CSV,

        "checkpoint":
            CHECKPOINT_JSONL,
    },
}


with open(
    OUTPUT_SUMMARY_JSON,
    "w",
    encoding="utf-8",
) as output_file:

    json.dump(
        summary_payload,
        output_file,
        indent=2,
        ensure_ascii=False,
        default=str,
    )


# =============================================================================
# 35. PDF REPORT HELPERS
# =============================================================================

pdf = PdfPages(
    OUTPUT_PDF
)


plt.rcParams.update(
    {
        "figure.dpi":
            120,

        "font.size":
            9,

        "axes.grid":
            False,
    }
)


page_counter = {
    "value":
        0
}


def create_page(
    title=None,
    landscape=False,
):

    figure = plt.figure(
        figsize=(
            (
                11,
                8.5,
            )
            if landscape
            else (
                8.5,
                11,
            )
        )
    )


    axis = figure.add_axes(
        [
            0,
            0,
            1,
            1,
        ]
    )


    axis.axis(
        "off"
    )


    if title:

        figure.text(
            0.5,
            0.96,
            title,
            ha="center",
            va="top",
            fontsize=14,
            fontweight="bold",
        )


    return figure


def save_page(
    figure,
):

    page_counter[
        "value"
    ] += 1


    figure.text(
        0.5,
        0.018,
        f"Page {page_counter['value']}",
        ha="center",
        va="bottom",
        fontsize=7.5,
        color="#666666",
    )


    pdf.savefig(
        figure,
        bbox_inches="tight",
    )


    plt.close(
        figure
    )


def add_text_pages(
    title,
    lines,
    font_size=8.4,
    lines_per_page=54,
    wrap_width=106,
):

    wrapped_lines = []


    for line in lines:

        if not line:

            wrapped_lines.append(
                ""
            )

        else:

            wrapped_lines.extend(
                textwrap.wrap(
                    str(
                        line
                    ),
                    width=wrap_width,
                    break_long_words=False,
                    break_on_hyphens=False,
                )
                or [""]
            )


    chunks = [
        wrapped_lines[
            start:
            start + lines_per_page
        ]
        for start in range(
            0,
            len(
                wrapped_lines
            ),
            lines_per_page,
        )
    ] or [[""]]


    for page_number, chunk in enumerate(
        chunks,
        start=1,
    ):

        page_title = title


        if len(
            chunks
        ) > 1:

            page_title += (
                f" - {page_number} "
                f"of {len(chunks)}"
            )


        figure = create_page(
            page_title
        )


        figure.text(
            0.055,
            0.915,
            "\n".join(
                chunk
            ),
            va="top",
            ha="left",
            fontsize=font_size,
            family="monospace",
            linespacing=1.19,
        )


        save_page(
            figure
        )


def add_table_pages(
    title,
    dataframe,
    rows_per_page=20,
    font_size=7.0,
    column_wrap_width=28,
    landscape=False,
):

    if (
        dataframe is None
        or dataframe.empty
    ):

        add_text_pages(
            title,
            [
                "No results were available."
            ],
        )


        return


    printable = dataframe.copy()


    for column in printable.columns:

        if pd.api.types.is_float_dtype(
            printable[
                column
            ]
        ):

            printable[
                column
            ] = printable[
                column
            ].map(
                lambda value:
                    ""
                    if pd.isna(
                        value
                    )
                    else f"{value:.4f}"
            )


    printable.columns = [
        readable_label(
            column
        )
        for column in printable.columns
    ]


    for column in printable.columns:

        printable[
            column
        ] = printable[
            column
        ].map(
            lambda value:
                ""
                if pd.isna(
                    value
                )
                else str(
                    value
                )
        )


        printable[
            column
        ] = printable[
            column
        ].map(
            lambda value:
                "\n".join(
                    textwrap.wrap(
                        value,
                        width=column_wrap_width,
                        break_long_words=False,
                        break_on_hyphens=False,
                    )
                )
                if value
                else ""
        )


    for start in range(
        0,
        len(
            printable
        ),
        rows_per_page,
    ):

        chunk = printable.iloc[
            start:
            start
            + rows_per_page
        ]


        figure = create_page(
            (
                title
                if start == 0
                else title
                + " - Continued"
            ),
            landscape=landscape,
        )


        axis = figure.add_axes(
            [
                0.025,
                0.065,
                0.95,
                0.84,
            ]
        )


        axis.axis(
            "off"
        )


        table = axis.table(
            cellText=chunk.values,
            colLabels=chunk.columns,
            cellLoc="left",
            colLoc="left",
            loc="upper left",
        )


        table.auto_set_font_size(
            False
        )


        table.set_fontsize(
            font_size
        )


        table.scale(
            1,
            1.38,
        )


        for (
            row_number,
            column_number,
        ), cell in table.get_celld().items():

            cell.set_edgecolor(
                "#AAB2BD"
            )


            if row_number == 0:

                cell.set_facecolor(
                    "#D9EAF7"
                )


                cell.set_text_props(
                    weight="bold"
                )


        save_page(
            figure
        )


def add_method_chart(
    metrics_df,
):

    if metrics_df.empty:

        return


    metric_columns = [
        "Exact accuracy",
        "Exact-or-related accuracy",
        "Balanced accuracy",
        "Macro F1",
        "Weighted F1",
        "Cohen's kappa",
    ]


    metric_columns = [
        column
        for column in metric_columns
        if column in metrics_df.columns
    ]


    figure = plt.figure(
        figsize=(
            10.5,
            7,
        )
    )


    axis = figure.add_subplot(
        111
    )


    x_positions = np.arange(
        len(
            metric_columns
        )
    )


    # Bar width scales with the number of methods so four bars do not overlap.
    width = min(
        0.24,
        0.80
        / max(
            len(
                metrics_df
            ),
            1,
        ),
    )


    colors = [
        "#2E6F9E",
        "#7A5195",
        "#B46A3C",
        "#4E8B57",
    ]


    for index, (
        _,
        row,
    ) in enumerate(
        metrics_df.iterrows()
    ):

        values = [
            row.get(
                metric,
                np.nan,
            )
            for metric in metric_columns
        ]


        axis.bar(
            x_positions
            + (
                index
                - (
                    len(
                        metrics_df
                    )
                    - 1
                )
                / 2
            )
            * width,
            values,
            width=width,
            label=row[
                "Model"
            ],
            color=colors[
                index
                % len(
                    colors
                )
            ],
        )


    axis.set_xticks(
        x_positions
    )


    axis.set_xticklabels(
        metric_columns,
        rotation=22,
        ha="right",
    )


    axis.set_ylabel(
        "Metric value"
    )


    axis.set_ylim(
        -0.05,
        1.05,
    )


    axis.set_title(
        "Primary-Cause Performance against Human Review",
        fontsize=13,
        fontweight="bold",
    )


    axis.grid(
        axis="y",
        alpha=0.25,
    )


    axis.legend(
        fontsize=8,
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_confusion_heatmap(
    title,
    matrix,
    labels,
):

    if matrix.size == 0:

        return


    figure = plt.figure(
        figsize=(
            11,
            8.5,
        )
    )


    axis = figure.add_subplot(
        111
    )


    image = axis.imshow(
        matrix,
        aspect="auto",
        cmap="Blues",
    )


    axis.set_xticks(
        range(
            len(
                labels
            )
        )
    )


    axis.set_xticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        rotation=45,
        ha="right",
        fontsize=6.3,
    )


    axis.set_yticks(
        range(
            len(
                labels
            )
        )
    )


    axis.set_yticklabels(
        [
            readable_label(
                label
            )
            for label in labels
        ],
        fontsize=6.3,
    )


    axis.set_xlabel(
        "Predicted primary cause"
    )


    axis.set_ylabel(
        "Human primary cause"
    )


    axis.set_title(
        title,
        fontsize=13,
        fontweight="bold",
    )


    maximum_value = (
        matrix.max()
        if matrix.size
        else 0
    )


    for row_number in range(
        matrix.shape[
            0
        ]
    ):

        for column_number in range(
            matrix.shape[
                1
            ]
        ):

            value = int(
                matrix[
                    row_number,
                    column_number,
                ]
            )


            if value > 0:

                color = (
                    "white"
                    if (
                        maximum_value
                        and value
                        > maximum_value
                        * 0.55
                    )
                    else "black"
                )


                axis.text(
                    column_number,
                    row_number,
                    str(
                        value
                    ),
                    ha="center",
                    va="center",
                    fontsize=6.2,
                    color=color,
                )


    figure.colorbar(
        image,
        ax=axis,
        label="Narrative count",
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_scatter_plot(
    dataframe,
):

    if (
        dataframe.empty
        or "Top_score"
        not in dataframe.columns
    ):

        return


    plot_df = dataframe[
        dataframe[
            "Top_score"
        ]
        .notna()
    ].copy()


    if plot_df.empty:

        return


    figure = plt.figure(
        figsize=(
            10,
            6.5,
        )
    )


    axis = figure.add_subplot(
        111
    )


    colors = np.where(
        plot_df[
            "RAG_EXACT_MATCH"
        ],
        "#4E8B57",
        "#C44E52",
    )


    axis.scatter(
        plot_df[
            "Top_score"
        ],
        plot_df[
            "cause_confidence"
        ],
        c=colors,
        alpha=0.75,
        edgecolor="black",
        linewidth=0.3,
    )


    axis.set_xlabel(
        "Highest retrieved-example hybrid score"
    )


    axis.set_ylabel(
        "RAG-Qwen cause confidence"
    )


    axis.set_title(
        "Retrieval Similarity, RAG Confidence, and Exact Agreement",
        fontsize=13,
        fontweight="bold",
    )


    axis.grid(
        alpha=0.22,
    )


    axis.text(
        0.02,
        0.98,
        "Green = exact human match\nRed = disagreement",
        transform=axis.transAxes,
        va="top",
        fontsize=8,
    )


    figure.tight_layout()


    save_page(
        figure
    )


def add_example_pages(
    title,
    dataframe,
    explanation,
    maximum_records=15,
):

    if dataframe.empty:

        add_text_pages(
            title,
            [
                explanation,
                "",
                "No examples were available.",
            ],
        )


        return


    lines = [
        explanation,
        "",
    ]


    for example_number, (
        _,
        row,
    ) in enumerate(
        dataframe.head(
            maximum_records
        ).iterrows(),
        start=1,
    ):

        lines.extend(
            [
                "=" * 84,
                (
                    f"Example {example_number} | "
                    f"Record {row.get('SOURCE_RECORD_ID', '')} | "
                    f"{row.get('SEVERITY_GROUP', '')} | "
                    f"{row.get('TIME_PERIOD', '')}"
                ),
                (
                    f"Human cause: "
                    f"{row.get('HUMAN_PRIMARY_CAUSE', '')}"
                ),
                (
                    f"Original Qwen: "
                    f"{row.get('ORIGINAL_QWEN_CAUSE', '')}"
                ),
                (
                    f"RAG-Qwen: "
                    f"{row.get('RAG_PRIMARY_CAUSE', '')}"
                ),
                (
                    f"RAG alternative: "
                    f"{row.get('RAG_ALTERNATIVE_CAUSE', '')}"
                ),
                (
                    f"RAG confidence: "
                    f"{safe_float(row.get('cause_confidence', np.nan)):.3f}"
                ),
                (
                    f"Retrieval-only: "
                    f"{row.get('RETRIEVAL_ONLY_CAUSE', '')} "
                    f"(vote share "
                    f"{safe_float(row.get('RETRIEVAL_ONLY_VOTE_SHARE', np.nan)):.3f})"
                ),
                (
                    f"Top retrieved cause: "
                    f"{row.get('TOP_RETRIEVED_CAUSE', '')}"
                ),
                (
                    f"Top retrieval score: "
                    f"{safe_float(row.get('TOP_RETRIEVAL_SCORE', np.nan)):.3f}"
                ),
                (
                    f"Retrieved causes: "
                    f"{row.get('RETRIEVED_CAUSES', '')}"
                ),
                (
                    f"Human uncertain: "
                    f"{row.get('HUMAN_CAUSE_UNCERTAIN', False)}"
                ),
                "",
                "Target narrative:",
                str(
                    row.get(
                        "NARRATIVE",
                        "",
                    )
                ),
                "",
                (
                    "RAG evidence from target: "
                    f"{row.get('cause_evidence_text', '')}"
                ),
                (
                    "Retrieval explanation: "
                    f"{row.get('retrieval_explanation', '')}"
                ),
                (
                    "Human comments: "
                    f"{row.get('HUMAN_REVIEW_COMMENTS', '')}"
                ),
                "",
            ]
        )


    add_text_pages(
        title,
        lines,
        font_size=7.5,
        lines_per_page=47,
        wrap_width=104,
    )


# =============================================================================
# 36. GENERATE DETAILED PDF REPORT
# =============================================================================

# Cover page.
figure = create_page()


figure.text(
    0.5,
    0.79,
    "Hybrid RAG Primary-Cause Classification",
    ha="center",
    fontsize=20,
    fontweight="bold",
)


figure.text(
    0.5,
    0.735,
    "Comparison with Retrieval-Only, Narrative-Only Qwen, Logistic Regression, and Human Review",
    ha="center",
    fontsize=11.5,
    color="#444444",
)


figure.text(
    0.5,
    0.69,
    datetime.now().strftime(
        "Generated %B %d, %Y at %H:%M"
    ),
    ha="center",
    fontsize=9.5,
    color="#666666",
)


cover_lines = [
    f"Human benchmark narratives: {len(results_df):,}",
    (
        "Human-reviewed retrieval references: "
        f"{int((reference_library_df['REFERENCE_SOURCE'] == 'HUMAN_REVIEWED').sum()):,}"
    ),
    (
        "High-confidence Qwen retrieval references: "
        f"{int((reference_library_df['REFERENCE_SOURCE'] == 'QWEN_HIGH_CONFIDENCE').sum()):,}"
    ),
    f"Maximum retrieved examples per target: {TOP_K_RETRIEVAL}",
    f"Semantic retrieval weight: {SEMANTIC_WEIGHT:.2f}",
    f"TF-IDF retrieval weight: {LEXICAL_WEIGHT:.2f}",
    f"Generator: {MODEL_ID}",
    f"Embedding model: {EMBEDDING_MODEL_ID}",
    f"Retrieval-only comparison: {RETRIEVAL_ONLY_METHOD_NAME}",
    "",
    "Objective:",
    "Test whether similar reviewed and high-confidence historical narratives",
    "improve primary-cause classification beyond narrative-only Qwen, and",
    "whether Qwen adds value beyond the retrieved labels alone.",
    "",
    "Important limitation:",
    "Human benchmark records are used through leave-one-out retrieval.",
    "The target record is excluded from its own retrieved examples.",
]


figure.text(
    0.09,
    0.60,
    "\n".join(
        cover_lines
    ),
    va="top",
    ha="left",
    fontsize=8.8,
    family="monospace",
    linespacing=1.22,
)


save_page(
    figure
)


add_text_pages(
    "Executive Summary",
    [
        "This experiment evaluates retrieval-augmented generation for primary",
        "accident-cause classification.",
        "",
        "For each independently reviewed target narrative, the retriever selected",
        "similar historical examples from a hybrid reference library containing",
        "human-reviewed records and high-confidence Code Block 2 results.",
        "",
        "The target record was excluded from its own retrieval results. Identical",
        "target text was also removed before ranking references.",
        "",
        "Semantic embedding similarity and TF-IDF lexical similarity were",
        "combined into a hybrid retrieval score.",
        "",
        "The retrieved examples were supplied to the same local Qwen model used",
        "in the earlier ontology-guided analysis. The prompt required evidence",
        "to come from the target narrative and prohibited transferring facts from",
        "retrieved incidents.",
        "",
        "A retrieval-only comparison was also run on the same retrieved examples.",
        "It assigns the cause with the highest similarity-weighted vote, with no",
        "generator. Comparing RAG-Qwen to retrieval-only shows whether Qwen adds",
        "value beyond copying the labels of the most similar historical records.",
        "",
        "RAG results are compared with retrieval-only, narrative-only Qwen, the",
        "independent human review, and the earlier logistic-regression baseline",
        "when its saved output is available.",
        "",
        "Exact agreement is the primary measure. Exact-or-related and top-two",
        "results are reported as sensitivity measures.",
    ],
)


add_text_pages(
    "RAG Experimental Design",
    [
        "1. Read the completed 100-record human-review workbook.",
        "",
        "2. Read high-confidence, cause-valid Code Block 2 results.",
        "",
        "3. Build one combined retrieval library with source-quality labels.",
        "",
        "4. Encode references and benchmark narratives using a local",
        "   Sentence-Transformers embedding model.",
        "",
        "5. Build a TF-IDF unigram and bigram representation of the same text.",
        "",
        "6. Calculate a hybrid score using 70% semantic similarity and 30%",
        "   lexical similarity.",
        "",
        "7. Retrieve up to five diverse references while limiting duplicate",
        "   cause categories and human examples.",
        "",
        "8. Exclude the target identifier and identical target text.",
        "",
        "9. Provide the target and retrieved examples to Qwen.",
        "",
        "10. Require target-narrative evidence and one ontology cause.",
        "",
        "11. Separately, classify each target by retrieval only: each retrieved",
        "    example votes for its accepted cause, weighted by hybrid score",
        "    multiplied by reference confidence.",
        "",
        "12. Compare RAG-Qwen, retrieval-only, narrative-only Qwen, logistic",
        "    regression, and independent human labels.",
    ],
)


reference_distribution_df = (
    reference_library_df[
        "REFERENCE_SOURCE"
    ]
    .value_counts()
    .rename_axis(
        "Reference source"
    )
    .reset_index(
        name="Records"
    )
)


cause_reference_distribution_df = (
    reference_library_df
    .groupby(
        [
            "REFERENCE_SOURCE",
            "REFERENCE_CAUSE",
        ]
    )
    .size()
    .reset_index(
        name="Records"
    )
    .sort_values(
        [
            "REFERENCE_SOURCE",
            "Records",
        ],
        ascending=[
            True,
            False,
        ],
    )
)


add_table_pages(
    "Hybrid Reference-Library Sources",
    reference_distribution_df,
    rows_per_page=10,
    font_size=7.4,
    column_wrap_width=42,
)


add_table_pages(
    "Reference-Library Cause Distribution",
    cause_reference_distribution_df,
    rows_per_page=28,
    font_size=6.9,
    column_wrap_width=36,
    landscape=True,
)


add_table_pages(
    "Overall Method Comparison",
    overall_metrics_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=34,
    landscape=True,
)


add_method_chart(
    overall_metrics_df
)


add_table_pages(
    "Comparison for Human Labels Not Marked Uncertain",
    certain_metrics_df,
    rows_per_page=12,
    font_size=6.8,
    column_wrap_width=34,
    landscape=True,
)


add_table_pages(
    "RAG Changes Relative to Narrative-Only Qwen",
    change_summary_df,
    rows_per_page=14,
    font_size=7.1,
    column_wrap_width=52,
)


add_text_pages(
    "How to Interpret the RAG Change Results",
    [
        "RAG can help, have no effect, or reduce performance.",
        "",
        "An improvement occurs when retrieval changes an incorrect original Qwen",
        "classification into the human-selected cause.",
        "",
        "A reduction occurs when retrieval changes an exact original Qwen match",
        "into a disagreement.",
        "",
        "The number of changes should therefore be considered together with the",
        "net change in exact accuracy, not treated as improvement by itself.",
        "",
        "Retrieved examples may be relevant while still describing a different",
        "event mechanism. Strong retrieval similarity does not guarantee that",
        "the retrieved cause applies to the target.",
        "",
        "The RAG evidence field is checked against the target narrative to detect",
        "whether the model copied unsupported evidence from a reference record.",
    ],
)


add_table_pages(
    "RAG-Qwen versus Retrieval-Only",
    retrieval_only_comparison_df,
    rows_per_page=14,
    font_size=7.1,
    column_wrap_width=56,
)


add_text_pages(
    "How to Interpret the Retrieval-Only Comparison",
    [
        "Retrieval-only uses the same retrieved examples given to RAG-Qwen, but",
        "assigns the cause with the highest similarity-weighted vote. It does",
        "not read or reason over the target narrative beyond retrieval.",
        "",
        "If RAG-Qwen outperforms retrieval-only, the generator is adding value by",
        "applying the ontology to the target narrative rather than copying the",
        "labels of similar records.",
        "",
        "If retrieval-only performs about as well as RAG-Qwen, most of the RAG",
        "benefit is coming from the retrieved labels, and a lower-cost retrieval",
        "workflow may be sufficient for some uses.",
        "",
        "If retrieval-only performs much worse, similar narratives often carry",
        "different causes, which supports keeping the generator in the loop.",
        "",
        "The retriever caps examples per cause, so the vote is a conservative",
        "nearest-neighbor style baseline rather than a tuned k-NN classifier.",
    ],
)


add_table_pages(
    "Category-Level Primary-Cause Performance",
    category_metrics_df,
    rows_per_page=30,
    font_size=6.5,
    column_wrap_width=28,
    landscape=True,
)


add_confusion_heatmap(
    "Hybrid RAG-Qwen versus Human Review",
    rag_confusion_matrix,
    evaluation_labels,
)


add_confusion_heatmap(
    "Retrieval-Only Weighted Vote versus Human Review",
    retrieval_only_confusion_matrix,
    evaluation_labels,
)


add_confusion_heatmap(
    "Narrative-Only Qwen versus Human Review",
    qwen_confusion_matrix,
    evaluation_labels,
)


if logistic_results_file:

    add_confusion_heatmap(
        "TF-IDF Logistic Regression versus Human Review",
        logistic_confusion_matrix,
        evaluation_labels,
    )


add_table_pages(
    "Performance by Injury Severity",
    severity_comparison_df,
    rows_per_page=12,
    font_size=6.7,
    column_wrap_width=30,
    landscape=True,
)


add_table_pages(
    "Performance by Time Period",
    time_comparison_df,
    rows_per_page=12,
    font_size=6.7,
    column_wrap_width=30,
    landscape=True,
)


if not retrieval_band_performance_df.empty:

    add_table_pages(
        "Performance by Retrieval-Similarity Band",
        retrieval_band_performance_df,
        rows_per_page=12,
        font_size=6.9,
        column_wrap_width=35,
        landscape=True,
    )


    add_scatter_plot(
        results_df
    )


if not retrieval_quality_by_target_df.empty:

    retrieval_summary_df = pd.DataFrame(
        [
            {
                "Measure":
                    "Targets with at least one retrieved reference",

                "Value":
                    int(
                        retrieval_quality_by_target_df[
                            "Retrieved_records"
                        ]
                        .gt(
                            0
                        )
                        .sum()
                    ),
            },
            {
                "Measure":
                    "Mean top retrieval score",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Top_score"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean exact-cause retrieval rate",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Exact_retrieval_rate"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean exact-or-related retrieval rate",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Related_retrieval_rate"
                        ].mean()
                    ),
            },
            {
                "Measure":
                    "Mean human-reviewed references per target",

                "Value":
                    float(
                        retrieval_quality_by_target_df[
                            "Human_reference_count"
                        ].mean()
                    ),
            },
        ]
    )


    add_table_pages(
        "Retrieval-Quality Summary",
        retrieval_summary_df,
        rows_per_page=12,
        font_size=7.2,
        column_wrap_width=48,
    )


improved_examples_df = results_df[
    results_df[
        "RAG_IMPROVED_EXACT"
    ]
].copy()


reduced_examples_df = results_df[
    results_df[
        "RAG_REDUCED_EXACT"
    ]
].copy()


rag_disagreement_df = results_df[
    ~results_df[
        "RAG_EXACT_MATCH"
    ]
].copy()


uncertain_examples_df = results_df[
    results_df[
        "HUMAN_CAUSE_UNCERTAIN"
    ]
].copy()


add_example_pages(
    "Examples Where Retrieval Improved the Exact Classification",
    improved_examples_df,
    (
        "The original narrative-only Qwen classification differed from the "
        "human label, while RAG-Qwen exactly matched it. These examples show "
        "where similar historical narratives may have clarified terminology "
        "or a category boundary."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Examples Where Retrieval Reduced Exact Performance",
    reduced_examples_df,
    (
        "The original Qwen classification exactly matched the human label, "
        "but RAG-Qwen changed the result. These examples are important because "
        "they show how relevant-looking retrieval can introduce anchoring or "
        "category-transfer errors."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Remaining RAG-Qwen Disagreements",
    rag_disagreement_df,
    (
        "These cases did not exactly match the human primary-cause label. "
        "Review should consider the alternative cause, retrieved examples, "
        "human uncertainty, and whether the categories are closely related."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_example_pages(
    "Human-Uncertain RAG Cases",
    uncertain_examples_df,
    (
        "The human reviewer marked these primary-cause decisions uncertain. "
        "Differences are retained for transparency but should not automatically "
        "be interpreted as definitive RAG errors."
    ),
    maximum_records=TOP_DISAGREEMENT_EXAMPLES,
)


add_text_pages(
    "Interpretation Framework",
    [
        "The central question is whether retrieval improves classification over",
        "the same Qwen model operating on the target narrative alone.",
        "",
        "The cleanest evidence of benefit is:",
        "",
        "1. Higher exact human agreement.",
        "2. Higher macro and weighted F1.",
        "3. Improvement in difficult cause categories.",
        "4. Traceable evidence copied from the target narrative.",
        "5. More improvements than reductions among changed predictions.",
        "6. Better performance than retrieval-only on the same examples.",
        "",
        "Retrieval quality must be evaluated separately from final classification.",
        "A reference can be semantically similar while having a different cause.",
        "",
        "RAG should be considered useful only when retrieved examples improve the",
        "target classification without causing the model to transfer unsupported",
        "facts from historical cases.",
        "",
        "If retrieval improves only exact-or-related agreement but not exact",
        "agreement, it may be helping with broad mechanism families without",
        "consistently resolving the final ontology boundary.",
    ],
)


add_text_pages(
    "Limitations",
    [
        "1. The human-reviewed reference library is small and is used through",
        "   leave-one-out retrieval rather than as a fully separate training set.",
        "",
        "2. High-confidence Code Block 2 labels were generated by Qwen and are",
        "   not independently verified ground truth.",
        "",
        "3. Retrieval may reinforce errors or dominant categories already present",
        "   in the Code Block 2 reference library.",
        "",
        "4. Similar narratives can describe different immediate mechanisms.",
        "",
        "5. The hybrid retrieval weights and minimum score are starting settings",
        "   rather than optimized values.",
        "",
        "6. Sentence embeddings and TF-IDF represent similarity differently and",
        "   may retrieve examples related by equipment, body part, or activity",
        "   rather than primary cause.",
        "",
        "7. The experiment evaluates primary cause only. Retrieval should not be",
        "   assumed to solve missing information about contributing factors or",
        "   corrective actions.",
        "",
        "8. Human labels can be uncertain, and multiple related causes may be",
        "   defensible for some short narratives.",
        "",
        "9. Exact-or-related metrics depend on predefined cause relationships and",
        "   remain secondary to exact agreement.",
    ],
)


add_text_pages(
    "Potential Business Use",
    [
        "RAG may provide value beyond classification by retrieving similar",
        "historical incidents for safety review.",
        "",
        "A practical workflow could present:",
        "",
        "1. The proposed primary cause.",
        "2. The target-narrative evidence.",
        "3. The most similar reviewed historical incidents.",
        "4. The source and review quality of each retrieved example.",
        "5. An alternative primary cause when the category boundary is unclear.",
        "",
        "This could support investigation preparation, lessons-learned searches,",
        "and review of recurring accident mechanisms.",
        "",
        "The RAG output should remain decision support. Retrieved incidents do not",
        "prove that the target event had the same conditions or root causes.",
    ],
)


add_text_pages(
    "Technical Appendix",
    [
        f"Generator model: {MODEL_ID}",
        f"Embedding model: {EMBEDDING_MODEL_ID}",
        f"Prompt version: {PROMPT_VERSION}",
        f"Benchmark records: {len(results_df):,}",
        f"Hybrid reference records: {len(reference_library_df):,}",
        f"Top-K retrieval: {TOP_K_RETRIEVAL}",
        f"Semantic weight: {SEMANTIC_WEIGHT:.2f}",
        f"Lexical weight: {LEXICAL_WEIGHT:.2f}",
        f"Minimum retrieval score: {MIN_RETRIEVAL_SCORE:.2f}",
        f"Minimum Code Block 2 confidence: "
        f"{MIN_BLOCK2_CAUSE_CONFIDENCE:.2f}",
        f"Maximum generated tokens: {MAX_NEW_TOKENS}",
        f"Maximum validation attempts: {MAX_RETRIES}",
        f"Retrieval-only method: {RETRIEVAL_ONLY_METHOD_NAME}",
        f"Retrieval-only uses reference confidence: "
        f"{RETRIEVAL_ONLY_USE_REFERENCE_CONFIDENCE}",
        f"Retrieval-only fallback cause: {RETRIEVAL_ONLY_FALLBACK_CAUSE}",
        "",
        f"PDF: {os.path.abspath(OUTPUT_PDF)}",
        f"Results CSV: {os.path.abspath(OUTPUT_RESULTS_CSV)}",
        f"Retrieval details: {os.path.abspath(OUTPUT_RETRIEVAL_CSV)}",
        f"Checkpoint: {os.path.abspath(CHECKPOINT_JSONL)}",
        f"Summary JSON: {os.path.abspath(OUTPUT_SUMMARY_JSON)}",
    ],
    font_size=8.2,
)


# =============================================================================
# 37. PDF METADATA AND CLOSE
# =============================================================================

metadata = pdf.infodict()


metadata[
    "Title"
] = (
    "MSHA Hybrid RAG Primary-Cause Classification Comparison Report"
)


metadata[
    "Author"
] = (
    "Python, Sentence-Transformers, and Qwen2.5-7B-Instruct"
)


metadata[
    "Subject"
] = (
    "Comparison of hybrid retrieval-augmented generation, retrieval-only, "
    "narrative-only Qwen, logistic regression, and independent human "
    "primary-cause labels"
)


metadata[
    "Keywords"
] = (
    "MSHA, retrieval-augmented generation, RAG, retrieval-only, Qwen, TF-IDF, "
    "sentence embeddings, accident cause, human validation"
)


metadata[
    "CreationDate"
] = datetime.now()


pdf.close()


# =============================================================================
# 38. FINAL CONSOLE SUMMARY
# =============================================================================

print("\n" + "=" * 95)
print("HYBRID RAG PRIMARY-CAUSE ANALYSIS COMPLETE")
print("=" * 95)


print(
    f"Benchmark records: "
    f"{len(results_df):,}"
)


print(
    f"Reference-library records: "
    f"{len(reference_library_df):,}"
)


print(
    f"Successful RAG classifications: "
    f"{int(results_df['RAG_PROCESSING_STATUS'].eq('SUCCESS').sum()):,}"
)


print(
    f"RAG exact human matches: "
    f"{int(results_df['RAG_EXACT_MATCH'].sum()):,}"
)


print(
    f"RAG exact-or-related human matches: "
    f"{int(results_df['RAG_EXACT_OR_RELATED'].sum()):,}"
)


print(
    f"Retrieval-only exact human matches: "
    f"{int(results_df['RETRIEVAL_ONLY_EXACT_MATCH'].sum()):,}"
)


print(
    f"Retrieval-only exact-or-related human matches: "
    f"{int(results_df['RETRIEVAL_ONLY_EXACT_OR_RELATED'].sum()):,}"
)


print(
    f"RAG improvements over original Qwen: "
    f"{int(results_df['RAG_IMPROVED_EXACT'].sum()):,}"
)


print(
    f"RAG reductions versus original Qwen: "
    f"{int(results_df['RAG_REDUCED_EXACT'].sum()):,}"
)


print(
    f"RAG correct where retrieval-only was not: "
    f"{int(results_df['RAG_IMPROVED_OVER_RETRIEVAL_ONLY'].sum()):,}"
)


print(
    f"Retrieval-only correct where RAG was not: "
    f"{int(results_df['RAG_REDUCED_VS_RETRIEVAL_ONLY'].sum()):,}"
)


print(
    f"PDF report: "
    f"{os.path.abspath(OUTPUT_PDF)}"
)


print(
    f"Results CSV: "
    f"{os.path.abspath(OUTPUT_RESULTS_CSV)}"
)


print(
    f"Retrieval details: "
    f"{os.path.abspath(OUTPUT_RETRIEVAL_CSV)}"
)


print(
    f"Checkpoint: "
    f"{os.path.abspath(CHECKPOINT_JSONL)}"
)


print("=" * 95)


# =============================================================================
# 39. NOTEBOOK DISPLAYS
# =============================================================================

display(
    overall_metrics_df
)


display(
    certain_metrics_df
)


display(
    change_summary_df
)


display(
    retrieval_only_comparison_df
)


display(
    category_metrics_df
)


display(
    severity_comparison_df
)


if not retrieval_band_performance_df.empty:

    display(
        retrieval_band_performance_df
    )
````
